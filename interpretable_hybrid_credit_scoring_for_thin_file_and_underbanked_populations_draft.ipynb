This repository contains an **independent**, professional-grade Python implementation of the research methodology from the **arXiv preprint** entitled **"Interpretable hybrid credit scoring for thin-file and underbanked populations: A fairness-aware residual learning framework with evidence from East African financial inclusion data"** by:

*   **Belise Kanziga**
*   **Yaé U. Gaba**
*   **Olivier Kanamugire**

The paper provides no code; this project translates its mathematical specifications into robust, versioned, executable software artifacts. The project delivers a **decision rule whose auditability is a measured quantity rather than an assertion**: a logistic-regression scorecard corrected by a gradient-boosted learner fitted on its raw response residuals, decomposed at every prediction into an interpretability ratio $\rho(x)$ that reports the share of the decision attributable to the linear branch, together with a fairness audit conducted **at the granularity of the three interpretability regions** — the contribution that aggregate fairness metrics structurally cannot provide.

## Introduction

This project provides a Python implementation of the analytical framework presented in Kanziga, Gaba and Kanamugire (2026). The core of this repository is the iPython Notebook `interpretable_hybrid_credit_scoring_for_thin_file_and_underbanked_populations_draft.ipynb`, which contains a comprehensive suite of **170 callables (165 functions, 2 classes and 3 class methods)** across Tasks 1 to 23, including the two orchestration layers.

Credit scoring in thin-file and underbanked populations poses a specific institutional problem. Logistic-regression scorecards remain the regulatory default because their coefficients *are* the explanation, and adverse-action notices can be generated directly from them. Gradient-boosted ensembles predict better but explain worse, and post-hoc attributions describe a surrogate rather than the decision rule itself. The residual-learning hybrid resolves the tension additively: the scorecard produces $p_{\text{LR}}(x)$, a shallow booster learns the raw response residual $r_i = y_i - p_{\text{LR}}(x_i)$, and the two combine into a single prediction whose linear share is computable per borrower.

That per-borrower share is what makes the framework auditable, and it is also what creates the risk this paper isolates. If the opaque region is not distributed evenly across the population, a lender can satisfy aggregate fairness tests while systematically routing protected subgroups into the part of the model where explanations are weakest. On the East African data the routing gaps are large: **18 percentage points** rural versus urban, **32 points** primary-or-less versus secondary-and-above, **22 points** Ugandan versus Kenyan, and essentially none by gender.

This codebase operationalizes the framework, allowing users to:

-   Validate the study configuration against **32 required top-level sections**, the equation-parameter map covering Equations (1)–(12) plus five unnumbered equations, and every pinned hyperparameter, halting on the first failure with a descriptive error naming the full dotted key path.
-   Build a Weight-of-Evidence scorecard branch under a strict information barrier: monotonic bin boundaries, WoE tables, imputation medians and modes, and winsorization bounds all estimated on the **training partition only** and frozen for validation and test.
-   Fit the residual booster with early stopping monitored on the **validation AUC of the provisional hybrid** $p_{\text{LR}}(x) + \hat{r}(x)$ — a criterion no native boosting library can express, implemented here as a custom training callback.
-   Select the adaptive weighting triple $(\alpha_{\min}, \alpha_{\max}, \gamma)$ by exhaustive search over all **27** grid points with five-fold stratified cross-validation, refitting the scorecard, the residuals and the booster inside every fold.
-   Decompose every prediction into $\rho(x)$ and assign it to one of three interpretability regions at the thresholds $0.9$ and $0.7$.
-   Audit fairness **within each region**, with disparate impact and equalised-odds differences computed against configurable reference groups and degenerate cells reported as explicitly undefined (`"---"`) rather than silently zero.
-   Quantify uncertainty with a stratified paired bootstrap ($B = 2000$) that resamples within class strata and the Sun–Xu fast midrank DeLong test for two correlated ROC curves.
-   Segment the thin-file subpopulation by a frozen composite index and test the two pre-registered hypotheses, including the one the manuscript reports as **not supported**.
-   Execute the entire workflow through the single top-level orchestrator `run_research_pipeline(...) -> Dict[str, Any]`.

> **Reproduction status — read this before running anything.** The notebook ships with a schema-conformant **synthetic corpus** generated in the *Study Inputs* cell from `np.random.default_rng(71)`, because the workspace carries no raw CSVs. It reproduces the published row counts, column order and class balance exactly, so every code path executes end to end — but the Zindi label it produces is a deterministic threshold on a linear function of six features, and no metric computed on it is a reproduction of the paper. Supplying the genuine Zindi and Taiwan files is a prerequisite for reproduction, not an optimisation. See [Input Data Structure](#input-data-structure).

## Theoretical Background

The implemented methods combine Statistical Machine Learning, Mathematical Statistics, Numerical Optimization, Algorithmic Fairness, Credit-Risk Scorecard Methodology, and Reproducible Research Engineering.

**1. Weight-of-Evidence Encoding and Information Value.** The linear branch consumes a WoE-transformed matrix. For bin or level $k$ with $g_k$ observations of the negative class and $b_k$ of the positive class, against training totals $G$ and $B$:

$$
\text{WoE}_k = \ln\!\left(\frac{g_k/G}{b_k/B}\right), \qquad \text{IV} = \sum_{k}\left(\frac{g_k}{G} - \frac{b_k}{B}\right)\text{WoE}_k .
$$

The class convention is pinned by the configuration: $g_k$ counts $y = 0$ and $b_k$ counts $y = 1$. Any bin or level with a zero count in either class makes the logarithm undefined and receives the documented **neutral value** $\text{WoE}_k = 0$, which is also the value assigned to a category absent from the training partition. Features are retained when $\text{IV} \ge 0.02$. Continuous predictors are binned from twenty quantile bins, merged until the WoE sequence is monotone, then merged again to satisfy a minimum bin frequency of five percent of training observations, and monotonicity is re-enforced afterwards because frequency merging can break it.

**2. The Hybrid Prediction Rule.** The scorecard produces $p_{\text{LR}}(x) = \sigma(\hat{\beta}^\top x + \hat{\beta}_0)$ under L2-regularised maximum likelihood with balanced class weights, the inverse-regularisation constant $C$ selected by five-fold stratified cross-validation over $\{0.01, 0.1, 1, 10\}$. The booster is fitted on the raw response residual $r_i = y_i - p_{\text{LR}}(x_i)$ under squared-error loss, and the two branches combine as

$$
p_{\text{Hybrid}}(x) = \mathrm{clip}\!\left(p_{\text{LR}}(x) + \alpha^*(x)\,\hat{r}(x),\; 0,\; 1\right), \qquad \hat{r}(x) = f_{\text{GBM}}(x) = \sum_{m=1}^{M}\nu\, h_m(x).
$$

The residual definition is load-bearing. Because $r_i$ is a *raw response* residual rather than a gradient, the booster's output lives on the probability scale, which is what permits the additive decomposition below to be read directly as a share.

**3. Adaptive Weighting.** The weight on the residual correction is not constant. It is driven by an uncertainty proxy that peaks where the scorecard is least decisive:

$$
u_{\text{LR}}(x) = 4\,p_{\text{LR}}(x)\left(1 - p_{\text{LR}}(x)\right) \in [0,1], \qquad \alpha^*(x) = \alpha_{\min} + \left(\alpha_{\max} - \alpha_{\min}\right)u_{\text{LR}}(x)^{\gamma}, \quad \gamma > 0 .
$$

The triple is selected over the Cartesian product $\{0.0, 0.1, 0.2\} \times \{0.3, 0.5, 0.8\} \times \{0.5, 1.0, 2.0\}$ — 27 candidates — by mean validation AUC across five stratified folds.

> **Registered divergence (1 of 5).** Section 3.4 of the manuscript specifies five-fold cross-validation for all three adaptive parameters; Table 1 states three-fold. The implementation resolves the inconsistency to **five-fold for every triple** and logs the resolution verbatim in the audit trail rather than choosing silently.

**4. The Interpretability Ratio and the Three-Region Partition.** Every prediction decomposes into a linear share:

$$
\rho(x) = \frac{\left|p_{\text{LR}}(x)\right|}{\left|p_{\text{LR}}(x)\right| + \left|\alpha^*(x)\,\hat{r}(x)\right|},
\qquad
\text{Region}(x) =
\begin{cases}
\text{Fully interpretable} & \rho(x) \ge 0.9,\\
\text{Partially interpretable} & 0.7 \le \rho(x) < 0.9,\\
\text{ML-driven} & \rho(x) < 0.7.
\end{cases}
$$

$\rho(x) = 1$ exactly when the residual correction vanishes. The thresholds are **indicative rather than theoretically derived**, and their sensitivity is evaluated empirically at $(0.85, 0.65)$ and $(0.95, 0.75)$.

> **Registered divergence (2 of 5).** The ratio is defined on the **probability scale**, not the logit scale. Borrowers with $p_{\text{LR}}(x)$ near one therefore tend to fall into the fully-interpretable region whenever the residual correction is bounded, which mechanically couples that region to the high-$p_{\text{LR}}$ tail. The empirical finding that the fully-interpretable region concentrates high-base-rate borrowers is consequently a joint property of the ratio's algebraic form and the scorecard's calibration on that tail — not a discovered feature independent of the definition. A logit-scale variant is left as future work by the manuscript and is not implemented here.

**5. Probability Calibration.** The hybrid produces values inside the unit interval but does not guarantee calibration. Platt scaling is applied post hoc, with the two scalars estimated by **unpenalised maximum likelihood** on the held-out validation fold:

$$
p_{\text{cal}}(x) = \frac{1}{1 + \exp\!\left(a\, p_{\text{Hybrid}}(x) + b\right)},
\qquad
\text{ECE} = \sum_{k=1}^{K}\frac{|B_k|}{n}\left|\mathrm{acc}(B_k) - \mathrm{conf}(B_k)\right|, \quad K = 10 .
$$

Region assignments are computed on the **uncalibrated** hybrid probabilities: calibration changes the probability scale, never $\rho(x)$ or region membership.

> **Registered divergence (3 of 5).** Equation (10) carries a **positive** exponent, whereas `sklearn.linear_model.LogisticRegression` parameterises a fitted binary model with a negative one. The two conventions differ by a sign on both scalars, so $a = -w$ and $b = -c$ for library coefficient $w$ and intercept $c$. Exporting the library pair under the manuscript's names makes Equation (10) evaluate to $1 - p_{\text{cal}}$ for every input, inverting every decision taken by a consumer who reconstructs calibration from the reported parameters. This implementation exports the **manuscript pair**, persists both under unambiguous keys with the negation relationship stated, and asserts the round trip numerically to $10^{-10}$ on every fit.

**6. Fairness at the Granularity of Interpretability Regions.** Predicted positives are $\hat{y} = \mathbf{1}[p_{\text{cal}}^{\text{test}} \ge 0.5]$. Against a reference group $g_{\text{ref}}$:

$$
\text{DI}(g) = \frac{\Pr[\hat{y} = 1 \mid A = g]}{\Pr[\hat{y} = 1 \mid A = g_{\text{ref}}]},
\qquad
\text{EOD} = \max_{g}\left\{\left|\text{TPR}(g) - \text{TPR}(g_{\text{ref}})\right|,\ \left|\text{FPR}(g) - \text{FPR}(g_{\text{ref}})\right|\right\},
$$

each computed **pooled and restricted to each region**. The routing quantity itself is $\hat{\Pr}\!\left[\rho(x) < 0.7 \mid A = g\right]$, with 95% percentile bootstrap intervals from $B = 500$ resamples drawn within $(\text{class} \times \text{attribute})$ strata.

Two disciplines govern the interpretation. First, **degenerate cells**: when the reference group has no predicted positives in a region, the TPR and FPR denominators collapse and the cell is reported as undefined, never as zero. Second, **post-treatment conditioning**: $\rho(x)$ is causally downstream of the features, so per-region disparate-impact and equalised-odds values are *descriptive stratifications*, not causal fairness measures. The four-fifths convention ($0.8$–$1.25$) is applied as the reporting band, following the U.S. EEOC Uniform Guidelines.

**7. Inference.** Two complementary vehicles, deliberately not interchangeable. The stratified paired bootstrap ($B = 2000$) resamples with replacement **within** the positive and negative test strata separately, preserving the marginal class distribution, and computes both models' AUCs on the same resample to form $\Delta\text{AUC}^{(b)}$; the interval is the 2.5th and 97.5th percentiles. Models are never refitted inside the bootstrap — the routine consumes pre-computed test predictions, which makes refitting structurally impossible rather than merely prohibited. The DeLong test uses the Sun–Xu fast midrank algorithm for the covariance of two correlated AUC estimates:

$$
z = \frac{\text{AUC}_1 - \text{AUC}_2}{\sqrt{\widehat{\mathrm{Var}}\!\left(\text{AUC}_1 - \text{AUC}_2\right)}},
$$

reported one-sided and two-sided. No multiple-comparison correction is applied, consistent with the descriptive audit intent, and that decision is recorded in the audit log rather than left implicit.

**8. Prediction-Level Explanations.** For observations in the ML-driven region the residual term dominates and the logistic coefficients no longer suffice. TreeSHAP attributions are computed on $f_{\text{GBM}}$ **only** — never on the full hybrid — so the additive decomposition survives, and the booster is sliced to its frozen best iteration before explanation so no tree absent from prediction is attributed:

$$
\overline{\left|\phi_j\right|} = \frac{1}{n_{\text{ML}}}\sum_{x \in \text{ML-region}}\left|\phi_j(x)\right| .
$$

Stability is screened by $B = 500$ observation-level bootstrap resamples, reporting each feature's rank standard deviation and the top-five set preservation rate, because post-hoc attributions are known to be attackable and an unstable ranking is a flag on the individual explanation rather than on the framework.

**9. Thin-File Segmentation.** The composite index sums three components standardised on the training partition only, with the threshold $\tau$ frozen at the training 25th percentile:

$$
\text{index} = z_{\text{education}} + z_{\text{cellphone}} + z_{\text{formaljob}}, \qquad z_j = \frac{x_j - \mu_j^{\text{train}}}{\sigma_j^{\text{train}}}, \qquad \text{thin-file} = \{\text{index} < \tau\}.
$$

> **Registered divergence (4 of 5).** The manuscript does not specify the standard-deviation estimator. The implementation uses the **population** form ($\text{ddof} = 0$) where the common library default is the sample form, a deterministic choice that shifts $\tau$ slightly and is recorded as a placeholder.

**10. Alternating-Minimisation Joint Optimisation.** The joint objective couples both branches:

$$
\mathcal{L}\!\left(\beta, f^{(t)}\right) = \sum_{i=1}^{N}\ell\!\left(y_i,\ \mathrm{clip}\!\left(p_{\text{LR}}(x_i;\beta) + \alpha^*(x_i;\beta)f^{(t)}(x_i),\,0,\,1\right)\right) + \lambda_{\text{LR}}\lVert\beta\rVert_2^2 ,
$$

with $\ell(y,p) = -[y\ln p + (1-y)\ln(1-p)]$. The fixed-point scheme alternates an L-BFGS update of $\beta$, residual recomputation, a booster refit, and a validation-AUC evaluation, converging at tolerance $10^{-3}$. The **mandatory rule** is that $\alpha^*(x;\beta)$ be recomputed from the updated $p_{\text{LR}}$ at every iteration; this implementation computes it *inside the objective*, which makes a stale weight unreachable rather than merely avoided.

> **Registered divergence (5 of 5).** The scheme is **not** coordinate descent on $\mathcal{L}$, because the booster sub-step minimises squared error on residuals rather than the joint cross-entropy. The manuscript states this explicitly and the implementation preserves the statement in its conclusion rather than overclaiming convergence. Two-stage training remains the operational default, and the scheme is not run on Taiwan.


## Features

The provided iPython Notebook (`interpretable_hybrid_credit_scoring_for_thin_file_and_underbanked_populations_draft.ipynb`) implements the full research pipeline, including:

-   **Modular, 23-Task Architecture:** 170 callables decomposed into 23 tasks, each with standalone step functions plus a task orchestrator, so a defect localises to a step rather than to a monolith. Every callable carries complete type hints, NumPy-style docstrings stating purpose, parameters, processes, outputs and raises, and per-line comments citing the governing equation.
-   **Configuration-Driven Design:** All study parameters live in `study_config.json` (mirrored as `config.yaml`) as the single source of truth — **32 top-level sections**, an `equation_parameter_map` covering Equations (1)–(12) plus five unnumbered equations, and every `UNSPECIFIED IN MANUSCRIPT` value annotated with its rationale.
-   **Strict Information Barrier:** Every transformation statistic — medians, modes, winsorization bounds, monotonic bin edges, WoE tables, one-hot level sets, the selected $C$, the adaptive triple, the Platt pair — is estimated on training and frozen. The orchestrator enforces this with before-and-after SHA-256 digests of every frozen artifact, partition-access declarations checked *before* each stage is dispatched, and an optional runtime cross-check that recomputes the frozen statistics from training rows alone.
-   **Hybrid-AUC Early Stopping:** The residual booster halts on the validation AUC of the provisional hybrid, which native boosting early stopping cannot express because its metric cannot observe $p_{\text{LR}}$. Implemented as an `xgboost.callback.TrainingCallback` subclass that predicts with an inclusive `iteration_range` each round and freezes the best iteration.
-   **Falsifiable Verification, Not Decorative:** The lightweight verifier evaluates the nine mandated intermediate quantities at the strength the specification declares, in exactly four comparison modes (exact equality, absolute tolerance, percentage-point tolerance, ordered sequence equality). Membership and threshold predicates are **deliberately unavailable**, because a check satisfied by every state the pipeline can produce carries no information. A **negative self-test** corrupts each quantity in turn and asserts the corresponding check fails; a check that cannot be made to fail does not ship.
-   **Two Verification Classes, Separated:** Structural failures — partition sizes, label domains, array alignments — halt the run. Manuscript-value failures are reproduction *findings* and are recorded without halting. Collapsing both behind a single switch would mean that disabling the reproduction comparison also disables every structural guarantee.
-   **Per-Stage Checkpointing and Telemetry:** After every stage the orchestrator serialises the frozen statistics to a deterministic, byte-stable checkpoint (sorted keys, fixed separators), records wall-clock time, resident-set size and its delta, and captures every warning the stage emitted rather than letting a convergence warning vanish to the console.
-   **Degenerate-Cell Discipline:** A module-level `UNDEFINED = "---"` sentinel is returned whenever a reference group has no predicted positives in a region — the requirement fairness implementations most often get wrong by writing `0.0` into a cell whose denominator vanished.
-   **Dual-Parameterisation Calibration Export:** Platt parameters are persisted in both the manuscript and library conventions under distinct keys, with the negation relationship stated in the artifact, so a supervisor reconstructing calibration from the reported pair cannot obtain an inverted model.
-   **Refit-Proof Inference:** The bootstrap signature accepts probability vectors rather than estimators, so the prohibition on refitting inside the resampling loop is enforced by design rather than by convention.
-   **Robustness Suite:** Alternative interpretability thresholds, alternative decision thresholds ($0.4$, $0.6$), a LightGBM residual-learner sensitivity check, and reference-group sensitivity across female, rural, primary-or-less and Rwanda.
-   **Single Top-Level Interface:** `run_research_pipeline(...)` executes the entire study and returns a consolidated artifact dictionary.

## Methodology Implemented

The core analytical steps directly implement the methodology from the paper:

1.  **Validation (Task 1):** `orchestrate_validate_ingested_parameters` verifies the 32 required configuration sections, the equation-parameter map, every pinned hyperparameter under `isinstance` type contracts and `math.isclose` float identities, and both raw frames — shape, column order (reordering rather than halting, and recording the event), label dtype and positive rate, zero missingness, country composition, and the verbatim preservation of the source typo `"Divorced/Seperated"`. Nested key access is guarded so a malformed configuration raises a descriptive `ValueError` naming the full dotted path, never a bare `KeyError`.
2.  **Cleansing (Task 2):** `orchestrate_cleanse_data` drops `uniqueid` and `year` from a copy, strips whitespace on every object column, re-validates each value against the declared exhaustive level sets, enforces integer dtype contracts, and persists a cleansing audit carrying dropped columns, the **structured** level-validation results, before-and-after SHA-256 digests and a UTC timestamp.
3.  **Partitioning (Task 3):** `orchestrate_create_partitions` applies a chained stratified split — 60/40, then 50/50 on the holdout — mapping holdout-relative indices back to original positions and asserting the three arrays cover the row range exactly once. Zindi yields **14,114 / 4,705 / 4,705**; Taiwan yields **18,000 / 6,000 / 6,000**.
4.  **Preprocessing (Task 4):** `orchestrate_fit_preprocessing_stats` computes medians for numeric features and the first sorted mode for categoricals (a determinism rule, since pandas returns modes in ascending order), and the 1st/99th training percentiles, then applies the frozen statistics as $\min(\max(x, q_{01}), q_{99})$.
5.  **WoE Encoding (Task 5):** `orchestrate_fit_woe_encoding` bins from twenty quantile bins with a progressive fallback on `qcut` failure, enforces monotonicity by merging the violating adjacent pair with the closest WoE values, enforces the five-percent frequency floor by merging the **least-frequent** violating bin toward its closest-WoE neighbour, then re-enforces monotonicity. Interval keys are asserted unique so a rounding collision cannot silently overwrite a bin.
6.  **IV Filtering and One-Hot Encoding (Task 6):** `orchestrate_iv_filter_and_ohe` computes the Information Values, retains features at $\text{IV} \ge 0.02$, and builds the boosting branch's one-hot matrix with frozen level sets, reindexing each partition against the frozen column list with zero fill so an unseen level yields the all-zero indicator vector in both drift directions.
7.  **Logistic Baseline (Task 7):** `orchestrate_fit_logistic` selects $C$ by five-fold stratified CV, fits with `lbfgs`, L2 and balanced class weights at `max_iter=1000`, and logs a convergence warning when the solver reaches the iteration cap — a diagnostic the specification does not require and which separates an under-converged coefficient vector from a converged one.
8.  **Residuals (Task 8):** `orchestrate_compute_residuals` computes $r_i = y_i - p_{\text{LR}}(x_i)$ on all three partitions, asserts containment in $[-1, 1]$, and **halts** when the training residual standard deviation falls below $10^{-3}$, which would indicate an essentially constant scorecard.
9.  **Residual Booster (Task 9):** `orchestrate_fit_residual_xgboost` fits `reg:squarederror` at depth 3, learning rate 0.03, 400 estimators, subsample and column-subsample 0.8, $\lambda = 2$, with patience-50 early stopping on the provisional hybrid validation AUC, then asserts $\mathrm{corr}(r, \hat{r}) > 0$.
10. **Standalone Comparator (Task 10):** `orchestrate_fit_standalone_xgboost` fits `binary:logistic` at depth 4 with $\text{scale\_pos\_weight} = n_0/n_1$ computed on training and recorded exactly.
11. **Adaptive Weighting (Task 11):** `orchestrate_select_adaptive_weights` enumerates all 27 triples under five-fold stratified CV, refitting the scorecard, residuals and booster within each fold, and freezes the per-observation $\alpha^*$ arrays.
12. **Hybrid and Decomposition (Task 12):** `orchestrate_build_hybrid_and_rho` forms the clipped hybrid, computes $\rho(x)$ with a zero-denominator guard returning $\rho = 1$ (the correct limit when both contributions vanish), assigns regions, and aggregates region shares, within-region positive rates and the full $\rho$ distribution.
13. **Calibration (Task 13):** `orchestrate_fit_platt` fits the unpenalised Platt model on validation, asserts the Equation (10) round trip and **strict** monotonicity with an explicit positive-slope check, applies the frozen model to test, and fits the LR+Platt ablation that isolates calibration gain from residual reshaping.
14. **Evaluation (Task 14):** `orchestrate_compute_metrics` computes AUC, $\text{Gini} = 2\,\text{AUC} - 1$ and minority-class F1 at threshold 0.5, plus Brier and ECE over ten equal-width bins with reliability-diagram data, for all five candidate models, and assembles Tables 2 and 3.
15. **Fairness Audit (Task 15):** `orchestrate_run_fairness_audit` recodes the four protected attributes with their reference groups, computes region routing rates with stratified bootstrap intervals, and computes pooled and per-region DI and EOD under the degenerate-cell rule.
16. **Inference (Task 16):** `orchestrate_run_bootstrap_delong` runs the stratified paired bootstrap and the Sun–Xu DeLong test, consolidating Table 9 and recording the deliberate absence of multiple-comparison correction.
17. **Attribution (Task 17):** `orchestrate_run_shap_analysis` restricts to $\rho(x) < 0.7$, explains the sliced residual booster only, ranks by mean absolute SHAP, and runs the 500-resample stability screen with conditional diagnostics.
18. **Thin-File Analysis (Task 18):** `orchestrate_run_thin_file` builds the frozen composite index, applies $\tau$ to test, computes subgroup discrimination and calibration with paired bootstrap intervals, and tests both pre-registered hypotheses — preserving the finding that H2 is **not** supported, with the suppression mechanism stated rather than a bare null.
19. **Joint Optimisation (Task 19):** `orchestrate_run_alternating_min` runs the fixed-point scheme with mandatory $\alpha^*$ recomputation, records the convergence trajectory and per-iteration wall-clock cost, and documents that two-stage training remains the deployment default. It is not run on Taiwan.
20. **Pipeline Orchestration (Task 20):** `run_full_study` builds the 19-entry callable registry, executes it in strict dependency order with each callable invoked exactly once, enforces partition-access declarations before dispatch and the information barrier after, checkpoints the frozen statistics, logs time, memory and warnings, runs the verifier after **every** stage, and writes the reproduction report against the declared headline values.
21. **Robustness (Task 21):** `orchestrate_run_robustness` runs the interpretability-threshold, decision-threshold, LightGBM and reference-group sensitivity analyses, reiterating the post-treatment conditioning caveat and the degenerate-cell rule.
22. **Continuity Benchmark (Task 22):** `orchestrate_taiwan_reproduction` applies the placeholder preprocessing protocol and the identical model chain to the Taiwan data, then walks the six-item fidelity checklist on any deviation.
23. **Top-Level Orchestration (Task 23):** `run_research_pipeline` composes the Tasks 1–19 chain, the robustness suite and the continuity benchmark, consolidating every artifact into a single run manifest.

> **Registered divergence — Taiwan preprocessing.** The manuscript states that Taiwan preprocessing is inherited without modification from prior work and does not restate it. Every Taiwan preprocessing decision in this implementation is therefore a documented placeholder, logged as `UNSPECIFIED IN MANUSCRIPT`, and Taiwan deviations are recorded transparently rather than resolved by adjusting the Zindi pipeline — which the protocol explicitly forbids.

## Core Components (Notebook Structure)

The notebook is structured as a logical pipeline with modular orchestrator functions for each major task. Every function carries detailed type hints, technical docstrings, granular per-line comments citing the governing equations, input validation, and comprehensive error handling. Verified structure: **45 cells — 19 markdown and 26 code.**

-   **Paper preamble (cells 0–15):** Markdown cells summarising the paper — abstract, equations, algorithms, network architectures, system IPO, reproduction steps, essential sections, skills, and the four data-structure specifications.
-   **Import block (cells 16–17):** The `# Import Essential Modules` heading and the consolidated import cell. `from __future__ import annotations` is the first executable statement of the notebook.
-   **Implementation anchor (cells 18–19):** The `# Implementation` heading and the `# Draft 1` anchor, below which every generated cell is inserted. Cell order is part of the specification.
-   **Study Inputs (cell 20):** Loads `study_config.json` and constructs `zindi_raw` and `taiwan_raw`. **This is the cell that must be replaced with real data ingestion** — see [Input Data Structure](#input-data-structure).
-   **Per-task cells (cells 21–43):** One cell per task, Tasks 1 to 23, each carrying its step functions, module-level helpers and constants, and the task orchestrator, under banner separators.
-   **Study Execution (cell 44):** Configures the notebook-level logger and invokes `run_research_pipeline`, then reports the stages executed and the reproduction pass count.

The notebook shares **one namespace**: a name defined in the Task 3 cell is visible to the Task 8 cell without any import. Seven symbols are intentionally defined in two cells each (`_frame_digest`, `_partition_frames`, `PIPELINE_ORDER`, `RHO_ML_THRESHOLD`, `REGION_FULLY`, `REGION_PARTIAL`, `REGION_ML`); all are byte-identical at the AST level, so the later binding winning is behaviour-neutral.

## Key Callable: `run_research_pipeline`

The project is designed around a single, top-level user-facing interface function:

-   **`run_research_pipeline(config, zindi_raw, taiwan_raw, output_dir, run_taiwan=False, verify_manuscript_targets=True) -> Dict[str, Any]`:** This master orchestrator runs the entire study. A single call transforms the configuration and the two raw frames into the complete artifact archive: Step 1 executes the Tasks 1–19 chain through the Task 20 registry, in strict dependency order with each callable invoked exactly once, under frozen-state snapshots and the test-partition information barrier; Step 2 runs the Task 21 robustness suite and, when `run_taiwan=True`, the Task 22 continuity benchmark on an isolated state with its own nested output directory and its own reproduction report; Step 3 consolidates every artifact and persists the run manifest. The returned dictionary carries `stages_executed`, `pipeline_order`, the `chain` block (validation report, cleansing audit, both partition manifests, preprocessing statistics, IV values, retained features, logistic branch, residual summary, residual booster, standalone comparator, adaptive block, region statistics and the Platt parameters in both conventions), `robustness`, `taiwan`, `execution_log` and `reproduction_report`.

The underlying Task 20 entry point, `run_full_study(config, zindi_raw, taiwan_raw, output_dir, run_taiwan, verify_manuscript_targets, verify_structural, verify_manuscript_values, verify_frozen_statistics)`, exposes the two independent verification classes and the runtime leakage cross-check for callers who need finer control than the top-level wrapper provides.

## Prerequisites

-   Python 3.11+.
-   Core dependencies, with the versions this implementation was verified against: `numpy` (2.3.5), `pandas` (2.3.3), `scipy` (1.16.3), `scikit-learn` (1.7.2), `xgboost` (3.4.1), `lightgbm` (4.7.0), `shap` (0.52.0).
-   `psutil` is required for the per-stage memory log on Windows, where the POSIX `resource` module is unavailable. The import is guarded and the memory probe degrades to `None` rather than raising, so its absence never aborts a study.
-   A CPU-only laptop is sufficient. The dominant cost is the Task 11 adaptive grid search — 27 triples × 5 folds, each refitting the scorecard, the residuals and the booster — which accounts for roughly 200 seconds of a ~4-minute end-to-end run at Zindi scale.
-   **The two source datasets.** The **Zindi Financial Inclusion in Africa** competition data (23,524 respondents across Kenya, Rwanda, Tanzania and Uganda, survey years 2016–2018, derived from Financial Sector Deepening Network survey instruments) and the **Taiwan Credit Default** data from the UCI Machine Learning Repository (30,000 accounts, 25 columns). **No code can manufacture these files.** The notebook currently substitutes a synthetic corpus in their absence; replacing that substitution with real ingestion is the single prerequisite for reproduction.

## Input Data Structure

The pipeline requires the configuration plus two raw frames, all validated by Task 1:

1.  **`config: Dict[str, Any]`** — Parsed from `study_config.json`. **32 top-level sections**, from `study_metadata` through `equation_parameter_map` and `code_and_data_availability`. Task 1 halts on the first failure.

2.  **`zindi_raw: pd.DataFrame`** — Shape **(23,524, 13)** in the declared column order: `uniqueid`, `country`, `year`, `location_type`, `cellphone_access`, `gender_of_respondent`, `relationship_with_head`, `marital_status`, `education_level`, `job_type`, `household_size`, `age_of_respondent`, `bank_account`. The label is `bank_account` (`int64`; **1 = owns a formal bank account**) at a positive rate of **0.1408**. Country composition must be Rwanda 8,735 / Tanzania 6,620 / Kenya 6,068 / Uganda 2,101. `uniqueid` and `year` are dropped in Task 2 as non-predictive — `year` is perfectly collinear with `country`. Zero missing values are permitted. The source typo `"Divorced/Seperated"` in `marital_status` **must be preserved verbatim**; correcting it fails validation.

    > **Label semantics.** The label is account *ownership*, not default. The manuscript reframes it deliberately as an eligibility-gate proxy: account ownership and loan default sit on opposite sides of the lender's selection funnel. This reframing is the paper's argument, and consumers of the output must not read `bank_account = 1` as a default event.

3.  **`taiwan_raw: pd.DataFrame`** — Shape **(30,000, 25)**, label `default.payment.next.month` at a positive rate of approximately **0.22**. `PAY_0` is present and `PAY_1` is absent, matching the source; `SEX` ⊆ {1,2}, `EDUCATION` ⊆ {0,…,6}, `MARRIAGE` ⊆ {0,1,2,3}. The `ID` column is dropped in Task 2, leaving 24 columns.

> **The synthetic-corpus substitution, stated plainly.** In the shipped notebook the *Study Inputs* cell does not read either file. It constructs both frames from `np.random.default_rng(71)`, reproducing the published row counts, column order and class balance exactly so that every downstream contract is satisfied. The Zindi label is then assigned as the top 14.08% of a deterministic logistic score in six features — `age_of_respondent`, `household_size`, `education_level`, `cellphone_access`, `job_type`, `country` — which makes the label a noiseless threshold on those features. The consequence is that the Information Value filter retains exactly those six and discards the other four as noise, and every discrimination metric is inflated far above the published figures. **Replace the cell with a real ingestion path before drawing any conclusion.** A provenance gate that fails closed when the raw frames are generator-derived is the recommended first extension.

## **Usage Example**

Here is a granular, step-by-step guide to executing the end-to-end pipeline for **"Interpretable hybrid credit scoring for thin-file and underbanked populations"**. This example demonstrates how to (i) load `study_config.json` from the working directory into a Python dictionary, (ii) construct the two raw study frames at the published composition, and (iii) execute the full study via the `run_research_pipeline(...) → Dict[str, Any]` interface.

> **Environment assumption (mandatory).** All pipeline callables from Tasks 1–23 are assumed to be defined **in a single Jupyter notebook** cell-context. There is **no `.py` module** to import, so this example imports only the libraries needed for configuration loading, frame construction and result inspection. `study_config.json` is assumed to be saved in the working directory under that exact name.

> **Verification status.** The execution in Step 3 was run end to end against the real orchestrator before being written into this document. All 19 registry stages completed, 19 per-stage checkpoints were written, and the reported figures below are the actual outputs of that run.

### **Step 1: Loading the Configuration (`study_config.json`)**

`study_config.json` is the single frozen source of truth. It is read into memory and converted into the nested `config` dictionary that every task orchestrator consumes.

```python
# ==============================================================================
# Step 1 - Load study_config.json from the Working Directory
# ==============================================================================

# Standard library and third-party imports required by this example only;
# every pipeline callable is already defined in the notebook context.
import json
from pathlib import Path
from typing import Any, Dict, List, Tuple

import numpy as np
import pandas as pd


def load_study_configuration(
    filepath: str = "study_config.json",
) -> Dict[str, Any]:
    """
    Load the frozen study configuration from the working directory.

    Purpose:
        Ingests the thirty-two configuration sections consumed by every task
        orchestrator - study metadata and headline reproduction targets,
        dataset parameters, raw data schemas, partitioning, preprocessing,
        WoE and Information Value settings, the four model configurations,
        adaptive weighting, hybrid prediction, interpretability thresholds,
        calibration, evaluation metrics, protected attributes, the fairness
        audit, thin-file parameters, joint optimisation, bootstrap and
        DeLong inference, SHAP settings, the training protocol, plotting
        defaults and the equation-parameter map - and converts them into
        the nested dictionary the pipeline consumes.

    Inputs:
        filepath (str): Path to the JSON file, default 'study_config.json',
            assumed to be in the working directory.

    Processes:
        1. Validate the argument type and the file's existence.
        2. Parse the document with an explicit UTF-8 decode, because the
           configuration carries mathematical symbols that the Windows
           console default codepage cannot represent.
        3. Verify that the required top-level sections are present.
        4. Report the section inventory to the operator.

    Outputs:
        Dict[str, Any]: The complete nested configuration dictionary.

    Raises:
        TypeError: If filepath is not a string.
        FileNotFoundError: If the configuration file is absent.
        json.JSONDecodeError: If the document is malformed.
        ValueError: If a required top-level section is missing.
    """
    # Reject a non-string path before touching the filesystem
    if not isinstance(filepath, str):
        raise TypeError(f"filepath must be a string; got {type(filepath)}.")
    # Resolve the path relative to the working directory
    config_path: Path = Path(filepath)
    # A missing configuration would silently break every downstream gate
    if not config_path.is_file():
        raise FileNotFoundError(
            f"Configuration file '{filepath}' not found in the working "
            f"directory {Path.cwd()}."
        )
    # Open with an explicit encoding; the default codepage would fail here
    with open(config_path, "r", encoding="utf-8") as handle:
        # Parse the document into the nested configuration dictionary
        config: Dict[str, Any] = json.load(handle)
    # The pipeline reads these sections unconditionally
    required_sections: Tuple[str, ...] = (
        "study_metadata", "dataset_parameters", "raw_data_schemas",
        "partitioning_parameters", "preprocessing_parameters",
        "woe_parameters", "iv_parameters",
        "logistic_regression_parameters", "residual_xgboost_parameters",
        "standalone_xgboost_parameters", "adaptive_weighting_parameters",
        "hybrid_prediction_parameters", "interpretability_parameters",
        "calibration_parameters", "evaluation_metrics_parameters",
        "protected_attribute_parameters", "fairness_audit_parameters",
        "thin_file_parameters", "joint_optimization_parameters",
        "bootstrap_inference_parameters", "delong_test_parameters",
        "shap_explanation_parameters", "equation_parameter_map",
    )
    # Fail loudly on a truncated configuration rather than at first use
    absent: List[str] = [s for s in required_sections if s not in config]
    if absent:
        raise ValueError(
            f"study_config.json is missing required section(s): {absent}."
        )
    # Report the inventory so the operator can see what was loaded
    print(f"Loaded configuration from {filepath}: "
          f"{len(config)} top-level sections.")
    return config


# Load the frozen study configuration from the working directory
config: Dict[str, Any] = load_study_configuration("study_config.json")

# Confirm the gates that depend most directly on the configuration
print(f"  zindi n              : "
      f"{config['dataset_parameters']['zindi_fsd']['n_observations']}")
print(f"  zindi positive rate  : "
      f"{config['dataset_parameters']['zindi_fsd']['positive_class_rate']}")
print(f"  partition seed       : "
      f"{config['partitioning_parameters']['random_seed']}")
print(f"  expected train n     : "
      f"{config['partitioning_parameters']['zindi_train_n']}")
print(f"  headline targets     : "
      f"{list(config['study_metadata']['headline_results'])}")
```

**Discussion of Step 1.** The explicit UTF-8 decode is not stylistic. The configuration carries mathematical symbols that the Windows console default codepage cannot represent, and an unguarded read fails with a `UnicodeDecodeError` on the encoding rather than on any real defect. The loader fails loudly on a missing file or a truncated section rather than proceeding with an empty dictionary, consistent with the pipeline's fail-fast gates.

### **Step 2: Constructing the Two Raw Study Frames**

Task 1 validates both frames against `config["dataset_parameters"]` and **halts** on mismatch, so the frames must hit the published composition *by construction* rather than by sampling luck.

#### The binding constraint that dictates the construction

`validate_zindi_raw` enforces, among other checks:

$$
n_{\text{observed}} = 23{,}524 \quad \text{(exact equality)},
$$
$$
\left|\hat{\pi}_{\text{positive}} - 0.1408\right| \le 10^{-4}, \qquad \hat{\pi}_{\text{positive}} = \frac{1}{n}\sum_{i=1}^{n}\mathbf{1}\{y_i = 1\},
$$
$$
\text{country counts} = \{\text{Rwanda}: 8735,\ \text{Tanzania}: 6620,\ \text{Kenya}: 6068,\ \text{Uganda}: 2101\}.
$$

A tolerance of $10^{-4}$ on the positive rate at $n = 23{,}524$ admits a drift of roughly two observations. Sampling labels as `Bernoulli(0.1408)` would fail that band with overwhelming probability; the construction therefore allocates an **exact count** and shuffles.

```python
# ==============================================================================
# Step 2 - Construct the Two Raw Study Frames at the Published Composition
# ==============================================================================

# NOTE: this constructs a SCHEMA-CONFORMANT corpus, not the real data. It
# satisfies every Task 1 contract so the pipeline executes end to end, but the
# metrics it produces are NOT a reproduction of the paper. Replace this step
# with real ingestion of the Zindi and Taiwan files before drawing conclusions.

def build_zindi_frame(config: Dict[str, Any], seed: int = 71) -> pd.DataFrame:
    """
    Construct the raw Zindi frame at the published composition.

    Purpose:
        Produces a (23524, 13) frame satisfying every Task 1 contract: the
        exact row count, the declared column order, the published country
        composition, the 0.1408 positive rate, integer dtypes on the three
        integer columns, object dtypes on the eight categorical columns and
        zero missing values.

    Inputs:
        config (Dict[str, Any]): The loaded study configuration, read for
            the declared categorical level sets and column order.
        seed (int): The generator seed, fixing the construction.

    Processes:
        1. Allocate the country vector at the exact published counts.
        2. Draw the demographic and infrastructure covariates.
        3. Assign the label by EXACT count so the rate gate is deterministic.
        4. Reorder the columns to the declared raw column order.

    Outputs:
        pd.DataFrame: The raw Zindi frame, shape (23524, 13).

    Raises:
        KeyError: If the configuration lacks the Zindi dataset block.
    """
    # The published composition is the construction target
    zindi_block: Dict[str, Any] = config["dataset_parameters"]["zindi_fsd"]
    # A dedicated generator keeps the construction reproducible
    rng: np.random.Generator = np.random.default_rng(seed)
    # The declared row count
    n: int = int(zindi_block["n_observations"])
    # The declared categorical level sets
    levels: Dict[str, List[str]] = zindi_block["categorical_levels"]
    # Allocate the country vector at the EXACT published counts
    country: np.ndarray = np.concatenate(
        [[name] * count
         for name, count in zindi_block["country_composition"].items()]
    )
    # Age is a bounded integer covariate
    age: np.ndarray = np.round(
        rng.normal(40, 14, n).clip(16, 85)).astype(np.int64)
    # Household size is a small positive integer
    household_size: np.ndarray = rng.integers(1, 12, n).astype(np.int64)
    # Education, cellphone access and job type are categorical
    education: np.ndarray = rng.choice(
        ["No formal education", "Primary education",
         "Secondary education", "Tertiary education"],
        n, p=[0.35, 0.30, 0.25, 0.10])
    cellphone: np.ndarray = rng.choice(["Yes", "No"], n, p=[0.7, 0.3])
    job_type: np.ndarray = rng.choice(
        ["Government", "Formally employed Private sector",
         "Informally employed", "Self employed",
         "Farming and Fishing", "Remittance Dependent"], n)
    # EXACT positive count so the 1e-4 rate gate passes deterministically
    n_positive: int = int(round(
        float(zindi_block["positive_class_rate"]) * n))
    # Allocate the label vector, then shuffle so positives are not contiguous
    bank_account: np.ndarray = np.zeros(n, dtype=np.int64)
    bank_account[:n_positive] = 1
    rng.shuffle(bank_account)
    # Assemble the frame; year is collinear with country by construction
    frame: pd.DataFrame = pd.DataFrame({
        "uniqueid": [f"uniqueid_{i:05d}" for i in range(n)],
        "country": country,
        "year": np.where(country == "Kenya", 2018,
                         np.where(country == "Rwanda", 2016, 2017)),
        "location_type": rng.choice(levels["location_type"], n),
        "cellphone_access": cellphone,
        "gender_of_respondent": rng.choice(
            levels["gender_of_respondent"], n),
        "relationship_with_head": rng.choice(
            levels["relationship_with_head"], n),
        # The source typo 'Divorced/Seperated' MUST survive verbatim
        "marital_status": rng.choice(levels["marital_status"], n),
        "education_level": education,
        "job_type": job_type,
        "household_size": household_size,
        "age_of_respondent": age,
        "bank_account": bank_account,
    })
    # Reorder to the declared raw column order; Task 1 checks this
    return frame[zindi_block["raw_column_order"]]


def build_taiwan_frame(config: Dict[str, Any], seed: int = 71) -> pd.DataFrame:
    """
    Construct the raw Taiwan continuity frame at the published composition.

    Purpose:
        Produces a (30000, 25) frame satisfying the Task 1 Taiwan contract:
        the declared 25-column order, the label at a positive rate of
        approximately 0.22, PAY_0 present and PAY_1 absent, and the declared
        domains on SEX, EDUCATION and MARRIAGE.

    Inputs:
        config (Dict[str, Any]): The loaded configuration, read for the
            declared Taiwan schema column list.
        seed (int): The generator seed, fixing the construction.

    Processes:
        1. Draw the limit, demographic, repayment and amount columns.
        2. Assign the label by EXACT count at the published rate.
        3. Reorder to the declared schema column order.

    Outputs:
        pd.DataFrame: The raw Taiwan frame, shape (30000, 25).

    Raises:
        KeyError: If the configuration lacks the Taiwan schema block.
    """
    # The declared schema fixes the column order Task 1 validates
    schema_columns: List[str] = [
        column["name"] for column in
        config["raw_data_schemas"]["taiwan_credit_schema"]["columns"]
    ]
    # A dedicated generator keeps the construction reproducible
    rng: np.random.Generator = np.random.default_rng(seed)
    # The declared row count for the continuity benchmark
    n: int = 30000
    # Assemble the identifier, credit limit and demographics
    frame: pd.DataFrame = pd.DataFrame({
        "ID": np.arange(1, n + 1),
        "LIMIT_BAL": rng.integers(10000, 1000000, n),
        "SEX": rng.choice([1, 2], n),
        "EDUCATION": rng.choice([0, 1, 2, 3, 4, 5, 6], n),
        "MARRIAGE": rng.choice([0, 1, 2, 3], n),
        "AGE": rng.integers(21, 79, n),
    })
    # PAY_0 is present and PAY_1 is absent, matching the source
    for index in [0, 2, 3, 4, 5, 6]:
        frame[f"PAY_{index}"] = rng.integers(-2, 9, n)
    # The six billing and six payment amount columns
    for index in range(1, 7):
        frame[f"BILL_AMT{index}"] = rng.integers(0, 100000, n)
        frame[f"PAY_AMT{index}"] = rng.integers(0, 50000, n)
    # EXACT positive count at the published default rate
    default_flag: np.ndarray = np.zeros(n, dtype=np.int64)
    default_flag[rng.permutation(n)[:6600]] = 1
    frame["default.payment.next.month"] = default_flag
    # Reorder to the declared schema column order
    return frame[schema_columns]


# Construct both frames once for the pipeline run
zindi_raw: pd.DataFrame = build_zindi_frame(config)
taiwan_raw: pd.DataFrame = build_taiwan_frame(config)

# Report each frame against its published composition
print(f"zindi_raw  : {zindi_raw.shape}  "
      f"positive_rate={zindi_raw['bank_account'].mean():.4f}  "
      f"missing={int(zindi_raw.isnull().sum().sum())}")
print(f"taiwan_raw : {taiwan_raw.shape}  "
      f"positive_rate="
      f"{taiwan_raw['default.payment.next.month'].mean():.4f}")
print(f"countries  : "
      f"{ {k: int(v) for k, v in zindi_raw['country'].value_counts().items()} }")
```

**Discussion of Step 3.** The orchestrator call is deliberately thin because every intermediate callable runs internally. Four behaviours are worth stating explicitly:

-   **The verifier runs after every stage, not once at the end.** A structural failure therefore halts at the stage that produced it and writes a diagnostic naming that stage, rather than surfacing after the full run with no attribution. A manuscript-value discrepancy is recorded and the run continues, because a reproduction discrepancy is a finding rather than a fault.
-   **The information barrier is enforced in both directions.** Partition-access declarations are checked *before* each stage is dispatched — a fitting-sequence stage that declares test access is rejected — and the test indices, labels and row order are digested and compared *after*. Setting `verify_frozen_statistics=True` on the underlying `run_full_study` additionally recomputes the frozen medians and winsorization bounds from training rows alone, which is the only check that directly detects estimation over held-out rows.
-   **The split asymmetry is not incidental.** $C$ is selected by cross-validation within training; the residual booster early-stops on validation; Platt is fitted on validation; AUC, Brier, ECE, the fairness audit and all inference are computed on test. No selection decision touches test.
-   **Region assignments are computed on uncalibrated probabilities.** Calibration shifts the probability scale and therefore F1 and Brier, but it must not shift $\rho(x)$ or region membership, and the implementation preserves that separation.

> **Realism note.** At Zindi scale the end-to-end run takes roughly four minutes on a CPU-only machine, dominated by the Task 11 adaptive grid search (27 triples × 5 folds ≈ 200 s) and the Task 16 bootstrap (2,000 resamples ≈ 19 s). Setting `run_taiwan=True` adds the continuity benchmark. Separately: the frames constructed in Step 2 reproduce the published *composition* but carry no real dependence structure, so they exercise every code path without reproducing the manuscript's AUC values. Reproducing those figures requires the genuine Zindi and Taiwan files, which no generator can manufacture.

### **Summary of the Execution Flow**

1.  **Configuration Loading.** `load_study_configuration("study_config.json")` parses the thirty-two sections under an explicit UTF-8 decode and validates that the pipeline-critical sections are present, failing loudly otherwise.
2.  **Frame Construction.** Both frames are built at the published composition by **exact count**, so the Task 1 rate gates pass deterministically rather than by sampling luck, with the declared column order, dtypes and level sets satisfied.
3.  **Validation and Cleansing (Tasks 1–2).** The configuration, both frames, the country composition, the label rates and every categorical level set are validated; `uniqueid` and `year` are dropped; whitespace is stripped; dtype contracts are enforced; and a cleansing audit carrying the structured level-validation results and before/after digests is persisted before any fitting.
4.  **Partitioning and Preprocessing (Tasks 3–4).** The chained stratified 60/40 then 50/50 split yields 14,114 / 4,705 / 4,705 under seed 20260830, with coverage asserted exactly; medians, modes and 1st/99th percentiles are frozen on training and applied to all partitions.
5.  **Encoding (Tasks 5–6).** Monotonic binning from twenty quantile bins with least-frequent-bin merging and a re-enforced monotonicity pass; WoE by Equation (2) with the neutral-zero rule; the $\text{IV} \ge 0.02$ filter; and the one-hot boosting matrix with frozen level sets and the all-zero indicator rule.
6.  **Model Fitting (Tasks 7–11).** $C$ selection by five-fold CV; the residual booster with hybrid-AUC early stopping; the standalone comparator at $\text{scale\_pos\_weight} = n_0/n_1$; and the 27-triple adaptive grid search.
7.  **Decomposition, Calibration and Evaluation (Tasks 12–14).** The clipped hybrid, $\rho(x)$ with its zero-denominator guard, the three-region partition, Platt scaling with the Equation (10) round-trip assertion, and the five-model metric tables.
8.  **Audit and Inference (Tasks 15–18).** The per-region fairness audit with degenerate-cell handling; the stratified paired bootstrap and DeLong test; TreeSHAP on the sliced residual booster with the stability screen; and the thin-file segmentation with both pre-registered hypotheses.
9.  **Joint Optimisation and Orchestration (Tasks 19–23).** The alternating fixed-point scheme with mandatory $\alpha^*$ recomputation; the registry execution with checkpointing, telemetry and per-stage verification; the robustness suite; the optional continuity benchmark; and the consolidated run manifest.

## Customization

The pipeline is highly customizable via the `study_config.json` file (or its `config.yaml` rendering). Users can modify:

-   **Partitioning and preprocessing:** `partitioning_parameters` (fractions and `random_seed`); `preprocessing_parameters` (imputation rules, winsorization percentiles, and the monotonic-binning placeholder — starting bin count, monotonicity rule, and the five-percent frequency floor).
-   **Scorecard encoding:** `woe_parameters` (the pinned class convention $g_k = y{=}0$, the neutral-zero rare-category rule) and `iv_parameters` (the $0.02$ retention threshold).
-   **Models:** `logistic_regression_parameters` (the $C$ grid, solver, penalty, class weighting, `max_iter`); `residual_xgboost_parameters` (objective, depth, $\nu$, subsampling, $\lambda$, early-stopping patience); `standalone_xgboost_parameters`; `lightgbm_sensitivity_parameters`.
-   **Adaptive weighting and decomposition:** `adaptive_weighting_parameters` (the three grids and the CV fold count); `interpretability_parameters` (the $0.9$ and $0.7$ region thresholds and the $\rho$ formula record).
-   **Calibration and evaluation:** `calibration_parameters` (the ECE bin count $K = 10$); `evaluation_metrics_parameters` (the decision threshold and the metric set).
-   **Fairness:** `protected_attribute_parameters` (the four attributes, their category mappings and reference groups) and `fairness_audit_parameters` (the four-fifths band, the degenerate-cell sentinel, and the routing-rate bootstrap count).
-   **Inference:** `bootstrap_inference_parameters` ($B = 2000$ for AUC differences, $500$ for fairness quantities) and `delong_test_parameters`.
-   **Interpretability and segmentation:** `shap_explanation_parameters` (the stability resample count and the top-$k$ set size) and `thin_file_parameters` (the ordinal education mapping, the formal-job mapping and the 25th-percentile threshold rule).
-   **Verification:** the two independent flags exposed on `run_full_study` — `verify_structural` (defaults true and should stay true) and `verify_manuscript_values` — plus `verify_frozen_statistics` for the runtime leakage cross-check.

> **The registered divergences are exposed, not hidden.** The 3-fold/5-fold resolution is logged verbatim in the audit trail; the Platt parameterisation is persisted under both conventions with the negation relationship stated; the Taiwan placeholder protocol is flagged `UNSPECIFIED IN MANUSCRIPT` on every decision; the probability-scale coupling of $\rho(x)$ is documented at its point of use; and the $\text{ddof} = 0$ estimator in the thin-file index is recorded as a placeholder.

Every parameter marked `UNSPECIFIED IN MANUSCRIPT` carries its rationale in the configuration and in the run manifest.

## Contributing

Contributions are welcome. Please fork the repository, create a feature branch, and submit a pull request with a clear description of your changes. Adherence to PEP 8, complete type hinting, technical docstrings, input validation, and error handling is required. Contributors must additionally:

-   **Never use the test partition for model selection.** $C$ selection, early stopping, the adaptive triple and calibration are all validation or training decisions. AUC, Brier, ECE, the fairness audit and every inference statistic are test quantities.
-   **Never fabricate provider data silently.** The synthetic corpus in the *Study Inputs* cell is a stopgap and is documented as one; any replacement must fail loudly on a missing file rather than substituting.
-   **Make every new assertion falsifiable.** Any check added to the verifier must be accompanied by a corruption case in the negative self-test proving it can fail. A predicate satisfied by every state the pipeline can produce must not ship.
-   **Preserve the information barrier.** Every transformation statistic is estimated on training only and frozen. New stages must declare the partitions they read for estimation in `ADAPTER_FITTING_PARTITIONS`.
-   **Report degenerate cells as undefined.** Never write `0.0` into a fairness cell whose denominator vanished.
-   **Document any methodological deviation** at its point of departure, as the five existing registered divergences are.

## Recommended Extensions

-   **Real-data provenance gate:** A check that fails closed when the raw frames are generator-derived, so the synthetic corpus can never silently reach a results table. This is the highest-value single addition to the repository.
-   **Logit-scale interpretability ratio:** The manuscript flags the probability-scale coupling and leaves a logit-scale variant as future work. Implementing both and reporting the region partition under each would separate the algebraic artefact from the empirical finding.
-   **True functional-gradient joint optimisation:** The current scheme is not coordinate descent on the joint loss, because the booster sub-step minimises squared error. Flowing gradients through both branches is named as future work by the paper.
-   **Multi-group and intersectional fairness:** The audit is currently pairwise against a reference group; extending to intersectional cells (for example rural × primary-or-less) is where routing disparities in this population are most likely to compound.
-   **Cost-sensitive threshold optimisation:** F1 at $0.5$ is a reporting convention, not a lending decision. Expected-cost thresholding with asymmetric misclassification costs would make the operational trade-off explicit.
-   **Adverse-action note generation:** For the fully-interpretable region the scorecard coefficients *are* the explanation; emitting regulator-ready reason codes directly from the WoE-scaled contributions would close the loop from audit to compliance artifact.
-   **Region-aware routing constraints:** The audit detects routing disparity but does not act on it. A constrained variant that caps the ML-driven share within any protected subgroup would convert a diagnostic into a control.

# Paper

Title: "*Interpretable hybrid credit scoring for thin-file and underbanked populations: A fairness-aware residual learning framework with evidence from East African financial inclusion data*"

Authors: Belise Kanziga, Yaé U. Gaba, Olivier Kanamugire

E-Journal Submission Date: 27 August 2026

Link: https://arxiv.org/abs/2608.26837

Abstract:

We extend a residual-learning hybrid credit scoring framework (logistic regression scorecard plus a gradient-boosting correction on its residuals, decomposed at each prediction into an interpretability ratio $\rho(x)$ that measures the share attributable to the linear branch) along three axes: an East African empirical instantiation on the Zindi Financial Inclusion in Africa data (Kenya, Rwanda, Tanzania, Uganda); a fairness audit at the granularity of the framework's three interpretability regions; and a thin-file segmentation analysis. On the Taiwan Credit Default benchmark retained for continuity, the calibrated hybrid attains $\text{AUC} = 0.776$ ($\Delta\text{AUC} = +0.057$ vs. standalone logistic regression, $+0.001$ vs. standalone XGBoost), reduces Brier Score by 23\%, and concentrates the highest-default-rate borrowers (69.5\%) in the fully interpretable region. On Zindi, the calibrated hybrid attains $\text{AUC} = 0.869$ ($\Delta\text{AUC} = +0.015$ vs. LR, $p < 0.001$; $-0.004$ vs. XGBoost), cuts Brier from 0.158 to 0.085 (a 46\% reduction), and replicates the regional routing pattern. The fairness audit detects severe routing into the opaque ML-driven region along socioeconomic axes: rural respondents by 18 percentage points relative to urban, primary-or-less-educated by 32 points relative to secondary-and-above, and Ugandan respondents by 22 points relative to Kenyan, while gender shows essentially no routing disparity. The audit pipeline surfaces subgroup-routing violations that aggregate fairness metrics miss, in a form directly usable by African central-bank supervisors of digital credit.

# Paper Summary

# Granular Step-by-Step Summary of the Paper

## 1. Paper identification and overarching aim

The paper is titled **“Interpretable hybrid credit scoring for thin-file and underbanked populations: A fairness-aware residual learning framework with evidence from East African financial inclusion data.”** The authors are Belise Kanziga, Yaé U. Gaba, and Olivier Kanamugire, affiliated with AIMS Kigali, Sefako Makgatho University, and AIRINA Labs. The paper is dated August 27, 2026.

The central objective is to extend an existing residual-learning hybrid credit-scoring framework — originally developed for logistic-regression scorecards plus gradient-boosting residual corrections — along three axes:

1. an **East African empirical application** using the Zindi Financial Inclusion in Africa dataset;
2. a **fairness audit at the granularity of the hybrid model’s interpretability regions**;
3. a **thin-file segmentation analysis**.

A fourth contribution is a **joint optimization formulation** of the two-stage estimation procedure via alternating minimization.

The motivating policy context is the tightening regulation of digital credit in Africa, including the Central Bank of Kenya’s 2022 Digital Credit Providers Regulations, Rwanda’s BNR framework, Nigeria’s FCCPC framework, South Africa’s NCR regime, and Ghana’s BoG sandbox. Regulators require lenders to document the basis of automated credit decisions, but many African borrowers are “thin-file” or “underbanked,” meaning they have little or no formal credit-bureau data. The paper argues that machine-learning models improve accuracy but undermine auditability, while pure logistic scorecards are auditable but underutilize alternative data. The hybrid framework is designed to combine both.

## 2. Intellectual background and related work

The paper positions itself against three literatures:

### 2.1 Credit-scoring methodology
- Logistic regression scorecards have been the regulatory default because of coefficient interpretability and Weight-of-Evidence transformations.
- Machine-learning ensembles such as XGBoost and LightGBM improve predictive performance on tabular data but lack scorecard-style transparency.
- Post-hoc explainability methods like LIME and SHAP exist but have documented instability and adversarial vulnerability.
- Hybrid modelling is the key alternative: retain an interpretable baseline and add a nonlinear correction.
- The closest comparator is Dumitrescu et al. (2022), Penalized Logistic Tree Regression (PLTR). The paper distinguishes itself from PLTR by using a **prediction-level additive decomposition** rather than embedding tree-derived features inside the logistic regression, and by defining a closed-form **per-prediction interpretability ratio** \(\rho(x)\).

### 2.2 Alternative-data scoring in sub-Saharan Africa
- Mobile-money and mobile-phone metadata are increasingly used for credit decisions.
- The contribution here is not simply that alternative data are predictive, but that a hybrid framework can allocate decision influence between an interpretable baseline and an opaque residual learner in an auditable way.

### 2.3 Fairness in algorithmic lending
- The paper draws on equalized odds, disparate-impact ratios, and the four-fifths rule.
- It argues that existing fairness audits typically evaluate aggregate metrics only, potentially missing **subgroup routing** into the opaque ML-driven region of a hybrid model.
- This paper claims to be the first to audit fairness at the level of interpretability regions.

## 3. Core methodological framework

### 3.1 Logistic regression scorecard
- The interpretable baseline is logistic regression:
  \[
  p_{\text{LR}}(x) = \sigma(\beta^\top x + \beta_0)
  \]
- Features are transformed using Weight of Evidence (WoE):
  \[
  \text{WoE}_k = \ln\left(\frac{g_k/G}{b_k/B}\right)
  \]
- The logistic regression uses L2 regularization and balanced class weighting.

### 3.2 Gradient boosting on residuals
- The residual learner is trained on raw response residuals:
  \[
  r_i = y_i - p_{\text{LR}}(x_i)
  \]
- XGBoost is used to estimate \(\hat{r}(x)\), with LightGBM as a robustness check.
- Tree depth is deliberately shallow (3–4) so that the boosting component corrects rather than replaces the logistic baseline.

### 3.3 Hybrid prediction and interpretability ratio
- The final hybrid prediction is:
  \[
  p_{\text{Hybrid}}(x) = \text{clip}\left(p_{\text{LR}}(x) + \alpha^*(x)\hat{r}(x), 0, 1\right)
  \]
- The **interpretability ratio** is:
  \[
  \rho(x) = \frac{|p_{\text{LR}}(x)|}{|p_{\text{LR}}(x)| + |\alpha^*(x)\hat{r}(x)|}
  \]
- Predictions are partitioned into three regions:
  - Fully interpretable: \(\rho(x) \ge 0.9\)
  - Partially interpretable: \(0.7 \le \rho(x) < 0.9\)
  - ML-driven: \(\rho(x) < 0.7\)

The paper explicitly warns that \(\rho(x)\) is measured on the probability scale, which mechanically tends to place high-\(p_{\text{LR}}\) borrowers in the fully interpretable region. Thus the observation that the fully interpretable region contains high-default-rate borrowers is partly an algebraic property, not an independent empirical discovery.

### 3.4 Adaptive ensemble weighting
- The weight \(\alpha^*(x)\) depends on the local uncertainty of the logistic baseline:
  \[
  u_{\text{LR}}(x) = 4p_{\text{LR}}(x)(1 - p_{\text{LR}}(x))
  \]
  \[
  \alpha^*(x) = \alpha_{\min} + (\alpha_{\max} - \alpha_{\min}) u_{\text{LR}}(x)^\gamma
  \]
- Hyperparameters are selected by cross-validation on AUC.

### 3.5 Probability calibration
- Platt scaling is applied post-hoc:
  \[
  p_{\text{cal}}(x) = \frac{1}{1 + \exp(a p_{\text{Hybrid}}(x) + b)}
  \]
- Calibration is measured using Brier score and Expected Calibration Error (ECE).

### 3.6 Prediction-level explanations
- For ML-driven observations, TreeSHAP is applied to the residual learner.
- Stability of SHAP rankings is assessed by bootstrap resampling.
- Unstable explanations are flagged.

## 4. Methodological extensions introduced in this paper

### 4.1 Hard/soft feature-space split
- On Zindi, features are divided into:
  - Hard demographic block: age, gender, marital status, education, household size, location type, country.
  - Soft behavioural block: cellphone access, job type, relationship with head of household.
- The logistic baseline is fit on hard features only; the residual learner sees both.
- This is presented as a methodological note rather than a headline contribution, because the soft block is modest.

### 4.2 Fairness audit per interpretability region
This is the empirical core of the paper. For each protected attribute:
- Gender: male vs. female
- Location: urban vs. rural
- Education: primary-or-less vs. secondary-and-above
- Country: Kenya, Rwanda, Tanzania, Uganda

The audit computes:

1. **Region routing rate**: \(\Pr[\rho(x) < 0.7 \mid A = g]\), i.e., the probability that a member of group \(g\) is routed into the ML-driven region.
2. **Disparate-impact ratio**: \(\Pr[\hat{y}=1 \mid A=g] / \Pr[\hat{y}=1 \mid A=g_{\text{ref}}]\).
3. **Equalized-odds difference**: maximum of absolute TPR and FPR differences against the reference group.

All statistics are reported pooled and per interpretability region, with 95% bootstrap confidence intervals.

The paper notes an important statistical caveat: conditioning on \(\rho(x)\) is conditioning on a post-treatment variable, because \(\rho(x)\) is causally downstream of \(X\). Therefore per-region DI and EOD are descriptive stratifications, not causal fairness metrics.

### 4.3 Thin-file segmentation
- Thin-file is operationally defined as the bottom quartile of a composite index of education, cellphone access, and formal-job status.
- The paper evaluates whether the residual learner contributes more for thin-file borrowers.

### 4.4 Joint optimization via alternating minimization
- The two-stage training is reformulated as a joint loss:
  \[
  \mathcal{L}(\beta, f_{\text{GBM}}) = \sum_i \ell(y_i, p_{\text{LR}}(x_i;\beta) + \alpha^*(x_i)f_{\text{GBM}}(x_i)) + \lambda_{\text{LR}}\|\beta\|_2^2 + \Omega(f_{\text{GBM}})
  \]
- The implemented algorithm is an alternating fixed-point iteration, not exact joint coordinate descent:
  1. Hold the residual learner fixed and update \(\beta\) by L-BFGS.
  2. Hold \(\beta\) fixed and refit the residual learner on updated residuals.
  3. Iterate until validation AUC change is below tolerance.

The authors explicitly acknowledge this is not true functional-gradient descent on the joint loss.

## 5. Datasets

### 5.1 Zindi Financial Inclusion in Africa (FSD)
- 23,524 household-survey respondents from Kenya, Rwanda, Tanzania, and Uganda.
- FinScope surveys conducted 2016–2018.
- Binary label: ownership of a bank account; positive rate 14.08%.
- Zero missing values.
- Features: country, location type, cellphone access, gender, relationship with head, marital status, education level, job type, household size, age.

### 5.2 Target reframing
- The label is not default, but account ownership.
- The paper defends this as a deliberate proxy for the **eligibility gate** that digital lenders face when deciding which thin-file borrowers to underwrite.
- The accuracy–interpretability–fairness structure is argued to be identical to default prediction.
- Zindi offers protected-attribute coverage that no public African default dataset currently provides.
- The paper acknowledges limitations: magnitudes of fairness effects may not transfer to default prediction; account ownership and default sit on different sides of the selection funnel.

### 5.3 Taiwan Credit Default dataset
- 30,000 credit-card observations with 23 features and a default label.
- Retained for continuity with the prior thesis and to benchmark the framework on a default-prediction setting.

### 5.4 Kiva African subset
- Originally planned as a robustness check with actual loan-adverse outcomes.
- Excluded from the present paper due to timeline constraints.
- The authors state the audit pipeline is dataset-agnostic and could be applied to Kiva in future work.

## 6. Experimental setup

### 6.1 Preprocessing
- Missing values are imputed using training-partition medians and modes.
- WoE-encoded categorical variables are used for logistic regression; one-hot encoding is used for the boosting learner.
- Continuous variables are winsorized at the 1st and 99th percentiles.
- Feature selection uses Information Value threshold IV ≥ 0.02.
- On Zindi, all ten features pass the filter; the strongest are education level (IV = 0.97), cellphone access (0.71), and job type (0.63).
- Year and uniqueid are dropped.
- A 60/20/20 stratified split is used.

### 6.2 Model configurations
- Logistic regression: L2, balanced class weights, grid over C.
- Residual XGBoost: squared-error objective, depth 3–4, learning rate 0.03–0.05, subsample 0.8, column subsample 0.8.
- Standalone XGBoost: depth 4, 400 trees, learning rate 0.03.
- Hybrid grid search over \(\alpha_{\min}\), \(\alpha_{\max}\), and \(\gamma\).

### 6.3 Evaluation metrics
- AUC, Gini, F1, Brier score, ECE.
- Bootstrap confidence intervals and DeLong tests for AUC differences.

## 7. Results

### 7.1 Predictive performance on Taiwan
- Logistic regression: AUC = 0.719, Brier = 0.207.
- Standalone XGBoost: AUC = 0.775, Brier = 0.177.
- Hybrid calibrated: AUC = 0.776, Brier = 0.136.
- The hybrid matches XGBoost on discrimination (\(\Delta\)AUC = +0.001, 95% CI \([-0.004, +0.006]\), \(p = 0.37\)).
- The hybrid beats logistic regression by +0.057 AUC.
- Brier score is reduced by roughly 23% relative to XGBoost.
- ECE falls to 0.011 from 0.234 for LR.

### 7.2 Predictive performance on Zindi
- Logistic regression: AUC = 0.854, Brier = 0.158.
- LR + Platt calibration: AUC = 0.854, Brier = 0.088, ECE = 0.016.
- Standalone XGBoost: AUC = 0.873, Brier = 0.143.
- Hybrid uncalibrated: AUC = 0.869, Brier = 0.092, F1 = 0.529.
- Hybrid Platt-calibrated: AUC = 0.869, Brier = 0.085, ECE = 0.024.
- The hybrid improves on LR by +0.015 AUC (95% CI \([+0.011, +0.020]\), \(p < 10^{-9}\)).
- The hybrid is statistically indistinguishable from XGBoost on AUC (\(\Delta = -0.004\), 95% CI \([-0.009, +0.001]\), \(p = 0.92\)).
- The Brier reduction from LR to calibrated hybrid is about 46%.
- Importantly, the ablation row shows that most of the calibration gain comes from Platt scaling the LR baseline, while the residual learner contributes the discrimination gain, not the calibration gain.

### 7.3 Interpretability regions

#### Taiwan
- Fully interpretable: 6.3% of test observations, default rate 69.5%.
- Partially interpretable: 28.6%, default rate 33.4%.
- ML-driven: 65.1%, default rate 12.5%.
- Per-region hybrid AUC exceeds per-region LR AUC, especially in the partially and ML-driven regions.

#### Zindi
- Fully interpretable: 6.4% of test observations, positive rate 53.8%.
- Partially interpretable: 34.1%, positive rate 19.7%.
- ML-driven: 59.4%, positive rate 6.5%.
- Per-region hybrid AUC:
  - Fully: 0.888 vs. LR 0.869.
  - Partially: 0.830 vs. LR 0.821.
  - ML-driven: 0.752 vs. LR 0.730.

The qualitative pattern is stable across datasets: the interpretable region is small but contains the highest-risk / highest-eligibility borrowers; the ML-driven region contains the majority of observations but at low base rates.

### 7.4 Fairness audit findings

The fairness audit is the central empirical contribution.

#### Routing rates
- Gender:
  - Male: 59.5% routed to ML-driven region.
  - Female: 59.4%.
  - Routing disparity: essentially zero.

- Location:
  - Urban: 48.2%.
  - Rural: 66.4%.
  - Gap: 18 percentage points.

- Education:
  - Secondary and above: 35.9%.
  - Primary or less: 68.3%.
  - Gap: 32 percentage points.

- Country:
  - Kenya: 48.4%.
  - Rwanda: 60.9%.
  - Tanzania: 63.8%.
  - Uganda: 70.2%.
  - Largest gap vs. Kenya: 22 percentage points.

The paper interprets this as a systematic routing disparity along socioeconomic axes, not along gender.

#### Disparate-impact ratios (pooled)
- Female vs. male: 0.429.
- Rural vs. urban: 0.309.
- Primary-or-less vs. secondary-and-above: 0.025.
- Rwanda vs. Kenya: 0.190.
- Tanzania vs. Kenya: 0.291.
- Uganda vs. Kenya: 0.216.

All except gender violate the four-fifths rule. The education ratio of 0.025 means the model predicts formal-banking-system membership for primary-or-less-educated respondents at 2.5% the rate of secondary-and-above respondents. The paper stresses this largely reflects structural inequality in the population, not model-created bias.

#### Equalized-odds differences (pooled)
- Gender: 0.104.
- Location: 0.238.
- Education: 0.537.
- Country: 0.266.

Per-region values are reported but should be read descriptively. The ML-driven region has undefined DI and EOD because the reference group has no predicted positives in that region.

#### Mechanism
The TreeSHAP analysis on the ML-driven region shows that the residual learner’s most important features are:
1. cellphone access = no;
2. age of respondent;
3. relationship with head = head of household;
4. country = Kenya;
5. country = Rwanda;
6. education = no formal education;
7. cellphone access = yes;
8. education = secondary;
9. job = farming and fishing;
10. gender = female.

The top two feature ranks are perfectly stable across 500 bootstrap resamples, and the top-five set is preserved in 74% of resamples. The residual learner is not adding truly alternative signal; it is nonlinearly reweighting the same infrastructure and demographic features already in the LR baseline. This explains why disadvantaged subgroups are both more uncertain under the LR baseline and more heavily corrected by the residual learner.

### 7.5 Thin-file segmentation results

- Thin-file subgroup: \(n = 1,559\), 33.1% of the Zindi test set; positive rate 2.4%.
- LR AUC: 0.782.
- XGBoost AUC: 0.856.
- Hybrid AUC: 0.840.
- Hybrid vs. LR \(\Delta\)AUC: +0.058, four times the full-population gain of +0.015.
- Calibration on thin-file: Brier = 0.023, ECE = 0.028.

However, the hypothesised shift in \(\rho(x)\) toward the ML-driven region does not occur. Mean \(\rho\) on thin-file is 0.718 vs. 0.703 on the full test set, a small shift in the opposite direction. The paper explains this by noting that thin-file borrowers are confidently predicted negative by the LR baseline, so the adaptive weight \(\alpha^*(x)\) is small and the residual learner has little influence. The hypothesis that “thin-file” equals “alternative-data-dominated” is rejected for this dataset.

### 7.6 Alternating minimization

- The alternating fixed-point scheme converges on Zindi in 3 iterations.
- Final \(\Delta\)AUC vs. two-stage training: +0.0006.
- The improvement is smaller than the convergence tolerance, so the paper does not claim two-stage training is suboptimal.
- The joint scheme costs about 4× the wall-clock time of two-stage training and is therefore not recommended as the deployment default.

### 7.7 Statistical inference

- Taiwan hybrid vs. LR: \(\Delta\)AUC = +0.057, 95% CI \([+0.048, +0.066]\), \(p < 0.001\).
- Taiwan hybrid vs. XGBoost: \(\Delta\)AUC = +0.001, 95% CI \([-0.004, +0.006]\), \(p = 0.37\).
- Zindi hybrid vs. LR: \(\Delta\)AUC = +0.015, 95% CI \([+0.011, +0.020]\), \(p < 10^{-9}\).
- Zindi hybrid vs. XGBoost: \(\Delta\)AUC = \(-0.004\), 95% CI \([-0.009, +0.001]\), \(p = 0.92\).

The paper reports DeLong one-sided \(p\)-values with the alternative hypothesis that hybrid AUC exceeds comparator AUC.

## 8. Discussion and implications

### 8.1 What the audit reveals
- The framework’s routing pattern is stable across Taiwan and Zindi despite different label semantics and class balances.
- The fully interpretable region is small but contains the highest-base-rate borrowers.
- The fairness audit shows that aggregate metrics can appear balanced while protected subgroups are disproportionately routed into the opaque ML-driven region.
- The SHAP analysis closes the loop: the same features that drive residual-learner corrections are the features associated with rural, low-education, and non-Kenyan status.

### 8.2 Regulatory implications
- The framework operationalizes per-prediction explanations required by CBK, BNR, FCCPC, and NCR.
- Supervisors are advised to require per-region routing-rate disaggregation, not just aggregate disparate-impact statistics.
- The paper acknowledges that audit pipelines detect disparity but do not automatically disentangle model-introduced disparity from structural population disparity.

### 8.3 Practical implications

**For lenders:**
- Publishing aggregate DI alone is a weaker fairness claim than the framework can support.
- The governance cost, not compute cost, is the main constraint.

**For supervisors:**
- A phased path is suggested: self-attestation, sample-based review, then routine examination.
- Supervisory capacity varies widely across African jurisdictions.

**For borrowers:**
- The framework makes routing asymmetry visible but does not directly fix it.
- Rural, low-education, and smaller-market borrowers are less likely to receive fully interpretable decisions.

### 8.4 Relation to PLTR
- PLTR is feature-level: tree-derived effects enter the logistic specification.
- This framework is prediction-level: the residual learner is a separate additive term.
- The prediction-level construction enables the closed-form \(\rho(x)\) decomposition.
- The paper does not claim strict predictive dominance over PLTR.

## 9. Limitations

1. **Zindi target reframing**: account ownership is not default; fairness-effect magnitudes may not transfer to default prediction.
2. **Single African dataset**: only East Africa is covered; West and Southern Africa may differ.
3. **Joint optimization**: compute cost is 4× higher, with no detectable AUC gain.
4. **Per-region fairness metrics**: conditioning on \(\rho(x)\) is conditioning on a post-treatment variable; per-region values are descriptive, not causal.
5. **Structural vs. model-created disparity**: the audit detects disparities but does not identify their origin without additional causal analysis.

## 10. Future work

The most consequential future direction is a **West African extension**:
- Regulatory environments differ: BCEAO in UEMOA countries, Nigeria’s FCCPC, Ghana’s BoG sandbox.
- Candidate data: Kiva African subset with actual loan-adverse outcomes, or partnerships with Nigerian/Ghanaian digital lenders.
- Protected-attribute coverage on Kiva is weaker than on Zindi.
- Pre-registered hypotheses:
  1. Socioeconomic routing disparities reproduce.
  2. Gender-routing disparity emerges where the underlying gender account-ownership gap is large.
  3. Per-region disparate-impact patterns hold.

Additional future directions include:
- Widening the soft feature block with mobile-money transaction trails.
- Studying the interaction between the fairness audit and cost-sensitive classification thresholds.

## 11. Conclusion

The paper demonstrates that a residual-learning hybrid scorecard can match standalone XGBoost on discrimination while retaining a per-prediction interpretability decomposition. On Zindi, the calibrated hybrid achieves AUC = 0.869, matching XGBoost, improving on logistic regression, and reducing the Brier score by 46%. The fairness audit at the granularity of interpretability regions is operationally feasible and surfaces routing disparities — rural, low-education, and Ugandan respondents are disproportionately routed to the opaque ML-driven region — that aggregate fairness metrics would miss. The framework is positioned as directly relevant to African digital-credit supervision, with a clear roadmap for West African replication and extension.

from the manuscript's data contract, so it is the first place that is hash-sealed.

### Task 3 — `orchestrate_create_partitions(config, zindi_cleansed, taiwan_cleansed, unseen_levels, output_dir) → Dict[str, Any]`

*Workbook anchor: cell 23, L7356.*

**Inputs.** The two cleansed frames; the `unseen_levels` registry carried forward; `config`; `output_dir`.

**Process.** *Step 1 — the stratified splitter.* `implement_stratified_splitter` chains two

→ 60% training block + 40% holdout; then `test_size=0.5` on the holdout (same seed) → 20% validation + 20%

not the algorithm) and is recorded in the manifest. *Step 2 — Zindi integrity.* `split_zindi_and_verify`

**Role.** Implements the protocol guarantee of Section 5.3: *"To prevent information leakage, all

### Task 4 — `orchestrate_fit_preprocessing_stats(config, zindi_frame, zindi_manifest, taiwan_frame, taiwan_manifest, output_dir) → Dict[str, Any]`

*Workbook anchor: cell 24, L7738.*

**Inputs.** The cleansed frames with their frozen partition manifests (the manifests supply the training-row
index sets; only those rows may contribute statistics); `config`; `output_dir`.

**Process.** *Step 1 — feature-set derivation.* `derive_taiwan_feature_sets` establishes the Taiwan
continuous/categorical block split the placeholder protocol requires (the thesis pipeline is not restated in
the paper; the split is declared and flagged UNSPECIFIED IN MANUSCRIPT). *Step 2 — imputation statistics.*
`compute_imputation_stats` computes the median of the numerical features and the mode of every categorical
feature via `Series.mode().iloc[0]` (first sorted mode, for determinism), on the training partition only;
the statistics are stored even though Zindi has zero missing values, for interface uniformity and for
Taiwan. *Step 3 — winsorization statistics.* `compute_winsorization_stats` computes the 1st and 99th
percentiles of the numerical features via `numpy.percentile` on the training partition only, records the
exact percentile values in the audit log, and applies the clipping bounds unrounded — a percentile of a
discrete integer column may be an unobserved boundary value, and the raw value, not a rounded one, is
frozen. *Step 4 — frozen application.* `apply_frozen_stats` (via `_fit_dataset_stats`) fills missing values
with the frozen medians/modes and clips the numerical features at the frozen bounds, identically on
training, validation and test; every Taiwan placeholder decision is logged and flagged; the frozen
`preprocessing_stats` dictionary is persisted to JSON.

**Transformation.** Clean partitions → partitions whose missing cells and tails are governed exclusively by
training-learned constants, plus the immutable `preprocessing_stats` block (medians, modes, `winsor_bounds`)
that every later transformation consumes. Validation and test rows are transformed by statistics they never
contributed to.

**Role.** Implements Section 5.1 verbatim: *"Missing values are imputed by median (numerical) and mode
(categorical) on the training partition only, and the imputation values are then applied to validation and
test"*; *"Numerical features are winsorized at the 1st and 99th percentiles of the training partition"*. The
training-only freeze is the leakage-prevention contract in its purest form — *"the discipline of computing
and freezing every transformation statistic on the training partition only, to guarantee that no information
from validation or test partitions enters model estimation"* — and the exact-value audit log is what makes
later drift diagnosable rather than mysterious.

### Task 5 — `orchestrate_fit_woe_encoding(config, zindi_train_df, zindi_y, zindi_frames, zindi_logistic_features, taiwan_train_df, taiwan_y, taiwan_frames, taiwan_logistic_features, taiwan_continuous, taiwan_categorical, output_dir) → Dict[str, Any]`

*Workbook anchor: cell 25, L8654.*

**Inputs.** Per dataset: the training partition frame, its label vector, the full partition-frame
dictionary, the ordered logistic-branch feature tuple; for Taiwan additionally the explicit continuous and
categorical block tuples; `config`; `output_dir`.

**Process.** *Step 1 — monotonic binning of the continuous features.* `fit_monotonic_bins` starts from 20
quantile bins on the training partition (`_qcut_edges`, `_bins_from_edges`), computes each bin's WoE via
`woe_from_counts` under Equation (2) with the pinned class convention ($g$ counts $y=0$, $b$ counts $y=1$,
$G, B$ the training totals), and enforces monotonicity with `_enforce_monotonicity`: whenever the WoE
sequence changes direction, the adjacent pair with the closest WoE values is merged (`_merge_bins`, test
`_is_monotone`), repeated until monotone; `_enforce_min_frequency` then enforces a minimum bin frequency of
5% of training observations; `_interval_label` and `assert_interval_key_uniqueness` keep bin keys stable and
unique. *Step 2 — WoE scores.* `compute_woe_scores` recomputes $g_k, b_k, G, B$ under the frozen boundaries
on training and forms the final WoE for every bin and every declared categorical level via Equation (2); a
zero count in either class takes the documented neutral WoE = 0 treatment; `_assert_woe_key_coverage` checks
key coverage before freezing. *Step 3 — frozen application.* `apply_woe_mapping` bins each validation/test
observation by the frozen cut points (continuous) or raw level (categorical) and replaces it with the
training-learned WoE; any level or bin unseen in training maps to neutral 0.0 and is logged; identical
column ordering across the three matrices is asserted before persistence.

**Transformation.** Raw numeric and categorical columns → monotone-risk WoE scores. Each bin or level $k$ is
replaced by the training-frozen value

$$\mathrm{WoE}_k = \ln\!\left(\frac{g_k/G}{b_k/B}\right), \tag{2}$$

so the logistic branch receives a matrix whose every column is a monotone log-odds contrast against the
training population.

**Role.** Implements the scorecard encoding of Section 3.1 — *"WoE-encoded features and L2 regularization on
β together preserve the auditable per-feature decomposition that regulators require of a scorecard"* — under
Section 5.1's leakage rules: *"Rare WoE values encountered in held-out folds that do not appear in training
are assigned a neutral WoE = 0, preserving the log-odds-zero contribution and preventing information
leakage."* The monotonicity enforcement is the industry convention (Siddiqi 2012; Thomas 2017) that makes
the coefficient table a risk ladder rather than an arbitrary contrast set.

### Task 6 — `orchestrate_iv_filter_and_ohe(config, train_df, y, woe_map, cut_points, woe_matrices, partition_frames, categorical_features, continuous_features, output_dir, verify_manuscript_targets) → Dict[str, Any]`

*Workbook anchor: cell 26, L9173.*

**Inputs.** The training partition frame and labels; the frozen WoE artefacts from Task 5; the WoE matrices
of all three partitions; the raw partition frames; the categorical and continuous feature tuples;
`output_dir`; the manuscript-target verification flag. (Invoked once per dataset — Zindi and Taiwan.)

**Process.** *Step 1 — Information Value and the retention filter.* `compute_information_values` computes,
per feature on the training partition,

$$\mathrm{IV} = \sum_k \left(\frac{g_k}{G} - \frac{b_k}{B}\right)\mathrm{WoE}_k,$$

from the frozen WoE values exactly; `apply_iv_filter` retains features with $\mathrm{IV} \ge 0.02$ and
filters the WoE matrices to the retained columns; `assert_zindi_iv_targets` (when
`verify_manuscript_targets`) asserts that all ten Zindi features pass and that the three strongest IVs match
the manuscript — `education_level` ≈ 0.97, `cellphone_access` ≈ 0.71, `job_type` ≈ 0.63 — within absolute
tolerance 0.02. *Step 2 — one-hot fitting.* `fit_one_hot_encoding` expands every categorical feature into
indicator columns via `pandas.get_dummies(drop_first=False)` on training, so every training-observed level
receives a column; the sorted unique levels per feature are frozen; continuous features are appended
unchanged (rare levels are accepted, not collapsed — the manuscript specifies no rare-level handling). *Step
3 — frozen application.* `apply_one_hot_encoding` applies `get_dummies` to validation and test, reindexes
against the frozen training column list with zero fill (a value absent from the frozen vocabulary yields an
all-zero block), and asserts `X_ohe_val.columns.equals(X_ohe_train.columns)` and likewise for test. The IV
table, retained feature set and frozen column list are persisted.

**Transformation.** Raw categorical levels → a dense indicator design aligned to a frozen training
vocabulary: $X_{\mathrm{ohe}}^{\mathrm{train/val/test}}$ with identical column ordering, while the logistic
branch continues on the WoE matrix filtered to the retained set — the two-representation split the
manuscript prescribes.

**Role.** Implements Section 5.1's feature economy verbatim: *"For Zindi specifically, all ten input
features (country, location_type, cellphone_access, gender_of_respondent, relationship_with_head,
marital_status, education_level, job_type, household_size, age_of_respondent) survive the IV ≥ 0.02 filter;
the strongest predictor by IV is education_level (IV = 0.97), followed by cellphone_access (0.71) and
job_type (0.63)"* — the three-value assertion is this orchestrator's precision check. It also implements the
independent representation rule: *"for the residual learner, categorical features are one-hot encoded, which
permits the boosting model to capture nonlinear interactions among categorical levels."* The IV filter binds
the logistic branch's feature economy; the boosting branch keeps the full set.

### Task 7 — `orchestrate_fit_logistic(config, X_woe_train, y_train, X_woe_val, y_val, X_woe_test, y_test, output_dir, feature_names) → Dict[str, Any]`

*Workbook anchor: cell 27, L9600.*

**Inputs.** The WoE design matrices of all three partitions with aligned label vectors; the ordered
`feature_names` of the WoE matrix (the coefficient vector must later attach to feature identities for
adverse-action explanations); `config`; `output_dir`.

**Process.** *Step 1 — C selection.* `select_C_cv` runs `StratifiedKFold(n_splits=5, shuffle=True, random_state=42)`;
for each $C \in \{0.01, 0.1, 1, 10\}$ it fits `LogisticRegression(solver='lbfgs', penalty='l2', class_weight='balanced', C=C, max_iter=1000)`
on the fold's training rows, scores AUC on the fold's validation rows, and selects the argmax mean; `lbfgs`
non-convergence at extreme $C$ is absorbed by `max_iter` and logged. *Step 2 — the final fit.*
`fit_final_logistic` refits the same estimator on the full training partition at the selected $C$ and
extracts $\hat{\beta}$ and $\hat{\beta}_0$, persisted for adverse-action explanations. *Step 3 — probability
generation.* `predict_lr_partitions` applies `predict_proba` to all three matrices, forming

$$p_{\mathrm{LR}}(x) = \sigma\!\left(\hat{\beta}^{\top} x + \hat{\beta}_0\right) = \frac{1}{1 + \exp\!\left(-(\hat{\beta}^{\top} x + \hat{\beta}_0)\right)}, \tag{1}$$

records the validation AUC for comparison against the manuscript's test value ($\approx 0.854$), and asserts
every probability lies in the open interval $(0, 1)$.

**Transformation.** WoE matrix → probability vector through the sigmoid link on the linear predictor. The
same pass yields the explanation object $(\hat{\beta}, \hat{\beta}_0, \text{feature names})$, so the
interpretable branch produces its score and its coefficient table simultaneously.

**Role.** Implements the baseline of Section 3.1 — *"The logistic regression baseline estimates
$p_{\mathrm{LR}}(x_i) = \sigma(\beta^\top x_i + \beta_0)$ … with $\beta \in \mathbb{R}^p$ fitted by maximum
likelihood under balanced class weighting"* — under the Table 1 configuration (`lbfgs`, L2, balanced, $C$
grid $\{0.01, 0.1, 1, 10\}$). The balanced weighting is kept exactly as specified; its probability-scale
consequence (the raw LR's $\mathrm{ECE} = 0.234$ on the imbalanced Zindi target) is deliberately left
uncorrected here — calibration belongs to Task 13, and the paper's ablation design requires the uncalibrated
baseline to be measured as it is.

### Task 8 — `orchestrate_compute_residuals(labels, p_lr) → Dict[str, Any]`

*Workbook anchor: cell 28, L9907.*

**Inputs.** Two dictionaries keyed by partition (train/val/test): the binary label vectors and the Task 7
probability vectors. No configuration enters; the step is arithmetic plus validation.

**Process.** *Step 1 — training residuals.* `compute_training_residuals` forms, per training observation,

$$r_i = y_i - p_{\mathrm{LR}}(x_i), \tag{3}$$

the regression target of the residual learner. *Step 2 — validation and test residuals.*
`compute_partition_residuals` applies the identical formula to validation and test; these are read-only
artefacts, retained for evaluation and for the alternating-minimization procedure, never used to fit the
booster. *Step 3 — distribution validation.* `validate_residual_distribution` asserts $r_i \in [-1, 1]$ (a
probability minus a binary label is bounded), computes mean, standard deviation, minimum, maximum, and halts
if the training standard deviation falls below $10^{-3}$ — a near-constant LR would make residual learning
meaningless — logging the summary into the run manifest.

**Transformation.** Label and probability vectors → signed, bounded residual vectors
$r^{\mathrm{train/val/test}}$, positionally aligned to the partitions; the training vector becomes the
boosting branch's regression target, the other two are evaluation-only artefacts.

**Role.** Implements Equation (3) and Section 3.2's residual-definition argument verbatim: *"Pearson or
deviance residuals are more standard for GLM diagnostics, but the raw response residual is adopted here
because it provides a direct regression target for the secondary learner and aligns with the
residual-learning paradigm popularized for deep networks by He et al."* The orchestrator enforces the
algebraic implication of that choice — the residuals are heteroscedastic by construction, hence the
squared-error objective in Task 9 — and the variance floor protects the residual-learning premise from a
degenerate baseline.

<br>

### Task 9 — `orchestrate_fit_residual_xgboost(config, X_ohe_matrices, residuals, p_lr, labels, lr_only_val_auc, output_dir) → Dict[str, Any]`

*Workbook anchor: cell 29, L10313.*

**Inputs.** The one-hot matrices of all three partitions; the three residual vectors; the LR probability and
label vectors (the provisional-hybrid scoring used by early stopping); the LR-only validation AUC (the
contribution-check baseline); `config`; `output_dir`.

**Process.** *Step 1 — configure and fit.* `fit_residual_booster` builds `XGBRegressor(objective='reg:squarederror', n_estimators=400, max_depth=3, learning_rate=0.03, subsample=0.8, colsample_bytree=0.8, reg_lambda=2)`
and trains it on $X_{\mathrm{ohe}}^{\mathrm{train}}$ against $r^{\mathrm{train}}$ under the custom
`_HybridAucEarlyStopping` monitor: if the config declares early stopping, training halts after `patience=50`
rounds without improvement in the validation AUC of the provisional hybrid $p_{\mathrm{LR}}(x) +
\hat{r}(x)$; otherwise the fixed 400 estimators are used. The paper's ranges (300–500 trees, depth 3–4, rate
0.03–0.05) resolve to the config placeholders, flagged UNSPECIFIED IN MANUSCRIPT. *Step 2 — predictions at
the frozen iteration.* `predict_residuals` predicts on all three matrices at the frozen best iteration,
records the actual tree count, and asserts no NaN in any vector. *Step 3 — contribution validation.*
`validate_residual_contribution` computes the Pearson correlation between $r$ and $\hat{r}$ on training
(`scipy.stats.pearsonr`), asserts it positive, and records the provisional hybrid validation AUC against
`lr_only_val_auc`.

**Transformation.** One-hot features → residual-score vector

$$\hat{r}(x) = f_{\mathrm{GBM}}(x) = \sum_{m=1}^{M} \nu\, h_m(x), \tag{4}$$

the nonlinear correction term of the hybrid — with $h_m$ the regression tree fitted at iteration $m$ and
$\nu$ the learning rate.

**Role.** Implements Section 3.2 under the Section 5.2 configuration, whose design rationale is
load-bearing: *"The shallow tree depths (3–4) for the residual learner are deliberate: deeper trees allow
the boosting component to dominate the prediction and erode the interpretability ratio. Low learning rates
(0.03–0.05) encourage gradual residual correction rather than aggressive nonlinear replacement of the
logistic baseline."* The hybrid-AUC early-stopping target operationalises that sentence: the model is
selected on the quantity the paper cares about — the combined prediction — not on the residual MSE alone.

### Task 10 — `orchestrate_fit_standalone_xgboost(config, X_ohe_matrices, labels, output_dir, verify_manuscript_targets) → Dict[str, Any]`

*Workbook anchor: cell 30, L10663.*

**Inputs.** The one-hot matrices of all three partitions with aligned binary labels; the config's standalone
hyperparameters; the verification flag; `output_dir`.

**Process.** *Step 1 — configuration.* `configure_standalone` builds `XGBClassifier(n_estimators=400, max_depth=4, learning_rate=0.03, subsample=0.8, colsample_bytree=0.8, reg_lambda=2, objective='binary:logistic', eval_metric='auc')`
and computes $\text{scale\_pos\_weight} = n_{\text{neg}}/n_{\text{pos}}$ on the training partition only,
recording the exact value. *Step 2 — fit and predict.* `fit_and_predict` fits on
$X_{\mathrm{ohe}}^{\mathrm{train}}$ with the aligned labels and produces probability scores for all three

**Role.** Implements the Table 2 comparator row (Zindi XGBoost AUC 0.873, Brier 0.143, F1 0.513) and Section

role is to separate the two claims the paper makes about the hybrid: matching the forest on rank-ordering

### Task 11 — `orchestrate_select_adaptive_weights(config, X_woe_train, y_train, X_ohe_train, selected_C, p_lr, output_dir) → Dict[str, Any]`

*Workbook anchor: cell 31, L11118.*

**Inputs.** The WoE and one-hot training matrices with training labels; the Task 7 selected $C$; the
final-model LR probability vectors of all partitions; `config`; `output_dir`.

**Process.** *Step 1 — the uncertainty proxy and the weight schedule.* `compute_uncertainty` and
`compute_adaptive_weight` implement

$$u_{\mathrm{LR}}(x) = 4 \cdot p_{\mathrm{LR}}(x)\,(1 - p_{\mathrm{LR}}(x)) \in [0, 1], \tag{8}$$

$$\alpha^{*}(x) = \alpha_{\min} + (\alpha_{\max} - \alpha_{\min})\, u_{\mathrm{LR}}(x)^{\gamma}, \tag{9}$$

with $\gamma > 0$ controlling concavity and the output verified bounded. *Step 2 — grid search.*
`grid_search_adaptive_weights` enumerates the 27 triples ($\alpha_{\min} \in \{0.0, 0.1, 0.2\}$,
$\alpha_{\max} \in \{0.3, 0.5, 0.8\}$, $\gamma \in \{0.5, 1.0, 2.0\}$) and runs 5-fold stratified CV per
triple: per fold it fits the LR, computes fold-training residuals, fits the residual booster on them,
computes fold-validation $u_{\mathrm{LR}}$ and $\alpha^{*}$, forms the Equation (5) hybrid and scores
validation AUC; the argmax triple is selected. *Step 3 — freezing.* `freeze_adaptive_weights` recomputes
$u_{\mathrm{LR}}$ and $\alpha^{*}$ for every observation of all three partitions from the final
full-training LR probabilities — never from fold models — asserts non-negativity and boundedness, and
persists the triple and the per-observation arrays with the full CV trace.

**Transformation.** Final LR probabilities → per-observation local-uncertainty weights that scale the
residual correction before it enters the hybrid sum; a three-dimensional grid choice collapses to one
selected triple with a complete selection record.

**Role.** Implements Section 3.4 — *"The weight $\alpha^{*}(x)$ in Eq. (5) is allowed to vary with the local
uncertainty of the logistic baseline"* — under the Section 5.2 grids. This is the mechanism that lets the
residual branch contribute most where the baseline is most uncertain, and the frozen $\alpha^{*}$ is what
makes the interpretability ratio of Equation (6) a per-prediction quantity rather than a global constant.

> **One registered deviation, disclosed rather than buried.** Section 5.2 states that "static and adaptive
> weighting hyperparameters are selected via 5-fold and 3-fold stratified cross-validation respectively", in
> tension with Section 3.4's 5-fold statement. The implementation standardises on 5-fold for the adaptive grid
> and logs the resolution; the choice is the more conservative of the two readings (more folds, smaller
> validation slices, identical seed discipline).

<br>

### Task 12 — `orchestrate_build_hybrid_and_rho(config, p_lr, r_hat, alpha_star, labels, output_dir, verify_manuscript_targets) → Dict[str, Any]`

*Workbook anchor: cell 32, L11499.*

**Inputs.** The three partition dictionaries of LR probabilities, residual predictions and frozen adaptive
weights; the label vectors; `config`; `output_dir`; the verification flag.

**Process.** *Step 1 — hybrid probabilities.* `compute_hybrid_probabilities` forms

$$p_{\mathrm{Hybrid}}(x) = \mathrm{clip}\!\left(p_{\mathrm{LR}}(x) + \alpha^{*}(x)\,\hat{r}(x),\, 0,\, 1\right), \tag{5}$$

via `numpy.clip` with NaN and range assertions, storing the *unclipped* additive components for Step 2.
*Step 2 — ratio and regions.* `compute_rho_and_regions` computes, from the unclipped components,

$$\rho(x) = \frac{|p_{\mathrm{LR}}(x)|}{|p_{\mathrm{LR}}(x)| + |\alpha^{*}(x)\,\hat{r}(x)|} \in [0, 1], \tag{6}$$

with the zero-denominator guard $\rho := 1$ when the denominator falls below $10^{-12}$ (theoretically
impossible, numerically reachable), and assigns the three regions,

$$\mathrm{Region}(x) = \begin{cases} \text{Fully interpretable} & \rho(x) \ge 0.9,\\ \text{Partially interpretable} & 0.7 \le \rho(x) < 0.9,\\ \text{ML-driven} & \rho(x) < 0.7; \end{cases} \tag{7}$$

regions are computed on the uncalibrated scale — Platt calibration (Task 13) changes only the probability
scale, never $\rho$ or membership. *Step 3 — test aggregation.* `aggregate_region_stats` computes per-region
shares, within-region positive-class rates, mean $\rho$ per region and the full $\rho$ distribution,
verifying Table 3's Zindi values (shares 6.4% / 34.1% / 59.4%; positive rates 53.8% / 19.7% / 6.5%; each
within $\pm 1$ percentage point) and asserting the bimodality of the $\rho$ distribution (primary mode below
0.7, secondary mode near 1.0) when the flag is set.

**Transformation.** Three per-observation vectors → one clipped hybrid score, one auditability scalar
$\rho(x) \in [0,1]$, and one three-level region label, per observation. The unclipped components are
retained because $\rho$ is a function of the pre-clip additive parts; clipping would distort the ratio at
the boundary.

**Role.** Implements the paper's second methodological pillar — Equations (5), (6), (7) — together with
Section 3.3's central interpretive caution, stated nearly verbatim there: *"the ratio is defined on the
probability scale rather than the logit scale, so borrowers for whom $p_{\mathrm{LR}}(x)$ is large (close to
one) tend to fall in the fully-interpretable region when the residual correction $\alpha^{*}(x)\hat{r}(x)$
is bounded, which mechanically couples the fully-interpretable region to the high-$p_{\mathrm{LR}}$ tail of
the score distribution"* — so the Table 3 concentration of high-base-rate borrowers in the fully
interpretable region (53.8% Zindi, 69.5% Taiwan) "is therefore a joint property of Eq. (6) and the
well-calibrated behaviour of the logistic-regression baseline on that tail", not an independent empirical
discovery.

<br>


### Task 13 — `orchestrate_fit_platt(p_hybrid, p_lr, labels, output_dir, verify_manuscript_targets) → Dict[str, Any]`

*Workbook anchor: cell 33, L12134.*

**Inputs.** The hybrid and LR probability vectors of all three partitions (validation for fitting, test for
application); the aligned label vectors; `output_dir`; the verification flag.

**Process.** *Step 1 — validation fit.* `fit_platt` (through `platt_equation_10`) fits a univariate logistic
regression of the binary labels on the validation hybrid probabilities under

$$p_{\mathrm{cal}}(x) = \frac{1}{1 + \exp\!\left(a \cdot p_{\mathrm{Hybrid}}(x) + b\right)}, \tag{10}$$

in an eight-substep internal procedure: input validation; unpenalised maximum-likelihood fitting (a
heavily-regularised fallback is documented for small validation sets); recovery of the library-convention
coefficients; derivation of the manuscript-convention $(a, b)$ by sign negation where the library convention
differs; a numerical round-trip assertion of the two conventions; strict-monotonicity verification of the
fitted map; logging; return. *Step 2 — frozen test application.* `apply_platt` evaluates the frozen logistic
function on the test hybrid probabilities; region assignments remain based on the *uncalibrated*
probabilities — calibration changes only the probability scale, never $\rho(x)$ or region membership — and
both the uncalibrated and calibrated test vectors are stored. *Step 3 — the LR+Platt ablation.*
`fit_lr_platt_ablation` fits a second Platt model on the validation *LR* probabilities (identical
procedure), applies it to the test LR probabilities, and records Brier and ECE (`compute_ece`, Equation (11)
with $K = 10$ bins); the expected Zindi values ($\approx 0.088$, $\approx 0.016$) are verified when the flag
is set.

**Transformation.** Uncalibrated hybrid probabilities → sigmoid-rescaled probabilities whose scale is
corrected toward the empirical positive rate while rank order is preserved; the ablation additionally yields
a second calibrated vector from the LR branch alone, isolating the calibration attribution.

**Role.** Implements Section 3.5 verbatim — *"We apply Platt scaling post-hoc, $p_{\mathrm{cal}}(x) =
\frac{1}{1 + \exp(a \cdot p_{\mathrm{Hybrid}}(x) + b)}$, with $a, b$ estimated on a held-out validation
fold"* — under Section 5.3's rule that all "Platt-calibration stages are performed exclusively on the
training and validation partitions". The ablation operationalises the study's principle of isolating "the
sources of predictive gains from calibration gains": the paper's Zindi numbers (LR+Platt Brier 0.088 versus
calibrated hybrid 0.085) attribute most of the calibration improvement to Platt scaling of the
class-weighted baseline, while the calibrated hybrid's ECE falls to 0.024 against 0.234 for the raw LR — the
tenfold reduction Table 2 and Figure 1 report.

<br>

### Task 14 — `orchestrate_compute_metrics(probabilities, labels_test, regions_test, p_lr_test, p_hybrid_test, region_stats, output_dir, verify_manuscript_targets) → Dict[str, Any]`

*Workbook anchor: cell 34, L12549.*

**Inputs.** A dictionary of test probability vectors for the five candidate models — LR, standalone XGBoost,
hybrid uncalibrated, hybrid Platt-calibrated, LR+Platt; the test labels; the test region assignments; the
test LR and hybrid probabilities (per-region AUC inputs); the Task 12 region statistics; `output_dir`; the
verification flag.

**Process.** *Step 1 — discrimination metrics.* `compute_discrimination_metrics` computes, per model on the
test partition, AUC (`roc_auc_score`), the Gini coefficient $\mathrm{Gini} = 2\,\mathrm{AUC} - 1$, and the
minority-class F1 at threshold 0.5 (`precision_recall_fscore_support`), aligned by test row order. *Step 2 —
calibration metrics.* `compute_calibration_metrics` computes the Brier score $\mathrm{Brier} =
\frac{1}{n}\sum_i (p_i - y_i)^2$ (`brier_score_loss`) and the Expected Calibration Error with $K = 10$
equal-width bins on $[0, 1]$,

$$\mathrm{ECE} = \sum_{k=1}^{K} \frac{|B_k|}{n}\,\bigl|\mathrm{acc}(B_k) - \mathrm{conf}(B_k)\bigr|, \tag{11}$$

with empty bins contributing zero, plus the reliability-diagram data (per-bin confidence and accuracy).
*Step 3 — per-region AUC and tables.* `compute_per_region_auc` computes LR and hybrid AUC within each of the
three interpretability regions; `assemble_results_tables` assembles Table 2 (predictive performance) and
Table 3 (region behaviour); `assert_headline_values` — when enabled — verifies the Zindi headline values
(calibrated hybrid AUC $0.869 \pm 0.005$, Brier $0.085 \pm 0.005$, ECE $0.024 \pm 0.005$) and, on deviation,
walks the fidelity-diagnosis checklist in its documented order: split seed → binning algorithm → rare-level
handling → calibration fitting → library versions → statistical-test implementation. Tables persist to JSON.

**Transformation.** Five probability vectors + region labels → the manuscript's reporting tables: rank
metrics (AUC, Gini), threshold metrics (F1 at 0.5), calibration metrics (Brier, ECE, reliability data) and
region-stratified AUCs.

**Role.** Implements Section 5.4's evaluation stack verbatim: *"Discrimination is evaluated via the Area
Under the ROC Curve (AUC), the Gini coefficient ($2\,\mathrm{AUC}-1$), and the minority-class F1-score.
Calibration is evaluated via the Brier score and the Expected Calibration Error of Eq. (11) with $K = 10$
equal-width probability bins, complemented by reliability diagrams."* The F1 convention follows Section
6.1's explicit caveat: *"F1 at 0.5 is reported for comparison with the methodological literature, not as the
operational decision rule"* — Platt scaling shifts probabilities toward the base rate and lowers the
0.5-threshold F1 even when rank-ordering is unchanged.

### Task 15 — `orchestrate_run_fairness_audit(config, test_frame, rho_test, regions_test, p_cal_test, y_test, output_dir, verify_manuscript_targets) → Dict[str, Any]`

*Workbook anchor: cell 35, L13137.*

**Inputs.** The test partition frame carrying the raw protected attributes; the test $\rho$ vector; the
region assignments; the calibrated hybrid test probabilities; the test labels; `config` (protected-attribute
mappings, reference groups); `output_dir`; the verification flag.

**Process.** *Step 1 — recode the protected attributes.* `recode_protected_attributes` builds the audit
frame with `row_id`, `rho`, `region`, `p_cal` and the protected columns: gender (as recorded), location
(urban/rural), education — collapsed per Section 4.5 to primary-or-less versus secondary-and-above, with
`Other` and `Don't know/Refused` mapped to secondary-and-above under a documented placeholder — and country
(Kenya, Rwanda, Tanzania, Uganda); `assert_cell_counts` checks the per-cell counts against Table 4 within
$\pm 5$ observations (the manuscript's counts depend on the exact split). *Step 2 — routing rates.*
`compute_routing_rates` (with `mask_indices`) estimates, per subgroup $g$, the routing rate
$\hat{\Pr}[\rho(x) < 0.7 \mid A = g]$ with 95% percentile bootstrap confidence intervals, $B = 500$
resamples drawn with replacement within (class × attribute) strata; an empty stratum in a resample falls
back to class-only stratification for that resample; the expected point estimates are verified and the
non-overlap of the rural/urban, primary/secondary and Uganda/Kenya intervals is checked when the flag is
set. *Step 3 — pooled and per-region DI and EOD.* `compute_di_and_eod` (with `_eod_for`) forms predicted
positives $\hat{y} = \mathbf{1}[p_{\mathrm{cal}} \ge 0.5]$ and computes, pooled and per region,

$$\mathrm{DI}(g) = \frac{\Pr[\hat{y} = 1 \mid A = g]}{\Pr[\hat{y} = 1 \mid A = g_{\mathrm{ref}}]}, \qquad \mathrm{EOD} = \max_{g}\left\{\bigl|\mathrm{TPR}(g) - \mathrm{TPR}(g_{\mathrm{ref}})\bigr|,\; \bigl|\mathrm{FPR}(g) - \mathrm{FPR}(g_{\mathrm{ref}})\bigr|\right\},$$

applying the degenerate-cell rule: a reference group with zero predicted positives in a region reports the
explicit undefined string "---", never zero. All reports persist with the post-treatment caveat reiterated
in the output.

**Transformation.** Test predictions and raw protected attributes → a per-region fairness profile: routing
rates with bootstrap uncertainty, disparate-impact ratios and equalized-odds differences, pooled and
stratified by the three interpretability regions (Tables 4–6 analogues).

**Role.** Implements Section 3.8 — "which is the empirical focus of the paper" — asking *"whether protected
subgroups (gender, urban/rural residence, education, country) are distributed evenly across those regions,
or whether some groups are systematically routed to the opaque ML-driven region where adverse-action
explanations become harder to produce"*. It reproduces the Section 6.3 findings (routing gaps of 18, 32 and
22 percentage points along location, education and country; gender essentially zero) and encodes two
statistical disciplines verbatim: per-region DI and EOD *"are consequently descriptive stratifications of
the audit surface rather than fairness metrics defined on well-behaved subpopulations"*, because
conditioning on $\rho(x)$ conditions on a variable *"causally downstream of $X$ and therefore a
post-treatment variable in the sense of Rosenbaum"*, while *"the routing-rate audit (list item 1) is not
affected, because $\rho(x)$ is the outcome rather than a conditioning variable"*; and the near-empty
predicted-positive cells of the ML-driven region are *"undefined, not zero, and must be reported explicitly
as such"*. The four-fifths rule ($\mathrm{DI} \in [0.8, 1.25]$) is reported as a reference threshold, not a
decision rule; reference groups are the majority subgroups.

### Task 16 — `orchestrate_run_bootstrap_delong(config, predictions, labels_test, output_dir, verify_manuscript_targets) → Dict[str, Any]`

*Workbook anchor: cell 36, L13519.*

**Inputs.** The frozen test prediction vectors of the hybrid, LR and standalone XGBoost; the test labels;
`config` (resample counts: $B = 2000$ for AUC differences); `output_dir`; the verification flag.

**Process.** *Step 1 — stratified paired bootstrap.* `stratified_paired_bootstrap_auc` runs $B = 2000$
resamples; each resample draws with replacement within the positive-class and negative-class test strata
separately (preserving the marginal class distribution), recomputes both AUCs on the resampled observations
and records $\Delta\mathrm{AUC}^{(b)}$; the 95% percentile interval is formed from the 2.5th and 97.5th
percentiles, for both comparisons (Hybrid vs LR, Hybrid vs XGBoost). *Step 2 — the DeLong test.*
`delong_test` implements the Sun–Xu fast midrank algorithm to estimate the covariance matrix of the two
correlated AUC estimates and computes

$$z = \frac{\widehat{\mathrm{AUC}}_1 - \widehat{\mathrm{AUC}}_2}{\sqrt{\widehat{\mathrm{Var}}(\widehat{\mathrm{AUC}}_1 - \widehat{\mathrm{AUC}}_2)}},$$

reporting the one-sided p-value for $H_1\!:\;\mathrm{AUC}_{\mathrm{Hybrid}} >
\mathrm{AUC}_{\mathrm{comparator}}$ and the two-sided p-value via the normal CDF; the implementation is
validated on a synthetic dataset with known AUCs before use. *Step 3 — consolidation.*
`consolidate_inference_table` assembles Table 9 (comparison, $\Delta\mathrm{AUC}$, bootstrap CI, one-sided
and two-sided DeLong p) for both datasets where applicable and records in the audit log that no
multiple-comparison correction is applied, consistent with the descriptive-audit intent; the table is
serialised to JSON.

**Transformation.** Paired test scores → per-comparison inference records: point $\Delta\mathrm{AUC}$,
percentile bootstrap CI, DeLong $z$ and one-/two-sided p-values.

**Role.** Implements Section 5.6's inference protocol — "Confidence intervals on all AUC differences,
disparate-impact ratios, and equalized-odds differences are constructed by stratified paired bootstrap on
the held-out test partition … For AUC-difference intervals we use $B = 2000$ resamples", with significance
"assessed by the paired DeLong test on correlated ROC curves" computed "by the fast midrank algorithm of Sun
and Xu" — and Section 6.5's headline evidence on Zindi: Hybrid vs LR $\Delta\mathrm{AUC} = +0.015$ (CI
$[+0.011, +0.020]$, $p < 10^{-9}$); Hybrid vs XGBoost $-0.004$ (CI $[-0.009, +0.001]$, $p = 0.92$); Taiwan:
$+0.057$ ($[+0.048, +0.066]$, $p < 0.001$) and $+0.001$ ($[-0.004, +0.006]$, $p = 0.37$). The bootstrap CI
is the primary interval statement; the DeLong test the primary hypothesis test; the absence of
multiple-comparison correction is a recorded design choice of the descriptive audit.

### Task 17 — `orchestrate_run_shap_analysis(config, booster, best_iteration, X_ohe_test, feature_names, regions_test, output_dir, verify_manuscript_targets) → Dict[str, Any]`

*Workbook anchor: cell 37, L13973.*

**Inputs.** The fitted residual XGBoost booster at its frozen best iteration; the one-hot test matrix; the

**Process.** *Step 1 — TreeSHAP attributions.* `compute_tree_shap` restricts

ranks features in descending order and, when the flag is set, asserts the top two ranks match the manuscript

**Transformation.** ML-driven-region features → per-observation additive attributions of the residual

**Role.** Implements Section 3.6 verbatim: *"We compute TreeSHAP attributions on $f_{\mathrm{GBM}}$ for

### Task 19 — `orchestrate_run_alternating_min(config, X_woe_train, X_ohe_train, y_train, X_woe_val, X_ohe_val, p_lr_val_initial, y_val, X_woe_test, X_ohe_test, y_test, beta_initial, beta_0_initial, booster_initial, best_iteration_initial, two_stage_test_auc, two_stage_fit_seconds, output_dir, verify_manuscript_targets) → Dict[str, Any]`

*Workbook anchor: cell 39, L15048.*

**Inputs.** The full WoE and one-hot design matrices of training, validation and test with labels; the
two-stage initialisation $(\beta^{(0)}, \beta_0^{(0)}, f_{\mathrm{GBM}}^{(0)}, \text{best iteration})$; the
two-stage test AUC and wall-clock fit time (the comparison baselines); `config`; `output_dir`; the
verification flag.

**Process.** *Step 1 — the fixed-point iteration.* `implement_fixed_point_iteration` (helpers `_sigmoid`,
`_joint_loss`) alternates, from the two-stage initialisation: (i) hold $f_{\mathrm{GBM}}$ fixed and update
$\beta$ by L-BFGS on the joint loss

$$\mathcal{L}(\beta, f_{\mathrm{GBM}}) = \sum_{i=1}^{N} \ell\!\left(y_i,\, \mathrm{clip}\!\left(p_{\mathrm{LR}}(x_i;\beta) + \alpha^{*}(x_i)\,f_{\mathrm{GBM}}(x_i),\, 0,\, 1\right)\right) + \lambda_{\mathrm{LR}}\|\beta\|_2^2 + \Omega(f_{\mathrm{GBM}}), \tag{12}$$

with $\ell$ the binary cross-entropy $\ell(y, p) = -y\ln p - (1-y)\ln(1-p)$ and $\Omega$ the XGBoost
complexity penalty — critically, $p_{\mathrm{LR}}$, $u_{\mathrm{LR}}$ and $\alpha^{*}$ are recomputed from
the *updated* $\beta$ before any hybrid prediction is formed; (ii) hold $\beta$ fixed, recompute the
residuals $r_i = y_i - p_{\mathrm{LR}}(x_i;\beta)$ and refit the residual booster on them; (iii) iterate
until the change in validation AUC falls below the $10^{-3}$ tolerance. *Step 2 — trajectory recording.*
`run_and_record_trajectory` records the iteration count, the validation-AUC sequence, the final test
$\Delta\mathrm{AUC}$ against the two-stage model, and per-iteration elapsed time via `time.perf_counter`
(the booster refit dominates runtime; expected total $\approx 4\times$ the two-stage fit). *Step 3 —
conclusion.* `document_conclusion` records convergence in three iterations on Zindi with a final
$\Delta\mathrm{AUC}$ of $+0.0006$ — smaller than the tolerance — so two-stage training remains the
operational deployment default; it states explicitly that step (ii) minimizes squared error on the residuals
rather than the joint binary cross-entropy, so the scheme is *not* coordinate descent on $\mathcal{L}$, and
that a true functional-gradient descent is left as future work. The scheme is not run on Taiwan. The report
persists.

**Transformation.** Two-stage parameter initialisation → a fixed-point sequence over $(\beta,
f_{\mathrm{GBM}})$ whose endpoint is scored on the test partition and compared against the two-stage
endpoint, with the full convergence trajectory and compute accounting.

**Role.** Implements Section 3.10's joint formulation — Equation (12) — and its deliberately honest Section
6.6 verdict: *"The tolerance is larger than the reported improvement, so the procedure halts before it can
produce a distinguishable change; we therefore do not interpret this as evidence that two-stage is globally
optimal, only that this particular fixed-point scheme does not improve on it in a detectable way, and that
the additional compute of the joint iteration (approximately four times the two-stage cost in our runs) is
not justified."* The methodological caveat is carried in the output, not buried: the residual-refit sub-step
is squared error on residuals, which is why the scheme "is not coordinate descent on $\mathcal{L}$" and why
"a true functional-gradient descent on the joint loss (which would flow gradients through both branches) is
left" as future work.

> **One registered caveat, disclosed rather than buried.** The convergence tolerance ($10^{-3}$ on
> validation AUC) exceeds the measured improvement ($+0.0006$ test $\Delta\mathrm{AUC}$); the orchestrator
> therefore refuses to claim joint superiority — exactly as the manuscript does — and records the $\approx
> 4\times$ wall-clock ratio as the compute cost of the methodological clarity the joint formulation buys.

## **Usage Example**

Here is a granular, step-by-step guide to executing the end-to-end pipeline for the **"Interpretable hybrid credit scoring for thin-file and underbanked populations: A fairness-aware residual learning framework with evidence from East African financial inclusion data"** study. This example demonstrates how to:
(i) synthetically generate the two raw study frames at high fidelity,
(ii) load `config.yaml` from the working directory into a Python dictionary, and
(iii) execute the full study via the `run_research_pipeline(config, zindi_raw, taiwan_raw, output_dir, run_taiwan, verify_manuscript_targets) → Dict[str, Any]` interface.

> **Environment assumption (mandatory).** All pipeline callables from Tasks 1–23 (`orchestrate_validate_ingested_parameters` … `orchestrate_run_alternating_min`, `run_full_study`, `orchestrate_run_robustness`, `orchestrate_taiwan_reproduction`, `run_research_pipeline`, and their step helpers) are assumed to be defined **in a single Jupyter notebook** cell-context. There is **no folder of `.py` modules** to import, so this example imports only the libraries needed for synthetic data generation, YAML loading and result inspection. `config.yaml` is assumed to be saved in the working directory under that exact name.

> **Verification status.** Every generator below was executed against the *real* Task 1, Task 2 and Task 3 orchestrators of the assessed workbook before being written into this document; all three gates passed at full published scale (23,524 Zindi rows + 30,000 Taiwan rows), and the transcripts embedded below are the actual outputs of those runs. The three defects the gates surfaced — the seed contract, the categorical dtype contract and a pandas-3 library-version drift — are documented at Step 1.4 rather than silently patched away.

<br>

### **Step 1: Synthetic Data Generation (`zindi_raw`, `taiwan_raw`)**

The pipeline ingests exactly two raw frames, and Task 1 validates them against the declared schema, **halting** on mismatch. The generators therefore hit the published structure *by construction*, not by sampling luck.

#### The binding constraints that dictate the generator design

`validate_zindi_raw` enforces `shape == (23524, 13)`, the exact country composition, zero missing values, the declared column order, and a positive-class rate within $10^{-4}$ of $0.1408$ via `math.isclose`:

$$n_{\text{Kenya}} + n_{\text{Rwanda}} + n_{\text{Tanzania}} + n_{\text{Uganda}} = 6068 + 8735 + 6620 + 2101 = 23{,}524,$$
$$\hat{\pi}_{+} = \frac{k_{+}}{n}, \qquad k_{+} = \operatorname{round}(0.1408 \times 23{,}524) = 3312, \qquad |\hat{\pi}_{+} - 0.1408| = 7.9 \times 10^{-6} < 10^{-4}.$$

Sampling labels as $\mathrm{Bernoulli}(0.1408)$ has standard deviation $\sqrt{p(1-p)/n} \approx 2.3 \times 10^{-3}$, which lands inside the $10^{-4}$ band only $\approx 3.5\%$ of the time; the generator therefore constructs **exact counts** ($k_{+}$ positives by rank assignment) and shuffles, so the gate passes deterministically. Taiwan uses the same discipline — $k_{+} = \operatorname{round}(0.22 \times 30{,}000) = 6{,}600$, giving $\hat{\pi}_{+} = 0.22$ on the nose (gate tolerance $10^{-2}$) — with the declared code sets reproduced exactly: `EDUCATION ∈ {0,…,6}` including the unknown codes, `MARRIAGE ∈ {0,1,2,3}`, the `PAY_0, PAY_2 … PAY_6` layout with **no `PAY_1`**, and `LIMIT_BAL` restricted to multiples of 10,000 NT dollars. The `year` column is a declared function of the country (Kenya 2018, Rwanda 2016, Tanzania 2017, Uganda 2017, mirroring the assessed workbook's preamble mapping), reproducing the manuscript's collinearity rationale; the exact public pairing is UNSPECIFIED IN MANUSCRIPT and immaterial because the column is dropped before modelling. Faker was considered for the synthesis, but every categorical column has a fixed, exhaustive level set declared in the configuration — no free text is ever synthesised — so a seeded NumPy `Generator` (the functionally similar model the instruction permits) is the accurate choice.

#### Generating the two raw frames

```python
# ==============================================================================
# Step 1 - High-Fidelity Synthetic Generation of the Two Raw Study Frames
# ==============================================================================

# Standard library and third-party imports required by the generators and the
# loader only; every pipeline callable is already defined in the notebook
# context. Faker was considered for synthetic generation, but every categorical
# column of this study has a FIXED, exhaustive level set declared in the
# configuration - no free text is ever synthesised - so a seeded NumPy
# Generator (the functionally similar model permitted by the instruction) is
# the accurate, deterministic choice.
import math
from pathlib import Path
from typing import Any, Dict, List, Tuple

import numpy as np
import pandas as pd
import yaml

# ------------------------------------------------------------------------------
# Declared schema constants.  These mirror config.yaml exactly
# (dataset_parameters.zindi_fsd and raw_data_schemas.taiwan_credit_schema);
# they are repeated here so the generators are self-contained before the
# configuration is loaded in Step 2.
# ------------------------------------------------------------------------------
ZINDI_N_ROWS: int = 23524
ZINDI_RAW_COLUMN_ORDER: List[str] = [
    "uniqueid", "country", "year", "location_type", "cellphone_access",
    "gender_of_respondent", "relationship_with_head", "marital_status",
    "education_level", "job_type", "household_size", "age_of_respondent",
    "bank_account",
]
# Exact published country composition (sums to 23,524 by construction)
COUNTRY_COMPOSITION: Dict[str, int] = {
    "Kenya": 6068, "Rwanda": 8735, "Tanzania": 6620, "Uganda": 2101,
}
# Country -> survey-year pairing (declared; column dropped before modelling)
COUNTRY_YEAR: Dict[str, int] = {
    "Kenya": 2018, "Rwanda": 2016, "Tanzania": 2017, "Uganda": 2017,
}
# Illustrative per-country level distributions (NOT published constants; the
# validated invariants are the level SETS, the composition and the rates)
URBAN_PROB: Dict[str, float] = {"Kenya": 0.55, "Rwanda": 0.25, "Tanzania": 0.30, "Uganda": 0.25}
CELLPHONE_PROB: Dict[str, float] = {"Kenya": 0.80, "Rwanda": 0.55, "Tanzania": 0.60, "Uganda": 0.55}
RELATIONSHIP_LEVELS: List[str] = [
    "Head of Household", "Spouse", "Child", "Parent",
    "Other relative", "Other non-relatives", "Don't know/Refused",
]
RELATIONSHIP_PROB: List[float] = [0.57, 0.20, 0.09, 0.05, 0.05, 0.03, 0.01]
MARITAL_LEVELS: List[str] = [
    "Married/Living together", "Single/Never married", "Widowed",
    "Divorced/Seperated", "Don't know/Refused",  # typo preserved verbatim
]
MARITAL_PROB: List[float] = [0.62, 0.22, 0.08, 0.06, 0.02]
EDUCATION_LEVELS: List[str] = [
    "No formal education", "Primary education", "Secondary education",
    "Tertiary education", "Other", "Don't know/Refused",
]
EDUCATION_PROB: List[float] = [0.15, 0.42, 0.28, 0.08, 0.05, 0.02]
EDUCATION_ORDINAL: Dict[str, int] = {
    "No formal education": 0, "Primary education": 1, "Secondary education": 2,
    "Tertiary education": 3, "Other": 1, "Don't know/Refused": 0,
}
JOB_LEVELS: List[str] = [
    "Government", "Formally employed Private sector", "Informally employed",
    "Self employed", "Farming and Fishing", "Remittance Dependent",
    "Other", "Don't know/Refused",
]
JOB_PROB: List[float] = [0.05, 0.10, 0.28, 0.22, 0.25, 0.05, 0.04, 0.01]
POSITIVE_RATE: float = 0.1408

TAIWAN_N_ROWS: int = 30000
TAIWAN_RAW_COLUMN_ORDER: List[str] = (
    ["ID", "LIMIT_BAL", "SEX", "EDUCATION", "MARRIAGE", "AGE"]
    + [f"PAY_{k}" for k in (0, 2, 3, 4, 5, 6)]
    + [f"BILL_AMT{k}" for k in range(1, 7)]
    + [f"PAY_AMT{k}" for k in range(1, 7)]
    + ["default.payment.next.month"]
)
TAIWAN_POSITIVE_RATE: float = 0.22


def generate_zindi_raw(n_rows: int = ZINDI_N_ROWS, seed: int = 42) -> pd.DataFrame:
    """
    Generate a schema-exact synthetic Zindi FSD raw frame.

    Purpose:
        Produce the direct, unprocessed stand-in for the Zindi Financial
        Inclusion in Africa competition training file: 13 columns in the
        declared order, 23,524 rows, zero missing values, the exact published
        country composition, the 14.08% marginal positive-class rate within
        the 1e-4 tolerance Task 1 enforces, and the country/year collinearity
        the manuscript gives as the reason `year` is dropped.

    Inputs:
        n_rows (int): Row count; must equal the published 23,524 for the
            composition constants to apply.
        seed (int): Seed for the NumPy Generator (the functionally similar
            synthetic-data model; the categorical level sets are fixed by the
            config, so no free-text synthesiser such as Faker is required).

    Processes:
        1. Build per-country blocks whose sizes are the PUBLISHED counts
           (Kenya 6,068; Rwanda 8,735; Tanzania 6,620; Uganda 2,101), so the
           composition gate passes by construction, not by sampling luck.
        2. Pair each respondent's `year` with the country through the declared
           country->year mapping, reproducing the collinearity invariant.
        3. Sample the categorical columns from the declared exhaustive level
           sets (typo `Divorced/Seperated` preserved verbatim) and the
           numerical columns from plausible ranges.
        4. Construct the label with an education/cellphone/age/country
           dependence through a logistic score, then calibrate the marginal
           EXACTLY: the k = round(0.1408 * n) highest-scored rows are set to
           1, so |phat - 0.1408| < 1e-4 deterministically.
        5. Shuffle rows once, enforce dtypes and the declared column order,
           and self-verify every invariant before returning.

    Outputs:
        pd.DataFrame: (n_rows, 13) frame; object columns for the categoricals,
            int64 for year/household_size/age_of_respondent/bank_account, in
            the declared raw column order.

    Raises:
        ValueError: If n_rows differs from the published count or any
            self-verification fails.
    """
    # The composition constants only apply at the published row count
    if n_rows != ZINDI_N_ROWS:
        raise ValueError(f"n_rows must be {ZINDI_N_ROWS}; got {n_rows}.")
    # A dedicated generator keeps the synthetic run reproducible
    rng: np.random.Generator = np.random.default_rng(seed)
    # Per-country row indices: exact published block sizes
    country_parts: List[np.ndarray] = [
        np.full(cnt, name, dtype=object) for name, cnt in COUNTRY_COMPOSITION.items()
    ]
    # Concatenate the blocks into one country vector of length n_rows
    country: np.ndarray = np.concatenate(country_parts)
    # The year column is a pure function of the country (declared mapping)
    year: np.ndarray = np.array(
        [COUNTRY_YEAR[c] for c in country], dtype=np.int64)
    # Urban/rural per the FinScope instrument, country-conditional
    location_type: np.ndarray = np.array(
        ["Urban" if rng.random() < URBAN_PROB[c] else "Rural" for c in country],
        dtype=object)
    # Cellphone access, country-conditional
    cellphone_access: np.ndarray = np.array(
        ["Yes" if rng.random() < CELLPHONE_PROB[c] else "No" for c in country],
        dtype=object)
    # Gender as recorded
    gender_of_respondent: np.ndarray = rng.choice(
        ["Female", "Male"], size=n_rows, p=[0.59, 0.41]).astype(object)
    # Relationship to the head of household, declared level set
    relationship_with_head: np.ndarray = rng.choice(
        RELATIONSHIP_LEVELS, size=n_rows, p=RELATIONSHIP_PROB).astype(object)
    # Marital status, declared level set (typo preserved verbatim)
    marital_status: np.ndarray = rng.choice(
        MARITAL_LEVELS, size=n_rows, p=MARITAL_PROB).astype(object)
    # Education level, declared level set; drives the label below
    education_level: np.ndarray = rng.choice(
        EDUCATION_LEVELS, size=n_rows, p=EDUCATION_PROB).astype(object)
    # Employment category, declared level set
    job_type: np.ndarray = rng.choice(
        JOB_LEVELS, size=n_rows, p=JOB_PROB).astype(object)
    # Household size: 1 to 12 members
    household_size: np.ndarray = rng.integers(1, 13, n_rows).astype(np.int64)
    # Age: right-skewed adult distribution clipped to [18, 90]
    age_of_respondent: np.ndarray = np.clip(
        rng.normal(40.0, 15.0, n_rows).round(), 18, 90).astype(np.int64)
    # Logistic score for the label: strong education/cellphone signal plus
    # demographic and country terms, mirroring the paper's IV ordering
    score: np.ndarray = (
        -3.1
        + 1.35 * np.array([EDUCATION_ORDINAL[e] for e in education_level])
        + 0.90 * (cellphone_access == "Yes")
        + 0.020 * (age_of_respondent - 40)
        + 0.25 * (country == "Kenya")
        + 0.05 * (household_size - 5)
        + rng.normal(0.0, 0.8, n_rows))
    # EXACT marginal calibration: k highest scores are labelled 1
    k_positive: int = int(round(POSITIVE_RATE * n_rows))
    order: np.ndarray = np.argsort(-score, kind="stable")
    bank_account: np.ndarray = np.zeros(n_rows, dtype=np.int64)
    bank_account[order[:k_positive]] = 1
    # Unique-per-row identifier in the declared Zindi format
    uniqueid: np.ndarray = np.array(
        [f"uniqueid_{i:05d}" for i in range(1, n_rows + 1)], dtype=object)
    # Assemble in the declared raw column order
    zindi_raw: pd.DataFrame = pd.DataFrame({
        "uniqueid": uniqueid,
        "country": country,
        "year": year,
        "location_type": location_type,
        "cellphone_access": cellphone_access,
        "gender_of_respondent": gender_of_respondent,
        "relationship_with_head": relationship_with_head,
        "marital_status": marital_status,
        "education_level": education_level,
        "job_type": job_type,
        "household_size": household_size,
        "age_of_respondent": age_of_respondent,
        "bank_account": bank_account,
    })[ZINDI_RAW_COLUMN_ORDER]
    # Shuffle rows once for realism; composition is order-invariant
    zindi_raw = zindi_raw.iloc[rng.permutation(n_rows)].reset_index(drop=True)
    # EXPLICIT object casts: modern pandas infers the str dtype for string
    # arrays, but the Task 1 schema contract requires object dtype for every
    # categorical column (the gate rejects str; see Step 1.4). The cast is
    # the documented fix.
    for categorical_column in [
            "uniqueid", "country", "location_type", "cellphone_access",
            "gender_of_respondent", "relationship_with_head",
            "marital_status", "education_level", "job_type"]:
        zindi_raw[categorical_column] = zindi_raw[categorical_column].astype(object)
    # Self-verify the invariants Task 1 will check (fail fast, not silently)
    if zindi_raw.isnull().sum().sum() != 0:
        raise ValueError("synthetic Zindi frame contains missing values.")
    observed: Dict[str, int] = zindi_raw["country"].value_counts().to_dict()
    if observed != COUNTRY_COMPOSITION:
        raise ValueError(f"country composition drifted: {observed}.")
    observed_rate: float = float(zindi_raw["bank_account"].mean())
    if abs(observed_rate - POSITIVE_RATE) >= 1e-4:
        raise ValueError(f"positive rate {observed_rate} outside 1e-4 of 0.1408.")
    return zindi_raw


def generate_taiwan_raw(n_rows: int = TAIWAN_N_ROWS, seed: int = 43) -> pd.DataFrame:
    """
    Generate a schema-exact synthetic Taiwan Credit Default raw frame.

    Purpose:
        Produce the direct stand-in for the UCI 'default of credit card
        clients' file: 25 columns in the UCI order (ID + 23 predictors +
        label), 30,000 rows, no missing values, the declared categorical code
        sets (including the unknown codes 0/5/6 and 0), the PAY_0..PAY_6
        layout with NO PAY_1, and the ~22% positive rate within 1e-2.

    Inputs:
        n_rows (int): Row count; must equal the published 30,000.
        seed (int): Seed for the NumPy Generator (independent stream from the
            Zindi generator).

    Processes:
        1. Assign ID = 1..n (UCI layout).
        2. Sample LIMIT_BAL in NT-dollar multiples of 10,000, the categorical
           codes from their declared sets with unknown codes included, AGE
           over [21, 79], and repayment status PAY_0..PAY_6 over the declared
           {-2, -1, 0, 1..8} scale with plausible delinquency persistence.
        3. Draw bill and payment amounts from log-normal distributions in NT
           dollars (bill statements may be negative; payments are >= 0), cast
           to int64.
        4. Construct the label with a repayment-status-driven logistic score
           and calibrate the marginal EXACTLY to round(0.22 * n) = 6,600
           positives, so phat = 0.22 on the nose.
        5. Enforce dtypes and the UCI column order and self-verify.

    Outputs:
        pd.DataFrame: (n_rows, 25) frame; every column int64, in the UCI
            column order.

    Raises:
        ValueError: If n_rows differs from the published count or any
            self-verification fails.
    """
    # The composition constants only apply at the published row count
    if n_rows != TAIWAN_N_ROWS:
        raise ValueError(f"n_rows must be {TAIWAN_N_ROWS}; got {n_rows}.")
    # Independent stream so the two generators are separately reproducible
    rng: np.random.Generator = np.random.default_rng(seed)
    # Row identifier preserved from the UCI layout
    id_column: np.ndarray = np.arange(1, n_rows + 1, dtype=np.int64)
    # Credit limit in NT dollars: multiples of 10,000 up to 800,000
    limit_bal: np.ndarray = (rng.integers(1, 81, n_rows) * 10_000).astype(np.int64)
    # Gender code: 1 = male, 2 = female
    sex: np.ndarray = rng.choice([1, 2], size=n_rows, p=[0.40, 0.60]).astype(np.int64)
    # Education code: 1..4 valid; 0/5/6 unknown (declared set preserved)
    education: np.ndarray = rng.choice(
        [1, 2, 3, 4, 5, 6, 0], size=n_rows,
        p=[0.16, 0.47, 0.16, 0.11, 0.04, 0.03, 0.03]).astype(np.int64)
    # Marriage code: 1/2/3 valid; 0 unknown
    marriage: np.ndarray = rng.choice(
        [1, 2, 3, 0], size=n_rows, p=[0.53, 0.45, 0.01, 0.01]).astype(np.int64)
    # Age in years over the observed [21, 79] range
    age: np.ndarray = rng.integers(21, 80, n_rows).astype(np.int64)
    # Repayment status per month: {-2, -1, 0} overwhelmingly, 1..8 delinquency
    pay_probabilities: np.ndarray = np.array(
        [0.12, 0.12, 0.70, 0.03, 0.015, 0.010,
         0.005, 0.004, 0.003, 0.003, 0.005], dtype=float)
    pay_values: List[int] = [-2, -1, 0, 1, 2, 3, 4, 5, 6, 7, 8]
    pay_columns: Dict[str, np.ndarray] = {}
    # Delinquency persistence grows slightly with the lag, not the reverse
    for lag, name in enumerate(["PAY_0", "PAY_2", "PAY_3", "PAY_4", "PAY_5", "PAY_6"]):
        # Shift a small mass from 0 to the delinquency codes as the lag grows
        lagged_probabilities: np.ndarray = pay_probabilities.copy()
        lagged_probabilities[2] -= 0.004 * lag
        lagged_probabilities[3] += 0.001 * lag
        lagged_probabilities[4] += 0.001 * lag
        lagged_probabilities[5] += 0.0005 * lag
        lagged_probabilities[6] += 0.0005 * lag
        # Renormalise so the shifted mass sums to exactly one
        lagged_probabilities /= lagged_probabilities.sum()
        pay_columns[name] = rng.choice(
            pay_values, size=n_rows, p=lagged_probabilities).astype(np.int64)
    # Bill statements: log-normal NT dollars; ~2% negative (credit balances)
    bill_amts: Dict[str, np.ndarray] = {}
    for k in range(1, 7):
        amounts: np.ndarray = np.exp(rng.normal(9.5, 1.1, n_rows))
        negative_mask: np.ndarray = rng.random(n_rows) < 0.02
        amounts[negative_mask] = -amounts[negative_mask]
        bill_amts[f"BILL_AMT{k}"] = amounts.round().astype(np.int64)
    # Previous payments: smaller log-normal amounts; ~2% exact zeros
    pay_amt_columns: Dict[str, np.ndarray] = {}
    for k in range(1, 7):
        amounts = np.exp(rng.normal(7.6, 1.3, n_rows))
        amounts[rng.random(n_rows) < 0.02] = 0.0
        pay_amt_columns[f"PAY_AMT{k}"] = amounts.round().astype(np.int64)
    # Logistic label score driven by recent repayment status and the limit
    score: np.ndarray = (
        -2.9
        + 0.55 * pay_columns["PAY_0"]
        + 0.18 * pay_columns["PAY_2"]
        + 0.10 * pay_columns["PAY_3"]
        - 0.25 * np.log(limit_bal / 10_000.0 + 1.0)
        + 0.012 * (age - 21)
        + rng.normal(0.0, 0.9, n_rows))
    # EXACT marginal calibration: round(0.22 * 30000) = 6600 positives
    k_positive: int = int(round(TAIWAN_POSITIVE_RATE * n_rows))
    order: np.ndarray = np.argsort(-score, kind="stable")
    default_next_month: np.ndarray = np.zeros(n_rows, dtype=np.int64)
    default_next_month[order[:k_positive]] = 1
    # Assemble in the exact UCI column order
    taiwan_raw: pd.DataFrame = pd.DataFrame(
        {"ID": id_column, "LIMIT_BAL": limit_bal, "SEX": sex,
         "EDUCATION": education, "MARRIAGE": marriage, "AGE": age,
         **pay_columns, **bill_amts, **pay_amt_columns,
         "default.payment.next.month": default_next_month}
    )[TAIWAN_RAW_COLUMN_ORDER]
    # Self-verify the invariants Task 1 will check
    if taiwan_raw.isnull().sum().sum() != 0:
        raise ValueError("synthetic Taiwan frame contains missing values.")
    observed_rate: float = float(taiwan_raw["default.payment.next.month"].mean())
    if abs(observed_rate - TAIWAN_POSITIVE_RATE) >= 1e-2:
        raise ValueError(f"positive rate {observed_rate} outside 1e-2 of 0.22.")
    return taiwan_raw


# ---- Generate both raw frames ----------------------------------------------
zindi_raw: pd.DataFrame = generate_zindi_raw()
taiwan_raw: pd.DataFrame = generate_taiwan_raw()

# ---- Verification transcript (composition, rates, schema) -------------------
print("Zindi raw frame:", zindi_raw.shape, "| zero missing:",
      int(zindi_raw.isnull().sum().sum()))
print("country composition:", zindi_raw["country"].value_counts().to_dict())
print("positive rate:", round(float(zindi_raw["bank_account"].mean()), 6))
print("year-country pairs:", sorted(
    {(c, int(y)) for c, y in zip(zindi_raw['country'], zindi_raw['year'])}))
print("Divorced/Seperated preserved:", zindi_raw["marital_status"].isin(
    ["Divorced/Seperated"]).any())
print()
print("Taiwan raw frame:", taiwan_raw.shape, "| zero missing:",
      int(taiwan_raw.isnull().sum().sum()))
```

#### Three defects the real gates surfaced during verification

All three were found by running the generators against the assessed workbook's `orchestrate_validate_ingested_parameters`, `orchestrate_cleanse_data` and `orchestrate_create_partitions` before this document was written, and all three are corrected above or below.

**(a) The seed contract.** `validate_study_config` enforces $45 \le \text{seed} \le 2^{31}-1$; the specification draft's placeholder value `42` is rejected with the exact error:

> `ValueError: random_seed must satisfy 45 <= seed <= 2^31 - 1; received 42.`

The shipped `config.yaml` therefore carries the gate-compliant placeholder `20260830` (the study's prior artifact value), with the contract documented in the file's `REVISED` comment block. Both values remain UNSPECIFIED IN MANUSCRIPT placeholders; the gate, not the spec draft, is the binding contract.

**(b) Categorical dtype.** Constructing string columns through NumPy arrays lets modern pandas infer the `str` dtype, which the gate rejects with the exact error:

> `ValueError: zindi_raw['country'] must have object dtype; received str.`

The fix is the explicit `.astype(object)` cast over the nine categorical columns, applied above. A `str`-dtype column would reach the WoE encoder, whose map keys are built with `str(...)` comparisons, and a dtype mismatch there sends every lookup to the neutral fallback, silently destroying the feature.

**(c) A pandas-3 library-version drift.** The assessed workbook was authored against pandas 2.x, where `DataFrame.drop(columns=..., axis=1)` was legal; pandas 3 rejects the combination with `ValueError: Cannot specify both 'axis' and 'index'/'columns'` — the "library-version behaviour" channel of the study's own fidelity-diagnosis checklist. The minimal, documented compatibility shim — applied once, in the notebook, before any pipeline call — is:

```python
# ---- pandas >= 3 compatibility shim (documented library-version drift) ----
# pandas 3.x rejects DataFrame.drop(columns=..., axis=1); the assessed
# workbook was authored against pandas 2.x, where the combination was legal.
# The shim wraps DataFrame.drop once, rewriting ONLY that combination, so
# every callable in the notebook runs unmodified.
_ORIGINAL_DATAFRAME_DROP = pd.DataFrame.drop
def _pandas3_drop_compat(self, *args: Any, **kwargs: Any):
    # Rewrite only the rejected combination: named columns plus a redundant
    # axis keyword; everything else passes through untouched
    if "columns" in kwargs and "axis" in kwargs:
        # Copy the keyword dictionary so the caller's dict is never mutated
        compatible_kwargs = dict(kwargs)
        # pandas 3 no longer accepts the redundant axis alongside columns
        compatible_kwargs.pop("axis")
        # Delegate to the original bound implementation
        return _ORIGINAL_DATAFRAME_DROP(self, *args, **compatible_kwargs)
    # Unchanged behaviour for every other call signature
    return _ORIGINAL_DATAFRAME_DROP(self, *args, **kwargs)
pd.DataFrame.drop = _pandas3_drop_compat
```

With the shim applied, the three real gates pass on the synthetic frames; the transcript below is the actual output of the run:

### **Step 2: Loading the Configuration (`config.yaml`)**

`config.yaml` is the single frozen source of truth for the study; it is read into memory and converted into the nested `config` dictionary every orchestrator consumes. `yaml.safe_load` guarantees a plain object graph — dicts, lists and scalars only, never arbitrary Python types — and the loader fails loudly on a missing file or a truncated section rather than proceeding with an empty dictionary, consistent with the pipeline's fail-fast gates.

```python
# ==============================================================================
# Step 2 - Load config.yaml from the Working Directory into a Python Dictionary
# ==============================================================================

def load_study_configuration(filepath: str = "config.yaml") -> Dict[str, Any]:
    """
    Load the frozen study configuration from the working directory.

    Purpose:
        Ingest the 32 configuration sections consumed by the orchestrators -
        metadata and headline targets, dataset parameters, raw schemas,
        partitioning, preprocessing, WoE and IV, model configurations,
        interpretability, training protocol, calibration, metrics, protected
        attributes, fairness audit, thin-file segmentation, joint
        optimization, inference, SHAP, methodological notes, plotting, the
        equation-parameter map and availability - and convert them into the
        nested dictionary the pipeline consumes.

    Inputs:
        filepath (str): Path to the YAML file, default 'config.yaml',
            assumed to be saved in the working directory under that name.

    Processes:
        1. Validate the argument type and the file's existence.
        2. Parse the document with yaml.safe_load, which constructs only
           plain dict/list/scalar objects and never arbitrary Python types.
        3. Verify that all 32 required top-level sections are present.
        4. Report the section inventory to the operator.

    Outputs:
        Dict[str, Any]: The complete nested configuration dictionary.

    Raises:
        TypeError: If filepath is not a string.
        FileNotFoundError: If the configuration file is absent.
        yaml.YAMLError: If the document is malformed.
        ValueError: If a required top-level section is missing.
    """
    # Reject a non-string path before touching the filesystem
    if not isinstance(filepath, str):
        raise TypeError(f"filepath must be a string; got {type(filepath)}.")
    # Resolve the path relative to the working directory
    config_path: Path = Path(filepath)
    # A missing configuration would silently break every downstream gate
    if not config_path.is_file():
        raise FileNotFoundError(
            f"Configuration file '{filepath}' not found in the working "
            f"directory {Path.cwd()}."
        )
    try:
        # Open the document with an explicit encoding for portability
        with open(config_path, "r", encoding="utf-8") as handle:
            # safe_load constructs only plain containers and scalars
            config: Dict[str, Any] = yaml.safe_load(handle)
    except yaml.YAMLError as exc:
        # Surface the parser context rather than a bare failure
        raise yaml.YAMLError(
            f"Failed to parse YAML configuration '{filepath}': {exc}"
        ) from exc
    # The 32 sections the orchestrators read unconditionally
    required_sections: Tuple[str, ...] = (
        "study_metadata", "prompt_templates", "llm_configuration",
        "dataset_parameters", "raw_data_schemas", "partitioning_parameters",
        "preprocessing_parameters", "woe_parameters", "iv_parameters",
        "logistic_regression_parameters", "residual_xgboost_parameters",
        "standalone_xgboost_parameters", "lightgbm_sensitivity_parameters",
        "adaptive_weighting_parameters", "hybrid_prediction_parameters",
        "interpretability_parameters", "network_training_parameters",
        "calibration_parameters", "evaluation_metrics_parameters",
        "protected_attribute_parameters", "fairness_audit_parameters",
        "thin_file_parameters", "hard_soft_split_parameters",
        "joint_optimization_parameters", "bootstrap_inference_parameters",
        "delong_test_parameters", "shap_explanation_parameters",
        "training_protocol_parameters", "essential_methodological_notes",
        "plot_construction_parameters", "equation_parameter_map",
        "code_and_data_availability",
    )
    # Fail loudly on a truncated configuration rather than at first use
    absent: List[str] = [s for s in required_sections if s not in config]
    if absent:
        raise ValueError(
            f"config.yaml is missing required section(s): {absent}."
        )
    # Report the inventory so the operator can see what was loaded
    print(f"Loaded configuration from {filepath}: "
          f"{len(config)} top-level sections.")
    return config


# Load the frozen study configuration from the working directory
config: Dict[str, Any] = load_study_configuration("config.yaml")

# Confirm the parameters the downstream gates depend on most directly
print(f"  random_seed           : "
      f"{config['partitioning_parameters']['random_seed']}")
print(f"  C grid                : "
      f"{config['logistic_regression_parameters']['C_grid']}")
print(f"  region thresholds     : "
      f"rho >= {config['interpretability_parameters']['region_definitions']['fully_interpretable']['condition'].split('>= ')[1]}, "
      f"rho < {config['interpretability_parameters']['region_definitions']['ml_driven']['condition'].split('< ')[1]}")
print(f"  alpha grids           : "
      f"{config['adaptive_weighting_parameters']['alpha_min_grid']} / "
      f"{config['adaptive_weighting_parameters']['alpha_max_grid']} / "
      f"{config['adaptive_weighting_parameters']['gamma_grid']}")
print(f"  bootstrap resamples   : "
      f"AUC {config['bootstrap_inference_parameters']['auc_difference_resamples']}, "
      f"fairness {config['bootstrap_inference_parameters']['fairness_quantity_resamples']}")
print(f"  IV threshold          : "
      f"{config['preprocessing_parameters']['information_value_filter']['threshold']}")
```

### **Step 3: Executing the Pipeline (`run_research_pipeline`)**

With the two frames and the configuration in memory, the top-level orchestrator is invoked. All six parameters are covered below.

| Parameter | Type | Default | Role |
|---|---|---|---|
| `config` | `Dict[str, Any]` | *required* | The frozen configuration loaded in Step 2 |
| `zindi_raw` | `pd.DataFrame` | *required* | Raw Zindi FSD frame, (23,524, 13), declared order |
| `taiwan_raw` | `pd.DataFrame` | *required* | Raw UCI Taiwan frame, (30,000, 25), UCI order |
| `output_dir` | `Path` | *required* | Destination for the run manifest, checkpoints and reports |
| `run_taiwan` | `bool` | `False` | Gate on the Task 22 Taiwan continuity chain |
| `verify_manuscript_targets` | `bool` | `False` | Gate on the headline-result and Tables 3/4/9 assertions |

The two flags are deliberately independent. `run_taiwan=True` executes the Task 22 continuity-benchmark chain (placeholder preprocessing flagged UNSPECIFIED IN MANUSCRIPT, the identical model chain, deviation documentation) on top of the Zindi run. `verify_manuscript_targets=True` activates every manuscript-target assertion (AUC $0.869 \pm 0.005$, Brier $0.085 \pm 0.005$, ECE $0.024 \pm 0.005$, the region shares, the routing-gap point estimates, the SHAP top-five order, the thin-file profile) — with the consequence, stated honestly, that these assertions can only pass on the genuine Zindi/Taiwan files; on synthetic frames they would fail by design, which is why the default and the smoke run use `False`.

**Discussion of Step 3.** The orchestrator call is thin because every intermediate callable runs internally — validation, cleansing, the 60/20/20 stratification, frozen preprocessing, WoE/IV/OHE encoding, the LR baseline with C selection, the raw-residual XGBoost learner, the adaptive-weighting grid, the hybrid and $\rho$ decomposition, Platt calibration, the metric tables, the per-region fairness audit, the bootstrap/DeLong inference, the TreeSHAP stability screen, the thin-file analysis and the alternating-minimization check. Three behaviours are worth stating explicitly:

- **The test-partition information barrier is enforced, not conventional.** Inside `run_full_study`, the test partition lives under a protected state key whose SHA-256 hash is verified at every checkpoint; a callable that mutates the test indices or labels is detected, not tolerated.
- **Region assignments are frozen before calibration.** Task 12 computes $\rho(x)$ and the three regions from the *uncalibrated* hybrid probabilities; Task 13's Platt scaling changes only the probability scale, never $\rho$ or region membership — so the Task 15 fairness audit stratifies exactly the regions the manuscript reports (6.4% / 34.1% / 59.4% shares on Zindi).
- **The manifest is the audit instrument.** Every seed, hyperparameter selection, environment version and deviation is logged — including the documented placeholders (the gate-compliant seed, the pandas-3 shim, and every UNSPECIFIED IN MANUSCRIPT choice the configuration flags).

> **Realism note.** The synthetic frames reproduce the published *schema, composition and marginal rates* — and they pass the real Task 1–3 gates — but they carry no genuine dependence structure. The manuscript's headline numbers (calibrated hybrid AUC 0.869, Brier 0.085, the 18/32/22-percentage-point routing gaps, the SHAP top-feature ordering) therefore **will not reproduce** on this corpus, which is exactly why the smoke run uses `verify_manuscript_targets=False`. Reproducing those figures requires the genuine Zindi FSD and UCI Taiwan downloads, loaded into the same two frame variables with `verify_manuscript_targets=True`; the configuration, the pipeline and the gates are then identical. No synthetic generator can manufacture the manuscript's dependence structure, and none is claimed to.

### **Summary of the Execution Flow**

1. **Data Ingestion.** `generate_zindi_raw()` and `generate_taiwan_raw()` produce the two raw frames by **exact-count construction**: 23,524 × 13 and 30,000 × 25, zero missing values, declared column orders, the published country composition, the country↔year pairing, the verbatim `Divorced/Seperated` typo, the declared `EDUCATION`/`MARRIAGE` code sets, no `PAY_1`, positive rates 0.140792 (within $10^{-4}$ of 0.1408) and 0.22 (within $10^{-2}$).
2. **Configuration Loading.** `load_study_configuration("config.yaml")` parses the 32 sections with `yaml.safe_load` and validates that all of them are present, failing loudly otherwise.
3. **Validation & Cleansing (Tasks 1–2).** The orchestrators enforce the shape, order, dtype, level-set, composition and rate contracts; drop `uniqueid`/`year` and `ID` by named-column access; strip categorical whitespace while preserving the verbatim typo; enforce the `int64` contracts; and seal the SHA-256 audit record before any fitting.
4. **Partitioning (Task 3).** The chained `StratifiedShuffleSplit` freezes the 60/20/20 indices — Zindi 14,114 / 4,705 / 4,705 at test rate 0.1407; Taiwan 18,000 / 6,000 / 6,000 at 0.22 — and the test partition becomes read-only under the information barrier.
5. **Preprocessing (Tasks 4–6).** Training-only medians/modes and 1st/99th-percentile winsorization; monotonic binning and WoE encoding for the logistic branch with the neutral WoE = 0 rule; the IV ≥ 0.02 filter with the Zindi targets asserted; frozen one-hot encoding for the boosting branch.
6. **Modelling (Tasks 7–13).** C selection over {0.01, 0.1, 1, 10}; the balanced L2 `lbfgs` baseline producing $p_{\mathrm{LR}}$; raw residuals $r_i = y_i - p_{\mathrm{LR}}(x_i)$; the `reg:squarederror` residual XGBoost with hybrid-AUC early stopping; the standalone comparator with `scale_pos_weight`; the 27-triple adaptive-weight grid over $u_{\mathrm{LR}}$ and $\alpha^{*}$; the clipped hybrid, $\rho(x)$ and the three-region partition; Platt calibration with the LR+Platt ablation.
7. **Evaluation & Audit (Tasks 14–17).** AUC, Gini, F1 at 0.5, Brier and ECE with reliability data; per-region AUC; the per-region fairness audit (routing rates with $B = 500$ bootstrap CIs; pooled and per-region DI and EOD with the degenerate-cell "---" rule); the $B = 2000$ stratified paired bootstrap and the Sun–Xu DeLong test; TreeSHAP on the residual learner only, with the rank-stability screen.
8. **Extensions & Robustness (Tasks 18–21).** The thin-file composite index with the frozen bottom-quartile threshold and the H1/H2 verdicts; the alternating-minimization fixed-point scheme with its honest no-improvement conclusion; threshold, LightGBM and reference-group sensitivity analyses.
9. **Consolidation (Task 23, with Task 22 when flagged).** One stage-keyed artefacts dictionary, the persisted JSON run manifest carrying the execution log and the reproduction report, and the pass/fail comparison against the manuscript's headline results — the auditable record the governance claim rests on.

````python
# Synthetic Study Inputs

# ===================================================================================
# Synthetic Study Inputs
# ===================================================================================

# Load the frozen study configuration (single source of truth)
config: Dict[str, Any] = json.loads(
    Path.cwd().joinpath('study_config.json')
    .read_text(encoding='utf-8')
)

# Input corpus: the study's raw frames are ingested parameters; the
# workspace carries no raw CSVs, so the corpus is constructed
# schema-conformantly with a deterministic seed (documented
# placeholder), attempting to load real CSVs from data/ when present
_corpus_rng: np.random.Generator = np.random.default_rng(71)
_comp: Dict[str, int] = {
    'Rwanda': 8735, 'Tanzania': 6620,
    'Kenya': 6068, 'Uganda': 2101,
}
_n_rows: int = 23524
_country: np.ndarray = np.concatenate(
    [[c] * k for c, k in _comp.items()]
)
_levels: Dict[str, List[str]] = config[
    'dataset_parameters']['zindi_fsd']['categorical_levels'
]
_age: np.ndarray = np.round(
    _corpus_rng.normal(40, 14, _n_rows).clip(16, 85)
).astype(np.int64)
_hs: np.ndarray = _corpus_rng.integers(
    1, 12, _n_rows
).astype(np.int64)
_edu: np.ndarray = _corpus_rng.choice(
    ['No formal education', 'Primary education',
     'Secondary education', 'Tertiary education'],
    _n_rows, p=[0.35, 0.3, 0.25, 0.1],
)
_cell: np.ndarray = _corpus_rng.choice(
    ['Yes', 'No'], _n_rows, p=[0.7, 0.3]
)
_job: np.ndarray = _corpus_rng.choice(
    ['Government', 'Formally employed Private sector',
     'Informally employed', 'Self employed',
     'Farming and Fishing', 'Remittance Dependent'],
    _n_rows,
)
_logit: np.ndarray = (
    -2.6 + 0.035 * (_age - 40) + 0.06 * (_hs - 6)
    + 0.9 * (_edu == 'Tertiary education')
    + 0.5 * (_edu == 'Secondary education')
    + 0.6 * (_cell == 'Yes')
    + 0.4 * (_job == 'Government')
    + 0.3 * (_country == 'Kenya')
)
_prob: np.ndarray = 1.0 / (1.0 + np.exp(-_logit))
_n_pos: int = round(0.1408 * _n_rows)
_y: np.ndarray = np.zeros(_n_rows, dtype=np.int64)
_y[np.argsort(-_prob)[:_n_pos]] = 1
zindi_raw: pd.DataFrame = pd.DataFrame({
    'uniqueid': [f'uniqueid_{i:05d}' for i in range(_n_rows)],
    'country': _country,
    'year': np.where(
        _country == 'Kenya', 2018,
        np.where(_country == 'Rwanda', 2016, 2017),
    ),
    'location_type': _corpus_rng.choice(
        _levels['location_type'], _n_rows
    ),
    'cellphone_access': _cell,
    'gender_of_respondent': _corpus_rng.choice(
        _levels['gender_of_respondent'], _n_rows
    ),
    'relationship_with_head': _corpus_rng.choice(
        _levels['relationship_with_head'], _n_rows
    ),
    'marital_status': _corpus_rng.choice(
        _levels['marital_status'], _n_rows
    ),
    'education_level': _edu,
    'job_type': _job,
    'household_size': _hs,
    'age_of_respondent': _age,
    'bank_account': _y,
})
zindi_raw = zindi_raw[config['dataset_parameters']['zindi_fsd']
    ['raw_column_order']]
_tw_cols: List[str] = [
    c['name'] for c in config['raw_data_schemas']
    ['taiwan_credit_schema']['columns']
]
_n2: int = 30000
taiwan_raw: pd.DataFrame = pd.DataFrame(
    np.zeros((_n2, 1), dtype=np.int64),
    columns=['default.payment.next.month'],
)
taiwan_raw.insert(0, 'ID', np.arange(1, _n2 + 1))
taiwan_raw['LIMIT_BAL'] = _corpus_rng.integers(
    10000, 1000000, _n2
)
taiwan_raw['SEX'] = _corpus_rng.choice([1, 2], _n2)
taiwan_raw['EDUCATION'] = _corpus_rng.choice(
    [0, 1, 2, 3, 4, 5, 6], _n2
)
taiwan_raw['MARRIAGE'] = _corpus_rng.choice(
    [0, 1, 2, 3], _n2
)
taiwan_raw['AGE'] = _corpus_rng.integers(21, 79, _n2)
for _i in [0, 2, 3, 4, 5, 6]:
    taiwan_raw[f'PAY_{_i}'] = _corpus_rng.integers(-2, 9, _n2)
for _i in range(1, 7):
    taiwan_raw[f'BILL_AMT{_i}'] = _corpus_rng.integers(
        0, 100000, _n2
    )
    taiwan_raw[f'PAY_AMT{_i}'] = _corpus_rng.integers(
        0, 50000, _n2
    )
taiwan_raw = taiwan_raw[_tw_cols]
_tw_pos: np.ndarray = _corpus_rng.permutation(_n2)[:6600]
taiwan_raw['default.payment.next.month'] = 0
taiwan_raw.loc[_tw_pos, 'default.payment.next.month'] = 1
````

## Core Callables

````python
# Task 1: Validate Ingested Parameters and Data Quality

# ==============================================================================
# Task 1: Validate Ingested Parameters and Data Quality
# ==============================================================================


# -------------------------------------------------------------------------------------------------------------------------------
# Task 1, Step 1: Validate the study_config dictionary
# -------------------------------------------------------------------------------------------------------------------------------

# The frozen tuple of every top-level key required by the manuscript
REQUIRED_TOP_LEVEL_KEYS: Tuple[str, ...] = (
    "study_metadata",
    "prompt_templates",
    "llm_configuration",
    "dataset_parameters",
    "raw_data_schemas",
    "partitioning_parameters",
    "preprocessing_parameters",
    "woe_parameters",
    "iv_parameters",
    "logistic_regression_parameters",
    "residual_xgboost_parameters",
    "standalone_xgboost_parameters",
    "lightgbm_sensitivity_parameters",
    "adaptive_weighting_parameters",
    "hybrid_prediction_parameters",
    "interpretability_parameters",
    "network_training_parameters",
    "calibration_parameters",
    "evaluation_metrics_parameters",
    "protected_attribute_parameters",
    "fairness_audit_parameters",
    "thin_file_parameters",
    "hard_soft_split_parameters",
    "joint_optimization_parameters",
    "bootstrap_inference_parameters",
    "delong_test_parameters",
    "shap_explanation_parameters",
    "training_protocol_parameters",
    "essential_methodological_notes",
    "plot_construction_parameters",
    "equation_parameter_map",
    "code_and_data_availability",
)

# The equation ids that must be present in the equation_parameter_map
REQUIRED_EQUATION_IDS: Tuple[str, ...] = (
    "Eq1",
    "Eq2",
    "Eq3",
    "Eq4",
    "Eq5",
    "Eq6",
    "Eq7",
    "Eq8",
    "Eq9",
    "Eq10",
    "Eq11",
    "Eq12",
    "Unnumbered-A",
    "Unnumbered-B",
    "Unnumbered-C",
    "Unnumbered-D",
    "Unnumbered-E",
)


def _check(
    results: List[Dict[str, Any]],
    name: str,
    ok: bool,
    detail: str,
) -> None:
    """
    Append one structured check record to the results list.
    """
    # Compose the structured check record
    results.append({"check": name, "ok": ok, "detail": detail})
    # Log the check outcome at INFO level
    logger.info(f"{'PASS' if ok else 'FAIL'} {name}: {detail}")


def validate_study_config(
    config: Dict[str, Any],
) -> List[Dict[str, Any]]:
    """
    Validate the study configuration dictionary against its schema.

    Asserts presence of every required top-level key, type contracts,
    value ranges, grid membership, exact hyperparameter pins, non-null
    headline results, and coverage of Equations (1)-(12) plus the five
    unnumbered equations.  Halts on the FIRST failure with a descriptive
    ValueError rather than collecting multiple failures.

    Parameters
    ----------
    config : Dict[str, Any]
        The study configuration dictionary declared by the manuscript
        specification.  Must carry all 32 required top-level keys.

    Returns
    -------
    List[Dict[str, Any]]
        A list of structured check records, each with keys 'check',
        'ok' and 'detail'.

    Raises
    ------
    ValueError
        If any required key is missing, any type or range contract is
        violated, or the equation map is incomplete.
    """
    # Validate the config is a dictionary before any key access
    if not isinstance(config, dict):
        # Raise a descriptive error for a non-dictionary config
        raise ValueError(
            f"config must be a dict; received {type(config).__name__}."
        )
    # Collect the structured check records
    results: List[Dict[str, Any]] = []
    # Iterate over the required top-level keys in declared order
    for key in REQUIRED_TOP_LEVEL_KEYS:
        # Assert the required key is present in the config
        if key not in config:
            # Halt on the first missing key with the key named
            raise ValueError(
                f"study_config missing required top-level key '{key}'. "
                f"Halted on the first failure as prescribed."
            )
        # Record the presence check as passed
        _check(results, f"key_present:{key}", True, "present")
    # Enforce the isinstance type contracts on every pinned scalar
    # before any value check runs, so 400.0 or True cannot satisfy a
    # pin that requires an int (Blueprint 1, step 3)
    _enforce_pin_type_contracts(config)
    # Record the type-contract gate as passed
    _check(results, "pin_type_contracts", True, "isinstance gates passed")
    # Extract the partitioning parameters block through the guarded
    # accessor so a missing nested key raises with the dotted path
    partitioning: Dict[str, Any] = _get_nested(
        config, "partitioning_parameters", dict
    )
    # Sum the three split fractions for the unit identity check
    fraction_sum: float = (
        float(partitioning["train_fraction"])
        + float(partitioning["validation_fraction"])
        + float(partitioning["test_fraction"])
    )
    # Verify the fractions sum to one within machine precision
    if not math.isclose(fraction_sum, 1.0, abs_tol=1e-9):
        # Halt with the received sum named in the error
        raise ValueError(
            f"train + validation + test fractions must sum to 1.0; "
            f"received {fraction_sum}."
        )
    # Record the fraction identity check as passed
    _check(results, "fractions_sum_to_one", True, f"sum={fraction_sum}")
    # Extract the random seed from the partitioning block
    seed: int = int(partitioning["random_seed"])
    # Verify the seed lies in the declared range
    if not (45 <= seed <= 2 ** 31 - 1):
        # Halt with the received seed named in the error
        raise ValueError(
            f"random_seed must satisfy 45 <= seed <= 2^31 - 1; "
            f"received {seed}."
        )
    # Record the seed range check as passed
    _check(results, "random_seed_range", True, f"seed={seed}")
    # Extract the adaptive weighting block through the guarded accessor
    adaptive: Dict[str, Any] = _get_nested(
        config, "adaptive_weighting_parameters", dict
    )
    # Verify the alpha_min grid is a subset of the declared set
    if not set(adaptive["alpha_min_grid"]).issubset({0.0, 0.1, 0.2}):
        # Halt with the received grid named in the error
        raise ValueError(
            f"alpha_min_grid must be a subset of {{0.0, 0.1, 0.2}}; "
            f"received {adaptive['alpha_min_grid']}."
        )
    # Record the alpha_min grid check as passed
    _check(results, "alpha_min_grid", True, str(adaptive["alpha_min_grid"]))
    # Verify the alpha_max grid is a subset of the declared set
    if not set(adaptive["alpha_max_grid"]).issubset({0.3, 0.5, 0.8}):
        # Halt with the received grid named in the error
        raise ValueError(
            f"alpha_max_grid must be a subset of {{0.3, 0.5, 0.8}}; "
            f"received {adaptive['alpha_max_grid']}."
        )
    # Record the alpha_max grid check as passed
    _check(results, "alpha_max_grid", True, str(adaptive["alpha_max_grid"]))
    # Verify the gamma grid is a subset of the declared set
    if not set(adaptive["gamma_grid"]).issubset({0.5, 1.0, 2.0}):
        # Halt with the received grid named in the error
        raise ValueError(
            f"gamma_grid must be a subset of {{0.5, 1.0, 2.0}}; "
            f"received {adaptive['gamma_grid']}."
        )
    # Record the gamma grid check as passed
    _check(results, "gamma_grid", True, str(adaptive["gamma_grid"]))
    # Apply the alpha-bound inequality over the CARTESIAN PRODUCT of
    # the two grids (Blueprint 1, step 4), so every declared pair is
    # covered, not only the selected triple
    for alpha_min in adaptive["alpha_min_grid"]:
        # Iterate the alpha_max grid for the current alpha_min
        for alpha_max in adaptive["alpha_max_grid"]:
            # Assert the pair satisfies 0 <= min < max <= 1
            if not (0.0 <= alpha_min < alpha_max <= 1.0):
                # Halt naming the offending grid pair
                raise ValueError(
                    f"grid pair (alpha_min={alpha_min}, "
                    f"alpha_max={alpha_max}) violates "
                    f"0 <= alpha_min < alpha_max <= 1."
                )
    # Record the grid cross-product bound check as passed
    _check(
        results,
        "alpha_bounds_grid_cross_product",
        True,
        "all 9 grid pairs satisfy 0 <= min < max <= 1",
    )
    # Extract the selected triple through the guarded accessor
    selected: Dict[str, float] = _get_nested(
        config, "adaptive_weighting_parameters.selected_on_zindi", dict
    )
    # Verify the selected triple bounds satisfy 0 <= min < max <= 1
    if not (
        0.0 <= float(selected["alpha_min"])
        < float(selected["alpha_max"])
        <= 1.0
    ):
        # Halt with the received bounds named in the error
        raise ValueError(
            "alpha bounds must satisfy 0 <= alpha_min < alpha_max <= 1; "
            f"received {selected}."
        )
    # Record the selected-triple bound check as passed
    _check(results, "alpha_bounds_selected_triple", True, str(selected))
    # Verify gamma > 0 for every grid value AND the selected gamma
    for gamma in adaptive["gamma_grid"]:
        # Assert every grid gamma is positive
        if not gamma > 0:
            # Halt with the offending gamma named in the error
            raise ValueError(f"gamma must be > 0; received {gamma}.")
    # Assert the selected gamma is positive as well (Blueprint 1)
    if not float(selected["gamma"]) > 0:
        # Halt with the selected gamma named in the error
        raise ValueError(
            f"selected gamma must be > 0; received {selected['gamma']}."
        )
    # Record the gamma positivity checks as passed
    _check(
        results, "gamma_positive", True, "grid and selected gammas > 0"
    )
    # Extract the residual booster block through the guarded accessor
    residual: Dict[str, Any] = _get_nested(
        config, "residual_xgboost_parameters", dict
    )
    # Verify the estimator count pin of 400
    if residual["n_estimators"] != 400:
        # Halt with the received value named in the error
        raise ValueError(
            f"residual n_estimators must be 400; "
            f"received {residual['n_estimators']}."
        )
    # Verify the tree depth pin of 3
    if residual["max_depth"] != 3:
        # Halt with the received value named in the error
        raise ValueError(
            f"residual max_depth must be 3; "
            f"received {residual['max_depth']}."
        )
    # Verify the learning rate pin of 0.03
    if not math.isclose(float(residual["nu"]), 0.03, abs_tol=1e-12):
        # Halt with the received value named in the error
        raise ValueError(
            f"residual learning rate must be 0.03; "
            f"received {residual['nu']}."
        )
    # Verify the subsample pin of 0.8
    if not math.isclose(float(residual["subsample"]), 0.8, abs_tol=1e-12):
        # Halt with the received value named in the error
        raise ValueError(
            f"residual subsample must be 0.8; "
            f"received {residual['subsample']}."
        )
    # Verify the colsample_bytree pin of 0.8
    if not math.isclose(
        float(residual["colsample_bytree"]), 0.8, abs_tol=1e-12
    ):
        # Halt with the received value named in the error
        raise ValueError(
            f"residual colsample_bytree must be 0.8; "
            f"received {residual['colsample_bytree']}."
        )
    # Verify the reg_lambda pin of 2
    if residual["reg_lambda"] != 2:
        # Halt with the received value named in the error
        raise ValueError(
            f"residual reg_lambda must be 2; "
            f"received {residual['reg_lambda']}."
        )
    # Record the residual booster pins as passed
    _check(results, "residual_pins", True, "400/3/0.03/0.8/0.8/2")
    # Extract the bootstrap inference block through the guarded accessor
    bootstrap: Dict[str, Any] = _get_nested(
        config, "bootstrap_inference_parameters", dict
    )
    # Verify the AUC-difference resample count pin of 2000
    if bootstrap["auc_difference_resamples"] != 2000:
        # Halt with the received value named in the error
        raise ValueError(
            f"auc_difference_resamples must be 2000; "
            f"received {bootstrap['auc_difference_resamples']}."
        )
    # Verify the fairness resample count pin of 500
    if bootstrap["fairness_quantity_resamples"] != 500:
        # Halt with the received value named in the error
        raise ValueError(
            f"fairness_quantity_resamples must be 500; "
            f"received {bootstrap['fairness_quantity_resamples']}."
        )
    # Record the bootstrap pins as passed
    _check(results, "bootstrap_pins", True, "2000/500")
    # Extract the calibration block through the guarded accessor
    calibration: Dict[str, Any] = _get_nested(
        config, "calibration_parameters.evaluation", dict
    )
    # Verify the ECE bin count pin of 10
    if calibration["ece_bins"] != 10:
        # Halt with the received value named in the error
        raise ValueError(
            f"ece_bins must be 10; received {calibration['ece_bins']}."
        )
    # Record the ECE bin pin as passed
    _check(results, "ece_bins", True, "10")
    # Extract the fairness audit block through the guarded accessor
    fairness: Dict[str, Any] = _get_nested(
        config, "fairness_audit_parameters", dict
    )
    # Verify the four-fifths reference interval pin
    if fairness["four_fifths_reference_interval"] != [0.8, 1.25]:
        # Halt with the received interval named in the error
        raise ValueError(
            f"four_fifths_reference_interval must be [0.8, 1.25]; "
            f"received {fairness['four_fifths_reference_interval']}."
        )
    # Record the four-fifths pin as passed
    _check(results, "four_fifths_interval", True, "[0.8, 1.25]")
    # Extract the convergence tolerance pin from the joint optimization
    joint: Dict[str, Any] = _get_nested(
        config, "joint_optimization_parameters", dict
    )
    # Verify the convergence tolerance pin of 1e-3
    if not math.isclose(
        float(joint["convergence_tolerance"]), 1e-3, abs_tol=1e-12
    ):
        # Halt with the received tolerance named in the error
        raise ValueError(
            f"convergence_tolerance must be 1e-3; "
            f"received {joint['convergence_tolerance']}."
        )
    # Record the convergence tolerance pin as passed
    _check(results, "convergence_tolerance", True, "1e-3")
    # Extract the headline results block through the guarded accessor
    headline: Dict[str, Any] = _get_nested(
        config, "study_metadata.headline_results", dict
    )
    # Verify every headline result is non-null
    for key, value in headline.items():
        # Assert the headline value is not None
        if value is None:
            # Halt with the null headline named in the error
            raise ValueError(
                f"headline result '{key}' must be non-null; "
                f"received None."
            )
    # Record the headline non-null check as passed
    _check(results, "headline_non_null", True, f"{len(headline)} results")
    # Extract the equation map through the guarded accessor
    equation_map: Dict[str, Any] = _get_nested(
        config, "equation_parameter_map", dict
    )
    # Verify every required equation id is present in the map
    for equation_id in REQUIRED_EQUATION_IDS:
        # Assert the equation id is present in the map
        if equation_id not in equation_map:
            # Halt with the missing equation id named in the error
            raise ValueError(
                f"equation_parameter_map missing required entry "
                f"'{equation_id}'."
            )
    # Record the equation map coverage as passed
    _check(results, "equation_map_coverage", True, "17 entries present")
    # Return the structured validation results
    return results


def _get_nested(
    config: Dict[str, Any], dotted_path: str, expected_type: type
) -> Any:
    """
    Return a nested config value with a guarded dotted-path access.

    Walks the dotted path, raising a descriptive ValueError naming the
    full path and the required repair when a key is missing or the
    value's type does not match, so a malformed config can never escape
    as an uncaught KeyError (Blueprint 1, steps 1-2).

    Parameters
    ----------
    config : Dict[str, Any]
        The study configuration dictionary.
    dotted_path : str
        The dotted key path, e.g. 'calibration_parameters.evaluation'.
    expected_type : type
        The expected Python type of the resolved value.

    Returns
    -------
    Any
        The resolved nested value.

    Raises
    ------
    ValueError
        If any path segment is missing or the value type mismatches.
    """
    # Split the dotted path into its segments
    segments: List[str] = dotted_path.split(".")
    # Start the walk at the config root
    current: Any = config
    # Track the walked prefix for the error messages
    walked: List[str] = []
    # Descend the path segment by segment
    for segment in segments:
        # Append the segment to the walked prefix
        walked.append(segment)
        # Guard against a non-dict intermediate node
        if not isinstance(current, dict) or segment not in current:
            # Raise with the full dotted path and the repair
            raise ValueError(
                f"study_config missing or malformed nested key "
                f"'{'.'.join(walked)}' (declared by the specification). "
                f"Repair the configuration before proceeding."
            )
        # Descend into the next level
        current = current[segment]
    # Enforce the type contract at the resolved leaf
    if not isinstance(current, expected_type):
        # Raise with the received and expected types named
        raise ValueError(
            f"study_config['{dotted_path}'] must be of type "
            f"{expected_type.__name__}; received "
            f"{type(current).__name__} ({current!r})."
        )
    # Return the resolved value
    return current


def _enforce_pin_type_contracts(config: Dict[str, Any]) -> None:
    """
    Enforce the isinstance type contracts on every pinned scalar.

    Asserts int for n_estimators, max_depth, random_seed, resample
    counts and ece_bins; float-or-int for nu, subsample,
    colsample_bytree, reg_lambda and the convergence tolerance; and
    list for the four-fifths interval, so a float or bool can never
    satisfy an integer pin (Blueprint 1, step 3).

    Parameters
    ----------
    config : Dict[str, Any]
        The study configuration dictionary.

    Raises
    ------
    ValueError
        If any pinned scalar violates its type contract.
    """
    # Declare the type contract for every pinned scalar by dotted path
    type_spec: Dict[str, Tuple[type, ...]] = {
        "partitioning_parameters.random_seed": (int,),
        "residual_xgboost_parameters.n_estimators": (int,),
        "residual_xgboost_parameters.max_depth": (int,),
        "residual_xgboost_parameters.nu": (float, int),
        "residual_xgboost_parameters.subsample": (float, int),
        "residual_xgboost_parameters.colsample_bytree": (float, int),
        "residual_xgboost_parameters.reg_lambda": (int, float),
        "bootstrap_inference_parameters.auc_difference_resamples": (int,),
        "bootstrap_inference_parameters.fairness_quantity_resamples": (
            int,
        ),
        "calibration_parameters.evaluation.ece_bins": (int,),
        "fairness_audit_parameters.four_fifths_reference_interval": (
            list,
        ),
        "joint_optimization_parameters.convergence_tolerance": (
            float,
            int,
        ),
    }
    # Enforce each pinned scalar's type contract
    for dotted_path, allowed_types in type_spec.items():
        # Resolve the value through the guarded accessor
        value: Any = _get_nested(config, dotted_path, object)
        # Assert the value's type is in the allowed tuple; note that
        # bool is a subclass of int, so it must be excluded explicitly
        if not isinstance(value, allowed_types) or isinstance(value, bool):
            # Raise naming the path, the allowed types and the received
            raise ValueError(
                f"study_config['{dotted_path}'] must be of type "
                f"{[t.__name__ for t in allowed_types]}; received "
                f"{type(value).__name__} ({value!r})."
            )


# -------------------------------------------------------------------------------------------------------------------------------
# Task 1, Step 2: Validate the zindi_raw DataFrame
# -------------------------------------------------------------------------------------------------------------------------------


def validate_zindi_raw(
    zindi_raw: pd.DataFrame, config: Dict[str, Any]
) -> Tuple[pd.DataFrame, List[Dict[str, Any]]]:
    """
    Validate the raw Zindi FSD DataFrame against its declared schema.

    Asserts shape (23524, 13), ordered columns against raw_column_order
    (reordering when the set matches but the order differs), the label
    column identity and dtype, the positive-class rate 0.1408 within
    1e-4, zero missing values, the exact country composition, per-column
    categorical level subsets (the typo 'Divorced/Seperated' preserved
    verbatim), and the declared dtypes.

    Parameters
    ----------
    zindi_raw : pd.DataFrame
        The raw Zindi Financial Inclusion in Africa training frame,
        shape (23524, 13).
    config : Dict[str, Any]
        The validated study configuration dictionary.

    Returns
    -------
    Tuple[pd.DataFrame, List[Dict[str, Any]]]
        The (possibly reordered) zindi frame and the structured check
        records.

    Raises
    ------
    ValueError
        If any structural, distributional or level contract is violated.
    """
    # Validate the frame type before any pandas operation
    if not isinstance(zindi_raw, pd.DataFrame):
        # Raise a descriptive error for a non-DataFrame input
        raise ValueError(
            f"zindi_raw must be a pd.DataFrame; "
            f"received {type(zindi_raw).__name__}."
        )
    # Collect the structured check records
    results: List[Dict[str, Any]] = []
    # Extract the zindi schema block from the configuration
    schema: Dict[str, Any] = config["dataset_parameters"]["zindi_fsd"]
    # Verify the frame shape is exactly (23524, 13)
    if zindi_raw.shape != (23524, 13):
        # Halt with the received shape named in the error
        raise ValueError(
            f"zindi_raw shape must be (23524, 13); "
            f"received {zindi_raw.shape}."
        )
    # Record the shape check as passed
    _check(results, "shape", True, str(zindi_raw.shape))
    # Extract the declared raw column order from the schema
    declared_order: List[str] = schema["raw_column_order"]
    # Compare the frame's column set against the declared set
    if set(zindi_raw.columns) != set(declared_order):
        # Halt naming the differing columns
        raise ValueError(
            f"zindi_raw column set does not match the declared set; "
            f"missing={set(declared_order) - set(zindi_raw.columns)}, "
            f"extra={set(zindi_raw.columns) - set(declared_order)}."
        )
    # Compare the frame's column order against the declared order
    if list(zindi_raw.columns) != declared_order:
        # Record the reorder decision in the manifest
        logger.info(
            "zindi_raw column order differs; "
            "reordering to declared order"
        )
        # Reorder the frame to the declared column order
        zindi_raw = zindi_raw[declared_order].copy()
        # Record the reorder event as a passed check with a note
        _check(results, "column_order", True, "reordered to declared order")
    else:
        # Record the order check as passed without reordering
        _check(results, "column_order", True, "matches declared order")
    # Extract the label column name from the schema
    label: str = schema["label_column"]
    # Verify the label column is present in the frame
    if label not in zindi_raw.columns:
        # Halt naming the missing label column
        raise ValueError(
            f"zindi_raw missing label column '{label}'."
        )
    # Verify the label column has integer dtype
    if not pd.api.types.is_integer_dtype(zindi_raw[label]):
        # Halt with the received dtype named in the error
        raise ValueError(
            f"zindi_raw['{label}'] must have integer dtype; "
            f"received {zindi_raw[label].dtype}."
        )
    # Record the label identity and dtype check as passed
    _check(results, "label_column", True, f"{label} int")
    # Compute the marginal positive-class rate as the label mean
    positive_rate: float = float(zindi_raw[label].mean())
    # Verify the positive rate equals 0.1408 within 1e-4
    if not math.isclose(positive_rate, 0.1408, abs_tol=1e-4):
        # Halt with the received rate named in the error
        raise ValueError(
            f"zindi positive-class rate must be 0.1408 within 1e-4; "
            f"received {positive_rate:.6f}."
        )
    # Record the positive rate check as passed
    _check(results, "positive_rate", True, f"{positive_rate:.6f}")
    # Verify zero missing values in every column
    missing: pd.Series = zindi_raw.isnull().sum()
    # Assert the total missing count across all columns is zero
    if int(missing.sum()) != 0:
        # Halt with the missing counts named in the error
        raise ValueError(
            f"zindi_raw must have zero missing values; "
            f"received {missing.to_dict()}."
        )
    # Record the missingness check as passed
    _check(results, "missing_values", True, "zero missing")
    # Extract the declared country composition from the schema
    declared_composition: Dict[str, int] = schema["country_composition"]
    # Compute the observed country composition
    observed_composition: Dict[str, int] = (
        zindi_raw["country"].value_counts().to_dict()
    )
    # Verify the composition matches exactly
    if observed_composition != declared_composition:
        # Halt naming the composition discrepancy
        raise ValueError(
            f"zindi country composition must be {declared_composition}; "
            f"received {observed_composition}."
        )
    # Record the country composition check as passed
    _check(results, "country_composition", True, str(declared_composition))
    # Extract the declared categorical level sets from the schema
    levels: Dict[str, List[str]] = schema["categorical_levels"]
    # Iterate over the categorical columns to validate their levels
    for column, declared_levels in levels.items():
        # Verify the column exists in the frame
        if column not in zindi_raw.columns:
            # Halt naming the missing categorical column
            raise ValueError(
                f"zindi_raw missing categorical column '{column}'."
            )
        # Compute the set of unique values observed in the column
        observed_levels: set = set(zindi_raw[column].unique())
        # Verify the observed levels are a subset of the declared set
        if not observed_levels.issubset(set(declared_levels)):
            # Halt naming the undeclared levels
            raise ValueError(
                f"zindi_raw['{column}'] has undeclared levels "
                f"{observed_levels - set(declared_levels)}."
            )
        # Record the level check for the column as passed
        _check(
            results,
            f"levels:{column}",
            True,
            f"{len(observed_levels)} levels",
        )
    # Verify the literal typo 'Divorced/Seperated' is present verbatim
    if "Divorced/Seperated" not in set(zindi_raw["marital_status"].unique()):
        # Halt because the typo is a structural invariant of the raw file
        raise ValueError(
            "zindi_raw['marital_status'] must contain the verbatim typo "
            "'Divorced/Seperated'; it must never be auto-corrected."
        )
    # Record the typo presence check as passed
    _check(results, "typo_preserved", True, "'Divorced/Seperated' present")
    # Verify the integer dtype contract on the numeric columns
    for column in ("household_size", "age_of_respondent", "year"):
        # Assert the numeric column has integer dtype
        if not pd.api.types.is_integer_dtype(zindi_raw[column]):
            # Halt with the received dtype named in the error
            raise ValueError(
                f"zindi_raw['{column}'] must have integer dtype; "
                f"received {zindi_raw[column].dtype}."
            )
    # Record the integer dtype check as passed
    _check(results, "integer_dtypes", True, "household/age/year int64")
    # Verify the object dtype contract on the categorical columns
    categorical_columns: List[str] = [
        "country",
        "location_type",
        "cellphone_access",
        "gender_of_respondent",
        "relationship_with_head",
        "marital_status",
        "education_level",
        "job_type",
    ]
    # Iterate over the categorical columns to validate their dtypes
    for column in categorical_columns:
        # Assert the categorical column has object dtype
        if not pd.api.types.is_object_dtype(zindi_raw[column]):
            # Halt with the received dtype named in the error
            raise ValueError(
                f"zindi_raw['{column}'] must have object dtype; "
                f"received {zindi_raw[column].dtype}."
            )
    # Record the object dtype check as passed
    _check(results, "object_dtypes", True, "8 categorical columns object")
    # Return the possibly reordered frame and the check records
    return zindi_raw, results


# -------------------------------------------------------------------------------------------------------------------------------
# Task 1, Step 3: Validate the taiwan_raw DataFrame
# -------------------------------------------------------------------------------------------------------------------------------


def validate_taiwan_raw(
    taiwan_raw: pd.DataFrame, config: Dict[str, Any]
) -> Tuple[pd.DataFrame, List[Dict[str, Any]]]:
    """
    Validate the raw Taiwan Credit Default DataFrame against its schema.

    Asserts shape (30000, 25), the ordered 25-column layout, the label
    column identity, the positive-class rate approximately 0.22 within
    1e-2, the absence of PAY_1 and presence of PAY_0, and the categorical
    code sets for SEX, EDUCATION and MARRIAGE.

    Parameters
    ----------
    taiwan_raw : pd.DataFrame
        The raw UCI Taiwan credit default frame, shape (30000, 25).
    config : Dict[str, Any]
        The validated study configuration dictionary.

    Returns
    -------
    Tuple[pd.DataFrame, List[Dict[str, Any]]]
        The validated taiwan frame and the structured check records.

    Raises
    ------
    ValueError
        If any structural or code-set contract is violated.
    """
    # Validate the frame type before any pandas operation
    if not isinstance(taiwan_raw, pd.DataFrame):
        # Raise a descriptive error for a non-DataFrame input
        raise ValueError(
            f"taiwan_raw must be a pd.DataFrame; "
            f"received {type(taiwan_raw).__name__}."
        )
    # Collect the structured check records
    results: List[Dict[str, Any]] = []
    # Extract the taiwan schema block from the configuration
    schema: Dict[str, Any] = config["raw_data_schemas"]["taiwan_credit_schema"]
    # Verify the frame shape is exactly (30000, 25)
    if taiwan_raw.shape != (30000, 25):
        # Halt with the received shape named in the error
        raise ValueError(
            f"taiwan_raw shape must be (30000, 25); "
            f"received {taiwan_raw.shape}."
        )
    # Record the shape check as passed
    _check(results, "shape", True, str(taiwan_raw.shape))
    # Extract the declared ordered column names from the schema
    declared_columns: List[str] = [c["name"] for c in schema["columns"]]
    # Verify the ordered column names match the schema exactly
    if list(taiwan_raw.columns) != declared_columns:
        # Halt naming the column-order discrepancy
        raise ValueError(
            f"taiwan_raw column order must match the declared schema; "
            f"received {list(taiwan_raw.columns)}."
        )
    # Record the column order check as passed
    _check(results, "column_order", True, "matches schema")
    # Extract the label column name from the schema metadata so the
    # single source of truth is the configuration, never a literal
    label: str = config["raw_data_schemas"]["taiwan_credit_schema"][
        "columns"
    ][-1]["name"]
    # Verify the label column is present in the frame
    if label not in taiwan_raw.columns:
        # Halt naming the missing label column
        raise ValueError(f"taiwan_raw missing label column '{label}'.")
    # Record the label identity check as passed
    _check(results, "label_column", True, label)
    # Compute the marginal positive-class rate as the label mean
    positive_rate: float = float(taiwan_raw[label].mean())
    # Verify the positive rate is approximately 0.22 within 1e-2
    if not math.isclose(positive_rate, 0.22, abs_tol=1e-2):
        # Halt with the received rate named in the error
        raise ValueError(
            f"taiwan positive-class rate must be approx 0.22 within 1e-2; "
            f"received {positive_rate:.6f}."
        )
    # Record the positive rate check as passed
    _check(results, "positive_rate", True, f"{positive_rate:.6f}")
    # Verify zero missing values across all 25 columns
    missing: pd.Series = taiwan_raw.isnull().sum()
    # Assert the total missing count across all columns is zero
    if int(missing.sum()) != 0:
        # Halt with the missing counts named in the error
        raise ValueError(
            f"taiwan_raw must have zero missing values; "
            f"received {missing.to_dict()}."
        )
    # Record the missingness check as passed
    _check(results, "missing_values", True, "zero missing")
    # Verify the raw UCI schema has PAY_0 but no PAY_1
    if "PAY_0" not in taiwan_raw.columns or "PAY_1" in taiwan_raw.columns:
        # Halt naming the PAY schema violation
        raise ValueError(
            "taiwan_raw must contain PAY_0 and must NOT contain PAY_1; "
            "the raw UCI schema skips PAY_1."
        )
    # Record the PAY schema check as passed
    _check(results, "pay_schema", True, "PAY_0 present, PAY_1 absent")
    # Verify the SEX code set is a subset of {1, 2}
    if not set(taiwan_raw["SEX"].unique()).issubset({1, 2}):
        # Halt with the offending SEX codes named in the error
        raise ValueError(
            f"SEX codes must be a subset of {{1, 2}}; "
            f"received {sorted(set(taiwan_raw['SEX'].unique()))}."
        )
    # Record the SEX code check as passed
    _check(results, "sex_codes", True, "subset of {1, 2}")
    # Verify the EDUCATION code set is a subset of {0..6}
    if not set(taiwan_raw["EDUCATION"].unique()).issubset(
        {0, 1, 2, 3, 4, 5, 6}
    ):
        # Halt with the offending EDUCATION codes named in the error
        raise ValueError(
            f"EDUCATION codes must be a subset of {{0..6}}; "
            f"received {sorted(set(taiwan_raw['EDUCATION'].unique()))}."
        )
    # Record the EDUCATION code check as passed
    _check(results, "education_codes", True, "subset of {0..6}")
    # Verify the MARRIAGE code set is a subset of {0, 1, 2, 3}
    if not set(taiwan_raw["MARRIAGE"].unique()).issubset({0, 1, 2, 3}):
        # Halt with the offending MARRIAGE codes named in the error
        raise ValueError(
            f"MARRIAGE codes must be a subset of {{0, 1, 2, 3}}; "
            f"received {sorted(set(taiwan_raw['MARRIAGE'].unique()))}."
        )
    # Record the MARRIAGE code check as passed
    _check(results, "marriage_codes", True, "subset of {0, 1, 2, 3}")
    # Record the inherited-pipeline caveat in the manifest
    _check(
        results,
        "preprocessing_inherited",
        True,
        "Taiwan preprocessing inherited from Kanziga (2026); "
        "placeholder protocol will be used if thesis code is unavailable",
    )
    # Return the validated frame and the check records
    return taiwan_raw, results


# -------------------------------------------------------------------------------------------------------------------------------
# Task 1, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


def orchestrate_validate_ingested_parameters(
    config: Dict[str, Any],
    zindi_raw: pd.DataFrame,
    taiwan_raw: pd.DataFrame,
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Orchestrate the validation of the ingested parameters and data.

    Composes validate_study_config, validate_zindi_raw and
    validate_taiwan_raw in dependency order, persists the combined
    validation report to a JSON manifest in the output directory, and
    returns the structured report.  Raises on the first failed check.

    Parameters
    ----------
    config : Dict[str, Any]
        The study configuration dictionary to validate.
    zindi_raw : pd.DataFrame
        The raw Zindi FSD frame, shape (23524, 13).
    taiwan_raw : pd.DataFrame
        The raw Taiwan credit default frame, shape (30000, 25).
    output_dir : Path
        Directory where the validation manifest is persisted.

    Returns
    -------
    Dict[str, Any]
        A report with keys 'config', 'zindi', 'taiwan', 'zindi_frame'
        and 'manifest_path'.

    Raises
    ------
    ValueError
        If any validation check fails.
    """
    # Validate the output directory type before any filesystem work
    if not isinstance(output_dir, Path):
        # Raise a descriptive error for a non-Path output directory
        raise ValueError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Create the output directory including any missing parents
    output_dir.mkdir(parents=True, exist_ok=True)
    # Log the start of the validation orchestration
    logger.info("Task 1: validating ingested parameters and data quality")
    # Step 1: validate the study configuration dictionary
    config_results: List[Dict[str, Any]] = validate_study_config(config)
    # Step 2: validate the raw Zindi frame, possibly reordering it
    zindi_frame, zindi_results = validate_zindi_raw(zindi_raw, config)
    # Step 3: validate the raw Taiwan frame
    taiwan_frame, taiwan_results = validate_taiwan_raw(taiwan_raw, config)
    # Compose the combined validation report
    report: Dict[str, Any] = {
        "config": config_results,
        "zindi": zindi_results,
        "taiwan": taiwan_results,
    }
    # Build the manifest path inside the output directory
    manifest_path: Path = output_dir / "validation_report.json"
    # Serialise the report to a JSON manifest with a stable layout
    manifest_text: str = json.dumps(report, indent=2, ensure_ascii=False)
    # Persist the manifest to the output directory
    manifest_path.write_text(manifest_text, encoding="utf-8")
    # Log the successful persistence of the manifest
    logger.info(f"validation manifest written to {manifest_path}")
    # Return the full report including the validated frames
    return {
        "config": config_results,
        "zindi": zindi_results,
        "taiwan": taiwan_results,
        "zindi_frame": zindi_frame,
        "taiwan_frame": taiwan_frame,
        "manifest_path": str(manifest_path),
    }
````

````python
# Task 2: Cleanse the Data

# ==============================================================================
# Task 2: Cleanse the Data
# ==============================================================================


# -------------------------------------------------------------------------------------------------------------------------------
# Task 2, Step 1: Drop non-predictive columns from zindi_raw
# -------------------------------------------------------------------------------------------------------------------------------


def drop_non_predictive_columns(
    zindi_raw: pd.DataFrame, config: Dict[str, Any]
) -> pd.DataFrame:
    """
    Drop the non-predictive columns from the raw Zindi frame.

    Drops exactly the columns named in
    config["dataset_parameters"]["zindi_fsd"]["dropped_columns"]
    (uniqueid and year) from a COPY of the raw frame, then asserts that
    the remaining column set is exactly the ten feature columns plus the
    label.  The raw frame is never mutated.

    Parameters
    ----------
    zindi_raw : pd.DataFrame
        The validated raw Zindi frame, shape (23524, 13).
    config : Dict[str, Any]
        The validated study configuration dictionary.

    Returns
    -------
    pd.DataFrame
        The cleansed frame, shape (23524, 11), with the two
        non-predictive columns removed and row order preserved.

    Raises
    ------
    ValueError
        If a drop column is missing or the remaining column set is not
        exactly the declared feature set plus the label.
    """
    # Validate the frame type before any pandas operation
    if not isinstance(zindi_raw, pd.DataFrame):
        # Raise a descriptive error for a non-DataFrame input
        raise ValueError(
            f"zindi_raw must be a pd.DataFrame; "
            f"received {type(zindi_raw).__name__}."
        )
    # Extract the declared dropped columns from the configuration
    dropped: List[str] = config["dataset_parameters"]["zindi_fsd"][
        "dropped_columns"
    ]
    # Verify every declared drop column exists in the raw frame
    for column in dropped:
        # Assert the drop column is present before dropping it
        if column not in zindi_raw.columns:
            # Raise a descriptive error naming the missing column
            raise ValueError(
                f"cannot drop '{column}': column absent from zindi_raw."
            )
    # Take an explicit copy so the raw frame is never mutated
    cleansed: pd.DataFrame = zindi_raw.copy()
    # Drop the declared columns by name along the column axis
    cleansed = cleansed.drop(columns=dropped, axis=1)
    # Declare the exact expected remaining column set
    expected_columns: set = {
        "country",
        "location_type",
        "cellphone_access",
        "gender_of_respondent",
        "relationship_with_head",
        "marital_status",
        "education_level",
        "job_type",
        "household_size",
        "age_of_respondent",
        "bank_account",
    }
    # Verify the remaining column set matches the declared set exactly
    if set(cleansed.columns) != expected_columns:
        # Raise a descriptive error naming the set discrepancy
        raise ValueError(
            f"remaining column set must be {sorted(expected_columns)}; "
            f"received {sorted(cleansed.columns)}."
        )
    # Record the drop decision and rationale in the audit trail
    logger.info(
        f"dropped columns {dropped}: uniqueid is a unique-per-row "
        f"identifier; year is perfectly collinear with country"
    )
    # Return the cleansed frame with row order preserved
    return cleansed


# -------------------------------------------------------------------------------------------------------------------------------
# Task 2, Step 2: Standardize string categorical values
# -------------------------------------------------------------------------------------------------------------------------------


def standardize_categorical_values(
    cleansed: pd.DataFrame, config: Dict[str, Any]
) -> Tuple[pd.DataFrame, Dict[str, List[str]]]:
    """
    Strip whitespace from object columns and re-validate all levels.

    Applies Series.str.strip to every object-typed column, preserving
    internal case, internal spacing and the verbatim typo
    'Divorced/Seperated'; then re-validates every unique value against
    the declared exhaustive level sets and collects any unseen level
    into a registry keyed by column name.

    Parameters
    ----------
    cleansed : pd.DataFrame
        The cleansed Zindi frame, shape (23524, 11).
    config : Dict[str, Any]
        The validated study configuration dictionary.

    Returns
    -------
    Tuple[pd.DataFrame, Dict[str, List[str]]]
        The cleansed frame with stripped values and the unseen-levels
        registry keyed by column name.

    Raises
    ------
    ValueError
        If the frame type is invalid.
    """
    # Validate the frame type before any pandas operation
    if not isinstance(cleansed, pd.DataFrame):
        # Raise a descriptive error for a non-DataFrame input
        raise ValueError(
            f"cleansed must be a pd.DataFrame; "
            f"received {type(cleansed).__name__}."
        )
    # Take a working copy so the caller's frame is not mutated
    stripped: pd.DataFrame = cleansed.copy()
    # Select the object-typed columns for whitespace stripping
    object_columns: List[str] = [
        column
        for column in stripped.columns
        if pd.api.types.is_object_dtype(stripped[column])
    ]
    # Strip leading and trailing whitespace from every object column
    for column in object_columns:
        # Apply the vectorised strip to the column's string values
        stripped[column] = stripped[column].str.strip()
    # Initialise the unseen-levels registry keyed by column name
    unseen_levels: Dict[str, List[str]] = {}
    # Extract the declared exhaustive level sets from the configuration
    levels: Dict[str, List[str]] = config["dataset_parameters"]["zindi_fsd"][
        "categorical_levels"
    ]
    # Re-validate every categorical column against its declared levels
    for column, declared_levels in levels.items():
        # Compute the observed unique values after stripping
        observed: set = set(stripped[column].unique())
        # Compute the unseen values relative to the declared set
        unseen: set = observed - set(declared_levels)
        # Record any unseen levels in the registry
        if unseen:
            # Store the sorted unseen values under the column name
            unseen_levels[column] = sorted(unseen)
            # Log the unseen levels for the audit trail
            logger.warning(f"unseen levels in {column}: {sorted(unseen)}")
    # Verify the typo survives stripping untouched
    if "Divorced/Seperated" not in set(stripped["marital_status"].unique()):
        # Raise because the typo is a structural invariant
        raise ValueError(
            "the verbatim typo 'Divorced/Seperated' must survive "
            "whitespace stripping; it must never be corrected."
        )
    # Verify the three binary categorical columns explicitly
    binary_checks: Dict[str, set] = {
        "gender_of_respondent": {"Male", "Female"},
        "location_type": {"Urban", "Rural"},
        "cellphone_access": {"Yes", "No"},
    }
    # Iterate over the binary columns to enforce their exact level sets
    for column, expected in binary_checks.items():
        # Assert the observed level set equals the expected set exactly
        if set(stripped[column].unique()) != expected:
            # Raise a descriptive error naming the offending column
            raise ValueError(
                f"{column} levels must be exactly {expected}; "
                f"received {set(stripped[column].unique())}."
            )
    # Return the stripped frame and the unseen-levels registry
    return stripped, unseen_levels


# -------------------------------------------------------------------------------------------------------------------------------
# Task 2, Step 3: Enforce dtype contracts and create the cleansing audit log
# -------------------------------------------------------------------------------------------------------------------------------


def _frame_digest(frame: pd.DataFrame) -> str:
    """
    Return a deterministic SHA-256 digest of a DataFrame.
    """
    # Serialise the frame to CSV bytes for a stable digest
    csv_bytes: bytes = frame.to_csv(index=False).encode("utf-8")
    # Hash the serialised bytes and return the hex digest
    return hashlib.sha256(csv_bytes).hexdigest()


def enforce_dtype_contracts(
    zindi_cleansed: pd.DataFrame,
    taiwan_raw: pd.DataFrame,
    output_dir: Path,
    unseen_levels: Optional[Dict[str, List[str]]] = None,
    config: Optional[Dict[str, Any]] = None,
) -> Tuple[pd.DataFrame, pd.DataFrame, Dict[str, Any]]:
    """
    Enforce integer dtype contracts and persist the cleansing audit.

    Casts the Zindi integer columns and all Taiwan numeric columns to
    int64, drops the Taiwan ID column, asserts the 24-column Taiwan
    frame, and persists the cleansed frames plus a cleansing audit
    record (dropped columns, structured level-validation results,
    SHA-256 digests before and after, timestamp) to the output
    directory before any fitting (Blueprint 2).

    Parameters
    ----------
    zindi_cleansed : pd.DataFrame
        The stripped Zindi frame, shape (23524, 11).
    taiwan_raw : pd.DataFrame
        The validated Taiwan raw frame, shape (30000, 25).
    output_dir : Path
        Directory where the cleansed artifacts are persisted.
    unseen_levels : Optional[Dict[str, List[str]]], optional
        The unseen-levels registry produced by
        standardize_categorical_values, keyed by column name.
    config : Optional[Dict[str, Any]], optional
        The study configuration dictionary supplying the declared
        categorical level sets for the structured audit block.

    Returns
    -------
    Tuple[pd.DataFrame, pd.DataFrame, Dict[str, Any]]
        The dtype-enforced Zindi frame, the Taiwan frame with ID
        dropped, and the cleansing audit record.
    """
    # Create the output directory including any missing parents
    output_dir.mkdir(parents=True, exist_ok=True)
    # Capture the pre-cast digests of both frames for the audit record
    zindi_before: str = _frame_digest(zindi_cleansed)
    taiwan_before: str = _frame_digest(taiwan_raw)
    # Copy the Zindi frame before applying any cast
    zindi_cast: pd.DataFrame = zindi_cleansed.copy()
    # Cast the declared Zindi integer columns to int64
    for column in ("household_size", "age_of_respondent", "bank_account"):
        # Apply the integer cast to the column
        zindi_cast[column] = zindi_cast[column].astype("int64")
        # Assert the cast produced an integer dtype
        if not pd.api.types.is_integer_dtype(zindi_cast[column]):
            # Raise a descriptive error for a failed cast
            raise ValueError(
                f"cast of zindi column '{column}' to int64 failed; "
                f"received {zindi_cast[column].dtype}."
            )
    # Copy the Taiwan frame before applying any cast
    taiwan_cast: pd.DataFrame = taiwan_raw.copy()
    # Drop the non-predictive ID column from the Taiwan frame
    taiwan_cast = taiwan_cast.drop(columns=["ID"], axis=1)
    # Assert the Taiwan frame now carries exactly 24 columns
    if taiwan_cast.shape[1] != 24:
        # Raise a descriptive error for the wrong column count
        raise ValueError(
            f"Taiwan frame must have 24 columns after ID drop; "
            f"received {taiwan_cast.shape[1]}."
        )
    # Cast every remaining Taiwan column to int64
    for column in taiwan_cast.columns:
        # Apply the integer cast to the column
        taiwan_cast[column] = taiwan_cast[column].astype("int64")
        # Assert the cast produced an integer dtype
        if not pd.api.types.is_integer_dtype(taiwan_cast[column]):
            # Raise a descriptive error for a failed cast
            raise ValueError(
                f"cast of taiwan column '{column}' to int64 failed; "
                f"received {taiwan_cast[column].dtype}."
            )
    # Capture the post-cast digests of both frames for the audit record
    zindi_after: str = _frame_digest(zindi_cast)
    taiwan_after: str = _frame_digest(taiwan_cast)
    # Build the structured level-validation block (Blueprint 2, step 2):
    # per-column observed levels, declared levels, unseen values and the
    # pass/fail outcome of the binary and typo checks
    level_validation: Dict[str, Any] = _build_level_validation(
        zindi_cast, config, unseen_levels
    )
    # Compose the cleansing audit record with all five components
    audit_record: Dict[str, Any] = {
        "dropped_columns": {
            "zindi": ["uniqueid", "year"],
            "taiwan": ["ID"],
        },
        "level_validation": level_validation,
        "hashes": {
            "zindi_before": zindi_before,
            "zindi_after": zindi_after,
            "taiwan_before": taiwan_before,
            "taiwan_after": taiwan_after,
        },
        "timestamp": datetime.now(timezone.utc).isoformat(),
    }
    # Build the cleansed output directory path
    cleansed_dir: Path = output_dir / "cleansed"
    # Create the cleansed output directory including parents
    cleansed_dir.mkdir(parents=True, exist_ok=True)
    # Persist the cleansed Zindi frame as CSV before any fitting
    zindi_path: Path = cleansed_dir / "zindi_cleansed.csv"
    # Write the Zindi frame to the CSV artifact
    zindi_cast.to_csv(zindi_path, index=False)
    # Persist the cleansed Taiwan frame as CSV before any fitting
    taiwan_path: Path = cleansed_dir / "taiwan_cleansed.csv"
    # Write the Taiwan frame to the CSV artifact
    taiwan_cast.to_csv(taiwan_path, index=False)
    # Build the audit record path inside the output directory
    audit_path: Path = cleansed_dir / "cleansing_audit.json"
    # Serialise the audit record to JSON with a stable layout
    audit_text: str = json.dumps(audit_record, indent=2, ensure_ascii=False)
    # Persist the audit record to the output directory
    audit_path.write_text(audit_text, encoding="utf-8")
    # Log the persistence of the cleansed artifacts
    logger.info(f"cleansed frames and audit persisted to {cleansed_dir}")
    # Return the dtype-enforced frames and the audit record
    return zindi_cast, taiwan_cast, audit_record


def _build_level_validation(
    zindi_cleansed: pd.DataFrame,
    config: Optional[Dict[str, Any]],
    unseen_levels: Optional[Dict[str, List[str]]],
) -> Dict[str, Any]:
    """
    Build the structured level-validation block of the audit record.

    Records, per categorical column, the observed level set, the
    declared level set (from the configuration when supplied), the
    unseen values (empty when none), the pass/fail outcome of the
    exact-set checks on the three binary columns, and the verbatim
    typo-preservation outcome (Blueprint 2, step 2).

    Parameters
    ----------
    zindi_cleansed : pd.DataFrame
        The cleansed Zindi frame.
    config : Optional[Dict[str, Any]]
        The study configuration dictionary, or None.
    unseen_levels : Optional[Dict[str, List[str]]]
        The unseen-levels registry, or None.

    Returns
    -------
    Dict[str, Any]
        The structured level-validation block.
    """
    # Declare the categorical columns of the Zindi frame
    categorical_columns: Tuple[str, ...] = (
        "country",
        "location_type",
        "cellphone_access",
        "gender_of_respondent",
        "relationship_with_head",
        "marital_status",
        "education_level",
        "job_type",
    )
    # Extract the declared level sets from the configuration when given
    declared_sets: Dict[str, List[str]] = {}
    # Read the declared levels only when the config is supplied
    if config is not None:
        # Extract the declared categorical level sets
        declared_sets = config["dataset_parameters"]["zindi_fsd"][
            "categorical_levels"
        ]
    # Normalise the unseen registry to an empty dict when absent
    unseen: Dict[str, List[str]] = unseen_levels or {}
    # Collect the per-column validation records
    columns: Dict[str, Dict[str, Any]] = {}
    # Iterate over the categorical columns
    for column in categorical_columns:
        # Compute the observed level set of the column
        observed: List[str] = sorted(
            zindi_cleansed[column].unique().tolist()
        )
        # Look up the declared level set for the column
        declared: List[str] = declared_sets.get(column, [])
        # Look up the unseen values for the column
        unseen_values: List[str] = unseen.get(column, [])
        # Record the column's structured validation entry
        columns[column] = {
            "observed_levels": observed,
            "declared_levels": declared,
            "unseen_values": unseen_values,
            "pass": len(unseen_values) == 0,
        }
    # Evaluate the exact-set checks on the three binary columns
    binary_checks: Dict[str, bool] = {
        "gender_of_respondent": (
            set(zindi_cleansed["gender_of_respondent"].unique())
            == {"Male", "Female"}
        ),
        "location_type": (
            set(zindi_cleansed["location_type"].unique())
            == {"Urban", "Rural"}
        ),
        "cellphone_access": (
            set(zindi_cleansed["cellphone_access"].unique())
            == {"Yes", "No"}
        ),
    }
    # Evaluate the verbatim typo-preservation outcome
    typo_preserved: bool = (
        "Divorced/Seperated"
        in set(zindi_cleansed["marital_status"].unique())
    )
    # Compose the structured level-validation block
    return {
        "columns": columns,
        "binary_column_checks": binary_checks,
        "typo_preserved_verbatim": typo_preserved,
        "typo_string": "Divorced/Seperated",
    }


# -------------------------------------------------------------------------------------------------------------------------------
# Task 2, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


def orchestrate_cleanse_data(
    config: Dict[str, Any],
    zindi_raw: pd.DataFrame,
    taiwan_raw: pd.DataFrame,
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Orchestrate the data cleansing stage.

    Composes drop_non_predictive_columns, standardize_categorical_values
    and enforce_dtype_contracts, persisting the cleansed frames and the
    cleansing audit record before any fitting occurs.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated study configuration dictionary.
    zindi_raw : pd.DataFrame
        The validated raw Zindi frame, shape (23524, 13).
    taiwan_raw : pd.DataFrame
        The validated raw Taiwan frame, shape (30000, 25).
    output_dir : Path
        Directory where the cleansed artifacts are persisted.

    Returns
    -------
    Dict[str, Any]
        A state-ready dict with keys 'zindi', 'taiwan',
        'unseen_levels', 'audit_record' and 'output_dir'.
    """
    # Validate the output directory type before any filesystem work
    if not isinstance(output_dir, Path):
        # Raise a descriptive error for a non-Path output directory
        raise ValueError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Step 1: drop the non-predictive columns from the raw Zindi frame
    zindi_dropped: pd.DataFrame = drop_non_predictive_columns(
        zindi_raw, config
    )
    # Step 2: strip whitespace and re-validate the categorical levels
    zindi_stripped, unseen_levels = standardize_categorical_values(
        zindi_dropped, config
    )
    # Step 3: enforce dtype contracts and persist frames plus audit;
    # the unseen-levels registry and the configuration flow into the
    # structured level-validation block (Blueprint 2, step 5)
    zindi_cast, taiwan_cast, audit_record = enforce_dtype_contracts(
        zindi_stripped, taiwan_raw, output_dir,
        unseen_levels=unseen_levels,
        config=config,
    )
    # Return the state-ready cleansing outputs
    return {
        "zindi": zindi_cast,
        "taiwan": taiwan_cast,
        "unseen_levels": unseen_levels,
        "audit_record": audit_record,
        "output_dir": str(output_dir),
    }
````

````python
# Task 3: Create Stratified Train/Validation/Test Partitions

# ==============================================================================
# Task 3: Create Stratified Train/Validation/Test Partitions
# ==============================================================================


# -------------------------------------------------------------------------------------------------------------------------------
# Task 3, Step 1: Implement the stratified splitter
# -------------------------------------------------------------------------------------------------------------------------------


def implement_stratified_splitter(
    dataframe: pd.DataFrame,
    y: np.ndarray,
    fractions: Dict[str, float],
    seed: int,
) -> Dict[str, np.ndarray]:
    """
    Implement the chained stratified train/validation/test splitter.

    Applies StratifiedShuffleSplit twice: a 60/40 split separates the
    training remainder from a temporary holdout, and a 50/50 split of
    the holdout yields the validation and test partitions.  Returns
    positional integer index arrays.

    Parameters
    ----------
    dataframe : pd.DataFrame
        The cleansed frame whose row count defines the index range.
    y : np.ndarray
        The binary label vector aligned positionally with the frame.
    fractions : Dict[str, float]
        The train/validation/test fractions (0.6 / 0.2 / 0.2).
    seed : int
        The global deterministic seed.

    Returns
    -------
    Dict[str, np.ndarray]
        Positional index arrays under keys train_idx, val_idx, test_idx.

    Raises
    ------
    ValueError
        If the frame and label lengths disagree or the split fails.
    """
    # Validate the label length matches the frame row count
    if len(y) != len(dataframe):
        # Raise a descriptive error for the alignment violation
        raise ValueError(
            f"label length {len(y)} must equal frame rows {len(dataframe)}."
        )
    # Validate the label vector is binary with values in {0, 1}
    # (a non-binary label would silently corrupt the stratification)
    if not set(np.unique(y)).issubset({0, 1}):
        # Raise a descriptive error naming the offending values
        raise ValueError(
            f"y must be binary with values in {{0, 1}}; received "
            f"{sorted(set(np.unique(y)))}."
        )
    # Validate the fractions sum to one within machine precision
    fraction_sum: float = (
        float(fractions["train_fraction"])
        + float(fractions["validation_fraction"])
        + float(fractions["test_fraction"])
    )
    # Assert the fractions identity before any split
    if abs(fraction_sum - 1.0) > 1e-9:
        # Raise a descriptive error for the invalid fractions
        raise ValueError(
            f"fractions must sum to 1.0; received {fraction_sum}."
        )
    # Validate every fraction lies in the open interval (0, 1)
    for name, fraction in fractions.items():
        # Assert the fraction is strictly between zero and one
        if not 0.0 < float(fraction) < 1.0:
            # Raise a descriptive error naming the offending fraction
            raise ValueError(
                f"{name} must lie in (0, 1); received {fraction}."
            )
    # Compute the temporary holdout size as the complement of training
    holdout_size: float = (
        float(fractions["validation_fraction"])
        + float(fractions["test_fraction"])
    )
    # First split: separate 60% training from the 40% temporary holdout
    first_splitter = StratifiedShuffleSplit(
        n_splits=1, test_size=holdout_size, random_state=seed
    )
    # Unpack the single train/holdout split
    train_idx, holdout_idx = next(first_splitter.split(dataframe, y))
    # Slice the holdout rows and labels by position for the second split
    holdout_frame: pd.DataFrame = dataframe.iloc[holdout_idx]
    holdout_y: np.ndarray = y[holdout_idx]
    # Second split: separate the holdout into equal validation and test
    second_splitter = StratifiedShuffleSplit(
        n_splits=1, test_size=0.5, random_state=seed
    )
    # Unpack the single validation/test split of the holdout
    val_in_holdout, test_in_holdout = next(
        second_splitter.split(holdout_frame, holdout_y)
    )
    # Map the holdout-relative indices back to original positional indices
    val_idx: np.ndarray = holdout_idx[val_in_holdout]
    test_idx: np.ndarray = holdout_idx[test_in_holdout]
    # Assert the three partition sizes sum to the frame row count
    total_size: int = (
        len(train_idx) + len(val_idx) + len(test_idx)
    )
    # Verify the sizes exhaust the frame exactly
    if total_size != len(dataframe):
        # Raise a descriptive error for the size-sum violation
        raise ValueError(
            f"partition sizes must sum to the frame row count "
            f"{len(dataframe)}; received {total_size}."
        )
    # Return the positional index arrays
    return {
        "train_idx": train_idx,
        "val_idx": val_idx,
        "test_idx": test_idx,
    }


# -------------------------------------------------------------------------------------------------------------------------------
# Task 3, Step 2: Split Zindi and verify partition integrity
# -------------------------------------------------------------------------------------------------------------------------------


def split_zindi_and_verify(
    zindi_cleansed: pd.DataFrame,
    config: Dict[str, Any],
    unseen_levels: Dict[str, List[str]],
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Split the cleansed Zindi frame and verify partition integrity.

    Applies the chained splitter with the configured seed, asserts the
    training size 14114, the validation and test sizes 4705 within one,
    and the test positive-class rate 0.1407 within 1e-3; enforces the
    unseen-levels training halt; persists the frozen partition manifest.

    Parameters
    ----------
    zindi_cleansed : pd.DataFrame
        The cleansed Zindi frame, shape (23524, 11).
    config : Dict[str, Any]
        The validated study configuration dictionary.
    unseen_levels : Dict[str, List[str]]
        The unseen-levels registry from Task 2.
    output_dir : Path
        Directory where the partition manifest is persisted.

    Returns
    -------
    Dict[str, Any]
        The frozen partition manifest with index arrays, sizes and
        positive-class rates.

    Raises
    ------
    ValueError
        If any integrity assertion fails or unseen levels land in the
        training partition.
    """
    # Extract the label column name from the configuration
    label: str = config["dataset_parameters"]["zindi_fsd"]["label_column"]
    # Extract the label vector as a positional numpy array
    y: np.ndarray = zindi_cleansed[label].to_numpy(dtype=np.int64)
    # Extract the partitioning parameters from the configuration
    partitioning: Dict[str, Any] = config["partitioning_parameters"]
    # Compose the fractions dict from the configuration
    fractions: Dict[str, float] = {
        "train_fraction": partitioning["train_fraction"],
        "validation_fraction": partitioning["validation_fraction"],
        "test_fraction": partitioning["test_fraction"],
    }
    # Apply the chained stratified splitter with the configured seed
    indices: Dict[str, np.ndarray] = implement_stratified_splitter(
        zindi_cleansed, y, fractions, int(partitioning["random_seed"])
    )
    # Assert the training partition size is 14114 within one; the
    # strict value sits at the rounding boundary (0.6 * 23524 =
    # 14114.4), so the tolerance band 14113..14115 is applied and any
    # deviation from the expected 14114 is logged (assessment remedy)
    n_train: int = len(indices["train_idx"])
    # Check the training size against the tolerance band
    if not (14113 <= n_train <= 14115):
        # Raise a descriptive error for the training size violation
        raise ValueError(
            f"zindi training size must be 14114 within one; "
            f"received {n_train}."
        )
    # Log a deviation note when the expected value is not exact
    if n_train != 14114:
        # Log the rounding-boundary deviation for the audit trail
        logger.info(
            f"zindi training size {n_train} differs from the expected "
            f"14114 by {n_train - 14114}; the split sits at the "
            f"rounding boundary 0.6 * 23524 = 14114.4"
        )
    # Assert the validation and test sizes are 4705 within one
    for name in ("val_idx", "test_idx"):
        # Check the partition size against the tolerance band
        if not (4704 <= len(indices[name]) <= 4706):
            # Raise a descriptive error for the size violation
            raise ValueError(
                f"zindi {name} size must be 4705 within one; "
                f"received {len(indices[name])}."
            )
    # Enforce the unseen-levels training halt rule
    if unseen_levels:
        # Compute the training row slice for level inspection
        train_rows: pd.DataFrame = zindi_cleansed.iloc[indices["train_idx"]]
        # Iterate over the unseen-level registry entries
        for column, values in unseen_levels.items():
            # Check whether any unseen value landed in training rows
            if train_rows[column].isin(values).any():
                # Raise naming the offending column and values
                raise ValueError(
                    f"unseen levels {values} in '{column}' occur in the "
                    f"training partition; this contradicts the fixed-levels "
                    f"claim and halts the pipeline."
                )
    # Compute the per-partition positive-class rates
    rates: Dict[str, float] = {
        name: float(y[indices[name]].mean())
        for name in ("train_idx", "val_idx", "test_idx")
    }
    # Assert the test positive-class rate is 0.1407 within 1e-3
    if abs(rates["test_idx"] - 0.1407) > 1e-3:
        # Raise a descriptive error for the rate violation
        raise ValueError(
            f"zindi test positive-class rate must be 0.1407 within 1e-3; "
            f"received {rates['test_idx']:.6f}."
        )
    # Assemble the frozen partition manifest
    manifest: Dict[str, Any] = {
        "dataset": "zindi",
        "seed": int(partitioning["random_seed"]),
        "method": "chained StratifiedShuffleSplit (60/40 then 50/50); "
        "rational placeholder documented in manifest",
        "sizes": {
            "train": int(len(indices["train_idx"])),
            "val": int(len(indices["val_idx"])),
            "test": int(len(indices["test_idx"])),
        },
        "positive_rates": {
            "train": rates["train_idx"],
            "val": rates["val_idx"],
            "test": rates["test_idx"],
        },
        "train_idx": indices["train_idx"].tolist(),
        "val_idx": indices["val_idx"].tolist(),
        "test_idx": indices["test_idx"].tolist(),
        "test_usage": "reserved strictly for final evaluation, Platt "
        "application, fairness auditing, bootstrap resampling of "
        "predictions, and SHAP explanation computation; never for fitting",
    }
    # Build the manifest path inside the output directory
    manifest_path: Path = output_dir / "zindi_partition_manifest.json"
    # Serialise the manifest to JSON with a stable layout
    manifest_text: str = json.dumps(manifest, indent=2, ensure_ascii=False)
    # Persist the manifest to the output directory immediately
    manifest_path.write_text(manifest_text, encoding="utf-8")
    # Log the partition summary for the audit trail
    logger.info(
        f"zindi partitions: train={manifest['sizes']['train']} "
        f"val={manifest['sizes']['val']} test={manifest['sizes']['test']}"
    )
    # Return the frozen partition manifest
    return manifest


# -------------------------------------------------------------------------------------------------------------------------------
# Task 3, Step 3: Split Taiwan and verify partition integrity
# -------------------------------------------------------------------------------------------------------------------------------


def split_taiwan_and_verify(
    taiwan_cleansed: pd.DataFrame,
    config: Dict[str, Any],
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Split the cleansed Taiwan frame with the identical splitter.

    Applies the same chained splitter and seed as the Zindi split,
    records the three partition sizes and positive-class rates in the
    manifest (internal reproducibility metadata, not reported in the
    manuscript), and applies the same read-only test discipline.

    Parameters
    ----------
    taiwan_cleansed : pd.DataFrame
        The cleansed Taiwan frame, shape (30000, 24).
    config : Dict[str, Any]
        The validated study configuration dictionary.
    output_dir : Path
        Directory where the partition manifest is persisted.

    Returns
    -------
    Dict[str, Any]
        The frozen Taiwan partition manifest.
    """
    # Extract the label column name from the configuration
    label: str = "default.payment.next.month"
    # Extract the label vector as a positional numpy array
    y: np.ndarray = taiwan_cleansed[label].to_numpy(dtype=np.int64)
    # Extract the partitioning parameters from the configuration
    partitioning: Dict[str, Any] = config["partitioning_parameters"]
    # Compose the fractions dict from the configuration
    fractions: Dict[str, float] = {
        "train_fraction": partitioning["train_fraction"],
        "validation_fraction": partitioning["validation_fraction"],
        "test_fraction": partitioning["test_fraction"],
    }
    # Apply the identical chained splitter with the identical seed
    indices: Dict[str, np.ndarray] = implement_stratified_splitter(
        taiwan_cleansed, y, fractions, int(partitioning["random_seed"])
    )
    # Compute the per-partition positive-class rates
    rates: Dict[str, float] = {
        name: float(y[indices[name]].mean())
        for name in ("train_idx", "val_idx", "test_idx")
    }
    # Assemble the frozen Taiwan partition manifest
    manifest: Dict[str, Any] = {
        "dataset": "taiwan",
        "seed": int(partitioning["random_seed"]),
        "method": "chained StratifiedShuffleSplit (60/40 then 50/50); "
        "rational placeholder documented in manifest",
        "caveat": "Taiwan preprocessing and partitioning are "
        "placeholder-based and may cause numerical differences from the "
        "published thesis results; partition sizes are not reported in "
        "the manuscript and are internal reproducibility metadata",
        "sizes": {
            "train": int(len(indices["train_idx"])),
            "val": int(len(indices["val_idx"])),
            "test": int(len(indices["test_idx"])),
        },
        "positive_rates": {
            "train": rates["train_idx"],
            "val": rates["val_idx"],
            "test": rates["test_idx"],
        },
        "train_idx": indices["train_idx"].tolist(),
        "val_idx": indices["val_idx"].tolist(),
        "test_idx": indices["test_idx"].tolist(),
        "test_usage": "read-only; reserved strictly for final evaluation",
    }
    # Build the manifest path inside the output directory
    manifest_path: Path = output_dir / "taiwan_partition_manifest.json"
    # Serialise the manifest to JSON with a stable layout
    manifest_text: str = json.dumps(manifest, indent=2, ensure_ascii=False)
    # Persist the manifest to the output directory immediately
    manifest_path.write_text(manifest_text, encoding="utf-8")
    # Log the partition summary for the audit trail
    logger.info(
        f"taiwan partitions: train={manifest['sizes']['train']} "
        f"val={manifest['sizes']['val']} test={manifest['sizes']['test']}"
    )
    # Return the frozen Taiwan partition manifest
    return manifest


# -------------------------------------------------------------------------------------------------------------------------------
# Task 3, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


def orchestrate_create_partitions(
    config: Dict[str, Any],
    zindi_cleansed: pd.DataFrame,
    taiwan_cleansed: pd.DataFrame,
    unseen_levels: Dict[str, List[str]],
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Orchestrate the creation of both frozen partition manifests.

    Composes implement_stratified_splitter, split_zindi_and_verify and
    split_taiwan_and_verify, persisting both manifests to the output
    directory.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated study configuration dictionary.
    zindi_cleansed : pd.DataFrame
        The cleansed Zindi frame, shape (23524, 11).
    taiwan_cleansed : pd.DataFrame
        The cleansed Taiwan frame, shape (30000, 24).
    unseen_levels : Dict[str, List[str]]
        The unseen-levels registry from Task 2.
    output_dir : Path
        Directory where the partition manifests are persisted.

    Returns
    -------
    Dict[str, Any]
        A state-ready dict with keys 'zindi_partitions',
        'taiwan_partitions' and 'output_dir'.
    """
    # Validate the output directory type before any filesystem work
    if not isinstance(output_dir, Path):
        # Raise a descriptive error for a non-Path output directory
        raise ValueError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Create the output directory including any missing parents
    output_dir.mkdir(parents=True, exist_ok=True)
    # Step 1: split the cleansed Zindi frame and verify its integrity
    zindi_manifest: Dict[str, Any] = split_zindi_and_verify(
        zindi_cleansed, config, unseen_levels, output_dir
    )
    # Step 2: split the cleansed Taiwan frame with the identical splitter
    taiwan_manifest: Dict[str, Any] = split_taiwan_and_verify(
        taiwan_cleansed, config, output_dir
    )
    # Assert both manifest files exist before returning
    for manifest_path in (
        output_dir / "zindi_partition_manifest.json",
        output_dir / "taiwan_partition_manifest.json",
    ):
        # Verify the manifest file was persisted
        if not manifest_path.is_file():
            # Raise a descriptive error naming the missing manifest
            raise FileNotFoundError(
                f"partition manifest missing at {manifest_path}."
            )
    # Return the state-ready partitioning outputs
    return {
        "zindi_partitions": zindi_manifest,
        "taiwan_partitions": taiwan_manifest,
        "output_dir": str(output_dir),
    }
````

````python
# Task 4: Fit Imputation and Winsorization Statistics on Training

# ==============================================================================
# Task 4: Fit Imputation and Winsorization Statistics on Training
# ==============================================================================


# The Zindi numeric features that receive median imputation and
# winsorization; the Taiwan feature sets are derived from the schema
_NUMERIC_FEATURES: Tuple[str, ...] = ("household_size", "age_of_respondent")

# The Zindi categorical features that receive mode imputation
_CATEGORICAL_FEATURES: Tuple[str, ...] = (
    "country",
    "location_type",
    "cellphone_access",
    "gender_of_respondent",
    "relationship_with_head",
    "marital_status",
    "education_level",
    "job_type",
)


def derive_taiwan_feature_sets(
    config: Dict[str, Any],
) -> Tuple[Tuple[str, ...], Tuple[str, ...]]:
    """
    Derive the Taiwan numeric and categorical feature sets from schema.
    """
    # Extract the Taiwan schema column list from the configuration
    schema_columns: List[Dict[str, Any]] = config["raw_data_schemas"][
        "taiwan_credit_schema"
    ]["columns"]
    # Collect the numeric and categorical feature names
    numeric: List[str] = []
    categorical: List[str] = []
    # Classify every schema column by its declared role and dtype
    for column in schema_columns:
        # Skip the row identifier and the label column
        if column["name"] in ("ID", "default.payment.next.month"):
            # Row identifiers and labels are not imputation features
            continue
        # Classify categorical-code columns as categorical
        if "categorical code" in column["dtype"]:
            # Add the column to the categorical feature set
            categorical.append(column["name"])
        else:
            # Add the column to the numeric feature set
            numeric.append(column["name"])
    # Return the derived feature sets as frozen tuples
    return tuple(numeric), tuple(categorical)

# -------------------------------------------------------------------------------------------------------------------------------
# Task 4, Step 1: Compute imputation statistics on the Zindi training partition
# -------------------------------------------------------------------------------------------------------------------------------


def compute_imputation_stats(
    train_df: pd.DataFrame,
    numeric_features: Tuple[str, ...],
    categorical_features: Tuple[str, ...],
) -> Dict[str, Dict[str, Any]]:
    """
    Compute median and mode imputation statistics on training only.

    Computes the median of every numeric feature and the first-sorted
    mode of every categorical feature on the supplied training
    partition.  The function contract receives training rows only, so no
    validation or test observation can enter the statistics.

    Parameters
    ----------
    train_df : pd.DataFrame
        The training partition of the cleansed frame.
    numeric_features : Tuple[str, ...]
        The numeric feature names receiving median imputation.
    categorical_features : Tuple[str, ...]
        The categorical feature names receiving mode imputation.

    Returns
    -------
    Dict[str, Dict[str, Any]]
        A dict with 'medians' (numeric column -> median) and 'modes'
        (categorical column -> mode) blocks.

    Raises
    ------
    ValueError
        If the frame type is invalid or a required column is missing.
    """
    # Validate the frame type before any pandas operation
    if not isinstance(train_df, pd.DataFrame):
        # Raise a descriptive error for a non-DataFrame input
        raise ValueError(
            f"train_df must be a pd.DataFrame; "
            f"received {type(train_df).__name__}."
        )
    # Validate every required numeric column is present
    for column in numeric_features:
        # Assert the numeric column exists in the frame
        if column not in train_df.columns:
            # Raise a descriptive error naming the missing column
            raise ValueError(
                f"train_df missing numeric column '{column}'."
            )
    # Validate every required categorical column is present
    for column in categorical_features:
        # Assert the categorical column exists in the frame
        if column not in train_df.columns:
            # Raise a descriptive error naming the missing column
            raise ValueError(
                f"train_df missing categorical column '{column}'."
            )
    # Compute the median of every numeric feature on training only
    medians: Dict[str, Any] = {
        column: float(train_df[column].median())
        for column in numeric_features
    }
    # Compute the first sorted mode of every categorical feature
    modes: Dict[str, Any] = {}
    # Iterate over the categorical features to compute their modes
    for column in categorical_features:
        # Take the first (sorted) mode for determinism
        modes[column] = train_df[column].mode().iloc[0]
    # Log the computed statistics for the audit trail
    logger.info(f"imputation medians: {medians}")
    # Return the frozen imputation statistics block
    return {"medians": medians, "modes": modes}


# -------------------------------------------------------------------------------------------------------------------------------
# Task 4, Step 2: Compute winsorization statistics on the Zindi training partition
# -------------------------------------------------------------------------------------------------------------------------------


def compute_winsorization_stats(
    train_df: pd.DataFrame,
    numeric_features: Tuple[str, ...],
) -> Dict[str, Tuple[float, float]]:
    """
    Compute the 1st and 99th percentile winsorization bounds on training.

    Computes numpy.percentile(x, 1) and numpy.percentile(x, 99) for
    every declared numeric feature on the training partition only.  The
    raw percentile values are recorded without rounding.

    Parameters
    ----------
    train_df : pd.DataFrame
        The training partition of the cleansed frame.
    numeric_features : Tuple[str, ...]
        The numeric feature names receiving winsorization.

    Returns
    -------
    Dict[str, Tuple[float, float]]
        Feature name -> (lower, upper) winsorization bounds.

    Raises
    ------
    ValueError
        If a required feature is missing or non-finite.
    """
    # Validate the frame type before any pandas operation
    if not isinstance(train_df, pd.DataFrame):
        # Raise a descriptive error for a non-DataFrame input
        raise ValueError(
            f"train_df must be a pd.DataFrame; "
            f"received {type(train_df).__name__}."
        )
    # Collect the winsorization bounds keyed by feature name
    winsor_bounds: Dict[str, Tuple[float, float]] = {}
    # Iterate over the numeric features to compute their bounds
    for column in numeric_features:
        # Assert the feature column exists in the frame
        if column not in train_df.columns:
            # Raise a descriptive error naming the missing feature
            raise ValueError(
                f"train_df missing numeric feature '{column}'."
            )
        # Extract the column values as a float array
        values: np.ndarray = train_df[column].to_numpy(dtype=np.float64)
        # Assert the column is finite before percentile computation
        if not np.isfinite(values).all():
            # Raise a descriptive error for non-finite values
            raise ValueError(
                f"column '{column}' contains non-finite values; "
                f"percentiles are undefined."
            )
        # Compute the 1st percentile of the training values
        lower: float = float(np.percentile(values, 1))
        # Compute the 99th percentile of the training values
        upper: float = float(np.percentile(values, 99))
        # Record the raw unrounded bounds for the feature
        winsor_bounds[column] = (lower, upper)
    # Log the exact percentile values for the audit log
    logger.info(f"winsorization bounds: {winsor_bounds}")
    # Return the frozen winsorization bounds block
    return winsor_bounds


# -------------------------------------------------------------------------------------------------------------------------------
# Task 4, Step 3: Apply frozen statistics to validation, test, and Taiwan
# -------------------------------------------------------------------------------------------------------------------------------


def apply_frozen_stats(
    frame: pd.DataFrame,
    stats: Dict[str, Any],
    dataset: str,
) -> pd.DataFrame:
    """
    Apply frozen imputation and winsorization statistics to a frame.

    Fills missing numeric values with the frozen medians, missing
    categorical values with the frozen modes, and clips the two numeric
    features with x_winsorized = min(max(x, q01), q99) using the frozen
    training bounds.  No statistic is recomputed here.

    Parameters
    ----------
    frame : pd.DataFrame
        The cleansed frame to transform (any partition).
    stats : Dict[str, Any]
        The frozen preprocessing stats for this dataset, carrying
        'medians', 'modes' and 'winsor_bounds'.
    dataset : str
        Dataset name for the audit log ('zindi' or 'taiwan').

    Returns
    -------
    pd.DataFrame
        The transformed frame with identical shape and row order.

    Raises
    ------
    ValueError
        If the frame type is invalid or the stats blocks are missing.
    """
    # Validate the frame type before any pandas operation
    if not isinstance(frame, pd.DataFrame):
        # Raise a descriptive error for a non-DataFrame input
        raise ValueError(
            f"frame must be a pd.DataFrame; "
            f"received {type(frame).__name__}."
        )
    # Validate the stats blocks are present
    for block in ("medians", "modes", "winsor_bounds"):
        # Assert the required stats block exists
        if block not in stats:
            # Raise a descriptive error naming the missing block
            raise ValueError(
                f"frozen stats missing block '{block}' for {dataset}."
            )
    # Capture the input shape for the integrity assertion
    input_shape: Tuple[int, int] = frame.shape
    # Take a working copy so the caller's frame is not mutated
    transformed: pd.DataFrame = frame.copy()
    # Fill missing numeric values with the frozen medians
    for column, median in stats["medians"].items():
        # Apply the frozen median fill to the numeric column
        transformed[column] = transformed[column].fillna(median)
    # Fill missing categorical values with the frozen modes
    for column, mode in stats["modes"].items():
        # Apply the frozen mode fill to the categorical column
        transformed[column] = transformed[column].fillna(mode)
    # Clip the numeric features with the frozen winsorization bounds
    for column, (lower, upper) in stats["winsor_bounds"].items():
        # Apply the winsorization clip: min(max(x, q01), q99)
        transformed[column] = transformed[column].clip(lower=lower, upper=upper)
    # Assert the transformation preserved the frame shape
    if transformed.shape != input_shape:
        # Raise a descriptive error for the shape violation
        raise ValueError(
            f"transformation changed shape from {input_shape} to "
            f"{transformed.shape}; row order integrity violated."
        )
    # Log the application for the audit trail
    logger.info(f"frozen stats applied to {dataset} frame (shape unchanged)")
    # Return the transformed frame
    return transformed


def _fit_dataset_stats(
    train_df: pd.DataFrame,
    numeric_features: Tuple[str, ...],
    categorical_features: Tuple[str, ...],
) -> Dict[str, Any]:
    """
    Fit the full frozen stats block for one dataset.
    """
    # Compute the imputation statistics on the training partition
    imputation: Dict[str, Dict[str, Any]] = compute_imputation_stats(
        train_df, numeric_features, categorical_features
    )
    # Compute the winsorization bounds on the training partition
    winsor_bounds: Dict[str, Tuple[float, float]] = compute_winsorization_stats(
        train_df, numeric_features
    )
    # Compose the complete frozen stats block for the dataset
    return {
        "medians": imputation["medians"],
        "modes": imputation["modes"],
        "winsor_bounds": winsor_bounds,
    }


# -------------------------------------------------------------------------------------------------------------------------------
# Task 4, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


def orchestrate_fit_preprocessing_stats(
    config: Dict[str, Any],
    zindi_frame: pd.DataFrame,
    zindi_manifest: Dict[str, Any],
    taiwan_frame: pd.DataFrame,
    taiwan_manifest: Dict[str, Any],
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Orchestrate the fitting and application of preprocessing statistics.

    Slices each training partition by its frozen manifest indices,
    computes imputation and winsorization statistics on training only,
    applies them frozen to the full frames, and persists the stats JSON
    to the output directory.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated study configuration dictionary.
    zindi_frame : pd.DataFrame
        The cleansed Zindi frame, shape (23524, 11).
    zindi_manifest : Dict[str, Any]
        The frozen Zindi partition manifest.
    taiwan_frame : pd.DataFrame
        The cleansed Taiwan frame, shape (30000, 24).
    taiwan_manifest : Dict[str, Any]
        The frozen Taiwan partition manifest.
    output_dir : Path
        Directory where the stats JSON is persisted.

    Returns
    -------
    Dict[str, Any]
        A state-ready dict with keys 'zindi_frame', 'taiwan_frame',
        'preprocessing_stats' and 'output_dir'.
    """
    # Validate the output directory type before any filesystem work
    if not isinstance(output_dir, Path):
        # Raise a descriptive error for a non-Path output directory
        raise ValueError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Create the output directory including any missing parents
    output_dir.mkdir(parents=True, exist_ok=True)
    # Derive the Taiwan numeric and categorical feature sets from schema
    taiwan_numeric, taiwan_categorical = derive_taiwan_feature_sets(config)
    # Slice the Zindi training partition by the frozen manifest indices
    zindi_train: pd.DataFrame = zindi_frame.iloc[
        np.array(zindi_manifest["train_idx"], dtype=np.int64)
    ]
    # Fit the frozen Zindi stats on the Zindi training partition only
    zindi_stats: Dict[str, Any] = _fit_dataset_stats(
        zindi_train, _NUMERIC_FEATURES, _CATEGORICAL_FEATURES
    )
    # Slice the Taiwan training partition by the frozen manifest indices
    taiwan_train: pd.DataFrame = taiwan_frame.iloc[
        np.array(taiwan_manifest["train_idx"], dtype=np.int64)
    ]
    # Fit the frozen Taiwan stats on the Taiwan training partition only
    # (placeholder protocol: identical procedure, Taiwan-specific fits)
    taiwan_stats: Dict[str, Any] = _fit_dataset_stats(
        taiwan_train, taiwan_numeric, taiwan_categorical
    )
    # Log the Taiwan placeholder decision explicitly
    logger.info(
        "Taiwan preprocessing is UNSPECIFIED IN MANUSCRIPT; placeholder "
        "protocol applied: training-only medians, modes, and 1st/99th "
        "percentiles, frozen for validation and test"
    )
    # Apply the frozen Zindi stats to the full Zindi frame
    zindi_transformed: pd.DataFrame = apply_frozen_stats(
        zindi_frame, zindi_stats, "zindi"
    )
    # Apply the frozen Taiwan stats to the full Taiwan frame
    taiwan_transformed: pd.DataFrame = apply_frozen_stats(
        taiwan_frame, taiwan_stats, "taiwan"
    )
    # Compose the per-dataset frozen stats block
    preprocessing_stats: Dict[str, Any] = {
        "zindi": zindi_stats,
        "taiwan": taiwan_stats,
    }
    # Build the stats JSON path inside the output directory
    stats_path: Path = output_dir / "preprocessing_stats.json"
    # Serialise the frozen stats to JSON with a stable layout
    stats_text: str = json.dumps(
        preprocessing_stats, indent=2, ensure_ascii=False, default=str
    )
    # Persist the frozen stats to the output directory
    stats_path.write_text(stats_text, encoding="utf-8")
    # Assert the stats JSON exists before returning
    if not stats_path.is_file():
        # Raise a descriptive error for the missing stats artifact
        raise FileNotFoundError(
            f"preprocessing stats missing at {stats_path}."
        )
    # Return the state-ready preprocessing outputs
    return {
        "zindi_frame": zindi_transformed,
        "taiwan_frame": taiwan_transformed,
        "preprocessing_stats": preprocessing_stats,
        "output_dir": str(output_dir),
    }
````

````python
# Task 5: Monotonic Binning and WoE Encoding (Logistic Branch)

# ==============================================================================
# Task 5: Monotonic Binning and WoE Encoding (Logistic Branch)
# ==============================================================================


# The initial quantile bin count per the placeholder algorithm
_INITIAL_BIN_COUNT: int = 20

# The minimum bin frequency as a fraction of training observations
_MIN_BIN_FREQUENCY: float = 0.05


def woe_from_counts(g: int, b: int, G: int, B: int) -> float:
    """
    Compute the Weight of Evidence for one bin via Equation (2).

    Implements WoE_k = ln((g_k / G) / (b_k / B)) with the pinned class
    convention g = count of y=0, b = count of y=1, G and B the totals.
    A zero count in either class makes the logarithm undefined; the
    documented neutral rule assigns WoE = 0.

    Parameters
    ----------
    g : int
        Count of y=0 observations in the bin.
    b : int
        Count of y=1 observations in the bin.
    G : int
        Total y=0 count on the training partition.
    B : int
        Total y=1 count on the training partition.

    Returns
    -------
    float
        The bin's Weight of Evidence, or 0.0 under the neutral rule.
    """
    # Apply the neutral rule when either class count is zero
    if g == 0 or b == 0 or G == 0 or B == 0:
        # Log the neutral assignment for the audit trail
        logger.info(f"neutral WoE=0 applied (g={g}, b={b}, G={G}, B={B})")
        # Return the neutral zero score
        return 0.0
    # Compute the good share: g_k / G
    good_share: float = g / G
    # Compute the bad share: b_k / B
    bad_share: float = b / B
    # Compute the WoE as the log of the odds ratio (Equation 2)
    return math.log(good_share / bad_share)


def _interval_label(left: float, right: float) -> str:
    """
    Render a deterministic bin identifier string from its edges.

    This helper is the SINGLE shared interval-identifier formatter of
    the WoE pipeline: compute_woe_scores, apply_woe_mapping and the
    Task 6 Information-Value computation all import it, so no duplicate
    formatting can drift.  The two-decimal rendering is retained, and
    callers constructing maps MUST assert key uniqueness because two
    distinct intervals whose edges round to the same string would
    otherwise collide silently as dict keys.

    Parameters
    ----------
    left : float
        The left bin edge.
    right : float
        The right bin edge.

    Returns
    -------
    str
        The half-open interval identifier string.
    """
    # Format the half-open interval label with two decimals
    return f"({left:.2f}, {right:.2f}]"


def assert_interval_key_uniqueness(
    intervals: List[Tuple[float, float]],
) -> None:
    """
    Assert that rendered interval keys are unique for the edges.

    Raises a descriptive ValueError when two distinct intervals render
    to the same two-decimal key, which would silently collide as dict
    keys (assessment remedy for _interval_label).

    Parameters
    ----------
    intervals : List[Tuple[float, float]]
        The (left, right) edge pairs of the frozen bins.

    Raises
    ------
    ValueError
        If any two distinct intervals collide on their rendered key.
    """
    # Collect the rendered keys and their source intervals
    rendered: Dict[str, Tuple[float, float]] = {}
    # Iterate over the edge pairs
    for left, right in intervals:
        # Render the interval key through the shared helper
        key: str = _interval_label(left, right)
        # Detect a collision with a distinct interval
        if key in rendered and rendered[key] != (left, right):
            # Raise naming the colliding intervals
            raise ValueError(
                f"interval key collision: '{key}' is rendered by both "
                f"{rendered[key]} and {(left, right)}; two distinct "
                f"bins would silently share one dict key."
            )
        # Register the rendered key
        rendered[key] = (left, right)


def _merge_bins(
    bins: List[Dict[str, Any]], index: int
) -> List[Dict[str, Any]]:
    """
    Merge adjacent bins at index with their counts summed.
    """
    # Fetch the left and right bins to merge
    left_bin: Dict[str, Any] = bins[index]
    right_bin: Dict[str, Any] = bins[index + 1]
    # Compose the merged bin with combined counts and extended edges
    merged: Dict[str, Any] = {
        "lo": left_bin["lo"],
        "hi": right_bin["hi"],
        "g": left_bin["g"] + right_bin["g"],
        "b": left_bin["b"] + right_bin["b"],
        "n": left_bin["n"] + right_bin["n"],
        "G": left_bin["G"],
        "B": left_bin["B"],
    }
    # Recompute the merged bin's WoE from the combined counts
    merged["woe"] = woe_from_counts(
        merged["g"], merged["b"], merged["G"], merged["B"]
    )
    # Build the new bin list without the two merged bins
    new_bins: List[Dict[str, Any]] = (
        bins[:index] + [merged] + bins[index + 2:]
    )
    # Return the reduced bin list
    return new_bins


def _is_monotone(woes: List[float]) -> bool:
    """
    Return True when a WoE sequence is monotone (either direction).
    """
    # Compute the signs of consecutive differences
    diffs: List[float] = [
        woes[i + 1] - woes[i] for i in range(len(woes) - 1)
    ]
    # A sequence is monotone when no sign change occurs
    non_zero: List[float] = [d for d in diffs if abs(d) > 1e-12]
    # Collect the sign pattern of the non-zero differences
    signs: List[int] = [1 if d > 0 else -1 for d in non_zero]
    # The sequence is monotone when all signs agree
    return len(set(signs)) <= 1


def _enforce_monotonicity(
    bins: List[Dict[str, Any]], max_iterations: int = 200
) -> List[Dict[str, Any]]:
    """
    Merge adjacent bins until the WoE sequence is monotone.
    """
    # Iterate with a bounded loop to guarantee termination
    for _ in range(max_iterations):
        # Extract the current WoE sequence
        woes: List[float] = [b["woe"] for b in bins]
        # Stop when the sequence is already monotone
        if _is_monotone(woes):
            # The monotonicity constraint is satisfied
            break
        # Find the direction-violating adjacent pairs
        violating: List[int] = []
        # Iterate over the internal bin positions
        for i in range(1, len(bins) - 1):
            # Compute the two consecutive differences at position i
            left_diff: float = woes[i] - woes[i - 1]
            right_diff: float = woes[i + 1] - woes[i]
            # Record a direction change when both differences are non-zero
            if abs(left_diff) > 1e-12 and abs(right_diff) > 1e-12:
                # Check whether the differences carry opposite signs
                if (left_diff > 0) != (right_diff > 0):
                    # Record the violating pair starting at i
                    violating.append(i)
        # Halt if no violating pair exists (defensive)
        if not violating:
            # Raise because the constraint cannot be satisfied
            raise ValueError(
                "monotonicity enforcement found no violating pair but "
                "the sequence is not monotone."
            )
        # Choose the violating pair with the closest WoE values
        best_index: int = min(
            violating,
            key=lambda i: abs(bins[i]["woe"] - bins[i + 1]["woe"]),
        )
        # Log the merge decision for the audit trail
        logger.info(
            f"monotonicity merge at bins {best_index}-{best_index + 1}"
        )
        # Merge the closest-WoE violating pair
        bins = _merge_bins(bins, best_index)
    # Verify the final sequence is monotone
    if not _is_monotone([b["woe"] for b in bins]):
        # Raise because the constraint could not be satisfied
        raise ValueError(
            "monotonicity could not be enforced within the iteration bound."
        )
    # Return the monotone bin list
    return bins


def _enforce_min_frequency(
    bins: List[Dict[str, Any]], min_count: int, max_iterations: int = 200
) -> List[Dict[str, Any]]:
    """
    Enforce the minimum bin frequency by least-frequent-bin merging.

    Implements the config-declared placeholder algorithm (preprocessing
    binning step 4): every bin must hold at least min_count =
    ceil(0.05 * n_train) training observations; when a bin violates the
    floor, the LEAST-FREQUENT violating bin (argmin over n, leftmost on
    ties) is merged with its closest-WoE neighbour (left neighbour on
    equal gaps; the only available direction at either boundary), the
    merged WoE is recomputed from the combined counts via Equation (2),
    and the loop repeats until the floor holds (Blueprint 3).

    Parameters
    ----------
    bins : List[Dict[str, Any]]
        The bin records, each carrying 'lo', 'hi', 'g', 'b', 'n', 'G',
        'B' and 'woe'.
    min_count : int
        The minimum per-bin observation count
        (ceil(0.05 * n_train), floored at 1).
    max_iterations : int, optional
        The bounded-loop cap (200 per the placeholder).

    Returns
    -------
    List[Dict[str, Any]]
        The frequency-compliant bin list.

    Raises
    ------
    ValueError
        If the floor cannot be satisfied within the iteration bound.
    """
    # Iterate with a bounded loop to guarantee termination
    for _ in range(max_iterations):
        # Identify the violating bins: n below the floor
        small: List[int] = [
            i for i, b in enumerate(bins) if b["n"] < min_count
        ]
        # Stop when no bin violates the frequency floor
        if not small:
            # The frequency constraint is satisfied
            break
        # Select the LEAST-FREQUENT violating bin (argmin over n); the
        # leftmost bin wins the tie as the documented deterministic
        # tie-break (Blueprint 3, step 2)
        index: int = min(small, key=lambda i: (bins[i]["n"], i))
        # Determine the merge direction with the closest-WoE neighbour
        left_woe_gap: float = (
            abs(bins[index]["woe"] - bins[index - 1]["woe"])
            if index > 0
            else float("inf")
        )
        # Compute the right neighbour's WoE gap
        right_woe_gap: float = (
            abs(bins[index]["woe"] - bins[index + 1]["woe"])
            if index < len(bins) - 1
            else float("inf")
        )
        # Merge toward the neighbour with the closer WoE; the left
        # neighbour wins equal gaps as the documented tie-break, and a
        # boundary bin takes its only available direction
        if left_woe_gap <= right_woe_gap:
            # Merge with the left neighbour
            merge_index: int = index - 1
        else:
            # Merge with the right neighbour
            merge_index = index
        # Log the frequency merge decision for the audit trail
        logger.info(
            f"frequency merge at bins {merge_index}-{merge_index + 1} "
            f"(n={bins[index]['n']} < {min_count})"
        )
        # Merge the chosen adjacent pair (WoE recomputed from combined
        # counts via Equation (2) inside _merge_bins)
        bins = _merge_bins(bins, merge_index)
    # Verify the final frequency floor is satisfied
    if any(b["n"] < min_count for b in bins):
        # Raise because the floor could not be satisfied
        raise ValueError(
            "minimum bin frequency could not be satisfied within the "
            "iteration bound."
        )
    # Return the frequency-compliant bin list
    return bins


def fit_monotonic_bins(
    train_df: pd.DataFrame,
    y: np.ndarray,
    continuous_features: Tuple[str, ...],
    initial_bin_count: int = _INITIAL_BIN_COUNT,
    min_frequency: float = _MIN_BIN_FREQUENCY,
) -> Dict[str, Any]:
    """
    Fit frozen monotonic bin boundaries on the training partition.

    Starts from initial_bin_count quantile bins, computes the WoE of
    every bin via Equation (2) with the pinned class convention,
    enforces monotonicity by closest-WoE adjacent merging, enforces the
    minimum bin frequency by closest-WoE merging, and freezes the final
    edges per feature.

    Parameters
    ----------
    train_df : pd.DataFrame
        The training partition of the cleansed frame.
    y : np.ndarray
        The training label vector, positionally aligned.
    continuous_features : Tuple[str, ...]
        The continuous feature names entering the logistic branch.
    initial_bin_count : int, optional
        The starting quantile bin count (20 per the placeholder).
    min_frequency : float, optional
        The minimum bin frequency as a fraction (0.05 per the spec).

    Returns
    -------
    Dict[str, Any]
        A dict with 'cut_points' (feature -> sorted edges), 'bin_records'
        (feature -> per-bin records) and 'min_count'.

    Raises
    ------
    ValueError
        If the label length mismatches or binning fails.
    """
    # Validate the label length matches the frame row count
    if len(y) != len(train_df):
        # Raise a descriptive error for the alignment violation
        raise ValueError(
            f"label length {len(y)} must equal frame rows {len(train_df)}."
        )
    # Compute the minimum per-bin observation count
    min_count: int = max(1, int(math.ceil(min_frequency * len(train_df))))
    # Collect the cut points and bin records per feature
    cut_points: Dict[str, List[float]] = {}
    bin_records: Dict[str, List[Dict[str, Any]]] = {}
    # Iterate over the continuous features to fit their bins
    for feature in continuous_features:
        # Validate the feature exists in the frame
        if feature not in train_df.columns:
            # Raise a descriptive error naming the missing feature
            raise ValueError(
                f"train_df missing continuous feature '{feature}'."
            )
        # Extract the feature values as a float array
        values: np.ndarray = train_df[feature].to_numpy(dtype=np.float64)
        # Compute the total class counts on the training partition
        G: int = int(np.sum(y == 0))
        B: int = int(np.sum(y == 1))
        # Initialise the quantile binning attempt
        edges: Optional[np.ndarray] = None
        # Try progressively fewer bins until quantiles are well-defined
        for trial in range(initial_bin_count, 1, -1):
            # Attempt the quantile binning with duplicate-edge dropping
            try:
                # Compute the trial quantile edges for the trial count
                edges = _qcut_edges(values, trial)
                # Stop the trial loop on a successful binning
                break
            except ValueError:
                # Reduce the bin count and retry
                continue
        # Halt if no valid quantile binning could be produced
        if edges is None:
            # Raise a descriptive error for the binning failure
            raise ValueError(
                f"quantile binning failed for feature '{feature}'."
            )
        # Build the initial bin records from the quantile edges
        bins: List[Dict[str, Any]] = _bins_from_edges(values, y, edges, G, B)
        # Enforce the monotonicity constraint by closest-WoE merging
        bins = _enforce_monotonicity(bins)
        # Enforce the minimum frequency floor by closest-WoE merging
        bins = _enforce_min_frequency(bins, min_count)
        # Re-enforce monotonicity in case frequency merging broke it
        bins = _enforce_monotonicity(bins)
        # Extract the final sorted unique edges from the merged bins
        final_edges: List[float] = sorted(
            {round(b["lo"], 10) for b in bins}
            | {round(bins[-1]["hi"], 10)}
        )
        # Freeze the cut points for the feature
        cut_points[feature] = final_edges
        # Freeze the bin records for the feature
        bin_records[feature] = bins
        # Log the fitted bin summary for the audit trail
        logger.info(
            f"feature '{feature}': {len(bins)} monotone bins, "
            f"edges={final_edges}"
        )
    # Return the frozen binning artifacts
    return {
        "cut_points": cut_points,
        "bin_records": bin_records,
        "min_count": min_count,
    }


def _qcut_edges(values: np.ndarray, q: int) -> np.ndarray:
    """
    Return the sorted unique quantile edges for a feature.
    """
    # Compute the quantile binning with duplicate-edge dropping
    bins: pd.Categorical = pd.qcut(values, q=q, duplicates="drop")
    # Extract the interval categories and their edges
    intervals: pd.IntervalIndex = pd.IntervalIndex(bins.categories)
    # Collect the left and right edges of every interval
    edges: List[float] = [float(iv.left) for iv in intervals]
    # Append the right edge of the final interval
    edges.append(float(intervals[-1].right))
    # Return the sorted unique edges as an array
    return np.array(sorted(set(edges)))


def _bins_from_edges(
    values: np.ndarray,
    y: np.ndarray,
    edges: np.ndarray,
    G: int,
    B: int,
) -> List[Dict[str, Any]]:
    """
    Build initial bin records from a set of edges.
    """
    # Assign each observation to an interval via the edges
    assigned: pd.Categorical = pd.cut(
        values, bins=edges, include_lowest=True, right=True
    )
    # Collect the constructed bin records
    bins: List[Dict[str, Any]] = []
    # Iterate over the interval categories in order
    for interval in assigned.categories:
        # Build the mask of observations in this interval
        mask: np.ndarray = assigned == interval
        # Count the negative and positive observations in the bin
        g: int = int(np.sum((y == 0) & mask))
        b: int = int(np.sum((y == 1) & mask))
        # Compute the bin's WoE via Equation (2)
        woe: float = woe_from_counts(g, b, G, B)
        # Append the bin record with its edges and counts
        bins.append(
            {
                "lo": float(interval.left),
                "hi": float(interval.right),
                "g": g,
                "b": b,
                "n": int(np.sum(mask)),
                "G": G,
                "B": B,
                "woe": woe,
            }
        )
    # Return the ordered bin records
    return bins


# -------------------------------------------------------------------------------------------------------------------------------
# Task 5, Step 2: Compute WoE scores per bin and per categorical
# level on training
# -------------------------------------------------------------------------------------------------------------------------------


def compute_woe_scores(
    train_df: pd.DataFrame,
    y: np.ndarray,
    cut_points: Dict[str, List[float]],
    categorical_features: Tuple[str, ...],
    declared_levels: Optional[Dict[str, List[str]]] = None,
) -> Dict[str, Dict[str, float]]:
    """
    Compute the frozen WoE mapping on the training partition.

    Recomputes per-bin counts under the frozen boundaries and per-level
    counts for every level of the UNION of the DECLARED level sets
    (from the configuration, when supplied) and the observed training
    levels, then computes the final WoE per identifier via Equation (2).
    Declared-but-unseen levels are stored EXPLICITLY as the neutral 0.0
    through the shared woe_from_counts helper, so the frozen map is a
    complete lookup that never relies on the NaN-fill path for a
    declared level (Blueprint 4).  Interval identifiers route through
    the single shared _interval_label helper, and key coverage is
    asserted after construction.

    Parameters
    ----------
    train_df : pd.DataFrame
        The training partition of the cleansed frame.
    y : np.ndarray
        The training label vector, positionally aligned.
    cut_points : Dict[str, List[float]]
        The frozen bin boundaries from Step 1.
    categorical_features : Tuple[str, ...]
        The categorical feature names entering the logistic branch.
    declared_levels : Optional[Dict[str, List[str]]], optional
        The declared exhaustive level sets per categorical feature,
        from config["dataset_parameters"]["zindi_fsd"]
        ["categorical_levels"].

    Returns
    -------
    Dict[str, Dict[str, float]]
        The woe_map: feature -> {identifier -> WoE}; every declared
        level of every categorical feature and every frozen bin of
        every continuous feature is present.

    Raises
    ------
    ValueError
        If a feature is missing, the label length mismatches, or the
        key-coverage assertion fails.
    """
    # Validate the label length matches the frame row count
    if len(y) != len(train_df):
        # Raise a descriptive error for the alignment violation
        raise ValueError(
            f"label length {len(y)} must equal frame rows {len(train_df)}."
        )
    # Compute the total class counts on the training partition
    G: int = int(np.sum(y == 0))
    B: int = int(np.sum(y == 1))
    # Collect the per-feature WoE mappings
    woe_map: Dict[str, Dict[str, float]] = {}
    # Compute the WoE for every continuous feature's frozen bins
    for feature, edges in cut_points.items():
        # Validate the feature exists in the frame
        if feature not in train_df.columns:
            # Raise a descriptive error naming the missing feature
            raise ValueError(
                f"train_df missing continuous feature '{feature}'."
            )
        # Guard against two-decimal key collisions among the edges
        # (assessment remedy for the shared identifier helper)
        assert_interval_key_uniqueness(
            [
                (float(edges[i]), float(edges[i + 1]))
                for i in range(len(edges) - 1)
            ]
        )
        # Extract the feature values as a float array
        values: np.ndarray = train_df[feature].to_numpy(dtype=np.float64)
        # Build the per-feature mapping dict
        woe_map[feature] = {}
        # Assign each observation to a frozen interval
        assigned: pd.Categorical = pd.cut(
            values, bins=edges, include_lowest=True, right=True
        )
        # Iterate over the frozen intervals
        for interval in assigned.categories:
            # Build the mask of observations in this interval
            mask: np.ndarray = assigned == interval
            # Count the negative and positive observations in the bin
            g: int = int(np.sum((y == 0) & mask))
            b: int = int(np.sum((y == 1) & mask))
            # Render the identifier through the shared helper
            identifier: str = _interval_label(
                float(interval.left), float(interval.right)
            )
            # Compute and store the bin's WoE
            woe_map[feature][identifier] = woe_from_counts(g, b, G, B)
    # Compute the WoE for every categorical feature over the UNION of
    # the declared and the observed level sets (Blueprint 4, step 1)
    for feature in categorical_features:
        # Validate the feature exists in the frame
        if feature not in train_df.columns:
            # Raise a descriptive error naming the missing feature
            raise ValueError(
                f"train_df missing categorical feature '{feature}'."
            )
        # Build the union of the declared and observed levels
        declared: List[str] = (
            declared_levels.get(feature, []) if declared_levels else []
        )
        # Collect the observed training levels
        observed: List[str] = sorted(
            train_df[feature].unique().astype(str).tolist()
        )
        # Form the union preserving a deterministic order
        union_levels: List[str] = sorted(set(declared) | set(observed))
        # Build the per-feature mapping dict
        woe_map[feature] = {}
        # Iterate over the union of the levels
        for level in union_levels:
            # Build the mask of observations with this level
            mask: np.ndarray = (
                train_df[feature].astype(str) == level
            ).to_numpy()
            # Count the negative and positive observations in the level
            g: int = int(np.sum((y == 0) & mask))
            b: int = int(np.sum((y == 1) & mask))
            # Compute and store the level's WoE; a declared-but-unseen
            # level receives the neutral 0.0 explicitly through the
            # shared helper (Blueprint 4, step 2)
            woe_map[feature][level] = woe_from_counts(g, b, G, B)
    # Assert the key coverage: every declared level and every frozen
    # bin must be present in the map (Blueprint 4, step 4)
    _assert_woe_key_coverage(woe_map, cut_points, declared_levels,
                             categorical_features)
    # Return the frozen WoE mapping
    return woe_map


def _assert_woe_key_coverage(
    woe_map: Dict[str, Dict[str, float]],
    cut_points: Dict[str, List[float]],
    declared_levels: Optional[Dict[str, List[str]]],
    categorical_features: Tuple[str, ...],
) -> None:
    """
    Assert every declared level and frozen bin is present in the map.

    Raises a descriptive ValueError naming any missing key, so the
    frozen WoE lookup is provably complete (Blueprint 4, step 4).

    Parameters
    ----------
    woe_map : Dict[str, Dict[str, float]]
        The constructed WoE mapping.
    cut_points : Dict[str, List[float]]
        The frozen bin boundaries.
    declared_levels : Optional[Dict[str, List[str]]]
        The declared level sets, or None.
    categorical_features : Tuple[str, ...]
        The categorical feature names.

    Raises
    ------
    ValueError
        If any declared level or frozen bin is missing from the map.
    """
    # Collect the missing keys
    missing: List[str] = []
    # Check every declared level of every categorical feature
    if declared_levels is not None:
        # Iterate over the categorical features
        for feature in categorical_features:
            # Iterate over the declared levels of the feature
            for level in declared_levels.get(feature, []):
                # Record a missing declared level
                if level not in woe_map.get(feature, {}):
                    # Append the missing key
                    missing.append(f"{feature}:{level}")
    # Check every frozen bin of every continuous feature
    for feature, edges in cut_points.items():
        # Iterate over the frozen edge pairs
        for i in range(len(edges) - 1):
            # Render the bin identifier through the shared helper
            key: str = _interval_label(
                float(edges[i]), float(edges[i + 1])
            )
            # Record a missing frozen bin
            if key not in woe_map.get(feature, {}):
                # Append the missing key
                missing.append(f"{feature}:{key}")
    # Raise on any missing key
    if missing:
        # Raise naming the missing keys
        raise ValueError(
            f"woe_map key coverage violated; missing keys: {missing}."
        )


# -------------------------------------------------------------------------------------------------------------------------------
# Task 5, Step 3: Apply frozen WoE mapping to validation and test partitions
# -------------------------------------------------------------------------------------------------------------------------------


def apply_woe_mapping(
    partition_frames: Dict[str, pd.DataFrame],
    woe_map: Dict[str, Dict[str, float]],
    cut_points: Dict[str, List[float]],
    feature_order: Tuple[str, ...],
) -> Tuple[Dict[str, pd.DataFrame], List[str]]:
    """
    Apply the frozen WoE mapping to every partition.

    Assigns continuous values to frozen bins and categorical values to
    frozen levels, mapping each through the frozen WoE table; any value
    never observed in training receives the neutral WoE of 0.0 and is
    logged.  The three matrices share identical column ordering.

    Parameters
    ----------
    partition_frames : Dict[str, pd.DataFrame]
        Frames keyed by 'train', 'val', 'test'.
    woe_map : Dict[str, Dict[str, float]]
        The frozen WoE mapping from Step 2.
    cut_points : Dict[str, List[float]]
        The frozen bin boundaries from Step 1.
    feature_order : Tuple[str, ...]
        The ordered logistic-branch feature names.

    Returns
    -------
    Tuple[Dict[str, pd.DataFrame], List[str]]
        The WoE-transformed matrices keyed by partition and the log of
        neutral-assignment events.

    Raises
    ------
    ValueError
        If a feature is missing from the mapping.
    """
    # Validate every ordered feature is present in the WoE mapping
    for feature in feature_order:
        # Assert the feature has a frozen mapping
        if feature not in woe_map:
            # Raise a descriptive error naming the missing feature
            raise ValueError(
                f"woe_map missing feature '{feature}'."
            )
    # Collect the neutral-assignment log entries
    neutral_log: List[str] = []
    # Build the transformed matrices keyed by partition
    matrices: Dict[str, pd.DataFrame] = {}
    # Iterate over the partition frames
    for partition, frame in partition_frames.items():
        # Build the transformed frame column by column
        columns: Dict[str, np.ndarray] = {}
        # Iterate over the ordered features
        for feature in feature_order:
            # Look up the feature's frozen mapping
            mapping: Dict[str, float] = woe_map[feature]
            # Route continuous features through the frozen bin edges
            if feature in cut_points:
                # Extract the feature values as a float array
                values: np.ndarray = frame[feature].to_numpy(dtype=np.float64)
                # Assign each value to a frozen interval
                assigned: pd.Categorical = pd.cut(
                    values,
                    bins=cut_points[feature],
                    include_lowest=True,
                    right=True,
                )
                # Build an interval-keyed lookup so the categorical's
                # Interval objects resolve to the frozen WoE scores; a
                # string-keyed map would silently produce NaN for every
                # observation because Series.map does not render labels
                interval_map: Dict[Any, float] = {
                    interval: mapping[
                        _interval_label(
                            float(interval.left), float(interval.right)
                        )
                    ]
                    for interval in assigned.categories
                }
                # Map each interval through the frozen WoE table
                mapped: pd.Series = pd.Series(assigned).map(interval_map)
            else:
                # Map each categorical value through the frozen table
                mapped = frame[feature].map(mapping)
            # Identify the neutral assignments (NaN after mapping)
            neutral_mask: np.ndarray = mapped.isna().to_numpy()
            # Log the neutral assignments for the audit trail
            if neutral_mask.any():
                # Record the count of neutral assignments
                neutral_log.append(
                    f"{partition}:{feature}:{int(neutral_mask.sum())}"
                )
                # Replace the neutral assignments with 0.0
                mapped = mapped.fillna(0.0)
            # Store the mapped column as a float array
            columns[feature] = mapped.to_numpy(dtype=np.float64)
        # Assemble the partition matrix with the declared column order
        matrices[partition] = pd.DataFrame(columns, columns=feature_order)
    # Assert identical column ordering across the three matrices
    train_columns: List[str] = list(matrices["train"].columns)
    # Verify the validation and test columns match the training columns
    for partition in ("val", "test"):
        # Assert the partition's columns equal the training columns
        if list(matrices[partition].columns) != train_columns:
            # Raise a descriptive error for the column drift
            raise ValueError(
                f"column ordering drift in partition '{partition}'."
            )
    # Return the transformed matrices and the neutral-assignment log
    return matrices, neutral_log


# -------------------------------------------------------------------------------------------------------------------------------
# Task 5, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


def orchestrate_fit_woe_encoding(
    config: Dict[str, Any],
    zindi_train_df: pd.DataFrame,
    zindi_y: np.ndarray,
    zindi_frames: Dict[str, pd.DataFrame],
    zindi_logistic_features: Tuple[str, ...],
    taiwan_train_df: pd.DataFrame,
    taiwan_y: np.ndarray,
    taiwan_frames: Dict[str, pd.DataFrame],
    taiwan_logistic_features: Tuple[str, ...],
    taiwan_continuous: Tuple[str, ...],
    taiwan_categorical: Tuple[str, ...],
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Orchestrate the monotonic binning and WoE encoding stage.

    Fits the monotonic bins and the WoE mapping on each training
    partition, applies them frozen to all partitions of both datasets,
    and persists the frozen artifacts to JSON.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated study configuration dictionary.
    zindi_train_df : pd.DataFrame
        The Zindi training partition.
    zindi_y : np.ndarray
        The Zindi training labels.
    zindi_frames : Dict[str, pd.DataFrame]
        The Zindi partition frames keyed by 'train', 'val', 'test'.
    zindi_logistic_features : Tuple[str, ...]
        The Zindi logistic-branch feature names.
    taiwan_train_df : pd.DataFrame
        The Taiwan training partition.
    taiwan_y : np.ndarray
        The Taiwan training labels.
    taiwan_frames : Dict[str, pd.DataFrame]
        The Taiwan partition frames keyed by 'train', 'val', 'test'.
    taiwan_logistic_features : Tuple[str, ...]
        The Taiwan logistic-branch feature names.
    taiwan_continuous : Tuple[str, ...]
        The Taiwan continuous feature names.
    taiwan_categorical : Tuple[str, ...]
        The Taiwan categorical feature names.
    output_dir : Path
        Directory where the frozen artifacts are persisted.

    Returns
    -------
    Dict[str, Any]
        A state-ready dict with 'zindi' and 'taiwan' blocks carrying
        cut_points, woe_map and matrices, plus 'neutral_log' and
        'output_dir'.
    """
    # Validate the output directory type before any filesystem work
    if not isinstance(output_dir, Path):
        # Raise a descriptive error for a non-Path output directory
        raise ValueError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Create the output directory including any missing parents
    output_dir.mkdir(parents=True, exist_ok=True)
    # Separate the Zindi continuous and categorical logistic features
    zindi_continuous: Tuple[str, ...] = tuple(
        f for f in zindi_logistic_features
        if f in ("household_size", "age_of_respondent")
    )
    zindi_categorical: Tuple[str, ...] = tuple(
        f for f in zindi_logistic_features if f not in zindi_continuous
    )
    # Step 1: fit the monotonic bins on the Zindi training partition
    zindi_bins: Dict[str, Any] = fit_monotonic_bins(
        zindi_train_df, zindi_y, zindi_continuous
    )
    # Step 2: compute the Zindi WoE mapping on the training partition
    zindi_woe_map: Dict[str, Dict[str, float]] = compute_woe_scores(
        zindi_train_df, zindi_y, zindi_bins["cut_points"],
        zindi_categorical,
        declared_levels=config["dataset_parameters"]["zindi_fsd"][
            "categorical_levels"
        ],
    )
    # Step 3: apply the frozen Zindi mapping to all three partitions
    zindi_matrices, neutral_log = apply_woe_mapping(
        zindi_frames, zindi_woe_map, zindi_bins["cut_points"],
        zindi_logistic_features,
    )
    # Fit the Taiwan bins on the Taiwan training partition
    taiwan_bins: Dict[str, Any] = fit_monotonic_bins(
        taiwan_train_df, taiwan_y, taiwan_continuous
    )
    # Compute the Taiwan WoE mapping on the training partition
    taiwan_woe_map: Dict[str, Dict[str, float]] = compute_woe_scores(
        taiwan_train_df, taiwan_y, taiwan_bins["cut_points"],
        taiwan_categorical,
        declared_levels=None,
    )
    # Apply the frozen Taiwan mapping to all three Taiwan partitions
    taiwan_matrices, taiwan_neutral = apply_woe_mapping(
        taiwan_frames, taiwan_woe_map, taiwan_bins["cut_points"],
        taiwan_logistic_features,
    )
    # Extend the neutral log with the Taiwan entries
    neutral_log.extend(taiwan_neutral)
    # Compose the frozen artifacts payload
    artifacts: Dict[str, Any] = {
        "zindi": {
            "cut_points": zindi_bins["cut_points"],
            "bin_records": zindi_bins["bin_records"],
            "woe_map": zindi_woe_map,
            "matrices": {
                partition: matrix.to_dict(orient="split")
                for partition, matrix in zindi_matrices.items()
            },
        },
        "taiwan": {
            "cut_points": taiwan_bins["cut_points"],
            "bin_records": taiwan_bins["bin_records"],
            "woe_map": taiwan_woe_map,
            "matrices": {
                partition: matrix.to_dict(orient="split")
                for partition, matrix in taiwan_matrices.items()
            },
        },
        "neutral_log": neutral_log,
    }
    # Build the artifacts path inside the output directory
    artifacts_path: Path = output_dir / "woe_artifacts.json"
    # Serialise the frozen artifacts to JSON with a stable layout
    artifacts_text: str = json.dumps(
        artifacts, indent=2, ensure_ascii=False, default=str
    )
    # Persist the frozen artifacts to the output directory
    artifacts_path.write_text(artifacts_text, encoding="utf-8")
    # Assert the artifacts file exists before returning
    if not artifacts_path.is_file():
        # Raise a descriptive error for the missing artifacts
        raise FileNotFoundError(
            f"woe artifacts missing at {artifacts_path}."
        )
    # Return the state-ready WoE encoding outputs
    return {
        "zindi": {
            "cut_points": zindi_bins["cut_points"],
            "bin_records": zindi_bins["bin_records"],
            "woe_map": zindi_woe_map,
            "matrices": zindi_matrices,
        },
        "taiwan": {
            "cut_points": taiwan_bins["cut_points"],
            "bin_records": taiwan_bins["bin_records"],
            "woe_map": taiwan_woe_map,
            "matrices": taiwan_matrices,
        },
        "neutral_log": neutral_log,
        "output_dir": str(output_dir),
    }
````

````python
# Task 6: IV Filtering and One-Hot Encoding (Boosting Branch)

# ==============================================================================
# Task 6: IV Filtering and One-Hot Encoding (Boosting Branch)
# ==============================================================================


# -------------------------------------------------------------------------------------------------------------------------------
# Task 6, Step 1: Compute Information Values and apply the feature-selection filter
# -------------------------------------------------------------------------------------------------------------------------------


def compute_information_values(
    train_df: pd.DataFrame,
    y: np.ndarray,
    woe_map: Dict[str, Dict[str, float]],
    cut_points: Dict[str, List[float]],
) -> Dict[str, float]:
    """
    Compute the Information Value of every feature on training.

    Implements IV = sum_k ((g_k/G) - (b_k/B)) * WoE_k over the frozen
    bin and level identifiers of the woe_map, with g_k and b_k
    recomputed on the training partition.

    Parameters
    ----------
    train_df : pd.DataFrame
        The training partition of the cleansed frame.
    y : np.ndarray
        The training label vector, positionally aligned.
    woe_map : Dict[str, Dict[str, float]]
        The frozen WoE mapping from Task 5.
    cut_points : Dict[str, List[float]]
        The frozen bin boundaries from Task 5.

    Returns
    -------
    Dict[str, float]
        Feature name -> Information Value.

    Raises
    ------
    ValueError
        If the label length mismatches or a feature is missing.
    """
    # Validate the label length matches the frame row count
    if len(y) != len(train_df):
        # Raise a descriptive error for the alignment violation
        raise ValueError(
            f"label length {len(y)} must equal frame rows {len(train_df)}."
        )
    # Compute the total class counts on the training partition
    G: int = int(np.sum(y == 0))
    B: int = int(np.sum(y == 1))
    # Collect the per-feature IV values
    iv_values: Dict[str, float] = {}
    # Iterate over the features of the frozen WoE mapping
    for feature, mapping in woe_map.items():
        # Validate the feature exists in the frame
        if feature not in train_df.columns:
            # Raise a descriptive error naming the missing feature
            raise ValueError(
                f"train_df missing feature '{feature}'."
            )
        # Accumulate the IV contributions for the feature
        iv: float = 0.0
        # Route continuous features through the frozen bin edges
        if feature in cut_points:
            # Extract the feature values as a float array
            values: np.ndarray = train_df[feature].to_numpy(dtype=np.float64)
            # Assign each value to a frozen interval
            assigned: pd.Categorical = pd.cut(
                values,
                bins=cut_points[feature],
                include_lowest=True,
                right=True,
            )
            # Iterate over the frozen intervals
            for interval in assigned.categories:
                # Build the mask of observations in this interval
                mask: np.ndarray = assigned == interval
                # Count the negative and positive observations
                g: int = int(np.sum((y == 0) & mask))
                b: int = int(np.sum((y == 1) & mask))
                # Render the identifier through the SHARED helper from
                # Task 5, so the IV keys can never drift from the WoE
                # mapping keys (assessment remedy)
                identifier: str = _interval_label(
                    float(interval.left), float(interval.right)
                )
                # Accumulate the IV contribution with explicit error
                # raising on degenerate divisions (numerical-safety
                # standard): (g/G - b/B) * WoE_k
                with np.errstate(divide="raise", invalid="raise"):
                    # Compute the weighted contribution of the bin
                    contribution: float = (g / G - b / B) * mapping[
                        identifier
                    ]
                # Accumulate the bin contribution into the IV
                iv += contribution
        else:
            # Iterate over the observed levels of the categorical feature
            for level in sorted(train_df[feature].unique()):
                # Build the mask of observations with this level
                mask = (train_df[feature] == level).to_numpy()
                # Count the negative and positive observations
                g = int(np.sum((y == 0) & mask))
                b = int(np.sum((y == 1) & mask))
                # Accumulate the IV contribution with explicit error
                # raising on degenerate divisions: (g/G - b/B) * WoE_k
                with np.errstate(divide="raise", invalid="raise"):
                    # Compute the weighted contribution of the level
                    contribution = (g / G - b / B) * mapping[str(level)]
                # Accumulate the level contribution into the IV
                iv += contribution
        # Record the feature's Information Value
        iv_values[feature] = iv
    # Return the per-feature IV dict
    return iv_values


def apply_iv_filter(
    iv_values: Dict[str, float], threshold: float
) -> List[str]:
    """
    Retain the features whose IV meets the selection threshold.
    """
    # Validate the threshold is a non-negative number
    if threshold < 0:
        # Raise a descriptive error for a negative threshold
        raise ValueError(
            f"IV threshold must be non-negative; received {threshold}."
        )
    # Retain the features with IV >= threshold in mapping order
    retained: List[str] = [
        feature
        for feature, iv in iv_values.items()
        if iv >= threshold
    ]
    # Log the retention decision for the audit trail
    logger.info(
        f"IV filter: retained {len(retained)} of {len(iv_values)} "
        f"features at threshold {threshold}"
    )
    # Return the retained feature list
    return retained


def assert_zindi_iv_targets(
    iv_values: Dict[str, float], tolerance: float = 0.02
) -> None:
    """
    Assert the Zindi top-3 IV targets from the manuscript.

    Asserts that education_level, cellphone_access and job_type carry
    IV values of approximately 0.97, 0.71 and 0.63 within the given
    absolute tolerance.

    Parameters
    ----------
    iv_values : Dict[str, float]
        The computed per-feature IV values.
    tolerance : float, optional
        Absolute tolerance for each target (0.02 per the spec).

    Raises
    ------
    ValueError
        If any target IV deviates beyond the tolerance.
    """
    # Declare the manuscript target values for the three strongest IVs
    targets: Dict[str, float] = {
        "education_level": 0.97,
        "cellphone_access": 0.71,
        "job_type": 0.63,
    }
    # Verify each target feature is present in the IV dict
    for feature, target in targets.items():
        # Assert the feature was computed
        if feature not in iv_values:
            # Raise a descriptive error naming the missing feature
            raise ValueError(
                f"IV target feature '{feature}' missing from iv_values."
            )
        # Assert the computed IV lies within the tolerance of the target
        if abs(iv_values[feature] - target) > tolerance:
            # Raise naming the received and expected values
            raise ValueError(
                f"IV target violation: '{feature}' expected {target} "
                f"within {tolerance}; received {iv_values[feature]:.4f}. "
                f"Run the fidelity-diagnosis checklist: split seed, "
                f"binning algorithm, rare-level handling."
            )
    # Log the target verification for the audit trail
    logger.info("Zindi IV targets verified within tolerance 0.02")


# -------------------------------------------------------------------------------------------------------------------------------
# Task 6, Step 2: One-hot encode categorical features on training
# -------------------------------------------------------------------------------------------------------------------------------


def fit_one_hot_encoding(
    train_df: pd.DataFrame,
    categorical_features: Tuple[str, ...],
    continuous_features: Tuple[str, ...],
) -> Tuple[pd.DataFrame, Dict[str, List[str]], List[str]]:
    """
    Fit the one-hot encoding on the training partition.

    Expands every categorical feature into binary indicator columns with
    drop_first=False, appends the continuous features unchanged, and
    freezes the sorted unique level sets and the exact column order.

    Parameters
    ----------
    train_df : pd.DataFrame
        The training partition of the cleansed frame.
    categorical_features : Tuple[str, ...]
        The categorical feature names for the boosting branch.
    continuous_features : Tuple[str, ...]
        The continuous feature names for the boosting branch.

    Returns
    -------
    Tuple[pd.DataFrame, Dict[str, List[str]], List[str]]
        The training matrix, the frozen level sets, and the frozen
        column list.

    Raises
    ------
    ValueError
        If a declared feature is missing from the frame.
    """
    # Validate every declared feature is present in the frame
    for feature in categorical_features + continuous_features:
        # Assert the feature column exists
        if feature not in train_df.columns:
            # Raise a descriptive error naming the missing feature
            raise ValueError(
                f"train_df missing boosting feature '{feature}'."
            )
    # Freeze the sorted unique level set per categorical feature
    level_sets: Dict[str, List[str]] = {
        feature: sorted(train_df[feature].unique().tolist())
        for feature in categorical_features
    }
    # Expand the categorical features into indicator columns
    ohe_categorical: pd.DataFrame = pd.get_dummies(
        train_df[list(categorical_features)], drop_first=False
    )
    # Append the continuous features unchanged
    ohe_continuous: pd.DataFrame = train_df[list(continuous_features)].copy()
    # Compose the full training matrix
    matrix: pd.DataFrame = pd.concat(
        [ohe_categorical, ohe_continuous], axis=1
    )
    # Freeze the exact column order of the training matrix
    column_list: List[str] = matrix.columns.tolist()
    # Assert the matrix is free of NaN values
    if matrix.isna().any().any():
        # Raise a descriptive error for the NaN contamination
        raise ValueError(
            "one-hot training matrix contains NaN values."
        )
    # Log the fitted matrix geometry for the audit trail
    logger.info(
        f"one-hot training matrix shape {matrix.shape}, "
        f"{len(column_list)} columns"
    )
    # Return the matrix, the level sets and the column list
    return matrix, level_sets, column_list


# -------------------------------------------------------------------------------------------------------------------------------
# Task 6, Step 3: Apply the frozen one-hot encoding to validation and test
# -------------------------------------------------------------------------------------------------------------------------------


def apply_one_hot_encoding(
    partition_frames: Dict[str, pd.DataFrame],
    categorical_features: Tuple[str, ...],
    continuous_features: Tuple[str, ...],
    column_list: List[str],
) -> Dict[str, pd.DataFrame]:
    """
    Apply the frozen one-hot encoding to every partition.

    Expands each partition, reindexes the expansion against the frozen
    training column list with zero fill (an unseen level produces an
    all-zero indicator block), appends the continuous features in the
    frozen order, and asserts column equality with the training matrix.

    Parameters
    ----------
    partition_frames : Dict[str, pd.DataFrame]
        Frames keyed by 'train', 'val', 'test'.
    categorical_features : Tuple[str, ...]
        The categorical feature names for the boosting branch.
    continuous_features : Tuple[str, ...]
        The continuous feature names for the boosting branch.
    column_list : List[str]
        The frozen training column list.

    Returns
    -------
    Dict[str, pd.DataFrame]
        The one-hot matrices keyed by partition.

    Raises
    ------
    ValueError
        If a partition's columns drift from the frozen list.
    """
    # Build the transformed matrices keyed by partition
    matrices: Dict[str, pd.DataFrame] = {}
    # Iterate over the partition frames
    for partition, frame in partition_frames.items():
        # Expand the partition's categorical features into indicators
        ohe_categorical: pd.DataFrame = pd.get_dummies(
            frame[list(categorical_features)], drop_first=False
        )
        # Append the continuous features in the frozen order
        ohe_continuous: pd.DataFrame = frame[list(continuous_features)].copy()
        # Compose the partition's unaligned matrix
        unaligned: pd.DataFrame = pd.concat(
            [ohe_categorical, ohe_continuous], axis=1
        )
        # Reindex against the frozen training column list with zero fill;
        # unseen levels produce extra columns that are dropped here,
        # leaving the all-zero indicator block for the frozen columns
        aligned: pd.DataFrame = unaligned.reindex(
            columns=column_list, fill_value=0
        )
        # Assert the aligned columns equal the frozen list exactly
        if not aligned.columns.equals(pd.Index(column_list)):
            # Raise a descriptive error for the column drift
            raise ValueError(
                f"column drift in partition '{partition}' after reindex."
            )
        # Assert the matrix is free of NaN values
        if aligned.isna().any().any():
            # Raise a descriptive error for the NaN contamination
            raise ValueError(
                f"one-hot matrix for '{partition}' contains NaN values."
            )
        # Store the aligned matrix for the partition
        matrices[partition] = aligned
    # Return the aligned one-hot matrices
    return matrices


# -------------------------------------------------------------------------------------------------------------------------------
# Task 6, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


def orchestrate_iv_filter_and_ohe(
    config: Dict[str, Any],
    train_df: pd.DataFrame,
    y: np.ndarray,
    woe_map: Dict[str, Dict[str, float]],
    cut_points: Dict[str, List[float]],
    woe_matrices: Dict[str, pd.DataFrame],
    partition_frames: Dict[str, pd.DataFrame],
    categorical_features: Tuple[str, ...],
    continuous_features: Tuple[str, ...],
    output_dir: Path,
    verify_manuscript_targets: bool = True,
) -> Dict[str, Any]:
    """
    Orchestrate the IV filtering and one-hot encoding stage.

    Computes the IVs, applies the retention threshold, filters the WoE
    matrices to the retained columns, fits the one-hot encoding on
    training and applies it frozen to all partitions, then persists the
    artifacts.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated study configuration dictionary.
    train_df : pd.DataFrame
        The training partition of the cleansed frame.
    y : np.ndarray
        The training label vector.
    woe_map : Dict[str, Dict[str, float]]
        The frozen WoE mapping from Task 5.
    cut_points : Dict[str, List[float]]
        The frozen bin boundaries from Task 5.
    woe_matrices : Dict[str, pd.DataFrame]
        The WoE-transformed matrices from Task 5.
    partition_frames : Dict[str, pd.DataFrame]
        The partition frames keyed by 'train', 'val', 'test'.
    categorical_features : Tuple[str, ...]
        The boosting-branch categorical features.
    continuous_features : Tuple[str, ...]
        The boosting-branch continuous features.
    output_dir : Path
        Directory where the artifacts are persisted.
    verify_manuscript_targets : bool, optional
        Whether to assert the manuscript Zindi IV targets (True for
        real-data runs; False for synthetic smoke runs).

    Returns
    -------
    Dict[str, Any]
        A state-ready dict with iv_values, retained_features, level_sets,
        ohe_matrices, filtered_woe_matrices and output_dir.
    """
    # Validate the output directory type before any filesystem work
    if not isinstance(output_dir, Path):
        # Raise a descriptive error for a non-Path output directory
        raise ValueError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Create the output directory including any missing parents
    output_dir.mkdir(parents=True, exist_ok=True)
    # Step 1: compute the per-feature Information Values on training
    iv_values: Dict[str, float] = compute_information_values(
        train_df, y, woe_map, cut_points
    )
    # Apply the IV retention threshold from the configuration
    threshold: float = float(
        config["preprocessing_parameters"]["information_value_filter"][
            "threshold"
        ]
    )
    # Retain the features meeting the threshold
    retained_features: List[str] = apply_iv_filter(iv_values, threshold)
    # Assert the manuscript Zindi IV targets when enabled
    if verify_manuscript_targets:
        # Assert the top-3 IV targets within the declared tolerance
        assert_zindi_iv_targets(iv_values)
    # Filter the WoE matrices to the retained feature columns
    filtered_woe: Dict[str, pd.DataFrame] = {
        partition: matrix[retained_features]
        for partition, matrix in woe_matrices.items()
    }
    # Assert the filtered column count equals the retained feature count
    for partition, matrix in filtered_woe.items():
        # Verify the column count matches the retained set
        if matrix.shape[1] != len(retained_features):
            # Raise a descriptive error for the count mismatch
            raise ValueError(
                f"WoE matrix '{partition}' has {matrix.shape[1]} columns "
                f"but the IV filter retained {len(retained_features)}."
            )
    # Step 2: fit the one-hot encoding on the training partition
    ohe_train, level_sets, column_list = fit_one_hot_encoding(
        partition_frames["train"], categorical_features, continuous_features
    )
    # Step 3: apply the frozen encoding to all three partitions
    ohe_matrices: Dict[str, pd.DataFrame] = apply_one_hot_encoding(
        partition_frames, categorical_features, continuous_features,
        column_list,
    )
    # Compose the persisted artifacts payload
    artifacts: Dict[str, Any] = {
        "iv_values": iv_values,
        "threshold": threshold,
        "retained_features": retained_features,
        "level_sets": level_sets,
        "column_list": column_list,
        "ohe_shapes": {
            partition: list(matrix.shape)
            for partition, matrix in ohe_matrices.items()
        },
    }
    # Build the artifacts path inside the output directory
    artifacts_path: Path = output_dir / "iv_ohe_artifacts.json"
    # Serialise the artifacts to JSON with a stable layout
    artifacts_text: str = json.dumps(
        artifacts, indent=2, ensure_ascii=False, default=str
    )
    # Persist the artifacts to the output directory
    artifacts_path.write_text(artifacts_text, encoding="utf-8")
    # Assert the artifacts file exists before returning
    if not artifacts_path.is_file():
        # Raise a descriptive error for the missing artifacts
        raise FileNotFoundError(
            f"iv/ohe artifacts missing at {artifacts_path}."
        )
    # Return the state-ready IV and one-hot outputs
    return {
        "iv_values": iv_values,
        "retained_features": retained_features,
        "level_sets": level_sets,
        "column_list": column_list,
        "ohe_matrices": ohe_matrices,
        "filtered_woe_matrices": filtered_woe,
        "output_dir": str(output_dir),
    }
````

````python
# Task 7: Fit Logistic Regression Baseline with C Selection

# ==============================================================================
# Task 7: Fit Logistic Regression Baseline with C Selection
# ==============================================================================


# -------------------------------------------------------------------------------------------------------------------------------
# Task 7, Step 1: Select C by 5-fold stratified cross-validation
# -------------------------------------------------------------------------------------------------------------------------------


def select_C_cv(
    X_woe_train: np.ndarray,
    y_train: np.ndarray,
    C_grid: List[float],
    seed: int,
    n_folds: int = 5,
    max_iter: int = 1000,
) -> Tuple[float, List[Dict[str, Any]]]:
    """
    Select the logistic C by 5-fold stratified cross-validation.

    For every C in the grid, fits LogisticRegression(solver='lbfgs',
    penalty='l2', class_weight='balanced', C=current, max_iter=1000) on
    each stratified fold's training rows, computes AUC on the fold's
    validation rows, averages the five fold AUCs, and selects the C with
    the highest mean validation AUC.

    Parameters
    ----------
    X_woe_train : np.ndarray
        The WoE-transformed training matrix, shape (n_train, p).
    y_train : np.ndarray
        The training label vector, shape (n_train,).
    C_grid : List[float]
        The candidate inverse-regularisation values.
    seed : int
        The global deterministic seed.
    n_folds : int, optional
        The number of stratified folds (5 per the manuscript).
    max_iter : int, optional
        The lbfgs iteration cap (1000, a documented placeholder).

    Returns
    -------
    Tuple[float, List[Dict[str, Any]]]
        The selected C and the full CV table with per-fold AUCs.

    Raises
    ------
    ValueError
        If the grid is empty or the input shapes are inconsistent.
    """
    # Validate the C grid is non-empty
    if not C_grid:
        # Raise a descriptive error for the empty grid
        raise ValueError("C_grid must contain at least one candidate.")
    # Validate the label length matches the matrix rows
    if len(y_train) != X_woe_train.shape[0]:
        # Raise a descriptive error for the alignment violation
        raise ValueError(
            f"label length {len(y_train)} must equal matrix rows "
            f"{X_woe_train.shape[0]}."
        )
    # Build the stratified fold splitter with the global seed
    splitter = StratifiedKFold(
        n_splits=n_folds, shuffle=True, random_state=seed
    )
    # Collect the CV table rows
    cv_table: List[Dict[str, Any]] = []
    # Iterate over the candidate C values
    for C in C_grid:
        # Collect the per-fold AUCs for this C
        fold_aucs: List[float] = []
        # Iterate over the stratified folds
        for train_idx, val_idx in splitter.split(X_woe_train, y_train):
            # Slice the fold training rows and labels
            X_fold_train: np.ndarray = X_woe_train[train_idx]
            y_fold_train: np.ndarray = y_train[train_idx]
            # Slice the fold validation rows and labels
            X_fold_val: np.ndarray = X_woe_train[val_idx]
            y_fold_val: np.ndarray = y_train[val_idx]
            # Fit the logistic regression on the fold training rows
            with warnings.catch_warnings():
                # Capture convergence warnings for the audit trail
                warnings.simplefilter("always")
                # Fit the regularised balanced logistic regression
                model = LogisticRegression(
                    solver="lbfgs",
                    penalty="l2",
                    class_weight="balanced",
                    C=C,
                    max_iter=max_iter,
                )
                # Fit the model on the fold training data
                model.fit(X_fold_train, y_fold_train)
                # Check whether lbfgs hit the iteration cap
                if model.n_iter_[0] >= max_iter:
                    # Log the convergence warning for the audit trail
                    logger.warning(
                        f"C={C}: lbfgs reached max_iter={max_iter} "
                        f"(n_iter={model.n_iter_[0]})"
                    )
            # Predict the positive-class probabilities on the fold
            p_fold: np.ndarray = model.predict_proba(X_fold_val)[:, 1]
            # Compute the fold validation AUC
            fold_auc: float = float(roc_auc_score(y_fold_val, p_fold))
            # Record the fold AUC
            fold_aucs.append(fold_auc)
        # Compute the mean AUC across the five folds
        mean_auc: float = float(np.mean(fold_aucs))
        # Record the CV table row for this C
        cv_table.append(
            {
                "C": float(C),
                "fold_aucs": fold_aucs,
                "mean_auc": mean_auc,
            }
        )
        # Log the C result for the audit trail
        logger.info(f"C={C}: mean AUC {mean_auc:.5f}")
    # Select the C with the highest mean validation AUC
    best_row: Dict[str, Any] = max(cv_table, key=lambda row: row["mean_auc"])
    # Extract the selected C from the best row
    selected_C: float = float(best_row["C"])
    # Log the selection decision for the audit trail
    logger.info(f"selected C={selected_C} (mean AUC {best_row['mean_auc']:.5f})")
    # Return the selected C and the full CV table
    return selected_C, cv_table


# -------------------------------------------------------------------------------------------------------------------------------
# Task 7, Step 2: Fit the final logistic regression on the full training partition
# -------------------------------------------------------------------------------------------------------------------------------


def fit_final_logistic(
    X_woe_train: np.ndarray,
    y_train: np.ndarray,
    C: float,
    max_iter: int = 1000,
) -> Tuple[LogisticRegression, np.ndarray, float]:
    """
    Fit the final logistic regression on the full training partition.

    Fits LogisticRegression with the selected C, solver lbfgs, penalty
    l2, class_weight balanced and max_iter 1000, and extracts the
    coefficient vector beta and the intercept beta_0 for adverse-action
    explanation use.

    Parameters
    ----------
    X_woe_train : np.ndarray
        The full WoE-transformed training matrix, shape (n_train, p).
    y_train : np.ndarray
        The training label vector, shape (n_train,).
    C : float
        The selected inverse-regularisation value.
    max_iter : int, optional
        The lbfgs iteration cap (1000, a documented placeholder).

    Returns
    -------
    Tuple[LogisticRegression, np.ndarray, float]
        The fitted estimator, the coefficient vector beta (shape (p,))
        and the intercept beta_0.

    Raises
    ------
    ValueError
        If the fitted model did not converge.
    """
    # Fit the final logistic regression with the selected C
    with warnings.catch_warnings():
        # Capture convergence warnings for the audit trail
        warnings.simplefilter("always")
        # Configure the final logistic regression estimator
        model = LogisticRegression(
            solver="lbfgs",
            penalty="l2",
            class_weight="balanced",
            C=C,
            max_iter=max_iter,
        )
        # Fit the model on the full training partition
        model.fit(X_woe_train, y_train)
        # Raise if lbfgs failed to converge within the iteration cap
        if model.n_iter_[0] >= max_iter:
            # Raise a descriptive error for the convergence failure
            raise ValueError(
                f"final logistic regression did not converge within "
                f"max_iter={max_iter} (n_iter={model.n_iter_[0]})."
            )
    # Extract the coefficient vector with a shape assertion
    coef_matrix: np.ndarray = model.coef_
    # Assert the coefficient matrix has the binary shape (1, p)
    if coef_matrix.shape[0] != 1:
        # Raise a descriptive error for the unexpected shape
        raise ValueError(
            f"expected binary coef_ shape (1, p); received {coef_matrix.shape}."
        )
    # Flatten the coefficient vector to shape (p,)
    coef: np.ndarray = coef_matrix[0]
    # Extract the intercept scalar
    intercept: float = float(model.intercept_[0])
    # Log the fitted model summary for the audit trail
    logger.info(
        f"final logistic fit: p={len(coef)} coefficients, "
        f"intercept={intercept:.5f}, converged"
    )
    # Return the estimator, coefficients and intercept
    return model, coef, intercept


# -------------------------------------------------------------------------------------------------------------------------------
# Task 7, Step 3: Generate p_LR for all partitions
# -------------------------------------------------------------------------------------------------------------------------------


def predict_lr_partitions(
    model: LogisticRegression,
    partition_matrices: Dict[str, np.ndarray],
    labels: Dict[str, np.ndarray],
) -> Tuple[Dict[str, np.ndarray], float]:
    """
    Generate p_LR for every partition from the fitted estimator.

    Computes p_LR(x) = 1 / (1 + exp(-(beta^T x + beta_0))) via the
    positive column of predict_proba for each partition, asserts all
    probabilities lie in the open interval (0, 1), and records the
    validation AUC.

    Parameters
    ----------
    model : LogisticRegression
        The fitted final logistic regression.
    partition_matrices : Dict[str, np.ndarray]
        The WoE matrices keyed by 'train', 'val', 'test'.
    labels : Dict[str, np.ndarray]
        The label vectors keyed by the same partition names.

    Returns
    -------
    Tuple[Dict[str, np.ndarray], float]
        The per-partition probability vectors and the validation AUC.

    Raises
    ------
    ValueError
        If any probability falls outside the open unit interval.
    """
    # Collect the per-partition probability vectors
    p_lr: Dict[str, np.ndarray] = {}
    # Iterate over the partition matrices
    for partition, matrix in partition_matrices.items():
        # Predict the positive-class probabilities (Equation 1)
        probabilities: np.ndarray = model.predict_proba(matrix)[:, 1]
        # Assert every probability lies in the open interval (0, 1)
        if not np.all((probabilities > 0.0) & (probabilities < 1.0)):
            # Raise a descriptive error for the bounds violation
            raise ValueError(
                f"p_LR for '{partition}' must lie in the open interval "
                f"(0, 1); bounds violated."
            )
        # Store the probability vector in partition row order
        p_lr[partition] = probabilities
    # Validate the validation labels are present for the AUC
    if "val" not in labels:
        # Raise a descriptive error naming the missing labels
        raise ValueError(
            "labels dict must carry a 'val' entry for the AUC."
        )
    # Compute the validation AUC of the fitted LR
    val_auc: float = float(
        roc_auc_score(labels["val"], p_lr["val"])
    )
    # Log the validation AUC for the audit trail
    logger.info(f"LR validation AUC: {val_auc:.5f}")
    # Return the probability vectors and the validation AUC
    return p_lr, val_auc


# -------------------------------------------------------------------------------------------------------------------------------
# Task 7, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


def orchestrate_fit_logistic(
    config: Dict[str, Any],
    X_woe_train: np.ndarray,
    y_train: np.ndarray,
    X_woe_val: np.ndarray,
    y_val: np.ndarray,
    X_woe_test: np.ndarray,
    y_test: np.ndarray,
    output_dir: Path,
    feature_names: Optional[List[str]] = None,
) -> Dict[str, Any]:
    """
    Orchestrate the logistic regression baseline fitting stage.

    Composes C selection, the final fit and probability generation,
    persisting the CV table and the fitted coefficients to JSON.  The
    test partition is scored only, never fitted.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated study configuration dictionary.
    X_woe_train : np.ndarray
        The filtered WoE training matrix.
    y_train : np.ndarray
        The training labels.
    X_woe_val : np.ndarray
        The filtered WoE validation matrix.
    y_val : np.ndarray
        The validation labels.
    X_woe_test : np.ndarray
        The filtered WoE test matrix.
    y_test : np.ndarray
        The test labels.
    output_dir : Path
        Directory where the artifacts are persisted.
    feature_names : Optional[List[str]], optional
        The WoE matrix column names for explanation use.

    Returns
    -------
    Dict[str, Any]
        A state-ready dict with C, cv_table, model, coef, intercept,
        p_lr, val_auc and output_dir.
    """
    # Validate the output directory type before any filesystem work
    if not isinstance(output_dir, Path):
        # Raise a descriptive error for a non-Path output directory
        raise ValueError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Create the output directory including any missing parents
    output_dir.mkdir(parents=True, exist_ok=True)
    # Extract the C grid from the configuration
    C_grid: List[float] = config["logistic_regression_parameters"]["C_grid"]
    # Extract the global seed from the configuration
    seed: int = int(config["partitioning_parameters"]["random_seed"])
    # Step 1: select C by 5-fold stratified cross-validation
    selected_C, cv_table = select_C_cv(
        X_woe_train, y_train, C_grid, seed
    )
    # Step 2: fit the final logistic regression on the full training set
    model, coef, intercept = fit_final_logistic(
        X_woe_train, y_train, selected_C
    )
    # Step 3: generate p_LR for all partitions and record the val AUC
    partition_matrices: Dict[str, np.ndarray] = {
        "train": X_woe_train,
        "val": X_woe_val,
        "test": X_woe_test,
    }
    # Compose the per-partition label vectors
    labels: Dict[str, np.ndarray] = {
        "train": y_train,
        "val": y_val,
        "test": y_test,
    }
    # Compute the per-partition probability vectors
    p_lr, val_auc = predict_lr_partitions(
        model, partition_matrices, labels
    )
    # Compose the persisted artifacts payload
    artifacts: Dict[str, Any] = {
        "C": selected_C,
        "cv_table": cv_table,
        "coef": coef.tolist(),
        "intercept": intercept,
        "feature_names": feature_names,
        "val_auc": val_auc,
    }
    # Build the artifacts path inside the output directory
    artifacts_path: Path = output_dir / "logistic_artifacts.json"
    # Serialise the artifacts to JSON with a stable layout
    artifacts_text: str = json.dumps(
        artifacts, indent=2, ensure_ascii=False, default=str
    )
    # Persist the artifacts to the output directory
    artifacts_path.write_text(artifacts_text, encoding="utf-8")
    # Return the state-ready logistic outputs
    return {
        "C": selected_C,
        "cv_table": cv_table,
        "model": model,
        "coef": coef,
        "intercept": intercept,
        "p_lr": p_lr,
        "val_auc": val_auc,
        "feature_names": feature_names,
        "output_dir": str(output_dir),
    }
````

````python
# Task 8: Compute Raw Response Residuals

# ==============================================================================
# Task 8: Compute Raw Response Residuals
# ==============================================================================


# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Step 1: Compute training residuals
# -------------------------------------------------------------------------------------------------------------------------------


def compute_training_residuals(
    y_train: np.ndarray, p_lr_train: np.ndarray
) -> np.ndarray:
    """
    Compute the raw response residuals on the training partition.

    Implements Equation (3): r_i = y_i - p_LR(x_i), elementwise.  The
    result is the regression target of the residual learner.

    Parameters
    ----------
    y_train : np.ndarray
        The training label vector, shape (n_train,).
    p_lr_train : np.ndarray
        The training LR probabilities, shape (n_train,).

    Returns
    -------
    np.ndarray
        The training residual vector, shape (n_train,).

    Raises
    ------
    ValueError
        If the inputs are misaligned or non-finite.
    """
    # Validate the two input vectors have equal length
    if len(y_train) != len(p_lr_train):
        # Raise a descriptive error for the alignment violation
        raise ValueError(
            f"y_train length {len(y_train)} must equal p_lr_train length "
            f"{len(p_lr_train)}."
        )
    # Validate both inputs are finite
    if not np.isfinite(y_train).all() or not np.isfinite(p_lr_train).all():
        # Raise a descriptive error for non-finite inputs
        raise ValueError(
            "y_train and p_lr_train must both be finite."
        )
    # Compute the raw response residual: r_i = y_i - p_LR(x_i) (Eq. 3)
    residuals: np.ndarray = y_train - p_lr_train
    # Log the residual computation for the audit trail
    logger.info(f"training residuals computed: n={len(residuals)}")
    # Return the training residual vector
    return residuals


# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Step 2: Compute validation and test residuals
# -------------------------------------------------------------------------------------------------------------------------------


def compute_partition_residuals(
    labels: Dict[str, np.ndarray],
    p_lr: Dict[str, np.ndarray],
) -> Dict[str, np.ndarray]:
    """
    Compute the raw response residuals for every partition.

    Applies Equation (3) to the validation and test partitions with
    their own probabilities.  These residuals are not used to fit the
    residual learner; they are retained for evaluation and for the
    alternating-minimization procedure.

    Parameters
    ----------
    labels : Dict[str, np.ndarray]
        The label vectors keyed by 'train', 'val', 'test'.
    p_lr : Dict[str, np.ndarray]
        The LR probability vectors keyed by the same partition names.

    Returns
    -------
    Dict[str, np.ndarray]
        The residual vectors keyed by partition.

    Raises
    ------
    ValueError
        If a partition is missing or its inputs are misaligned.
    """
    # Validate both dicts carry the same partition keys
    if set(labels) != set(p_lr):
        # Raise a descriptive error for the key mismatch
        raise ValueError(
            f"labels keys {set(labels)} must equal p_lr keys {set(p_lr)}."
        )
    # Collect the per-partition residual vectors
    residuals: Dict[str, np.ndarray] = {}
    # Iterate over the partitions
    for partition in labels:
        # Extract the partition's label and probability vectors
        y_part: np.ndarray = labels[partition]
        p_part: np.ndarray = p_lr[partition]
        # Validate the two vectors have equal length
        if len(y_part) != len(p_part):
            # Raise a descriptive error for the alignment violation
            raise ValueError(
                f"partition '{partition}': label length {len(y_part)} "
                f"must equal p_lr length {len(p_part)}."
            )
        # Compute the raw response residual: r_i = y_i - p_LR(x_i) (Eq. 3)
        residuals[partition] = y_part - p_part
        # Record the partition summary statistics for the audit
        logger.info(
            f"partition '{partition}': residual mean "
            f"{float(np.mean(residuals[partition])):.5f}, std "
            f"{float(np.std(residuals[partition])):.5f}"
        )
    # Return the per-partition residual vectors
    return residuals


# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Step 3: Validate residual distributions
# -------------------------------------------------------------------------------------------------------------------------------


def validate_residual_distribution(
    r_train: np.ndarray,
) -> Dict[str, float]:
    """
    Validate the training residual distribution.

    Asserts all residuals lie in [-1, 1] (mathematically guaranteed
    since y in {0, 1} and p_LR in (0, 1)), computes the mean, standard
    deviation, minimum and maximum, and halts when the standard
    deviation is below 1e-3 because a constant LR would make residual
    learning meaningless.

    Parameters
    ----------
    r_train : np.ndarray
        The training residual vector, shape (n_train,).

    Returns
    -------
    Dict[str, float]
        The residual summary: mean, std, min, max.

    Raises
    ------
    ValueError
        If any residual lies outside [-1, 1] or the standard deviation
        is below 1e-3.
    """
    # Validate the residuals lie in [-1, 1] within float tolerance
    tolerance: float = 1e-9
    # Check the lower and upper bounds with the tolerance
    if np.any(r_train < -1.0 - tolerance) or np.any(r_train > 1.0 + tolerance):
        # Raise a descriptive error for the bounds violation
        raise ValueError(
            "training residuals must lie in [-1, 1]; bounds violated. "
            "This is mathematically guaranteed by y in {0, 1} and "
            "p_LR in (0, 1); a violation indicates an upstream error."
        )
    # Compute the residual summary statistics
    summary: Dict[str, float] = {
        "mean": float(np.mean(r_train)),
        "std": float(np.std(r_train)),
        "min": float(np.min(r_train)),
        "max": float(np.max(r_train)),
    }
    # Halt when the LR is essentially constant (degenerate residual)
    if summary["std"] < 1e-3:
        # Raise a descriptive error for the degenerate model
        raise ValueError(
            f"training residual standard deviation {summary['std']:.6f} "
            f"is below 1e-3; the LR is essentially constant and residual "
            f"learning is meaningless. Inspect the logistic baseline."
        )
    # Log the residual summary for the audit trail
    logger.info(f"training residual summary: {summary}")
    # Return the residual summary
    return summary


# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


def orchestrate_compute_residuals(
    labels: Dict[str, np.ndarray],
    p_lr: Dict[str, np.ndarray],
) -> Dict[str, Any]:
    """
    Orchestrate the raw response residual computation stage.

    Composes the training residual computation, the partition residual
    computation and the distribution validation, returning all three
    residual vectors plus the training summary.

    Parameters
    ----------
    labels : Dict[str, np.ndarray]
        The label vectors keyed by 'train', 'val', 'test'.
    p_lr : Dict[str, np.ndarray]
        The LR probability vectors keyed by the same partition names.

    Returns
    -------
    Dict[str, Any]
        A state-ready dict with 'residuals' (per partition) and
        'residual_summary' (training distribution).
    """
    # Step 1: compute the training residuals (regression target)
    r_train: np.ndarray = compute_training_residuals(
        labels["train"], p_lr["train"]
    )
    # Step 2: compute the validation and test residuals
    residuals: Dict[str, np.ndarray] = compute_partition_residuals(
        labels, p_lr
    )
    # Step 3: validate the training residual distribution
    residual_summary: Dict[str, float] = validate_residual_distribution(
        r_train
    )
    # Return the state-ready residual outputs
    return {
        "residuals": residuals,
        "residual_summary": residual_summary,
    }
````

````python
# Task 9: Fit Residual XGBoost Learner

# ==============================================================================
# Task 9: Fit Residual XGBoost Learner
# ==============================================================================


class _HybridAucEarlyStopping(xgb.callback.TrainingCallback):
    """
    Early-stopping callback monitoring the provisional hybrid AUC.

    At every boosting round the callback predicts r_hat on the
    validation matrix with iteration_range=(0, epoch + 1), forms the
    provisional hybrid p_LR(x) + r_hat(x), and computes its validation
    AUC.  Training halts after patience rounds without improvement, and
    the best iteration is recorded for freezing.
    """

    def __init__(
        self,
        dval: xgb.DMatrix,
        p_lr_val: np.ndarray,
        y_val: np.ndarray,
        patience: int = 50,
    ) -> None:
        """
        Initialise the callback with the hybrid monitor inputs.

        Parameters
        ----------
        dval : xgb.DMatrix
            The validation DMatrix for per-round predictions.
        p_lr_val : np.ndarray
            The validation LR probabilities for the hybrid.
        y_val : np.ndarray
            The validation labels for the AUC.
        patience : int, optional
            Rounds without improvement before halting (50 per config).
        """
        # Validate the patience is a positive integer
        if patience < 1:
            # Raise a descriptive error for the invalid patience
            raise ValueError(
                f"patience must be >= 1; received {patience}."
            )
        # Validate the monitor vector lengths agree
        if len(p_lr_val) != len(y_val):
            # Raise a descriptive error for the alignment violation
            raise ValueError(
                f"p_lr_val length {len(p_lr_val)} must equal y_val "
                f"length {len(y_val)}."
            )
        # Store the validation DMatrix for per-round predictions
        self.dval: xgb.DMatrix = dval
        # Store the validation LR probabilities
        self.p_lr_val: np.ndarray = p_lr_val
        # Store the validation labels
        self.y_val: np.ndarray = y_val
        # Store the patience window
        self.patience: int = patience
        # Track the best hybrid AUC observed so far
        self.best_auc: float = -1.0
        # Track the epoch at which the best AUC was observed
        self.best_iteration: int = 0
        # Track the epoch of the last improvement
        self._last_improvement: int = 0

    def after_iteration(
        self,
        model: Any,
        epoch: int,
        evals_log: Dict[str, Dict[str, Any]],
    ) -> bool:
        """
        Monitor the hybrid AUC and signal early stopping.
        """
        # Predict r_hat up to the current round (Equation 4 partial sum)
        r_hat_val: np.ndarray = model.predict(
            self.dval, iteration_range=(0, epoch + 1)
        )
        # Form the provisional hybrid: p_LR(x) + r_hat(x) (Eq. 5 form)
        hybrid: np.ndarray = self.p_lr_val + r_hat_val
        # Compute the provisional hybrid validation AUC
        auc: float = float(roc_auc_score(self.y_val, hybrid))
        # Update the best observed AUC and its epoch
        if auc > self.best_auc:
            # Record the improved AUC
            self.best_auc = auc
            # Record the epoch of the improvement
            self.best_iteration = epoch
            # Record the last-improvement epoch
            self._last_improvement = epoch
        # Halt when the patience window elapsed without improvement
        if epoch - self._last_improvement >= self.patience:
            # Log the early-stopping decision for the audit trail
            logger.info(
                f"early stopping at epoch {epoch}: best hybrid AUC "
                f"{self.best_auc:.5f} at iteration {self.best_iteration}"
            )
            # Return True to stop training
            return True
        # Continue training
        return False


# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Step 1: Configure and fit the residual booster
# -------------------------------------------------------------------------------------------------------------------------------


def fit_residual_booster(
    X_ohe_train: np.ndarray,
    r_train: np.ndarray,
    X_ohe_val: np.ndarray,
    p_lr_val: np.ndarray,
    y_val: np.ndarray,
    config: Dict[str, Any],
) -> Tuple[xgb.Booster, int]:
    """
    Fit the residual XGBoost learner with hybrid-AUC early stopping.

    Configures the booster per the config pins (reg:squarederror, 400
    estimators cap, depth 3, learning rate 0.03, subsample 0.8,
    colsample_bytree 0.8, reg_lambda 2, fixed seed) and trains with a
    custom callback monitoring the provisional hybrid validation AUC
    p_LR(x) + r_hat(x) with patience 50, freezing the model at the best
    iteration.

    Parameters
    ----------
    X_ohe_train : np.ndarray
        The one-hot encoded training matrix.
    r_train : np.ndarray
        The training raw response residuals (Equation 3 targets).
    X_ohe_val : np.ndarray
        The one-hot encoded validation matrix.
    p_lr_val : np.ndarray
        The validation LR probabilities for the hybrid monitor.
    y_val : np.ndarray
        The validation labels for the hybrid monitor.
    config : Dict[str, Any]
        The validated study configuration dictionary.

    Returns
    -------
    Tuple[xgb.Booster, int]
        The fitted booster and the frozen best iteration.

    Raises
    ------
    ValueError
        If the input shapes are inconsistent.
    """
    # Validate the training matrix and residual lengths agree
    if len(r_train) != X_ohe_train.shape[0]:
        # Raise a descriptive error for the alignment violation
        raise ValueError(
            f"r_train length {len(r_train)} must equal matrix rows "
            f"{X_ohe_train.shape[0]}."
        )
    # Validate the validation inputs agree in length
    if not (len(y_val) == len(p_lr_val) == X_ohe_val.shape[0]):
        # Raise a descriptive error for the alignment violation
        raise ValueError(
            "validation labels, LR probabilities and matrix rows must "
            "agree in length."
        )
    # Extract the residual booster parameters from the configuration
    params: Dict[str, Any] = config["residual_xgboost_parameters"]
    # Compute the validation residuals as the eval-set labels
    r_val: np.ndarray = y_val - p_lr_val
    # Build the training DMatrix with the residual targets
    dtrain: xgb.DMatrix = xgb.DMatrix(X_ohe_train, label=r_train)
    # Build the validation DMatrix with the residual targets
    dval: xgb.DMatrix = xgb.DMatrix(X_ohe_val, label=r_val)
    # Extract the early-stopping patience from the configuration
    patience: int = int(params["early_stopping"]["patience"])
    # Construct the hybrid-AUC early-stopping callback
    early_stopper = _HybridAucEarlyStopping(
        dval=dval, p_lr_val=p_lr_val, y_val=y_val, patience=patience
    )
    # Compose the booster parameters from the configuration pins
    booster_params: Dict[str, Any] = {
        "objective": params["objective"],
        "max_depth": params["max_depth"],
        "learning_rate": params["nu"],
        "subsample": params["subsample"],
        "colsample_bytree": params["colsample_bytree"],
        "reg_lambda": params["reg_lambda"],
        "seed": int(config["partitioning_parameters"]["random_seed"]),
    }
    # Train the booster with the hybrid-AUC early stopping callback
    booster: xgb.Booster = xgb.train(
        params=booster_params,
        dtrain=dtrain,
        num_boost_round=params["n_estimators"],
        evals=[(dval, "val")],
        callbacks=[early_stopper],
        verbose_eval=False,
    )
    # Extract the frozen best iteration from the callback
    best_iteration: int = early_stopper.best_iteration
    # Log the fitted booster summary for the audit trail
    logger.info(
        f"residual booster fitted: best hybrid AUC "
        f"{early_stopper.best_auc:.5f} at iteration {best_iteration}"
    )
    # Return the fitted booster and the frozen best iteration
    return booster, best_iteration


# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Step 2: Generate r_hat(x) for all partitions
# -------------------------------------------------------------------------------------------------------------------------------


def predict_residuals(
    booster: xgb.Booster,
    best_iteration: int,
    partition_matrices: Dict[str, np.ndarray],
) -> Dict[str, np.ndarray]:
    """
    Generate r_hat(x) for every partition at the frozen iteration.

    Computes r_hat(x) = sum_{m=1}^{M*} nu * h_m(x) (Equation 4) with
    M* the frozen best iteration, via booster.predict with
    iteration_range=(0, best_iteration + 1).

    Parameters
    ----------
    booster : xgb.Booster
        The fitted residual booster.
    best_iteration : int
        The frozen best iteration from early stopping.
    partition_matrices : Dict[str, np.ndarray]
        The one-hot matrices keyed by 'train', 'val', 'test'.

    Returns
    -------
    Dict[str, np.ndarray]
        The r_hat vectors keyed by partition.

    Raises
    ------
    ValueError
        If any prediction vector contains NaN.
    """
    # Collect the per-partition prediction vectors
    r_hat: Dict[str, np.ndarray] = {}
    # Iterate over the partition matrices
    for partition, matrix in partition_matrices.items():
        # Build a DMatrix for the partition matrix
        dmatrix: xgb.DMatrix = xgb.DMatrix(matrix)
        # Predict with the frozen iteration range (Equation 4)
        predictions: np.ndarray = booster.predict(
            dmatrix, iteration_range=(0, best_iteration + 1)
        )
        # Assert the predictions are NaN-free
        if not np.isfinite(predictions).all():
            # Raise a descriptive error for the NaN contamination
            raise ValueError(
                f"r_hat predictions for '{partition}' contain non-finite "
                f"values."
            )
        # Store the prediction vector in partition row order
        r_hat[partition] = predictions
    # Log the actual number of trees used for the audit trail
    logger.info(
        f"residual predictions at best iteration {best_iteration} "
        f"({best_iteration + 1} trees)"
    )
    # Return the per-partition prediction vectors
    return r_hat


# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Step 3: Validate the residual learner's contribution
# -------------------------------------------------------------------------------------------------------------------------------


def validate_residual_contribution(
    r_train: np.ndarray,
    r_hat_train: np.ndarray,
    p_lr_val: np.ndarray,
    r_hat_val: np.ndarray,
    y_val: np.ndarray,
    lr_only_val_auc: float,
) -> Tuple[float, float]:
    """
    Validate that the residual learner adds discrimination.

    Computes the Pearson correlation between r and r_hat on the
    training partition (asserted positive) and the provisional hybrid
    validation AUC on p_LR(x) + r_hat(x), compared against the LR-only
    AUC.

    Parameters
    ----------
    r_train : np.ndarray
        The training residual targets.
    r_hat_train : np.ndarray
        The training r_hat predictions.
    p_lr_val : np.ndarray
        The validation LR probabilities.
    r_hat_val : np.ndarray
        The validation r_hat predictions.
    y_val : np.ndarray
        The validation labels.
    lr_only_val_auc : float
        The LR-only validation AUC for comparison.

    Returns
    -------
    Tuple[float, float]
        The training correlation and the provisional hybrid AUC.

    Raises
    ------
    ValueError
        If the correlation is non-positive or the inputs are degenerate.
    """
    # Validate the training vectors agree in length
    if len(r_train) != len(r_hat_train):
        # Raise a descriptive error for the alignment violation
        raise ValueError(
            f"r_train length {len(r_train)} must equal r_hat_train "
            f"length {len(r_hat_train)}."
        )
    # Guard against zero-variance inputs before the correlation
    if np.std(r_train) == 0.0 or np.std(r_hat_train) == 0.0:
        # Raise a descriptive error for the degenerate inputs
        raise ValueError(
            "zero-variance residual inputs make the correlation "
            "undefined."
        )
    # Compute the Pearson correlation between r and r_hat
    correlation: float = float(pearsonr(r_train, r_hat_train)[0])
    # Assert the correlation is positive (the learner must be learning)
    if not correlation > 0.0:
        # Raise a descriptive error naming the investigation targets
        raise ValueError(
            f"training correlation r(r, r_hat) must be positive; "
            f"received {correlation:.5f}. Investigate the booster "
            f"configuration and feature matrix alignment."
        )
    # Form the provisional hybrid validation scores
    hybrid_val: np.ndarray = p_lr_val + r_hat_val
    # Compute the provisional hybrid validation AUC
    provisional_auc: float = float(roc_auc_score(y_val, hybrid_val))
    # Log the contribution summary for the audit trail
    logger.info(
        f"corr(r, r_hat)={correlation:.5f}; provisional hybrid val AUC "
        f"{provisional_auc:.5f} vs LR-only {lr_only_val_auc:.5f}"
    )
    # Return the correlation and the provisional AUC
    return correlation, provisional_auc


# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


def orchestrate_fit_residual_xgboost(
    config: Dict[str, Any],
    X_ohe_matrices: Dict[str, np.ndarray],
    residuals: Dict[str, np.ndarray],
    p_lr: Dict[str, np.ndarray],
    labels: Dict[str, np.ndarray],
    lr_only_val_auc: float,
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Orchestrate the residual XGBoost learner fitting stage.

    Composes the hybrid-AUC-early-stopped fit, the frozen-iteration
    predictions and the contribution validation, persisting the
    hyperparameters and scalars to JSON.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated study configuration dictionary.
    X_ohe_matrices : Dict[str, np.ndarray]
        The one-hot matrices keyed by 'train', 'val', 'test'.
    residuals : Dict[str, np.ndarray]
        The residual vectors keyed by the same partition names.
    p_lr : Dict[str, np.ndarray]
        The LR probability vectors keyed by partition.
    labels : Dict[str, np.ndarray]
        The label vectors keyed by partition.
    lr_only_val_auc : float
        The LR-only validation AUC for comparison.
    output_dir : Path
        Directory where the artifacts are persisted.

    Returns
    -------
    Dict[str, Any]
        A state-ready dict with booster, best_iteration, r_hat,
        correlation, provisional_auc and output_dir.
    """
    # Validate the output directory type before any filesystem work
    if not isinstance(output_dir, Path):
        # Raise a descriptive error for a non-Path output directory
        raise ValueError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Create the output directory including any missing parents
    output_dir.mkdir(parents=True, exist_ok=True)
    # Step 1: fit the residual booster with hybrid-AUC early stopping
    booster, best_iteration = fit_residual_booster(
        X_ohe_matrices["train"],
        residuals["train"],
        X_ohe_matrices["val"],
        p_lr["val"],
        labels["val"],
        config,
    )
    # Step 2: generate r_hat for all partitions at the frozen iteration
    r_hat: Dict[str, np.ndarray] = predict_residuals(
        booster, best_iteration, X_ohe_matrices
    )
    # Step 3: validate the residual learner's contribution
    correlation, provisional_auc = validate_residual_contribution(
        residuals["train"],
        r_hat["train"],
        p_lr["val"],
        r_hat["val"],
        labels["val"],
        lr_only_val_auc,
    )
    # Compose the persisted artifacts payload
    artifacts: Dict[str, Any] = {
        "objective": config["residual_xgboost_parameters"]["objective"],
        "n_estimators": config["residual_xgboost_parameters"]["n_estimators"],
        "max_depth": config["residual_xgboost_parameters"]["max_depth"],
        "learning_rate": config["residual_xgboost_parameters"]["nu"],
        "subsample": config["residual_xgboost_parameters"]["subsample"],
        "colsample_bytree": config["residual_xgboost_parameters"][
            "colsample_bytree"
        ],
        "reg_lambda": config["residual_xgboost_parameters"]["reg_lambda"],
        "patience": config["residual_xgboost_parameters"]["early_stopping"][
            "patience"
        ],
        "best_iteration": best_iteration,
        "correlation_train": correlation,
        "provisional_hybrid_val_auc": provisional_auc,
        "lr_only_val_auc": lr_only_val_auc,
    }
    # Build the artifacts path inside the output directory
    artifacts_path: Path = output_dir / "residual_booster_artifacts.json"
    # Serialise the artifacts to JSON with a stable layout
    artifacts_text: str = json.dumps(
        artifacts, indent=2, ensure_ascii=False, default=str
    )
    # Persist the artifacts to the output directory
    artifacts_path.write_text(artifacts_text, encoding="utf-8")
    # Return the state-ready residual learner outputs
    return {
        "booster": booster,
        "best_iteration": best_iteration,
        "r_hat": r_hat,
        "correlation_train": correlation,
        "provisional_auc": provisional_auc,
        "output_dir": str(output_dir),
    }
````

````python
# Task 10: Fit Standalone XGBoost Comparator

# ==============================================================================
# Task 10: Fit Standalone XGBoost Comparator
# ==============================================================================


# -------------------------------------------------------------------------------------------------------------------------------
# Task 10, Step 1: Configure the standalone classifier
# -------------------------------------------------------------------------------------------------------------------------------


def configure_standalone(
    config: Dict[str, Any],
    n_negative: int,
    n_positive: int,
) -> Tuple[Dict[str, Any], float]:
    """
    Configure the standalone XGBoost classifier parameters.

    Builds the parameter dict from the config pins (400 estimators,
    depth 4, learning rate 0.03, subsample 0.8, colsample_bytree 0.8,
    reg_lambda 2, binary:logistic, eval_metric auc) and computes
    scale_pos_weight = n0 / n1 on the training partition.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated study configuration dictionary.
    n_negative : int
        The training count of y=0.
    n_positive : int
        The training count of y=1.

    Returns
    -------
    Tuple[Dict[str, Any], float]
        The classifier parameter dict and the scale_pos_weight value.

    Raises
    ------
    ValueError
        If the positive count is zero.
    """
    # Validate the positive count is non-zero before the ratio
    if n_positive <= 0:
        # Raise a descriptive error for the degenerate denominator
        raise ValueError(
            f"n_positive must be > 0 to compute scale_pos_weight; "
            f"received {n_positive}."
        )
    # Compute the positive-class weight: n0 / n1 (config formula)
    scale_pos_weight: float = n_negative / n_positive
    # Extract the standalone parameter block from the configuration
    standalone: Dict[str, Any] = config["standalone_xgboost_parameters"]
    # Compose the classifier parameter dict from the config pins
    params: Dict[str, Any] = {
        "n_estimators": standalone["n_estimators"],
        "max_depth": standalone["max_depth"],
        "learning_rate": standalone["learning_rate"],
        "subsample": standalone["subsample"],
        "colsample_bytree": standalone["colsample_bytree"],
        "reg_lambda": standalone["reg_lambda"],
        "objective": standalone["objective"],
        "eval_metric": standalone["eval_metric"],
        "scale_pos_weight": scale_pos_weight,
        "random_state": int(config["partitioning_parameters"]["random_seed"]),
    }
    # Log the exact weight value for the audit trail
    logger.info(
        f"standalone scale_pos_weight = n0/n1 = {scale_pos_weight:.4f}"
    )
    # Return the parameter dict and the weight
    return params, scale_pos_weight


# -------------------------------------------------------------------------------------------------------------------------------
# Task 10, Step 2: Fit and predict
# -------------------------------------------------------------------------------------------------------------------------------


def fit_and_predict(
    params: Dict[str, Any],
    X_ohe_train: np.ndarray,
    y_train: np.ndarray,
    partition_matrices: Dict[str, np.ndarray],
) -> Tuple[XGBClassifier, Dict[str, np.ndarray]]:
    """
    Fit the standalone classifier and predict on every partition.

    Fits XGBClassifier with the configured parameters on the one-hot
    training matrix and binary labels, then generates probability
    scores for training, validation and test via the positive column of
    predict_proba.  No post-hoc calibration is applied.

    Parameters
    ----------
    params : Dict[str, Any]
        The configured classifier parameters.
    X_ohe_train : np.ndarray
        The one-hot encoded training matrix.
    y_train : np.ndarray
        The training labels.
    partition_matrices : Dict[str, np.ndarray]
        The one-hot matrices keyed by 'train', 'val', 'test'.

    Returns
    -------
    Tuple[XGBClassifier, Dict[str, np.ndarray]]
        The fitted classifier and the per-partition probability vectors.

    Raises
    ------
    ValueError
        If the label length mismatches the training rows.
    """
    # Validate the training label length matches the matrix rows
    if len(y_train) != X_ohe_train.shape[0]:
        # Raise a descriptive error for the alignment violation
        raise ValueError(
            f"y_train length {len(y_train)} must equal matrix rows "
            f"{X_ohe_train.shape[0]}."
        )
    # Construct the standalone classifier from the parameters
    classifier: XGBClassifier = XGBClassifier(**params)
    # Fit the classifier on the training matrix and labels
    classifier.fit(X_ohe_train, y_train)
    # Collect the per-partition probability vectors
    probabilities: Dict[str, np.ndarray] = {}
    # Iterate over the partition matrices
    for partition, matrix in partition_matrices.items():
        # Predict the positive-class probabilities
        probs: np.ndarray = classifier.predict_proba(matrix)[:, 1]
        # Assert the probabilities are finite
        if not np.isfinite(probs).all():
            # Raise a descriptive error for non-finite probabilities
            raise ValueError(
                f"standalone probabilities for '{partition}' contain "
                f"non-finite values."
            )
        # Store the probability vector in partition row order
        probabilities[partition] = probs
    # Log the fit completion for the audit trail
    logger.info(
        f"standalone XGBoost fitted; probabilities generated for "
        f"{sorted(probabilities)}"
    )
    # Return the classifier and the probability vectors
    return classifier, probabilities


# -------------------------------------------------------------------------------------------------------------------------------
# Task 10, Step 3: Record the benchmark performance
# -------------------------------------------------------------------------------------------------------------------------------


def record_benchmark(
    probs_test: np.ndarray,
    y_test: np.ndarray,
    verify_manuscript_targets: bool = True,
    tolerance: float = 0.01,
) -> Dict[str, float]:
    """
    Record the standalone benchmark metrics on the test partition.

    Computes AUC, Brier and minority-class F1 at threshold 0.5, and
    asserts the manuscript targets (AUC 0.873, Brier 0.143, F1 0.513)
    within the given tolerance when verification is enabled.

    Parameters
    ----------
    probs_test : np.ndarray
        The standalone test probabilities.
    y_test : np.ndarray
        The test labels.
    verify_manuscript_targets : bool, optional
        Whether to assert the manuscript targets (True for real runs).
    tolerance : float, optional
        Absolute tolerance for each target (0.01 per the spec).

    Returns
    -------
    Dict[str, float]
        The metrics dict with auc, brier and f1.

    Raises
    ------
    ValueError
        If a manuscript target is violated when verification is on.
    """
    # Compute the test AUC (rank-order discrimination)
    auc: float = float(roc_auc_score(y_test, probs_test))
    # Compute the test Brier score
    brier: float = float(brier_score_loss(y_test, probs_test))
    # Build the binary predictions at threshold 0.5
    preds: np.ndarray = (probs_test >= 0.5).astype(np.int64)
    # Compute precision, recall and F1 per class at threshold 0.5
    precision, recall, f1_scores, _ = precision_recall_fscore_support(
        y_test, preds, labels=[0, 1], zero_division=0
    )
    # Select the minority-class F1 (the positive class index 1)
    f1: float = float(f1_scores[1])
    # Compose the metrics dict
    metrics: Dict[str, float] = {
        "auc": auc,
        "brier": brier,
        "f1": f1,
    }
    # Assert the manuscript targets when verification is enabled
    if verify_manuscript_targets:
        # Declare the manuscript target values
        targets: Dict[str, float] = {
            "auc": 0.873,
            "brier": 0.143,
            "f1": 0.513,
        }
        # Verify each target within the tolerance
        for name, target in targets.items():
            # Assert the computed metric lies within the tolerance
            if abs(metrics[name] - target) > tolerance:
                # Raise naming the fidelity-diagnosis checklist
                raise ValueError(
                    f"standalone XGBoost {name} expected {target} within "
                    f"{tolerance}; received {metrics[name]:.4f}. Run the "
                    f"fidelity-diagnosis checklist: split seed, binning "
                    f"algorithm, rare-level handling, calibration fitting, "
                    f"library versions, statistical-test implementation."
                )
    # Log the benchmark metrics for the audit trail
    logger.info(
        f"standalone test metrics: AUC {auc:.4f}, Brier {brier:.4f}, "
        f"F1 {f1:.4f}"
    )
    # Return the metrics dict
    return metrics


# -------------------------------------------------------------------------------------------------------------------------------
# Task 10, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


def orchestrate_fit_standalone_xgboost(
    config: Dict[str, Any],
    X_ohe_matrices: Dict[str, np.ndarray],
    labels: Dict[str, np.ndarray],
    output_dir: Path,
    verify_manuscript_targets: bool = True,
) -> Dict[str, Any]:
    """
    Orchestrate the standalone XGBoost comparator fitting stage.

    Composes the configuration, the fit-and-predict step and the
    benchmark recording, persisting the weight, metrics and
    hyperparameters to JSON.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated study configuration dictionary.
    X_ohe_matrices : Dict[str, np.ndarray]
        The one-hot matrices keyed by 'train', 'val', 'test'.
    labels : Dict[str, np.ndarray]
        The label vectors keyed by the same partition names.
    output_dir : Path
        Directory where the artifacts are persisted.
    verify_manuscript_targets : bool, optional
        Whether to assert the manuscript benchmark targets.

    Returns
    -------
    Dict[str, Any]
        A state-ready dict with classifier, probs, metrics,
        scale_pos_weight and output_dir.
    """
    # Validate the output directory type before any filesystem work
    if not isinstance(output_dir, Path):
        # Raise a descriptive error for a non-Path output directory
        raise ValueError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Create the output directory including any missing parents
    output_dir.mkdir(parents=True, exist_ok=True)
    # Validate the matrices and labels carry identical partition keys
    if set(X_ohe_matrices) != set(labels):
        # Raise a descriptive error for the key mismatch
        raise ValueError(
            f"matrix keys {set(X_ohe_matrices)} must equal label keys "
            f"{set(labels)}."
        )
    # Compute the training class counts for the weight
    n_negative: int = int(np.sum(labels["train"] == 0))
    n_positive: int = int(np.sum(labels["train"] == 1))
    # Step 1: configure the standalone classifier
    params, scale_pos_weight = configure_standalone(
        config, n_negative, n_positive
    )
    # Step 2: fit and predict on every partition
    classifier, probs = fit_and_predict(
        params, X_ohe_matrices["train"], labels["train"], X_ohe_matrices
    )
    # Step 3: record the benchmark performance on the test partition
    metrics: Dict[str, float] = record_benchmark(
        probs["test"], labels["test"], verify_manuscript_targets
    )
    # Compose the persisted artifacts payload
    artifacts: Dict[str, Any] = {
        "scale_pos_weight": scale_pos_weight,
        "metrics": metrics,
        "params": {key: value for key, value in params.items()},
    }
    # Build the artifacts path inside the output directory
    artifacts_path: Path = output_dir / "standalone_xgb_artifacts.json"
    # Serialise the artifacts to JSON with a stable layout
    artifacts_text: str = json.dumps(
        artifacts, indent=2, ensure_ascii=False, default=str
    )
    # Persist the artifacts to the output directory
    artifacts_path.write_text(artifacts_text, encoding="utf-8")
    # Return the state-ready standalone outputs
    return {
        "classifier": classifier,
        "probs": probs,
        "metrics": metrics,
        "scale_pos_weight": scale_pos_weight,
        "output_dir": str(output_dir),
    }
````

````python
# Task 11: Adaptive Weighting Hyperparameter Selection

# ==============================================================================
# Task 11: Adaptive Weighting Hyperparameter Selection
# ==============================================================================


# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Step 1: Implement the uncertainty proxy and weight mapping
# -------------------------------------------------------------------------------------------------------------------------------


def compute_uncertainty(p_lr: np.ndarray) -> np.ndarray:
    """
    Compute the LR uncertainty proxy via Equation (8).

    Implements u_LR(x) = 4 * p_LR(x) * (1 - p_LR(x)) in [0, 1].

    Parameters
    ----------
    p_lr : np.ndarray
        The LR probability vector, shape (n,), values in (0, 1).

    Returns
    -------
    np.ndarray
        The uncertainty vector u_LR, shape (n,), values in [0, 1].

    Raises
    ------
    ValueError
        If p_lr lies outside the open unit interval.
    """
    # Validate the probabilities lie in the open unit interval
    if not np.all((p_lr > 0.0) & (p_lr < 1.0)):
        # Raise a descriptive error for the bounds violation
        raise ValueError(
            "p_lr must lie in the open interval (0, 1) for the "
            "uncertainty proxy."
        )
    # Compute the Bernoulli-variance proxy: u = 4 p (1 - p) (Eq. 8)
    uncertainty: np.ndarray = 4.0 * p_lr * (1.0 - p_lr)
    # Assert the proxy lies in [0, 1] within float tolerance
    if not np.all((uncertainty >= 0.0) & (uncertainty <= 1.0 + 1e-9)):
        # Raise a descriptive error for the range violation
        raise ValueError(
            "u_LR must lie in [0, 1]; range violated."
        )
    # Return the uncertainty vector
    return uncertainty


def compute_adaptive_weight(
    p_lr: np.ndarray,
    alpha_min: float,
    alpha_max: float,
    gamma: float,
) -> np.ndarray:
    """
    Compute the adaptive ensemble weight via Equation (9).

    Implements alpha*(x) = alpha_min + (alpha_max - alpha_min) *
    u_LR(x)^gamma with gamma > 0.

    Parameters
    ----------
    p_lr : np.ndarray
        The LR probability vector, shape (n,).
    alpha_min : float
        The lower weight bound in [0, 1).
    alpha_max : float
        The upper weight bound in (alpha_min, 1].
    gamma : float
        The concavity exponent, strictly positive.

    Returns
    -------
    np.ndarray
        The adaptive weight vector, shape (n,), bounded by alpha_min
        and alpha_max.

    Raises
    ------
    ValueError
        If the hyperparameter constraints are violated.
    """
    # Validate the alpha bounds satisfy 0 <= min < max <= 1
    if not (0.0 <= alpha_min < alpha_max <= 1.0):
        # Raise a descriptive error for the bound violation
        raise ValueError(
            f"alpha bounds must satisfy 0 <= alpha_min < alpha_max <= 1; "
            f"received min={alpha_min}, max={alpha_max}."
        )
    # Validate the gamma exponent is strictly positive
    if not gamma > 0:
        # Raise a descriptive error for the exponent violation
        raise ValueError(
            f"gamma must be > 0; received {gamma}."
        )
    # Compute the uncertainty proxy via Equation (8)
    uncertainty: np.ndarray = compute_uncertainty(p_lr)
    # Compute the adaptive weight via Equation (9)
    weights: np.ndarray = (
        alpha_min + (alpha_max - alpha_min) * np.power(uncertainty, gamma)
    )
    # Assert the weights are bounded by the selected alpha bounds
    if not np.all((weights >= alpha_min - 1e-9) & (weights <= alpha_max + 1e-9)):
        # Raise a descriptive error for the bound violation
        raise ValueError(
            f"alpha* must be bounded by [{alpha_min}, {alpha_max}]; "
            f"bounds violated."
        )
    # Return the adaptive weight vector
    return weights


# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Step 2: Grid search over (alpha_min, alpha_max, gamma) with 5-fold stratified CV
# -------------------------------------------------------------------------------------------------------------------------------


def grid_search_adaptive_weights(
    X_woe_train: np.ndarray,
    y_train: np.ndarray,
    X_ohe_train: np.ndarray,
    selected_C: float,
    config: Dict[str, Any],
) -> Tuple[Tuple[float, float, float], List[Dict[str, Any]]]:
    """
    Grid-search the adaptive weighting triple by 5-fold stratified CV.

    Enumerates the 27 triples of the declared grids; for each triple,
    runs 5-fold stratified CV: fit the LR on the fold training rows,
    compute fold-training residuals, fit the residual booster on those
    residuals, compute the fold-validation hybrid prediction
    p_Hybrid = clip(p_LR + alpha* * r_hat, 0, 1) (Equation 5), and
    compute the fold-validation AUC.  Selects the triple with the
    maximum mean AUC.

    Parameters
    ----------
    X_woe_train : np.ndarray
        The WoE-transformed training matrix.
    y_train : np.ndarray
        The training labels.
    X_ohe_train : np.ndarray
        The one-hot encoded training matrix.
    selected_C : float
        The logistic C selected by Task 7.
    config : Dict[str, Any]
        The validated study configuration dictionary.

    Returns
    -------
    Tuple[Tuple[float, float, float], List[Dict[str, Any]]]
        The selected (alpha_min, alpha_max, gamma) triple and the full
        grid table.

    Raises
    ------
    ValueError
        If the label length mismatches the matrix rows.
    """
    # Validate the label length matches both matrices
    if len(y_train) != X_woe_train.shape[0] or len(y_train) != X_ohe_train.shape[0]:
        # Raise a descriptive error for the alignment violation
        raise ValueError(
            f"y_train length {len(y_train)} must equal both matrix row "
            f"counts ({X_woe_train.shape[0]}, {X_ohe_train.shape[0]})."
        )
    # Extract the adaptive weighting block from the configuration
    adaptive: Dict[str, Any] = config["adaptive_weighting_parameters"]
    # Extract the three grids from the configuration
    alpha_min_grid: List[float] = adaptive["alpha_min_grid"]
    alpha_max_grid: List[float] = adaptive["alpha_max_grid"]
    gamma_grid: List[float] = adaptive["gamma_grid"]
    # Extract the residual booster parameters for the fold fits
    residual_params: Dict[str, Any] = config["residual_xgboost_parameters"]
    # Extract the global seed for deterministic fold splits
    seed: int = int(config["partitioning_parameters"]["random_seed"])
    # Build the fixed 5-fold stratified splitter (resolution: the config
    # resolves the manuscript's 3-fold/5-fold inconsistency to 5-fold)
    splitter = StratifiedKFold(n_splits=5, shuffle=True, random_state=seed)
    # Log the CV-strategy resolution for the audit trail
    logger.info(
        "grid search uses 5-fold stratified CV for every triple; "
        "the config resolves the manuscript's 3-fold/5-fold "
        "inconsistency to 5-fold"
    )
    # Build the booster parameter dict for the fold fits
    booster_params: Dict[str, Any] = {
        "objective": residual_params["objective"],
        "max_depth": residual_params["max_depth"],
        "learning_rate": residual_params["nu"],
        "subsample": residual_params["subsample"],
        "colsample_bytree": residual_params["colsample_bytree"],
        "reg_lambda": residual_params["reg_lambda"],
        "seed": seed,
    }
    # Collect the grid table rows
    grid_table: List[Dict[str, Any]] = []
    # Enumerate the full Cartesian product of the three grids
    for alpha_min, alpha_max, gamma in itertools.product(
        alpha_min_grid, alpha_max_grid, gamma_grid
    ):
        # Skip degenerate triples defensively (cannot occur in the grids)
        if not (0.0 <= alpha_min < alpha_max <= 1.0) or not gamma > 0:
            # Log the skipped degenerate triple
            logger.warning(f"skipping degenerate triple {(alpha_min, alpha_max, gamma)}")
            # Continue to the next triple
            continue
        # Collect the per-fold AUCs for this triple
        fold_aucs: List[float] = []
        # Iterate over the fixed stratified folds
        for fold_train, fold_val in splitter.split(X_woe_train, y_train):
            # Slice the fold training matrices and labels
            X_w_fold: np.ndarray = X_woe_train[fold_train]
            X_o_fold: np.ndarray = X_ohe_train[fold_train]
            y_fold: np.ndarray = y_train[fold_train]
            # Fit the LR on the fold training rows with the selected C
            lr_model = LogisticRegression(
                solver="lbfgs",
                penalty="l2",
                class_weight="balanced",
                C=selected_C,
                max_iter=1000,
            )
            # Fit the fold LR
            lr_model.fit(X_w_fold, y_fold)
            # Compute the fold-training LR probabilities
            p_lr_fold_train: np.ndarray = lr_model.predict_proba(X_w_fold)[:, 1]
            # Compute the fold-training residuals (Equation 3)
            r_fold: np.ndarray = y_fold - p_lr_fold_train
            # Build the fold-training DMatrix for the booster
            dtrain_fold: xgb.DMatrix = xgb.DMatrix(X_o_fold, label=r_fold)
            # Fit the residual booster on the fold-training residuals
            booster_fold: xgb.Booster = xgb.train(
                params=booster_params,
                dtrain=dtrain_fold,
                num_boost_round=residual_params["n_estimators"],
                verbose_eval=False,
            )
            # Predict the fold-validation residuals
            r_hat_val: np.ndarray = booster_fold.predict(
                xgb.DMatrix(X_ohe_train[fold_val])
            )
            # Compute the fold-validation LR probabilities
            p_lr_val: np.ndarray = lr_model.predict_proba(
                X_woe_train[fold_val]
            )[:, 1]
            # Compute the fold-validation uncertainty (Equation 8)
            u_val: np.ndarray = compute_uncertainty(p_lr_val)
            # Compute the fold-validation adaptive weight (Equation 9)
            alpha_val: np.ndarray = (
                alpha_min + (alpha_max - alpha_min) * np.power(u_val, gamma)
            )
            # Form the hybrid prediction (Equation 5 with clipping)
            hybrid_val: np.ndarray = np.clip(
                p_lr_val + alpha_val * r_hat_val, 0.0, 1.0
            )
            # Compute the fold-validation AUC
            fold_auc: float = float(
                roc_auc_score(y_train[fold_val], hybrid_val)
            )
            # Record the fold AUC
            fold_aucs.append(fold_auc)
        # Compute the mean AUC across the five folds
        mean_auc: float = float(np.mean(fold_aucs))
        # Record the grid table row for this triple
        grid_table.append(
            {
                "alpha_min": alpha_min,
                "alpha_max": alpha_max,
                "gamma": gamma,
                "fold_aucs": fold_aucs,
                "mean_auc": mean_auc,
            }
        )
        # Log the triple result for the audit trail
        logger.info(
            f"triple ({alpha_min}, {alpha_max}, {gamma}): "
            f"mean AUC {mean_auc:.5f}"
        )
    # Select the triple with the maximum mean validation AUC
    best_row: Dict[str, Any] = max(
        grid_table, key=lambda row: row["mean_auc"]
    )
    # Compose the selected triple
    selected_triple: Tuple[float, float, float] = (
        float(best_row["alpha_min"]),
        float(best_row["alpha_max"]),
        float(best_row["gamma"]),
    )
    # Log the selection for the audit trail
    logger.info(
        f"selected triple {selected_triple} "
        f"(mean AUC {best_row['mean_auc']:.5f})"
    )
    # Return the selected triple and the full grid table
    return selected_triple, grid_table


# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Step 3: Freeze the selected weights
# -------------------------------------------------------------------------------------------------------------------------------


def freeze_adaptive_weights(
    p_lr: Dict[str, np.ndarray],
    selected_triple: Tuple[float, float, float],
) -> Dict[str, np.ndarray]:
    """
    Freeze the adaptive weights from the FINAL LR probabilities.

    Computes u_LR and alpha* for every observation in all partitions
    using the final LR p_LR vectors and the selected triple.  The
    final-LR vectors must come from the full-training model, never from
    a fold model.

    Parameters
    ----------
    p_lr : Dict[str, np.ndarray]
        The final LR probability vectors keyed by partition.
    selected_triple : Tuple[float, float, float]
        The selected (alpha_min, alpha_max, gamma) triple.

    Returns
    -------
    Dict[str, np.ndarray]
        The alpha* arrays keyed by partition.

    Raises
    ------
    ValueError
        If the triple is degenerate.
    """
    # Unpack the selected triple
    alpha_min, alpha_max, gamma = selected_triple
    # Validate the triple satisfies the hyperparameter constraints
    if not (0.0 <= alpha_min < alpha_max <= 1.0) or not gamma > 0:
        # Raise a descriptive error for the degenerate triple
        raise ValueError(
            f"selected triple {(alpha_min, alpha_max, gamma)} is "
            f"degenerate."
        )
    # Collect the frozen alpha* arrays keyed by partition
    alpha_star: Dict[str, np.ndarray] = {}
    # Iterate over the final LR probability vectors
    for partition, p_part in p_lr.items():
        # Compute the partition's adaptive weight via Equation (9)
        alpha_star[partition] = compute_adaptive_weight(
            p_part, alpha_min, alpha_max, gamma
        )
    # Log the freezing for the audit trail
    logger.info(
        f"alpha* frozen for partitions {sorted(alpha_star)} from the "
        f"final LR model with triple {selected_triple}"
    )
    # Return the frozen alpha* arrays
    return alpha_star


# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


def orchestrate_select_adaptive_weights(
    config: Dict[str, Any],
    X_woe_train: np.ndarray,
    y_train: np.ndarray,
    X_ohe_train: np.ndarray,
    selected_C: float,
    p_lr: Dict[str, np.ndarray],
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Orchestrate the adaptive weighting hyperparameter selection.

    Composes the grid search and the weight freezing, persisting the
    grid table and the selected triple to JSON.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated study configuration dictionary.
    X_woe_train : np.ndarray
        The WoE-transformed training matrix.
    y_train : np.ndarray
        The training labels.
    X_ohe_train : np.ndarray
        The one-hot encoded training matrix.
    selected_C : float
        The logistic C selected by Task 7.
    p_lr : Dict[str, np.ndarray]
        The final LR probability vectors keyed by partition.
    output_dir : Path
        Directory where the artifacts are persisted.

    Returns
    -------
    Dict[str, Any]
        A state-ready dict with selected_triple, grid_table,
        alpha_star and output_dir.
    """
    # Validate the output directory type before any filesystem work
    if not isinstance(output_dir, Path):
        # Raise a descriptive error for a non-Path output directory
        raise ValueError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Create the output directory including any missing parents
    output_dir.mkdir(parents=True, exist_ok=True)
    # Step 2: grid-search the adaptive weighting triple
    selected_triple, grid_table = grid_search_adaptive_weights(
        X_woe_train, y_train, X_ohe_train, selected_C, config
    )
    # Step 3: freeze the adaptive weights from the final LR probabilities
    alpha_star: Dict[str, np.ndarray] = freeze_adaptive_weights(
        p_lr, selected_triple
    )
    # Compose the persisted artifacts payload
    artifacts: Dict[str, Any] = {
        "selected_triple": list(selected_triple),
        "grid_table": grid_table,
    }
    # Build the artifacts path inside the output directory
    artifacts_path: Path = output_dir / "adaptive_weights_artifacts.json"
    # Serialise the artifacts to JSON with a stable layout
    artifacts_text: str = json.dumps(
        artifacts, indent=2, ensure_ascii=False, default=str
    )
    # Persist the artifacts to the output directory
    artifacts_path.write_text(artifacts_text, encoding="utf-8")
    # Return the state-ready adaptive weighting outputs
    return {
        "selected_triple": selected_triple,
        "grid_table": grid_table,
        "alpha_star": alpha_star,
        "output_dir": str(output_dir),
    }
````

````python
# Task 12: Hybrid Prediction and Interpretability Decomposition

# ==============================================================================
# Task 12: Hybrid Prediction and Interpretability Decomposition
# ==============================================================================


# The interpretability region definitions from Equation (7)
RHO_FULLY: float = 0.9
RHO_PARTIAL: float = 0.7

# The region label strings used across the pipeline
REGION_FULLY: str = "Fully interpretable"
REGION_PARTIAL: str = "Partially interpretable"
REGION_ML: str = "ML-driven"


# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 1: Compute hybrid probabilities
# -------------------------------------------------------------------------------------------------------------------------------


def compute_hybrid_probabilities(
    p_lr: np.ndarray,
    r_hat: np.ndarray,
    alpha_star: np.ndarray,
) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    """
    Compute the hybrid probabilities via Equation (5).

    Implements p_Hybrid(x) = clip(p_LR(x) + alpha*(x) * r_hat(x), 0, 1)
    and returns the unclipped additive components for the rho
    computation.

    Parameters
    ----------
    p_lr : np.ndarray
        The LR probability vector, shape (n,).
    r_hat : np.ndarray
        The residual prediction vector, shape (n,).
    alpha_star : np.ndarray
        The adaptive weight vector, shape (n,).

    Returns
    -------
    Tuple[np.ndarray, np.ndarray, np.ndarray]
        The hybrid probabilities, the unclipped p_LR and the unclipped
        weighted residual alpha* * r_hat.

    Raises
    ------
    ValueError
        If the vectors are misaligned or non-finite.
    """
    # Validate the three vectors have equal length
    if not (len(p_lr) == len(r_hat) == len(alpha_star)):
        # Raise a descriptive error for the alignment violation
        raise ValueError(
            f"p_lr, r_hat and alpha_star must share length; received "
            f"{len(p_lr)}, {len(r_hat)}, {len(alpha_star)}."
        )
    # Validate the inputs are finite
    if not (
        np.isfinite(p_lr).all()
        and np.isfinite(r_hat).all()
        and np.isfinite(alpha_star).all()
    ):
        # Raise a descriptive error for non-finite inputs
        raise ValueError(
            "p_lr, r_hat and alpha_star must all be finite."
        )
    # Compute the weighted residual correction: alpha* * r_hat (Eq. 5)
    weighted_residual: np.ndarray = alpha_star * r_hat
    # Compute the unclipped hybrid sum (Equation 5 additive form)
    unclipped_sum: np.ndarray = p_lr + weighted_residual
    # Clip the sum to the unit interval (Equation 5)
    hybrid: np.ndarray = np.clip(unclipped_sum, 0.0, 1.0)
    # Assert the hybrid contains no NaN
    if np.isnan(hybrid).any():
        # Raise a descriptive error for the NaN contamination
        raise ValueError("hybrid probabilities contain NaN.")
    # Assert the hybrid lies in [0, 1]
    if not np.all((hybrid >= 0.0) & (hybrid <= 1.0)):
        # Raise a descriptive error for the range violation
        raise ValueError("hybrid probabilities must lie in [0, 1].")
    # Return the hybrid and the unclipped components
    return hybrid, p_lr, weighted_residual


# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 2: Compute the interpretability ratio and region assignments
# -------------------------------------------------------------------------------------------------------------------------------


def compute_rho_and_regions(
    p_lr_unclipped: np.ndarray,
    weighted_residual: np.ndarray,
) -> Tuple[np.ndarray, np.ndarray]:
    """
    Compute the interpretability ratio and region assignments.

    Implements Equation (6): rho(x) = |p_LR(x)| / (|p_LR(x)| +
    |alpha*(x) * r_hat(x)|), with rho = 1 under the zero-denominator
    guard, and assigns regions per Equation (7): Fully for rho >= 0.9,
    Partially for 0.7 <= rho < 0.9, ML-driven for rho < 0.7.

    Parameters
    ----------
    p_lr_unclipped : np.ndarray
        The unclipped LR probability vector, shape (n,).
    weighted_residual : np.ndarray
        The unclipped weighted residual vector, shape (n,).

    Returns
    -------
    Tuple[np.ndarray, np.ndarray]
        The rho vector and the region label vector.

    Raises
    ------
    ValueError
        If the components are misaligned or non-finite.
    """
    # Validate the two components have equal length
    if len(p_lr_unclipped) != len(weighted_residual):
        # Raise a descriptive error for the alignment violation
        raise ValueError(
            f"p_lr and weighted residual must share length; received "
            f"{len(p_lr_unclipped)} and {len(weighted_residual)}."
        )
    # Validate the components are finite
    if not (
        np.isfinite(p_lr_unclipped).all()
        and np.isfinite(weighted_residual).all()
    ):
        # Raise a descriptive error for non-finite components
        raise ValueError(
            "rho components must all be finite."
        )
    # Compute the absolute value of the linear share (Equation 6)
    abs_lr: np.ndarray = np.abs(p_lr_unclipped)
    # Compute the absolute value of the residual share (Equation 6)
    abs_residual: np.ndarray = np.abs(weighted_residual)
    # Compute the denominator of the ratio (Equation 6)
    denominator: np.ndarray = abs_lr + abs_residual
    # Apply the zero-denominator guard: rho = 1 below the threshold
    safe_denominator: np.ndarray = np.where(
        denominator < 1e-12, 1.0, denominator
    )
    # Compute the interpretability ratio (Equation 6)
    rho: np.ndarray = np.where(
        denominator < 1e-12, 1.0, abs_lr / safe_denominator
    )
    # Initialise the region label vector
    regions: np.ndarray = np.empty(len(rho), dtype=object)
    # Assign the Fully interpretable region: rho >= 0.9 (Equation 7)
    regions[rho >= RHO_FULLY] = REGION_FULLY
    # Assign the Partially interpretable region: 0.7 <= rho < 0.9 (Eq. 7)
    regions[(rho >= RHO_PARTIAL) & (rho < RHO_FULLY)] = REGION_PARTIAL
    # Assign the ML-driven region: rho < 0.7 (Equation 7)
    regions[rho < RHO_PARTIAL] = REGION_ML
    # Return the rho vector and the region labels
    return rho, regions


# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 3: Aggregate region statistics on test
# -------------------------------------------------------------------------------------------------------------------------------


def aggregate_region_stats(
    regions: np.ndarray,
    y: np.ndarray,
    rho: np.ndarray,
    verify_manuscript_targets: bool = True,
) -> Dict[str, Any]:
    """
    Aggregate the interpretability region statistics on the test set.

    Computes the share of observations in each region, the within-region
    positive-class rate, the per-region mean rho, the full rho
    distribution and per-region counts, and optionally asserts the
    manuscript Table 3 targets within 1 percentage point.

    Parameters
    ----------
    regions : np.ndarray
        The region label vector for the test partition.
    y : np.ndarray
        The test label vector.
    rho : np.ndarray
        The test rho vector.
    verify_manuscript_targets : bool, optional
        Whether to assert the Table 3 targets (True for real runs).

    Returns
    -------
    Dict[str, Any]
        The region statistics: shares, rates, mean rho, distribution
        and counts.

    Raises
    ------
    ValueError
        If the inputs are misaligned.
    """
    # Validate the inputs share length
    if not (len(regions) == len(y) == len(rho)):
        # Raise a descriptive error for the alignment violation
        raise ValueError(
            f"regions, y and rho must share length; received "
            f"{len(regions)}, {len(y)}, {len(rho)}."
        )
    # Assemble a temporary analysis frame positionally
    frame: pd.DataFrame = pd.DataFrame(
        {"region": regions, "y": y, "rho": rho}
    )
    # Group the frame by region
    grouped: pd.DataFrameGroupBy = frame.groupby("region")
    # Compute the per-region observation counts
    counts: Dict[str, int] = {
        str(name): int(subframe.shape[0])
        for name, subframe in grouped
    }
    # Compute the per-region observation shares
    shares: Dict[str, float] = {
        str(name): float(subframe.shape[0] / len(frame))
        for name, subframe in grouped
    }
    # Compute the per-region positive-class rates
    rates: Dict[str, float] = {}
    # Iterate over the groups to compute the label means
    for name, subframe in grouped:
        # Guard the rate against an empty region
        rates[str(name)] = (
            float(subframe["y"].mean()) if subframe.shape[0] > 0 else 0.0
        )
    # Compute the per-region mean rho
    mean_rho: Dict[str, float] = {
        str(name): float(subframe["rho"].mean())
        for name, subframe in grouped
    }
    # Compute the full rho distribution histogram on [0, 1]
    hist_counts, hist_edges = np.histogram(rho, bins=20, range=(0.0, 1.0))
    # Compose the region statistics payload
    stats: Dict[str, Any] = {
        "counts": counts,
        "shares": shares,
        "rates": rates,
        "mean_rho": mean_rho,
        "rho_distribution": {
            "hist_counts": hist_counts.tolist(),
            "hist_edges": hist_edges.tolist(),
        },
    }
    # Assert the manuscript Table 3 targets when verification is enabled
    if verify_manuscript_targets:
        # Declare the Table 3 target shares and rates
        target_shares: Dict[str, float] = {
            REGION_FULLY: 0.064,
            REGION_PARTIAL: 0.341,
            REGION_ML: 0.594,
        }
        target_rates: Dict[str, float] = {
            REGION_FULLY: 0.538,
            REGION_PARTIAL: 0.197,
            REGION_ML: 0.065,
        }
        # Verify each target share within 1 percentage point
        for region, target in target_shares.items():
            # Assert the share lies within the tolerance
            if abs(stats["shares"].get(region, 0.0) - target) > 0.01:
                # Raise naming the fidelity-diagnosis checklist
                raise ValueError(
                    f"Table 3 share for '{region}' expected {target} "
                    f"within 0.01; received {stats['shares'].get(region)}."
                )
        # Verify each target rate within 1 percentage point
        for region, target in target_rates.items():
            # Assert the rate lies within the tolerance
            if abs(stats["rates"].get(region, 0.0) - target) > 0.01:
                # Raise naming the fidelity-diagnosis checklist
                raise ValueError(
                    f"Table 3 rate for '{region}' expected {target} "
                    f"within 0.01; received {stats['rates'].get(region)}."
                )
    # Report the qualitative bimodality of the rho distribution
    primary_mode_bin: int = int(np.argmax(hist_counts[:14]))
    # Report the secondary-mode presence in the top bins
    secondary_mode_bin: int = int(np.argmax(hist_counts[14:]) + 14)
    # Log the distribution characterisation for the audit trail
    logger.info(
        f"rho distribution: primary mode bin {primary_mode_bin} "
        f"(<0.7), secondary mode bin {secondary_mode_bin} (near 1.0)"
    )
    # Return the region statistics
    return stats


# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


def orchestrate_build_hybrid_and_rho(
    config: Dict[str, Any],
    p_lr: Dict[str, np.ndarray],
    r_hat: Dict[str, np.ndarray],
    alpha_star: Dict[str, np.ndarray],
    labels: Dict[str, np.ndarray],
    output_dir: Path,
    verify_manuscript_targets: bool = True,
) -> Dict[str, Any]:
    """
    Orchestrate the hybrid prediction and interpretability stage.

    Composes the hybrid probability computation, the rho and region
    computation and the test region aggregation, persisting the region
    statistics to JSON.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated study configuration dictionary.
    p_lr : Dict[str, np.ndarray]
        The LR probability vectors keyed by partition.
    r_hat : Dict[str, np.ndarray]
        The residual prediction vectors keyed by partition.
    alpha_star : Dict[str, np.ndarray]
        The adaptive weight vectors keyed by partition.
    labels : Dict[str, np.ndarray]
        The label vectors keyed by the same partition names.
    output_dir : Path
        Directory where the artifacts are persisted.
    verify_manuscript_targets : bool, optional
        Whether to assert the Table 3 targets (True for real runs).

    Returns
    -------
    Dict[str, Any]
        A state-ready dict with p_hybrid, components, rho, regions and
        region_stats.
    """
    # Validate the output directory type before any filesystem work
    if not isinstance(output_dir, Path):
        # Raise a descriptive error for a non-Path output directory
        raise ValueError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Create the output directory including any missing parents
    output_dir.mkdir(parents=True, exist_ok=True)
    # Validate the input dicts carry identical partition keys
    key_sets: set = set(p_lr)
    # Assert the other dicts share the same key set
    for name, mapping in (
        ("r_hat", r_hat),
        ("alpha_star", alpha_star),
        ("labels", labels),
    ):
        # Assert the mapping's key set equals the reference set
        if set(mapping) != key_sets:
            # Raise a descriptive error for the key mismatch
            raise ValueError(
                f"{name} keys {set(mapping)} must equal p_lr keys "
                f"{key_sets}."
            )
    # Collect the per-partition hybrid and component vectors
    p_hybrid: Dict[str, np.ndarray] = {}
    components: Dict[str, Dict[str, np.ndarray]] = {}
    # Iterate over the partitions
    for partition in p_lr:
        # Step 1: compute the hybrid probabilities and components
        hybrid, p_unclipped, weighted_residual = (
            compute_hybrid_probabilities(
                p_lr[partition], r_hat[partition], alpha_star[partition]
            )
        )
        # Store the hybrid probabilities for the partition
        p_hybrid[partition] = hybrid
        # Store the unclipped components for the partition
        components[partition] = {
            "p_lr": p_unclipped,
            "weighted_residual": weighted_residual,
        }
    # Collect the per-partition rho and region vectors
    rho: Dict[str, np.ndarray] = {}
    regions: Dict[str, np.ndarray] = {}
    # Iterate over the partitions
    for partition in p_lr:
        # Step 2: compute the rho and region assignments
        rho_part, regions_part = compute_rho_and_regions(
            components[partition]["p_lr"],
            components[partition]["weighted_residual"],
        )
        # Store the rho vector for the partition
        rho[partition] = rho_part
        # Store the region vector for the partition
        regions[partition] = regions_part
    # Step 3: aggregate the region statistics on the test partition
    region_stats: Dict[str, Any] = aggregate_region_stats(
        regions["test"], labels["test"], rho["test"],
        verify_manuscript_targets,
    )
    # Build the artifacts path inside the output directory
    artifacts_path: Path = output_dir / "hybrid_rho_artifacts.json"
    # Serialise the region stats to JSON with a stable layout
    artifacts_text: str = json.dumps(
        region_stats, indent=2, ensure_ascii=False, default=str
    )
    # Persist the region stats to the output directory
    artifacts_path.write_text(artifacts_text, encoding="utf-8")
    # Return the state-ready hybrid and interpretability outputs
    return {
        "p_hybrid": p_hybrid,
        "components": components,
        "rho": rho,
        "regions": regions,
        "region_stats": region_stats,
        "output_dir": str(output_dir),
    }
````

````python
# Task 13: Platt Calibration

# ==============================================================================
# Task 13: Platt Calibration
# ==============================================================================


def compute_ece(
    probabilities: np.ndarray, y: np.ndarray, n_bins: int = 10
) -> float:
    """
    Compute the Expected Calibration Error via Equation (11).

    Implements ECE = sum_k (|B_k| / n) * |acc(B_k) - conf(B_k)| over K
    equal-width probability bins on [0, 1].

    Parameters
    ----------
    probabilities : np.ndarray
        The predicted probabilities, shape (n,).
    y : np.ndarray
        The binary labels, shape (n,).
    n_bins : int, optional
        The number of equal-width bins (10 per the manuscript).

    Returns
    -------
    float
        The Expected Calibration Error.
    """
    # Validate the inputs share length
    if len(probabilities) != len(y):
        # Raise a descriptive error for the alignment violation
        raise ValueError(
            f"probabilities length {len(probabilities)} must equal "
            f"labels length {len(y)}."
        )
    # Build the equal-width bin edges on [0, 1]
    edges: np.ndarray = np.linspace(0.0, 1.0, n_bins + 1)
    # Assign each probability to its bin (the last edge is inclusive)
    bin_ids: np.ndarray = np.clip(
        np.digitize(probabilities, edges[1:-1]), 0, n_bins - 1
    )
    # Accumulate the ECE contributions
    ece: float = 0.0
    # Iterate over the bins
    for k in range(n_bins):
        # Build the mask of observations in bin k
        mask: np.ndarray = bin_ids == k
        # Skip empty bins (zero weight contribution)
        if not mask.any():
            # Continue to the next bin
            continue
        # Compute the bin's empirical positive rate
        acc_k: float = float(np.mean(y[mask]))
        # Compute the bin's mean predicted probability
        conf_k: float = float(np.mean(probabilities[mask]))
        # Accumulate the weighted absolute gap (Equation 11)
        ece += (mask.sum() / len(y)) * abs(acc_k - conf_k)
    # Return the ECE
    return ece


# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 1: Fit Platt parameters on the validation partition
# -------------------------------------------------------------------------------------------------------------------------------


def platt_equation_10(
    p_hybrid: np.ndarray, a: float, b: float
) -> np.ndarray:
    """
    Evaluate Equation (10) directly in the manuscript's parameterisation.

    Implements the manuscript's probability-calibration equation verbatim,

        p_cal(x) = 1 / (1 + exp(a * p_Hybrid(x) + b)),

    where the exponent carries a POSITIVE sign on the linear predictor.
    This is the closed form a supervisor, an adverse-action engine or any
    downstream consumer would evaluate when handed the frozen calibration
    parameters, so it is provided as a first-class callable rather than
    left implicit.  Evaluation is routed through the logistic sigmoid
    `scipy.special.expit`, using the identity

        1 / (1 + exp(z)) = expit(-z),

    because a naive `1 / (1 + np.exp(z))` overflows to infinity for large
    positive z and silently returns exactly 0.0, whereas `expit` is
    numerically stable across the whole real line.

    Parameters
    ----------
    p_hybrid : np.ndarray
        The uncalibrated hybrid probabilities to be calibrated, shape
        (n,).  Values are expected in the closed unit interval.
    a : float
        The manuscript-convention scaling parameter of Equation (10).
    b : float
        The manuscript-convention shift parameter of Equation (10).

    Returns
    -------
    np.ndarray
        The calibrated probabilities, shape (n,), in the open unit
        interval.

    Raises
    ------
    ValueError
        If the probability vector is not one-dimensional, or if either
        parameter is not finite.
    """
    # Coerce the input to a float64 array so integer input cannot silently
    # truncate the linear predictor
    probabilities: np.ndarray = np.asarray(p_hybrid, dtype=np.float64)
    # Validate the array is one-dimensional, since Equation (10) is a
    # scalar map applied element-wise
    if probabilities.ndim != 1:
        # Raise a descriptive error naming the offending shape
        raise ValueError(
            f"p_hybrid must be one-dimensional; received shape "
            f"{probabilities.shape}."
        )
    # Validate both scalars are finite before they enter the exponent
    if not (math.isfinite(a) and math.isfinite(b)):
        # Raise a descriptive error naming the non-finite parameters
        raise ValueError(
            f"Equation (10) parameters must be finite; received a={a}, b={b}."
        )
    # Form the linear predictor z = a * p + b exactly as Equation (10) writes it
    linear_predictor: np.ndarray = a * probabilities + b
    # Return 1 / (1 + exp(z)) via the stable identity expit(-z)
    return expit(-linear_predictor)


def fit_platt(
    p_val: np.ndarray, y_val: np.ndarray
) -> Tuple[float, float, LogisticRegression]:
    """
    Fit the Platt scaling parameters on the validation partition.

    Fits a univariate logistic regression of the binary validation labels
    on the validation hybrid probabilities by unpenalised maximum
    likelihood, then exports the frozen calibration parameters in the
    MANUSCRIPT convention of Equation (10),

        p_cal(x) = 1 / (1 + exp(a * p_Hybrid(x) + b)).

    Convention discipline.  `sklearn.linear_model.LogisticRegression`
    parameterises a fitted binary model with a NEGATIVE exponent,

        p(x) = 1 / (1 + exp(-(w * x + c))),

    so the library pair (w, c) and the manuscript pair (a, b) differ by a
    sign on both scalars: a = -w and b = -c.  Negating both the
    coefficient and the intercept negates the entire linear predictor and
    therefore flips the exponent sign, which is the algebraic identity the
    export relies on.  Exporting the library pair under the manuscript's
    names — the defect this implementation replaces — makes Equation (10)
    evaluate to 1 - p_cal for every input, inverting every decision taken
    by any consumer that reconstructs calibration from the reported
    parameters.  The returned (a, b) are therefore the manuscript pair,
    and the identity is asserted numerically before return.

    Monotonicity discipline.  Calibration must be rank-preserving so that
    the area under the ROC curve is invariant under it.  That invariance
    holds only for a STRICTLY INCREASING map.  A negative library slope
    means the hybrid probabilities are inversely associated with the
    outcome on the validation fold, which sends the calibrated test AUC to
    one minus its former value; a zero slope means calibration has
    destroyed all discrimination.  Both are modelling failures that must
    halt the pipeline, so both are rejected explicitly.

    Parameters
    ----------
    p_val : np.ndarray
        The validation hybrid probabilities, shape (n_val,), in the closed
        unit interval.
    y_val : np.ndarray
        The validation binary labels, shape (n_val,), containing both
        classes.

    Returns
    -------
    Tuple[float, float, LogisticRegression]
        The frozen MANUSCRIPT-convention scaling parameter a, the frozen
        MANUSCRIPT-convention shift parameter b, and the fitted estimator.
        The library-convention pair remains recoverable from the returned
        estimator as `model.coef_[0][0]` and `model.intercept_[0]`.

    Raises
    ------
    ValueError
        If the inputs are misaligned; if the labels do not contain both
        classes; if the probabilities are non-finite or outside the unit
        interval; if the optimiser did not converge; if the round-trip
        identity between Equation (10) and the fitted estimator fails; or
        if the fitted calibration map is not strictly increasing.
    """
    # --- Step 1: validate the inputs -------------------------------------
    # Coerce both inputs to arrays so that list input cannot bypass the checks
    probabilities: np.ndarray = np.asarray(p_val, dtype=np.float64)
    # Coerce the labels to a 64-bit integer array for the class-count check
    labels: np.ndarray = np.asarray(y_val, dtype=np.int64)
    # Validate the inputs share length, since the fit pairs them element-wise
    if probabilities.shape[0] != labels.shape[0]:
        # Raise a descriptive error for the alignment violation
        raise ValueError(
            f"p_val length {probabilities.shape[0]} must equal y_val length "
            f"{labels.shape[0]}."
        )
    # Validate the probability vector is one-dimensional before reshaping
    if probabilities.ndim != 1:
        # Raise a descriptive error naming the offending shape
        raise ValueError(
            f"p_val must be one-dimensional; received shape "
            f"{probabilities.shape}."
        )
    # Validate every probability is finite, since a NaN silently poisons the fit
    if not np.isfinite(probabilities).all():
        # Raise a descriptive error naming the count of non-finite entries
        raise ValueError(
            f"p_val must be finite; found "
            f"{int((~np.isfinite(probabilities)).sum())} non-finite values."
        )
    # Validate every probability lies in the closed unit interval
    if float(probabilities.min()) < 0.0 or float(probabilities.max()) > 1.0:
        # Raise a descriptive error naming the observed range
        raise ValueError(
            f"p_val must lie in [0, 1]; observed range "
            f"[{float(probabilities.min())}, {float(probabilities.max())}]."
        )
    # Collect the distinct label values to confirm the fold carries both classes
    observed_classes: np.ndarray = np.unique(labels)
    # Validate both classes are present: a single-class fold makes the
    # Bernoulli likelihood unbounded and the maximum-likelihood fit meaningless
    if not np.array_equal(observed_classes, np.array([0, 1], dtype=np.int64)):
        # Raise a descriptive error naming the classes actually observed
        raise ValueError(
            f"y_val must contain both classes {{0, 1}} for the Platt "
            f"likelihood to be bounded; observed {observed_classes.tolist()}."
        )
    # --- Step 2: fit by unpenalised maximum likelihood --------------------
    # Reshape the probability vector into a single-column design matrix
    X_val: np.ndarray = probabilities.reshape(-1, 1)
    # Configure the univariate logistic regression with the penalty EXPLICITLY
    # disabled: the library default applies an L2 penalty, which would yield a
    # maximum-a-posteriori estimate that shrinks the slope toward zero and
    # under-corrects the calibration, not the maximum-likelihood estimate the
    # instruction demands
    platt_model: LogisticRegression = LogisticRegression(
        penalty=None, max_iter=1000
    )
    # Fit the Platt model on the validation partition
    platt_model.fit(X_val, labels)
    # Confirm the quasi-Newton optimiser converged rather than terminating at
    # the iteration cap, since a capped fit yields unconverged scalars
    if int(platt_model.n_iter_[0]) >= 1000:
        # Raise a descriptive error naming the iteration count reached
        raise ValueError(
            f"Platt maximum-likelihood fit failed to converge within 1000 "
            f"iterations (n_iter={int(platt_model.n_iter_[0])}); the exported "
            f"calibration parameters would be unconverged."
        )
    # --- Step 3: recover the library-convention parameters ----------------
    # Extract the library coefficient w of the negative-exponent form
    slope_sklearn: float = float(platt_model.coef_[0][0])
    # Extract the library intercept c of the negative-exponent form
    intercept_sklearn: float = float(platt_model.intercept_[0])
    # --- Step 4: derive the manuscript-convention parameters by negation ---
    # Negate the coefficient to obtain the manuscript's a of Equation (10)
    a: float = -slope_sklearn
    # Negate the intercept to obtain the manuscript's b of Equation (10)
    b: float = -intercept_sklearn
    # --- Step 5: assert the round trip numerically ------------------------
    # Evaluate Equation (10) directly with the negated pair over the fold
    p_cal_equation: np.ndarray = platt_equation_10(probabilities, a, b)
    # Evaluate the fitted estimator's own positive-class probabilities
    p_cal_estimator: np.ndarray = platt_model.predict_proba(X_val)[:, 1]
    # Measure the largest element-wise discrepancy between the two routes
    max_discrepancy: float = float(
        np.max(np.abs(p_cal_equation - p_cal_estimator))
    )
    # Require agreement to one part in ten billion, appropriate for double
    # precision on the unit interval; this converts the sign convention from a
    # documentation claim into a tested property
    if max_discrepancy > 1e-10:
        # Raise a descriptive error naming the maximum absolute discrepancy
        raise ValueError(
            f"Equation (10) round-trip failed: evaluating "
            f"p_cal = 1 / (1 + exp(a * p + b)) with the exported pair "
            f"(a={a}, b={b}) disagrees with the fitted estimator by "
            f"{max_discrepancy:.3e} (tolerance 1e-10).  The sign convention "
            f"of the exported calibration parameters is wrong."
        )
    # --- Step 6: assert strict monotonicity correctly ---------------------
    # Require the library slope to be strictly positive, which is the exact
    # condition for a strictly increasing calibration map
    if slope_sklearn <= 0.0:
        # Raise a descriptive error explaining that this is a modelling failure
        raise ValueError(
            f"Platt calibration slope must be strictly positive; fitted "
            f"library slope is {slope_sklearn}.  A non-positive slope means "
            f"the hybrid probabilities are inversely associated with the "
            f"outcome on the validation fold (a modelling failure, not a "
            f"calibration failure): it sends the calibrated test AUC to "
            f"1 - AUC, and a zero slope destroys all discrimination.  Halt "
            f"and investigate the hybrid branch rather than proceeding."
        )
    # Order the fold by its input probability so consecutive differences trace
    # the calibration map along its own domain
    order: np.ndarray = np.argsort(probabilities, kind="stable")
    # Compute the consecutive differences of the calibrated probabilities
    diffs: np.ndarray = np.diff(p_cal_estimator[order])
    # Require every difference to be non-negative up to a numerical tolerance,
    # a ONE-SIDED test that a decreasing or flat map cannot satisfy
    if not bool(np.all(diffs >= -1e-12)):
        # Raise a descriptive error naming the most negative difference found
        raise ValueError(
            f"Platt calibration must be strictly increasing in the input "
            f"probability; the most negative consecutive difference is "
            f"{float(diffs.min()):.3e}."
        )
    # --- Step 7/8: log and return ----------------------------------------
    # Log both parameterisations and their relationship for the audit trail
    logger.info(
        f"Platt parameters frozen (Eq. 10 convention): a={a:.5f}, b={b:.5f}; "
        f"sklearn convention: slope={slope_sklearn:.5f}, "
        f"intercept={intercept_sklearn:.5f}; a=-slope and b=-intercept; "
        f"Eq.(10) round-trip max discrepancy {max_discrepancy:.3e}"
    )
    # Return the MANUSCRIPT pair and the fitted estimator; application must
    # continue to route through the estimator, whose implementation is stable
    # at extreme linear-predictor values
    return a, b, platt_model


# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 2: Apply the frozen Platt model to the test partition
# -------------------------------------------------------------------------------------------------------------------------------


def apply_platt(
    p_test: np.ndarray, a: float, b: float, platt_model: LogisticRegression
) -> np.ndarray:
    """
    Apply the frozen Platt model to the test partition.

    Computes the calibrated test probabilities via Equation (10) with
    the frozen (a, b) and asserts ranking invariance of the monotone
    transform.  Region assignments remain based on the uncalibrated
    probabilities.

    Parameters
    ----------
    p_test : np.ndarray
        The test hybrid probabilities, shape (n_test,).
    a : float
        The frozen scaling parameter.
    b : float
        The frozen shift parameter.
    platt_model : LogisticRegression
        The fitted Platt estimator for numerically stable application.

    Returns
    -------
    np.ndarray
        The calibrated test probabilities, shape (n_test,).

    Raises
    ------
    ValueError
        If the calibrated output is non-finite.
    """
    # Reshape the test probabilities to the design matrix form
    X_test: np.ndarray = p_test.reshape(-1, 1)
    # Apply the frozen Platt sigmoid (Equation 10) through the fitted
    # estimator, whose implementation is numerically stable at extreme
    # linear-predictor values where a naive exponential would overflow
    p_cal: np.ndarray = platt_model.predict_proba(X_test)[:, 1]
    # Assert the calibrated probabilities are finite
    if not np.isfinite(p_cal).all():
        # Raise a descriptive error for non-finite probabilities
        raise ValueError(
            "calibrated test probabilities contain non-finite values."
        )
    # Assert that the supplied manuscript-convention pair reproduces the
    # estimator's own output through Equation (10); this pins the exported
    # parameters to the applied transform at the point of use rather than
    # trusting the caller to have passed a consistent pair
    p_cal_equation: np.ndarray = platt_equation_10(
        np.asarray(p_test, dtype=np.float64), a, b
    )
    # Measure the largest element-wise discrepancy between the two routes
    max_discrepancy: float = float(np.max(np.abs(p_cal_equation - p_cal)))
    # Reject any pair that does not satisfy Equation (10) for this estimator
    if max_discrepancy > 1e-10:
        # Raise a descriptive error naming the discrepancy and the pair
        raise ValueError(
            f"the supplied calibration pair (a={a}, b={b}) does not satisfy "
            f"Equation (10) for the supplied estimator; maximum absolute "
            f"discrepancy {max_discrepancy:.3e} exceeds 1e-10.  The pair "
            f"must be the MANUSCRIPT convention, a = -coef_ and "
            f"b = -intercept_."
        )
    # Assert ranking invariance with a ONE-SIDED test: calibration is
    # rank-preserving only when the map is strictly increasing, so a
    # decreasing or flat map must be rejected rather than accepted
    order_test: np.ndarray = np.argsort(p_test, kind="stable")
    # Compute the consecutive differences of the sorted calibrated probs
    diffs_test: np.ndarray = np.diff(p_cal[order_test])
    # Require every consecutive difference to be non-negative
    if not bool(np.all(diffs_test >= -1e-12)):
        # Raise a descriptive error naming the most negative difference
        raise ValueError(
            f"calibration must not change ranking; the calibrated mapping "
            f"is not increasing in the input probability (most negative "
            f"consecutive difference {float(diffs_test.min()):.3e})."
        )
    # Log the calibration application for the audit trail
    logger.info(
        f"Platt applied to test: n={len(p_cal)}, "
        f"mean shift {(float(np.mean(p_cal)) - float(np.mean(p_test))):.5f}"
    )
    # Return the calibrated test probabilities
    return p_cal


# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 3: Fit the LR+Platt ablation
# -------------------------------------------------------------------------------------------------------------------------------


def fit_lr_platt_ablation(
    p_lr_val: np.ndarray,
    y_val: np.ndarray,
    p_lr_test: np.ndarray,
    y_test: np.ndarray,
    verify_manuscript_targets: bool = True,
) -> Dict[str, Any]:
    """
    Fit the LR+Platt ablation as the attribution control.

    Fits a second Platt model on the validation LR probabilities,
    applies it to the test LR probabilities, and records the resulting
    Brier and ECE; the expected Zindi values (Brier 0.088, ECE 0.016)
    are asserted when verification is enabled.

    Parameters
    ----------
    p_lr_val : np.ndarray
        The validation LR probabilities.
    y_val : np.ndarray
        The validation labels.
    p_lr_test : np.ndarray
        The test LR probabilities.
    y_test : np.ndarray
        The test labels.
    verify_manuscript_targets : bool, optional
        Whether to assert the ablation targets (True for real runs).

    Returns
    -------
    Dict[str, Any]
        The ablation metrics and the calibrated ablation probabilities.

    Raises
    ------
    ValueError
        If a manuscript target is violated when verification is on.
    """
    # Fit the ablation Platt model with the identical Step 1 procedure
    a_ab, b_ab, platt_ab = fit_platt(p_lr_val, y_val)
    # Apply the ablation Platt model to the test LR probabilities
    p_ablation_test: np.ndarray = apply_platt(
        p_lr_test, a_ab, b_ab, platt_ab
    )
    # Compute the ablation Brier score on the test partition
    brier: float = float(brier_score_loss(y_test, p_ablation_test))
    # Compute the ablation ECE on the test partition (Equation 11)
    ece: float = float(compute_ece(p_ablation_test, y_test, n_bins=10))
    # Compose the ablation metrics
    metrics: Dict[str, float] = {"brier": brier, "ece": ece}
    # Assert the manuscript targets when verification is enabled
    if verify_manuscript_targets:
        # Verify the Brier target of 0.088 within 0.01
        if abs(brier - 0.088) > 0.01:
            # Raise naming the fidelity-diagnosis checklist
            raise ValueError(
                f"LR+Platt ablation Brier expected 0.088 within 0.01; "
                f"received {brier:.4f}."
            )
        # Verify the ECE target of 0.016 within 0.01
        if abs(ece - 0.016) > 0.01:
            # Raise naming the fidelity-diagnosis checklist
            raise ValueError(
                f"LR+Platt ablation ECE expected 0.016 within 0.01; "
                f"received {ece:.4f}."
            )
    # Log the ablation result for the audit trail
    logger.info(f"LR+Platt ablation: Brier {brier:.4f}, ECE {ece:.4f}")
    # Return the ablation metrics and calibrated probabilities
    return {
        "metrics": metrics,
        "p_ablation_test": p_ablation_test,
        "a": a_ab,
        "b": b_ab,
    }


# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


def orchestrate_fit_platt(
    p_hybrid: Dict[str, np.ndarray],
    p_lr: Dict[str, np.ndarray],
    labels: Dict[str, np.ndarray],
    output_dir: Path,
    verify_manuscript_targets: bool = True,
) -> Dict[str, Any]:
    """
    Orchestrate the Platt calibration stage.

    Composes the validation fit, the test application and the LR+Platt
    ablation, persisting the frozen (a, b) and the ablation metrics to
    JSON.  Region assignments remain based on the uncalibrated hybrid
    probabilities.

    Parameters
    ----------
    p_hybrid : Dict[str, np.ndarray]
        The hybrid probabilities keyed by partition.
    p_lr : Dict[str, np.ndarray]
        The LR probabilities keyed by the same partition names.
    labels : Dict[str, np.ndarray]
        The label vectors keyed by partition.
    output_dir : Path
        Directory where the artifacts are persisted.
    verify_manuscript_targets : bool, optional
        Whether to assert the ablation targets (True for real runs).

    Returns
    -------
    Dict[str, Any]
        A state-ready dict with a, b, p_cal_test, p_hybrid (retained),
        ablation and output_dir.
    """
    # Validate the output directory type before any filesystem work
    if not isinstance(output_dir, Path):
        # Raise a descriptive error for a non-Path output directory
        raise ValueError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Create the output directory including any missing parents
    output_dir.mkdir(parents=True, exist_ok=True)
    # Step 1: fit the Platt parameters on the validation partition
    a, b, platt_model = fit_platt(
        p_hybrid["val"], labels["val"]
    )
    # Step 2: apply the frozen Platt model to the test partition
    p_cal_test: np.ndarray = apply_platt(
        p_hybrid["test"], a, b, platt_model
    )
    # Step 3: fit the LR+Platt ablation as the attribution control
    ablation: Dict[str, Any] = fit_lr_platt_ablation(
        p_lr["val"], labels["val"], p_lr["test"], labels["test"],
        verify_manuscript_targets,
    )
    # Recover the library-convention pair from the fitted estimator so the
    # artifact can name BOTH parameterisations explicitly; ambiguity between
    # the two sign conventions is precisely what makes a calibration export
    # dangerous to a downstream consumer
    slope_sklearn: float = float(platt_model.coef_[0][0])
    # Recover the library-convention intercept from the fitted estimator
    intercept_sklearn: float = float(platt_model.intercept_[0])
    # Compose the persisted artifacts payload carrying both parameterisations
    # under unambiguous keys, together with the relationship note
    artifacts: Dict[str, Any] = {
        "a": a,
        "b": b,
        "a_manuscript": a,
        "b_manuscript": b,
        "a_sklearn": slope_sklearn,
        "b_sklearn": intercept_sklearn,
        "parameterisation_note": (
            "'a_manuscript'/'b_manuscript' (exported as 'a'/'b') satisfy "
            "Equation (10) p_cal = 1 / (1 + exp(a * p_Hybrid + b)). "
            "'a_sklearn'/'b_sklearn' are sklearn's coef_/intercept_ for the "
            "negative-exponent form p = 1 / (1 + exp(-(a * p_Hybrid + b))). "
            "The two differ by a sign on both scalars: "
            "a_manuscript = -a_sklearn and b_manuscript = -b_sklearn. "
            "Use the manuscript pair with Equation (10)."
        ),
        "ablation_metrics": ablation["metrics"],
    }
    # Build the artifacts path inside the output directory
    artifacts_path: Path = output_dir / "platt_artifacts.json"
    # Serialise the artifacts to JSON with a stable layout
    artifacts_text: str = json.dumps(
        artifacts, indent=2, ensure_ascii=False, default=str
    )
    # Persist the artifacts to the output directory
    artifacts_path.write_text(artifacts_text, encoding="utf-8")
    # Return the state-ready calibration outputs, carrying BOTH
    # parameterisations forward so the run manifest can report them without
    # re-deriving the sign relationship downstream
    return {
        "a": a,
        "b": b,
        "a_sklearn": slope_sklearn,
        "b_sklearn": intercept_sklearn,
        "p_cal_test": p_cal_test,
        "p_hybrid": p_hybrid,
        "ablation": ablation,
        "output_dir": str(output_dir),
    }
````

````python
# Task 14: Compute Evaluation Metrics

# ==============================================================================
# Task 14: Compute Evaluation Metrics
# ==============================================================================


# The decision threshold for the F1 reporting (manuscript convention)
_F1_THRESHOLD: float = 0.5


# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Step 1: Compute discrimination metrics for all five candidate models
# -------------------------------------------------------------------------------------------------------------------------------


def compute_discrimination_metrics(
    probabilities: Dict[str, np.ndarray], y: np.ndarray
) -> Dict[str, Dict[str, float]]:
    """
    Compute AUC, Gini and F1 for every candidate model.

    Computes AUC (rank-order), Gini = 2 * AUC - 1, and minority-class
    F1 at threshold 0.5 for each model's test probability vector.

    Parameters
    ----------
    probabilities : Dict[str, np.ndarray]
        The test probability vectors keyed by model name.
    y : np.ndarray
        The test label vector.

    Returns
    -------
    Dict[str, Dict[str, float]]
        Model name -> {auc, gini, f1}.

    Raises
    ------
    ValueError
        If a probability vector is misaligned with the labels.
    """
    # Collect the per-model discrimination metrics
    metrics: Dict[str, Dict[str, float]] = {}
    # Iterate over the candidate models
    for model_name, probs in probabilities.items():
        # Assert the probability vector aligns with the labels
        if len(probs) != len(y):
            # Raise a descriptive error for the alignment violation
            raise ValueError(
                f"model '{model_name}' probabilities length {len(probs)} "
                f"must equal labels length {len(y)}."
            )
        # Compute the test AUC (rank-order discrimination)
        auc: float = float(roc_auc_score(y, probs))
        # Compute the Gini coefficient: Gini = 2 * AUC - 1
        gini: float = 2.0 * auc - 1.0
        # Build the binary predictions at threshold 0.5
        preds: np.ndarray = (probs >= _F1_THRESHOLD).astype(np.int64)
        # Compute per-class precision, recall and F1 at threshold 0.5
        _, _, f1_scores, _ = precision_recall_fscore_support(
            y, preds, labels=[0, 1], zero_division=0
        )
        # Select the minority-class (positive) F1
        f1: float = float(f1_scores[1])
        # Record the model's metrics
        metrics[model_name] = {"auc": auc, "gini": gini, "f1": f1}
    # Return the per-model discrimination metrics
    return metrics


# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Step 2: Compute calibration metrics
# -------------------------------------------------------------------------------------------------------------------------------


def compute_calibration_metrics(
    probabilities: Dict[str, np.ndarray], y: np.ndarray
) -> Dict[str, Dict[str, Any]]:
    """
    Compute Brier, ECE and reliability data for every candidate model.

    Computes Brier = (1/n) * sum (p_i - y_i)^2, ECE with K = 10
    equal-width bins (Equation 11), and the reliability-diagram data
    (per-bin confidence and accuracy).

    Parameters
    ----------
    probabilities : Dict[str, np.ndarray]
        The test probability vectors keyed by model name.
    y : np.ndarray
        The test label vector.

    Returns
    -------
    Dict[str, Dict[str, Any]]
        Model name -> {brier, ece, reliability}.

    Raises
    ------
    ValueError
        If a probability vector is misaligned or out of range.
    """
    # Collect the per-model calibration metrics
    metrics: Dict[str, Dict[str, Any]] = {}
    # Iterate over the candidate models
    for model_name, probs in probabilities.items():
        # Assert the probability vector aligns with the labels
        if len(probs) != len(y):
            # Raise a descriptive error for the alignment violation
            raise ValueError(
                f"model '{model_name}' probabilities length {len(probs)} "
                f"must equal labels length {len(y)}."
            )
        # Assert the probabilities lie in [0, 1]
        if not np.all((probs >= 0.0) & (probs <= 1.0)):
            # Raise a descriptive error for the range violation
            raise ValueError(
                f"model '{model_name}' probabilities must lie in [0, 1]."
            )
        # Compute the Brier score: (1/n) * sum (p - y)^2
        brier: float = float(brier_score_loss(y, probs))
        # Compute the ECE with K = 10 equal-width bins (Equation 11)
        ece: float = float(compute_ece(probs, y, n_bins=10))
        # Build the reliability-diagram bins on [0, 1]
        bin_edges: np.ndarray = np.linspace(0.0, 1.0, 11)
        # Assign each probability to its bin
        bin_ids: np.ndarray = np.clip(
            np.digitize(probs, bin_edges[1:-1]), 0, 9
        )
        # Collect the per-bin confidence and accuracy
        confidence: List[float] = []
        accuracy: List[float] = []
        # Iterate over the ten bins
        for k in range(10):
            # Build the mask of observations in bin k
            mask: np.ndarray = bin_ids == k
            # Record zero confidence and accuracy for empty bins
            if not mask.any():
                # Append the empty-bin confidence
                confidence.append(0.0)
                # Append the empty-bin accuracy
                accuracy.append(0.0)
                # Continue to the next bin
                continue
            # Record the bin's mean confidence
            confidence.append(float(np.mean(probs[mask])))
            # Record the bin's empirical accuracy
            accuracy.append(float(np.mean(y[mask])))
        # Record the model's calibration metrics
        metrics[model_name] = {
            "brier": brier,
            "ece": ece,
            "reliability": {"confidence": confidence, "accuracy": accuracy},
        }
    # Return the per-model calibration metrics
    return metrics


# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Step 3: Compute per-region AUC and assemble the results tables
# -------------------------------------------------------------------------------------------------------------------------------


def compute_per_region_auc(
    regions: np.ndarray,
    p_lr_test: np.ndarray,
    p_hybrid_test: np.ndarray,
    y_test: np.ndarray,
) -> Dict[str, Dict[str, Optional[float]]]:
    """
    Compute the LR and Hybrid AUC within each interpretability region.

    Computes the LR AUC and the Hybrid AUC restricted to each of the
    three interpretability regions via boolean region masks.

    Parameters
    ----------
    regions : np.ndarray
        The test region label vector.
    p_lr_test : np.ndarray
        The test LR probabilities.
    p_hybrid_test : np.ndarray
        The test hybrid probabilities.
    y_test : np.ndarray
        The test labels.

    Returns
    -------
    Dict[str, Dict[str, Optional[float]]]
        Region -> {lr_auc, hybrid_auc}.

    Raises
    ------
    ValueError
        If the inputs are misaligned.
    """
    # Validate the inputs share length
    if not (len(regions) == len(p_lr_test) == len(p_hybrid_test) == len(y_test)):
        # Raise a descriptive error for the alignment violation
        raise ValueError(
            "regions, probabilities and labels must share length."
        )
    # Collect the per-region AUCs
    region_auc: Dict[str, Dict[str, Optional[float]]] = {}
    # Iterate over the three region names
    for region_name in (
        "Fully interpretable",
        "Partially interpretable",
        "ML-driven",
    ):
        # Build the region mask
        mask: np.ndarray = regions == region_name
        # Guard a region with fewer than two classes for the AUC
        if mask.sum() < 2 or len(np.unique(y_test[mask])) < 2:
            # Log the guarded region
            logger.warning(f"region '{region_name}': AUC undefined (n={mask.sum()})")
            # Record the guarded region with None AUCs
            region_auc[region_name] = {"lr_auc": None, "hybrid_auc": None}
            # Continue to the next region
            continue
        # Compute the LR AUC within the region
        lr_auc: float = float(roc_auc_score(y_test[mask], p_lr_test[mask]))
        # Compute the Hybrid AUC within the region
        hybrid_auc: float = float(
            roc_auc_score(y_test[mask], p_hybrid_test[mask])
        )
        # Record the region's AUCs
        region_auc[region_name] = {"lr_auc": lr_auc, "hybrid_auc": hybrid_auc}
    # Return the per-region AUCs
    return region_auc


def assemble_results_tables(
    discrimination: Dict[str, Dict[str, float]],
    calibration: Dict[str, Dict[str, Any]],
    region_stats: Dict[str, Any],
    region_auc: Dict[str, Dict[str, Optional[float]]],
) -> Dict[str, Any]:
    """
    Assemble Table 2 and Table 3 payloads from the computed metrics.
    """
    # Assemble Table 2: per-model predictive performance
    table_2: Dict[str, Dict[str, float]] = {}
    # Iterate over the models in the discrimination metrics
    for model_name, disc in discrimination.items():
        # Compose the Table 2 row for the model
        table_2[model_name] = {
            "auc": disc["auc"],
            "gini": disc["gini"],
            "f1": disc["f1"],
            "brier": calibration[model_name]["brier"],
            "ece": calibration[model_name]["ece"],
        }
    # Assemble Table 3: region behaviour
    table_3: Dict[str, Dict[str, Any]] = {}
    # Iterate over the regions in the region statistics
    for region_name in ("Fully interpretable", "Partially interpretable", "ML-driven"):
        # Compose the Table 3 row for the region
        table_3[region_name] = {
            "share": region_stats["shares"].get(region_name, 0.0),
            "positive_rate": region_stats["rates"].get(region_name, 0.0),
            "mean_rho": region_stats["mean_rho"].get(region_name, 0.0),
            "lr_auc": region_auc.get(region_name, {}).get("lr_auc"),
            "hybrid_auc": region_auc.get(region_name, {}).get("hybrid_auc"),
        }
    # Return the assembled tables
    return {"table_2": table_2, "table_3": table_3}


def assert_headline_values(
    table_2: Dict[str, Dict[str, Any]], tolerance: float = 0.005
) -> None:
    """
    Assert the Zindi headline values within the given tolerance.
    """
    # Extract the calibrated hybrid row from Table 2
    hybrid_row: Dict[str, Any] = table_2["Hybrid (Platt calibrated)"]
    # Assert the calibrated hybrid AUC target of 0.869
    if abs(hybrid_row["auc"] - 0.869) > tolerance:
        # Raise naming the fidelity-diagnosis checklist
        raise ValueError(
            f"Zindi calibrated hybrid AUC expected 0.869 within "
            f"{tolerance}; received {hybrid_row['auc']:.4f}. Run the "
            f"fidelity-diagnosis checklist in order: split seed; "
            f"binning algorithm; rare-level handling; calibration "
            f"fitting; library versions; statistical-test implementation."
        )
    # Assert the calibrated hybrid Brier target of 0.085
    if abs(hybrid_row["brier"] - 0.085) > tolerance:
        # Raise naming the fidelity-diagnosis checklist
        raise ValueError(
            f"Zindi calibrated hybrid Brier expected 0.085 within "
            f"{tolerance}; received {hybrid_row['brier']:.4f}."
        )
    # Assert the calibrated hybrid ECE target of 0.024
    if abs(hybrid_row["ece"] - 0.024) > tolerance:
        # Raise naming the fidelity-diagnosis checklist
        raise ValueError(
            f"Zindi calibrated hybrid ECE expected 0.024 within "
            f"{tolerance}; received {hybrid_row['ece']:.4f}."
        )


# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


def orchestrate_compute_metrics(
    probabilities: Dict[str, np.ndarray],
    labels_test: np.ndarray,
    regions_test: np.ndarray,
    p_lr_test: np.ndarray,
    p_hybrid_test: np.ndarray,
    region_stats: Dict[str, Any],
    output_dir: Path,
    verify_manuscript_targets: bool = True,
) -> Dict[str, Any]:
    """
    Orchestrate the evaluation metrics computation stage.

    Composes the discrimination, calibration and region metrics steps,
    assembles Table 2 and Table 3, asserts the headline values when
    enabled, and persists the tables to JSON.

    Parameters
    ----------
    probabilities : Dict[str, np.ndarray]
        The test probability vectors keyed by the five model names.
    labels_test : np.ndarray
        The test labels.
    regions_test : np.ndarray
        The test region assignments.
    p_lr_test : np.ndarray
        The test LR probabilities.
    p_hybrid_test : np.ndarray
        The test hybrid probabilities.
    region_stats : Dict[str, Any]
        The region statistics from Task 12.
    output_dir : Path
        Directory where the artifacts are persisted.
    verify_manuscript_targets : bool, optional
        Whether to assert the headline values (True for real runs).

    Returns
    -------
    Dict[str, Any]
        A state-ready dict with discrimination, calibration,
        region_auc, tables and output_dir.
    """
    # Validate the output directory type before any filesystem work
    if not isinstance(output_dir, Path):
        # Raise a descriptive error for a non-Path output directory
        raise ValueError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Create the output directory including any missing parents
    output_dir.mkdir(parents=True, exist_ok=True)
    # Step 1: compute the discrimination metrics for all five models
    discrimination: Dict[str, Dict[str, float]] = compute_discrimination_metrics(
        probabilities, labels_test
    )
    # Step 2: compute the calibration metrics for all five models
    calibration: Dict[str, Dict[str, Any]] = compute_calibration_metrics(
        probabilities, labels_test
    )
    # Step 3: compute the per-region AUCs
    region_auc: Dict[str, Dict[str, Optional[float]]] = compute_per_region_auc(
        regions_test, p_lr_test, p_hybrid_test, labels_test
    )
    # Assemble Table 2 and Table 3
    tables: Dict[str, Any] = assemble_results_tables(
        discrimination, calibration, region_stats, region_auc
    )
    # Assert the headline values when verification is enabled
    if verify_manuscript_targets:
        # Assert the Zindi headline values within the tolerance
        assert_headline_values(tables["table_2"])
    # Build the artifacts path inside the output directory
    artifacts_path: Path = output_dir / "metrics_artifacts.json"
    # Serialise the tables and metrics to JSON with a stable layout
    artifacts_text: str = json.dumps(
        {
            "tables": tables,
            "discrimination": discrimination,
            "calibration": calibration,
            "region_auc": region_auc,
        },
        indent=2,
        ensure_ascii=False,
        default=str,
    )
    # Persist the artifacts to the output directory
    artifacts_path.write_text(artifacts_text, encoding="utf-8")
    # Return the state-ready metrics outputs
    return {
        "discrimination": discrimination,
        "calibration": calibration,
        "region_auc": region_auc,
        "tables": tables,
        "output_dir": str(output_dir),
    }
````

````python
# Task 15: Fairness Audit per Interpretability Region

# ==============================================================================
# Task 15: Fairness Audit per Interpretability Region
# ==============================================================================


# The routing threshold from Equation (7) (ML-driven region)
RHO_ML_THRESHOLD: float = 0.7

# The decision threshold for the predicted positives
DECISION_THRESHOLD: float = 0.5

# The region names
REGIONS: Tuple[str, ...] = (
    "Fully interpretable",
    "Partially interpretable",
    "ML-driven",
)

# The undefined-cell sentinel
UNDEFINED: str = "---"


# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Step 1: Recode protected attributes on the test partition
# -------------------------------------------------------------------------------------------------------------------------------


def recode_protected_attributes(
    test_frame: pd.DataFrame, config: Dict[str, Any]
) -> pd.DataFrame:
    """
    Recode the protected attributes into audit columns.

    Creates the audit columns gender, location, education and country
    from the config mappings, asserting that every raw value is covered.
    The original test partition is never mutated.

    Parameters
    ----------
    test_frame : pd.DataFrame
        The test partition frame with the raw protected attributes.
    config : Dict[str, Any]
        The validated study configuration dictionary.

    Returns
    -------
    pd.DataFrame
        The analysis frame with the four audit columns appended.

    Raises
    ------
    ValueError
        If any raw value is uncovered by its mapping.
    """
    # Take a working copy so the test partition is never mutated
    analysis: pd.DataFrame = test_frame.copy()
    # Extract the protected attribute mappings from the configuration
    protected: Dict[str, Any] = config["protected_attribute_parameters"]
    # Map the gender raw values to the audit buckets
    gender_map: Dict[str, str] = protected["gender"]["raw_mapping"]
    # Assert every observed gender value is covered by the mapping
    if not set(analysis["gender_of_respondent"]).issubset(gender_map):
        # Raise naming the uncovered gender values
        raise ValueError(
            f"uncovered gender values: "
            f"{set(analysis['gender_of_respondent']) - set(gender_map)}."
        )
    # Apply the gender mapping
    analysis["gender"] = analysis["gender_of_respondent"].map(gender_map)
    # Map the location raw values to the audit buckets
    location_map: Dict[str, str] = protected["location"]["raw_mapping"]
    # Assert every observed location value is covered by the mapping
    if not set(analysis["location_type"]).issubset(location_map):
        # Raise naming the uncovered location values
        raise ValueError(
            f"uncovered location values: "
            f"{set(analysis['location_type']) - set(location_map)}."
        )
    # Apply the location mapping
    analysis["location"] = analysis["location_type"].map(location_map)
    # Extract the education bucket mapping from the configuration
    education_map: Dict[str, str] = protected["education"][
        "raw_category_to_audit_bucket"
    ]
    # Assert every observed education value is covered by the mapping
    if not set(analysis["education_level"]).issubset(education_map):
        # Raise naming the uncovered education values
        raise ValueError(
            f"uncovered education values: "
            f"{set(analysis['education_level']) - set(education_map)}."
        )
    # Apply the education mapping (Other and Don't know/Refused map to
    # secondary or above per the documented placeholder)
    analysis["education"] = analysis["education_level"].map(education_map)
    # Keep the raw country values as the audit column
    analysis["country"] = analysis["country"].astype(str)
    # Log the recoding completion for the audit trail
    logger.info("protected attributes recoded into audit columns")
    # Return the analysis frame
    return analysis


def assert_cell_counts(
    analysis: pd.DataFrame, verify_manuscript_targets: bool = True
) -> None:
    """
    Assert the per-cell counts against Table 4 within 5 observations.
    """
    # Compute the observed per-cell counts
    observed: Dict[str, int] = {
        "male": int((analysis["gender"] == "male").sum()),
        "female": int((analysis["gender"] == "female").sum()),
        "urban": int((analysis["location"] == "urban").sum()),
        "rural": int((analysis["location"] == "rural").sum()),
        "secondary": int((analysis["education"] == "secondary or above").sum()),
        "primary": int((analysis["education"] == "primary or less").sum()),
    }
    # Add the country counts to the observed dict
    for country in ("Kenya", "Rwanda", "Tanzania", "Uganda"):
        # Record the country count
        observed[country] = int((analysis["country"] == country).sum())
    # Assert the Table 4 targets when verification is enabled
    if verify_manuscript_targets:
        # Declare the Table 4 target counts
        targets: Dict[str, int] = {
            "male": 1913,
            "female": 2792,
            "urban": 1809,
            "rural": 2896,
            "secondary": 1283,
            "primary": 3422,
            "Kenya": 1209,
            "Rwanda": 1746,
            "Tanzania": 1277,
            "Uganda": 473,
        }
        # Verify each target count within 5 observations
        for subgroup, target in targets.items():
            # Assert the observed count lies within the tolerance
            if abs(observed.get(subgroup, 0) - target) > 5:
                # Raise naming the fidelity-diagnosis checklist
                raise ValueError(
                    f"Table 4 count for '{subgroup}' expected {target} "
                    f"within 5; received {observed.get(subgroup)}."
                )
    # Log the observed counts for the audit trail
    logger.info(f"audit cell counts: {observed}")


# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Step 2: Compute region routing rates with bootstrap confidence intervals
# -------------------------------------------------------------------------------------------------------------------------------


def compute_routing_rates(
    analysis: pd.DataFrame,
    config: Dict[str, Any],
    n_resamples: int = 500,
) -> Dict[str, Dict[str, Dict[str, Any]]]:
    """
    Compute per-subgroup routing rates with bootstrap CIs.

    Computes Pr[rho < 0.7 | A = g] per subgroup with 95% percentile
    bootstrap confidence intervals from n_resamples drawn with
    replacement within (class x attribute) strata, with a class-only
    fallback for empty strata.

    Parameters
    ----------
    analysis : pd.DataFrame
        The analysis frame with rho, the label and the audit columns.
    config : Dict[str, Any]
        The validated study configuration dictionary.
    n_resamples : int, optional
        The bootstrap resample count (500 per the config).

    Returns
    -------
    Dict[str, Dict[str, Dict[str, Any]]]
        Attribute -> subgroup -> {rate, ci_low, ci_high, n}.

    Raises
    ------
    ValueError
        If the analysis frame lacks required columns.
    """
    # Validate the required columns are present
    for column in ("rho", "bank_account", "gender", "location", "education", "country"):
        # Assert the analysis column exists
        if column not in analysis.columns:
            # Raise a descriptive error naming the missing column
            raise ValueError(
                f"analysis frame missing column '{column}'."
            )
    # Extract the reference groups from the configuration
    protected: Dict[str, Any] = config["protected_attribute_parameters"]
    # Extract the global seed for the bootstrap
    seed: int = int(config["partitioning_parameters"]["random_seed"])
    # Build the rng for reproducible resampling
    rng: np.random.Generator = np.random.default_rng(seed)
    # Build the routing indicator: rho below the ML threshold
    routed: np.ndarray = (analysis["rho"].to_numpy() < RHO_ML_THRESHOLD).astype(
        np.int64
    )
    # Extract the label vector for the stratification
    y: np.ndarray = analysis["bank_account"].to_numpy(dtype=np.int64)
    # Collect the routing report per attribute
    routing: Dict[str, Dict[str, Dict[str, Any]]] = {}
    # Iterate over the four protected attributes
    for attribute, block in (
        ("gender", protected["gender"]),
        ("location", protected["location"]),
        ("education", protected["education"]),
        ("country", protected["country"]),
    ):
        # Extract the attribute values
        values: np.ndarray = analysis[attribute].to_numpy(dtype=object)
        # Collect the per-subgroup routing entries
        routing[attribute] = {}
        # Iterate over the declared categories of the attribute
        for subgroup in block["categories"]:
            # Build the subgroup mask
            mask: np.ndarray = values == subgroup
            # Guard a subgroup with no observations
            if not mask.any():
                # Log the empty subgroup
                logger.warning(f"subgroup '{subgroup}' of '{attribute}' is empty")
                # Record the undefined entry
                routing[attribute][subgroup] = {
                    "rate": None,
                    "ci_low": None,
                    "ci_high": None,
                    "n": 0,
                }
                # Continue to the next subgroup
                continue
            # Compute the point estimate of the routing rate
            rate: float = float(np.mean(routed[mask]))
            # Compute the positive-class share for the class strata
            pos_count: int = int(np.sum(y[mask] == 1))
            neg_count: int = int(np.sum(y[mask] == 0))
            # Collect the resampled rates for the percentile CI
            resampled_rates: List[float] = []
            # Iterate over the bootstrap resamples
            for _ in range(n_resamples):
                # Draw positive indices within the subgroup
                pos_idx: np.ndarray = mask_indices(mask, y, 1, rng, pos_count)
                # Draw negative indices within the subgroup
                neg_idx: np.ndarray = mask_indices(mask, y, 0, rng, neg_count)
                # Combine the resampled indices
                sample_idx: np.ndarray = np.concatenate([pos_idx, neg_idx])
                # Fall back to class-only strata when the combined sample
                # is empty (the (class x attribute) cell vanished)
                if len(sample_idx) == 0:
                    # Draw a class-only resample from the subgroup mask
                    sample_idx = rng.choice(
                        np.flatnonzero(mask), size=int(mask.sum()), replace=True
                    )
                    # Log the fallback for the audit trail
                    logger.info(
                        f"bootstrap fallback to class-only strata for "
                        f"'{subgroup}' of '{attribute}'"
                    )
                # Compute the resampled routing rate
                resampled_rates.append(float(np.mean(routed[sample_idx])))
            # Compute the 2.5th and 97.5th percentiles of the resamples
            ci_low: float = float(np.percentile(resampled_rates, 2.5))
            ci_high: float = float(np.percentile(resampled_rates, 97.5))
            # Record the subgroup's routing entry
            routing[attribute][subgroup] = {
                "rate": rate,
                "ci_low": ci_low,
                "ci_high": ci_high,
                "n": int(mask.sum()),
            }
    # Return the routing report
    return routing


def mask_indices(
    mask: np.ndarray,
    y: np.ndarray,
    cls: int,
    rng: np.random.Generator,
    count: int,
) -> np.ndarray:
    """
    Draw a with-replacement sample of a class within a mask.
    """
    # Locate the class indices inside the mask
    class_idx: np.ndarray = np.flatnonzero(mask & (y == cls))
    # Return an empty array when the class cell is empty
    if len(class_idx) == 0:
        # Return the empty index array
        return np.array([], dtype=np.int64)
    # Draw the class indices with replacement
    return rng.choice(class_idx, size=count, replace=True)


# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Step 3: Compute pooled and per-region DI and EOD
# -------------------------------------------------------------------------------------------------------------------------------


def compute_di_and_eod(
    analysis: pd.DataFrame, config: Dict[str, Any]
) -> Dict[str, Any]:
    """
    Compute pooled and per-region DI and EOD with the degenerate rule.

    Computes DI(g) = Pr[y_hat = 1 | A = g] / Pr[y_hat = 1 | A = g_ref]
    and EOD = max_g { |TPR(g) - TPR(g_ref)|, |FPR(g) - FPR(g_ref)| },
    pooled and per region, reporting '---' for degenerate cells.

    Parameters
    ----------
    analysis : pd.DataFrame
        The analysis frame with the calibrated probability, the label,
        the regions and the audit columns.
    config : Dict[str, Any]
        The validated study configuration dictionary.

    Returns
    -------
    Dict[str, Any]
        The DI and EOD tables plus the reference groups.
    """
    # Extract the reference groups from the configuration
    protected: Dict[str, Any] = config["protected_attribute_parameters"]
    # Build the predicted positives at the decision threshold
    y_hat: np.ndarray = (
        analysis["p_cal"].to_numpy() >= DECISION_THRESHOLD
    ).astype(np.int64)
    # Extract the label vector
    y: np.ndarray = analysis["bank_account"].to_numpy(dtype=np.int64)
    # Extract the region vector
    regions: np.ndarray = analysis["region"].to_numpy(dtype=object)
    # Collect the DI table
    di_table: Dict[str, Dict[str, Any]] = {}
    # Collect the EOD table
    eod_table: Dict[str, Dict[str, Any]] = {}
    # Iterate over the four protected attributes
    for attribute, block in (
        ("gender", protected["gender"]),
        ("location", protected["location"]),
        ("education", protected["education"]),
        ("country", protected["country"]),
    ):
        # Extract the attribute values
        values: np.ndarray = analysis[attribute].to_numpy(dtype=object)
        # Extract the reference subgroup
        ref: str = block["reference_group"]
        # Compute the reference positive rate
        ref_rate: float = float(np.mean(y_hat[values == ref]))
        # Compute the pooled DI per subgroup
        pooled_di: Dict[str, Any] = {}
        # Iterate over the declared categories
        for subgroup in block["categories"]:
            # Compute the subgroup positive rate
            subgroup_rate: float = float(np.mean(y_hat[values == subgroup]))
            # Apply the degenerate-cell rule for the pooled ratio
            if ref_rate == 0.0:
                # Record the undefined pooled DI
                pooled_di[subgroup] = UNDEFINED
            else:
                # Compute the pooled DI ratio
                pooled_di[subgroup] = subgroup_rate / ref_rate
        # Compute the per-region DI tables
        per_region_di: Dict[str, Dict[str, Any]] = {}
        # Iterate over the regions
        for region in REGIONS:
            # Build the region mask
            region_mask: np.ndarray = regions == region
            # Collect the region's DI per subgroup
            region_di: Dict[str, Any] = {}
            # Compute the region reference positive rate
            ref_region_rate: float = float(
                np.mean(y_hat[region_mask & (values == ref)])
            )
            # Iterate over the declared categories
            for subgroup in block["categories"]:
                # Compute the region subgroup positive rate
                subgroup_region_rate: float = float(
                    np.mean(y_hat[region_mask & (values == subgroup)])
                )
                # Apply the degenerate-cell rule for the region ratio
                if ref_region_rate == 0.0:
                    # Record the undefined region DI
                    region_di[subgroup] = UNDEFINED
                else:
                    # Compute the region DI ratio
                    region_di[subgroup] = subgroup_region_rate / ref_region_rate
            # Record the region's DI column
            per_region_di[region] = region_di
        # Record the attribute's DI table
        di_table[attribute] = {
            "reference": ref,
            "pooled": pooled_di,
            "per_region": per_region_di,
        }
        # Compute the pooled EOD
        pooled_eod: float = _eod_for(values, y_hat, y, ref, None)
        # Compute the per-region EOD values
        per_region_eod: Dict[str, float] = {}
        # Iterate over the regions
        for region in REGIONS:
            # Compute the region's EOD with the degenerate rule
            per_region_eod[region] = _eod_for(
                values, y_hat, y, ref, regions == region
            )
        # Record the attribute's EOD table
        eod_table[attribute] = {
            "reference": ref,
            "pooled": pooled_eod,
            "per_region": per_region_eod,
        }
    # Return the DI and EOD tables
    return {"di": di_table, "eod": eod_table}


def _eod_for(
    values: np.ndarray,
    y_hat: np.ndarray,
    y: np.ndarray,
    ref: str,
    region_mask: Optional[np.ndarray],
) -> Any:
    """
    Compute the EOD for an attribute under an optional region mask.
    """
    # Apply the region mask when provided
    mask: np.ndarray = region_mask if region_mask is not None else np.ones(
        len(values), dtype=bool
    )
    # Compute the reference TPR and FPR denominators
    ref_pred_pos: int = int(np.sum(y_hat[mask & (values == ref)]))
    ref_actual_pos: int = int(np.sum(y[mask & (values == ref)] == 1))
    ref_actual_neg: int = int(np.sum(y[mask & (values == ref)] == 0))
    # Apply the degenerate-cell rule: undefined when the reference has no
    # predicted positives or no actual members
    if ref_pred_pos == 0 or (ref_actual_pos == 0 and ref_actual_neg == 0):
        # Return the undefined sentinel
        return UNDEFINED
    # Compute the reference TPR
    tpr_ref: float = (
        np.mean(y_hat[mask & (values == ref) & (y == 1)])
        if ref_actual_pos > 0
        else 0.0
    )
    # Compute the reference FPR
    fpr_ref: float = (
        np.mean(y_hat[mask & (values == ref) & (y == 0)])
        if ref_actual_neg > 0
        else 0.0
    )
    # Compute the maximum absolute gap over the non-reference subgroups
    max_gap: float = 0.0
    # Iterate over the observed subgroups
    for subgroup in np.unique(values[mask]):
        # Skip the reference subgroup itself
        if subgroup == ref:
            # Continue to the next subgroup
            continue
        # Compute the subgroup's actual counts
        sub_actual_pos: int = int(np.sum(y[mask & (values == subgroup)] == 1))
        sub_actual_neg: int = int(np.sum(y[mask & (values == subgroup)] == 0))
        # Compute the subgroup TPR
        tpr_g: float = (
            np.mean(y_hat[mask & (values == subgroup) & (y == 1)])
            if sub_actual_pos > 0
            else 0.0
        )
        # Compute the subgroup FPR
        fpr_g: float = (
            np.mean(y_hat[mask & (values == subgroup) & (y == 0)])
            if sub_actual_neg > 0
            else 0.0
        )
        # Accumulate the maximum absolute gap (Equation: EOD)
        max_gap = max(max_gap, abs(tpr_g - tpr_ref), abs(fpr_g - fpr_ref))
    # Return the EOD value as a native float for clean serialisation
    return float(max_gap)


# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


def orchestrate_run_fairness_audit(
    config: Dict[str, Any],
    test_frame: pd.DataFrame,
    rho_test: np.ndarray,
    regions_test: np.ndarray,
    p_cal_test: np.ndarray,
    y_test: np.ndarray,
    output_dir: Path,
    verify_manuscript_targets: bool = True,
) -> Dict[str, Any]:
    """
    Orchestrate the fairness audit stage.

    Composes the protected-attribute recoding, the routing-rate
    bootstrap and the DI/EOD computation, persisting all reports to
    JSON.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated study configuration dictionary.
    test_frame : pd.DataFrame
        The test partition frame with the raw protected attributes.
    rho_test : np.ndarray
        The test rho vector.
    regions_test : np.ndarray
        The test region assignments.
    p_cal_test : np.ndarray
        The calibrated test probabilities.
    y_test : np.ndarray
        The test labels.
    output_dir : Path
        Directory where the artifacts are persisted.
    verify_manuscript_targets : bool, optional
        Whether to assert the Table 4 counts (True for real runs).

    Returns
    -------
    Dict[str, Any]
        A state-ready dict with analysis, routing, di_eod and output_dir.
    """
    # Validate the output directory type before any filesystem work
    if not isinstance(output_dir, Path):
        # Raise a descriptive error for a non-Path output directory
        raise ValueError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Create the output directory including any missing parents
    output_dir.mkdir(parents=True, exist_ok=True)
    # Validate the input vector lengths agree with the test frame
    n_test: int = test_frame.shape[0]
    # Assert each input vector matches the frame row count
    for name, vector in (
        ("rho_test", rho_test),
        ("regions_test", regions_test),
        ("p_cal_test", p_cal_test),
        ("y_test", y_test),
    ):
        # Assert the vector length matches the frame
        if len(vector) != n_test:
            # Raise a descriptive error for the alignment violation
            raise ValueError(
                f"{name} length {len(vector)} must equal test frame rows "
                f"{n_test}."
            )
    # Step 1: recode the protected attributes into audit columns
    analysis: pd.DataFrame = recode_protected_attributes(test_frame, config)
    # Attach the rho, region, probability and label columns
    analysis["rho"] = rho_test
    analysis["region"] = regions_test
    analysis["p_cal"] = p_cal_test
    analysis["bank_account"] = y_test
    # Assert the per-cell counts against Table 4 when enabled
    assert_cell_counts(analysis, verify_manuscript_targets)
    # Step 2: compute the routing rates with bootstrap CIs
    routing: Dict[str, Dict[str, Dict[str, Any]]] = compute_routing_rates(
        analysis, config
    )
    # Step 3: compute the pooled and per-region DI and EOD
    di_eod: Dict[str, Any] = compute_di_and_eod(analysis, config)
    # Compose the persisted artifacts payload
    artifacts: Dict[str, Any] = {
        "routing": routing,
        "di_eod": di_eod,
        "caveats": {
            "post_treatment": (
                "per-region DI/EOD are descriptive stratifications "
                "because rho(x) is a post-treatment variable (Rosenbaum "
                "1984); they are not 'fairness within the region'."
            ),
            "degenerate_cells": (
                "'---' cells are undefined (vanishing denominators), "
                "never zero; the ML-driven region has near-empty "
                "predicted-positive cells for reference groups."
            ),
        },
    }
    # Build the artifacts path inside the output directory
    artifacts_path: Path = output_dir / "fairness_audit_artifacts.json"
    # Serialise the artifacts to JSON with a stable layout
    artifacts_text: str = json.dumps(
        artifacts, indent=2, ensure_ascii=False, default=str
    )
    # Persist the artifacts to the output directory
    artifacts_path.write_text(artifacts_text, encoding="utf-8")
    # Return the state-ready fairness audit outputs
    return {
        "analysis": analysis,
        "routing": routing,
        "di_eod": di_eod,
        "output_dir": str(output_dir),
    }
````

````python
# Task 16: Bootstrap Inference and DeLong Tests

# ==============================================================================
# Task 16: Bootstrap Inference and DeLong Tests
# ==============================================================================


# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Step 1: Perform stratified paired bootstrap for AUC differences
# -------------------------------------------------------------------------------------------------------------------------------


def stratified_paired_bootstrap_auc(
    predictions_1: np.ndarray,
    predictions_2: np.ndarray,
    y: np.ndarray,
    n_resamples: int = 2000,
    seed: int = 20260830,
) -> Dict[str, Any]:
    """
    Run the stratified paired bootstrap for a Delta-AUC comparison.

    Draws n_resamples with replacement within the positive-class and
    negative-class test strata separately, computes the AUC of both
    models on each resample and the Delta-AUC, and forms the 95%
    percentile confidence interval.

    Parameters
    ----------
    predictions_1 : np.ndarray
        The hybrid test predictions.
    predictions_2 : np.ndarray
        The comparator test predictions.
    y : np.ndarray
        The test labels.
    n_resamples : int, optional
        The resample count (2000 per the config).
    seed : int, optional
        The global deterministic seed.

    Returns
    -------
    Dict[str, Any]
        delta_auc, ci_low, ci_high, auc_1, auc_2 and the resample
        distribution.

    Raises
    ------
    ValueError
        If the inputs are misaligned or a class is absent.
    """
    # Validate the input lengths agree
    if not (len(predictions_1) == len(predictions_2) == len(y)):
        # Raise a descriptive error for the alignment violation
        raise ValueError(
            "predictions and labels must share length."
        )
    # Validate both classes are present for the stratification
    if len(np.unique(y)) != 2:
        # Raise a descriptive error for the missing class
        raise ValueError(
            "both classes must be present for the stratified bootstrap."
        )
    # Build the positive and negative index sets
    pos_idx: np.ndarray = np.flatnonzero(y == 1)
    neg_idx: np.ndarray = np.flatnonzero(y == 0)
    # Build the deterministic generator for the resampling
    rng: np.random.Generator = np.random.default_rng(seed)
    # Compute the full-sample AUCs of both models
    auc_1: float = float(roc_auc_score(y, predictions_1))
    auc_2: float = float(roc_auc_score(y, predictions_2))
    # Collect the resampled Delta-AUC values
    delta_distribution: List[float] = []
    # Iterate over the resamples
    for _ in range(n_resamples):
        # Draw the positive indices with replacement
        sample_pos: np.ndarray = rng.choice(pos_idx, size=len(pos_idx), replace=True)
        # Draw the negative indices with replacement
        sample_neg: np.ndarray = rng.choice(neg_idx, size=len(neg_idx), replace=True)
        # Concatenate the stratified sample indices
        sample: np.ndarray = np.concatenate([sample_pos, sample_neg])
        # Compute the hybrid AUC on the resample
        auc_1_b: float = float(roc_auc_score(y[sample], predictions_1[sample]))
        # Compute the comparator AUC on the resample
        auc_2_b: float = float(roc_auc_score(y[sample], predictions_2[sample]))
        # Record the resampled Delta-AUC
        delta_distribution.append(auc_1_b - auc_2_b)
    # Convert the distribution to an array
    deltas: np.ndarray = np.array(delta_distribution)
    # Compute the point estimate of the Delta-AUC
    delta_auc: float = auc_1 - auc_2
    # Compute the 95% percentile confidence interval
    ci_low: float = float(np.percentile(deltas, 2.5))
    ci_high: float = float(np.percentile(deltas, 97.5))
    # Return the bootstrap result payload
    return {
        "delta_auc": delta_auc,
        "ci_low": ci_low,
        "ci_high": ci_high,
        "auc_1": auc_1,
        "auc_2": auc_2,
        "n_resamples": n_resamples,
        "delta_distribution": deltas.tolist(),
    }


# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Step 2: Implement the DeLong test for correlated ROC curves
# -------------------------------------------------------------------------------------------------------------------------------


def delong_test(
    predictions_1: np.ndarray,
    predictions_2: np.ndarray,
    y: np.ndarray,
) -> Dict[str, float]:
    """
    Run the DeLong test via the Sun-Xu fast midrank algorithm.

    Computes the structural components V10 and V01 from pooled and
    within-group midranks, the AUC estimates (AUC_k = mean(V10_k)),
    the covariance matrix of the two AUC estimates, the z-statistic
    z = (AUC_1 - AUC_2) / sqrt(Var(AUC_1) + Var(AUC_2) - 2*Cov), and
    the one- and two-sided p-values.

    Parameters
    ----------
    predictions_1 : np.ndarray
        The hybrid test predictions.
    predictions_2 : np.ndarray
        The comparator test predictions.
    y : np.ndarray
        The test labels.

    Returns
    -------
    Dict[str, float]
        auc_1, auc_2, var_1, var_2, cov, z, p_one_sided, p_two_sided.

    Raises
    ------
    ValueError
        If the inputs are misaligned or a class is absent.
    """
    # Validate the input lengths agree
    if not (len(predictions_1) == len(predictions_2) == len(y)):
        # Raise a descriptive error for the alignment violation
        raise ValueError(
            "predictions and labels must share length."
        )
    # Validate both classes are present
    if len(np.unique(y)) != 2:
        # Raise a descriptive error for the missing class
        raise ValueError(
            "both classes must be present for the DeLong test."
        )
    # Build the positive and negative masks
    pos_mask: np.ndarray = y == 1
    neg_mask: np.ndarray = y == 0
    # Count the class sizes
    n1: int = int(pos_mask.sum())
    n0: int = int(neg_mask.sum())
    # Compute the pooled midranks per model (Sun-Xu fast midrank)
    midranks_1: np.ndarray = rankdata(predictions_1, method="average")
    midranks_2: np.ndarray = rankdata(predictions_2, method="average")
    # Compute the within-positive midranks per model
    within_1_pos: np.ndarray = rankdata(
        predictions_1[pos_mask], method="average"
    )
    within_2_pos: np.ndarray = rankdata(
        predictions_2[pos_mask], method="average"
    )
    # Compute the within-negative midranks per model
    within_1_neg: np.ndarray = rankdata(
        predictions_1[neg_mask], method="average"
    )
    within_2_neg: np.ndarray = rankdata(
        predictions_2[neg_mask], method="average"
    )
    # Compute the V10 components for positives (Sun-Xu identity)
    v10_1: np.ndarray = (midranks_1[pos_mask] - within_1_pos) / n0
    v10_2: np.ndarray = (midranks_2[pos_mask] - within_2_pos) / n0
    # Compute the V01 components for negatives (Sun-Xu identity)
    v01_1: np.ndarray = (n1 - midranks_1[neg_mask] + within_1_neg) / n1
    v01_2: np.ndarray = (n1 - midranks_2[neg_mask] + within_2_neg) / n1
    # Compute the AUC estimates via the component identity
    auc_1: float = float(np.mean(v10_1))
    auc_2: float = float(np.mean(v10_2))
    # Compute the per-model variances (DeLong decomposition)
    var_1: float = float(np.var(v10_1, ddof=1) / n1 + np.var(v01_1, ddof=1) / n0)
    var_2: float = float(np.var(v10_2, ddof=1) / n1 + np.var(v01_2, ddof=1) / n0)
    # Compute the covariance between the two AUC estimates
    cov: float = float(
        np.cov(v10_1, v10_2, ddof=1)[0, 1] / n1
        + np.cov(v01_1, v01_2, ddof=1)[0, 1] / n0
    )
    # Compute the variance of the AUC difference
    var_diff: float = var_1 + var_2 - 2.0 * cov
    # Guard a zero variance (identical models)
    if var_diff <= 0.0:
        # Return the degenerate result with z = 0
        return {
            "auc_1": auc_1,
            "auc_2": auc_2,
            "var_1": var_1,
            "var_2": var_2,
            "cov": cov,
            "z": 0.0,
            "p_one_sided": 0.5,
            "p_two_sided": 1.0,
        }
    # Compute the z-statistic of the AUC difference
    z: float = (auc_1 - auc_2) / np.sqrt(var_diff)
    # Compute the one-sided p-value: H1: AUC_1 > AUC_2
    p_one_sided: float = float(1.0 - norm.cdf(z))
    # Compute the two-sided p-value
    p_two_sided: float = float(2.0 * (1.0 - norm.cdf(abs(z))))
    # Return the DeLong result payload
    return {
        "auc_1": auc_1,
        "auc_2": auc_2,
        "var_1": var_1,
        "var_2": var_2,
        "cov": cov,
        "z": float(z),
        "p_one_sided": p_one_sided,
        "p_two_sided": p_two_sided,
    }


# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Step 3: Consolidate the inference results
# -------------------------------------------------------------------------------------------------------------------------------


def consolidate_inference_table(
    pairwise: Dict[str, Dict[str, Any]],
) -> pd.DataFrame:
    """
    Assemble the consolidated inference table (Table 9).
    """
    # Collect the table rows
    rows: List[Dict[str, Any]] = []
    # Iterate over the pairwise comparison results
    for comparison, result in pairwise.items():
        # Compose the Table 9 row for the comparison
        rows.append(
            {
                "comparison": comparison,
                "delta_auc": result["bootstrap"]["delta_auc"],
                "ci_low": result["bootstrap"]["ci_low"],
                "ci_high": result["bootstrap"]["ci_high"],
                "p_delong_one_sided": result["delong"]["p_one_sided"],
                "p_delong_two_sided": result["delong"]["p_two_sided"],
            }
        )
    # Build the DataFrame from the rows
    table: pd.DataFrame = pd.DataFrame(rows)
    # Return the consolidated table
    return table


# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


def orchestrate_run_bootstrap_delong(
    config: Dict[str, Any],
    predictions: Dict[str, np.ndarray],
    labels_test: np.ndarray,
    output_dir: Path,
    verify_manuscript_targets: bool = True,
) -> Dict[str, Any]:
    """
    Orchestrate the bootstrap and DeLong inference stage.

    Runs the stratified paired bootstrap and the DeLong test for both
    pairwise comparisons (Hybrid vs LR, Hybrid vs XGBoost), consolidates
    Table 9, and persists it to JSON.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated study configuration dictionary.
    predictions : Dict[str, np.ndarray]
        The test predictions keyed by 'hybrid', 'lr', 'xgb'.
    labels_test : np.ndarray
        The test labels.
    output_dir : Path
        Directory where the artifacts are persisted.
    verify_manuscript_targets : bool, optional
        Whether to assert the manuscript values (True for real runs).

    Returns
    -------
    Dict[str, Any]
        A state-ready dict with pairwise, table_9 and output_dir.
    """
    # Validate the output directory type before any filesystem work
    if not isinstance(output_dir, Path):
        # Raise a descriptive error for a non-Path output directory
        raise ValueError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Create the output directory including any missing parents
    output_dir.mkdir(parents=True, exist_ok=True)
    # Validate the required prediction keys are present
    for key in ("hybrid", "lr", "xgb"):
        # Assert the prediction key exists
        if key not in predictions:
            # Raise a descriptive error naming the missing key
            raise ValueError(
                f"predictions dict missing key '{key}'."
            )
    # Extract the resample counts from the configuration
    bootstrap_params: Dict[str, Any] = config["bootstrap_inference_parameters"]
    n_resamples: int = int(bootstrap_params["auc_difference_resamples"])
    # Extract the global seed from the configuration
    seed: int = int(config["partitioning_parameters"]["random_seed"])
    # Collect the pairwise results
    pairwise: Dict[str, Dict[str, Any]] = {}
    # Iterate over the two pairwise comparisons
    for name, key in (("Hybrid vs Logistic Regression", "lr"),
                      ("Hybrid vs XGBoost", "xgb")):
        # Step 1: run the stratified paired bootstrap for the comparison
        bootstrap_result: Dict[str, Any] = stratified_paired_bootstrap_auc(
            predictions["hybrid"], predictions[key], labels_test,
            n_resamples, seed,
        )
        # Step 2: run the DeLong test for the comparison
        delong_result: Dict[str, float] = delong_test(
            predictions["hybrid"], predictions[key], labels_test
        )
        # Record the comparison's results
        pairwise[name] = {
            "bootstrap": bootstrap_result,
            "delong": delong_result,
        }
    # Step 3: consolidate the inference table
    table_9: pd.DataFrame = consolidate_inference_table(pairwise)
    # Record the no-multiple-comparison decision in the audit log
    logger.info(
        "no multiple-comparison correction applied, consistent with the "
        "descriptive audit intent (Section 5.6)"
    )
    # Build the artifacts path inside the output directory
    artifacts_path: Path = output_dir / "inference_artifacts.json"
    # Serialise the table and the pairwise results to JSON
    artifacts_text: str = json.dumps(
        {
            "table_9": table_9.to_dict(orient="records"),
            "pairwise": pairwise,
            "note": "no multiple-comparison correction applied; "
            "bootstrap CI is the primary interval statement and the "
            "DeLong test the primary hypothesis test",
        },
        indent=2,
        ensure_ascii=False,
        default=str,
    )
    # Persist the artifacts to the output directory
    artifacts_path.write_text(artifacts_text, encoding="utf-8")
    # Return the state-ready inference outputs
    return {
        "pairwise": pairwise,
        "table_9": table_9,
        "output_dir": str(output_dir),
    }
````

````python
# Task 17: TreeSHAP Explanations on the ML-Driven Region

# ==============================================================================
# Task 17: TreeSHAP Explanations on the ML-Driven Region
# ==============================================================================


# The ML-driven region threshold from Equation (7)
RHO_ML_THRESHOLD: float = 0.7


# -------------------------------------------------------------------------------------------------------------------------------
# Task 17, Step 1: Compute TreeSHAP attributions on the residual learner
# -------------------------------------------------------------------------------------------------------------------------------


def compute_tree_shap(
    booster: xgb.Booster,
    best_iteration: int,
    X_ohe_ml: np.ndarray,
) -> Tuple[np.ndarray, float]:
    """
    Compute TreeSHAP attributions on the frozen residual learner.

    Slices the booster to the frozen best iteration, builds a
    shap.TreeExplainer on the residual regressor ONLY (never on the
    full hybrid, preserving the additive decomposition of Equation (5)),
    and computes the SHAP values for the ML-driven region observations.

    Parameters
    ----------
    booster : xgb.Booster
        The fitted residual booster.
    best_iteration : int
        The frozen best iteration from early stopping.
    X_ohe_ml : np.ndarray
        The one-hot encoded matrix restricted to the ML-driven region.

    Returns
    -------
    Tuple[np.ndarray, float]
        The SHAP matrix (n_ML, p) and the base value.

    Raises
    ------
    ValueError
        If the region subset is empty.
    """
    # Validate the region subset is non-empty
    if X_ohe_ml.shape[0] == 0:
        # Raise a descriptive error for the empty region
        raise ValueError(
            "the ML-driven region subset is empty; SHAP cannot be "
            "computed."
        )
    # Slice the booster to the frozen iteration so the explanation
    # covers exactly the model that produced the predictions
    frozen_booster: xgb.Booster = booster[0: best_iteration + 1]
    # Build the TreeSHAP explainer on the residual learner only
    explainer = shap.TreeExplainer(frozen_booster)
    # Compute the SHAP values for the ML-driven observations
    shap_values: np.ndarray = explainer.shap_values(X_ohe_ml)
    # Extract the base (expected) value of the explainer
    base_value: float = float(explainer.expected_value)
    # Verify local accuracy: sum of SHAP plus base equals the prediction
    predictions: np.ndarray = frozen_booster.predict(xgb.DMatrix(X_ohe_ml))
    # Compute the reconstructed predictions from the SHAP values
    reconstructed: np.ndarray = shap_values.sum(axis=1) + base_value
    # Assert local accuracy within a relative tolerance
    if not np.allclose(reconstructed, predictions, rtol=1e-4, atol=1e-6):
        # Raise a descriptive error for the local-accuracy violation
        raise ValueError(
            "TreeSHAP local accuracy violated: the SHAP reconstruction "
            "does not equal the model's predictions."
        )
    # Log the SHAP computation for the audit trail
    logger.info(
        f"TreeSHAP computed: {shap_values.shape[0]} observations x "
        f"{shap_values.shape[1]} features, base={base_value:.5f}"
    )
    # Return the SHAP matrix and the base value
    return shap_values, base_value


# -------------------------------------------------------------------------------------------------------------------------------
# Task 17, Step 2: Rank features by mean absolute SHAP
# -------------------------------------------------------------------------------------------------------------------------------


def rank_features_by_mean_abs_shap(
    shap_values: np.ndarray,
    feature_names: List[str],
    verify_manuscript_targets: bool = True,
) -> Dict[str, Any]:
    """
    Rank features by mean absolute SHAP.

    Computes mean_abs_j = (1/n_ML) * sum_x |phi_j(x)| per feature and
    ranks the features in descending order; optionally verifies the
    manuscript's expected top five.

    Parameters
    ----------
    shap_values : np.ndarray
        The SHAP matrix, shape (n_ML, p).
    feature_names : List[str]
        The frozen one-hot feature names.
    verify_manuscript_targets : bool, optional
        Whether to assert the manuscript top-5 (True for real runs).

    Returns
    -------
    Dict[str, Any]
        The mean-absolute-SHAP series and the descending ranking.

    Raises
    ------
    ValueError
        If the matrix and names are misaligned or non-finite.
    """
    # Validate the matrix and names align
    if shap_values.shape[1] != len(feature_names):
        # Raise a descriptive error for the alignment violation
        raise ValueError(
            f"SHAP matrix has {shap_values.shape[1]} columns but "
            f"{len(feature_names)} feature names were supplied."
        )
    # Assert the SHAP matrix is finite
    if not np.isfinite(shap_values).all():
        # Raise a descriptive error for non-finite values
        raise ValueError(
            "SHAP matrix contains non-finite values."
        )
    # Compute the mean absolute SHAP per feature (the ranking statistic)
    mean_abs: np.ndarray = np.mean(np.abs(shap_values), axis=0)
    # Build the labelled series
    series: pd.Series = pd.Series(mean_abs, index=feature_names)
    # Rank the features in descending order
    ranking: pd.Series = series.sort_values(ascending=False)
    # Verify the manuscript top-5 when enabled
    if verify_manuscript_targets:
        # Declare the expected top-5 features
        expected_top5: List[str] = [
            "cellphone_access_No",
            "age_of_respondent",
            "relationship_with_head_Head of Household",
            "country_Kenya",
            "country_Rwanda",
        ]
        # Extract the observed top-5 feature names
        observed_top5: List[str] = ranking.index[:5].tolist()
        # Assert the observed top-5 equals the expected set
        if set(observed_top5) != set(expected_top5):
            # Raise naming the fidelity-diagnosis checklist
            raise ValueError(
                f"expected top-5 features {expected_top5}; received "
                f"{observed_top5}."
            )
    # Log the top-5 ranking for the audit trail
    logger.info(f"top-5 mean |SHAP|: {ranking.index[:5].tolist()}")
    # Return the series and the ranking
    return {
        "mean_abs_series": series,
        "ranking": ranking,
    }


# -------------------------------------------------------------------------------------------------------------------------------
# Task 17, Step 3: Perform the bootstrap stability screen and conditional diagnostics
# -------------------------------------------------------------------------------------------------------------------------------


def bootstrap_stability_screen(
    shap_values: np.ndarray,
    X_ohe_ml: np.ndarray,
    feature_names: List[str],
    n_resamples: int = 500,
    seed: int = 20260830,
    verify_manuscript_targets: bool = True,
) -> Dict[str, Any]:
    """
    Run the SHAP rank-stability bootstrap and conditional diagnostics.

    Draws n_resamples observation-level bootstrap resamples WITH
    replacement from the ML-driven region, recomputes the mean-absolute
    ranking per resample, computes the per-feature rank standard
    deviation, the top-5 set preservation rate, and the conditional
    mean SHAP diagnostics for the Kenya and Rwanda country dummies.

    Parameters
    ----------
    shap_values : np.ndarray
        The SHAP matrix, shape (n_ML, p).
    X_ohe_ml : np.ndarray
        The one-hot ML-driven matrix for the conditional diagnostics.
    feature_names : List[str]
        The frozen one-hot feature names.
    n_resamples : int, optional
        The resample count (500 per the config).
    seed : int, optional
        The global deterministic seed.
    verify_manuscript_targets : bool, optional
        Whether to assert the manuscript targets (True for real runs).

    Returns
    -------
    Dict[str, Any]
        The rank standard deviations, the set-preservation rate and
        the conditional diagnostics.

    Raises
    ------
    ValueError
        If the inputs are misaligned.
    """
    # Validate the matrix rows agree
    if shap_values.shape[0] != X_ohe_ml.shape[0]:
        # Raise a descriptive error for the alignment violation
        raise ValueError(
            f"SHAP rows {shap_values.shape[0]} must equal matrix rows "
            f"{X_ohe_ml.shape[0]}."
        )
    # Build the deterministic generator for the resampling
    rng: np.random.Generator = np.random.default_rng(seed)
    # Compute the full-sample mean absolute SHAP ranking
    full_mean_abs: np.ndarray = np.mean(np.abs(shap_values), axis=0)
    # Compute the full-sample rank order (descending)
    full_ranks: np.ndarray = np.argsort(-full_mean_abs)
    # Extract the full-sample top-5 feature set
    full_top5: set = set(feature_names[i] for i in full_ranks[:5])
    # Collect the per-resample rank matrices
    rank_matrix: List[np.ndarray] = []
    # Count the resamples preserving the top-5 set
    top5_preserved: int = 0
    # Iterate over the bootstrap resamples
    for _ in range(n_resamples):
        # Draw observation indices with replacement
        sample_idx: np.ndarray = rng.choice(
            shap_values.shape[0], size=shap_values.shape[0], replace=True
        )
        # Compute the resampled mean absolute SHAP
        resampled_mean: np.ndarray = np.mean(
            np.abs(shap_values[sample_idx]), axis=0
        )
        # Compute the resampled rank order (descending)
        resampled_ranks: np.ndarray = np.argsort(-resampled_mean)
        # Record the resampled rank vector
        rank_matrix.append(resampled_ranks)
        # Check the top-5 set preservation
        resampled_top5: set = set(
            feature_names[i] for i in resampled_ranks[:5]
        )
        # Count the preserved top-5 sets
        if resampled_top5 == full_top5:
            # Increment the preservation counter
            top5_preserved += 1
    # Convert the rank matrix to an array of shape (B, p)
    ranks: np.ndarray = np.array(rank_matrix)
    # Compute the per-feature rank standard deviation across resamples
    rank_sd: pd.Series = pd.Series(
        np.std(ranks, axis=0, ddof=1), index=feature_names
    )
    # Compute the top-5 set preservation rate
    preservation_rate: float = top5_preserved / n_resamples
    # Compute the conditional mean SHAP for the Kenya dummy
    kenya_col: Optional[int] = (
        feature_names.index("country_Kenya")
        if "country_Kenya" in feature_names
        else None
    )
    # Compute the conditional mean SHAP for the Rwanda dummy
    rwanda_col: Optional[int] = (
        feature_names.index("country_Rwanda")
        if "country_Rwanda" in feature_names
        else None
    )
    # Build the Kenya indicator mask from the one-hot matrix
    kenya_mask: np.ndarray = (
        X_ohe_ml[:, kenya_col] == 1.0 if kenya_col is not None
        else np.zeros(X_ohe_ml.shape[0], dtype=bool)
    )
    # Build the Rwanda indicator mask from the one-hot matrix
    rwanda_mask: np.ndarray = (
        X_ohe_ml[:, rwanda_col] == 1.0 if rwanda_col is not None
        else np.zeros(X_ohe_ml.shape[0], dtype=bool)
    )
    # Compute the conditional mean SHAP for Kenya among Kenyans
    kenya_conditional: float = (
        float(np.mean(shap_values[kenya_mask, kenya_col]))
        if kenya_col is not None and kenya_mask.any()
        else 0.0
    )
    # Compute the conditional mean SHAP for Rwanda among Rwandans
    rwanda_conditional: float = (
        float(np.mean(shap_values[rwanda_mask, rwanda_col]))
        if rwanda_col is not None and rwanda_mask.any()
        else 0.0
    )
    # Verify the manuscript targets when enabled
    if verify_manuscript_targets:
        # Verify the top-2 ranks are perfectly stable
        if rank_sd.iloc[full_ranks[0]] > 0.005 or rank_sd.iloc[full_ranks[1]] > 0.005:
            # Raise naming the fidelity-diagnosis checklist
            raise ValueError(
                "expected top-2 rank standard deviations of 0.00; "
                f"received {rank_sd.iloc[full_ranks[0]]:.3f}, "
                f"{rank_sd.iloc[full_ranks[1]]:.3f}."
            )
        # Verify the top-5 set preservation rate of 74.0%
        if abs(preservation_rate - 0.74) > 0.03:
            # Raise naming the fidelity-diagnosis checklist
            raise ValueError(
                f"expected top-5 set preservation 0.74 within 0.03; "
                f"received {preservation_rate:.3f}."
            )
        # Verify the Kenya conditional diagnostic of approximately -0.055
        if abs(kenya_conditional - (-0.055)) > 0.01:
            # Raise naming the fidelity-diagnosis checklist
            raise ValueError(
                f"expected Kenya conditional SHAP approx -0.055; "
                f"received {kenya_conditional:.4f}."
            )
        # Verify the Rwanda conditional diagnostic of approximately +0.019
        if abs(rwanda_conditional - 0.019) > 0.01:
            # Raise naming the fidelity-diagnosis checklist
            raise ValueError(
                f"expected Rwanda conditional SHAP approx +0.019; "
                f"received {rwanda_conditional:.4f}."
            )
    # Log the stability summary for the audit trail
    logger.info(
        f"top-5 set preservation {preservation_rate:.3f}; Kenya "
        f"conditional {kenya_conditional:.4f}; Rwanda conditional "
        f"{rwanda_conditional:.4f}"
    )
    # Return the stability report
    return {
        "rank_sd": rank_sd,
        "top5_preservation_rate": preservation_rate,
        "kenya_conditional_shap": kenya_conditional,
        "rwanda_conditional_shap": rwanda_conditional,
    }


# -------------------------------------------------------------------------------------------------------------------------------
# Task 17, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


def orchestrate_run_shap_analysis(
    config: Dict[str, Any],
    booster: xgb.Booster,
    best_iteration: int,
    X_ohe_test: np.ndarray,
    feature_names: List[str],
    regions_test: np.ndarray,
    output_dir: Path,
    verify_manuscript_targets: bool = True,
) -> Dict[str, Any]:
    """
    Orchestrate the TreeSHAP explanation stage.

    Restricts the test matrix to the ML-driven region, composes the
    SHAP computation, the feature ranking and the bootstrap stability
    screen, and persists the aggregated report to JSON.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated study configuration dictionary.
    booster : xgb.Booster
        The fitted residual booster.
    best_iteration : int
        The frozen best iteration.
    X_ohe_test : np.ndarray
        The one-hot encoded test matrix.
    feature_names : List[str]
        The frozen one-hot feature names.
    regions_test : np.ndarray
        The test region assignments.
    output_dir : Path
        Directory where the artifacts are persisted.
    verify_manuscript_targets : bool, optional
        Whether to assert the manuscript targets (True for real runs).

    Returns
    -------
    Dict[str, Any]
        A state-ready dict with shap_values, ranking, stability and
        output_dir.
    """
    # Validate the output directory type before any filesystem work
    if not isinstance(output_dir, Path):
        # Raise a descriptive error for a non-Path output directory
        raise ValueError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Create the output directory including any missing parents
    output_dir.mkdir(parents=True, exist_ok=True)
    # Build the ML-driven region mask (rho < 0.7)
    ml_mask: np.ndarray = regions_test == "ML-driven"
    # Validate the ML-driven region is non-empty
    if not ml_mask.any():
        # Raise a descriptive error for the empty region
        raise ValueError(
            "the ML-driven region is empty on the test partition; "
            "the SHAP analysis cannot run."
        )
    # Restrict the test matrix to the ML-driven region
    X_ohe_ml: np.ndarray = X_ohe_test[ml_mask]
    # Step 1: compute the TreeSHAP attributions on the residual learner
    shap_values, base_value = compute_tree_shap(
        booster, best_iteration, X_ohe_ml
    )
    # Step 2: rank the features by mean absolute SHAP
    ranking_result: Dict[str, Any] = rank_features_by_mean_abs_shap(
        shap_values, feature_names, verify_manuscript_targets
    )
    # Step 3: run the bootstrap stability screen and diagnostics
    stability: Dict[str, Any] = bootstrap_stability_screen(
        shap_values, X_ohe_ml, feature_names,
        n_resamples=int(
            config["shap_explanation_parameters"]["stability_screen_resamples"]
        ),
        seed=int(config["partitioning_parameters"]["random_seed"]),
        verify_manuscript_targets=verify_manuscript_targets,
    )
    # Compose the persisted aggregated report
    report: Dict[str, Any] = {
        "n_ml": int(ml_mask.sum()),
        "base_value": base_value,
        "mean_abs_shap": ranking_result["mean_abs_series"].to_dict(),
        "ranking": ranking_result["ranking"].index.tolist(),
        "rank_sd": {
            name: float(value)
            for name, value in stability["rank_sd"].to_dict().items()
        },
        "top5_preservation_rate": stability["top5_preservation_rate"],
        "kenya_conditional_shap": stability["kenya_conditional_shap"],
        "rwanda_conditional_shap": stability["rwanda_conditional_shap"],
        "interpretation": (
            "SHAP values are contributions to the residual learner "
            "r_hat(x) in the margin space, preserving the additive "
            "decomposition of Equation (5); they are NOT probability-"
            "scale attributions."
        ),
    }
    # Build the artifacts path inside the output directory
    artifacts_path: Path = output_dir / "shap_artifacts.json"
    # Serialise the aggregated report to JSON with a stable layout
    artifacts_text: str = json.dumps(
        report, indent=2, ensure_ascii=False, default=str
    )
    # Persist the report to the output directory
    artifacts_path.write_text(artifacts_text, encoding="utf-8")
    # Return the state-ready SHAP outputs
    return {
        "shap_values": shap_values,
        "ranking": ranking_result["ranking"],
        "stability": stability,
        "output_dir": str(output_dir),
    }
````

````python
# Task 18: Thin-File Segmentation Analysis

# ==============================================================================
# Task 18: Thin-File Segmentation Analysis
# ==============================================================================


# The composite index components
INDEX_COMPONENTS: Tuple[str, ...] = ("education", "cellphone", "formal_job")


# -------------------------------------------------------------------------------------------------------------------------------
# Task 18, Step 1: Construct the thin-file composite index
# -------------------------------------------------------------------------------------------------------------------------------


def construct_thin_file_index(
    train_df: pd.DataFrame, config: Dict[str, Any]
) -> Tuple[np.ndarray, Dict[str, Dict[str, float]], float]:
    """
    Construct the thin-file composite index on the training partition.

    Maps education_level to the config's ordinal scale, builds the
    cellphone binary and the formal-job indicator, z-score standardizes
    each component on the training partition via
    z_j = (x_j - mean_j_train) / std_j_train, forms the index as the
    sum of the three standardized components, and freezes the training
    25th percentile threshold tau.

    Parameters
    ----------
    train_df : pd.DataFrame
        The training partition of the Zindi frame.
    config : Dict[str, Any]
        The validated study configuration dictionary.

    Returns
    -------
    Tuple[np.ndarray, Dict[str, Dict[str, float]], float]
        The training index vector, the frozen standardization stats
        (mean and std per component) and the frozen threshold tau.

    Raises
    ------
    ValueError
        If a component standard deviation is zero.
    """
    # Extract the thin-file parameters from the configuration
    thin_file: Dict[str, Any] = config["thin_file_parameters"]
    # Extract the ordinal education mapping
    education_map: Dict[str, int] = thin_file["education_ordinal_mapping"]
    # Assert every observed education level is covered by the mapping
    if not set(train_df["education_level"]).issubset(education_map):
        # Raise naming the uncovered education levels
        raise ValueError(
            f"uncovered education levels: "
            f"{set(train_df['education_level']) - set(education_map)}."
        )
    # Build the ordinal education component
    education: pd.Series = train_df["education_level"].map(education_map)
    # Build the cellphone binary component: 1 if Yes else 0
    cellphone: pd.Series = (train_df["cellphone_access"] == "Yes").astype(
        np.float64
    )
    # Extract the formal-job indicator mapping
    job_map: Dict[str, int] = thin_file["formal_job_indicator_mapping"]
    # Build the formal-job indicator component
    formal_job: pd.Series = train_df["job_type"].map(job_map).astype(
        np.float64
    )
    # Collect the raw components
    components: Dict[str, pd.Series] = {
        "education": education,
        "cellphone": cellphone,
        "formal_job": formal_job,
    }
    # Collect the frozen standardization statistics
    stats: Dict[str, Dict[str, float]] = {}
    # Compute the per-component z-score statistics on training only
    for name in INDEX_COMPONENTS:
        # Compute the training mean of the component
        mean: float = float(components[name].mean())
        # Compute the training standard deviation of the component
        std: float = float(components[name].std(ddof=0))
        # Guard a zero standard deviation
        if std == 0.0:
            # Raise a descriptive error for the degenerate component
            raise ValueError(
                f"component '{name}' has zero training variance; the "
                f"z-score is undefined."
            )
        # Freeze the component's statistics
        stats[name] = {"mean": mean, "std": std}
    # Compute the z-scored components and the index sum
    index: np.ndarray = np.zeros(len(train_df), dtype=np.float64)
    # Iterate over the components to accumulate the z-scores
    for name in INDEX_COMPONENTS:
        # Compute the component's z-scores: z = (x - mean)/std
        z_scores: np.ndarray = (
            components[name].to_numpy(dtype=np.float64) - stats[name]["mean"]
        ) / stats[name]["std"]
        # Accumulate the z-scores into the composite index
        index += z_scores
    # Compute the frozen training 25th percentile threshold
    tau: float = float(np.percentile(index, 25))
    # Log the frozen statistics for the audit trail
    logger.info(
        f"thin-file stats frozen: {stats}, tau={tau:.5f}; all mappings "
        f"are UNSPECIFIED IN MANUSCRIPT placeholders"
    )
    # Return the index, the stats and the threshold
    return index, stats, tau


# -------------------------------------------------------------------------------------------------------------------------------
# Task 18, Step 2: Apply the threshold to the test partition and compute subgroup performance
# -------------------------------------------------------------------------------------------------------------------------------


def apply_threshold_and_compute(
    test_df: pd.DataFrame,
    frozen_stats: Dict[str, Dict[str, float]],
    tau: float,
    model_probs_test: Dict[str, np.ndarray],
    y_test: np.ndarray,
    config: Dict[str, Any],
    verify_manuscript_targets: bool = True,
) -> Dict[str, Any]:
    """
    Apply the frozen threshold and compute subgroup performance.

    Applies the frozen standardization statistics and tau to the test
    partition, defines the thin-file subgroup as index < tau, and
    computes AUC, Brier, ECE and pairwise Delta-AUC bootstrap CIs on
    the subgroup.

    Parameters
    ----------
    test_df : pd.DataFrame
        The test partition frame.
    frozen_stats : Dict[str, Dict[str, float]]
        The frozen standardization statistics from Step 1.
    tau : float
        The frozen training threshold.
    model_probs_test : Dict[str, np.ndarray]
        The per-model test probabilities keyed by 'lr', 'xgb',
        'hybrid_cal'.
    y_test : np.ndarray
        The test labels.
    config : Dict[str, Any]
        The validated study configuration dictionary.
    verify_manuscript_targets : bool, optional
        Whether to assert the manuscript profile (True for real runs).

    Returns
    -------
    Dict[str, Any]
        The subgroup mask, profile and performance metrics.

    Raises
    ------
    ValueError
        If the subgroup is empty.
    """
    # Extract the thin-file parameters from the configuration
    thin_file: Dict[str, Any] = config["thin_file_parameters"]
    # Extract the ordinal education mapping
    education_map: Dict[str, int] = thin_file["education_ordinal_mapping"]
    # Build the ordinal education component on the test partition
    education: pd.Series = test_df["education_level"].map(education_map)
    # Build the cellphone binary component on the test partition
    cellphone: pd.Series = (test_df["cellphone_access"] == "Yes").astype(
        np.float64
    )
    # Extract the formal-job indicator mapping
    job_map: Dict[str, int] = thin_file["formal_job_indicator_mapping"]
    # Build the formal-job indicator component on the test partition
    formal_job: pd.Series = test_df["job_type"].map(job_map).astype(
        np.float64
    )
    # Collect the raw components
    components: Dict[str, pd.Series] = {
        "education": education,
        "cellphone": cellphone,
        "formal_job": formal_job,
    }
    # Build the frozen test index
    test_index: np.ndarray = np.zeros(len(test_df), dtype=np.float64)
    # Iterate over the components to accumulate the frozen z-scores
    for name in INDEX_COMPONENTS:
        # Compute the frozen z-scores using the training statistics
        z_scores: np.ndarray = (
            components[name].to_numpy(dtype=np.float64)
            - frozen_stats[name]["mean"]
        ) / frozen_stats[name]["std"]
        # Accumulate the z-scores into the test index
        test_index += z_scores
    # Define the thin-file subgroup: index below the frozen threshold
    subgroup_mask: np.ndarray = test_index < tau
    # Validate the subgroup is non-empty
    if not subgroup_mask.any():
        # Raise a descriptive error for the empty subgroup
        raise ValueError(
            "the thin-file subgroup is empty on the test partition."
        )
    # Compute the subgroup profile
    n_subgroup: int = int(subgroup_mask.sum())
    share: float = n_subgroup / len(test_df)
    positive_rate: float = float(np.mean(y_test[subgroup_mask]))
    # Compose the profile
    profile: Dict[str, Any] = {
        "n": n_subgroup,
        "share": share,
        "positive_rate": positive_rate,
    }
    # Assert the manuscript profile when verification is enabled
    if verify_manuscript_targets:
        # Assert the subgroup size within 20 observations
        if abs(n_subgroup - 1559) > 20:
            # Raise naming the fidelity-diagnosis checklist
            raise ValueError(
                f"expected thin-file n=1559 within 20; received "
                f"{n_subgroup}."
            )
        # Assert the share within 1 percentage point
        if abs(share - 0.331) > 0.01:
            # Raise naming the fidelity-diagnosis checklist
            raise ValueError(
                f"expected thin-file share 0.331 within 0.01; received "
                f"{share:.4f}."
            )
        # Assert the positive rate within 0.5 percentage points
        if abs(positive_rate - 0.024) > 0.005:
            # Raise naming the fidelity-diagnosis checklist
            raise ValueError(
                f"expected thin-file positive rate 0.024 within 0.005; "
                f"received {positive_rate:.4f}."
            )
    # Compute the subgroup performance metrics per model
    metrics: Dict[str, Dict[str, float]] = {}
    # Iterate over the candidate models
    for model_name, probs in model_probs_test.items():
        # Assert the probability vector aligns with the labels
        if len(probs) != len(y_test):
            # Raise a descriptive error for the alignment violation
            raise ValueError(
                f"model '{model_name}' probabilities length {len(probs)} "
                f"must equal labels length {len(y_test)}."
            )
        # Restrict the probabilities to the subgroup
        probs_sub: np.ndarray = probs[subgroup_mask]
        # Compute the subgroup AUC
        auc: float = float(roc_auc_score(y_test[subgroup_mask], probs_sub))
        # Compute the subgroup Brier score
        brier: float = float(
            brier_score_loss(y_test[subgroup_mask], probs_sub)
        )
        # Compute the subgroup ECE with K = 10 bins
        ece: float = float(
            compute_ece(probs_sub, y_test[subgroup_mask], n_bins=10)
        )
        # Record the model's subgroup metrics
        metrics[model_name] = {"auc": auc, "brier": brier, "ece": ece}
    # Compute the pairwise Delta-AUC bootstrap CIs on the subgroup
    bootstrap_params: Dict[str, Any] = config["bootstrap_inference_parameters"]
    n_resamples: int = int(bootstrap_params["auc_difference_resamples"])
    # Extract the global seed from the configuration
    seed: int = int(config["partitioning_parameters"]["random_seed"])
    # Build the subgroup-restricted hybrid-vs-LR bootstrap
    delta_hybrid_lr: Dict[str, Any] = stratified_paired_bootstrap_auc(
        model_probs_test["hybrid_cal"][subgroup_mask],
        model_probs_test["lr"][subgroup_mask],
        y_test[subgroup_mask],
        n_resamples=n_resamples,
        seed=seed,
    )
    # Record the pairwise Delta-AUC result
    pairwise: Dict[str, Any] = {
        "hybrid_vs_lr": {
            "delta_auc": delta_hybrid_lr["delta_auc"],
            "ci_low": delta_hybrid_lr["ci_low"],
            "ci_high": delta_hybrid_lr["ci_high"],
        }
    }
    # Log the subgroup summary for the audit trail
    logger.info(
        f"thin-file subgroup: n={n_subgroup}, share={share:.4f}, "
        f"rate={positive_rate:.4f}"
    )
    # Return the subgroup report
    return {
        "subgroup_mask": subgroup_mask,
        "profile": profile,
        "metrics": metrics,
        "pairwise": pairwise,
    }


# -------------------------------------------------------------------------------------------------------------------------------
# Task 18, Step 3: Test the two pre-registered hypotheses
# -------------------------------------------------------------------------------------------------------------------------------


def test_hypotheses(
    subgroup_metrics: Dict[str, Dict[str, float]],
    full_test_metrics: Dict[str, Dict[str, float]],
    subgroup_mean_rho: float,
    full_mean_rho: float,
) -> Dict[str, Any]:
    """
    Test the two pre-registered thin-file hypotheses.

    H1: the hybrid AUC gain over LR is larger on the thin-file subgroup
    than on the full population (supported by construction when the
    subgroup delta exceeds the full delta).  H2: the rho distribution is
    shifted toward smaller values on the subgroup (not supported when
    the subgroup mean rho is not below the full mean).

    Parameters
    ----------
    subgroup_metrics : Dict[str, Dict[str, float]]
        The subgroup metrics keyed by model.
    full_test_metrics : Dict[str, Dict[str, float]]
        The full-test metrics keyed by model.
    subgroup_mean_rho : float
        The mean rho on the thin-file subgroup.
    full_mean_rho : float
        The mean rho on the full test set.

    Returns
    -------
    Dict[str, Any]
        The H1 and H2 verdicts with their evidence.

    Raises
    ------
    ValueError
        If a required model key is missing.
    """
    # Validate the required model keys are present
    for metrics in (subgroup_metrics, full_test_metrics):
        # Assert the hybrid and LR keys exist
        for key in ("hybrid_cal", "lr"):
            # Raise a descriptive error for a missing key
            if key not in metrics:
                # Name the missing key in the error
                raise ValueError(
                    f"metrics dict missing key '{key}'."
                )
    # Compute the subgroup hybrid-vs-LR AUC delta
    subgroup_delta: float = (
        subgroup_metrics["hybrid_cal"]["auc"]
        - subgroup_metrics["lr"]["auc"]
    )
    # Compute the full-population hybrid-vs-LR AUC delta
    full_delta: float = (
        full_test_metrics["hybrid_cal"]["auc"]
        - full_test_metrics["lr"]["auc"]
    )
    # H1 is supported when the subgroup delta exceeds the full delta
    h1_supported: bool = subgroup_delta > full_delta
    # Compose the H1 verdict
    h1: Dict[str, Any] = {
        "supported": h1_supported,
        "subgroup_delta_auc": subgroup_delta,
        "full_delta_auc": full_delta,
        "interpretation": (
            "the residual learner contributes more where the linear "
            "scorecard has the thinnest signal"
        ),
    }
    # H2 is supported when the subgroup mean rho is BELOW the full mean
    h2_supported: bool = subgroup_mean_rho < full_mean_rho
    # Compose the H2 verdict
    h2: Dict[str, Any] = {
        "supported": h2_supported,
        "subgroup_mean_rho": subgroup_mean_rho,
        "full_mean_rho": full_mean_rho,
        "interpretation": (
            "the thin-file population is confidently predicted negative "
            "by the LR baseline, so u_LR(x) and alpha*(x) remain low and "
            "the residual correction is suppressed; H2 was pre-registered "
            "but is NOT supported when the subgroup rho is not shifted "
            "downward"
        ),
    }
    # Log the hypothesis verdicts for the audit trail
    logger.info(
        f"H1 supported={h1_supported} (subgroup {subgroup_delta:.4f} vs "
        f"full {full_delta:.4f}); H2 supported={h2_supported} "
        f"(subgroup rho {subgroup_mean_rho:.4f} vs full "
        f"{full_mean_rho:.4f})"
    )
    # Return the hypothesis verdicts
    return {"h1": h1, "h2": h2}


# -------------------------------------------------------------------------------------------------------------------------------
# Task 18, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


def orchestrate_run_thin_file(
    config: Dict[str, Any],
    train_df: pd.DataFrame,
    test_df: pd.DataFrame,
    model_probs_test: Dict[str, np.ndarray],
    y_test: np.ndarray,
    rho_test: np.ndarray,
    full_test_metrics: Dict[str, Dict[str, float]],
    output_dir: Path,
    verify_manuscript_targets: bool = True,
) -> Dict[str, Any]:
    """
    Orchestrate the thin-file segmentation analysis stage.

    Composes the index construction, the subgroup performance
    computation and the hypothesis tests, persisting the report to
    JSON.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated study configuration dictionary.
    train_df : pd.DataFrame
        The training partition frame.
    test_df : pd.DataFrame
        The test partition frame.
    model_probs_test : Dict[str, np.ndarray]
        The per-model test probabilities.
    y_test : np.ndarray
        The test labels.
    rho_test : np.ndarray
        The test rho vector.
    full_test_metrics : Dict[str, Dict[str, float]]
        The full-test metrics from Task 14.
    output_dir : Path
        Directory where the artifacts are persisted.
    verify_manuscript_targets : bool, optional
        Whether to assert the manuscript targets (True for real runs).

    Returns
    -------
    Dict[str, Any]
        A state-ready dict with frozen_stats, tau, subgroup and
        hypotheses.
    """
    # Validate the output directory type before any filesystem work
    if not isinstance(output_dir, Path):
        # Raise a descriptive error for a non-Path output directory
        raise ValueError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Create the output directory including any missing parents
    output_dir.mkdir(parents=True, exist_ok=True)
    # Step 1: construct the composite index on the training partition
    train_index, frozen_stats, tau = construct_thin_file_index(
        train_df, config
    )
    # Step 2: apply the frozen threshold and compute subgroup performance
    subgroup: Dict[str, Any] = apply_threshold_and_compute(
        test_df, frozen_stats, tau, model_probs_test, y_test, config,
        verify_manuscript_targets,
    )
    # Compute the mean rho on the subgroup and the full test set
    subgroup_mean_rho: float = float(
        np.mean(rho_test[subgroup["subgroup_mask"]])
    )
    # Compute the full-test mean rho
    full_mean_rho: float = float(np.mean(rho_test))
    # Step 3: test the two pre-registered hypotheses
    hypotheses: Dict[str, Any] = test_hypotheses(
        subgroup["metrics"], full_test_metrics,
        subgroup_mean_rho, full_mean_rho,
    )
    # Compose the persisted report
    report: Dict[str, Any] = {
        "frozen_stats": frozen_stats,
        "tau": tau,
        "profile": subgroup["profile"],
        "metrics": subgroup["metrics"],
        "pairwise": subgroup["pairwise"],
        "mean_rho": {
            "subgroup": subgroup_mean_rho,
            "full_test": full_mean_rho,
        },
        "hypotheses": hypotheses,
    }
    # Build the artifacts path inside the output directory
    artifacts_path: Path = output_dir / "thin_file_artifacts.json"
    # Serialise the report to JSON with a stable layout
    artifacts_text: str = json.dumps(
        report, indent=2, ensure_ascii=False, default=str
    )
    # Persist the report to the output directory
    artifacts_path.write_text(artifacts_text, encoding="utf-8")
    # Return the state-ready thin-file outputs
    return {
        "frozen_stats": frozen_stats,
        "tau": tau,
        "subgroup": subgroup,
        "hypotheses": hypotheses,
        "output_dir": str(output_dir),
    }
````

````python
# Task 19: Alternating-Minimization Joint Optimization

# ==============================================================================
# Task 19: Alternating-Minimization Joint Optimization
# ==============================================================================


def _sigmoid(z: np.ndarray) -> np.ndarray:
    """
    Compute the logistic sigmoid: sigma(z) = 1 / (1 + exp(-z)).
    """
    # Clip the argument to prevent overflow in the exponential
    z_clipped: np.ndarray = np.clip(z, -500.0, 500.0)
    # Compute the sigmoid (Equation 1)
    return 1.0 / (1.0 + np.exp(-z_clipped))


def _joint_loss(
    beta_full: np.ndarray,
    X_woe: np.ndarray,
    X_ohe: np.ndarray,
    y: np.ndarray,
    booster: xgb.Booster,
    best_iteration: int,
    alpha_min: float,
    alpha_max: float,
    gamma: float,
    lambda_lr: float,
) -> float:
    """
    Compute the joint loss L(beta, f) of Equation (12).

    Implements L(beta, f) = sum_i ell(y_i, clip(p_LR(x_i; beta) +
    alpha*(x_i; beta) * f(x_i), 0, 1)) + lambda_LR * ||beta||_2^2 with
    ell(y, p) = -[y ln p + (1 - y) ln(1 - p)].  The MANDATORY alpha*
    recomputation happens inside this closure: p_LR is recomputed from
    the current beta, then u_LR (Equation 8) and alpha* (Equation 9)
    before the fused prediction.

    Parameters
    ----------
    beta_full : np.ndarray
        The full parameter vector [beta_0, beta_1..beta_p].
    X_woe : np.ndarray
        The WoE training matrix.
    X_ohe : np.ndarray
        The one-hot training matrix.
    y : np.ndarray
        The training labels.
    booster : xgb.Booster
        The current residual booster.
    best_iteration : int
        The booster's frozen best iteration.
    alpha_min, alpha_max, gamma : float
        The frozen adaptive weighting hyperparameters.
    lambda_lr : float
        The L2 penalty on the coefficients.

    Returns
    -------
    float
        The joint loss value.
    """
    # Split the full vector into the intercept and the coefficients
    beta_0: float = float(beta_full[0])
    beta: np.ndarray = beta_full[1:]
    # Compute the linear predictor: z = beta^T x + beta_0 (Equation 1)
    z: np.ndarray = X_woe @ beta + beta_0
    # Compute the LR probabilities from the CURRENT beta (Eq. 1)
    p_lr: np.ndarray = _sigmoid(z)
    # Compute the uncertainty proxy from the current p_LR (Eq. 8)
    u_lr: np.ndarray = 4.0 * p_lr * (1.0 - p_lr)
    # Compute the adaptive weight from the current beta (Eq. 9)
    alpha_star: np.ndarray = (
        alpha_min + (alpha_max - alpha_min) * np.power(u_lr, gamma)
    )
    # Predict the residual corrections with the current booster
    r_hat: np.ndarray = booster.predict(
        xgb.DMatrix(X_ohe), iteration_range=(0, best_iteration + 1)
    )
    # Form the fused prediction with clipping (Equation 5 / 12)
    fused: np.ndarray = np.clip(p_lr + alpha_star * r_hat, 1e-12, 1.0 - 1e-12)
    # Compute the binary cross-entropy: ell(y, p) (Equation 12)
    cross_entropy: float = float(
        -np.mean(y * np.log(fused) + (1.0 - y) * np.log(1.0 - fused))
    )
    # Compute the L2 penalty term: lambda_LR * ||beta||_2^2 (Eq. 12)
    penalty: float = lambda_lr * float(np.dot(beta, beta))
    # Return the joint loss
    return cross_entropy + penalty


# -------------------------------------------------------------------------------------------------------------------------------
# Task 19, Step 1: Implement the fixed-point iteration
# -------------------------------------------------------------------------------------------------------------------------------


def implement_fixed_point_iteration(
    X_woe_train: np.ndarray,
    X_ohe_train: np.ndarray,
    y_train: np.ndarray,
    X_woe_val: np.ndarray,
    X_ohe_val: np.ndarray,
    p_lr_val_initial: np.ndarray,
    y_val: np.ndarray,
    beta_initial: np.ndarray,
    beta_0_initial: float,
    booster_initial: xgb.Booster,
    best_iteration_initial: int,
    config: Dict[str, Any],
    tolerance: float = 1e-3,
    max_iterations: int = 10,
) -> Dict[str, Any]:
    """
    Implement the alternating fixed-point iteration on the joint loss.

    Alternates: (i) hold f fixed and update beta by L-BFGS-B on the
    joint loss (Equation 12) with the mandatory alpha* recomputation;
    (ii) recompute the residuals r_i = y_i - p_LR(x_i; beta); (iii)
    refit the residual booster on the new residuals with the same
    hyperparameters; (iv) evaluate the validation AUC after each full
    cycle, halting when the change falls below the tolerance.

    Parameters
    ----------
    X_woe_train : np.ndarray
        The WoE training matrix.
    X_ohe_train : np.ndarray
        The one-hot training matrix.
    y_train : np.ndarray
        The training labels.
    X_woe_val : np.ndarray
        The WoE validation matrix.
    X_ohe_val : np.ndarray
        The one-hot validation matrix.
    p_lr_val_initial : np.ndarray
        The initial validation LR probabilities.
    y_val : np.ndarray
        The validation labels.
    beta_initial : np.ndarray
        The initial coefficient vector from two-stage training.
    beta_0_initial : float
        The initial intercept from two-stage training.
    booster_initial : xgb.Booster
        The initial residual booster from two-stage training.
    best_iteration_initial : int
        The initial booster's frozen best iteration.
    config : Dict[str, Any]
        The validated study configuration dictionary.
    tolerance : float, optional
        The convergence tolerance on the validation AUC change (1e-3).
    max_iterations : int, optional
        The fixed-point iteration cap.

    Returns
    -------
    Dict[str, Any]
        The final beta, beta_0, booster, best iteration, convergence
        history and per-iteration timings.

    Raises
    ------
    ValueError
        If the iteration fails to converge within the cap.
    """
    # Extract the adaptive weighting hyperparameters from the config
    adaptive: Dict[str, Any] = config["adaptive_weighting_parameters"]
    # Extract the selected triple from the config
    selected: Dict[str, float] = adaptive["selected_on_zindi"]
    # Unpack the selected hyperparameters
    alpha_min: float = float(selected["alpha_min"])
    alpha_max: float = float(selected["alpha_max"])
    gamma: float = float(selected["gamma"])
    # Extract the selected logistic C for the L2 penalty convention
    # (sklearn's penalty is (1 / (2C)) * ||beta||^2)
    C_selected: float = float(config["logistic_regression_parameters"]["C_grid"][0])
    # Compute the L2 penalty coefficient
    lambda_lr: float = 1.0 / (2.0 * C_selected)
    # Initialise the current parameters from the two-stage outputs
    beta: np.ndarray = beta_initial.copy()
    beta_0: float = beta_0_initial
    booster: xgb.Booster = booster_initial
    best_iteration: int = best_iteration_initial
    # Initialise the validation LR probabilities from the two-stage fit
    p_lr_val: np.ndarray = p_lr_val_initial.copy()
    # Collect the convergence history
    history: List[Dict[str, Any]] = []
    # Track the previous validation AUC for the tolerance check
    previous_val_auc: Optional[float] = None
    # Run the fixed-point iteration
    for iteration in range(max_iterations):
        # Record the cycle start time
        start_time: float = time.perf_counter()
        # Step (i): hold f fixed and update beta by L-BFGS-B on the loss
        result = minimize(
            fun=_joint_loss,
            x0=np.concatenate([[beta_0], beta]),
            args=(
                X_woe_train, X_ohe_train, y_train, booster,
                best_iteration, alpha_min, alpha_max, gamma, lambda_lr,
            ),
            method="L-BFGS-B",
            options={"maxiter": 200},
        )
        # Raise if the optimizer failed
        if not result.success:
            # Raise a descriptive error for the optimization failure
            raise ValueError(
                f"L-BFGS-B failed at iteration {iteration}: "
                f"{result.message}."
            )
        # Extract the updated full parameter vector
        beta_full: np.ndarray = np.asarray(result.x)
        # Update the intercept and the coefficients
        beta_0 = float(beta_full[0])
        beta = beta_full[1:]
        # Step (ii): recompute the training LR probabilities
        p_lr_train: np.ndarray = _sigmoid(X_woe_train @ beta + beta_0)
        # Recompute the residuals under the updated beta (Equation 3)
        residuals_train: np.ndarray = y_train - p_lr_train
        # Recompute the validation LR probabilities under the updated
        # beta (the mandatory alpha* recomputation requires them)
        p_lr_val = _sigmoid(X_woe_val @ beta + beta_0)
        # Step (iii): refit the residual booster on the new residuals
        booster, best_iteration = fit_residual_booster(
            X_ohe_train, residuals_train, X_ohe_val, p_lr_val, y_val,
            config,
        )
        # Predict the validation residual corrections
        r_hat_val: np.ndarray = booster.predict(
            xgb.DMatrix(X_ohe_val), iteration_range=(0, best_iteration + 1)
        )
        # Recompute the validation alpha* from the updated beta
        u_val: np.ndarray = 4.0 * p_lr_val * (1.0 - p_lr_val)
        # Compute the validation adaptive weights (Equation 9)
        alpha_val: np.ndarray = (
            alpha_min + (alpha_max - alpha_min) * np.power(u_val, gamma)
        )
        # Step (iv): form the validation hybrid and compute its AUC
        hybrid_val: np.ndarray = np.clip(
            p_lr_val + alpha_val * r_hat_val, 0.0, 1.0
        )
        # Compute the validation AUC of the updated hybrid
        val_auc: float = float(roc_auc_score(y_val, hybrid_val))
        # Record the cycle end time
        elapsed: float = time.perf_counter() - start_time
        # Record the iteration's convergence entry
        history.append(
            {
                "iteration": iteration,
                "val_auc": val_auc,
                "elapsed_seconds": elapsed,
                "beta_norm": float(np.dot(beta, beta)),
            }
        )
        # Log the iteration summary for the audit trail
        logger.info(
            f"iteration {iteration}: val AUC {val_auc:.5f} "
            f"({elapsed:.2f}s)"
        )
        # Check the convergence criterion on the validation AUC change
        if previous_val_auc is not None:
            # Compute the validation AUC change
            auc_change: float = abs(val_auc - previous_val_auc)
            # Halt when the change falls below the tolerance
            if auc_change < tolerance:
                # Log the convergence for the audit trail
                logger.info(
                    f"converged at iteration {iteration}: "
                    f"|Delta val AUC| = {auc_change:.6f} < {tolerance}"
                )
                # Return the converged parameters and history
                return {
                    "beta": beta,
                    "beta_0": beta_0,
                    "booster": booster,
                    "best_iteration": best_iteration,
                    "history": history,
                    "val_auc_final": val_auc,
                }
        # Update the previous validation AUC
        previous_val_auc = val_auc
    # Raise if the iteration cap was reached without convergence
    raise ValueError(
        f"alternating minimization did not converge within "
        f"{max_iterations} iterations."
    )


# -------------------------------------------------------------------------------------------------------------------------------
# Task 19, Step 2: Run and record the convergence trajectory
# -------------------------------------------------------------------------------------------------------------------------------


def run_and_record_trajectory(
    fixed_point_result: Dict[str, Any],
    X_woe_test: np.ndarray,
    X_ohe_test: np.ndarray,
    y_test: np.ndarray,
    two_stage_test_auc: float,
    two_stage_fit_seconds: float,
    config: Dict[str, Any],
    verify_manuscript_targets: bool = True,
) -> Dict[str, Any]:
    """
    Record the convergence trajectory and the final comparison.

    Computes the final test-set Delta-AUC of the joint scheme against
    the two-stage model and the wall-clock ratio, and optionally
    verifies the manuscript targets (3 iterations, +0.0006, ~4x).

    Parameters
    ----------
    fixed_point_result : Dict[str, Any]
        The converged fixed-point result from Step 1.
    X_woe_test : np.ndarray
        The WoE test matrix.
    X_ohe_test : np.ndarray
        The one-hot test matrix.
    y_test : np.ndarray
        The test labels.
    two_stage_test_auc : float
        The two-stage hybrid test AUC.
    two_stage_fit_seconds : float
        The two-stage fit wall-clock time.
    config : Dict[str, Any]
        The validated study configuration dictionary.
    verify_manuscript_targets : bool, optional
        Whether to assert the manuscript targets (True for real runs).

    Returns
    -------
    Dict[str, Any]
        The trajectory summary, the Delta-AUC and the wall-clock ratio.
    """
    # Unpack the converged parameters
    beta: np.ndarray = fixed_point_result["beta"]
    beta_0: float = fixed_point_result["beta_0"]
    booster: xgb.Booster = fixed_point_result["booster"]
    best_iteration: int = fixed_point_result["best_iteration"]
    # Extract the adaptive weighting hyperparameters from the config
    selected: Dict[str, float] = config["adaptive_weighting_parameters"][
        "selected_on_zindi"
    ]
    # Unpack the selected hyperparameters
    alpha_min: float = float(selected["alpha_min"])
    alpha_max: float = float(selected["alpha_max"])
    gamma: float = float(selected["gamma"])
    # Compute the test LR probabilities under the final beta
    p_lr_test: np.ndarray = _sigmoid(X_woe_test @ beta + beta_0)
    # Compute the test uncertainty (Equation 8)
    u_test: np.ndarray = 4.0 * p_lr_test * (1.0 - p_lr_test)
    # Compute the test adaptive weights (Equation 9)
    alpha_test: np.ndarray = (
        alpha_min + (alpha_max - alpha_min) * np.power(u_test, gamma)
    )
    # Predict the test residual corrections
    r_hat_test: np.ndarray = booster.predict(
        xgb.DMatrix(X_ohe_test), iteration_range=(0, best_iteration + 1)
    )
    # Form the joint test hybrid
    hybrid_test: np.ndarray = np.clip(
        p_lr_test + alpha_test * r_hat_test, 0.0, 1.0
    )
    # Compute the joint test AUC
    joint_test_auc: float = float(roc_auc_score(y_test, hybrid_test))
    # Compute the final test-set Delta-AUC against two-stage
    delta_auc: float = joint_test_auc - two_stage_test_auc
    # Compute the total joint wall-clock time
    joint_seconds: float = sum(
        entry["elapsed_seconds"] for entry in fixed_point_result["history"]
    )
    # Compute the wall-clock ratio against the two-stage fit
    wall_clock_ratio: float = (
        joint_seconds / two_stage_fit_seconds
        if two_stage_fit_seconds > 0
        else 0.0
    )
    # Count the iterations to convergence
    n_iterations: int = len(fixed_point_result["history"])
    # Verify the manuscript targets when enabled
    if verify_manuscript_targets:
        # Verify the iteration count of 3
        if n_iterations != 3:
            # Raise naming the fidelity-diagnosis checklist
            raise ValueError(
                f"expected 3 iterations to convergence; received "
                f"{n_iterations}."
            )
        # Verify the Delta-AUC of +0.0006 within 0.001
        if abs(delta_auc - 0.0006) > 0.001:
            # Raise naming the fidelity-diagnosis checklist
            raise ValueError(
                f"expected joint Delta-AUC +0.0006 within 0.001; "
                f"received {delta_auc:.5f}."
            )
    # Log the trajectory summary for the audit trail
    logger.info(
        f"joint scheme: {n_iterations} iterations, test Delta-AUC "
        f"{delta_auc:+.5f}, wall-clock ratio {wall_clock_ratio:.2f}x"
    )
    # Return the trajectory summary
    return {
        "n_iterations": n_iterations,
        "val_auc_sequence": [
            entry["val_auc"] for entry in fixed_point_result["history"]
        ],
        "joint_test_auc": joint_test_auc,
        "two_stage_test_auc": two_stage_test_auc,
        "delta_auc_vs_two_stage": delta_auc,
        "wall_clock_ratio": wall_clock_ratio,
        "joint_seconds": joint_seconds,
        "two_stage_fit_seconds": two_stage_fit_seconds,
    }


# -------------------------------------------------------------------------------------------------------------------------------
# Task 19, Step 3: Document the conclusion
# -------------------------------------------------------------------------------------------------------------------------------


def document_conclusion(trajectory: Dict[str, Any]) -> Dict[str, Any]:
    """
    Compose the joint-optimization conclusion for the report.
    """
    # Compose the conclusion text from the measured trajectory
    conclusion: str = (
        "The alternating fixed-point scheme does not improve on "
        f"two-stage training in a detectable way: the measured "
        f"Delta-AUC of {trajectory['delta_auc_vs_two_stage']:+.5f} is "
        "smaller than the convergence tolerance of 1e-3, so the "
        "procedure halts before producing a distinguishable change. "
        "Two-stage training remains the operational deployment default. "
        "Step (iii) minimizes squared error on residuals rather than "
        "the joint binary cross-entropy, so the scheme is NOT exact "
        "coordinate descent on the joint loss; a true functional-"
        "gradient descent that flows gradients through both branches is "
        "left as future work. The scheme is not run on Taiwan."
    )
    # Return the conclusion entry
    return {
        "conclusion": conclusion,
        "methodological_status": "transparency artifact, not a "
        "performance improvement",
    }


# -------------------------------------------------------------------------------------------------------------------------------
# Task 19, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


def orchestrate_run_alternating_min(
    config: Dict[str, Any],
    X_woe_train: np.ndarray,
    X_ohe_train: np.ndarray,
    y_train: np.ndarray,
    X_woe_val: np.ndarray,
    X_ohe_val: np.ndarray,
    p_lr_val_initial: np.ndarray,
    y_val: np.ndarray,
    X_woe_test: np.ndarray,
    X_ohe_test: np.ndarray,
    y_test: np.ndarray,
    beta_initial: np.ndarray,
    beta_0_initial: float,
    booster_initial: xgb.Booster,
    best_iteration_initial: int,
    two_stage_test_auc: float,
    two_stage_fit_seconds: float,
    output_dir: Path,
    verify_manuscript_targets: bool = True,
) -> Dict[str, Any]:
    """
    Orchestrate the alternating-minimization joint optimization.

    Composes the fixed-point iteration, the trajectory recording and
    the conclusion, persisting the report to JSON.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated study configuration dictionary.
    X_woe_train, X_ohe_train, y_train : np.ndarray
        The training matrices and labels.
    X_woe_val, X_ohe_val, p_lr_val_initial, y_val : np.ndarray
        The validation matrices, initial LR probabilities and labels.
    X_woe_test, X_ohe_test, y_test : np.ndarray
        The test matrices and labels.
    beta_initial : np.ndarray
        The initial two-stage coefficient vector.
    beta_0_initial : float
        The initial two-stage intercept.
    booster_initial : xgb.Booster
        The initial two-stage booster.
    best_iteration_initial : int
        The initial booster's frozen iteration.
    two_stage_test_auc : float
        The two-stage hybrid test AUC.
    two_stage_fit_seconds : float
        The two-stage fit wall-clock time.
    output_dir : Path
        Directory where the artifacts are persisted.
    verify_manuscript_targets : bool, optional
        Whether to assert the manuscript targets (True for real runs).

    Returns
    -------
    Dict[str, Any]
        A state-ready dict with the trajectory, the conclusion and
        output_dir.
    """
    # Validate the output directory type before any filesystem work
    if not isinstance(output_dir, Path):
        # Raise a descriptive error for a non-Path output directory
        raise ValueError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Create the output directory including any missing parents
    output_dir.mkdir(parents=True, exist_ok=True)
    # Step 1: run the fixed-point iteration
    fixed_point_result: Dict[str, Any] = implement_fixed_point_iteration(
        X_woe_train, X_ohe_train, y_train,
        X_woe_val, X_ohe_val, p_lr_val_initial, y_val,
        beta_initial, beta_0_initial, booster_initial, best_iteration_initial,
        config,
    )
    # Step 2: record the convergence trajectory
    trajectory: Dict[str, Any] = run_and_record_trajectory(
        fixed_point_result, X_woe_test, X_ohe_test, y_test,
        two_stage_test_auc, two_stage_fit_seconds, config,
        verify_manuscript_targets,
    )
    # Step 3: document the conclusion
    conclusion: Dict[str, Any] = document_conclusion(trajectory)
    # Compose the persisted report
    report: Dict[str, Any] = {
        "trajectory": trajectory,
        "conclusion": conclusion,
        "history": fixed_point_result["history"],
    }
    # Build the artifacts path inside the output directory
    artifacts_path: Path = output_dir / "joint_optimization_artifacts.json"
    # Serialise the report to JSON with a stable layout
    artifacts_text: str = json.dumps(
        report, indent=2, ensure_ascii=False, default=str
    )
    # Persist the report to the output directory
    artifacts_path.write_text(artifacts_text, encoding="utf-8")
    # Return the state-ready joint optimization outputs
    return {
        "trajectory": trajectory,
        "conclusion": conclusion,
        "output_dir": str(output_dir),
    }
````

````python
# Task 20: Create the Orchestrator Function

# ==============================================================================
# Task 20: Create the Orchestrator Function
# ==============================================================================


# The canonical execution order of the 19 registry callables
PIPELINE_ORDER: Tuple[str, ...] = (
    "validate_inputs",
    "cleanse_data",
    "split_partitions",
    "fit_preprocessing_stats",
    "fit_woe_encoding",
    "fit_ohe_encoding",
    "fit_logistic",
    "compute_residuals",
    "fit_residual_xgboost",
    "fit_standalone_xgboost",
    "select_adaptive_weights",
    "build_hybrid_and_rho",
    "fit_platt",
    "compute_metrics",
    "run_fairness_audit",
    "run_bootstrap_delong",
    "run_shap_analysis",
    "run_thin_file",
    "run_alternating_min",
)

# The Zindi logistic-branch features (all ten for the reproduction
# pipeline, matching Task 6's all-ten-pass assertion)
ZINDI_LOGISTIC_FEATURES: Tuple[str, ...] = (
    "country",
    "location_type",
    "cellphone_access",
    "gender_of_respondent",
    "relationship_with_head",
    "marital_status",
    "education_level",
    "job_type",
    "household_size",
    "age_of_respondent",
)

# The Zindi boosting-branch categorical and continuous features
ZINDI_CATEGORICAL: Tuple[str, ...] = (
    "country",
    "location_type",
    "cellphone_access",
    "gender_of_respondent",
    "relationship_with_head",
    "marital_status",
    "education_level",
    "job_type",
)

# The Zindi boosting-branch continuous features
ZINDI_CONTINUOUS: Tuple[str, ...] = ("household_size", "age_of_respondent")


def _frame_digest(frame: pd.DataFrame) -> str:
    """
    Return a deterministic SHA-256 digest of a DataFrame.
    """
    # Serialise the frame to CSV bytes for a stable digest
    csv_bytes: bytes = frame.to_csv(index=False).encode("utf-8")
    # Hash the serialised bytes and return the hex digest
    return hashlib.sha256(csv_bytes).hexdigest()


def _partition_frames(
    frame: pd.DataFrame, manifest: Dict[str, Any]
) -> Dict[str, pd.DataFrame]:
    """
    Slice a frame into train/val/test frames by the manifest indices.
    """
    # Slice each partition by the frozen positional indices
    return {
        partition: frame.iloc[np.array(manifest[f"{partition}_idx"], dtype=np.int64)]
        for partition in ("train", "val", "test")
    }


# -------------------------------------------------------------------------------------------------------------------------------
# Task 20, Step 1: Define the orchestrator signature and callable registry
# -------------------------------------------------------------------------------------------------------------------------------


def _adapter_validate_inputs(state: Dict[str, Any]) -> Dict[str, Any]:
    """
    Adapter: validate the ingested parameters and data quality.
    """
    # Invoke the Task 1 orchestrator and record the results
    result: Dict[str, Any] = orchestrate_validate_ingested_parameters(
        state["config"], state["zindi_raw"], state["taiwan_raw"],
        Path(state["output_dir"]) / "t1_validate",
    )
    # Write the validated frames back into the state
    state["zindi_raw"] = result["zindi_frame"]
    state["taiwan_raw"] = result["taiwan_frame"]
    # Record the validation report in the state
    state["validation_report"] = {
        "config": result["config"],
        "zindi": result["zindi"],
        "taiwan": result["taiwan"],
    }
    # Return the state for the pipeline
    return state


def _adapter_cleanse_data(state: Dict[str, Any]) -> Dict[str, Any]:
    """
    Adapter: cleanse both datasets and persist the audit record.
    """
    # Invoke the Task 2 orchestrator
    result: Dict[str, Any] = orchestrate_cleanse_data(
        state["config"], state["zindi_raw"], state["taiwan_raw"],
        Path(state["output_dir"]) / "t2_cleanse",
    )
    # Store the cleansed frames and the unseen-levels registry
    state["zindi_cleansed"] = result["zindi"]
    state["taiwan_cleansed"] = result["taiwan"]
    state["unseen_levels"] = result["unseen_levels"]
    # Record the cleansing audit in the state
    state["cleansing_audit"] = result["audit_record"]
    # Return the state for the pipeline
    return state


def _adapter_split_partitions(state: Dict[str, Any]) -> Dict[str, Any]:
    """
    Adapter: create both frozen partition manifests.
    """
    # Invoke the Task 3 orchestrator
    result: Dict[str, Any] = orchestrate_create_partitions(
        state["config"], state["zindi_cleansed"], state["taiwan_cleansed"],
        state["unseen_levels"], Path(state["output_dir"]) / "t3_split",
    )
    # Store the partition manifests
    state["zindi_partitions"] = result["zindi_partitions"]
    state["taiwan_partitions"] = result["taiwan_partitions"]
    # Return the state for the pipeline
    return state


def _adapter_fit_preprocessing_stats(state: Dict[str, Any]) -> Dict[str, Any]:
    """
    Adapter: fit and apply the frozen preprocessing statistics.
    """
    # Invoke the Task 4 orchestrator
    result: Dict[str, Any] = orchestrate_fit_preprocessing_stats(
        state["config"], state["zindi_cleansed"], state["zindi_partitions"],
        state["taiwan_cleansed"], state["taiwan_partitions"],
        Path(state["output_dir"]) / "t4_preprocessing",
    )
    # Store the transformed frames and the frozen statistics
    state["zindi_transformed"] = result["zindi_frame"]
    state["taiwan_transformed"] = result["taiwan_frame"]
    state["preprocessing_stats"] = result["preprocessing_stats"]
    # Return the state for the pipeline
    return state


def _adapter_fit_woe_encoding(state: Dict[str, Any]) -> Dict[str, Any]:
    """
    Adapter: fit the monotonic bins and the WoE mapping.
    """
    # Extract the Zindi training frame and labels
    zindi_frames: Dict[str, pd.DataFrame] = _partition_frames(
        state["zindi_transformed"], state["zindi_partitions"]
    )
    # Extract the Zindi label column name
    label: str = state["config"]["dataset_parameters"]["zindi_fsd"][
        "label_column"
    ]
    # Extract the Zindi training labels
    zindi_y_train: np.ndarray = zindi_frames["train"][label].to_numpy(
        dtype=np.int64
    )
    # Extract the Taiwan label column name
    taiwan_label: str = "default.payment.next.month"
    # Build the Taiwan partition frames
    taiwan_frames: Dict[str, pd.DataFrame] = _partition_frames(
        state["taiwan_transformed"], state["taiwan_partitions"]
    )
    # Extract the Taiwan training labels
    taiwan_y_train: np.ndarray = taiwan_frames["train"][
        taiwan_label
    ].to_numpy(dtype=np.int64)
    # Derive the Taiwan feature sets from the config schema
    taiwan_numeric: List[str] = []
    taiwan_categorical: List[str] = []
    # Classify the Taiwan schema columns
    for column in state["config"]["raw_data_schemas"][
        "taiwan_credit_schema"
    ]["columns"]:
        # Skip the row identifier and the label
        if column["name"] in ("ID", taiwan_label):
            # Continue to the next schema column
            continue
        # Classify categorical-code columns as categorical
        if "categorical code" in column["dtype"]:
            # Add the column to the categorical set
            taiwan_categorical.append(column["name"])
        else:
            # Add the column to the numeric set
            taiwan_numeric.append(column["name"])
    # Invoke the Task 5 orchestrator for both datasets
    result: Dict[str, Any] = orchestrate_fit_woe_encoding(
        state["config"],
        zindi_frames["train"], zindi_y_train, zindi_frames,
        ZINDI_LOGISTIC_FEATURES,
        taiwan_frames["train"], taiwan_y_train, taiwan_frames,
        tuple(taiwan_numeric + taiwan_categorical),
        tuple(taiwan_numeric), tuple(taiwan_categorical),
        Path(state["output_dir"]) / "t5_woe",
    )
    # Store the WoE artifacts
    state["zindi_woe"] = result["zindi"]
    state["taiwan_woe"] = result["taiwan"]
    state["neutral_log"] = result["neutral_log"]
    # Store the Zindi partition frames for downstream tasks
    state["zindi_frames"] = zindi_frames
    state["taiwan_frames"] = taiwan_frames
    # Return the state for the pipeline
    return state


def _adapter_fit_ohe_encoding(state: Dict[str, Any]) -> Dict[str, Any]:
    """
    Adapter: compute IVs and fit the one-hot encoding.
    """
    # Extract the Zindi label column name
    label: str = state["config"]["dataset_parameters"]["zindi_fsd"][
        "label_column"
    ]
    # Extract the Zindi training labels
    zindi_y_train: np.ndarray = state["zindi_frames"]["train"][
        label
    ].to_numpy(dtype=np.int64)
    # Invoke the Task 6 orchestrator on the Zindi pipeline
    result: Dict[str, Any] = orchestrate_iv_filter_and_ohe(
        state["config"],
        state["zindi_frames"]["train"], zindi_y_train,
        state["zindi_woe"]["woe_map"], state["zindi_woe"]["cut_points"],
        state["zindi_woe"]["matrices"], state["zindi_frames"],
        ZINDI_CATEGORICAL, ZINDI_CONTINUOUS,
        Path(state["output_dir"]) / "t6_ohe",
        verify_manuscript_targets=state["verify_manuscript_targets"],
    )
    # Store the IV and one-hot artifacts
    state["iv_values"] = result["iv_values"]
    state["retained_features"] = result["retained_features"]
    state["level_sets"] = result["level_sets"]
    state["column_list"] = result["column_list"]
    state["ohe_matrices"] = result["ohe_matrices"]
    state["filtered_woe_matrices"] = result["filtered_woe_matrices"]
    # Return the state for the pipeline
    return state


def _adapter_fit_logistic(state: Dict[str, Any]) -> Dict[str, Any]:
    """
    Adapter: fit the logistic baseline with C selection.
    """
    # Extract the Zindi label column name
    label: str = state["config"]["dataset_parameters"]["zindi_fsd"][
        "label_column"
    ]
    # Build the per-partition filtered WoE matrices as arrays
    matrices: Dict[str, np.ndarray] = {
        partition: state["filtered_woe_matrices"][partition].to_numpy(
            dtype=np.float64
        )
        for partition in ("train", "val", "test")
    }
    # Build the per-partition label vectors
    labels: Dict[str, np.ndarray] = {
        partition: state["zindi_frames"][partition][label].to_numpy(
            dtype=np.int64
        )
        for partition in ("train", "val", "test")
    }
    # Record the two-stage fit start time
    t_start: float = time.perf_counter()
    # Invoke the Task 7 orchestrator
    result: Dict[str, Any] = orchestrate_fit_logistic(
        state["config"],
        matrices["train"], labels["train"],
        matrices["val"], labels["val"],
        matrices["test"], labels["test"],
        Path(state["output_dir"]) / "t7_logistic",
        feature_names=state["retained_features"],
    )
    # Record the logistic fit wall-clock time
    state["two_stage_fit_seconds"] = time.perf_counter() - t_start
    # Store the logistic artifacts
    state["logistic"] = result
    # Return the state for the pipeline
    return state


def _adapter_compute_residuals(state: Dict[str, Any]) -> Dict[str, Any]:
    """
    Adapter: compute the raw response residuals.
    """
    # Extract the Zindi label column name
    label: str = state["config"]["dataset_parameters"]["zindi_fsd"][
        "label_column"
    ]
    # Build the per-partition label vectors
    labels: Dict[str, np.ndarray] = {
        partition: state["zindi_frames"][partition][label].to_numpy(
            dtype=np.int64
        )
        for partition in ("train", "val", "test")
    }
    # Invoke the Task 8 orchestrator
    result: Dict[str, Any] = orchestrate_compute_residuals(
        labels, state["logistic"]["p_lr"]
    )
    # Store the residual artifacts
    state["residuals"] = result["residuals"]
    state["residual_summary"] = result["residual_summary"]
    # Return the state for the pipeline
    return state


def _adapter_fit_residual_xgboost(state: Dict[str, Any]) -> Dict[str, Any]:
    """
    Adapter: fit the residual XGBoost learner.
    """
    # Extract the Zindi label column name
    label: str = state["config"]["dataset_parameters"]["zindi_fsd"][
        "label_column"
    ]
    # Build the per-partition one-hot matrices as arrays
    matrices: Dict[str, np.ndarray] = {
        partition: state["ohe_matrices"][partition].to_numpy(dtype=np.float64)
        for partition in ("train", "val", "test")
    }
    # Build the per-partition label vectors
    labels: Dict[str, np.ndarray] = {
        partition: state["zindi_frames"][partition][label].to_numpy(
            dtype=np.int64
        )
        for partition in ("train", "val", "test")
    }
    # Invoke the Task 9 orchestrator
    result: Dict[str, Any] = orchestrate_fit_residual_xgboost(
        state["config"], matrices, state["residuals"],
        state["logistic"]["p_lr"], labels, state["logistic"]["val_auc"],
        Path(state["output_dir"]) / "t9_residual",
    )
    # Store the residual learner artifacts
    state["residual_booster"] = result["booster"]
    state["best_iteration"] = result["best_iteration"]
    state["r_hat"] = result["r_hat"]
    state["correlation_train"] = result["correlation_train"]
    state["provisional_auc"] = result["provisional_auc"]
    # Return the state for the pipeline
    return state


def _adapter_fit_standalone_xgboost(state: Dict[str, Any]) -> Dict[str, Any]:
    """
    Adapter: fit the standalone XGBoost comparator.
    """
    # Extract the Zindi label column name
    label: str = state["config"]["dataset_parameters"]["zindi_fsd"][
        "label_column"
    ]
    # Build the per-partition one-hot matrices as arrays
    matrices: Dict[str, np.ndarray] = {
        partition: state["ohe_matrices"][partition].to_numpy(dtype=np.float64)
        for partition in ("train", "val", "test")
    }
    # Build the per-partition label vectors
    labels: Dict[str, np.ndarray] = {
        partition: state["zindi_frames"][partition][label].to_numpy(
            dtype=np.int64
        )
        for partition in ("train", "val", "test")
    }
    # Invoke the Task 10 orchestrator
    result: Dict[str, Any] = orchestrate_fit_standalone_xgboost(
        state["config"], matrices, labels,
        Path(state["output_dir"]) / "t10_standalone",
        verify_manuscript_targets=state["verify_manuscript_targets"],
    )
    # Store the standalone artifacts
    state["standalone_xgb"] = result
    # Return the state for the pipeline
    return state


def _adapter_select_adaptive_weights(state: Dict[str, Any]) -> Dict[str, Any]:
    """
    Adapter: grid-search and freeze the adaptive weights.
    """
    # Extract the Zindi label column name
    label: str = state["config"]["dataset_parameters"]["zindi_fsd"][
        "label_column"
    ]
    # Build the training matrices as arrays
    X_woe_train: np.ndarray = state["filtered_woe_matrices"][
        "train"
    ].to_numpy(dtype=np.float64)
    X_ohe_train: np.ndarray = state["ohe_matrices"]["train"].to_numpy(
        dtype=np.float64
    )
    # Extract the training labels
    y_train: np.ndarray = state["zindi_frames"]["train"][label].to_numpy(
        dtype=np.int64
    )
    # Invoke the Task 11 orchestrator
    result: Dict[str, Any] = orchestrate_select_adaptive_weights(
        state["config"], X_woe_train, y_train, X_ohe_train,
        state["logistic"]["C"], state["logistic"]["p_lr"],
        Path(state["output_dir"]) / "t11_adaptive",
    )
    # Store the adaptive weighting artifacts
    state["selected_triple"] = result["selected_triple"]
    state["grid_table"] = result["grid_table"]
    state["alpha_star"] = result["alpha_star"]
    # Return the state for the pipeline
    return state


def _adapter_build_hybrid_and_rho(state: Dict[str, Any]) -> Dict[str, Any]:
    """
    Adapter: build the hybrid predictions and the rho decomposition.
    """
    # Extract the Zindi label column name
    label: str = state["config"]["dataset_parameters"]["zindi_fsd"][
        "label_column"
    ]
    # Build the per-partition label vectors
    labels: Dict[str, np.ndarray] = {
        partition: state["zindi_frames"][partition][label].to_numpy(
            dtype=np.int64
        )
        for partition in ("train", "val", "test")
    }
    # Invoke the Task 12 orchestrator
    result: Dict[str, Any] = orchestrate_build_hybrid_and_rho(
        state["config"], state["logistic"]["p_lr"], state["r_hat"],
        state["alpha_star"], labels,
        Path(state["output_dir"]) / "t12_hybrid",
        verify_manuscript_targets=state["verify_manuscript_targets"],
    )
    # Store the hybrid and interpretability artifacts
    state["p_hybrid"] = result["p_hybrid"]
    state["components"] = result["components"]
    state["rho"] = result["rho"]
    state["regions"] = result["regions"]
    state["region_stats"] = result["region_stats"]
    # Return the state for the pipeline
    return state


def _adapter_fit_platt(state: Dict[str, Any]) -> Dict[str, Any]:
    """
    Adapter: fit and apply the Platt calibration.
    """
    # Extract the Zindi label column name
    label: str = state["config"]["dataset_parameters"]["zindi_fsd"][
        "label_column"
    ]
    # Build the per-partition label vectors
    labels: Dict[str, np.ndarray] = {
        partition: state["zindi_frames"][partition][label].to_numpy(
            dtype=np.int64
        )
        for partition in ("train", "val", "test")
    }
    # Invoke the Task 13 orchestrator
    result: Dict[str, Any] = orchestrate_fit_platt(
        state["p_hybrid"], state["logistic"]["p_lr"], labels,
        Path(state["output_dir"]) / "t13_platt",
        verify_manuscript_targets=state["verify_manuscript_targets"],
    )
    # Store the calibration artifacts; 'platt_a'/'platt_b' are the MANUSCRIPT
    # convention satisfying Equation (10), and the sklearn pair is carried
    # alongside under distinct keys so the two can never be confused
    state["platt_a"] = result["a"]
    state["platt_b"] = result["b"]
    state["platt_a_sklearn"] = result["a_sklearn"]
    state["platt_b_sklearn"] = result["b_sklearn"]
    state["p_cal_test"] = result["p_cal_test"]
    state["ablation"] = result["ablation"]
    # Return the state for the pipeline
    return state


def _adapter_compute_metrics(state: Dict[str, Any]) -> Dict[str, Any]:
    """
    Adapter: compute the evaluation metrics and the result tables.
    """
    # Extract the Zindi label column name
    label: str = state["config"]["dataset_parameters"]["zindi_fsd"][
        "label_column"
    ]
    # Extract the test labels
    y_test: np.ndarray = state["zindi_frames"]["test"][label].to_numpy(
        dtype=np.int64
    )
    # Compose the five-model test probability dict
    probabilities: Dict[str, np.ndarray] = {
        "Logistic regression": state["logistic"]["p_lr"]["test"],
        "XGBoost": state["standalone_xgb"]["probs"]["test"],
        "Hybrid (uncalibrated)": state["p_hybrid"]["test"],
        "Hybrid (Platt calibrated)": state["p_cal_test"],
        "LR + Platt": state["ablation"]["p_ablation_test"],
    }
    # Invoke the Task 14 orchestrator
    result: Dict[str, Any] = orchestrate_compute_metrics(
        probabilities, y_test, state["regions"]["test"],
        state["logistic"]["p_lr"]["test"], state["p_hybrid"]["test"],
        state["region_stats"], Path(state["output_dir"]) / "t14_metrics",
        verify_manuscript_targets=state["verify_manuscript_targets"],
    )
    # Store the metrics artifacts
    state["discrimination"] = result["discrimination"]
    state["calibration"] = result["calibration"]
    state["region_auc"] = result["region_auc"]
    state["tables"] = result["tables"]
    # Return the state for the pipeline
    return state


def _adapter_run_fairness_audit(state: Dict[str, Any]) -> Dict[str, Any]:
    """
    Adapter: run the fairness audit per interpretability region.
    """
    # Extract the Zindi label column name
    label: str = state["config"]["dataset_parameters"]["zindi_fsd"][
        "label_column"
    ]
    # Extract the test labels
    y_test: np.ndarray = state["zindi_frames"]["test"][label].to_numpy(
        dtype=np.int64
    )
    # Invoke the Task 15 orchestrator on the test partition
    result: Dict[str, Any] = orchestrate_run_fairness_audit(
        state["config"], state["zindi_frames"]["test"],
        state["rho"]["test"], state["regions"]["test"],
        state["p_cal_test"], y_test,
        Path(state["output_dir"]) / "t15_fairness",
        verify_manuscript_targets=state["verify_manuscript_targets"],
    )
    # Store the fairness artifacts
    state["analysis"] = result["analysis"]
    state["routing"] = result["routing"]
    state["di_eod"] = result["di_eod"]
    # Return the state for the pipeline
    return state


def _adapter_run_bootstrap_delong(state: Dict[str, Any]) -> Dict[str, Any]:
    """
    Adapter: run the bootstrap inference and the DeLong tests.
    """
    # Extract the Zindi label column name
    label: str = state["config"]["dataset_parameters"]["zindi_fsd"][
        "label_column"
    ]
    # Extract the test labels
    y_test: np.ndarray = state["zindi_frames"]["test"][label].to_numpy(
        dtype=np.int64
    )
    # Compose the prediction dict for the pairwise comparisons
    predictions: Dict[str, np.ndarray] = {
        "hybrid": state["p_hybrid"]["test"],
        "lr": state["logistic"]["p_lr"]["test"],
        "xgb": state["standalone_xgb"]["probs"]["test"],
    }
    # Invoke the Task 16 orchestrator
    result: Dict[str, Any] = orchestrate_run_bootstrap_delong(
        state["config"], predictions, y_test,
        Path(state["output_dir"]) / "t16_inference",
        verify_manuscript_targets=state["verify_manuscript_targets"],
    )
    # Store the inference artifacts
    state["pairwise"] = result["pairwise"]
    state["table_9"] = result["table_9"]
    # Return the state for the pipeline
    return state


def _adapter_run_shap_analysis(state: Dict[str, Any]) -> Dict[str, Any]:
    """
    Adapter: run the TreeSHAP analysis on the ML-driven region.
    """
    # Invoke the Task 17 orchestrator
    result: Dict[str, Any] = orchestrate_run_shap_analysis(
        state["config"], state["residual_booster"], state["best_iteration"],
        state["ohe_matrices"]["test"].to_numpy(dtype=np.float64),
        state["column_list"], state["regions"]["test"],
        Path(state["output_dir"]) / "t17_shap",
        verify_manuscript_targets=state["verify_manuscript_targets"],
    )
    # Store the SHAP artifacts
    state["shap_values"] = result["shap_values"]
    state["shap_ranking"] = result["ranking"]
    state["shap_stability"] = result["stability"]
    # Return the state for the pipeline
    return state


def _adapter_run_thin_file(state: Dict[str, Any]) -> Dict[str, Any]:
    """
    Adapter: run the thin-file segmentation analysis.
    """
    # Extract the Zindi label column name
    label: str = state["config"]["dataset_parameters"]["zindi_fsd"][
        "label_column"
    ]
    # Extract the test labels
    y_test: np.ndarray = state["zindi_frames"]["test"][label].to_numpy(
        dtype=np.int64
    )
    # Compose the per-model test probability dict
    model_probs: Dict[str, np.ndarray] = {
        "lr": state["logistic"]["p_lr"]["test"],
        "xgb": state["standalone_xgb"]["probs"]["test"],
        "hybrid_cal": state["p_cal_test"],
    }
    # Build the full-test metrics from the Task 14 tables
    full_metrics: Dict[str, Dict[str, float]] = {
        "hybrid_cal": state["tables"]["table_2"]["Hybrid (Platt calibrated)"],
        "lr": state["tables"]["table_2"]["Logistic regression"],
    }
    # Invoke the Task 18 orchestrator
    result: Dict[str, Any] = orchestrate_run_thin_file(
        state["config"], state["zindi_frames"]["train"],
        state["zindi_frames"]["test"], model_probs, y_test,
        state["rho"]["test"], full_metrics,
        Path(state["output_dir"]) / "t18_thin_file",
        verify_manuscript_targets=state["verify_manuscript_targets"],
    )
    # Store the thin-file artifacts
    state["thin_file_stats"] = result["frozen_stats"]
    state["thin_file_tau"] = result["tau"]
    state["thin_file"] = result["subgroup"]
    state["thin_file_hypotheses"] = result["hypotheses"]
    # Return the state for the pipeline
    return state


def _adapter_run_alternating_min(state: Dict[str, Any]) -> Dict[str, Any]:
    """
    Adapter: run the alternating-minimization joint optimization.
    """
    # Extract the Zindi label column name
    label: str = state["config"]["dataset_parameters"]["zindi_fsd"][
        "label_column"
    ]
    # Build the per-partition matrices and labels
    X_woe: Dict[str, np.ndarray] = {
        partition: state["filtered_woe_matrices"][partition].to_numpy(
            dtype=np.float64
        )
        for partition in ("train", "val", "test")
    }
    # Build the per-partition one-hot matrices
    X_ohe: Dict[str, np.ndarray] = {
        partition: state["ohe_matrices"][partition].to_numpy(dtype=np.float64)
        for partition in ("train", "val", "test")
    }
    # Build the per-partition label vectors
    labels: Dict[str, np.ndarray] = {
        partition: state["zindi_frames"][partition][label].to_numpy(
            dtype=np.int64
        )
        for partition in ("train", "val", "test")
    }

    # Compute the two-stage hybrid test AUC
    two_stage_test_auc: float = float(
        roc_auc_score(labels["test"], state["p_hybrid"]["test"])
    )
    # Invoke the Task 19 orchestrator
    result: Dict[str, Any] = orchestrate_run_alternating_min(
        state["config"],
        X_woe["train"], X_ohe["train"], labels["train"],
        X_woe["val"], X_ohe["val"], state["logistic"]["p_lr"]["val"],
        labels["val"], X_woe["test"], X_ohe["test"], labels["test"],
        state["logistic"]["coef"], state["logistic"]["intercept"],
        state["residual_booster"], state["best_iteration"],
        two_stage_test_auc, state["two_stage_fit_seconds"],
        Path(state["output_dir"]) / "t19_joint",
        verify_manuscript_targets=state["verify_manuscript_targets"],
    )
    # Store the joint optimization artifacts
    state["joint_trajectory"] = result["trajectory"]
    state["joint_conclusion"] = result["conclusion"]
    # Return the state for the pipeline
    return state


def build_registry() -> Dict[str, Callable[[Dict[str, Any]], Dict[str, Any]]]:
    """
    Build the 19-entry callable registry with pre-flight checks.
    """
    # Compose the registry mapping the canonical names to the adapters
    registry: Dict[str, Callable[[Dict[str, Any]], Dict[str, Any]]] = {
        "validate_inputs": _adapter_validate_inputs,
        "cleanse_data": _adapter_cleanse_data,
        "split_partitions": _adapter_split_partitions,
        "fit_preprocessing_stats": _adapter_fit_preprocessing_stats,
        "fit_woe_encoding": _adapter_fit_woe_encoding,
        "fit_ohe_encoding": _adapter_fit_ohe_encoding,
        "fit_logistic": _adapter_fit_logistic,
        "compute_residuals": _adapter_compute_residuals,
        "fit_residual_xgboost": _adapter_fit_residual_xgboost,
        "fit_standalone_xgboost": _adapter_fit_standalone_xgboost,
        "select_adaptive_weights": _adapter_select_adaptive_weights,
        "build_hybrid_and_rho": _adapter_build_hybrid_and_rho,
        "fit_platt": _adapter_fit_platt,
        "compute_metrics": _adapter_compute_metrics,
        "run_fairness_audit": _adapter_run_fairness_audit,
        "run_bootstrap_delong": _adapter_run_bootstrap_delong,
        "run_shap_analysis": _adapter_run_shap_analysis,
        "run_thin_file": _adapter_run_thin_file,
        "run_alternating_min": _adapter_run_alternating_min,
    }
    # Pre-flight: assert the registry has exactly the 19 required entries
    if set(registry) != set(PIPELINE_ORDER):
        # Raise a descriptive error for the registry mismatch
        raise ValueError(
            f"registry names must equal the pipeline order; missing "
            f"{set(PIPELINE_ORDER) - set(registry)}, extra "
            f"{set(registry) - set(PIPELINE_ORDER)}."
        )
    # Return the validated registry
    return registry


@dataclass(frozen=True)
class VerificationCheck:
    """
    One expected intermediate value mandated by Task 20, Step 3.

    Purpose.  Task 20 Step 3 enumerates nine intermediate quantities the
    lightweight verifier must compare against declared expectations.  This
    record expresses ONE such comparison as DATA rather than as imperative
    code, so the complete check set can be diffed line by line against the
    task list without reading control flow.  That auditability is the
    reason the check set is externalised: the defect this design replaces
    was a verifier whose checks were individually plausible but which, read
    as a set, covered one of nine mandated quantities.

    Process.  At verification time the resolver walks `state_path` through
    the shared state dictionary to obtain the actual value, then applies
    `comparison` against `expected` with `tolerance` where the mode
    requires one.  A record is emitted carrying the outcome together with
    both compared values.

    Attributes
    ----------
    name : str
        The key under which this check's record is filed in the verifier
        output, and the human-readable name shown in diagnostics.
    state_path : Tuple[Union[str, int], ...]
        The successive keys or indices to walk through the shared state to
        reach the actual value.  An empty tuple marks a check whose actual
        value is supplied by a dedicated resolver instead.
    expected : Any
        The value the specification declares for this quantity.
    comparison : str
        One of exactly four modes: "equals" for integer counts and
        categorical values; "within_abs" for real quantities with an
        absolute margin; "within_pp" for quantities expressed in
        percentage points; "sequence_equals" for rankings and parameter
        triples where order carries meaning.  Membership and threshold
        modes are deliberately absent: they are the modes that produced
        the tautological checks this design replaces.
    tolerance : float
        The absolute or percentage-point margin.  Ignored by the "equals"
        and "sequence_equals" modes.
    source_task : int
        The task number the expectation originates from, so a failure
        report can name the stage responsible.
    structural : bool
        True when failure indicates a broken pipeline and must halt the
        run; False when failure indicates a reproduction discrepancy,
        which is a finding to be recorded rather than a fault to raise on.
    """

    # The record key and human-readable check name
    name: str
    # The successive keys/indices to walk through the shared state
    state_path: Tuple[Union[str, int], ...]
    # The value declared by the specification
    expected: Any
    # The comparison mode, restricted to the four permitted forms
    comparison: str
    # The absolute or percentage-point margin, unused by exact modes
    tolerance: float = 0.0
    # The originating task number, for failure attribution
    source_task: int = 0
    # Whether failure is a structural fault (halt) or a reproduction finding
    structural: bool = False

    def __post_init__(self) -> None:
        """
        Validate the record on construction.

        Rejects any comparison mode outside the four permitted forms and
        any negative tolerance, so a malformed check specification fails at
        import time rather than silently degrading a verification run.

        Raises
        ------
        ValueError
            If the comparison mode is unrecognised or the tolerance is
            negative.
        """
        # Restrict the comparison mode to exactly the four permitted forms
        if self.comparison not in (
            "equals", "within_abs", "within_pp", "sequence_equals"
        ):
            # Raise a descriptive error naming the offending mode
            raise ValueError(
                f"check '{self.name}' declares unsupported comparison "
                f"'{self.comparison}'; permitted modes are 'equals', "
                f"'within_abs', 'within_pp', 'sequence_equals'."
            )
        # Reject a negative tolerance, which would make every comparison pass
        if self.tolerance < 0.0:
            # Raise a descriptive error naming the offending tolerance
            raise ValueError(
                f"check '{self.name}' declares negative tolerance "
                f"{self.tolerance}."
            )


# The nine intermediate quantities Task 20 Step 3 mandates, expressed as data.
# Each carries the value declared by the specification, so the set can be
# audited against the task list by inspection.
VERIFICATION_CHECKS: Tuple[VerificationCheck, ...] = (
    # (1) The Zindi training partition size, an exact structural invariant
    VerificationCheck(
        name="zindi_train_n",
        state_path=("zindi_partitions", "sizes", "train"),
        expected=14114,
        comparison="equals",
        source_task=3,
        structural=True,
    ),
    # (2) The Task 6 information values of the three highest-IV features
    VerificationCheck(
        name="iv_education_level",
        state_path=("iv_values", "education_level"),
        expected=0.97,
        comparison="within_abs",
        tolerance=0.02,
        source_task=6,
    ),
    # (2) continued: the cellphone-access information value
    VerificationCheck(
        name="iv_cellphone_access",
        state_path=("iv_values", "cellphone_access"),
        expected=0.71,
        comparison="within_abs",
        tolerance=0.02,
        source_task=6,
    ),
    # (2) continued: the job-type information value
    VerificationCheck(
        name="iv_job_type",
        state_path=("iv_values", "job_type"),
        expected=0.63,
        comparison="within_abs",
        tolerance=0.02,
        source_task=6,
    ),
    # (3) The selected adaptive triple, compared as an ORDERED sequence so a
    # permuted or partially-correct triple cannot pass
    VerificationCheck(
        name="alpha_triple_selected",
        state_path=("selected_triple",),
        expected=(0.2, 0.8, 1.0),
        comparison="sequence_equals",
        source_task=11,
    ),
    # (4) The three Table 3 region shares, in percentage points
    VerificationCheck(
        name="region_shares",
        state_path=(),
        expected=(6.4, 34.1, 59.4),
        comparison="within_pp",
        tolerance=1.0,
        source_task=12,
    ),
    # (4) continued: the three within-region positive rates, percentage points
    VerificationCheck(
        name="region_positive_rates",
        state_path=(),
        expected=(53.8, 19.7, 6.5),
        comparison="within_pp",
        tolerance=1.0,
        source_task=12,
    ),
    # (5) The eight Table 4 routing-rate point estimates
    VerificationCheck(
        name="routing_point_estimates",
        state_path=(),
        expected=(
            0.664, 0.482, 0.683, 0.359, 0.702, 0.484, 0.594, 0.595,
        ),
        comparison="within_abs",
        tolerance=0.01,
        source_task=15,
    ),
    # (6) The paired Delta-AUC confidence interval, hybrid against LR
    VerificationCheck(
        name="delta_auc_ci_hybrid_vs_lr",
        state_path=(),
        expected=(0.011, 0.020),
        comparison="within_abs",
        tolerance=0.01,
        source_task=16,
    ),
    # (6) continued: the interval for hybrid against standalone XGBoost
    VerificationCheck(
        name="delta_auc_ci_hybrid_vs_xgb",
        state_path=(),
        expected=(-0.009, 0.001),
        comparison="within_abs",
        tolerance=0.01,
        source_task=16,
    ),
    # (7) The top-five attribution ranking, compared as an ORDERED sequence
    VerificationCheck(
        name="shap_top_five",
        state_path=(),
        expected=(
            "cellphone_access_No",
            "age_of_respondent",
            "relationship_with_head_Head of Household",
            "country_Kenya",
            "country_Rwanda",
        ),
        comparison="sequence_equals",
        source_task=17,
    ),
    # (8) The thin-file subgroup size with its declared tolerance
    VerificationCheck(
        name="thin_file_n",
        state_path=("thin_file", "profile", "n"),
        expected=1559,
        comparison="within_abs",
        tolerance=20.0,
        source_task=18,
    ),
    # (9) The alternating-minimisation iteration count, an exact comparison
    VerificationCheck(
        name="joint_iterations",
        state_path=("joint_trajectory", "n_iterations"),
        expected=3,
        comparison="equals",
        source_task=19,
    ),
)

# The partitions each pipeline stage is permitted to read FOR ESTIMATION.
# This is the access declaration required by the test-partition information
# barrier.  It records fitting reads only: applying an already-frozen
# transform to validation or test is always permitted and is not recorded
# here, because the barrier exists to prevent held-out data from influencing
# estimation, not from being scored.
ADAPTER_FITTING_PARTITIONS: Dict[str, Tuple[str, ...]] = {
    # Input validation estimates nothing; it inspects the raw frames
    "validate_inputs": (),
    # Cleansing estimates nothing and runs before the split exists
    "cleanse_data": (),
    # The splitter creates the partitions rather than reading them
    "split_partitions": (),
    # Medians, modes and winsorization bounds are training-only statistics
    "fit_preprocessing_stats": ("train",),
    # Bin boundaries and WoE tables are frozen on training
    "fit_woe_encoding": ("train",),
    # Information values and one-hot level sets are frozen on training
    "fit_ohe_encoding": ("train",),
    # The logistic coefficients and the selected C come from training
    "fit_logistic": ("train",),
    # Residuals are a pure transform; only the training vector feeds fitting
    "compute_residuals": ("train",),
    # The residual booster fits on training and early-stops on validation AUC
    "fit_residual_xgboost": ("train", "val"),
    # The standalone comparator fits on training alone
    "fit_standalone_xgboost": ("train",),
    # The adaptive grid search cross-validates WITHIN the training partition
    "select_adaptive_weights": ("train",),
    # Hybrid assembly and the interpretability ratio estimate nothing
    "build_hybrid_and_rho": (),
    # Platt scaling is fitted on the validation partition by construction
    "fit_platt": ("val",),
    # Metric computation is evaluation, not estimation
    "compute_metrics": (),
    # The fairness audit is evaluation, not estimation
    "run_fairness_audit": (),
    # Bootstrap and DeLong inference consume pre-computed test predictions
    "run_bootstrap_delong": (),
    # Attribution is computed on the frozen booster, estimating nothing
    "run_shap_analysis": (),
    # The thin-file z-score moments and threshold are frozen on training
    "run_thin_file": ("train",),
    # Joint optimisation refits on training and monitors validation AUC
    "run_alternating_min": ("train", "val"),
}

# The stages that constitute the fitting sequence, from input validation
# through calibration.  No stage in this set may declare test access.
FITTING_SEQUENCE_STAGES: Tuple[str, ...] = (
    "validate_inputs", "cleanse_data", "split_partitions",
    "fit_preprocessing_stats", "fit_woe_encoding", "fit_ohe_encoding",
    "fit_logistic", "compute_residuals", "fit_residual_xgboost",
    "fit_standalone_xgboost", "select_adaptive_weights",
    "build_hybrid_and_rho", "fit_platt",
)

# The state keys carrying the frozen statistics that Task 20 Step 2 requires
# to be serialised to a checkpoint after every callable.
FROZEN_STATISTIC_KEYS: Tuple[str, ...] = (
    # Partition index arrays for both datasets
    "zindi_partitions", "taiwan_partitions",
    # Imputation medians, imputation modes and winsorization bounds
    "preprocessing_stats",
    # Monotonic bin boundaries and the frozen WoE lookup tables
    "zindi_woe", "taiwan_woe",
    # Information values and the surviving feature set
    "iv_values", "retained_features",
    # One-hot level sets and the frozen column list
    "level_sets", "column_list",
    # The logistic branch, carrying the selected inverse-regularisation C
    "logistic",
    # The selected adaptive weighting triple
    "selected_triple",
    # The frozen Platt parameters in both parameterisations
    "platt_a", "platt_b", "platt_a_sklearn", "platt_b_sklearn",
    # The frozen thin-file standardisation moments and threshold
    "thin_file_stats", "thin_file_tau",
)


# -------------------------------------------------------------------------------------------------------------------------------
# Task 20, Step 2: Implement the sequential pipeline with frozen-state passing
# -------------------------------------------------------------------------------------------------------------------------------


def _digest_array(array: np.ndarray) -> str:
    """
    Compute a shape- and dtype-sensitive SHA-256 digest of an array.

    Folds the dtype string and the shape tuple into the hash alongside the
    raw buffer, so that a reshape or a dtype change cannot produce a
    colliding digest.  Digesting `tobytes()` alone would let an array of
    the same bytes reinterpreted under a different dtype, or reshaped,
    compare equal to the original.

    Parameters
    ----------
    array : np.ndarray
        The array to digest.  A C-contiguous copy is taken when required,
        because `tobytes()` on a non-contiguous view would serialise in an
        order that depends on the view rather than on the logical content.

    Returns
    -------
    str
        The hexadecimal SHA-256 digest.

    Raises
    ------
    TypeError
        If the argument is not a NumPy array.
    """
    # Reject non-array input rather than digesting an unintended object
    if not isinstance(array, np.ndarray):
        # Raise a descriptive error naming the received type
        raise TypeError(
            f"_digest_array expects np.ndarray; received "
            f"{type(array).__name__}."
        )
    # Force C-contiguity so the byte serialisation reflects logical order
    contiguous: np.ndarray = np.ascontiguousarray(array)
    # Initialise the hash with the dtype string so a reinterpretation differs
    hasher = hashlib.sha256(str(contiguous.dtype).encode("utf-8"))
    # Fold the shape in so a reshape of identical bytes differs
    hasher.update(str(contiguous.shape).encode("utf-8"))
    # Fold in the raw buffer contents
    hasher.update(contiguous.tobytes())
    # Return the hexadecimal digest
    return hasher.hexdigest()


def _snapshot_hashes(state: Dict[str, Any]) -> Dict[str, str]:
    """
    Capture SHA-256 digests of every frozen artifact in the state.

    Task 20 Step 2 requires a before-and-after digest of each key frame so
    that no callable can mutate a frozen artifact undetected.  Coverage is
    extended beyond dataframes to the encoded design matrices and the
    label vectors, because those are array-typed: an in-place mutation of
    a training design matrix would otherwise pass the check untouched,
    while the step's stated guarantee is that NO frozen artifact is
    mutated.

    Parameters
    ----------
    state : Dict[str, Any]
        The shared pipeline state.

    Returns
    -------
    Dict[str, str]
        Digests keyed by a stable identifier naming the artifact.  Keys
        absent from the state are simply omitted, so the mapping grows as
        the pipeline populates the state.
    """
    # Collect the digests of every protected artifact present
    digests: Dict[str, str] = {}
    # Digest the frame-bearing state keys first
    for key in (
        "zindi_raw", "taiwan_raw", "zindi_cleansed", "taiwan_cleansed",
        "zindi_transformed", "taiwan_transformed",
    ):
        # Digest the frame when present and of the expected type
        if key in state and isinstance(state[key], pd.DataFrame):
            # Record the frame digest
            digests[key] = _frame_digest(state[key])
    # Digest the per-partition frames held under the partition dictionaries
    for container in ("zindi_frames", "taiwan_frames"):
        # Proceed only when the container is a populated mapping
        if isinstance(state.get(container), dict):
            # Digest each partition frame under a compound key
            for partition, frame in state[container].items():
                # Record the digest when the entry is a dataframe
                if isinstance(frame, pd.DataFrame):
                    # Record the per-partition frame digest
                    digests[f"{container}.{partition}"] = _frame_digest(frame)
    # Digest the array-typed encoded matrices, which the dataframe sweep misses
    for container in (
        "ohe_matrices", "filtered_woe_matrices", "zindi_woe", "taiwan_woe",
    ):
        # Proceed only when the container is a populated mapping
        if isinstance(state.get(container), dict):
            # Walk the mapping looking for array-valued entries
            for entry_key, entry_value in state[container].items():
                # Digest any NumPy array encountered at this level
                if isinstance(entry_value, np.ndarray):
                    # Record the array digest under a compound key
                    digests[f"{container}.{entry_key}"] = _digest_array(
                        entry_value
                    )
                # Descend one level for nested per-partition mappings
                elif isinstance(entry_value, dict):
                    # Digest each array held in the nested mapping
                    for sub_key, sub_value in entry_value.items():
                        # Record the digest when the leaf is an array
                        if isinstance(sub_value, np.ndarray):
                            # Record the nested array digest
                            digests[
                                f"{container}.{entry_key}.{sub_key}"
                            ] = _digest_array(sub_value)
    # Return the captured digests
    return digests


def _verify_test_indices_unchanged(state: Dict[str, Any]) -> None:
    """
    Verify the test partition's indices, labels and row order are unchanged.

    This is the INVARIANCE half of the test-partition information barrier.
    It establishes that the test index array, the test label vector and the
    row ordering of the test frames are byte-identical to the baseline
    recorded immediately after the split.  It detects tampering; it does
    NOT detect a fitting statistic computed over test rows, which leaves
    every one of these artifacts untouched.  That access question is the
    responsibility of `_enforce_partition_access`, and the two are kept
    separate so a reader can tell which guarantee is in force.

    Parameters
    ----------
    state : Dict[str, Any]
        The shared pipeline state.  The baseline digest map is recorded
        under the private key `_test_barrier_baseline` on first call.

    Returns
    -------
    None
        The function is a guard; it returns nothing and raises on failure.

    Raises
    ------
    ValueError
        If any protected test artifact changed after the baseline was set.
    """
    # Skip the check until the partition manifest exists, since the early
    # pipeline stages have no test partition to protect
    if "zindi_partitions" not in state:
        # There is nothing to verify yet
        return
    # Collect the digests of every protected test artifact
    current: Dict[str, str] = {}
    # Digest the test index array, whose byte order encodes the row order
    current["test_idx"] = _digest_array(
        np.asarray(state["zindi_partitions"]["test_idx"], dtype=np.int64)
    )
    # Digest the test label vector when the partition frames are populated
    if isinstance(state.get("zindi_frames"), dict):
        # Retrieve the test frame if present
        test_frame: Any = state["zindi_frames"].get("test")
        # Digest the label column and the row index when the frame exists
        if isinstance(test_frame, pd.DataFrame):
            # Resolve the declared label column name from the configuration
            label_column: str = state["config"]["dataset_parameters"][
                "zindi_fsd"
            ]["label_column"]
            # Digest the test labels when the column is present
            if label_column in test_frame.columns:
                # Record the label digest
                current["test_labels"] = _digest_array(
                    test_frame[label_column].to_numpy(dtype=np.int64)
                )
            # Digest the test frame's row index to pin the row ordering
            current["test_row_order"] = _digest_array(
                np.asarray(test_frame.index.to_numpy())
            )
    # Retrieve the recorded baseline, absent on the first invocation
    baseline: Optional[Dict[str, str]] = state.get("_test_barrier_baseline")
    # Record the baseline on the first checkpoint after the split
    if baseline is None:
        # Store the baseline digest map for subsequent comparisons
        state["_test_barrier_baseline"] = dict(current)
        # Preserve the legacy single-digest key for backward compatibility
        state["_test_idx_digest"] = current["test_idx"]
        # Nothing further to verify on the first call
        return
    # Compare every artifact that was present in the baseline
    changed: List[str] = [
        key for key, digest in current.items()
        if key in baseline and baseline[key] != digest
    ]
    # Raise naming every artifact that changed
    if changed:
        # Raise a descriptive error for the barrier violation
        raise ValueError(
            f"test-partition information barrier violated: the following "
            f"protected test artifacts changed during the pipeline: "
            f"{sorted(changed)}."
        )
    # Extend the baseline with artifacts that appeared after it was recorded
    for key, digest in current.items():
        # Add any newly available artifact so later stages are also pinned
        if key not in baseline:
            # Record the newly available artifact's digest
            baseline[key] = digest


def _enforce_partition_access(stage: str) -> None:
    """
    Enforce the declared partition-access contract for one pipeline stage.

    This is the ACCESS-CONTROL half of the test-partition information
    barrier, and it is the half that speaks to the methodological rule the
    barrier exists to enforce: the absolute prohibition on test data
    influencing any model-selection, hyperparameter or calibration
    decision.  Every stage declares, in `ADAPTER_FITTING_PARTITIONS`, the
    partitions it may read FOR ESTIMATION.  Any stage in the fitting
    sequence that declares test access is rejected before it is invoked.

    Scope of the guarantee, stated honestly.  This is a static check on the
    declaration table.  It proves that the pipeline's declared intent
    excludes test data from estimation, and it fails loudly if a future
    edit grants a fitting stage test access.  It does not, by itself,
    prove that a stage's body honours its own declaration; that is what
    `_verify_frozen_statistics_from_training` tests at runtime.

    Parameters
    ----------
    stage : str
        The canonical pipeline stage name about to be dispatched.

    Returns
    -------
    None
        The function is a guard; it returns nothing and raises on failure.

    Raises
    ------
    ValueError
        If the stage has no declaration, or if a fitting-sequence stage
        declares access to the test partition.
    """
    # Reject an undeclared stage: silence here would defeat the barrier
    if stage not in ADAPTER_FITTING_PARTITIONS:
        # Raise a descriptive error naming the undeclared stage
        raise ValueError(
            f"stage '{stage}' has no partition-access declaration; every "
            f"pipeline stage must declare the partitions it reads for "
            f"estimation before it may be dispatched."
        )
    # Retrieve the partitions this stage may read for estimation
    declared: Tuple[str, ...] = ADAPTER_FITTING_PARTITIONS[stage]
    # Reject test access for any stage in the fitting sequence
    if stage in FITTING_SEQUENCE_STAGES and "test" in declared:
        # Raise a descriptive error naming the offending stage
        raise ValueError(
            f"test-partition information barrier violated: fitting-sequence "
            f"stage '{stage}' declares estimation access to the test "
            f"partition {declared}.  No stage from input validation through "
            f"calibration may read test data for estimation."
        )


def _verify_frozen_statistics_from_training(state: Dict[str, Any]) -> None:
    """
    Recompute frozen preprocessing statistics from training rows alone.

    This is the runtime leakage detector.  A statistic contaminated by
    validation or test rows will not reproduce when recomputed from the
    training slice alone, so recomputation is the only check that directly
    evidences the leakage-prevention discipline rather than inferring it
    from an unchanged digest.  The Zindi imputation medians and the
    winsorization bounds are recomputed and compared against the frozen
    values, because those are the statistics whose definition is exactly
    reproducible from the cleansed frame.

    The cleansed frame is used rather than the transformed frame, because
    the statistics were fitted before the frozen transform was applied;
    recomputing on the transformed frame would compare a post-winsorization
    quantity against a pre-winsorization one.

    Parameters
    ----------
    state : Dict[str, Any]
        The shared pipeline state.  The check is skipped when the cleansed
        frame, the partition manifest or the statistics are not yet
        present.

    Returns
    -------
    None
        The function is a guard; it returns nothing and raises on failure.

    Raises
    ------
    ValueError
        If any recomputed statistic disagrees with its frozen counterpart.
    """
    # Skip until every input the check needs is available in the state
    if not (
        isinstance(state.get("zindi_cleansed"), pd.DataFrame)
        and "zindi_partitions" in state
        and isinstance(state.get("preprocessing_stats"), dict)
    ):
        # The check has nothing to verify yet
        return
    # Retrieve the Zindi block of the frozen statistics
    frozen: Dict[str, Any] = state["preprocessing_stats"].get("zindi", {})
    # Skip when the Zindi block has not been populated
    if not frozen:
        # There is nothing to verify yet
        return
    # Materialise the training row positions from the frozen manifest
    train_idx: np.ndarray = np.asarray(
        state["zindi_partitions"]["train_idx"], dtype=np.int64
    )
    # Slice the cleansed frame down to the training rows only
    train_frame: pd.DataFrame = state["zindi_cleansed"].iloc[train_idx]
    # Collect every disagreement so the error can name all of them at once
    disagreements: List[str] = []
    # Recompute each frozen median from the training slice
    for column, stored_median in frozen.get("medians", {}).items():
        # Skip a column absent from the cleansed frame
        if column not in train_frame.columns:
            # Nothing to recompute for this column
            continue
        # Recompute the median over the training rows alone
        recomputed: float = float(train_frame[column].median())
        # Compare against the frozen value at double-precision tolerance
        if not math.isclose(recomputed, float(stored_median), abs_tol=1e-9):
            # Record the disagreement with both values
            disagreements.append(
                f"median[{column}]: frozen={float(stored_median)!r}, "
                f"recomputed_from_training={recomputed!r}"
            )
    # Recompute each frozen winsorization bound pair from the training slice
    for column, stored_bounds in frozen.get("winsor_bounds", {}).items():
        # Skip a column absent from the cleansed frame
        if column not in train_frame.columns:
            # Nothing to recompute for this column
            continue
        # Recompute the 1st percentile over the training rows alone
        lower: float = float(np.percentile(train_frame[column], 1))
        # Recompute the 99th percentile over the training rows alone
        upper: float = float(np.percentile(train_frame[column], 99))
        # Compare the lower bound against the frozen value
        if not math.isclose(lower, float(stored_bounds[0]), abs_tol=1e-9):
            # Record the lower-bound disagreement
            disagreements.append(
                f"winsor_lower[{column}]: frozen={float(stored_bounds[0])!r}, "
                f"recomputed_from_training={lower!r}"
            )
        # Compare the upper bound against the frozen value
        if not math.isclose(upper, float(stored_bounds[1]), abs_tol=1e-9):
            # Record the upper-bound disagreement
            disagreements.append(
                f"winsor_upper[{column}]: frozen={float(stored_bounds[1])!r}, "
                f"recomputed_from_training={upper!r}"
            )
    # Raise naming every statistic that failed to reproduce
    if disagreements:
        # Raise a descriptive error listing all disagreements
        raise ValueError(
            "test-partition information barrier violated: the following "
            "frozen statistics do not reproduce from the training partition "
            "alone, which indicates they were estimated over held-out rows: "
            + "; ".join(disagreements)
        )


def _verify_test_barrier(
    state: Dict[str, Any], verify_frozen_statistics: bool = False
) -> None:
    """
    Enforce the test-partition information barrier.

    Composes the two halves of the barrier required by Task 20 Step 2.
    The invariance half confirms the test indices, labels and row order are
    unchanged since the split.  The leakage half optionally recomputes the
    frozen preprocessing statistics from training rows alone.  The
    access-control half is enforced separately at dispatch by
    `_enforce_partition_access`, because it must run BEFORE a stage is
    invoked rather than after it returns.

    Parameters
    ----------
    state : Dict[str, Any]
        The shared pipeline state.
    verify_frozen_statistics : bool, optional
        Whether to run the runtime recomputation cross-check.  It is more
        expensive than digest comparison, so it is gated; it is the only
        check that directly detects estimation over held-out rows.

    Returns
    -------
    None
        The function is a guard; it returns nothing and raises on failure.

    Raises
    ------
    ValueError
        If a protected test artifact changed, or if a frozen statistic
        fails to reproduce from the training partition alone.
    """
    # Verify the protected test artifacts are unchanged since the split
    _verify_test_indices_unchanged(state)
    # Optionally recompute the frozen statistics from training rows alone
    if verify_frozen_statistics:
        # Run the runtime leakage cross-check
        _verify_frozen_statistics_from_training(state)


def _measure_rss_bytes() -> Optional[int]:
    """
    Measure the process's peak resident-set size in bytes.

    Task 20 Step 2 requires the per-callable memory footprint to be logged.
    The POSIX `resource` interface is preferred where it exists; on Windows
    it is absent, so the cross-platform process-inspection library is used
    instead.  When neither is available the probe degrades to None rather
    than raising, because a missing memory reading must not abort a study.

    Returns
    -------
    Optional[int]
        The resident-set size in bytes, or None when no probe is
        available.  On Linux the POSIX interface reports kibibytes and is
        scaled here; on macOS it already reports bytes, so the platform is
        consulted before scaling.
    """
    # Prefer the POSIX resource interface when the platform provides it
    if _resource_module is not None:
        # Read the peak resident-set size of the calling process
        usage = _resource_module.getrusage(_resource_module.RUSAGE_SELF)
        # Capture the raw peak value, whose unit is platform-dependent
        peak: int = int(usage.ru_maxrss)
        # Linux reports kibibytes; macOS reports bytes, so scale on Linux only
        if not sys.platform.startswith("darwin"):
            # Convert kibibytes to bytes on Linux and other POSIX platforms
            return peak * 1024
        # Return the macOS value unscaled
        return peak
    # Fall back to the process-inspection library, which works on Windows
    if _psutil_module is not None:
        # Return the current resident-set size of this process in bytes
        return int(_psutil_module.Process().memory_info().rss)
    # Report no measurement rather than raising when no probe exists
    return None


def _json_safe(value: Any) -> Any:
    """
    Convert a state value into a deterministically JSON-serialisable form.

    Checkpoints must be byte-stable across runs so that two executions of
    the same pipeline can be compared by file difference.  NumPy scalars
    and arrays, pandas indices and sets are therefore normalised to plain
    Python containers, and every mapping is emitted with sorted keys by the
    caller's serialiser.

    Parameters
    ----------
    value : Any
        An arbitrary value drawn from the shared pipeline state.

    Returns
    -------
    Any
        A structure composed only of dicts, lists, str, int, float, bool
        and None.  Any object not otherwise convertible is rendered via
        `repr`, so serialisation cannot fail and silently lose a
        checkpoint.
    """
    # Convert NumPy arrays to nested lists, which JSON can represent
    if isinstance(value, np.ndarray):
        # Emit the array as a plain nested list
        return value.tolist()
    # Convert NumPy scalar integers to Python ints
    if isinstance(value, (np.integer,)):
        # Emit the scalar as a Python int
        return int(value)
    # Convert NumPy scalar floats to Python floats
    if isinstance(value, (np.floating,)):
        # Emit the scalar as a Python float
        return float(value)
    # Convert NumPy booleans to Python bools
    if isinstance(value, (np.bool_,)):
        # Emit the scalar as a Python bool
        return bool(value)
    # Recurse through mappings, normalising keys to strings for JSON
    if isinstance(value, dict):
        # Emit a dictionary with string keys and converted values
        return {str(k): _json_safe(v) for k, v in value.items()}
    # Convert sets to sorted lists so ordering is deterministic
    if isinstance(value, (set, frozenset)):
        # Emit the set as a sorted list of converted members
        return sorted(_json_safe(v) for v in value)
    # Recurse through ordered sequences, preserving their order
    if isinstance(value, (list, tuple)):
        # Emit the sequence as a list of converted members
        return [_json_safe(v) for v in value]
    # Pass through the scalar types JSON represents natively
    if isinstance(value, (str, int, float, bool)) or value is None:
        # Emit the scalar unchanged
        return value
    # Render anything else by repr so serialisation cannot fail
    return repr(value)


def _write_stage_checkpoint(
    state: Dict[str, Any], order: int, stage: str
) -> Optional[Path]:
    """
    Serialise the frozen statistics present in the state to a checkpoint.

    Task 20 Step 2 requires that after each callable returns, the
    orchestrator serialise to JSON all frozen statistics: the imputation
    medians and modes, the winsorization bounds, the bin boundaries, the
    WoE tables, the one-hot level sets, the selected regularisation
    constant, the selected adaptive triple, the calibration parameters and
    the partition index arrays.  This is the orchestrator-level checkpoint
    sequence that allows a specific stage to be diffed against a prior run.

    Serialisation is deterministic — sorted keys and a fixed separator set
    — so that two runs of the same pipeline produce byte-identical
    checkpoints and any difference between runs is visible by comparison.

    Parameters
    ----------
    state : Dict[str, Any]
        The shared pipeline state, read immediately after the stage
        returned.
    order : int
        The zero-based execution position of the stage, used to prefix the
        filename so a directory listing sorts into execution order.
    stage : str
        The canonical stage name, used in the filename.

    Returns
    -------
    Optional[Path]
        The path written, or None when the state carries no output
        directory and checkpointing is therefore not possible.

    Raises
    ------
    OSError
        If the checkpoint directory cannot be created or the file written.
    """
    # Skip checkpointing when no output directory has been established
    if "output_dir" not in state:
        # Report that no checkpoint was written
        return None
    # Resolve the checkpoint directory beneath the run's output directory
    checkpoint_dir: Path = Path(state["output_dir"]) / "checkpoints"
    # Create the checkpoint directory including any missing parents
    checkpoint_dir.mkdir(parents=True, exist_ok=True)
    # Collect the frozen statistics that are present at this point
    payload: Dict[str, Any] = {
        key: _json_safe(state[key])
        for key in FROZEN_STATISTIC_KEYS
        if key in state
    }
    # Compose the checkpoint record with its provenance header
    record: Dict[str, Any] = {
        "execution_order": order,
        "stage": stage,
        "frozen_statistics_present": sorted(payload),
        "frozen_statistics": payload,
    }
    # Build the checkpoint path, zero-padded so a listing sorts correctly
    checkpoint_path: Path = checkpoint_dir / f"{order:02d}_{stage}.json"
    # Serialise deterministically: sorted keys and fixed separators
    checkpoint_text: str = json.dumps(
        record, indent=2, ensure_ascii=False, sort_keys=True,
        separators=(",", ": "), default=str,
    )
    # Persist the checkpoint to the output directory
    checkpoint_path.write_text(checkpoint_text, encoding="utf-8")
    # Return the path written for the execution log
    return checkpoint_path


def run_pipeline(state: Dict[str, Any]) -> Dict[str, Any]:
    """
    Execute the registry in strict dependency order with snapshots.

    Implements Task 20 Step 2 in full.  For every stage, in the canonical
    dependency order and with each callable invoked exactly once, the
    orchestrator:

      1. enforces the declared partition-access contract BEFORE dispatch,
         so a fitting-sequence stage that declares test access is rejected
         rather than detected after the fact;
      2. captures before-and-after digests of every frozen artifact and
         raises on any mutation;
      3. serialises the frozen statistics to a per-stage checkpoint;
      4. records wall-clock time, resident-set size and its delta, and
         every warning the stage emitted;
      5. enforces the test-partition information barrier; and
      6. runs the lightweight verifier, halting on a structural failure and
         recording a manuscript-value discrepancy without halting.

    Parameters
    ----------
    state : Dict[str, Any]
        The shared pipeline state, carrying at minimum the configuration,
        the raw frames and the output directory.

    Returns
    -------
    Dict[str, Any]
        The completed state, extended with `execution_log` and
        `verification_by_stage`.

    Raises
    ------
    ValueError
        If a stage mutates a frozen artifact, violates the information
        barrier, breaches its partition-access declaration, or fails a
        structural verification check.
    """
    # Build the callable registry, which self-validates against the order
    registry: Dict[str, Callable[[Dict[str, Any]], Dict[str, Any]]] = (
        build_registry()
    )
    # Collect the per-callable execution log
    execution_log: List[Dict[str, Any]] = []
    # Collect the per-stage verification records for later reporting
    verification_by_stage: Dict[str, Any] = {}
    # Read the barrier cross-check flag, defaulting to off for cost reasons
    verify_frozen: bool = bool(state.get("verify_frozen_statistics", False))
    # Establish the memory baseline before the first stage runs
    previous_rss: Optional[int] = _measure_rss_bytes()
    # Iterate over the canonical pipeline order
    for order, name in enumerate(PIPELINE_ORDER):
        # Enforce the declared partition-access contract BEFORE dispatch
        _enforce_partition_access(name)
        # Capture the pre-callable artifact digests
        before: Dict[str, str] = _snapshot_hashes(state)
        # Capture the pre-callable state key set for the added-keys log
        keys_before: Set[str] = set(state)
        # Record the callable start time
        t_start: float = time.perf_counter()
        # Invoke the callable exactly once inside a warnings-recording context
        with warnings.catch_warnings(record=True) as captured:
            # Ensure every warning is recorded rather than deduplicated away
            warnings.simplefilter("always")
            # Invoke the stage exactly once
            state = registry[name](state)
        # Record the callable elapsed time
        elapsed: float = time.perf_counter() - t_start
        # Capture the post-callable artifact digests
        after: Dict[str, str] = _snapshot_hashes(state)
        # Detect any mutation of a frozen artifact present before the call
        mutations: List[str] = [
            key for key in before
            if key in after and before[key] != after[key]
        ]
        # Raise on any frozen-artifact mutation
        if mutations:
            # Raise a descriptive error naming the mutated artifacts
            raise ValueError(
                f"callable '{name}' mutated frozen frames: {mutations}."
            )
        # Enforce the test-partition information barrier after the stage
        _verify_test_barrier(state, verify_frozen_statistics=verify_frozen)
        # Serialise the frozen statistics present after this stage
        checkpoint_path: Optional[Path] = _write_stage_checkpoint(
            state, order, name
        )
        # Measure the resident-set size after the stage completed
        current_rss: Optional[int] = _measure_rss_bytes()
        # Compute the memory delta attributable to this stage when measurable
        rss_delta: Optional[int] = (
            None if current_rss is None or previous_rss is None
            else current_rss - previous_rss
        )
        # Normalise the captured warnings into a serialisable record
        warning_records: List[Dict[str, str]] = [
            {
                "category": type(w.message).__name__,
                "message": str(w.message),
                "module": str(w.filename),
            }
            for w in captured
        ]
        # Run the lightweight verifier over the state as it now stands
        verification: Dict[str, Any] = run_lightweight_verifier(
            state, stage=name
        )
        # Record this stage's verification outcome for the final report
        verification_by_stage[name] = verification
        # Halt immediately on a structural failure, naming the stage
        if verification["failed"]:
            # Emit the mandated fidelity diagnostic before raising
            _write_fidelity_diagnostic(state, name, verification)
            # Raise a descriptive error naming the stage and failed checks
            raise ValueError(
                f"structural verification failed at stage '{name}': "
                f"{verification['failed']}.  Run the fidelity-diagnosis "
                f"checklist in order: split seed and partition indices; "
                f"binning algorithm variant; rare-level handling; "
                f"calibration fitting details; library-version behaviour; "
                f"statistical-test implementation variant."
            )
        # Record a manuscript-value discrepancy without halting the run
        if verification["manuscript_failed"]:
            # Log the discrepancy as a reproduction finding, not a fault
            logger.warning(
                f"stage '{name}': manuscript-value discrepancies "
                f"{verification['manuscript_failed']} (recorded, not fatal)"
            )
        # Record the execution log entry for this stage
        execution_log.append(
            {
                "callable": name,
                "execution_order": order,
                "elapsed_seconds": elapsed,
                "rss_bytes": current_rss,
                "rss_delta_bytes": rss_delta,
                "warnings": warning_records,
                "checkpoint": (
                    None if checkpoint_path is None else str(checkpoint_path)
                ),
                "state_keys_added": sorted(
                    before_state_keys(state, keys_before)
                ),
                # Count only the checks that were actually EVALUATED as
                # passed.  Counting not-yet-applicable checks as passes
                # would report full coverage at a stage where nothing was
                # compared, which is the precise reporting defect this
                # remediation exists to remove.
                "verification_passed": sum(
                    1 for record in verification["records"].values()
                    if record.get("ok") is True
                ),
                "verification_evaluated": sum(
                    1 for record in verification["records"].values()
                    if record.get("ok") is not None
                ),
                "verification_not_applicable": len(
                    verification["not_applicable"]
                ),
                "verification_declared": len(verification["records"]),
            }
        )
        # Carry the memory reading forward as the next stage's baseline
        previous_rss = current_rss
        # Log the callable completion for the audit trail
        logger.info(
            f"callable '{name}' completed in {elapsed:.2f}s"
            + ("" if rss_delta is None else f" (rss delta {rss_delta:+d} B)")
            + (
                f", {len(warning_records)} warning(s)"
                if warning_records else ""
            )
        )
    # Store the execution log in the state
    state["execution_log"] = execution_log
    # Store the per-stage verification records in the state
    state["verification_by_stage"] = verification_by_stage
    # Return the completed state
    return state


def before_state_keys(
    state: Dict[str, Any], before: Set[str]
) -> Set[str]:
    """
    Return the state keys added by the stage that just executed.

    Differences the post-invocation key set against the key set captured
    BEFORE the stage ran.  The previous implementation compared the state
    against itself at call time and therefore reported an empty set for
    every stage; the pre-invocation set must be supplied by the caller for
    the difference to carry information.

    Parameters
    ----------
    state : Dict[str, Any]
        The shared pipeline state as it stands after the stage returned.
    before : Set[str]
        The state key set captured immediately before the stage was
        invoked.

    Returns
    -------
    Set[str]
        The keys present after the stage that were absent before it.

    Raises
    ------
    TypeError
        If the pre-invocation key set is not a set, which would silently
        reduce the difference to a meaningless comparison.
    """
    # Reject a non-set baseline, which would make the difference meaningless
    if not isinstance(before, (set, frozenset)):
        # Raise a descriptive error naming the received type
        raise TypeError(
            f"before_state_keys expects the pre-invocation key set as a set; "
            f"received {type(before).__name__}."
        )
    # Return the keys the stage added to the state
    return set(state) - set(before)


# -------------------------------------------------------------------------------------------------------------------------------
# Task 20, Step 3: Verification, checkpointing, and reporting
# -------------------------------------------------------------------------------------------------------------------------------


def _resolve_check_actual(
    state: Dict[str, Any], check: VerificationCheck
) -> Tuple[bool, Any]:
    """
    Resolve one check's actual value from the shared pipeline state.

    Walks the check's declared `state_path` through the state, and for the
    checks whose quantity is assembled from several state entries applies a
    dedicated derivation.  A check whose inputs are not yet present is
    reported as unresolved so the verifier can mark it not-yet-applicable,
    which is the ONLY correct handling of an unavailable quantity: gating
    on presence preserves the check's strength, whereas weakening its
    predicate until it passes destroys it.

    Parameters
    ----------
    state : Dict[str, Any]
        The shared pipeline state.
    check : VerificationCheck
        The check whose actual value is being resolved.

    Returns
    -------
    Tuple[bool, Any]
        A resolved flag and the actual value.  The value is None when the
        flag is False.
    """
    # Derive the three Table 3 region shares as PERCENTAGE POINTS.  The state
    # stores shares as fractions of the test partition, whereas the
    # specification declares them as percentages, so the resolver rescales by
    # one hundred; comparing a fraction against a percentage would make the
    # check fail permanently for the wrong reason.
    if check.name == "region_shares":
        # Report unresolved until the region statistics exist
        if not isinstance(state.get("region_stats"), dict):
            # The quantity is not yet available
            return False, None
        # Retrieve the per-region share block, keyed 'shares' in the state
        shares: Any = state["region_stats"].get("shares")
        # Report unresolved when the block is absent
        if not isinstance(shares, dict):
            # The quantity is not yet available
            return False, None
        # Return the shares as percentages in the canonical region order
        return True, tuple(
            float(shares.get(region, float("nan"))) * 100.0
            for region in (
                "Fully interpretable", "Partially interpretable", "ML-driven"
            )
        )
    # Derive the three within-region positive rates as PERCENTAGE POINTS.  The
    # state stores these under the key 'rates' as fractions, so the resolver
    # both selects the correct key and rescales.
    if check.name == "region_positive_rates":
        # Report unresolved until the region statistics exist
        if not isinstance(state.get("region_stats"), dict):
            # The quantity is not yet available
            return False, None
        # Retrieve the per-region positive-rate block, keyed 'rates'
        rates: Any = state["region_stats"].get("rates")
        # Report unresolved when the block is absent
        if not isinstance(rates, dict):
            # The quantity is not yet available
            return False, None
        # Return the rates as percentages in the canonical region order
        return True, tuple(
            float(rates.get(region, float("nan"))) * 100.0
            for region in (
                "Fully interpretable", "Partially interpretable", "ML-driven"
            )
        )
    # Derive the eight routing-rate point estimates in the declared order
    if check.name == "routing_point_estimates":
        # Report unresolved until the routing audit exists
        if not isinstance(state.get("routing"), dict):
            # The quantity is not yet available
            return False, None
        # Declare the (attribute, subgroup) pairs in the specification order
        wanted: Tuple[Tuple[str, str], ...] = (
            ("location", "rural"), ("location", "urban"),
            ("education", "primary or less"),
            ("education", "secondary or above"),
            ("country", "Uganda"), ("country", "Kenya"),
            ("gender", "female"), ("gender", "male"),
        )
        # Collect each subgroup's point estimate in that order
        estimates: List[float] = []
        # Walk the declared pairs, tolerating either record layout
        for attribute, subgroup in wanted:
            # Retrieve the attribute block, reporting unresolved if absent
            block: Any = state["routing"].get(attribute)
            # Report unresolved when the attribute block is missing
            if not isinstance(block, dict):
                # The quantity is not yet available
                return False, None
            # Retrieve the subgroup entry from the attribute block
            entry: Any = block.get(subgroup)
            # Report unresolved when the subgroup entry is missing
            if entry is None:
                # The quantity is not yet available
                return False, None
            # Accept either a bare float or a record carrying a point estimate
            estimates.append(
                float(entry) if isinstance(entry, (int, float))
                else float(entry.get("rate", entry.get("point_estimate")))
            )
        # Return the eight estimates as an ordered tuple
        return True, tuple(estimates)
    # Derive either paired Delta-AUC confidence interval.  Task 16 stores the
    # consolidated inference table as a LIST of comparison records under
    # 'table_9', each carrying a human-readable 'comparison' label and the two
    # interval bounds; a dictionary layout under 'pairwise' is accepted as a
    # fallback so the resolver survives either representation.
    if check.name in (
        "delta_auc_ci_hybrid_vs_lr", "delta_auc_ci_hybrid_vs_xgb"
    ):
        # Select the substring identifying this check's comparison row
        marker: str = (
            "logistic" if check.name.endswith("lr") else "xgboost"
        )
        # Hold the located record, if any
        entry: Any = None
        # Prefer the consolidated table, which is the canonical layout
        if isinstance(state.get("table_9"), list):
            # Scan the table for the row whose comparison label matches
            entry = next(
                (
                    row for row in state["table_9"]
                    if isinstance(row, dict)
                    and marker in str(row.get("comparison", "")).lower()
                ),
                None,
            )
        # Fall back to the dictionary layout keyed by comparison name
        if entry is None and isinstance(state.get("pairwise"), dict):
            # Map the check name to its candidate dictionary keys
            candidates: Tuple[str, ...] = (
                ("hybrid_vs_lr", "hybrid_vs_logistic")
                if check.name.endswith("lr")
                else ("hybrid_vs_xgb", "hybrid_vs_standalone")
            )
            # Locate the first candidate key present in the block
            entry = next(
                (state["pairwise"][k] for k in candidates
                 if k in state["pairwise"]),
                None,
            )
        # Report unresolved when neither layout yielded a record
        if not isinstance(entry, dict):
            # The quantity is not yet available
            return False, None
        # Report unresolved when the interval bounds are absent
        if "ci_low" not in entry or "ci_high" not in entry:
            # The quantity is not yet available
            return False, None
        # Return the interval as an ordered (low, high) pair
        return True, (float(entry["ci_low"]), float(entry["ci_high"]))
    # Derive the ordered top-five attribution ranking
    if check.name == "shap_top_five":
        # Report unresolved until the attribution ranking exists
        ranking: Any = state.get("shap_ranking")
        # Report unresolved when the ranking is absent
        if ranking is None:
            # The quantity is not yet available
            return False, None
        # Accept a bare ordered sequence of feature names
        if isinstance(ranking, (list, tuple)):
            # Normalise the first five entries, tolerating record layouts
            names: List[str] = [
                item if isinstance(item, str) else str(item.get("feature"))
                for item in list(ranking)[:5]
            ]
            # Return the five names as an ordered tuple
            return True, tuple(names)
        # Accept a mapping carrying an explicit ordered ranking
        if isinstance(ranking, dict):
            # Locate the ordered sequence inside the mapping
            ordered: Any = ranking.get("ranking", ranking.get("features"))
            # Report unresolved when no ordered sequence is present
            if not isinstance(ordered, (list, tuple)):
                # The quantity is not yet available
                return False, None
            # Normalise the first five entries to plain names
            names = [
                item if isinstance(item, str) else str(item.get("feature"))
                for item in list(ordered)[:5]
            ]
            # Return the five names as an ordered tuple
            return True, tuple(names)
        # Report unresolved for any other layout
        return False, None
    # Walk the declared state path for every straightforwardly-located check
    cursor: Any = state
    # Descend one path element at a time
    for element in check.state_path:
        # Report unresolved when the cursor cannot be descended
        if isinstance(cursor, dict) and element in cursor:
            # Descend into the mapping
            cursor = cursor[element]
        elif isinstance(cursor, (list, tuple)) and isinstance(element, int) \
                and -len(cursor) <= element < len(cursor):
            # Descend into the sequence by index
            cursor = cursor[element]
        else:
            # The quantity is not yet available
            return False, None
    # Report the resolved value
    return True, cursor


def _compare_check(
    check: VerificationCheck, actual: Any
) -> Tuple[bool, Optional[float]]:
    """
    Apply one check's declared comparison mode to its actual value.

    Implements exactly the four permitted modes.  Membership and threshold
    comparisons are deliberately unavailable: a predicate satisfied by
    every state the pipeline can produce carries no information, and those
    two modes are what produced the tautological checks this design
    replaces.

    Parameters
    ----------
    check : VerificationCheck
        The check carrying the expected value, mode and tolerance.
    actual : Any
        The resolved actual value.

    Returns
    -------
    Tuple[bool, Optional[float]]
        The outcome and, for the numeric modes, the largest absolute
        discrepancy observed.  The discrepancy is None for exact modes.

    Raises
    ------
    ValueError
        If the comparison mode is unrecognised, which `VerificationCheck`
        construction should already have prevented.
    """
    # Apply exact equality for integer counts and categorical values
    if check.comparison == "equals":
        # Compare directly and report no numeric discrepancy
        return bool(actual == check.expected), None
    # Apply ordered sequence equality for rankings and parameter triples
    if check.comparison == "sequence_equals":
        # Reject a non-sequence actual value outright
        if not isinstance(actual, (list, tuple)):
            # The shapes do not match, so the check fails
            return False, None
        # Compare element counts before comparing members
        if len(actual) != len(check.expected):
            # A different length is a failure
            return False, None
        # Compare numerically where both members are numbers, else by equality
        for actual_item, expected_item in zip(actual, check.expected):
            # Use an exact tolerance for floating-point members
            if isinstance(actual_item, (int, float)) and isinstance(
                expected_item, (int, float)
            ):
                # Reject when the pair differs beyond double-precision noise
                if not math.isclose(
                    float(actual_item), float(expected_item), abs_tol=1e-12
                ):
                    # A numeric mismatch is a failure
                    return False, None
            # Fall back to equality for non-numeric members such as names
            elif actual_item != expected_item:
                # A categorical mismatch is a failure
                return False, None
        # Every member matched
        return True, 0.0
    # Apply the absolute or percentage-point tolerance modes
    if check.comparison in ("within_abs", "within_pp"):
        # Normalise both sides to tuples so scalars and vectors share a path
        actual_values: Tuple[Any, ...] = (
            tuple(actual) if isinstance(actual, (list, tuple)) else (actual,)
        )
        # Normalise the expected side the same way
        expected_values: Tuple[Any, ...] = (
            tuple(check.expected)
            if isinstance(check.expected, (list, tuple))
            else (check.expected,)
        )
        # Reject a length mismatch, which indicates a structural change
        if len(actual_values) != len(expected_values):
            # A different length is a failure
            return False, None
        # Track the largest absolute discrepancy across the vector
        worst: float = 0.0
        # Compare each pair against the declared tolerance
        for actual_item, expected_item in zip(actual_values, expected_values):
            # Reject a non-numeric actual value under a numeric mode
            if not isinstance(actual_item, (int, float)) or isinstance(
                actual_item, bool
            ):
                # A non-numeric value cannot satisfy a numeric comparison
                return False, None
            # Reject a non-finite actual value, which no tolerance admits
            if not math.isfinite(float(actual_item)):
                # A non-finite value is a failure
                return False, None
            # Measure this pair's absolute discrepancy
            discrepancy: float = abs(
                float(actual_item) - float(expected_item)
            )
            # Retain the largest discrepancy seen so far
            worst = max(worst, discrepancy)
        # Report the outcome against the declared tolerance
        return worst <= check.tolerance, worst
    # Reject an unrecognised mode, which construction should have prevented
    raise ValueError(
        f"check '{check.name}' declares unsupported comparison "
        f"'{check.comparison}'."
    )


def _write_fidelity_diagnostic(
    state: Dict[str, Any], stage: str, verification: Dict[str, Any]
) -> Optional[Path]:
    """
    Write the mandated diagnostic report for a failed verification.

    Task 20 Step 3 requires that on a failed verification the orchestrator
    write a diagnostic report walking a six-item fidelity checklist in a
    FIXED order.  The ordering is not decorative: it runs from the cheapest
    and most common cause to the most expensive and rarest, and following
    it in order is the intended diagnostic protocol.

    Parameters
    ----------
    state : Dict[str, Any]
        The shared pipeline state, consulted for the output directory.
    stage : str
        The stage at which verification failed, so the report names the
        stage responsible rather than only the run.
    verification : Dict[str, Any]
        The verifier output carrying the records and the failure lists.

    Returns
    -------
    Optional[Path]
        The path written, or None when no output directory is available.

    Raises
    ------
    OSError
        If the report cannot be written.
    """
    # Skip when no output directory has been established
    if "output_dir" not in state:
        # Report that no diagnostic was written
        return None
    # Compose the six-item checklist in the mandated order
    checklist: List[str] = [
        "1. split seed and partition indices",
        "2. binning algorithm variant",
        "3. rare-level handling",
        "4. calibration fitting details",
        "5. library-version behaviour",
        "6. statistical-test implementation variant",
    ]
    # Compose the diagnostic record naming the stage and the failures
    diagnostic: Dict[str, Any] = {
        "stage": stage,
        "structural_failures": verification.get("failed", []),
        "manuscript_failures": verification.get("manuscript_failed", []),
        "failed_records": {
            name: record
            for name, record in verification.get("records", {}).items()
            if record.get("ok") is False
        },
        "fidelity_diagnosis_checklist": checklist,
        "timestamp": datetime.now(timezone.utc).isoformat(),
    }
    # Resolve the diagnostic path beneath the run's output directory
    diagnostic_path: Path = (
        Path(state["output_dir"]) / f"fidelity_diagnostic_{stage}.json"
    )
    # Serialise the diagnostic deterministically
    diagnostic_text: str = json.dumps(
        _json_safe(diagnostic), indent=2, ensure_ascii=False,
        sort_keys=True, default=str,
    )
    # Persist the diagnostic report
    diagnostic_path.write_text(diagnostic_text, encoding="utf-8")
    # Return the path written
    return diagnostic_path


def run_lightweight_verifier(
    state: Dict[str, Any], stage: Optional[str] = None
) -> Dict[str, Any]:
    """
    Run the lightweight verifier over the declared state values.

    Evaluates every check in `VERIFICATION_CHECKS` at the strength the
    specification declares.  Each check is resolved from the shared state,
    compared under its declared mode and tolerance, and recorded with BOTH
    the expected and the actual value, so a passing record is auditable
    rather than merely reassuring.

    Two disciplines govern this implementation and are the reason it
    replaces its predecessor.  First, presence gating: a check whose inputs
    are not yet in the state is recorded as not-yet-applicable, never
    weakened until it passes on whatever happens to be available.  Second,
    failure-class separation: a structural failure indicates a broken
    pipeline and is surfaced for the caller to halt on, whereas a
    manuscript-value failure indicates a reproduction discrepancy and is
    recorded as a finding without halting.

    Parameters
    ----------
    state : Dict[str, Any]
        The shared pipeline state.
    stage : Optional[str], optional
        The stage after which the verifier is running.  Recorded on every
        outcome so a failure can be localised to the stage that produced
        it, which is the verifier's operational purpose.

    Returns
    -------
    Dict[str, Any]
        A mapping carrying `records` (one record per evaluated check),
        `failed` (structural failures only, for the caller to halt on),
        `manuscript_failed` (reproduction discrepancies), `not_applicable`
        (checks whose inputs are absent) and `stage`.

    Raises
    ------
    ValueError
        If a check declares an unsupported comparison mode.
    """
    # Read the two independent verification flags from the state
    verify_structural: bool = bool(state.get("verify_structural", True))
    # Read the manuscript-value flag, defaulting to the legacy flag's value
    verify_manuscript_values: bool = bool(
        state.get(
            "verify_manuscript_values",
            state.get("verify_manuscript_targets", True),
        )
    )
    # Collect one record per evaluated check
    records: Dict[str, Dict[str, Any]] = {}
    # Collect the structural failures, which the caller must halt on
    failed: List[str] = []
    # Collect the manuscript-value discrepancies, which are findings
    manuscript_failed: List[str] = []
    # Collect the checks whose inputs are not yet present
    not_applicable: List[str] = []
    # Evaluate every declared check in specification order
    for check in VERIFICATION_CHECKS:
        # Skip a class of check the caller has switched off, recording why
        if check.structural and not verify_structural:
            # Record the deliberate skip so the reader can see it
            records[check.name] = {
                "check": check.name, "stage": stage, "ok": None,
                "status": "skipped_structural_disabled",
                "expected": check.expected, "actual": None,
                "tolerance": check.tolerance, "source_task": check.source_task,
            }
            # Move to the next check
            continue
        # Skip a manuscript-value check when that class is switched off
        if not check.structural and not verify_manuscript_values:
            # Record the deliberate skip so the reader can see it
            records[check.name] = {
                "check": check.name, "stage": stage, "ok": None,
                "status": "skipped_manuscript_disabled",
                "expected": check.expected, "actual": None,
                "tolerance": check.tolerance, "source_task": check.source_task,
            }
            # Move to the next check
            continue
        # Resolve the actual value from the shared state
        resolved, actual = _resolve_check_actual(state, check)
        # Record a not-yet-applicable check rather than weakening it
        if not resolved:
            # Record the check as awaiting its inputs
            records[check.name] = {
                "check": check.name, "stage": stage, "ok": None,
                "status": "not_applicable",
                "expected": check.expected, "actual": None,
                "tolerance": check.tolerance, "source_task": check.source_task,
            }
            # Track the check as not-yet-applicable
            not_applicable.append(check.name)
            # Move to the next check
            continue
        # Apply the declared comparison at full strength
        ok, discrepancy = _compare_check(check, actual)
        # Record the outcome carrying BOTH compared values
        records[check.name] = {
            "check": check.name,
            "stage": stage,
            "ok": bool(ok),
            "status": "passed" if ok else "failed",
            "expected": check.expected,
            "actual": actual,
            "tolerance": check.tolerance,
            "discrepancy": discrepancy,
            "comparison": check.comparison,
            "source_task": check.source_task,
            "structural": check.structural,
        }
        # Route a failure into its class
        if not ok:
            # A structural failure means the pipeline is broken
            if check.structural:
                # Track the structural failure for the caller to halt on
                failed.append(check.name)
            else:
                # Track the reproduction discrepancy as a finding
                manuscript_failed.append(check.name)
    # Count the checks that were actually evaluated
    evaluated: int = len(failed) + len(manuscript_failed) + sum(
        1 for r in records.values() if r.get("ok") is True
    )
    # Log the verification summary for the audit trail
    logger.info(
        f"lightweight verifier"
        + (f" [{stage}]" if stage else "")
        + f": {evaluated - len(failed) - len(manuscript_failed)}/{evaluated} "
        f"evaluated checks passed; {len(not_applicable)} not yet applicable; "
        f"{len(failed)} structural failure(s); "
        f"{len(manuscript_failed)} manuscript discrepancy(ies)"
    )
    # Return the verification outcome partitioned by failure class
    return {
        "records": records,
        "failed": failed,
        "manuscript_failed": manuscript_failed,
        "not_applicable": not_applicable,
        "stage": stage,
    }


def verifier_negative_self_test() -> Dict[str, Any]:
    """
    Prove every declared check can fail. A release gate, not a nicety.

    For each check in `VERIFICATION_CHECKS`, constructs a state in which
    that quantity is deliberately wrong and asserts the verifier reports
    that check as failed.  A check that cannot be made to fail carries no
    information regardless of how it reads, and must not ship.  The absence
    of this test is the root cause of the defect this implementation
    replaces: five of six checks in the predecessor were satisfied by every
    state the pipeline could produce.

    Process.  A minimal state is synthesised carrying a correct value for
    every check, so the baseline passes.  Each check is then corrupted in
    turn — a numeric expectation is displaced far beyond its tolerance, a
    sequence is permuted — and the verifier is re-run against the corrupted
    state.

    Returns
    -------
    Dict[str, Any]
        A mapping carrying `all_checks_can_fail`, the list of
        `checks_that_cannot_fail`, and the per-check outcome.

    Raises
    ------
    AssertionError
        If any declared check cannot be made to fail.
    """
    # Build a baseline state in which every declared check passes
    def _baseline() -> Dict[str, Any]:
        """Synthesise a state satisfying every declared check exactly."""
        # Compose the minimal state the resolvers require
        return {
            "verify_structural": True,
            "verify_manuscript_values": True,
            "zindi_partitions": {"sizes": {"train": 14114}},
            "iv_values": {
                "education_level": 0.97,
                "cellphone_access": 0.71,
                "job_type": 0.63,
            },
            "selected_triple": (0.2, 0.8, 1.0),
            # Shares and rates are stored as FRACTIONS, exactly as the
            # pipeline stores them, so the self-test exercises the same
            # rescaling path the real state takes
            "region_stats": {
                "shares": {
                    "Fully interpretable": 0.064,
                    "Partially interpretable": 0.341,
                    "ML-driven": 0.594,
                },
                "rates": {
                    "Fully interpretable": 0.538,
                    "Partially interpretable": 0.197,
                    "ML-driven": 0.065,
                },
            },
            # Routing subgroups are records carrying a 'rate' field, exactly
            # as Task 15 stores them
            "routing": {
                "location": {
                    "rural": {"rate": 0.664}, "urban": {"rate": 0.482},
                },
                "education": {
                    "primary or less": {"rate": 0.683},
                    "secondary or above": {"rate": 0.359},
                },
                "country": {
                    "Uganda": {"rate": 0.702}, "Kenya": {"rate": 0.484},
                },
                "gender": {
                    "female": {"rate": 0.594}, "male": {"rate": 0.595},
                },
            },
            # The consolidated inference table is a LIST of comparison
            # records, exactly as Task 16 stores it
            "table_9": [
                {
                    "comparison": "Hybrid vs Logistic Regression",
                    "ci_low": 0.011, "ci_high": 0.020,
                },
                {
                    "comparison": "Hybrid vs XGBoost",
                    "ci_low": -0.009, "ci_high": 0.001,
                },
            ],
            "shap_ranking": [
                "cellphone_access_No",
                "age_of_respondent",
                "relationship_with_head_Head of Household",
                "country_Kenya",
                "country_Rwanda",
            ],
            "thin_file": {"profile": {"n": 1559}},
            "joint_trajectory": {"n_iterations": 3},
        }

    # Corrupt one declared quantity in a supplied state
    def _corrupt(corrupt_state: Dict[str, Any], name: str) -> None:
        """Displace the named quantity far beyond any declared tolerance."""
        # Corrupt the training partition size
        if name == "zindi_train_n":
            # Displace the count by one, beyond an exact comparison
            corrupt_state["zindi_partitions"]["sizes"]["train"] = 14115
        # Corrupt any of the three information values
        elif name.startswith("iv_"):
            # Displace the value far beyond the 0.02 tolerance
            corrupt_state["iv_values"][name[3:]] = 0.0
        # Corrupt the selected adaptive triple
        elif name == "alpha_triple_selected":
            # Replace gamma with another grid member, as observed in the run
            corrupt_state["selected_triple"] = (0.2, 0.8, 0.5)
        # Corrupt the region shares
        elif name == "region_shares":
            # Displace one share far beyond the one-percentage-point tolerance
            corrupt_state["region_stats"]["shares"]["ML-driven"] = 0.200
        # Corrupt the within-region positive rates
        elif name == "region_positive_rates":
            # Displace one rate far beyond the one-percentage-point tolerance
            corrupt_state["region_stats"]["rates"]["ML-driven"] = 0.400
        # Corrupt the routing point estimates
        elif name == "routing_point_estimates":
            # Displace one estimate far beyond the 0.01 tolerance
            corrupt_state["routing"]["location"]["rural"]["rate"] = 0.100
        # Corrupt the hybrid-versus-LR interval
        elif name == "delta_auc_ci_hybrid_vs_lr":
            # Displace the lower bound far beyond tolerance
            corrupt_state["table_9"][0]["ci_low"] = -0.500
        # Corrupt the hybrid-versus-XGBoost interval
        elif name == "delta_auc_ci_hybrid_vs_xgb":
            # Displace the upper bound far beyond tolerance
            corrupt_state["table_9"][1]["ci_high"] = 0.500
        # Corrupt the attribution ranking by transposing its first two entries
        elif name == "shap_top_five":
            # Swap the top two names so the ORDER differs
            ranking: List[str] = list(corrupt_state["shap_ranking"])
            # Transpose the first two entries
            ranking[0], ranking[1] = ranking[1], ranking[0]
            # Store the permuted ranking
            corrupt_state["shap_ranking"] = ranking
        # Corrupt the thin-file subgroup size
        elif name == "thin_file_n":
            # Displace the count far beyond the tolerance of twenty
            corrupt_state["thin_file"]["profile"]["n"] = 1001
        # Corrupt the joint-optimisation iteration count
        elif name == "joint_iterations":
            # Displace the count, as observed in the run of record
            corrupt_state["joint_trajectory"]["n_iterations"] = 2
        else:
            # Refuse to pass silently on an unhandled check
            raise AssertionError(
                f"negative self-test has no corruption defined for check "
                f"'{name}'; every declared check must be provably falsifiable."
            )

    # Confirm the uncorrupted baseline passes every declared check
    baseline_result: Dict[str, Any] = run_lightweight_verifier(
        _baseline(), stage="self_test_baseline"
    )
    # Reject a baseline that does not pass, which would invalidate the test
    if baseline_result["failed"] or baseline_result["manuscript_failed"]:
        # Raise naming the checks that failed on a correct state
        raise AssertionError(
            f"negative self-test baseline must pass every check; failed "
            f"{baseline_result['failed']} and "
            f"{baseline_result['manuscript_failed']}."
        )
    # Collect the per-check outcome of the corruption sweep
    outcomes: Dict[str, bool] = {}
    # Corrupt each declared check in turn and require it to fail
    for check in VERIFICATION_CHECKS:
        # Start from a fresh, correct baseline for every check
        corrupt_state: Dict[str, Any] = copy.deepcopy(_baseline())
        # Corrupt exactly the quantity this check examines
        _corrupt(corrupt_state, check.name)
        # Re-run the verifier against the corrupted state
        result: Dict[str, Any] = run_lightweight_verifier(
            corrupt_state, stage="self_test"
        )
        # Record whether the corresponding check reported failure
        outcomes[check.name] = check.name in (
            result["failed"] + result["manuscript_failed"]
        )
    # Identify any check that could not be made to fail
    cannot_fail: List[str] = [
        name for name, detected in outcomes.items() if not detected
    ]
    # Reject the release when any check is unfalsifiable
    if cannot_fail:
        # Raise naming every check that carries no information
        raise AssertionError(
            f"the following declared checks cannot be made to fail and must "
            f"not ship: {cannot_fail}."
        )
    # Return the self-test outcome for the caller to record
    return {
        "all_checks_can_fail": True,
        "checks_that_cannot_fail": cannot_fail,
        "outcomes": outcomes,
        "n_checks": len(VERIFICATION_CHECKS),
    }


def _verify_equals(
    actual: Any, expected: Any, name: str
) -> Dict[str, Any]:
    """
    Verify an exact equality and compose the record.
    """
    # Compose the verification record
    return {
        "check": name,
        "expected": expected,
        "actual": actual,
        "ok": actual == expected,
    }


def _verify_in(actual: Any, allowed: List[Any], name: str) -> Dict[str, Any]:
    """
    Verify membership in an allowed set and compose the record.
    """
    # Compose the membership verification record
    return {
        "check": name,
        "allowed": allowed,
        "actual": actual,
        "ok": actual in allowed,
    }


def _verify_subset(
    actual: set, allowed: set, name: str
) -> Dict[str, Any]:
    """
    Verify a subset relation and compose the record.
    """
    # Compose the subset verification record
    return {
        "check": name,
        "allowed": sorted(allowed),
        "actual": sorted(actual),
        "ok": actual.issubset(allowed),
    }


def _verify_gt(actual: Any, floor: Any, name: str) -> Dict[str, Any]:
    """
    Verify a strict lower bound and compose the record.
    """
    # Compose the bound verification record
    return {
        "check": name,
        "floor": floor,
        "actual": actual,
        "ok": actual > floor,
    }


def build_reproduction_report(state: Dict[str, Any]) -> Dict[str, Any]:
    """
    Compare the reproduced metrics against the headline results.
    """
    # Extract the headline results from the configuration
    headline: Dict[str, Any] = state["config"]["study_metadata"][
        "headline_results"
    ]
    # Extract the Table 2 calibrated hybrid row
    hybrid_row: Dict[str, Any] = state["tables"]["table_2"][
        "Hybrid (Platt calibrated)"
    ]
    # Extract the routing gaps from the routing report
    routing: Dict[str, Any] = state["routing"]
    # Compute the routing gaps in percentage points
    rural_gap: float = (
        routing["location"]["rural"]["rate"]
        - routing["location"]["urban"]["rate"]
    ) * 100.0
    # Compute the education routing gap
    education_gap: float = (
        routing["education"]["primary or less"]["rate"]
        - routing["education"]["secondary or above"]["rate"]
    ) * 100.0
    # Compute the Uganda-Kenya routing gap
    uganda_gap: float = (
        routing["country"]["Uganda"]["rate"]
        - routing["country"]["Kenya"]["rate"]
    ) * 100.0
    # Compute the gender routing gap
    gender_gap: float = (
        routing["gender"]["female"]["rate"]
        - routing["gender"]["male"]["rate"]
    ) * 100.0
    # Map the reproduced values to the headline keys
    reproduced: Dict[str, Dict[str, Any]] = {
        "zindi_hybrid_auc": {
            "reproduced": hybrid_row["auc"],
            "target": headline["zindi_hybrid_auc"],
            "tolerance": 0.005,
        },
        "zindi_brier_calibrated_hybrid": {
            "reproduced": hybrid_row["brier"],
            "target": headline["zindi_brier_calibrated_hybrid"],
            "tolerance": 0.005,
        },
        "zindi_ece_calibrated_hybrid": {
            "reproduced": hybrid_row["ece"],
            "target": headline["zindi_ece_calibrated_hybrid"],
            "tolerance": 0.005,
        },
        "routing_gap_rural_vs_urban_pp": {
            "reproduced": rural_gap,
            "target": headline["routing_gap_rural_vs_urban_pp"],
            "tolerance": 3.0,
        },
        "routing_gap_primary_vs_secondary_pp": {
            "reproduced": education_gap,
            "target": headline["routing_gap_primary_vs_secondary_pp"],
            "tolerance": 3.0,
        },
        "routing_gap_uganda_vs_kenya_pp": {
            "reproduced": uganda_gap,
            "target": headline["routing_gap_uganda_vs_kenya_pp"],
            "tolerance": 3.0,
        },
        "gender_routing_gap_pp": {
            "reproduced": gender_gap,
            "target": headline["gender_routing_gap_pp"],
            "tolerance": 3.0,
        },
    }
    # Evaluate each headline comparison
    report_entries: Dict[str, Dict[str, Any]] = {}
    # Iterate over the reproduced values
    for name, entry in reproduced.items():
        # Compute the absolute discrepancy
        discrepancy: float = abs(entry["reproduced"] - entry["target"])
        # Compose the pass/fail entry with the discrepancy analysis
        report_entries[name] = {
            "pass": discrepancy <= entry["tolerance"],
            "reproduced": entry["reproduced"],
            "target": entry["target"],
            "tolerance": entry["tolerance"],
            "discrepancy": discrepancy,
        }
    # Compose the reproduction report
    report: Dict[str, Any] = {
        "dataset": "zindi",
        "entries": report_entries,
        "pass_count": sum(1 for e in report_entries.values() if e["pass"]),
        "total_count": len(report_entries),
        "verification": run_lightweight_verifier(state),
    }
    # Return the reproduction report
    return report


# -------------------------------------------------------------------------------------------------------------------------------
# Task 20, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


def run_full_study(
    config: Dict[str, Any],
    zindi_raw: pd.DataFrame,
    taiwan_raw: pd.DataFrame,
    output_dir: Path,
    run_taiwan: bool = False,
    verify_manuscript_targets: bool = True,
    verify_structural: Optional[bool] = None,
    verify_manuscript_values: Optional[bool] = None,
    verify_frozen_statistics: bool = False,
) -> Dict[str, Any]:
    """
    Run the full end-to-end study pipeline (the only public entry).

    Initialises the shared state, executes the 19-callable registry in
    strict dependency order with frozen-state snapshots, per-stage
    checkpoints and the test-partition information barrier, runs the
    lightweight verifier after every stage, builds the reproduction report
    against the headline results, optionally runs the continuity benchmark
    on the second dataset, and prints the summary table.

    Two verification classes, deliberately separated.  Structural checks —
    partition sizes, label domains, array alignments — indicate a broken
    pipeline and halt the run.  Manuscript-value checks indicate a
    reproduction discrepancy, which is a finding rather than a fault, and
    are recorded without halting.  Collapsing both classes behind one
    switch, as the previous design did, meant that disabling the
    reproduction comparison also disabled every structural guarantee.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated study configuration dictionary.
    zindi_raw : pd.DataFrame
        The raw Zindi FSD frame, shape (23524, 13).
    taiwan_raw : pd.DataFrame
        The raw Taiwan credit default frame, shape (30000, 25).
    output_dir : Path
        Directory where all artifacts are persisted.
    run_taiwan : bool, optional
        Whether to run the Taiwan continuity chain.  When True the
        identical callable chain is executed on the second dataframe in an
        isolated state and a SECOND reproduction report is written beneath
        `output_dir / "taiwan"`.
    verify_manuscript_targets : bool, optional
        The legacy flag, retained for signature compatibility.  It gates
        the manuscript-value assertions embedded in the task modules and,
        unless `verify_manuscript_values` overrides it, the verifier's
        manuscript-value class.
    verify_structural : Optional[bool], optional
        Whether the structural verification class runs.  Defaults to True
        and should stay True: a structural failure means every subsequent
        number is meaningless.
    verify_manuscript_values : Optional[bool], optional
        Whether the verifier's manuscript-value class runs.  Defaults to
        `verify_manuscript_targets`.  Failures are recorded, never raised.
    verify_frozen_statistics : bool, optional
        Whether the barrier additionally recomputes the frozen
        preprocessing statistics from the training partition alone.  This
        is the only check that directly detects estimation over held-out
        rows, and it is gated because it is more expensive than a digest
        comparison.

    Returns
    -------
    Dict[str, Any]
        The completed state with all artifacts, the reproduction report,
        the per-stage verification records and, when the continuity
        benchmark ran, the Taiwan outputs and its reproduction report.

    Raises
    ------
    ValueError
        If the output directory is not a Path, if any stage mutates a
        frozen artifact, if the information barrier is violated, or if any
        structural verification check fails.
    """
    # Validate the output directory type before any filesystem work
    if not isinstance(output_dir, Path):
        # Raise a descriptive error for a non-Path output directory
        raise ValueError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Create the output directory including any missing parents
    output_dir.mkdir(parents=True, exist_ok=True)
    # Resolve the structural verification flag, which defaults ON and governs
    # the checks whose failure means the pipeline itself is broken
    resolved_structural: bool = (
        True if verify_structural is None else bool(verify_structural)
    )
    # Resolve the manuscript-value flag, which defaults to the legacy flag so
    # existing callers keep their behaviour; its failures are recorded as
    # reproduction findings rather than raised as faults
    resolved_manuscript: bool = (
        bool(verify_manuscript_targets)
        if verify_manuscript_values is None
        else bool(verify_manuscript_values)
    )
    # Initialise the shared state
    state: Dict[str, Any] = {
        "config": config,
        "zindi_raw": zindi_raw,
        "taiwan_raw": taiwan_raw,
        "output_dir": str(output_dir),
        "run_taiwan": run_taiwan,
        "verify_manuscript_targets": verify_manuscript_targets,
        "verify_structural": resolved_structural,
        "verify_manuscript_values": resolved_manuscript,
        "verify_frozen_statistics": bool(verify_frozen_statistics),
    }
    # Run the sequential pipeline with frozen-state passing; the verifier now
    # runs after every stage inside this call, so a structural failure halts
    # at the stage that produced it rather than at the end of the study
    state = run_pipeline(state)
    # Run the lightweight verifier once more over the completed state
    verification: Dict[str, Any] = run_lightweight_verifier(
        state, stage="final"
    )
    # Raise on any structural verification failure
    if verification["failed"]:
        # Emit the mandated fidelity diagnostic before raising
        _write_fidelity_diagnostic(state, "final", verification)
        # Raise naming the failed structural checks
        raise ValueError(
            f"structural verification failed: {verification['failed']}. "
            f"Run the fidelity-diagnosis checklist in order: split seed; "
            f"binning algorithm; rare-level handling; calibration "
            f"fitting; library versions; statistical-test implementation."
        )
    # Record manuscript-value discrepancies as findings without halting
    if verification["manuscript_failed"]:
        # Emit the diagnostic so the discrepancies are documented on disk
        _write_fidelity_diagnostic(state, "final", verification)
        # Log the discrepancies as reproduction findings
        logger.warning(
            f"manuscript-value discrepancies recorded (not fatal): "
            f"{verification['manuscript_failed']}"
        )
    # Build the reproduction report against the headline results
    reproduction_report: Dict[str, Any] = build_reproduction_report(state)
    # Persist the reproduction report to the output directory
    report_path: Path = output_dir / "reproduction_report.json"
    # Serialise the report to JSON with a stable layout
    report_text: str = json.dumps(
        reproduction_report, indent=2, ensure_ascii=False, default=str
    )
    # Write the reproduction report
    report_path.write_text(report_text, encoding="utf-8")
    # Store the reproduction report in the state
    state["reproduction_report"] = reproduction_report
    # Hold the Taiwan continuity report, populated only when requested
    taiwan_report: Optional[Dict[str, Any]] = None
    # Execute the continuity benchmark when the flag is set.  The flag was
    # previously inert: it was written into the state and never read, so the
    # continuity benchmark never ran and no cross-dataset claim was supported
    # by any artifact.
    if run_taiwan:
        # Run the identical chain on the second dataset in an ISOLATED state
        taiwan_outputs: Dict[str, Any] = _run_taiwan_continuity_chain(
            state, output_dir, verify_manuscript_targets
        )
        # Build the second reproduction report against Taiwan's own targets
        taiwan_report = build_taiwan_reproduction_report(
            config, taiwan_outputs["taiwan_results"]
        )
        # Resolve the nested Taiwan output directory
        taiwan_dir: Path = output_dir / "taiwan"
        # Create the nested directory including any missing parents
        taiwan_dir.mkdir(parents=True, exist_ok=True)
        # Serialise the Taiwan report with the same structure as the primary
        taiwan_text: str = json.dumps(
            taiwan_report, indent=2, ensure_ascii=False, default=str
        )
        # Persist the second reproduction report
        (taiwan_dir / "reproduction_report.json").write_text(
            taiwan_text, encoding="utf-8"
        )
        # Store the Taiwan outputs and report in the state
        state["taiwan_reproduction"] = taiwan_outputs
        # Store the Taiwan reproduction report in the state
        state["taiwan_reproduction_report"] = taiwan_report
    # Print the summary table covering both datasets when Taiwan ran
    _print_summary_table(reproduction_report, taiwan_report)
    # Return the completed state
    return state


def _run_taiwan_continuity_chain(
    state: Dict[str, Any],
    output_dir: Path,
    verify_manuscript_targets: bool,
) -> Dict[str, Any]:
    """
    Run the identical callable chain on the Taiwan continuity dataset.

    Executes the second-dataset branch required by Task 20 Step 3 in an
    ISOLATED state.  Isolation is not stylistic: any frozen statistic
    leaking from the primary dataset into the continuity dataset would
    invalidate the comparison entirely, which is the specific failure this
    benchmark exists to detect.  A deep copy of the cleansed frame and the
    partition manifest is therefore taken, and no mutable object is shared
    with the primary state.

    The joint-optimisation stage is deliberately not run on the second
    dataset, consistent with the manuscript, which excludes it.

    Parameters
    ----------
    state : Dict[str, Any]
        The completed primary state, read for the cleansed Taiwan frame and
        the Taiwan partition manifest produced by the shared chain.
    output_dir : Path
        The primary output directory; Taiwan artifacts are written to a
        nested subdirectory beneath it so the two cannot collide.
    verify_manuscript_targets : bool
        Whether the Taiwan chain should assert its continuity targets.

    Returns
    -------
    Dict[str, Any]
        The Taiwan orchestrator's outputs, carrying `taiwan_results`, the
        `deviation_report` and the artifact paths.

    Raises
    ------
    ValueError
        If the primary state does not carry the cleansed Taiwan frame and
        the Taiwan partition manifest, which the shared chain must have
        produced before the continuity benchmark can run.
    """
    # Require the cleansed Taiwan frame produced by the shared chain
    if not isinstance(state.get("taiwan_cleansed"), pd.DataFrame):
        # Raise a descriptive error naming the missing prerequisite
        raise ValueError(
            "the Taiwan continuity chain requires 'taiwan_cleansed' in the "
            "state; the shared cleansing stage must run first."
        )
    # Require the Taiwan partition manifest produced by the shared chain
    if "taiwan_partitions" not in state:
        # Raise a descriptive error naming the missing prerequisite
        raise ValueError(
            "the Taiwan continuity chain requires 'taiwan_partitions' in the "
            "state; the shared splitting stage must run first."
        )
    # Resolve the nested output directory for the continuity artifacts
    taiwan_dir: Path = output_dir / "taiwan"
    # Create the nested directory including any missing parents
    taiwan_dir.mkdir(parents=True, exist_ok=True)
    # Log the continuity-benchmark start with its placeholder caveat
    logger.info(
        "running the Taiwan continuity benchmark on an isolated state; all "
        "Taiwan preprocessing decisions are UNSPECIFIED IN MANUSCRIPT "
        "placeholders"
    )
    # Deep-copy the partition manifest so the continuity chain cannot mutate
    # the primary state's frozen indices
    isolated_partitions: Dict[str, Any] = copy.deepcopy(
        state["taiwan_partitions"]
    )
    # Copy the cleansed frame so no buffer is shared with the primary state
    isolated_frame: pd.DataFrame = state["taiwan_cleansed"].copy(deep=True)
    # Invoke the Task 22 orchestrator, which applies the placeholder
    # preprocessing and then the identical model chain
    return orchestrate_taiwan_reproduction(
        copy.deepcopy(state["config"]),
        isolated_frame,
        isolated_partitions,
        taiwan_dir,
        verify_manuscript_targets=verify_manuscript_targets,
    )


def build_taiwan_reproduction_report(
    config: Dict[str, Any], taiwan_results: Dict[str, Any]
) -> Dict[str, Any]:
    """
    Compare the Taiwan continuity metrics against their own declared targets.

    Builds the SECOND reproduction report required by Task 20 Step 3.  The
    comparison is made against the Taiwan continuity targets, never against
    the primary dataset's targets: the calibrated hybrid AUC of 0.776
    within 0.005; the Brier score of 0.136 within 0.005; the calibration
    error of 0.011 within 0.003; and the fully-interpretable region share
    of 6.3 percent within one percentage point with a default rate of 69.5
    percent within two percentage points.

    Deviations are recorded, never forced into agreement: the protocol
    forbids adjusting the primary pipeline to make the continuity dataset
    match, and deviations are expected to arise from the placeholder
    preprocessing protocol.  Each failing entry therefore carries the
    six-item fidelity checklist in its fixed diagnostic order.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated study configuration, read for the declared headline
        values where it carries them.
    taiwan_results : Dict[str, Any]
        The Taiwan chain outputs carrying the metric tables and the hybrid
        region statistics.

    Returns
    -------
    Dict[str, Any]
        A report with the same structure as the primary reproduction
        report: `dataset`, `entries` keyed by headline name, `pass_count`,
        `total_count` and the diagnostic checklist.

    Raises
    ------
    ValueError
        If the Taiwan metric tables are absent, which means the continuity
        chain did not complete.
    """
    # Require the metrics block produced by the continuity chain
    if not isinstance(taiwan_results.get("metrics"), dict):
        # Raise a descriptive error naming the missing block
        raise ValueError(
            "the Taiwan continuity chain produced no 'metrics' block; the "
            "reproduction report cannot be built."
        )
    # Retrieve the declared headline values from the configuration
    headline: Dict[str, Any] = config["study_metadata"]["headline_results"]
    # Retrieve the calibrated hybrid row of the Taiwan Table 2
    hybrid_row: Dict[str, Any] = taiwan_results["metrics"]["tables"][
        "table_2"
    ]["Hybrid (Platt calibrated)"]
    # Collect the reproduced values against their declared targets
    reproduced: Dict[str, Dict[str, Any]] = {
        # The calibrated hybrid discrimination on the continuity dataset
        "taiwan_hybrid_auc": {
            "reproduced": float(hybrid_row["auc"]),
            "target": float(headline.get("taiwan_hybrid_auc", 0.776)),
            "tolerance": 0.005,
        },
        # The calibrated hybrid Brier score on the continuity dataset
        "taiwan_brier_calibrated_hybrid": {
            "reproduced": float(hybrid_row["brier"]),
            "target": float(
                headline.get("taiwan_brier_calibrated_hybrid", 0.136)
            ),
            "tolerance": 0.005,
        },
        # The calibrated hybrid expected calibration error
        "taiwan_ece_calibrated_hybrid": {
            "reproduced": float(hybrid_row["ece"]),
            "target": 0.011,
            "tolerance": 0.003,
        },
    }
    # Add the region-behaviour entries when the hybrid block carries them
    region_stats: Any = taiwan_results.get("hybrid", {}).get("region_stats")
    # Proceed only when the region statistics are present
    if isinstance(region_stats, dict):
        # Retrieve the per-region share block
        shares: Any = region_stats.get("shares", {})
        # Retrieve the per-region positive-rate block
        rates: Any = region_stats.get("rates", {})
        # Read the fully-interpretable share as a fraction before rescaling
        fully_share: float = float(
            shares.get("Fully interpretable", float("nan"))
        )
        # Read the fully-interpretable default rate as a fraction
        fully_rate: float = float(
            rates.get("Fully interpretable", float("nan"))
        )
        # Record the fully-interpretable region share as a percentage
        reproduced["taiwan_fully_interpretable_share_pct"] = {
            "reproduced": fully_share * 100.0,
            "target": 6.3,
            "tolerance": 1.0,
        }
        # Record the fully-interpretable region default rate as a percentage
        reproduced["taiwan_fully_interpretable_default_rate_pct"] = {
            "reproduced": fully_rate * 100.0,
            "target": 69.5,
            "tolerance": 2.0,
        }
    # Evaluate each entry against its tolerance and record the discrepancy
    entries: Dict[str, Dict[str, Any]] = {}
    # Walk every reproduced value
    for name, entry in reproduced.items():
        # Measure the absolute discrepancy against the declared target
        discrepancy: float = abs(entry["reproduced"] - entry["target"])
        # Compose the report entry carrying the outcome and the discrepancy
        entries[name] = {
            "pass": bool(discrepancy <= entry["tolerance"]),
            "reproduced": entry["reproduced"],
            "target": entry["target"],
            "tolerance": entry["tolerance"],
            "discrepancy": discrepancy,
        }
    # Count the entries that reproduced within tolerance
    pass_count: int = sum(1 for e in entries.values() if e["pass"])
    # Return the completed continuity report
    return {
        "dataset": "taiwan",
        "entries": entries,
        "pass_count": pass_count,
        "total_count": len(entries),
        "placeholder_caveat": (
            "Taiwan preprocessing is UNSPECIFIED IN MANUSCRIPT; every "
            "preprocessing decision is a documented placeholder, and "
            "deviations are expected to arise from that protocol.  The "
            "primary pipeline must NOT be adjusted to force agreement."
        ),
        "fidelity_diagnosis_checklist": [
            "1. split seed and partition indices",
            "2. binning algorithm variant",
            "3. rare-level handling",
            "4. calibration fitting details",
            "5. library-version behaviour",
            "6. statistical-test implementation variant",
        ],
    }


def _print_summary_table(
    report: Dict[str, Any],
    taiwan_report: Optional[Dict[str, Any]] = None,
) -> None:
    """
    Print the reproduced headline summary table for both datasets.

    Task 20 Step 3 requires a summary table of all reproduced headline
    values with their pass or fail status after the run completes.  When
    the continuity benchmark ran, the table covers BOTH datasets, each
    labelled, so the reader sees the continuity comparison alongside the
    primary results rather than having to open a second file.

    Parameters
    ----------
    report : Dict[str, Any]
        The primary reproduction report.
    taiwan_report : Optional[Dict[str, Any]], optional
        The continuity reproduction report, printed as a second labelled
        block when the benchmark ran.

    Returns
    -------
    None
        The function prints to standard output and returns nothing.
    """

    # Render one labelled report block
    def _print_block(label: str, block: Dict[str, Any]) -> None:
        """Print one dataset's entries beneath a labelled header."""
        # Print the dataset label for this block
        print(f"-- {label} " + "-" * max(0, 75 - len(label)))
        # Iterate over the block's entries
        for name, entry in block["entries"].items():
            # Print the pass/fail line for this headline value
            print(
                f"{'PASS' if entry['pass'] else 'FAIL'} {name}: "
                f"reproduced={entry['reproduced']:.4f} "
                f"target={entry['target']} "
                f"tolerance={entry['tolerance']} "
                f"discrepancy={entry['discrepancy']:.4f}"
            )
        # Print the block's pass count
        print(f"PASS {block['pass_count']}/{block['total_count']}")

    # Print the summary header
    print("=" * 78)
    # Print the report title
    print("REPRODUCTION SUMMARY")
    # Print the header separator
    print("=" * 78)
    # Print the primary dataset block
    _print_block("ZINDI (primary)", report)
    # Print the continuity dataset block when the benchmark ran
    if taiwan_report is not None:
        # Separate the two blocks visually
        print("")
        # Print the continuity dataset block
        _print_block("TAIWAN (continuity benchmark)", taiwan_report)
        # Print the placeholder caveat governing every Taiwan deviation
        print(
            "NOTE: Taiwan preprocessing is UNSPECIFIED IN MANUSCRIPT; "
            "deviations are expected."
        )
````

````python
# Task 21: Robustness Analyses

# ==============================================================================
# Task 21: Robustness Analyses
# ==============================================================================


# The primary interpretability thresholds from Equation (7)
RHO_FULLY_DEFAULT: float = 0.9
RHO_ML_DEFAULT: float = 0.7

# The region label strings
REGION_FULLY: str = "Fully interpretable"
REGION_PARTIAL: str = "Partially interpretable"
REGION_ML: str = "ML-driven"


# -------------------------------------------------------------------------------------------------------------------------------
# Task 21, Step 1: Threshold sensitivity for rho and the decision threshold
# -------------------------------------------------------------------------------------------------------------------------------


def threshold_sensitivity(
    analysis: pd.DataFrame,
    alternative_thresholds: List[Tuple[float, float]],
    decision_thresholds: List[float],
) -> Dict[str, Any]:
    """
    Recompute regions and fairness statistics under alternative thresholds.

    Rebuilds the region assignments under each alternative
    (rho_fully, rho_ml) pair, recomputes the region shares and the
    routing gaps, and recomputes F1, DI and EOD at each alternative
    decision threshold.

    Parameters
    ----------
    analysis : pd.DataFrame
        The test analysis frame with rho, p_cal, the label, the
        protected attributes and the baseline regions.
    alternative_thresholds : List[Tuple[float, float]]
        The alternative (rho_fully, rho_ml) pairs.
    decision_thresholds : List[float]
        The alternative decision thresholds (0.4, 0.6).

    Returns
    -------
    Dict[str, Any]
        The scenario tables for the region shares, routing gaps and
        decision-threshold statistics.
    """
    # Collect the threshold-scenario tables
    scenario_tables: Dict[str, Any] = {}
    # Iterate over the alternative rho threshold pairs
    for rho_fully, rho_ml in alternative_thresholds:
        # Build the scenario key
        scenario: str = f"rho_fully={rho_fully},rho_ml={rho_ml}"
        # Rebuild the region assignments under the alternative thresholds
        regions: np.ndarray = np.empty(len(analysis), dtype=object)
        # Extract the rho vector
        rho: np.ndarray = analysis["rho"].to_numpy(dtype=np.float64)
        # Assign the Fully region under the alternative threshold
        regions[rho >= rho_fully] = REGION_FULLY
        # Assign the Partially region under the alternative thresholds
        regions[(rho >= rho_ml) & (rho < rho_fully)] = REGION_PARTIAL
        # Assign the ML-driven region under the alternative threshold
        regions[rho < rho_ml] = REGION_ML
        # Compute the region shares under the scenario
        shares: Dict[str, float] = {
            region: float(np.mean(regions == region))
            for region in (REGION_FULLY, REGION_PARTIAL, REGION_ML)
        }
        # Compute the routing rate per subgroup under the scenario
        routing_gaps: Dict[str, float] = {}
        # Compute the rural-urban routing gap
        rural: float = float(
            np.mean(rho[analysis["location"].to_numpy() == "rural"] < rho_ml)
        )
        # Compute the urban routing rate
        urban: float = float(
            np.mean(rho[analysis["location"].to_numpy() == "urban"] < rho_ml)
        )
        # Record the rural-urban gap in percentage points
        routing_gaps["rural_vs_urban_pp"] = (rural - urban) * 100.0
        # Compute the primary-education routing rate
        primary: float = float(
            np.mean(
                rho[analysis["education"].to_numpy() == "primary or less"]
                < rho_ml
            )
        )
        # Compute the secondary-education routing rate
        secondary: float = float(
            np.mean(
                rho[analysis["education"].to_numpy() == "secondary or above"]
                < rho_ml
            )
        )
        # Record the education gap in percentage points
        routing_gaps["primary_vs_secondary_pp"] = (primary - secondary) * 100.0
        # Compute the Uganda routing rate
        uganda: float = float(
            np.mean(rho[analysis["country"].to_numpy() == "Uganda"] < rho_ml)
        )
        # Compute the Kenya routing rate
        kenya: float = float(
            np.mean(rho[analysis["country"].to_numpy() == "Kenya"] < rho_ml)
        )
        # Record the Uganda-Kenya gap in percentage points
        routing_gaps["uganda_vs_kenya_pp"] = (uganda - kenya) * 100.0
        # Record the scenario table
        scenario_tables[scenario] = {
            "shares": shares,
            "routing_gaps_pp": routing_gaps,
            "caveat": "alternative thresholds change the region sample "
            "sizes, especially the ML-driven region; small-sample "
            "caveats apply",
        }
    # Collect the decision-threshold tables
    decision_tables: Dict[str, Any] = {}
    # Extract the calibrated probabilities and the labels
    p_cal: np.ndarray = analysis["p_cal"].to_numpy(dtype=np.float64)
    y: np.ndarray = analysis["bank_account"].to_numpy(dtype=np.int64)
    # Iterate over the alternative decision thresholds
    for threshold in decision_thresholds:
        # Build the predicted positives at the alternative threshold
        y_hat: np.ndarray = (p_cal >= threshold).astype(np.int64)
        # Compute the F1 at the alternative threshold
        tp: int = int(np.sum((y_hat == 1) & (y == 1)))
        fp: int = int(np.sum((y_hat == 1) & (y == 0)))
        fn: int = int(np.sum((y_hat == 0) & (y == 1)))
        # Compute the precision at the threshold
        precision: float = tp / (tp + fp) if (tp + fp) > 0 else 0.0
        # Compute the recall at the threshold
        recall: float = tp / (tp + fn) if (tp + fn) > 0 else 0.0
        # Compute the F1 at the threshold
        f1: float = (
            2.0 * precision * recall / (precision + recall)
            if (precision + recall) > 0
            else 0.0
        )
        # Record the decision-threshold row
        decision_tables[f"threshold={threshold}"] = {
            "f1": f1,
            "predicted_positive_rate": float(np.mean(y_hat)),
            "note": "pooled DI/EOD shifts at alternative thresholds are "
            "recorded in the fairness tables; the manuscript's "
            "thresholds are indicative, so the qualitative stability of "
            "the routing pattern is the operational finding",
        }
    # Return the threshold-sensitivity report
    return {
        "scenario_tables": scenario_tables,
        "decision_tables": decision_tables,
    }


# -------------------------------------------------------------------------------------------------------------------------------
# Task 21, Step 2: LightGBM sensitivity check for the residual learner
# -------------------------------------------------------------------------------------------------------------------------------


def lightgbm_sensitivity(
    X_ohe_train: np.ndarray,
    r_train: np.ndarray,
    X_ohe_val: np.ndarray,
    X_ohe_test: np.ndarray,
    p_lr_train: np.ndarray,
    p_lr_val: np.ndarray,
    p_lr_test: np.ndarray,
    alpha_star: Dict[str, np.ndarray],
    y_val: np.ndarray,
    y_test: np.ndarray,
    config: Dict[str, Any],
) -> Dict[str, Any]:
    """
    Refit the residual learner with LightGBM and compare the hybrid.

    Fits LGBMRegressor with the mirror configuration, rebuilds the
    hybrid with the frozen adaptive weights, refits Platt on the
    validation partition, and recomputes the test AUC, Brier, ECE,
    region shares and routing gaps.

    Parameters
    ----------
    X_ohe_train, X_ohe_val, X_ohe_test : np.ndarray
        The one-hot matrices.
    r_train : np.ndarray
        The training residuals.
    p_lr_train, p_lr_val, p_lr_test : np.ndarray
        The LR probabilities per partition.
    alpha_star : Dict[str, np.ndarray]
        The frozen adaptive weights keyed by partition.
    y_val, y_test : np.ndarray
        The validation and test labels.
    config : Dict[str, Any]
        The validated study configuration dictionary.

    Returns
    -------
    Dict[str, Any]
        The LightGBM-based hybrid metrics and the XGBoost comparison.

    Raises
    ------
    ValueError
        If the input lengths are misaligned.
    """
    # Validate the training residual length
    if len(r_train) != X_ohe_train.shape[0]:
        # Raise a descriptive error for the alignment violation
        raise ValueError(
            f"r_train length {len(r_train)} must equal matrix rows "
            f"{X_ohe_train.shape[0]}."
        )
    # Extract the LightGBM mirror configuration from the config
    lgbm_params: Dict[str, Any] = config["lightgbm_sensitivity_parameters"][
        "configuration"
    ]
    # Parse the mirror configuration into a parameter dict
    params: Dict[str, Any] = {
        "objective": "regression",
        "learning_rate": 0.03,
        "num_leaves": 7,
        "subsample": 0.8,
        "feature_fraction": 0.8,
        "lambda_l2": 2,
        "random_state": int(config["partitioning_parameters"]["random_seed"]),
    }
    # Construct and fit the LightGBM residual learner
    lgbm_model: lgb.LGBMRegressor = lgb.LGBMRegressor(**params)
    # Fit the LightGBM learner on the training residuals
    lgbm_model.fit(X_ohe_train, r_train)
    # Predict the residuals on all three partitions
    r_hat_lgbm: Dict[str, np.ndarray] = {
        "train": lgbm_model.predict(X_ohe_train),
        "val": lgbm_model.predict(X_ohe_val),
        "test": lgbm_model.predict(X_ohe_test),
    }
    # Rebuild the hybrid with the frozen adaptive weights (Equation 5)
    p_hybrid_lgbm: Dict[str, np.ndarray] = {
        "train": np.clip(
            p_lr_train + alpha_star["train"] * r_hat_lgbm["train"], 0.0, 1.0
        ),
        "val": np.clip(
            p_lr_val + alpha_star["val"] * r_hat_lgbm["val"], 0.0, 1.0
        ),
        "test": np.clip(
            p_lr_test + alpha_star["test"] * r_hat_lgbm["test"], 0.0, 1.0
        ),
    }
    # Refit the Platt calibration on the validation hybrid
    a_lgbm, b_lgbm, platt_lgbm = fit_platt(p_hybrid_lgbm["val"], y_val)
    # Apply the calibration to the test hybrid
    p_cal_lgbm: np.ndarray = apply_platt(
        p_hybrid_lgbm["test"], a_lgbm, b_lgbm, platt_lgbm
    )
    # Compute the test AUC of the LightGBM-based calibrated hybrid
    auc_lgbm: float = float(roc_auc_score(y_test, p_cal_lgbm))
    # Compute the test Brier of the LightGBM-based hybrid
    brier_lgbm: float = float(brier_score_loss(y_test, p_cal_lgbm))
    # Compute the test ECE of the LightGBM-based hybrid
    ece_lgbm: float = float(compute_ece(p_cal_lgbm, y_test, n_bins=10))
    # Build the region shares from the uncalibrated hybrid components
    rho_lgbm: np.ndarray = np.abs(p_lr_test) / (
        np.abs(p_lr_test) + np.abs(alpha_star["test"] * r_hat_lgbm["test"])
    )
    # Guard the rho denominator
    rho_lgbm = np.where(np.isfinite(rho_lgbm), rho_lgbm, 1.0)
    # Compute the region shares under the default thresholds
    shares_lgbm: Dict[str, float] = {
        REGION_FULLY: float(np.mean(rho_lgbm >= RHO_FULLY_DEFAULT)),
        REGION_PARTIAL: float(
            np.mean((rho_lgbm >= RHO_ML_DEFAULT) & (rho_lgbm < RHO_FULLY_DEFAULT))
        ),
        REGION_ML: float(np.mean(rho_lgbm < RHO_ML_DEFAULT)),
    }
    # Compose the sensitivity report
    report: Dict[str, Any] = {
        "metrics": {
            "auc": auc_lgbm,
            "brier": brier_lgbm,
            "ece": ece_lgbm,
        },
        "region_shares": shares_lgbm,
        "configuration": params,
        "configuration_status": (
            "UNSPECIFIED IN MANUSCRIPT - mirror settings are rational "
            "placeholders; no LightGBM-based number is a headline result"
        ),
    }
    # Log the sensitivity summary for the audit trail
    logger.info(
        f"LightGBM sensitivity: AUC {auc_lgbm:.4f}, Brier "
        f"{brier_lgbm:.4f}, ECE {ece_lgbm:.4f}"
    )
    # Return the sensitivity report
    return report


# -------------------------------------------------------------------------------------------------------------------------------
# Task 21, Step 3: Reference-group sensitivity for the fairness audit
# -------------------------------------------------------------------------------------------------------------------------------


def reference_group_sensitivity(
    analysis: pd.DataFrame,
    alternative_references: Dict[str, str],
) -> Dict[str, Any]:
    """
    Re-run the fairness audit under alternative reference groups.

    Recomputes the routing gaps and the pooled DI/EOD under each
    alternative reference group (female, rural, primary-or-less,
    Rwanda), applying the degenerate-cell rule throughout.

    Parameters
    ----------
    analysis : pd.DataFrame
        The test analysis frame.
    alternative_references : Dict[str, str]
        Attribute -> alternative reference group.

    Returns
    -------
    Dict[str, Any]
        The per-scenario routing gaps, DI and EOD tables.
    """
    # Extract the predicted positives and the labels
    p_cal: np.ndarray = analysis["p_cal"].to_numpy(dtype=np.float64)
    y: np.ndarray = analysis["bank_account"].to_numpy(dtype=np.int64)
    y_hat: np.ndarray = (p_cal >= DECISION_THRESHOLD).astype(np.int64)
    # Collect the scenario tables
    scenario_tables: Dict[str, Any] = {}
    # Iterate over the alternative reference groups
    for attribute, ref in alternative_references.items():
        # Extract the attribute values
        values: np.ndarray = analysis[attribute].to_numpy(dtype=object)
        # Compute the reference positive rate
        ref_rate: float = float(np.mean(y_hat[values == ref]))
        # Compute the pooled DI for every observed subgroup
        pooled_di: Dict[str, Any] = {}
        # Iterate over the observed subgroups
        for subgroup in np.unique(values):
            # Compute the subgroup positive rate
            subgroup_rate: float = float(np.mean(y_hat[values == subgroup]))
            # Apply the degenerate-cell rule
            if ref_rate == 0.0:
                # Record the undefined ratio
                pooled_di[str(subgroup)] = UNDEFINED
            else:
                # Compute the DI ratio
                pooled_di[str(subgroup)] = subgroup_rate / ref_rate
        # Compute the pooled EOD under the alternative reference
        pooled_eod: Any = _eod_for(values, y_hat, y, ref, None)
        # Record the scenario table
        scenario_tables[attribute] = {
            "reference": ref,
            "pooled_di": pooled_di,
            "pooled_eod": pooled_eod,
            "note": (
                "routing-rate gaps and pooled DI/EOD are the least "
                "sensitive to reference-group choice, while per-region "
                "values can shift materially due to small denominators; "
                "per-region DI/EOD are descriptive stratifications "
                "because rho(x) is a post-treatment variable (Rosenbaum "
                "1984)"
            ),
        }
    # Return the reference-group sensitivity report
    return {"scenarios": scenario_tables}


# -------------------------------------------------------------------------------------------------------------------------------
# Task 21, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


def orchestrate_run_robustness(
    config: Dict[str, Any],
    analysis: pd.DataFrame,
    X_ohe_train: np.ndarray,
    X_ohe_val: np.ndarray,
    X_ohe_test: np.ndarray,
    r_train: np.ndarray,
    p_lr: Dict[str, np.ndarray],
    alpha_star: Dict[str, np.ndarray],
    y_val: np.ndarray,
    y_test: np.ndarray,
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Orchestrate the robustness analyses stage.

    Composes the threshold sensitivity, the LightGBM sensitivity and
    the reference-group sensitivity, persisting all reports to JSON.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated study configuration dictionary.
    analysis : pd.DataFrame
        The test analysis frame from Task 15.
    X_ohe_train, X_ohe_val, X_ohe_test : np.ndarray
        The one-hot matrices.
    r_train : np.ndarray
        The training residuals.
    p_lr : Dict[str, np.ndarray]
        The LR probabilities keyed by partition.
    alpha_star : Dict[str, np.ndarray]
        The frozen adaptive weights keyed by partition.
    y_val, y_test : np.ndarray
        The validation and test labels.
    output_dir : Path
        Directory where the artifacts are persisted.

    Returns
    -------
    Dict[str, Any]
        A state-ready dict with threshold_sensitivity,
        lightgbm_sensitivity, reference_sensitivity and output_dir.
    """
    # Validate the output directory type before any filesystem work
    if not isinstance(output_dir, Path):
        # Raise a descriptive error for a non-Path output directory
        raise ValueError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Create the output directory including any missing parents
    output_dir.mkdir(parents=True, exist_ok=True)
    # Step 1: threshold sensitivity for rho and the decision threshold
    threshold_report: Dict[str, Any] = threshold_sensitivity(
        analysis,
        alternative_thresholds=[(0.85, 0.65), (0.95, 0.75)],
        decision_thresholds=[0.4, 0.6],
    )
    # Step 2: LightGBM sensitivity for the residual learner
    lgbm_report: Dict[str, Any] = lightgbm_sensitivity(
        X_ohe_train, r_train, X_ohe_val, X_ohe_test,
        p_lr["train"], p_lr["val"], p_lr["test"],
        alpha_star, y_val, y_test, config,
    )
    # Step 3: reference-group sensitivity for the fairness audit
    reference_report: Dict[str, Any] = reference_group_sensitivity(
        analysis,
        alternative_references={
            "gender": "female",
            "location": "rural",
            "education": "primary or less",
            "country": "Rwanda",
        },
    )
    # Compose the persisted report
    report: Dict[str, Any] = {
        "threshold_sensitivity": threshold_report,
        "lightgbm_sensitivity": lgbm_report,
        "reference_sensitivity": reference_report,
    }
    # Build the artifacts path inside the output directory
    artifacts_path: Path = output_dir / "robustness_artifacts.json"
    # Serialise the report to JSON with a stable layout
    artifacts_text: str = json.dumps(
        report, indent=2, ensure_ascii=False, default=str
    )
    # Persist the report to the output directory
    artifacts_path.write_text(artifacts_text, encoding="utf-8")
    # Return the state-ready robustness outputs
    return {
        "threshold_sensitivity": threshold_report,
        "lightgbm_sensitivity": lgbm_report,
        "reference_sensitivity": reference_report,
        "output_dir": str(output_dir),
    }
````

````python
# Task 22: Taiwan Reproduction (Continuity Benchmark)

# ==============================================================================
# Task 22: Taiwan Reproduction (Continuity Benchmark)
# ==============================================================================


# The Taiwan label column name
TAIWAN_LABEL: str = "default.payment.next.month"


def _partition_frames(
    frame: pd.DataFrame, manifest: Dict[str, Any]
) -> Dict[str, pd.DataFrame]:
    """
    Slice a frame into train/val/test frames by the manifest indices.
    """
    # Slice each partition by the frozen positional indices
    return {
        partition: frame.iloc[
            np.array(manifest[f"{partition}_idx"], dtype=np.int64)
        ]
        for partition in ("train", "val", "test")
    }


# -------------------------------------------------------------------------------------------------------------------------------
# Task 22, Step 1: Preprocess Taiwan under the placeholder protocol
# -------------------------------------------------------------------------------------------------------------------------------


def preprocess_taiwan_placeholder(
    config: Dict[str, Any],
    taiwan_cleansed: pd.DataFrame,
    taiwan_partitions: Dict[str, Any],
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Preprocess Taiwan under the documented placeholder protocol.

    Applies the identical preprocessing callables to Taiwan: frozen
    imputation and winsorization statistics fit on the Taiwan training
    partition only, monotonic binning and WoE encoding for the logistic
    branch, and one-hot encoding for the boosting branch, all with
    training-only frozen statistics kept separate from the Zindi
    artifacts.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated study configuration dictionary.
    taiwan_cleansed : pd.DataFrame
        The cleansed Taiwan frame, shape (30000, 24).
    taiwan_partitions : Dict[str, Any]
        The frozen Taiwan partition manifest.
    output_dir : Path
        Directory where the Taiwan artifacts are persisted.

    Returns
    -------
    Dict[str, Any]
        The Taiwan frames, frozen stats, WoE matrices and one-hot
        matrices.
    """
    # Slice the Taiwan partition frames from the manifest
    taiwan_frames: Dict[str, pd.DataFrame] = _partition_frames(
        taiwan_cleansed, taiwan_partitions
    )
    # Derive the Taiwan numeric and categorical feature sets
    taiwan_numeric, taiwan_categorical = derive_taiwan_feature_sets(config)
    # Fit and apply the Taiwan preprocessing statistics on training only
    preprocessing: Dict[str, Any] = orchestrate_fit_preprocessing_stats(
        config, taiwan_cleansed, taiwan_partitions,
        taiwan_cleansed, taiwan_partitions,
        output_dir / "taiwan_t4_preprocessing",
    )
    # Rebuild the partition frames from the transformed frame
    taiwan_frames = _partition_frames(
        preprocessing["taiwan_frame"], taiwan_partitions
    )
    # Extract the Taiwan training labels
    y_train: np.ndarray = taiwan_frames["train"][TAIWAN_LABEL].to_numpy(
        dtype=np.int64
    )
    # Compose the Taiwan logistic-branch feature order
    taiwan_logistic_features: Tuple[str, ...] = tuple(
        taiwan_numeric + taiwan_categorical
    )
    # Run the WoE pipeline with the Taiwan feature sets
    woe_result: Dict[str, Any] = orchestrate_fit_woe_encoding(
        config,
        taiwan_frames["train"], y_train, taiwan_frames,
        taiwan_logistic_features,
        taiwan_frames["train"], y_train, taiwan_frames,
        taiwan_logistic_features,
        tuple(taiwan_numeric), tuple(taiwan_categorical),
        output_dir / "taiwan_t5_woe",
    )
    # Extract the Taiwan WoE artifacts
    taiwan_woe: Dict[str, Any] = woe_result["taiwan"]
    # Run the IV filter and the one-hot pipeline
    ohe_result: Dict[str, Any] = orchestrate_iv_filter_and_ohe(
        config,
        taiwan_frames["train"], y_train,
        taiwan_woe["woe_map"], taiwan_woe["cut_points"],
        taiwan_woe["matrices"], taiwan_frames,
        tuple(taiwan_categorical), tuple(taiwan_numeric),
        output_dir / "taiwan_t6_ohe",
        verify_manuscript_targets=False,
    )
    # Log the placeholder protocol decisions
    logger.info(
        "Taiwan preprocessing is UNSPECIFIED IN MANUSCRIPT; placeholder "
        "protocol applied: training-only medians, modes, 1st/99th "
        "percentiles, monotonic binning, WoE and one-hot encoding; "
        "numerical drift from the published thesis numbers is possible"
    )
    # Return the Taiwan preprocessing artifacts
    return {
        "frames": taiwan_frames,
        "preprocessing": preprocessing["preprocessing_stats"]["taiwan"],
        "woe": taiwan_woe,
        "ohe_matrices": ohe_result["ohe_matrices"],
        "filtered_woe_matrices": ohe_result["filtered_woe_matrices"],
        "retained_features": ohe_result["retained_features"],
        "column_list": ohe_result["column_list"],
    }


# -------------------------------------------------------------------------------------------------------------------------------
# Task 22, Step 2: Fit and evaluate the Taiwan pipeline
# -------------------------------------------------------------------------------------------------------------------------------


def fit_and_evaluate_taiwan(
    config: Dict[str, Any],
    taiwan_artifacts: Dict[str, Any],
    output_dir: Path,
    verify_manuscript_targets: bool = True,
) -> Dict[str, Any]:
    """
    Run the identical model chain on the Taiwan partitions.

    Executes the same callable chain as the Zindi pipeline: LR with C
    selection, residuals, residual XGBoost, standalone XGBoost,
    adaptive weight grid search, hybrid and rho, Platt calibration and
    metrics, then the bootstrap and DeLong inference.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated study configuration dictionary.
    taiwan_artifacts : Dict[str, Any]
        The Taiwan preprocessing artifacts from Step 1.
    output_dir : Path
        Directory where the Taiwan artifacts are persisted.
    verify_manuscript_targets : bool, optional
        Whether to assert the continuity targets (True for real runs).

    Returns
    -------
    Dict[str, Any]
        The Taiwan metrics, tables and inference results.
    """
    # Extract the Taiwan frames and labels
    frames: Dict[str, pd.DataFrame] = taiwan_artifacts["frames"]
    # Build the per-partition label vectors
    labels: Dict[str, np.ndarray] = {
        partition: frames[partition][TAIWAN_LABEL].to_numpy(dtype=np.int64)
        for partition in ("train", "val", "test")
    }
    # Build the per-partition filtered WoE matrices as arrays
    woe_matrices: Dict[str, np.ndarray] = {
        partition: taiwan_artifacts["filtered_woe_matrices"][
            partition
        ].to_numpy(dtype=np.float64)
        for partition in ("train", "val", "test")
    }
    # Build the per-partition one-hot matrices as arrays
    ohe_matrices: Dict[str, np.ndarray] = {
        partition: taiwan_artifacts["ohe_matrices"][partition].to_numpy(
            dtype=np.float64
        )
        for partition in ("train", "val", "test")
    }
    # Fit the logistic baseline with C selection
    logistic: Dict[str, Any] = orchestrate_fit_logistic(
        config, woe_matrices["train"], labels["train"],
        woe_matrices["val"], labels["val"],
        woe_matrices["test"], labels["test"],
        output_dir / "taiwan_t7_logistic",
        feature_names=taiwan_artifacts["retained_features"],
    )
    # Compute the raw response residuals
    residuals: Dict[str, Any] = orchestrate_compute_residuals(
        labels, logistic["p_lr"]
    )
    # Fit the residual XGBoost learner
    residual: Dict[str, Any] = orchestrate_fit_residual_xgboost(
        config, ohe_matrices, residuals["residuals"],
        logistic["p_lr"], labels, logistic["val_auc"],
        output_dir / "taiwan_t9_residual",
    )
    # Fit the standalone XGBoost comparator
    standalone: Dict[str, Any] = orchestrate_fit_standalone_xgboost(
        config, ohe_matrices, labels,
        output_dir / "taiwan_t10_standalone",
        verify_manuscript_targets=False,
    )
    # Grid-search and freeze the adaptive weights
    adaptive: Dict[str, Any] = orchestrate_select_adaptive_weights(
        config, woe_matrices["train"], labels["train"],
        ohe_matrices["train"], logistic["C"], logistic["p_lr"],
        output_dir / "taiwan_t11_adaptive",
    )
    # Build the hybrid predictions and the rho decomposition
    hybrid: Dict[str, Any] = orchestrate_build_hybrid_and_rho(
        config, logistic["p_lr"], residual["r_hat"],
        adaptive["alpha_star"], labels,
        output_dir / "taiwan_t12_hybrid",
        verify_manuscript_targets=False,
    )
    # Fit and apply the Platt calibration
    platt: Dict[str, Any] = orchestrate_fit_platt(
        hybrid["p_hybrid"], logistic["p_lr"], labels,
        output_dir / "taiwan_t13_platt",
        verify_manuscript_targets=False,
    )
    # Compose the five-model test probability dict
    probabilities: Dict[str, np.ndarray] = {
        "Logistic regression": logistic["p_lr"]["test"],
        "XGBoost": standalone["probs"]["test"],
        "Hybrid (uncalibrated)": hybrid["p_hybrid"]["test"],
        "Hybrid (Platt calibrated)": platt["p_cal_test"],
        "LR + Platt": platt["ablation"]["p_ablation_test"],
    }
    # Compute the evaluation metrics and the tables
    metrics: Dict[str, Any] = orchestrate_compute_metrics(
        probabilities, labels["test"], hybrid["regions"]["test"],
        logistic["p_lr"]["test"], hybrid["p_hybrid"]["test"],
        hybrid["region_stats"], output_dir / "taiwan_t14_metrics",
        verify_manuscript_targets=False,
    )
    # Run the bootstrap and DeLong inference
    inference: Dict[str, Any] = orchestrate_run_bootstrap_delong(
        config,
        {
            "hybrid": hybrid["p_hybrid"]["test"],
            "lr": logistic["p_lr"]["test"],
            "xgb": standalone["probs"]["test"],
        },
        labels["test"], output_dir / "taiwan_t16_inference",
        verify_manuscript_targets=False,
    )
    # Return the Taiwan pipeline results
    return {
        "logistic": logistic,
        "residual": residual,
        "standalone": standalone,
        "adaptive": adaptive,
        "hybrid": hybrid,
        "platt": platt,
        "metrics": metrics,
        "inference": inference,
    }


# -------------------------------------------------------------------------------------------------------------------------------
# Task 22, Step 3: Investigate and document any deviations
# -------------------------------------------------------------------------------------------------------------------------------


def investigate_deviations(
    taiwan_results: Dict[str, Any],
    verify_manuscript_targets: bool = True,
) -> Dict[str, Any]:
    """
    Investigate and document any continuity deviations.

    Compares the Taiwan continuity metrics against the manuscript
    targets within the declared tolerances when verification is
    enabled, and documents the fidelity-diagnosis trail.

    Parameters
    ----------
    taiwan_results : Dict[str, Any]
        The Taiwan pipeline results from Step 2.
    verify_manuscript_targets : bool, optional
        Whether to assert the continuity targets (True for real runs).

    Returns
    -------
    Dict[str, Any]
        The deviation analysis and the fidelity trail.
    """
    # Extract the Table 2 calibrated hybrid row
    hybrid_row: Dict[str, Any] = taiwan_results["metrics"]["tables"][
        "table_2"
    ]["Hybrid (Platt calibrated)"]
    # Extract the region stats
    region_stats: Dict[str, Any] = taiwan_results["hybrid"]["region_stats"]
    # Extract the pairwise inference results
    pairwise: Dict[str, Any] = taiwan_results["inference"]["pairwise"]
    # Compose the reproduced continuity values
    reproduced: Dict[str, Dict[str, Any]] = {
        "taiwan_hybrid_auc": {
            "reproduced": hybrid_row["auc"],
            "target": 0.776,
            "tolerance": 0.005,
        },
        "taiwan_brier": {
            "reproduced": hybrid_row["brier"],
            "target": 0.136,
            "tolerance": 0.005,
        },
        "taiwan_ece": {
            "reproduced": hybrid_row["ece"],
            "target": 0.011,
            "tolerance": 0.003,
        },
        "taiwan_fully_share": {
            "reproduced": region_stats["shares"].get(
                "Fully interpretable", 0.0
            ),
            "target": 0.063,
            "tolerance": 0.01,
        },
        "taiwan_fully_rate": {
            "reproduced": region_stats["rates"].get(
                "Fully interpretable", 0.0
            ),
            "target": 0.695,
            "tolerance": 0.02,
        },
    }
    # Evaluate each continuity comparison
    entries: Dict[str, Dict[str, Any]] = {}
    # Iterate over the reproduced values
    for name, entry in reproduced.items():
        # Compute the absolute discrepancy
        discrepancy: float = abs(entry["reproduced"] - entry["target"])
        # Compose the entry
        entries[name] = {
            "pass": discrepancy <= entry["tolerance"],
            "reproduced": entry["reproduced"],
            "target": entry["target"],
            "tolerance": entry["tolerance"],
            "discrepancy": discrepancy,
        }
    # Extract the Delta-AUC values from the pairwise results
    delta_vs_lr: float = pairwise["Hybrid vs Logistic Regression"][
        "bootstrap"
    ]["delta_auc"]
    # Extract the Delta-AUC versus XGBoost
    delta_vs_xgb: float = pairwise["Hybrid vs XGBoost"]["bootstrap"][
        "delta_auc"
    ]
    # Record the Delta-AUC entries
    entries["delta_auc_vs_lr"] = {
        "reproduced": delta_vs_lr,
        "target": 0.057,
        "tolerance": 0.01,
        "pass": abs(delta_vs_lr - 0.057) <= 0.01,
        "discrepancy": abs(delta_vs_lr - 0.057),
    }
    # Record the Delta-AUC versus XGBoost entry
    entries["delta_auc_vs_xgb"] = {
        "reproduced": delta_vs_xgb,
        "target": 0.001,
        "tolerance": 0.01,
        "pass": abs(delta_vs_xgb - 0.001) <= 0.01,
        "discrepancy": abs(delta_vs_xgb - 0.001),
    }
    # Compose the fidelity-diagnosis trail
    fidelity_trail: List[str] = [
        "1. split seed and partition indices",
        "2. binning algorithm variant",
        "3. rare-level WoE handling",
        "4. calibration fitting details",
        "5. library-version behavior",
        "6. DeLong implementation variant",
    ]
    # Compose the deviation report
    report: Dict[str, Any] = {
        "entries": entries,
        "pass_count": sum(1 for e in entries.values() if e["pass"]),
        "total_count": len(entries),
        "fidelity_trail": fidelity_trail,
        "deviation_policy": (
            "deviations are documented transparently as arising from the "
            "placeholder preprocessing protocol; the Zindi pipeline is "
            "NEVER modified to force Taiwan numbers to match; a "
            "transparent deviation is an acceptable outcome"
        ),
        "f1_asymmetry_note": (
            "the calibrated hybrid F1 can be lower than standalone "
            "XGBoost despite equal AUC because Platt scaling shifts the "
            "predicted probabilities toward the empirical default rate, "
            "reducing the share of observations above the 0.5 threshold"
        ),
        "placeholder_flag": "every Taiwan decision is UNSPECIFIED IN "
        "MANUSCRIPT",
    }
    # Log the deviation summary for the audit trail
    logger.info(
        f"Taiwan continuity: {report['pass_count']}/{report['total_count']} "
        f"entries within tolerance"
    )
    # Return the deviation report
    return report


# -------------------------------------------------------------------------------------------------------------------------------
# Task 22, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


def orchestrate_taiwan_reproduction(
    config: Dict[str, Any],
    taiwan_cleansed: pd.DataFrame,
    taiwan_partitions: Dict[str, Any],
    output_dir: Path,
    verify_manuscript_targets: bool = True,
) -> Dict[str, Any]:
    """
    Orchestrate the Taiwan continuity-benchmark reproduction.

    Composes the placeholder preprocessing, the identical model chain
    and the deviation documentation, persisting the Taiwan
    reproduction report to JSON.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated study configuration dictionary.
    taiwan_cleansed : pd.DataFrame
        The cleansed Taiwan frame, shape (30000, 24).
    taiwan_partitions : Dict[str, Any]
        The frozen Taiwan partition manifest.
    output_dir : Path
        Directory where the Taiwan artifacts are persisted.
    verify_manuscript_targets : bool, optional
        Whether to assert the continuity targets (True for real runs).

    Returns
    -------
    Dict[str, Any]
        A state-ready dict with taiwan_results, deviation_report and
        output_dir.
    """
    # Validate the output directory type before any filesystem work
    if not isinstance(output_dir, Path):
        # Raise a descriptive error for a non-Path output directory
        raise ValueError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Create the output directory including any missing parents
    output_dir.mkdir(parents=True, exist_ok=True)
    # Step 1: preprocess Taiwan under the placeholder protocol
    taiwan_artifacts: Dict[str, Any] = preprocess_taiwan_placeholder(
        config, taiwan_cleansed, taiwan_partitions, output_dir
    )
    # Step 2: fit and evaluate the Taiwan pipeline
    taiwan_results: Dict[str, Any] = fit_and_evaluate_taiwan(
        config, taiwan_artifacts, output_dir, verify_manuscript_targets
    )
    # Step 3: investigate and document any deviations
    deviation_report: Dict[str, Any] = investigate_deviations(
        taiwan_results, verify_manuscript_targets
    )
    # Build the report path inside the output directory
    report_path: Path = output_dir / "taiwan_reproduction_report.json"
    # Serialise the report to JSON with a stable layout
    report_text: str = json.dumps(
        deviation_report, indent=2, ensure_ascii=False, default=str
    )
    # Persist the Taiwan reproduction report
    report_path.write_text(report_text, encoding="utf-8")
    # Return the state-ready Taiwan outputs
    return {
        "taiwan_results": taiwan_results,
        "deviation_report": deviation_report,
        "report_path": str(report_path),
        "output_dir": str(output_dir),
    }
````

````python
# Task 23: Create the Top-Level End-to-End Research Pipeline Orchestrator

# ==============================================================================
# Task 23: Create the Top-Level End-to-End Research Pipeline Orchestrator
# ==============================================================================


# The canonical state keys the downstream steps consume from the chain
REQUIRED_CHAIN_KEYS: Tuple[str, ...] = (
    "analysis",
    "ohe_matrices",
    "residuals",
    "alpha_star",
    "logistic",
    "zindi_frames",
    "zindi_cleansed",
    "taiwan_cleansed",
    "zindi_partitions",
    "taiwan_partitions",
    "selected_triple",
    "tables",
    "reproduction_report",
)

# The canonical pipeline order recorded in the run manifest
PIPELINE_ORDER: Tuple[str, ...] = (
    "validate_inputs",
    "cleanse_data",
    "split_partitions",
    "fit_preprocessing_stats",
    "fit_woe_encoding",
    "fit_ohe_encoding",
    "fit_logistic",
    "compute_residuals",
    "fit_residual_xgboost",
    "fit_standalone_xgboost",
    "select_adaptive_weights",
    "build_hybrid_and_rho",
    "fit_platt",
    "compute_metrics",
    "run_fairness_audit",
    "run_bootstrap_delong",
    "run_shap_analysis",
    "run_thin_file",
    "run_alternating_min",
)


# -------------------------------------------------------------------------------------------------------------------------------
# Task 23, Step 1: Assemble the pipeline state and invoke the Tasks 1-19 chain
# -------------------------------------------------------------------------------------------------------------------------------


def run_tasks_1_to_19_chain(
    config: Dict[str, Any],
    zindi_raw: pd.DataFrame,
    taiwan_raw: pd.DataFrame,
    output_dir: Path,
    verify_manuscript_targets: bool,
) -> Dict[str, Any]:
    """
    Run the verified Tasks 1-19 chain via the Task 20 orchestrator.

    Delegates entirely to run_full_study, which executes the 19-callable
    registry in strict dependency order with the frozen-state snapshots
    and the test-partition information barrier, and validates that the
    returned state carries every key the downstream steps consume.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated study configuration dictionary.
    zindi_raw : pd.DataFrame
        The raw Zindi FSD frame, shape (23524, 13).
    taiwan_raw : pd.DataFrame
        The raw Taiwan credit default frame, shape (30000, 25).
    output_dir : Path
        Directory where all chain artifacts are persisted.
    verify_manuscript_targets : bool
        Whether the manuscript targets are asserted throughout.

    Returns
    -------
    Dict[str, Any]
        The completed chain state with all 19-stage artifacts.

    Raises
    ------
    ValueError
        If a required downstream state key is missing.
    """
    # Invoke the verified Tasks 1-19 chain exactly once
    state: Dict[str, Any] = run_full_study(
        config=config,
        zindi_raw=zindi_raw,
        taiwan_raw=taiwan_raw,
        output_dir=output_dir,
        run_taiwan=False,
        verify_manuscript_targets=verify_manuscript_targets,
    )
    # Validate every required downstream key is present in the state
    missing_keys: List[str] = [
        key for key in REQUIRED_CHAIN_KEYS if key not in state
    ]
    # Raise on any missing key so the failure is early and named
    if missing_keys:
        # Raise a descriptive error naming the missing keys
        raise ValueError(
            f"Tasks 1-19 chain state missing required keys: "
            f"{missing_keys}. Run the fidelity-diagnosis checklist in "
            f"order: split seed; binning algorithm; rare-level handling; "
            f"calibration fitting; library versions; statistical-test "
            f"implementation."
        )
    # Log the chain completion for the audit trail
    logger.info(
        f"Tasks 1-19 chain completed; state carries "
        f"{len(state)} keys"
    )
    # Return the completed chain state
    return state


# -------------------------------------------------------------------------------------------------------------------------------
# Task 23, Step 2: Run the robustness analyses and the Taiwan reproduction chain
# -------------------------------------------------------------------------------------------------------------------------------


def run_robustness_and_taiwan(
    state: Dict[str, Any],
    config: Dict[str, Any],
    output_dir: Path,
    run_taiwan: bool,
) -> Dict[str, Any]:
    """
    Run the Task 21 robustness and the Task 22 Taiwan reproduction.

    Invokes the verified robustness orchestrator on the chain artifacts
    and, when run_taiwan is True, the verified Taiwan reproduction
    orchestrator on the cleansed Taiwan frame and its partition
    manifest.

    Parameters
    ----------
    state : Dict[str, Any]
        The completed Tasks 1-19 chain state.
    config : Dict[str, Any]
        The validated study configuration dictionary.
    output_dir : Path
        Directory where the robustness and Taiwan artifacts are
        persisted.
    run_taiwan : bool
        Whether to run the Taiwan continuity chain.

    Returns
    -------
    Dict[str, Any]
        The state extended with the robustness and Taiwan reports.
    """
    # Convert the one-hot matrices to float64 arrays for the robustness
    # callables, which consume arrays (matching the Task 20 adapters)
    X_ohe_train: np.ndarray = state["ohe_matrices"]["train"].to_numpy(
        dtype=np.float64
    )
    # Convert the validation one-hot matrix to an array
    X_ohe_val: np.ndarray = state["ohe_matrices"]["val"].to_numpy(
        dtype=np.float64
    )
    # Convert the test one-hot matrix to an array
    X_ohe_test: np.ndarray = state["ohe_matrices"]["test"].to_numpy(
        dtype=np.float64
    )
    # Extract the training residuals for the LightGBM sensitivity check
    r_train: np.ndarray = state["residuals"]["train"]
    # Extract the LR probabilities for the robustness callables
    p_lr: Dict[str, np.ndarray] = state["logistic"]["p_lr"]
    # Extract the frozen adaptive weights
    alpha_star: Dict[str, np.ndarray] = state["alpha_star"]
    # Extract the validation labels from the Zindi frames
    y_val: np.ndarray = state["zindi_frames"]["val"]["bank_account"].to_numpy(
        dtype=np.int64
    )
    # Extract the test labels from the Zindi frames
    y_test: np.ndarray = state["zindi_frames"]["test"][
        "bank_account"
    ].to_numpy(dtype=np.int64)
    # Invoke the verified Task 21 robustness orchestrator once
    robustness: Dict[str, Any] = orchestrate_run_robustness(
        config=config,
        analysis=state["analysis"],
        X_ohe_train=X_ohe_train,
        X_ohe_val=X_ohe_val,
        X_ohe_test=X_ohe_test,
        r_train=r_train,
        p_lr=p_lr,
        alpha_star=alpha_star,
        y_val=y_val,
        y_test=y_test,
        output_dir=output_dir / "t23_robustness",
    )
    # Store the robustness report in the state
    state["robustness"] = robustness
    # Run the Taiwan reproduction chain when flagged
    if run_taiwan:
        # Invoke the verified Task 22 Taiwan orchestrator once
        taiwan: Dict[str, Any] = orchestrate_taiwan_reproduction(
            config=config,
            taiwan_cleansed=state["taiwan_cleansed"],
            taiwan_partitions=state["taiwan_partitions"],
            output_dir=output_dir / "t23_taiwan",
        )
        # Store the Taiwan report in the state
        state["taiwan_reproduction"] = taiwan
        # Log the Taiwan completion for the audit trail
        logger.info("Taiwan reproduction chain completed")
    # Return the extended state
    return state


# -------------------------------------------------------------------------------------------------------------------------------
# Task 23, Step 3: Consolidate the artifacts dictionary and persist the run manifest
# -------------------------------------------------------------------------------------------------------------------------------


def consolidate_artifacts(
    state: Dict[str, Any], output_dir: Path, run_taiwan: bool
) -> Dict[str, Any]:
    """
    Consolidate the artifacts of every executed step and persist the
    run manifest.

    Selects the JSON-safe artifact groups from the completed state,
    keys them by stage, attaches the execution log and the reproduction
    report, and persists a run manifest; the returned dictionary is the
    same object the caller receives.

    Parameters
    ----------
    state : Dict[str, Any]
        The completed pipeline state.
    output_dir : Path
        Directory where the run manifest is persisted.
    run_taiwan : bool
        Whether the Taiwan chain was executed.

    Returns
    -------
    Dict[str, Any]
        The consolidated artifacts dictionary keyed by stage.

    Raises
    ------
    ValueError
        If the consolidated dictionary is incomplete.
    """
    # Build the stages-executed list from the flag
    stages_executed: List[str] = ["chain", "robustness"]
    # Append the Taiwan stage when it was executed
    if run_taiwan:
        # Record the Taiwan stage in the executed list
        stages_executed.append("taiwan")
    # Compose the consolidated artifacts dictionary
    consolidated: Dict[str, Any] = {
        "stages_executed": stages_executed,
        "pipeline_order": list(PIPELINE_ORDER),
        "chain": {
            "validation_report": state.get("validation_report"),
            "cleansing_audit": state.get("cleansing_audit"),
            "zindi_partitions": state.get("zindi_partitions"),
            "taiwan_partitions": state.get("taiwan_partitions"),
            "preprocessing_stats": state.get("preprocessing_stats"),
            "iv_values": state.get("iv_values"),
            "retained_features": state.get("retained_features"),
            "logistic": {
                "C": state.get("logistic", {}).get("C"),
                "cv_table": state.get("logistic", {}).get("cv_table"),
                "coef": state.get("logistic", {}).get("coef"),
                "intercept": state.get("logistic", {}).get("intercept"),
                "val_auc": state.get("logistic", {}).get("val_auc"),
            },
            "residual_summary": state.get("residual_summary"),
            "residual_booster": {
                "best_iteration": state.get("best_iteration"),
                "correlation_train": state.get("correlation_train"),
                "provisional_auc": state.get("provisional_auc"),
            },
            "standalone": {
                "scale_pos_weight": state.get("standalone_xgb", {}).get(
                    "scale_pos_weight"
                ),
                "metrics": state.get("standalone_xgb", {}).get("metrics"),
            },
            "adaptive": {
                "selected_triple": state.get("selected_triple"),
                "grid_table": state.get("grid_table"),
            },
            "region_stats": state.get("region_stats"),
            "platt": {
                "a": state.get("platt_a"),
                "b": state.get("platt_b"),
                "a_manuscript": state.get("platt_a"),
                "b_manuscript": state.get("platt_b"),
                "a_sklearn": state.get("platt_a_sklearn"),
                "b_sklearn": state.get("platt_b_sklearn"),
                "parameterisation_note": (
                    "'a'/'b' are the MANUSCRIPT convention and satisfy "
                    "Equation (10) p_cal = 1 / (1 + exp(a * p_Hybrid + b)). "
                    "'a_sklearn'/'b_sklearn' are the negative-exponent "
                    "library convention; the pairs differ by a sign on both "
                    "scalars."
                ),
                "ablation_metrics": state.get("ablation", {}).get("metrics"),
            },
            "tables": state.get("tables"),
            "fairness": {
                "routing": state.get("routing"),
                "di_eod": state.get("di_eod"),
            },
            "inference": state.get("table_9"),
            "shap": {
                "ranking": state.get("shap_ranking"),
                "stability": state.get("shap_stability"),
            },
            "thin_file": {
                "profile": state.get("thin_file", {}).get("profile"),
                "metrics": state.get("thin_file", {}).get("metrics"),
                "hypotheses": state.get("thin_file_hypotheses"),
            },
            "joint_optimization": {
                "trajectory": state.get("joint_trajectory"),
                "conclusion": state.get("joint_conclusion"),
            },
        },
        "robustness": state.get("robustness"),
        "taiwan": state.get("taiwan_reproduction")
        if run_taiwan
        else None,
        "execution_log": state.get("execution_log"),
        "reproduction_report": state.get("reproduction_report"),
    }
    # Validate the consolidated dictionary carries the required stages
    for key in ("chain", "robustness", "execution_log", "reproduction_report"):
        # Assert the consolidated key is present
        if consolidated.get(key) is None:
            # Raise a descriptive error naming the missing stage
            raise ValueError(
                f"consolidated artifacts missing stage '{key}'."
            )
    # Build the manifest path inside the output directory
    manifest_path: Path = output_dir / "run_manifest.json"
    # Serialise the JSON-safe subset of the consolidated dictionary
    manifest_text: str = json.dumps(
        consolidated, indent=2, ensure_ascii=False, default=str
    )
    # Persist the run manifest to the output directory
    manifest_path.write_text(manifest_text, encoding="utf-8")
    # Record the manifest path in the consolidated dictionary
    consolidated["manifest_path"] = str(manifest_path)
    # Log the consolidation for the audit trail
    logger.info(
        f"run manifest persisted to {manifest_path} "
        f"({len(stages_executed)} stages)"
    )
    # Return the consolidated artifacts dictionary
    return consolidated


# -------------------------------------------------------------------------------------------------------------------------------
# Task 23, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


def run_research_pipeline(
    config: Dict[str, Any],
    zindi_raw: pd.DataFrame,
    taiwan_raw: pd.DataFrame,
    output_dir: Path,
    run_taiwan: bool = False,
    verify_manuscript_targets: bool = True,
) -> Dict[str, Any]:
    """
    Run the full end-to-end research pipeline (Tasks 1-22).

    The top-level orchestrator of the study: invokes the verified
    Tasks 1-19 chain via run_full_study, the Task 21 robustness
    analyses, and the Task 22 Taiwan reproduction when flagged, then
    consolidates the artifacts of every executed step into a single
    dictionary and persists the run manifest.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated study configuration dictionary.
    zindi_raw : pd.DataFrame
        The raw Zindi FSD frame, shape (23524, 13).
    taiwan_raw : pd.DataFrame
        The raw Taiwan credit default frame, shape (30000, 25).
    output_dir : Path
        Directory where all artifacts and the run manifest are
        persisted.
    run_taiwan : bool, optional
        Whether to run the Taiwan continuity chain.
    verify_manuscript_targets : bool, optional
        Whether the manuscript targets are asserted throughout.

    Returns
    -------
    Dict[str, Any]
        The consolidated artifacts dictionary of all executed steps,
        keyed by stage, with the run manifest path recorded.

    Raises
    ------
    ValueError
        If the output directory is invalid or a stage fails.
    """
    # Validate the output directory type before any filesystem work
    if not isinstance(output_dir, Path):
        # Raise a descriptive error for a non-Path output directory
        raise ValueError(
            f"output_dir must be a pathlib.Path; "
            f"received {type(output_dir).__name__}."
        )
    # Create the output directory including any missing parents
    output_dir.mkdir(parents=True, exist_ok=True)
    # Step 1: run the verified Tasks 1-19 chain
    state: Dict[str, Any] = run_tasks_1_to_19_chain(
        config=config,
        zindi_raw=zindi_raw,
        taiwan_raw=taiwan_raw,
        output_dir=output_dir,
        verify_manuscript_targets=verify_manuscript_targets,
    )
    # Step 2: run the robustness analyses and the Taiwan chain
    state = run_robustness_and_taiwan(
        state=state,
        config=config,
        output_dir=output_dir,
        run_taiwan=run_taiwan,
    )
    # Step 3: consolidate the artifacts and persist the run manifest
    consolidated: Dict[str, Any] = consolidate_artifacts(
        state=state, output_dir=output_dir, run_taiwan=run_taiwan
    )
    # Assert the manifest file exists before returning
    if not Path(consolidated["manifest_path"]).is_file():
        # Raise a descriptive error for the missing manifest
        raise FileNotFoundError(
            f"run manifest missing at {consolidated['manifest_path']}."
        )
    # Log the pipeline completion for the audit trail
    logger.info(
        "Task 23: full research pipeline completed; "
        f"{len(consolidated['stages_executed'])} stages executed"
    )
    # Return the consolidated artifacts dictionary
    return consolidated
````
