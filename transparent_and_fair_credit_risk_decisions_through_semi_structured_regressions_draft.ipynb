This repository contains an **independent**, professional-grade Python implementation of the research methodology from the **arXiv preprint** entitled **"Interpretable hybrid credit scoring for thin-file and underbanked populations: A fairness-aware residual learning framework with evidence from East African financial inclusion data"** by:

*   **Belise Kanziga**
*   **Yaé U. Gaba**
*   **Olivier Kanamugire**

The paper provides no code; this project translates its mathematical specifications into robust, versioned, executable software artifacts. The project delivers a **decision rule whose auditability is a measured quantity rather than an assertion**: a logistic-regression scorecard corrected by a gradient-boosted learner fitted on its raw response residuals, decomposed at every prediction into an interpretability ratio $\rho(x)$ that reports the share of the decision attributable to the linear branch, together with a fairness audit conducted **at the granularity of the three interpretability regions** — the contribution that aggregate fairness metrics structurally cannot provide.

## Introduction

This project provides a Python implementation of the analytical framework presented in Kanziga, Gaba and Kanamugire (2026). The core of this repository is the iPython Notebook `interpretable_hybrid_credit_scoring_for_thin_file_and_underbanked_populations_draft.ipynb`, which contains a comprehensive suite of **170 callables (165 functions, 2 classes and 3 class methods)** across Tasks 1 to 23, including the two orchestration layers.

Credit scoring in thin-file and underbanked populations poses a specific institutional problem. Logistic-regression scorecards remain the regulatory default because their coefficients *are* the explanation, and adverse-action notices can be generated directly from them. Gradient-boosted ensembles predict better but explain worse, and post-hoc attributions describe a surrogate rather than the decision rule itself. The residual-learning hybrid resolves the tension additively: the scorecard produces $p_{\text{LR}}(x)$, a shallow booster learns the raw response residual $r_i = y_i - p_{\text{LR}}(x_i)$, and the two combine into a single prediction whose linear share is computable per borrower.

That per-borrower share is what makes the framework auditable, and it is also what creates the risk this paper isolates. If the opaque region is not distributed evenly across the population, a lender can satisfy aggregate fairness tests while systematically routing protected subgroups into the part of the model where explanations are weakest. On the East African data the routing gaps are large: **18 percentage points** rural versus urban, **32 points** primary-or-less versus secondary-and-above, **22 points** Ugandan versus Kenyan, and essentially none by gender.

This codebase operationalizes the framework, allowing users to:

-   Validate the study configuration against **32 required top-level sections**, the equation-parameter map covering Equations (1)–(12) plus five unnumbered equations, and every pinned hyperparameter, halting on the first failure with a descriptive error naming the full dotted key path.
-   Build a Weight-of-Evidence scorecard branch under a strict information barrier: monotonic bin boundaries, WoE tables, imputation medians and modes, and winsorization bounds all estimated on the **training partition only** and frozen for validation and test.
-   Fit the residual booster with early stopping monitored on the **validation AUC of the provisional hybrid** $p_{\text{LR}}(x) + \hat{r}(x)$ — a criterion no native boosting library can express, implemented here as a custom training callback.
-   Select the adaptive weighting triple $(\alpha_{\min}, \alpha_{\max}, \gamma)$ by exhaustive search over all **27** grid points with five-fold stratified cross-validation, refitting the scorecard, the residuals and the booster inside every fold.
-   Decompose every prediction into $\rho(x)$ and assign it to one of three interpretability regions at the thresholds $0.9$ and $0.7$.
-   Audit fairness **within each region**, with disparate impact and equalised-odds differences computed against configurable reference groups and degenerate cells reported as explicitly undefined (`"---"`) rather than silently zero.
-   Quantify uncertainty with a stratified paired bootstrap ($B = 2000$) that resamples within class strata and the Sun–Xu fast midrank DeLong test for two correlated ROC curves.
-   Segment the thin-file subpopulation by a frozen composite index and test the two pre-registered hypotheses, including the one the manuscript reports as **not supported**.
-   Execute the entire workflow through the single top-level orchestrator `run_research_pipeline(...) -> Dict[str, Any]`.

> **Reproduction status — read this before running anything.** The notebook ships with a schema-conformant **synthetic corpus** generated in the *Study Inputs* cell from `np.random.default_rng(71)`, because the workspace carries no raw CSVs. It reproduces the published row counts, column order and class balance exactly, so every code path executes end to end — but the Zindi label it produces is a deterministic threshold on a linear function of six features, and no metric computed on it is a reproduction of the paper. Supplying the genuine Zindi and Taiwan files is a prerequisite for reproduction, not an optimisation. See [Input Data Structure](#input-data-structure).

## Theoretical Background

The implemented methods combine Statistical Machine Learning, Mathematical Statistics, Numerical Optimization, Algorithmic Fairness, Credit-Risk Scorecard Methodology, and Reproducible Research Engineering.

**1. Weight-of-Evidence Encoding and Information Value.** The linear branch consumes a WoE-transformed matrix. For bin or level $k$ with $g_k$ observations of the negative class and $b_k$ of the positive class, against training totals $G$ and $B$:

$$
\text{WoE}_k = \ln\!\left(\frac{g_k/G}{b_k/B}\right), \qquad \text{IV} = \sum_{k}\left(\frac{g_k}{G} - \frac{b_k}{B}\right)\text{WoE}_k .
$$

The class convention is pinned by the configuration: $g_k$ counts $y = 0$ and $b_k$ counts $y = 1$. Any bin or level with a zero count in either class makes the logarithm undefined and receives the documented **neutral value** $\text{WoE}_k = 0$, which is also the value assigned to a category absent from the training partition. Features are retained when $\text{IV} \ge 0.02$. Continuous predictors are binned from twenty quantile bins, merged until the WoE sequence is monotone, then merged again to satisfy a minimum bin frequency of five percent of training observations, and monotonicity is re-enforced afterwards because frequency merging can break it.

**2. The Hybrid Prediction Rule.** The scorecard produces $p_{\text{LR}}(x) = \sigma(\hat{\beta}^\top x + \hat{\beta}_0)$ under L2-regularised maximum likelihood with balanced class weights, the inverse-regularisation constant $C$ selected by five-fold stratified cross-validation over $\{0.01, 0.1, 1, 10\}$. The booster is fitted on the raw response residual $r_i = y_i - p_{\text{LR}}(x_i)$ under squared-error loss, and the two branches combine as

$$
p_{\text{Hybrid}}(x) = \mathrm{clip}\!\left(p_{\text{LR}}(x) + \alpha^*(x)\,\hat{r}(x),\; 0,\; 1\right), \qquad \hat{r}(x) = f_{\text{GBM}}(x) = \sum_{m=1}^{M}\nu\, h_m(x).
$$

The residual definition is load-bearing. Because $r_i$ is a *raw response* residual rather than a gradient, the booster's output lives on the probability scale, which is what permits the additive decomposition below to be read directly as a share.

**3. Adaptive Weighting.** The weight on the residual correction is not constant. It is driven by an uncertainty proxy that peaks where the scorecard is least decisive:

$$
u_{\text{LR}}(x) = 4\,p_{\text{LR}}(x)\left(1 - p_{\text{LR}}(x)\right) \in [0,1], \qquad \alpha^*(x) = \alpha_{\min} + \left(\alpha_{\max} - \alpha_{\min}\right)u_{\text{LR}}(x)^{\gamma}, \quad \gamma > 0 .
$$

The triple is selected over the Cartesian product $\{0.0, 0.1, 0.2\} \times \{0.3, 0.5, 0.8\} \times \{0.5, 1.0, 2.0\}$ — 27 candidates — by mean validation AUC across five stratified folds.

> **Registered divergence (1 of 5).** Section 3.4 of the manuscript specifies five-fold cross-validation for all three adaptive parameters; Table 1 states three-fold. The implementation resolves the inconsistency to **five-fold for every triple** and logs the resolution verbatim in the audit trail rather than choosing silently.

**4. The Interpretability Ratio and the Three-Region Partition.** Every prediction decomposes into a linear share:

$$
\rho(x) = \frac{\left|p_{\text{LR}}(x)\right|}{\left|p_{\text{LR}}(x)\right| + \left|\alpha^*(x)\,\hat{r}(x)\right|},
\qquad
\text{Region}(x) =
\begin{cases}
\text{Fully interpretable} & \rho(x) \ge 0.9,\\
\text{Partially interpretable} & 0.7 \le \rho(x) < 0.9,\\
\text{ML-driven} & \rho(x) < 0.7.
\end{cases}
$$

$\rho(x) = 1$ exactly when the residual correction vanishes. The thresholds are **indicative rather than theoretically derived**, and their sensitivity is evaluated empirically at $(0.85, 0.65)$ and $(0.95, 0.75)$.

> **Registered divergence (2 of 5).** The ratio is defined on the **probability scale**, not the logit scale. Borrowers with $p_{\text{LR}}(x)$ near one therefore tend to fall into the fully-interpretable region whenever the residual correction is bounded, which mechanically couples that region to the high-$p_{\text{LR}}$ tail. The empirical finding that the fully-interpretable region concentrates high-base-rate borrowers is consequently a joint property of the ratio's algebraic form and the scorecard's calibration on that tail — not a discovered feature independent of the definition. A logit-scale variant is left as future work by the manuscript and is not implemented here.

**5. Probability Calibration.** The hybrid produces values inside the unit interval but does not guarantee calibration. Platt scaling is applied post hoc, with the two scalars estimated by **unpenalised maximum likelihood** on the held-out validation fold:

$$
p_{\text{cal}}(x) = \frac{1}{1 + \exp\!\left(a\, p_{\text{Hybrid}}(x) + b\right)},
\qquad
\text{ECE} = \sum_{k=1}^{K}\frac{|B_k|}{n}\left|\mathrm{acc}(B_k) - \mathrm{conf}(B_k)\right|, \quad K = 10 .
$$

Region assignments are computed on the **uncalibrated** hybrid probabilities: calibration changes the probability scale, never $\rho(x)$ or region membership.

> **Registered divergence (3 of 5).** Equation (10) carries a **positive** exponent, whereas `sklearn.linear_model.LogisticRegression` parameterises a fitted binary model with a negative one. The two conventions differ by a sign on both scalars, so $a = -w$ and $b = -c$ for library coefficient $w$ and intercept $c$. Exporting the library pair under the manuscript's names makes Equation (10) evaluate to $1 - p_{\text{cal}}$ for every input, inverting every decision taken by a consumer who reconstructs calibration from the reported parameters. This implementation exports the **manuscript pair**, persists both under unambiguous keys with the negation relationship stated, and asserts the round trip numerically to $10^{-10}$ on every fit.

**6. Fairness at the Granularity of Interpretability Regions.** Predicted positives are $\hat{y} = \mathbf{1}[p_{\text{cal}}^{\text{test}} \ge 0.5]$. Against a reference group $g_{\text{ref}}$:

$$
\text{DI}(g) = \frac{\Pr[\hat{y} = 1 \mid A = g]}{\Pr[\hat{y} = 1 \mid A = g_{\text{ref}}]},
\qquad
\text{EOD} = \max_{g}\left\{\left|\text{TPR}(g) - \text{TPR}(g_{\text{ref}})\right|,\ \left|\text{FPR}(g) - \text{FPR}(g_{\text{ref}})\right|\right\},
$$

each computed **pooled and restricted to each region**. The routing quantity itself is $\hat{\Pr}\!\left[\rho(x) < 0.7 \mid A = g\right]$, with 95% percentile bootstrap intervals from $B = 500$ resamples drawn within $(\text{class} \times \text{attribute})$ strata.

Two disciplines govern the interpretation. First, **degenerate cells**: when the reference group has no predicted positives in a region, the TPR and FPR denominators collapse and the cell is reported as undefined, never as zero. Second, **post-treatment conditioning**: $\rho(x)$ is causally downstream of the features, so per-region disparate-impact and equalised-odds values are *descriptive stratifications*, not causal fairness measures. The four-fifths convention ($0.8$–$1.25$) is applied as the reporting band, following the U.S. EEOC Uniform Guidelines.

**7. Inference.** Two complementary vehicles, deliberately not interchangeable. The stratified paired bootstrap ($B = 2000$) resamples with replacement **within** the positive and negative test strata separately, preserving the marginal class distribution, and computes both models' AUCs on the same resample to form $\Delta\text{AUC}^{(b)}$; the interval is the 2.5th and 97.5th percentiles. Models are never refitted inside the bootstrap — the routine consumes pre-computed test predictions, which makes refitting structurally impossible rather than merely prohibited. The DeLong test uses the Sun–Xu fast midrank algorithm for the covariance of two correlated AUC estimates:

$$
z = \frac{\text{AUC}_1 - \text{AUC}_2}{\sqrt{\widehat{\mathrm{Var}}\!\left(\text{AUC}_1 - \text{AUC}_2\right)}},
$$

reported one-sided and two-sided. No multiple-comparison correction is applied, consistent with the descriptive audit intent, and that decision is recorded in the audit log rather than left implicit.

**8. Prediction-Level Explanations.** For observations in the ML-driven region the residual term dominates and the logistic coefficients no longer suffice. TreeSHAP attributions are computed on $f_{\text{GBM}}$ **only** — never on the full hybrid — so the additive decomposition survives, and the booster is sliced to its frozen best iteration before explanation so no tree absent from prediction is attributed:

$$
\overline{\left|\phi_j\right|} = \frac{1}{n_{\text{ML}}}\sum_{x \in \text{ML-region}}\left|\phi_j(x)\right| .
$$

Stability is screened by $B = 500$ observation-level bootstrap resamples, reporting each feature's rank standard deviation and the top-five set preservation rate, because post-hoc attributions are known to be attackable and an unstable ranking is a flag on the individual explanation rather than on the framework.

**9. Thin-File Segmentation.** The composite index sums three components standardised on the training partition only, with the threshold $\tau$ frozen at the training 25th percentile:

$$
\text{index} = z_{\text{education}} + z_{\text{cellphone}} + z_{\text{formaljob}}, \qquad z_j = \frac{x_j - \mu_j^{\text{train}}}{\sigma_j^{\text{train}}}, \qquad \text{thin-file} = \{\text{index} < \tau\}.
$$

> **Registered divergence (4 of 5).** The manuscript does not specify the standard-deviation estimator. The implementation uses the **population** form ($\text{ddof} = 0$) where the common library default is the sample form, a deterministic choice that shifts $\tau$ slightly and is recorded as a placeholder.

**10. Alternating-Minimisation Joint Optimisation.** The joint objective couples both branches:

$$
\mathcal{L}\!\left(\beta, f^{(t)}\right) = \sum_{i=1}^{N}\ell\!\left(y_i,\ \mathrm{clip}\!\left(p_{\text{LR}}(x_i;\beta) + \alpha^*(x_i;\beta)f^{(t)}(x_i),\,0,\,1\right)\right) + \lambda_{\text{LR}}\lVert\beta\rVert_2^2 ,
$$

with $\ell(y,p) = -[y\ln p + (1-y)\ln(1-p)]$. The fixed-point scheme alternates an L-BFGS update of $\beta$, residual recomputation, a booster refit, and a validation-AUC evaluation, converging at tolerance $10^{-3}$. The **mandatory rule** is that $\alpha^*(x;\beta)$ be recomputed from the updated $p_{\text{LR}}$ at every iteration; this implementation computes it *inside the objective*, which makes a stale weight unreachable rather than merely avoided.

> **Registered divergence (5 of 5).** The scheme is **not** coordinate descent on $\mathcal{L}$, because the booster sub-step minimises squared error on residuals rather than the joint cross-entropy. The manuscript states this explicitly and the implementation preserves the statement in its conclusion rather than overclaiming convergence. Two-stage training remains the operational default, and the scheme is not run on Taiwan.

## Features

The provided iPython Notebook (`interpretable_hybrid_credit_scoring_for_thin_file_and_underbanked_populations_draft.ipynb`) implements the full research pipeline, including:

-   **Modular, 23-Task Architecture:** 170 callables decomposed into 23 tasks, each with standalone step functions plus a task orchestrator, so a defect localises to a step rather than to a monolith. Every callable carries complete type hints, NumPy-style docstrings stating purpose, parameters, processes, outputs and raises, and per-line comments citing the governing equation.
-   **Configuration-Driven Design:** All study parameters live in `study_config.json` (mirrored as `config.yaml`) as the single source of truth — **32 top-level sections**, an `equation_parameter_map` covering Equations (1)–(12) plus five unnumbered equations, and every `UNSPECIFIED IN MANUSCRIPT` value annotated with its rationale.
-   **Strict Information Barrier:** Every transformation statistic — medians, modes, winsorization bounds, monotonic bin edges, WoE tables, one-hot level sets, the selected $C$, the adaptive triple, the Platt pair — is estimated on training and frozen. The orchestrator enforces this with before-and-after SHA-256 digests of every frozen artifact, partition-access declarations checked *before* each stage is dispatched, and an optional runtime cross-check that recomputes the frozen statistics from training rows alone.
-   **Hybrid-AUC Early Stopping:** The residual booster halts on the validation AUC of the provisional hybrid, which native boosting early stopping cannot express because its metric cannot observe $p_{\text{LR}}$. Implemented as an `xgboost.callback.TrainingCallback` subclass that predicts with an inclusive `iteration_range` each round and freezes the best iteration.
-   **Falsifiable Verification, Not Decorative:** The lightweight verifier evaluates the nine mandated intermediate quantities at the strength the specification declares, in exactly four comparison modes (exact equality, absolute tolerance, percentage-point tolerance, ordered sequence equality). Membership and threshold predicates are **deliberately unavailable**, because a check satisfied by every state the pipeline can produce carries no information. A **negative self-test** corrupts each quantity in turn and asserts the corresponding check fails; a check that cannot be made to fail does not ship.
-   **Two Verification Classes, Separated:** Structural failures — partition sizes, label domains, array alignments — halt the run. Manuscript-value failures are reproduction *findings* and are recorded without halting. Collapsing both behind a single switch would mean that disabling the reproduction comparison also disables every structural guarantee.
-   **Per-Stage Checkpointing and Telemetry:** After every stage the orchestrator serialises the frozen statistics to a deterministic, byte-stable checkpoint (sorted keys, fixed separators), records wall-clock time, resident-set size and its delta, and captures every warning the stage emitted rather than letting a convergence warning vanish to the console.
-   **Degenerate-Cell Discipline:** A module-level `UNDEFINED = "---"` sentinel is returned whenever a reference group has no predicted positives in a region — the requirement fairness implementations most often get wrong by writing `0.0` into a cell whose denominator vanished.
-   **Dual-Parameterisation Calibration Export:** Platt parameters are persisted in both the manuscript and library conventions under distinct keys, with the negation relationship stated in the artifact, so a supervisor reconstructing calibration from the reported pair cannot obtain an inverted model.
-   **Refit-Proof Inference:** The bootstrap signature accepts probability vectors rather than estimators, so the prohibition on refitting inside the resampling loop is enforced by design rather than by convention.
-   **Robustness Suite:** Alternative interpretability thresholds, alternative decision thresholds ($0.4$, $0.6$), a LightGBM residual-learner sensitivity check, and reference-group sensitivity across female, rural, primary-or-less and Rwanda.
-   **Single Top-Level Interface:** `run_research_pipeline(...)` executes the entire study and returns a consolidated artifact dictionary.

## Methodology Implemented

The core analytical steps directly implement the methodology from the paper:

1.  **Validation (Task 1):** `orchestrate_validate_ingested_parameters` verifies the 32 required configuration sections, the equation-parameter map, every pinned hyperparameter under `isinstance` type contracts and `math.isclose` float identities, and both raw frames — shape, column order (reordering rather than halting, and recording the event), label dtype and positive rate, zero missingness, country composition, and the verbatim preservation of the source typo `"Divorced/Seperated"`. Nested key access is guarded so a malformed configuration raises a descriptive `ValueError` naming the full dotted path, never a bare `KeyError`.
2.  **Cleansing (Task 2):** `orchestrate_cleanse_data` drops `uniqueid` and `year` from a copy, strips whitespace on every object column, re-validates each value against the declared exhaustive level sets, enforces integer dtype contracts, and persists a cleansing audit carrying dropped columns, the **structured** level-validation results, before-and-after SHA-256 digests and a UTC timestamp.
3.  **Partitioning (Task 3):** `orchestrate_create_partitions` applies a chained stratified split — 60/40, then 50/50 on the holdout — mapping holdout-relative indices back to original positions and asserting the three arrays cover the row range exactly once. Zindi yields **14,114 / 4,705 / 4,705**; Taiwan yields **18,000 / 6,000 / 6,000**.
4.  **Preprocessing (Task 4):** `orchestrate_fit_preprocessing_stats` computes medians for numeric features and the first sorted mode for categoricals (a determinism rule, since pandas returns modes in ascending order), and the 1st/99th training percentiles, then applies the frozen statistics as $\min(\max(x, q_{01}), q_{99})$.
5.  **WoE Encoding (Task 5):** `orchestrate_fit_woe_encoding` bins from twenty quantile bins with a progressive fallback on `qcut` failure, enforces monotonicity by merging the violating adjacent pair with the closest WoE values, enforces the five-percent frequency floor by merging the **least-frequent** violating bin toward its closest-WoE neighbour, then re-enforces monotonicity. Interval keys are asserted unique so a rounding collision cannot silently overwrite a bin.
6.  **IV Filtering and One-Hot Encoding (Task 6):** `orchestrate_iv_filter_and_ohe` computes the Information Values, retains features at $\text{IV} \ge 0.02$, and builds the boosting branch's one-hot matrix with frozen level sets, reindexing each partition against the frozen column list with zero fill so an unseen level yields the all-zero indicator vector in both drift directions.
7.  **Logistic Baseline (Task 7):** `orchestrate_fit_logistic` selects $C$ by five-fold stratified CV, fits with `lbfgs`, L2 and balanced class weights at `max_iter=1000`, and logs a convergence warning when the solver reaches the iteration cap — a diagnostic the specification does not require and which separates an under-converged coefficient vector from a converged one.
8.  **Residuals (Task 8):** `orchestrate_compute_residuals` computes $r_i = y_i - p_{\text{LR}}(x_i)$ on all three partitions, asserts containment in $[-1, 1]$, and **halts** when the training residual standard deviation falls below $10^{-3}$, which would indicate an essentially constant scorecard.
9.  **Residual Booster (Task 9):** `orchestrate_fit_residual_xgboost` fits `reg:squarederror` at depth 3, learning rate 0.03, 400 estimators, subsample and column-subsample 0.8, $\lambda = 2$, with patience-50 early stopping on the provisional hybrid validation AUC, then asserts $\mathrm{corr}(r, \hat{r}) > 0$.
10. **Standalone Comparator (Task 10):** `orchestrate_fit_standalone_xgboost` fits `binary:logistic` at depth 4 with $\text{scale\_pos\_weight} = n_0/n_1$ computed on training and recorded exactly.
11. **Adaptive Weighting (Task 11):** `orchestrate_select_adaptive_weights` enumerates all 27 triples under five-fold stratified CV, refitting the scorecard, residuals and booster within each fold, and freezes the per-observation $\alpha^*$ arrays.
12. **Hybrid and Decomposition (Task 12):** `orchestrate_build_hybrid_and_rho` forms the clipped hybrid, computes $\rho(x)$ with a zero-denominator guard returning $\rho = 1$ (the correct limit when both contributions vanish), assigns regions, and aggregates region shares, within-region positive rates and the full $\rho$ distribution.
13. **Calibration (Task 13):** `orchestrate_fit_platt` fits the unpenalised Platt model on validation, asserts the Equation (10) round trip and **strict** monotonicity with an explicit positive-slope check, applies the frozen model to test, and fits the LR+Platt ablation that isolates calibration gain from residual reshaping.
14. **Evaluation (Task 14):** `orchestrate_compute_metrics` computes AUC, $\text{Gini} = 2\,\text{AUC} - 1$ and minority-class F1 at threshold 0.5, plus Brier and ECE over ten equal-width bins with reliability-diagram data, for all five candidate models, and assembles Tables 2 and 3.
15. **Fairness Audit (Task 15):** `orchestrate_run_fairness_audit` recodes the four protected attributes with their reference groups, computes region routing rates with stratified bootstrap intervals, and computes pooled and per-region DI and EOD under the degenerate-cell rule.
16. **Inference (Task 16):** `orchestrate_run_bootstrap_delong` runs the stratified paired bootstrap and the Sun–Xu DeLong test, consolidating Table 9 and recording the deliberate absence of multiple-comparison correction.
17. **Attribution (Task 17):** `orchestrate_run_shap_analysis` restricts to $\rho(x) < 0.7$, explains the sliced residual booster only, ranks by mean absolute SHAP, and runs the 500-resample stability screen with conditional diagnostics.
18. **Thin-File Analysis (Task 18):** `orchestrate_run_thin_file` builds the frozen composite index, applies $\tau$ to test, computes subgroup discrimination and calibration with paired bootstrap intervals, and tests both pre-registered hypotheses — preserving the finding that H2 is **not** supported, with the suppression mechanism stated rather than a bare null.
19. **Joint Optimisation (Task 19):** `orchestrate_run_alternating_min` runs the fixed-point scheme with mandatory $\alpha^*$ recomputation, records the convergence trajectory and per-iteration wall-clock cost, and documents that two-stage training remains the deployment default. It is not run on Taiwan.
20. **Pipeline Orchestration (Task 20):** `run_full_study` builds the 19-entry callable registry, executes it in strict dependency order with each callable invoked exactly once, enforces partition-access declarations before dispatch and the information barrier after, checkpoints the frozen statistics, logs time, memory and warnings, runs the verifier after **every** stage, and writes the reproduction report against the declared headline values.
21. **Robustness (Task 21):** `orchestrate_run_robustness` runs the interpretability-threshold, decision-threshold, LightGBM and reference-group sensitivity analyses, reiterating the post-treatment conditioning caveat and the degenerate-cell rule.
22. **Continuity Benchmark (Task 22):** `orchestrate_taiwan_reproduction` applies the placeholder preprocessing protocol and the identical model chain to the Taiwan data, then walks the six-item fidelity checklist on any deviation.
23. **Top-Level Orchestration (Task 23):** `run_research_pipeline` composes the Tasks 1–19 chain, the robustness suite and the continuity benchmark, consolidating every artifact into a single run manifest.

> **Registered divergence — Taiwan preprocessing.** The manuscript states that Taiwan preprocessing is inherited without modification from prior work and does not restate it. Every Taiwan preprocessing decision in this implementation is therefore a documented placeholder, logged as `UNSPECIFIED IN MANUSCRIPT`, and Taiwan deviations are recorded transparently rather than resolved by adjusting the Zindi pipeline — which the protocol explicitly forbids.

## Core Components (Notebook Structure)

The notebook is structured as a logical pipeline with modular orchestrator functions for each major task. Every function carries detailed type hints, technical docstrings, granular per-line comments citing the governing equations, input validation, and comprehensive error handling. Verified structure: **45 cells — 19 markdown and 26 code.**

-   **Paper preamble (cells 0–15):** Markdown cells summarising the paper — abstract, equations, algorithms, network architectures, system IPO, reproduction steps, essential sections, skills, and the four data-structure specifications.
-   **Import block (cells 16–17):** The `# Import Essential Modules` heading and the consolidated import cell. `from __future__ import annotations` is the first executable statement of the notebook.
-   **Implementation anchor (cells 18–19):** The `# Implementation` heading and the `# Draft 1` anchor, below which every generated cell is inserted. Cell order is part of the specification.
-   **Study Inputs (cell 20):** Loads `study_config.json` and constructs `zindi_raw` and `taiwan_raw`. **This is the cell that must be replaced with real data ingestion** — see [Input Data Structure](#input-data-structure).
-   **Per-task cells (cells 21–43):** One cell per task, Tasks 1 to 23, each carrying its step functions, module-level helpers and constants, and the task orchestrator, under banner separators.
-   **Study Execution (cell 44):** Configures the notebook-level logger and invokes `run_research_pipeline`, then reports the stages executed and the reproduction pass count.

The notebook shares **one namespace**: a name defined in the Task 3 cell is visible to the Task 8 cell without any import. Seven symbols are intentionally defined in two cells each (`_frame_digest`, `_partition_frames`, `PIPELINE_ORDER`, `RHO_ML_THRESHOLD`, `REGION_FULLY`, `REGION_PARTIAL`, `REGION_ML`); all are byte-identical at the AST level, so the later binding winning is behaviour-neutral.

## Key Callable: `run_research_pipeline`

The project is designed around a single, top-level user-facing interface function:

-   **`run_research_pipeline(config, zindi_raw, taiwan_raw, output_dir, run_taiwan=False, verify_manuscript_targets=True) -> Dict[str, Any]`:** This master orchestrator runs the entire study. A single call transforms the configuration and the two raw frames into the complete artifact archive: Step 1 executes the Tasks 1–19 chain through the Task 20 registry, in strict dependency order with each callable invoked exactly once, under frozen-state snapshots and the test-partition information barrier; Step 2 runs the Task 21 robustness suite and, when `run_taiwan=True`, the Task 22 continuity benchmark on an isolated state with its own nested output directory and its own reproduction report; Step 3 consolidates every artifact and persists the run manifest. The returned dictionary carries `stages_executed`, `pipeline_order`, the `chain` block (validation report, cleansing audit, both partition manifests, preprocessing statistics, IV values, retained features, logistic branch, residual summary, residual booster, standalone comparator, adaptive block, region statistics and the Platt parameters in both conventions), `robustness`, `taiwan`, `execution_log` and `reproduction_report`.

The underlying Task 20 entry point, `run_full_study(config, zindi_raw, taiwan_raw, output_dir, run_taiwan, verify_manuscript_targets, verify_structural, verify_manuscript_values, verify_frozen_statistics)`, exposes the two independent verification classes and the runtime leakage cross-check for callers who need finer control than the top-level wrapper provides.

## Prerequisites

-   Python 3.11+.
-   Core dependencies, with the versions this implementation was verified against: `numpy` (2.3.5), `pandas` (2.3.3), `scipy` (1.16.3), `scikit-learn` (1.7.2), `xgboost` (3.4.1), `lightgbm` (4.7.0), `shap` (0.52.0).
-   `psutil` is required for the per-stage memory log on Windows, where the POSIX `resource` module is unavailable. The import is guarded and the memory probe degrades to `None` rather than raising, so its absence never aborts a study.
-   A CPU-only laptop is sufficient. The dominant cost is the Task 11 adaptive grid search — 27 triples × 5 folds, each refitting the scorecard, the residuals and the booster — which accounts for roughly 200 seconds of a ~4-minute end-to-end run at Zindi scale.
-   **The two source datasets.** The **Zindi Financial Inclusion in Africa** competition data (23,524 respondents across Kenya, Rwanda, Tanzania and Uganda, survey years 2016–2018, derived from Financial Sector Deepening Network survey instruments) and the **Taiwan Credit Default** data from the UCI Machine Learning Repository (30,000 accounts, 25 columns). **No code can manufacture these files.** The notebook currently substitutes a synthetic corpus in their absence; replacing that substitution with real ingestion is the single prerequisite for reproduction.

## Input Data Structure

The pipeline requires the configuration plus two raw frames, all validated by Task 1:

1.  **`config: Dict[str, Any]`** — Parsed from `study_config.json`. **32 top-level sections**, from `study_metadata` through `equation_parameter_map` and `code_and_data_availability`. Task 1 halts on the first failure.

2.  **`zindi_raw: pd.DataFrame`** — Shape **(23,524, 13)** in the declared column order: `uniqueid`, `country`, `year`, `location_type`, `cellphone_access`, `gender_of_respondent`, `relationship_with_head`, `marital_status`, `education_level`, `job_type`, `household_size`, `age_of_respondent`, `bank_account`. The label is `bank_account` (`int64`; **1 = owns a formal bank account**) at a positive rate of **0.1408**. Country composition must be Rwanda 8,735 / Tanzania 6,620 / Kenya 6,068 / Uganda 2,101. `uniqueid` and `year` are dropped in Task 2 as non-predictive — `year` is perfectly collinear with `country`. Zero missing values are permitted. The source typo `"Divorced/Seperated"` in `marital_status` **must be preserved verbatim**; correcting it fails validation.

    > **Label semantics.** The label is account *ownership*, not default. The manuscript reframes it deliberately as an eligibility-gate proxy: account ownership and loan default sit on opposite sides of the lender's selection funnel. This reframing is the paper's argument, and consumers of the output must not read `bank_account = 1` as a default event.

3.  **`taiwan_raw: pd.DataFrame`** — Shape **(30,000, 25)**, label `default.payment.next.month` at a positive rate of approximately **0.22**. `PAY_0` is present and `PAY_1` is absent, matching the source; `SEX` ⊆ {1,2}, `EDUCATION` ⊆ {0,…,6}, `MARRIAGE` ⊆ {0,1,2,3}. The `ID` column is dropped in Task 2, leaving 24 columns.

> **The synthetic-corpus substitution, stated plainly.** In the shipped notebook the *Study Inputs* cell does not read either file. It constructs both frames from `np.random.default_rng(71)`, reproducing the published row counts, column order and class balance exactly so that every downstream contract is satisfied. The Zindi label is then assigned as the top 14.08% of a deterministic logistic score in six features — `age_of_respondent`, `household_size`, `education_level`, `cellphone_access`, `job_type`, `country` — which makes the label a noiseless threshold on those features. The consequence is that the Information Value filter retains exactly those six and discards the other four as noise, and every discrimination metric is inflated far above the published figures. **Replace the cell with a real ingestion path before drawing any conclusion.** A provenance gate that fails closed when the raw frames are generator-derived is the recommended first extension.

## **Usage Example**

Here is a granular, step-by-step guide to executing the end-to-end pipeline for **"Interpretable hybrid credit scoring for thin-file and underbanked populations"**. This example demonstrates how to (i) load `study_config.json` from the working directory into a Python dictionary, (ii) construct the two raw study frames at the published composition, and (iii) execute the full study via the `run_research_pipeline(...) → Dict[str, Any]` interface.

> **Environment assumption (mandatory).** All pipeline callables from Tasks 1–23 are assumed to be defined **in a single Jupyter notebook** cell-context. There is **no `.py` module** to import, so this example imports only the libraries needed for configuration loading, frame construction and result inspection. `study_config.json` is assumed to be saved in the working directory under that exact name.

> **Verification status.** The execution in Step 3 was run end to end against the real orchestrator before being written into this document. All 19 registry stages completed, 19 per-stage checkpoints were written, and the reported figures below are the actual outputs of that run.

### **Step 1: Loading the Configuration (`study_config.json`)**

`study_config.json` is the single frozen source of truth. It is read into memory and converted into the nested `config` dictionary that every task orchestrator consumes.

```python
# ==============================================================================
# Step 1 - Load study_config.json from the Working Directory
# ==============================================================================

# Standard library and third-party imports required by this example only;
# every pipeline callable is already defined in the notebook context.
import json
from pathlib import Path
from typing import Any, Dict, List, Tuple

import numpy as np
import pandas as pd


def load_study_configuration(
    filepath: str = "study_config.json",
) -> Dict[str, Any]:
    """
    Load the frozen study configuration from the working directory.

    Purpose:
        Ingests the thirty-two configuration sections consumed by every task
        orchestrator - study metadata and headline reproduction targets,
        dataset parameters, raw data schemas, partitioning, preprocessing,
        WoE and Information Value settings, the four model configurations,
        adaptive weighting, hybrid prediction, interpretability thresholds,
        calibration, evaluation metrics, protected attributes, the fairness
        audit, thin-file parameters, joint optimisation, bootstrap and
        DeLong inference, SHAP settings, the training protocol, plotting
        defaults and the equation-parameter map - and converts them into
        the nested dictionary the pipeline consumes.

    Inputs:
        filepath (str): Path to the JSON file, default 'study_config.json',
            assumed to be in the working directory.

    Processes:
        1. Validate the argument type and the file's existence.
        2. Parse the document with an explicit UTF-8 decode, because the
           configuration carries mathematical symbols that the Windows
           console default codepage cannot represent.
        3. Verify that the required top-level sections are present.
        4. Report the section inventory to the operator.

    Outputs:
        Dict[str, Any]: The complete nested configuration dictionary.

    Raises:
        TypeError: If filepath is not a string.
        FileNotFoundError: If the configuration file is absent.
        json.JSONDecodeError: If the document is malformed.
        ValueError: If a required top-level section is missing.
    """
    # Reject a non-string path before touching the filesystem
    if not isinstance(filepath, str):
        raise TypeError(f"filepath must be a string; got {type(filepath)}.")
    # Resolve the path relative to the working directory
    config_path: Path = Path(filepath)
    # A missing configuration would silently break every downstream gate
    if not config_path.is_file():
        raise FileNotFoundError(
            f"Configuration file '{filepath}' not found in the working "
            f"directory {Path.cwd()}."
        )
    # Open with an explicit encoding; the default codepage would fail here
    with open(config_path, "r", encoding="utf-8") as handle:
        # Parse the document into the nested configuration dictionary
        config: Dict[str, Any] = json.load(handle)
    # The pipeline reads these sections unconditionally
    required_sections: Tuple[str, ...] = (
        "study_metadata", "dataset_parameters", "raw_data_schemas",
        "partitioning_parameters", "preprocessing_parameters",
        "woe_parameters", "iv_parameters",
        "logistic_regression_parameters", "residual_xgboost_parameters",
        "standalone_xgboost_parameters", "adaptive_weighting_parameters",
        "hybrid_prediction_parameters", "interpretability_parameters",
        "calibration_parameters", "evaluation_metrics_parameters",
        "protected_attribute_parameters", "fairness_audit_parameters",
        "thin_file_parameters", "joint_optimization_parameters",
        "bootstrap_inference_parameters", "delong_test_parameters",
        "shap_explanation_parameters", "equation_parameter_map",
    )
    # Fail loudly on a truncated configuration rather than at first use
    absent: List[str] = [s for s in required_sections if s not in config]
    if absent:
        raise ValueError(
            f"study_config.json is missing required section(s): {absent}."
        )
    # Report the inventory so the operator can see what was loaded
    print(f"Loaded configuration from {filepath}: "
          f"{len(config)} top-level sections.")
    return config


# Load the frozen study configuration from the working directory
config: Dict[str, Any] = load_study_configuration("study_config.json")

# Confirm the gates that depend most directly on the configuration
print(f"  zindi n              : "
      f"{config['dataset_parameters']['zindi_fsd']['n_observations']}")
print(f"  zindi positive rate  : "
      f"{config['dataset_parameters']['zindi_fsd']['positive_class_rate']}")
print(f"  partition seed       : "
      f"{config['partitioning_parameters']['random_seed']}")
print(f"  expected train n     : "
      f"{config['partitioning_parameters']['zindi_train_n']}")
print(f"  headline targets     : "
      f"{list(config['study_metadata']['headline_results'])}")
```

**Discussion of Step 1.** The explicit UTF-8 decode is not stylistic. The configuration carries mathematical symbols that the Windows console default codepage cannot represent, and an unguarded read fails with a `UnicodeDecodeError` on the encoding rather than on any real defect. The loader fails loudly on a missing file or a truncated section rather than proceeding with an empty dictionary, consistent with the pipeline's fail-fast gates.

### **Step 2: Constructing the Two Raw Study Frames**

Task 1 validates both frames against `config["dataset_parameters"]` and **halts** on mismatch, so the frames must hit the published composition *by construction* rather than by sampling luck.

#### The binding constraint that dictates the construction

`validate_zindi_raw` enforces, among other checks:

$$
n_{\text{observed}} = 23{,}524 \quad \text{(exact equality)},
$$
$$
\left|\hat{\pi}_{\text{positive}} - 0.1408\right| \le 10^{-4}, \qquad \hat{\pi}_{\text{positive}} = \frac{1}{n}\sum_{i=1}^{n}\mathbf{1}\{y_i = 1\},
$$
$$
\text{country counts} = \{\text{Rwanda}: 8735,\ \text{Tanzania}: 6620,\ \text{Kenya}: 6068,\ \text{Uganda}: 2101\}.
$$

A tolerance of $10^{-4}$ on the positive rate at $n = 23{,}524$ admits a drift of roughly two observations. Sampling labels as `Bernoulli(0.1408)` would fail that band with overwhelming probability; the construction therefore allocates an **exact count** and shuffles.

```python
# ==============================================================================
# Step 2 - Construct the Two Raw Study Frames at the Published Composition
# ==============================================================================

# NOTE: this constructs a SCHEMA-CONFORMANT corpus, not the real data. It
# satisfies every Task 1 contract so the pipeline executes end to end, but the
# metrics it produces are NOT a reproduction of the paper. Replace this step
# with real ingestion of the Zindi and Taiwan files before drawing conclusions.

def build_zindi_frame(config: Dict[str, Any], seed: int = 71) -> pd.DataFrame:
    """
    Construct the raw Zindi frame at the published composition.

    Purpose:
        Produces a (23524, 13) frame satisfying every Task 1 contract: the
        exact row count, the declared column order, the published country
        composition, the 0.1408 positive rate, integer dtypes on the three
        integer columns, object dtypes on the eight categorical columns and
        zero missing values.

    Inputs:
        config (Dict[str, Any]): The loaded study configuration, read for
            the declared categorical level sets and column order.
        seed (int): The generator seed, fixing the construction.

    Processes:
        1. Allocate the country vector at the exact published counts.
        2. Draw the demographic and infrastructure covariates.
        3. Assign the label by EXACT count so the rate gate is deterministic.
        4. Reorder the columns to the declared raw column order.

    Outputs:
        pd.DataFrame: The raw Zindi frame, shape (23524, 13).

    Raises:
        KeyError: If the configuration lacks the Zindi dataset block.
    """
    # The published composition is the construction target
    zindi_block: Dict[str, Any] = config["dataset_parameters"]["zindi_fsd"]
    # A dedicated generator keeps the construction reproducible
    rng: np.random.Generator = np.random.default_rng(seed)
    # The declared row count
    n: int = int(zindi_block["n_observations"])
    # The declared categorical level sets
    levels: Dict[str, List[str]] = zindi_block["categorical_levels"]
    # Allocate the country vector at the EXACT published counts
    country: np.ndarray = np.concatenate(
        [[name] * count
         for name, count in zindi_block["country_composition"].items()]
    )
    # Age is a bounded integer covariate
    age: np.ndarray = np.round(
        rng.normal(40, 14, n).clip(16, 85)).astype(np.int64)
    # Household size is a small positive integer
    household_size: np.ndarray = rng.integers(1, 12, n).astype(np.int64)
    # Education, cellphone access and job type are categorical
    education: np.ndarray = rng.choice(
        ["No formal education", "Primary education",
         "Secondary education", "Tertiary education"],
        n, p=[0.35, 0.30, 0.25, 0.10])
    cellphone: np.ndarray = rng.choice(["Yes", "No"], n, p=[0.7, 0.3])
    job_type: np.ndarray = rng.choice(
        ["Government", "Formally employed Private sector",
         "Informally employed", "Self employed",
         "Farming and Fishing", "Remittance Dependent"], n)
    # EXACT positive count so the 1e-4 rate gate passes deterministically
    n_positive: int = int(round(
        float(zindi_block["positive_class_rate"]) * n))
    # Allocate the label vector, then shuffle so positives are not contiguous
    bank_account: np.ndarray = np.zeros(n, dtype=np.int64)
    bank_account[:n_positive] = 1
    rng.shuffle(bank_account)
    # Assemble the frame; year is collinear with country by construction
    frame: pd.DataFrame = pd.DataFrame({
        "uniqueid": [f"uniqueid_{i:05d}" for i in range(n)],
        "country": country,
        "year": np.where(country == "Kenya", 2018,
                         np.where(country == "Rwanda", 2016, 2017)),
        "location_type": rng.choice(levels["location_type"], n),
        "cellphone_access": cellphone,
        "gender_of_respondent": rng.choice(
            levels["gender_of_respondent"], n),
        "relationship_with_head": rng.choice(
            levels["relationship_with_head"], n),
        # The source typo 'Divorced/Seperated' MUST survive verbatim
        "marital_status": rng.choice(levels["marital_status"], n),
        "education_level": education,
        "job_type": job_type,
        "household_size": household_size,
        "age_of_respondent": age,
        "bank_account": bank_account,
    })
    # Reorder to the declared raw column order; Task 1 checks this
    return frame[zindi_block["raw_column_order"]]


def build_taiwan_frame(config: Dict[str, Any], seed: int = 71) -> pd.DataFrame:
    """
    Construct the raw Taiwan continuity frame at the published composition.

    Purpose:
        Produces a (30000, 25) frame satisfying the Task 1 Taiwan contract:
        the declared 25-column order, the label at a positive rate of
        approximately 0.22, PAY_0 present and PAY_1 absent, and the declared
        domains on SEX, EDUCATION and MARRIAGE.

    Inputs:
        config (Dict[str, Any]): The loaded configuration, read for the
            declared Taiwan schema column list.
        seed (int): The generator seed, fixing the construction.

    Processes:
        1. Draw the limit, demographic, repayment and amount columns.
        2. Assign the label by EXACT count at the published rate.
        3. Reorder to the declared schema column order.

    Outputs:
        pd.DataFrame: The raw Taiwan frame, shape (30000, 25).

    Raises:
        KeyError: If the configuration lacks the Taiwan schema block.
    """
    # The declared schema fixes the column order Task 1 validates
    schema_columns: List[str] = [
        column["name"] for column in
        config["raw_data_schemas"]["taiwan_credit_schema"]["columns"]
    ]
    # A dedicated generator keeps the construction reproducible
    rng: np.random.Generator = np.random.default_rng(seed)
    # The declared row count for the continuity benchmark
    n: int = 30000
    # Assemble the identifier, credit limit and demographics
    frame: pd.DataFrame = pd.DataFrame({
        "ID": np.arange(1, n + 1),
        "LIMIT_BAL": rng.integers(10000, 1000000, n),
        "SEX": rng.choice([1, 2], n),
        "EDUCATION": rng.choice([0, 1, 2, 3, 4, 5, 6], n),
        "MARRIAGE": rng.choice([0, 1, 2, 3], n),
        "AGE": rng.integers(21, 79, n),
    })
    # PAY_0 is present and PAY_1 is absent, matching the source
    for index in [0, 2, 3, 4, 5, 6]:
        frame[f"PAY_{index}"] = rng.integers(-2, 9, n)
    # The six billing and six payment amount columns
    for index in range(1, 7):
        frame[f"BILL_AMT{index}"] = rng.integers(0, 100000, n)
        frame[f"PAY_AMT{index}"] = rng.integers(0, 50000, n)
    # EXACT positive count at the published default rate
    default_flag: np.ndarray = np.zeros(n, dtype=np.int64)
    default_flag[rng.permutation(n)[:6600]] = 1
    frame["default.payment.next.month"] = default_flag
    # Reorder to the declared schema column order
    return frame[schema_columns]


# Construct both frames once for the pipeline run
zindi_raw: pd.DataFrame = build_zindi_frame(config)
taiwan_raw: pd.DataFrame = build_taiwan_frame(config)

# Report each frame against its published composition
print(f"zindi_raw  : {zindi_raw.shape}  "
      f"positive_rate={zindi_raw['bank_account'].mean():.4f}  "
      f"missing={int(zindi_raw.isnull().sum().sum())}")
print(f"taiwan_raw : {taiwan_raw.shape}  "
      f"positive_rate="
      f"{taiwan_raw['default.payment.next.month'].mean():.4f}")
print(f"countries  : "
      f"{ {k: int(v) for k, v in zindi_raw['country'].value_counts().items()} }")
```

**Discussion of Step 3.** The orchestrator call is deliberately thin because every intermediate callable runs internally. Four behaviours are worth stating explicitly:

-   **The verifier runs after every stage, not once at the end.** A structural failure therefore halts at the stage that produced it and writes a diagnostic naming that stage, rather than surfacing after the full run with no attribution. A manuscript-value discrepancy is recorded and the run continues, because a reproduction discrepancy is a finding rather than a fault.
-   **The information barrier is enforced in both directions.** Partition-access declarations are checked *before* each stage is dispatched — a fitting-sequence stage that declares test access is rejected — and the test indices, labels and row order are digested and compared *after*. Setting `verify_frozen_statistics=True` on the underlying `run_full_study` additionally recomputes the frozen medians and winsorization bounds from training rows alone, which is the only check that directly detects estimation over held-out rows.
-   **The split asymmetry is not incidental.** $C$ is selected by cross-validation within training; the residual booster early-stops on validation; Platt is fitted on validation; AUC, Brier, ECE, the fairness audit and all inference are computed on test. No selection decision touches test.
-   **Region assignments are computed on uncalibrated probabilities.** Calibration shifts the probability scale and therefore F1 and Brier, but it must not shift $\rho(x)$ or region membership, and the implementation preserves that separation.

> **Realism note.** At Zindi scale the end-to-end run takes roughly four minutes on a CPU-only machine, dominated by the Task 11 adaptive grid search (27 triples × 5 folds ≈ 200 s) and the Task 16 bootstrap (2,000 resamples ≈ 19 s). Setting `run_taiwan=True` adds the continuity benchmark. Separately: the frames constructed in Step 2 reproduce the published *composition* but carry no real dependence structure, so they exercise every code path without reproducing the manuscript's AUC values. Reproducing those figures requires the genuine Zindi and Taiwan files, which no generator can manufacture.

### **Summary of the Execution Flow**

1.  **Configuration Loading.** `load_study_configuration("study_config.json")` parses the thirty-two sections under an explicit UTF-8 decode and validates that the pipeline-critical sections are present, failing loudly otherwise.
2.  **Frame Construction.** Both frames are built at the published composition by **exact count**, so the Task 1 rate gates pass deterministically rather than by sampling luck, with the declared column order, dtypes and level sets satisfied.
3.  **Validation and Cleansing (Tasks 1–2).** The configuration, both frames, the country composition, the label rates and every categorical level set are validated; `uniqueid` and `year` are dropped; whitespace is stripped; dtype contracts are enforced; and a cleansing audit carrying the structured level-validation results and before/after digests is persisted before any fitting.
4.  **Partitioning and Preprocessing (Tasks 3–4).** The chained stratified 60/40 then 50/50 split yields 14,114 / 4,705 / 4,705 under seed 20260830, with coverage asserted exactly; medians, modes and 1st/99th percentiles are frozen on training and applied to all partitions.
5.  **Encoding (Tasks 5–6).** Monotonic binning from twenty quantile bins with least-frequent-bin merging and a re-enforced monotonicity pass; WoE by Equation (2) with the neutral-zero rule; the $\text{IV} \ge 0.02$ filter; and the one-hot boosting matrix with frozen level sets and the all-zero indicator rule.
6.  **Model Fitting (Tasks 7–11).** $C$ selection by five-fold CV; the residual booster with hybrid-AUC early stopping; the standalone comparator at $\text{scale\_pos\_weight} = n_0/n_1$; and the 27-triple adaptive grid search.
7.  **Decomposition, Calibration and Evaluation (Tasks 12–14).** The clipped hybrid, $\rho(x)$ with its zero-denominator guard, the three-region partition, Platt scaling with the Equation (10) round-trip assertion, and the five-model metric tables.
8.  **Audit and Inference (Tasks 15–18).** The per-region fairness audit with degenerate-cell handling; the stratified paired bootstrap and DeLong test; TreeSHAP on the sliced residual booster with the stability screen; and the thin-file segmentation with both pre-registered hypotheses.
9.  **Joint Optimisation and Orchestration (Tasks 19–23).** The alternating fixed-point scheme with mandatory $\alpha^*$ recomputation; the registry execution with checkpointing, telemetry and per-stage verification; the robustness suite; the optional continuity benchmark; and the consolidated run manifest.

## Customization

The pipeline is highly customizable via the `study_config.json` file (or its `config.yaml` rendering). Users can modify:

-   **Partitioning and preprocessing:** `partitioning_parameters` (fractions and `random_seed`); `preprocessing_parameters` (imputation rules, winsorization percentiles, and the monotonic-binning placeholder — starting bin count, monotonicity rule, and the five-percent frequency floor).
-   **Scorecard encoding:** `woe_parameters` (the pinned class convention $g_k = y{=}0$, the neutral-zero rare-category rule) and `iv_parameters` (the $0.02$ retention threshold).
-   **Models:** `logistic_regression_parameters` (the $C$ grid, solver, penalty, class weighting, `max_iter`); `residual_xgboost_parameters` (objective, depth, $\nu$, subsampling, $\lambda$, early-stopping patience); `standalone_xgboost_parameters`; `lightgbm_sensitivity_parameters`.
-   **Adaptive weighting and decomposition:** `adaptive_weighting_parameters` (the three grids and the CV fold count); `interpretability_parameters` (the $0.9$ and $0.7$ region thresholds and the $\rho$ formula record).
-   **Calibration and evaluation:** `calibration_parameters` (the ECE bin count $K = 10$); `evaluation_metrics_parameters` (the decision threshold and the metric set).
-   **Fairness:** `protected_attribute_parameters` (the four attributes, their category mappings and reference groups) and `fairness_audit_parameters` (the four-fifths band, the degenerate-cell sentinel, and the routing-rate bootstrap count).
-   **Inference:** `bootstrap_inference_parameters` ($B = 2000$ for AUC differences, $500$ for fairness quantities) and `delong_test_parameters`.
-   **Interpretability and segmentation:** `shap_explanation_parameters` (the stability resample count and the top-$k$ set size) and `thin_file_parameters` (the ordinal education mapping, the formal-job mapping and the 25th-percentile threshold rule).
-   **Verification:** the two independent flags exposed on `run_full_study` — `verify_structural` (defaults true and should stay true) and `verify_manuscript_values` — plus `verify_frozen_statistics` for the runtime leakage cross-check.

> **The registered divergences are exposed, not hidden.** The 3-fold/5-fold resolution is logged verbatim in the audit trail; the Platt parameterisation is persisted under both conventions with the negation relationship stated; the Taiwan placeholder protocol is flagged `UNSPECIFIED IN MANUSCRIPT` on every decision; the probability-scale coupling of $\rho(x)$ is documented at its point of use; and the $\text{ddof} = 0$ estimator in the thin-file index is recorded as a placeholder.

Every parameter marked `UNSPECIFIED IN MANUSCRIPT` carries its rationale in the configuration and in the run manifest.

## Contributing

Contributions are welcome. Please fork the repository, create a feature branch, and submit a pull request with a clear description of your changes. Adherence to PEP 8, complete type hinting, technical docstrings, input validation, and error handling is required. Contributors must additionally:

-   **Never use the test partition for model selection.** $C$ selection, early stopping, the adaptive triple and calibration are all validation or training decisions. AUC, Brier, ECE, the fairness audit and every inference statistic are test quantities.
-   **Never fabricate provider data silently.** The synthetic corpus in the *Study Inputs* cell is a stopgap and is documented as one; any replacement must fail loudly on a missing file rather than substituting.
-   **Make every new assertion falsifiable.** Any check added to the verifier must be accompanied by a corruption case in the negative self-test proving it can fail. A predicate satisfied by every state the pipeline can produce must not ship.
-   **Preserve the information barrier.** Every transformation statistic is estimated on training only and frozen. New stages must declare the partitions they read for estimation in `ADAPTER_FITTING_PARTITIONS`.
-   **Report degenerate cells as undefined.** Never write `0.0` into a fairness cell whose denominator vanished.
-   **Document any methodological deviation** at its point of departure, as the five existing registered divergences are.

## Recommended Extensions

-   **Real-data provenance gate:** A check that fails closed when the raw frames are generator-derived, so the synthetic corpus can never silently reach a results table. This is the highest-value single addition to the repository.
-   **Logit-scale interpretability ratio:** The manuscript flags the probability-scale coupling and leaves a logit-scale variant as future work. Implementing both and reporting the region partition under each would separate the algebraic artefact from the empirical finding.
-   **True functional-gradient joint optimisation:** The current scheme is not coordinate descent on the joint loss, because the booster sub-step minimises squared error. Flowing gradients through both branches is named as future work by the paper.
-   **Multi-group and intersectional fairness:** The audit is currently pairwise against a reference group; extending to intersectional cells (for example rural × primary-or-less) is where routing disparities in this population are most likely to compound.
-   **Cost-sensitive threshold optimisation:** F1 at $0.5$ is a reporting convention, not a lending decision. Expected-cost thresholding with asymmetric misclassification costs would make the operational trade-off explicit.
-   **Adverse-action note generation:** For the fully-interpretable region the scorecard coefficients *are* the explanation; emitting regulator-ready reason codes directly from the WoE-scaled contributions would close the loop from audit to compliance artifact.
-   **Region-aware routing constraints:** The audit detects routing disparity but does not act on it. A constrained variant that caps the ML-driven share within any protected subgroup would convert a diagnostic into a control.

# Paper

Title: "*findr: Transparent and Fair Credit Risk Decisions through Semi-Structured Regressions*"

Authors: Victor Medina-Olivares, Stefan Lessmann, Jonathan Crook

E-Journal Submission Date: 25 August 2026

Link: https://arxiv.org/abs/2608.24582

Abstract:

Credit risk models increasingly need to combine predictive accuracy with transparent explanations and auditable fairness constraints. Logistic regression remains attractive because its coefficients are easy to interpret, but it can miss nonlinear structure. Flexible models can improve prediction, but their explanations are often post-hoc and may not describe the decision rule itself. We introduce findr, short for flexible, interpretable deep regression, a semi-structured framework for binary credit risk modelling that decomposes the logit into an interpretable structured component and an orthogonal neural residual. The orthogonalisation separates coefficient-based effects from residual nonlinear variation, while an in-processing Wasserstein penalty mitigates group disparities by comparing score distributions during training. The framework also includes diagnostics that measure the structured component's contribution to logit variation, decision agreement, and local directional consistency. We evaluate findr in a simulation study and on eight public credit datasets using score-level accuracy-fairness frontiers. The results show that findr behaves close to logistic regression when the signal is approximately linear, while recovering much of the predictive gain of neural models when nonlinear structure is relevant. The diagnostics identify when coefficient-based explanations remain close to the full fitted model and when residual variation must also be examined. These findings support semi-structured modelling as a practical way to make performance, fairness, and interpretability trade-offs explicit in credit risk decisions.


# Paper Summary

## Granular Step-by-Step Summary of the Paper

**Paper:** *`findr`: Transparent and Fair Credit Risk Decisions Through Semi-Structured Regressions*  
**Authors:** Victor Medina-Olivares, Stefan Lessmann, Jonathan Crook  
**Type:** Preprint, University of Edinburgh / Humboldt-Universität zu Berlin  
**Target audience:** Credit-risk modelers, financial regulators, fairness/interpretability researchers.

### Step 1: Problem Context and Motivation

The paper addresses a central tension in credit-risk modelling:

1. **Logistic regression (LR)** is the industry workhorse because it is probabilistic, auditable, and easy to communicate.
2. LR is restricted to a **linear log-odds** form, so it underfits real-world nonlinearities and feature interactions.
3. **Generalized additive models (GAMs)** relax linearity but do not naturally capture higher-order interactions.
4. **Flexible machine-learning models** (e.g., deep neural networks) are highly predictive but opaque; post-hoc explanations such as SHAP/LIME are not the decision rule itself.
5. **Regulatory pressure** (EU AI Act, EBA guidelines, CFPB Reg B) demands explainability, auditability, and monitoring of disparate impacts.
6. **Algorithmic fairness** criteria are grouped into *independence* (demographic parity), *separation* (error-rate parity), and *sufficiency* (calibration). These criteria generally cannot all hold simultaneously when group base rates differ and predictions are imperfect.
7. The paper focuses on **group-level fairness** because it is portfolio-auditable, and chooses **in-processing fairness** because post-processing at one threshold does not control the entire score distribution.

The proposed solution is `findr`, a **semi-structured regression** that combines:

- an interpretable linear logit,
- an orthogonal neural residual,
- an in-processing Wasserstein fairness penalty,
- interpretability diagnostics,
- score-level accuracy–fairness frontiers for benchmarking.

### Step 2: Review of Related Work and Identification of the Gap

The literature is organized into two streams:

#### A. Interpretable Credit-Risk Models
- LR, GAMs, GA2Ms, Neural Additive Models, GAMI-Net, NODE-GAM all make interpretability intrinsic but restrict interaction structure.
- **Semi-structured regression** (Rügamer et al., 2024) combines a structured linear predictor with a neural component.
- Existing semi-structured work has been used for time-to-default and multi-state delinquency models but not for fairness.

#### B. Fairness-Aware Machine Learning
- Pre-processing, in-processing, post-processing methods exist.
- In-processing methods include regularisation, constrained optimisation, adversarial training.
- **Wasserstein fair classification** (Jiang et al., 2020) penalises distributional differences between protected groups.
- The paper distinguishes its use of Wasserstein from **distributionally robust fairness**: it compares group score distributions *within the fitted model*, not perturbations of the data distribution.

**Gap:** No prior framework simultaneously provides:

1. an identifiable structured coefficient component,
2. a flexible neural residual,
3. full-score-distribution fairness control,
4. diagnostics showing whether the coefficient component is actually sufficient to explain decisions.

This is the gap `findr` fills.

### Step 3: The Semi-Structured Logistic Model

The model decomposes the total log-odds into two parts:

\[
\eta_i = \eta_i^{\text{str}} + \eta_i^{\text{unstr}},
\]

with

\[
\eta_i^{\text{str}} = \beta_0 + \mathbf{x}_i^\top \boldsymbol{\beta},
\]

and

\[
\eta_i^{\text{unstr}} = d_{\boldsymbol{\theta}}(\mathbf{z}_i).
\]

The default probability is:

\[
p(y_i = 1 \mid \eta_i) = \sigma(\eta_i), \quad \sigma(t) = \frac{1}{1 + e^{-t}}.
\]

Here:

- \(\mathbf{x}_i\): structured covariates.
- \(\mathbf{z}_i\): neural network inputs (may overlap with \(\mathbf{x}_i\) or be separate, e.g., text).
- \(\boldsymbol{\beta}\): interpretable coefficients.
- \(d_{\boldsymbol{\theta}}\): neural residual map.

### Step 4: Identifiability via Orthogonalisation

The key technical issue is **attribution ambiguity**: a neural network can represent linear effects, making the structured coefficients arbitrary. The paper solves this by projecting the neural representation onto the orthogonal complement of the structured design.

#### 4.1 Design Matrix
Define:

\[
\tilde{\mathbf{X}} = [\mathbf{1}_n \quad \mathbf{X}],
\]

where \(\mathbf{X} = (\mathbf{x}_1, \dots, \mathbf{x}_n)^\top\). The structured logit vector is:

\[
\boldsymbol{\eta}^{\text{str}} = \tilde{\mathbf{X}} \tilde{\boldsymbol{\beta}}.
\]

#### 4.2 QR Decomposition and Projection
Take the thin QR decomposition:

\[
\tilde{\mathbf{X}} = \mathbf{Q}\mathbf{R}, \quad \mathbf{Q}^\top\mathbf{Q} = \mathbf{I}_{p+1}.
\]

The orthogonal projector onto the column space of \(\tilde{\mathbf{X}}\) is:

\[
\mathbf{P} = \mathbf{Q}\mathbf{Q}^\top,
\]

and its complement is:

\[
\mathbf{P}^\perp = \mathbf{I}_n - \mathbf{P}.
\]

#### 4.3 Orthogonalised Encoder Outputs
Let the neural encoder outputs be:

\[
\mathbf{H}_{\boldsymbol{\alpha}} = (h_{\boldsymbol{\alpha}}(\mathbf{z}_1), \dots, h_{\boldsymbol{\alpha}}(\mathbf{z}_n))^\top.
\]

The orthogonalised outputs are:

\[
\mathbf{H}_{\boldsymbol{\alpha}}^\perp = \mathbf{P}^\perp \mathbf{H}_{\boldsymbol{\alpha}}.
\]

#### 4.4 Residual Logit
A linear residual layer \(r_{\boldsymbol{\gamma}}\) maps these orthogonalised features to the logit scale:

\[
\boldsymbol{\eta}^{\text{unstr}} = \mathbf{H}_{\boldsymbol{\alpha}}^\perp \boldsymbol{\gamma} + \gamma_0 \mathbf{1}_n.
\]

Because:

\[
\tilde{\mathbf{X}}^\top \mathbf{H}_{\boldsymbol{\alpha}}^\perp = \mathbf{0},
\]

the residual contains **no linear combination of the structured covariates** and no intercept effect (except \(\gamma_0\), which is absorbed into \(\beta_0\)). This guarantees the linear coefficients are uniquely attributable to the structured component.

### Step 5: In-Processing Fairness via Wasserstein Distances

The paper chooses Wasserstein distance because it compares **entire score distributions**, not just means or thresholded error rates.

#### 5.1 Empirical \(p\)-Wasserstein Distance
For two univariate empirical distributions \(\hat{\mu}\) and \(\hat{\nu}\), the \(p\)-Wasserstein distance is computed exactly by sorting scores and using a quantile grid:

\[
W_p^p(\hat{\mu}, \hat{\nu}) = \sum_{k=1}^K (q_k - q_{k-1}) \left| F_{\hat{\mu}}^{-1}(q_{k-1}) - F_{\hat{\nu}}^{-1}(q_{k-1}) \right|^p.
\]

This is differentiable almost everywhere, so gradients can be backpropagated through the fairness term.

#### 5.2 Fairness Criteria as Penalties
Let \(\hat{Y}\) be the predicted probability score. The paper supports three group-fairness criteria:

- **Equalised Odds (EO):**
  \[
  \sum_{y \in \{0,1\}} W\left(\hat{Y} \mid Y=y, A=0, \; \hat{Y} \mid Y=y, A=1\right).
  \]
- **Equal Opportunity (EOpp):**
  \[
  W\left(\hat{Y} \mid Y=0, A=0, \; \hat{Y} \mid Y=0, A=1\right).
  \]
- **Independence (Demographic Parity):**
  \[
  W\left(\hat{Y} \mid A=0, \; \hat{Y} \mid A=1\right).
  \]

The empirical analysis focuses on **Equalised Odds**, since it conditions on both outcome classes and is the most demanding.

### Step 6: Training Objective

The total loss is a weighted combination of predictive loss and fairness penalty:

\[
\mathcal{L}_{\text{total}} = \underbrace{\frac{1}{n} \sum_{i=1}^n \left[-y_i \log \hat{Y}_i - (1-y_i)\log(1-\hat{Y}_i)\right]}_{\mathcal{L}_{\text{pred}}}
+ \lambda \cdot \mathcal{L}_{\text{fair}}(\hat{\mathbf{Y}}, \mathbf{A}, \mathbf{Y}).
\]

Gradients are backpropagated through:

- the logistic loss,
- the orthogonalisation step,
- the Wasserstein penalty.

This means the fairness constraint acts on the **full probability score**, not just a decision threshold.

### Step 7: Score-Level Accuracy–Fairness Frontier

The paper introduces an empirical benchmark that compares fitted models on a **score-level accuracy–fairness plane**.

#### 7.1 Fairness Functional for EO
For a candidate score vector \(\mathbf{s}\), the equalised-odds unfairness is:

\[
\hat{\Phi}(\mathbf{s}; \mathbf{y}, \mathbf{a}) = \sum_{y \in \{0,1\}} W\left(\hat{\mu}_{y0}^{\mathbf{s}}, \hat{\mu}_{y1}^{\mathbf{s}}\right),
\]

where \(\hat{\mu}_{ya}^{\mathbf{s}}\) is the empirical distribution of scores among observations with \(Y=y, A=a\).

#### 7.2 Expected Log Loss Under a Reference Risk
The paper defines the reference risk \(\mathbf{r}\) as the true conditional probability \(P(Y=1 \mid \mathbf{X})\) in simulations, or the score from an unconstrained model in real data.

The expected Bernoulli log loss of \(\mathbf{s}\) under \(\mathbf{r}\) is:

\[
\hat{R}_{\log}(\mathbf{r}, \mathbf{s}) = -\frac{1}{n} \sum_{i=1}^n \left[r_i \log s_i + (1-r_i)\log(1-s_i)\right].
\]

This equals:

\[
\hat{R}_{\log}(\mathbf{r}, \mathbf{s}) = \hat{H}(\mathbf{r}) + \frac{1}{n} \sum_{i=1}^n \mathrm{KL}\{\mathrm{Bern}(r_i) \parallel \mathrm{Bern}(s_i)\},
\]

so the **excess log loss** is exactly the average KL divergence from the reference risk to the candidate score.

#### 7.3 Frontier Definition
The frontier is the lower boundary of the scalarised problem:

\[
\min_{\mathbf{s}} Q_{\xi}(\mathbf{s}) = \hat{R}_{\log}(\mathbf{r}, \mathbf{s}) + \xi \hat{\Phi}(\mathbf{s}; \mathbf{y}, \mathbf{a}),
\]

for a grid of \(\xi \ge 0\).

- At \(\xi = 0\), \(\mathbf{s} = \mathbf{r}\).
- Larger \(\xi\) drives \(\mathbf{s}\) toward lower unfairness.
- The frontier is a numerical score-vector benchmark, not a population-optimal bound.

#### 7.4 Accuracy Gap
For a fitted model with scores \(\mathbf{s}_m\):

\[
\Delta(\mathbf{s}_m) = \hat{R}_{\log}(\mathbf{r}, \mathbf{s}_m) - F\left(\hat{\Phi}(\mathbf{s}_m)\right),
\]

where \(F\) is the interpolated frontier value. A smaller gap means the fitted model is closer to the best achievable trade-off at that fairness level.

### Step 8: Interpretability Diagnostics

The paper provides three diagnostics to assess whether the structured component is sufficient as an explanation.

#### 8.1 Explained Variability Ratio (EVR)

\[
\mathrm{EVR} = \frac{\mathrm{Var}_n(\boldsymbol{\eta}^{\text{str}})}{\mathrm{Var}_n(\boldsymbol{\eta})}
= \frac{\mathrm{Var}_n(\boldsymbol{\eta}^{\text{str}})}{\mathrm{Var}_n(\boldsymbol{\eta}^{\text{str}})+\mathrm{Var}_n(\boldsymbol{\eta}^{\text{unstr}})}.
\]

EVR measures the share of logit-scale variation explained by the linear structured component. High EVR → the residual is relatively unimportant.

#### 8.2 Decision Disagreement Rate (DDR)

\[
\mathrm{DDR} = \frac{1}{n} \sum_{i=1}^n \mathbf{1}\left[\mathbf{1}[\eta_i^{\text{str}} \ge 0] \neq \mathbf{1}[\eta_i \ge 0]\right].
\]

DDR measures how often the structured component and the full model disagree on the binary default decision at threshold 0.5.

#### 8.3 Counterfactual Sign Error Rate (CSER)

For feature \(j\) with \(\beta_j \neq 0\), define a perturbation of \(\mathbf{x}_i\) in the direction of \(\beta_j\):

\[
\mathbf{x}_i^{(j,+)} = \mathbf{x}_i + d_j \delta \mathbf{e}_j, \quad d_j = \operatorname{sign}(\beta_j).
\]

Then:

\[
\mathrm{CSER}_j = \frac{1}{n} \sum_{i=1}^n \mathbf{1}\left\{ \eta(\mathbf{x}_i^{(j,+)}, \mathbf{z}_i) < \eta(\mathbf{x}_i, \mathbf{z}_i) \right\}.
\]

This measures the fraction of observations for which increasing the feature in the direction suggested by \(\beta_j\) actually *decreases* the full-model logit. CSER_min and CSER_max are reported across features.

These diagnostics together answer three questions:

1. How much of the score variation is attributable to coefficients? (EVR)
2. Do the coefficients produce the same decisions as the full model? (DDR)
3. Are local directional interpretations consistent? (CSER)

### Step 9: Simulation Study

#### 9.1 Data-Generating Process
- \(n = 10{,}000\), \(p = 6\) covariates.
- Sensitive attribute: \(A \sim \mathrm{Bern}(0.3)\).
- \(X_j \sim \mathcal{N}(0,1)\).
- **Unfairness sources:**
  - Covariate shift: \(X_1 \leftarrow X_1 + (1-A)\).
  - Baseline shift: \(B(A) = 0.2 \cdot \mathbf{1}\{A=1\}\).
- **Linear component:** \(L = b_0 + \boldsymbol{\beta}^\top \mathbf{X}\), with \(b_0 = -2\), \(\boldsymbol{\beta} = (-1, -0.5, 0, 0.25, 0.5, 1)^\top\).
- **Simple mode:** \(\eta = L + B(A)\) — purely linear signal.
- **Complex mode:** \(\eta = L + \mathrm{NL}(\mathbf{X}) + I(\mathbf{X}) + B(A)\) — includes nonlinear univariate terms and pairwise interactions.
- Outcome: \(Y \sim \mathrm{Bern}(\sigma(\eta))\).

#### 9.2 Models Compared
- **LR:** standard logistic regression.
- **NN:** two-layer feed-forward neural network as a standalone logistic model.
- **findr:** semi-structured model with linear structured logit + orthogonal neural residual.

All models are trained with the equalised-odds Wasserstein penalty. The fairness multiplier \(\lambda\) is varied over a grid from 0 to 4, with 10 random seeds.

#### 9.3 Key Simulation Results

**Simple mode (linear truth):**

- `findr` behaves almost identically to LR.
- EVR is \(1.000\) at \(\lambda=0\), and remains \(\ge 0.984\) at \(\lambda=1\).
- DDR is essentially zero.
- CSER is negligible.
- The neural residual has essentially nothing to learn.
- As \(\lambda\) increases, \(\beta_1\) (the feature with covariate shift) shrinks toward zero, illustrating that fairness adjustment operates through the structured coefficient when the signal is linear.

**Complex mode (nonlinear truth):**

- `findr` follows the neural model more than LR in predictive performance.
- EVR is much lower: approximately \(0.556\) at \(\lambda=0\), \(0.436\) at \(\lambda=0.5\), \(0.469\) at \(\lambda=1\).
- DDR becomes nonzero (around \(0.068\)–\(0.123\)).
- CSER_max increases with \(\lambda\), reaching \(0.424\) at \(\lambda=1\), showing that strong fairness regularisation can distort local directional interpretation.

**Interpretation:** `findr` recovers nonlinear structure with the residual while keeping the structured coefficients coherent with the linear part of the signal.

### Step 10: Empirical Application to Eight Credit Datasets

#### 10.1 Datasets
The sensitive attribute is **age**, binarised at 25 (`younger` vs. `older`).

| Dataset | n | Features | Default rate (%) | Sensitive share (%) |
|---|---|---|---|---|
| South German | 1,000 | 19 | 30.00 | 19.00 |
| Taiwan | 30,000 | 22 | 22.12 | 12.90 |
| GMSC | 150,000 | 9 | 6.68 | 2.02 |
| Vehicle Loan | 233,154 | 32 | 21.71 | 22.84 |
| MyHome | 7,000 | 7 | 40.00 | 7.03 |
| Thomas | 1,225 | 13 | 26.37 | 7.67 |
| UK Credit | 30,000 | 13 | 4.00 | 19.75 |
| PAKDD 2010 | 50,000 | 51 | 26.08 | 11.49 |

#### 10.2 Experimental Setup
- Train/validation/test split: 70/10/20.
- Continuous features standardised; categorical features encoded via weight of evidence.
- Models: LR, NN, `findr`.
- \(\lambda\) grid: 0 to 2.
- 10 random seeds.
- CatBoost included as an unpenalised, state-of-the-art tabular baseline.
- Main text reports Equalised Odds results.

#### 10.3 Performance Results

**Nonlinear regimes (Taiwan and GMSC):**

At \(\lambda=0\):

- Taiwan AUC: LR \(0.722\), NN \(0.777\), `findr` \(0.774\).
- GMSC AUC: LR \(0.669\), NN \(0.856\), `findr` \(0.850\).
- Brier scores follow the same pattern: flexible models achieve lower prediction error.

At \(\lambda=0.1\):

- Taiwan AUC: LR \(0.718\), NN \(0.772\), `findr` \(0.774\).
- GMSC AUC: LR \(0.669\), NN \(0.833\), `findr` \(0.849\).

**Linear regimes (remaining datasets):**

- LR, NN, and `findr` are much closer at \(\lambda=0\).
- Example: South German AUC around \(0.818\)–\(0.823\).
- `findr` often slightly improves on LR, but the gap is small.

**Fairness cost varies:**

- Taiwan and Vehicle Loan show large AUC drops at high \(\lambda\).
- GMSC retains useful nonlinear signal through moderate penalties.
- PAKDD 2010 shows AUC decline but Brier improvement, indicating the penalty produces less extreme probability forecasts.
- In many datasets, `findr` stays close to the frontier and is not systematically worse than LR or NN.

#### 10.4 Fairness Frontiers

The frontier plots show:

- In **Taiwan** and **GMSC**, `findr` and NN are closer to the frontier than LR.
- In **South German**, LR moves more directly toward the low-gap region.
- In **Thomas** and **South German**, fairness scales are wide, likely due to small sample sizes.
- **Vehicle Loan** and **UK Credit** have small gaps with moderate vertical movement.
- **PAKDD 2010** has initially small gaps but large excess loss cost when forcing the gap to zero.

#### 10.5 Interpretability Diagnostics

At \(\lambda=0\):

| Dataset | EVR | DDR | Min CSER | Max CSER |
|---|---|---|---|---|
| Taiwan | 0.527 | 0.084 | 0.088 | 0.470 |
| GMSC | 0.399 | 0.034 | 0.018 | 0.215 |
| South German | 0.957 | 0.030 | 0.000 | 0.158 |
| Vehicle Loan | 0.805 | 0.004 | 0.001 | 0.204 |
| MyHome | 0.935 | 0.022 | 0.000 | 0.004 |
| Thomas | 0.981 | 0.005 | 0.000 | 0.051 |
| UK Credit | 0.749 | 0.000 | 0.051 | 0.429 |
| PAKDD | 0.946 | 0.028 | 0.000 | 0.044 |

At \(\lambda=0.1\), the patterns persist, with modest changes.

**Interpretation:**

- In **MyHome**, **South German**, **Thomas**, **PAKDD**, EVR is high (\(>0.9\)), so the structured coefficients are close to a full explanation.
- In **Vehicle Loan** and **UK Credit**, EVR is moderate, but DDR is near zero: the residual contributes to continuous score variation without changing thresholded decisions.
- In **Taiwan** and **GMSC**, EVR is low (\(<0.55\)), and Taiwan has high DDR and CSER: the reader cannot rely on coefficients alone; the residual must be examined.
- In **GMSC**, DDR is low despite low EVR: the residual changes the score surface but not the final binary decisions.

### Step 11: Conclusion and Contributions

The paper makes several contributions:

1. **`findr` model:** combines an interpretable linear logit with an orthogonal neural residual, ensuring identifiable attribution.
2. **In-processing fairness:** uses exact univariate Wasserstein distances between protected-group score distributions, applied to the full score, not a threshold.
3. **Interpretability diagnostics:** EVR, DDR, CSER quantify whether the coefficient component is sufficient for score variation, decisions, and local directions.
4. **Score-level frontier:** provides an empirical accuracy–fairness benchmark for comparing fitted models at the same fairness level.
5. **Empirical validation:** simulation study plus eight real-world credit datasets.

The paper positions `findr` as a practical middle ground:

- When the signal is linear, it behaves like logistic regression.
- When nonlinear structure exists, it recovers much of the neural model's predictive gain while retaining interpretable coefficients.
- The diagnostics make explicit when the coefficients are sufficient and when the residual must be inspected.

### Step 12: Extensions Mentioned

- Multi-group fairness.
- Intersectional fairness.
- Individual fairness / counterfactual fairness.
- Variable selection under an explicit interpretability constraint.
- Incorporation of unstructured data (text, images) through the neural residual.
- Replacement of the linear structured component with more expressive structured models (e.g., GAMs).

### Step 13: Technical Appendix Highlights

The appendix provides further details for reproducibility:

- **Frontier construction:** uses a soft-cell approximation to the Wasserstein direction, a Polyak–Ruppert averaged heuristic, and refinement multipliers.
- **Hyperparameter search:** learning rate \(\{3\times10^{-4}, 10^{-3}, 3\times10^{-3}, 10^{-2}\}\), weight decay \(\{0,10^{-5},10^{-4},10^{-3}\}\), batch size \(\{32,64,128,256\}\), hidden widths \(\{[4], [16,8], [64,32], [128,64]\}\), activations \(\{\text{ReLU}, \text{GELU}, \tanh\}\), dropout \(\{0, 0.1, 0.25\}\).
- Models are trained with AdamW, binary cross-entropy with logits, up to 6,000 epochs, early stopping after 20 epochs without improvement, and configurations selected on validation.

## Overall Assessment

This is a technically rigorous and well-motivated contribution. It successfully formalises the trade-off between:

- interpretable linear coefficients,
- flexible neural residuals,
- distributional fairness.

The most valuable aspects are:

1. The **orthogonalisation mechanism**, which makes the linear/neural split mathematically unambiguous.
2. The **exact univariate Wasserstein penalty**, which is both theoretically clean and differentiable.
3. The **diagnostic suite**, which turns “interpretability” from a vague claim into measurable quantities.
4. The **frontier methodology**, which enables fair comparison across model classes on the same fairness and accuracy axes.

The empirical section demonstrates that `findr` is a viable practical tool, especially in nonlinear regimes where it recovers the neural advantage while preserving a readable coefficient component.

## **Discussion of the Inputs, Processes and Outputs (IPOs) of the Final Orchestrator Callables**

Below is a discussion of the Inputs, Processes and Outputs (IPOs) of the **fourteen final task-specific and top-level orchestrator callables** in the "*findr: Transparent and Fair Credit Risk Decisions through Semi-Structured Regressions*" end-to-end research pipeline (Medina-Olivares, Lessmann & Crook, 2026; arXiv:2608.24582).

**Scope note, stated for fidelity.** These fourteen are the *task* orchestrators, Task 1 → Task 13 plus the top-level wrapper. The pre-task input builders introduced when the synthetic fixture was removed — `ingest_real_datasets`, `build_simulation_input`, `build_study_config`, `assess_study_readiness` — are documented inside **Entry 14**, because they are precisely what the top-level entry point calls to manufacture the inputs consumed by Entry 1. They are not counted as separate entries, but they are not omitted either.

All signatures, equation numbers and LaTeX excerpts below were verified against the source module and `latex_context.txt` rather than recalled.

### Task 1 — `validate_inputs(config, raw_structured_features_by_dataset, raw_outcome_labels_by_dataset, raw_sensitive_attribute_by_dataset, raw_simulation_input, schema_continuous, schema_categorical, datasets_metadata=None, run_seeds=range(10), audit_log_path=PLACEHOLDER_AUDIT_PATH) → ValidationReport`

**Inputs:** `config: Dict` the frozen study configuration; three dicts of `pd.DataFrame` keyed by the eight real dataset names (`south_german, taiwan, gmsc, vehicle_loan, myhome, thomas, uk_credit, pakdd_2010`); `raw_simulation_input: pd.DataFrame` carrying `mode, seed, A, X1_post_shift, X2..X6, Y, linear_eta, nl_component, interaction_component, eta_true, r_true`; the per-dataset continuous/categorical schema lists; the ten accepted replication seeds.

**Process:** *Step 1 — key integrity.* `validate_dataset_key_symmetry` asserts three-way dictionary-key equality; `coerce_and_validate_record_ids` enforces that `record_id` is non-null, unique and **exactly** integral (the exactness test runs on the float64 representation *before* the int64 cast, so a fractional key raises rather than truncating); `check_key_set_equality` asserts the three `record_id` populations are identical, reporting the concrete dangling sets; `set_record_id_index` promotes the validated key to the index; `prepare_simulation_record_ids` generates a 1-based within-block counter via `groupby(["mode","seed"]).cumcount()+1`; `validate_simulation_domains` checks `mode ∈ {simple, complex}`, `A, Y ∈ {0,1}`, `r_true ∈ (0,1)`; `validate_dtypes_against_schema` compares against the *declared* schema rather than inferring. *Step 2 — composition gate.* `derive_sensitive_attribute` applies $A_i = \mathbf{1}\{\text{age}_i \le 25\}$, raising on a null age rather than defaulting a protected attribute to $A=0$; `compute_dataset_statistics` forms $\hat{\pi}_{\text{default}} = \frac{1}{n}\sum_i \mathbf{1}\{y_i=1\}$ and $\hat{\pi}_{\text{sensitive}} = \frac{1}{n}\sum_i \mathbf{1}\{A_i=1\}$; `validate_statistics_against_metadata` compares these against the **independent** published constants at $|\cdot| \le 0.005$ and halts on mismatch. *Step 3 — placeholder audit.* `record_placeholder_parameters` scans four channels (present paths, absent paths, live operative module constants, annotation-flagged leaves), raising if any operative threshold escapes `PLACEHOLDER_PATHS`; `write_placeholder_audit_log` seals the record to JSON.

**Transformation:** Three untrusted dicts of raw frames + one simulation frame → a `ValidationReport` carrying record_id-indexed validated frames, the simulation frame with its composite key, a `List[DatasetStatistics]` of observed-versus-published composition, and a `List[PlaceholderRecord]` with rationales. No frame is mutated until every check passes.

**Role:** Research gate for §5.1. The `record_id` join is what guarantees the exact row correspondence presupposed by

$$\tilde{\mathbf{X}} = [\mathbf{1}_n \ \mathbf{X}], \qquad \mathbf{X} = (\mathbf{x}_1, \dots, \mathbf{x}_n)^\top \in \mathbb{R}^{n \times p},$$

before any model is fitted. The composition gate enforces the paper's stated samples verbatim: *"South German Credit. $n = 1{,}000$ applicants with 19 features; default rate $30.0\%$; sensitive class share $19.0\%$"* and the seven analogous entries. **The gate is meaningful only because the metadata constants are independent of the loaded data** — deriving them from the frame reduces the check to a tautology that can never fail.

<br>

### Task 2 — `clean_raw_data(validation_report, datasets_metadata, schema_continuous, schema_categorical, indicator_suffix, cleansing_artifact_path) → CleanedDataBundle`

**Inputs:** The `ValidationReport` from Task 1; the published metadata; schema lists; the missing-indicator suffix.

**Process:** *Step 1 — referential integrity.* `drop_duplicate_record_ids` removes repeated index entries keeping the first; `inner_join_dataset_components` aligns the three frames **on the index, never on row order**, after a pairwise column-label collision check that raises rather than emitting duplicate columns; `verify_joined_counts` asserts $n_{\text{joined}} = n_{\text{features}} = n_{\text{labels}} = n_{\text{sensitive}}$; `deduplicate_simulation_block_rows` deduplicates on the non-key columns *within* each `(mode, seed)` block and raises above a $0.1\%$ block rate. *Step 2 — value hygiene.* `validate_numeric_finiteness` and `validate_plausibility_rules` enforce `age ∈ [18,100]`, `loan_amount ≥ 0`, `checking_balance` finite and permitted negative; `normalise_categorical_values` trims and lower-cases while preserving NaN; `populate_missing_indicators` creates the binary indicator **before** imputation so it records genuine missingness; `map_null_categoricals_to_unknown` maps NaN categoricals to an explicit level so no row is silently lost. *Step 3 — audit.* `compute_cleansing_report`, `audit_cleaning_against_metadata` (same $0.005$ tolerance), `write_cleansing_artifact`.

**Transformation:** Validated frames → one joined, cleansed, record_id-indexed frame per dataset, plus indicator columns that become legitimate design columns, plus a `CleansingReport` of pre/post counts and final rates.

**Role:** Implements the data-governance layer of §5.1. The ordering is the substance: the missing indicator must precede imputation, otherwise it records the imputation rather than the missingness, and the paper's per-dataset default rates and sensitive shares must survive cleansing unchanged.

<br>

### Task 3 — `preprocess_real_data(config, cleaned_bundle, datasets_metadata, schema_continuous, schema_categorical, split_config=None, woe_alpha=None, artifact_path=PREPROCESSING_ARTIFACT_PATH) → Dict[str, PreprocessedDataBundle]`

**Inputs:** The `CleanedDataBundle`; the split configuration $(0.70, 0.10, 0.20)$, `stratify_on_y=True`, `seed=42`; the WoE smoothing constant $\alpha = 0.5$.

**Process:** *Step 1 — stratified partition.* `stratified_record_id_split` performs two stratified stages using **integer** targets $\lfloor 0.20n \rceil$ and $\lfloor 0.10n \rceil$ rather than float ratios, because a float `test_size` can flip scikit-learn's rounding by one row; disjointness and full coverage are asserted, and thin minority cells warn rather than failing silently. *Step 2 — frozen transformations.* `compute_standardisation_stats` computes $\bar{X}_j^{\text{train}}, s_j^{\text{train}}$ **on training only**; `compute_woe_maps` forms, per category $c$ and under the default classical convention,

$$\mathrm{WoE}_c = \ln\!\left[\frac{(g_c + \alpha)/(G + 2\alpha)}{(b_c + \alpha)/(B + 2\alpha)}\right],$$

with $g_c, b_c$ the non-default and default counts in $c$ and $G, B$ the training totals; unseen categories receive the **pooled** training statistic. `compute_training_medians` and `impute_missing_values` complete the frozen set. `transform_partition` applies them in the order impute → WoE → standardise, standardising only the continuous list so WoE scores are not double-transformed. `validate_standardised_statistics` asserts the training mean is $0$ and s.d. $1$ post-transform. *Step 3 — design tensors.* `derive_sensitive_attribute_rawscale` applies $A_i = \mathbf{1}\{\text{age}_i \le 25\}$; `build_design_tensors` drops identifiers, casts to float64, and sets $\mathbf{Z} = \mathbf{X}$.

**Transformation:** Cleansed frames → per dataset $\mathbf{X}_{\text{train}}, \mathbf{X}_{\text{val}}, \mathbf{X}_{\text{test}}$, $\mathbf{y}_\cdot$, $\mathbf{a}_\cdot$, plus an immutable `PreprocessingStats` (means, s.d.s, medians, WoE maps, global fallbacks) serialised to JSON.

**Role:** Implements the information barrier. Every statistic is estimated on the training split and *frozen*; validation and test are transformed, never consulted. The paper's neural input convention is applied here — *"The vector $\mathbf{z}_i$ contains the inputs to the neural network, which may coincide with $\mathbf{x}_i$"* — and in all experiments $\mathbf{Z} = \mathbf{X}$, which is exactly what makes the orthogonalisation of Task 5 necessary rather than optional.

<br>

### Task 4 — `prepare_simulation_data(simulation_frame, split_config, n_observations=10000, p_covariates=6, run_seeds=None, b0=-2.0, beta=(-1,-0.5,0,0.25,0.5,1), baseline_shift=0.2, expected_modes=None) → SimulationDataBundle`

**Inputs:** The simulation frame; the same split configuration as Task 3; the DGP constants fixed by the manuscript.

**Process:** *Step 1 — structure.* `validate_columns_present` and `validate_simulation_structure` assert $n = 10{,}000$ and $p = 6$ per `(mode, seed)` block; `validate_covariate_shift_implication` verifies the implied shift through the group mean difference, accepting $|\mathbb{E}[X_1 \mid A=0] - \mathbb{E}[X_1 \mid A=1] - 1|$ within **four analytic standard errors** $4\sqrt{1/n_0 + 1/n_1}$ rather than a fixed absolute band; `derive_composite_key` enforces the `(mode, seed, record_id)` ordering. *Step 2 — mechanism verification.* `compute_true_logit` recomputes

$$\eta_i = L_i + \mathrm{NL}_i + I_i + B(A_i), \qquad L_i = b_0 + \boldsymbol{\beta}^\top \mathbf{x}_i, \qquad B(A_i) = 0.2 \cdot \mathbf{1}\{A_i = 1\},$$

on the **post-shift** design $[X_1^{\text{post}}, X_2, \dots, X_6]$; `stable_sigmoid` forms $r_i = \sigma(\eta_i)$ branchwise so no positive argument is ever exponentiated; `verify_true_logit_and_risk` asserts elementwise agreement with the supplied `eta_true` and `r_true` to $10^{-8}$, reporting the first offending row by its full composite key. *Step 3 — partition.* `split_simulation_block` applies the identical stratified machinery per block; `validate_partition_rates` confirms the default rate is preserved.

**Transformation:** One long simulation frame → a `SimulationDataBundle` of per-`(mode, seed)` $\mathbf{X}, \mathbf{Z}, \mathbf{y}, \mathbf{a}$ and the true reference risk $\mathbf{r}$, with the composite key retained.

**Role:** Implements §5.1 and guarantees the frontier's reference input. The paper fixes the generating law as $A \sim \mathrm{Bern}(0.3), \quad X_j \sim \mathcal{N}(0, 1) \quad (j = 1, \dots, 6)$, the shift $X_1 \leftarrow X_1 + (1-A)$, and two processes — **Simple mode.** $\eta = L + B(A)$; **Complex mode.** $\eta = L + \mathrm{NL}(\mathbf{X}) + I(\mathbf{X}) + B(A)$. Because §3.5 states the reference risk is the true conditional risk when available, the $10^{-8}$ verification is not decoration: a corrupted $r_i$ silently invalidates every frontier point downstream.

<br>

### Task 5 — `build_model(model_class, input_dim, hidden_layer_widths, activation, dropout_rate, X_train=None, device=None) → nn.Module`

**Inputs:** `model_class ∈ {LR, NN, findr}`; the structured width $p$; hidden widths from $\{[4],[16,8],[64,32],[128,64]\}$; activation from $\{\mathrm{ReLU}, \mathrm{GELU}, \tanh\}$; dropout from $\{0, 0.1, 0.25\}$; the **training** design $\mathbf{X}_{\text{train}}$, which is mandatory for `findr` because it defines the projector.

**Process:** Dispatches to `LogisticRegression` ($\eta_i = \beta_0 + \mathbf{x}_i^\top\boldsymbol{\beta}$ via a single biased linear layer), `NeuralLogistic` (encoder $h_{\boldsymbol{\alpha}}$ then a linear head, no orthogonalisation), or `FindrSemiStructured`. For `findr` the constructor forms $\tilde{\mathbf{X}} = [\mathbf{1}_n \ \mathbf{X}]$, takes the thin QR **in double precision**, certifies full column rank from $\min|\mathrm{diag}(\mathbf{R})|$ relative to $\max|\mathrm{diag}(\mathbf{R})|$, registers $\mathbf{Q}, \mathbf{R}$ as detached buffers, and verifies the annihilator against a **relative, self-calibrating** bound $\max(\tau_{\text{rel}}\|\cdot\|_F,\; c\sqrt{n}\,\varepsilon\|\cdot\|_F)$. The projector is applied associatively as $\mathbf{H} - \mathbf{Q}(\mathbf{Q}^\top\mathbf{H})$, never materialising the $n \times n$ operator.

**Transformation:** Configuration scalars + the training design → a parameterised `nn.Module` whose forward realises

$$\eta_i = \eta_i^{\text{str}} + \eta_i^{\text{unstr}}, \quad \eta_i^{\text{str}} = \beta_0 + \mathbf{x}_i^\top \boldsymbol{\beta}, \quad \eta_i^{\text{unstr}} = d_{\boldsymbol{\theta}}(\mathbf{z}_i) \tag{eq. 1}$$

$$\mathbf{H}_{\boldsymbol{\alpha}}^\perp = \mathbf{P}^\perp \mathbf{H}_{\boldsymbol{\alpha}}, \qquad \mathbf{P} = \mathbf{Q}\mathbf{Q}^\top, \quad \mathbf{P}^\perp = \mathbf{I}_n - \mathbf{P} \tag{eq. 2}$$

$$\boldsymbol{\eta}^{\text{unstr}} = r_{\boldsymbol{\gamma}}(\mathbf{H}_{\boldsymbol{\alpha}}^\perp) = \mathbf{H}_{\boldsymbol{\alpha}}^\perp \boldsymbol{\gamma} + \gamma_0 \mathbf{1}_n \tag{eq. 3}$$

**Role:** This is the architectural core of the paper. §3.2 states the identifiability problem exactly: *"When the same covariates enter both the structured and unstructured components in (1), the decomposition is not identifiable without additional constraints. Linear effects can then be represented either by the coefficient vector $\boldsymbol{\beta}$ or by the neural network, which makes attribution to the structured component ambiguous."* The resolution is the annihilation property: *"Since $\tilde{\mathbf{X}}^\top \mathbf{H}_{\boldsymbol{\alpha}}^\perp = \mathbf{0}$, the projected encoder features contain no intercept term and no linear combination of the covariates in $\mathbf{X}$."* Two implementation consequences follow and are honoured. First, $\gamma_0$ is *"absorbed into the structured intercept"*, so the residual layer carries `bias=False`. Second, because $\mathbf{P}^\perp$ is **constant** with respect to $\boldsymbol{\alpha}$, autograd realises $\partial\mathcal{L}/\partial\mathbf{H}_{\boldsymbol{\alpha}} = \mathbf{P}^\perp\,\partial\mathcal{L}/\partial\mathbf{H}_{\boldsymbol{\alpha}}^\perp$ with no custom backward. The relative tolerance matters at study scale: an absolute bound on a $\sqrt{n}$-growing float32 residual makes the model unconstructible above roughly twenty thousand training rows, which is five of the eight datasets.

<br>

### Task 6 — `total_loss(logits, y_true, a_true, fairness_multiplier=0.0, criterion="equalised_odds") → torch.Tensor`

**Inputs:** The logit vector $\boldsymbol{\eta}$; binary labels $\mathbf{y}$; the sensitive attribute $\mathbf{a}$; the multiplier $\lambda \ge 0$; the criterion.

**Process:** `bcce_with_logits` computes the prediction term in the numerically stable LogSumExp form $\frac{1}{n}\sum_i[\log(1+e^{\eta_i}) - y_i\eta_i]$, never forming $\sigma(\eta_i)$ before the logarithm. At $\lambda = 0$ the penalty is short-circuited to an exact zero. Otherwise $\mathbf{s} = \sigma(\boldsymbol{\eta})$ is formed for the penalty term only, and `fairness_penalty` partitions the sample into the outcome-by-group cells and calls `wasserstein_1` per cell pair. That routine sorts both supports, forms cumulative weights $\Omega_i = \sum_{r\le i}\omega_r$ and $\Upsilon_j = \sum_{\ell\le j}\upsilon_\ell$, merges them into the breakpoint grid $0 = q_0 < \dots < q_K = 1$, and integrates the absolute inverse-CDF gap using `searchsorted` on autograd-compatible primitives.

**Transformation:** Logits + labels + groups → one differentiable scalar,

$$\mathcal{L}_{\text{total}}(\boldsymbol{\theta}, \boldsymbol{\beta}) = \underbrace{\frac{1}{n}\sum_{i=1}^n\left[-y_i\log\hat{Y}_i - (1-y_i)\log(1-\hat{Y}_i)\right]}_{\mathcal{L}_{\text{pred}}} + \lambda\cdot\mathcal{L}_{\text{fair}}(\hat{\mathbf{Y}}, \mathbf{A}, \mathbf{Y}) \tag{eq. 7}$$

with the equalised-odds penalty from eq. (6), $\sum_{y\in\{0,1\}} W(\hat{Y}\mid Y=y, A=0,\; \hat{Y}\mid Y=y, A=1)$.

**Role:** Implements §3.4. The paper justifies the distributional choice: *"Because credit scores are one-dimensional, $W_p$ admits a closed-form characterisation via quantile matching and no entropic regularisation is required."* Equal opportunity is implemented on the $Y=0$ cell — not $Y=1$ — because the manuscript is explicit: *"Equal opportunity focuses only on the outcome group treated as opportunity-relevant in this application, here $Y = 0$, and therefore targets score disparities among non-defaulters."*

> **One registered deviation, disclosed rather than buried.** Eq. (5) as printed evaluates the quantile at the strip's *left* endpoint $q_{k-1}$. The paper's own quantile definition $F^{-1}_{\hat\mu}(q) = s_{(i)}$ for $i = \min\{r : \Omega_r \ge q\}$ is left-continuous, so the quantile is constant on $(q_{k-1}, q_k]$ and the *exact* integral requires the **right** endpoint $q_k$. The implementation uses $q_k$ and matches `scipy.stats.wasserstein_distance` to machine precision; the literal left-endpoint form deviates from the true distance by $0.2\%$–$36\%$ depending on the sample. The code is correct and the printed equation is imprecise, but this is a departure from the manuscript's text and is carried in the audit as a named methodological decision.

<br>

### Task 7 — `train_model_and_select_hyperparameters(model_class, X_train, y_train, a_train, X_val, y_val, a_val, X_test, config, fairness_multiplier=0.0, criterion="equalised_odds") → Tuple[Dict, List[FinalSeedResult], List[Dict]]`

**Inputs:** The design tensors; the configuration carrying the grid $\eta_{\text{lr}} \in \{3\!\times\!10^{-4}, 10^{-3}, 3\!\times\!10^{-3}, 10^{-2}\}$, $\lambda_{\text{wd}} \in \{0, 10^{-5}, 10^{-4}, 10^{-3}\}$, $b \in \{32,64,128,256\}$, widths, activations, dropout; the multiplier $\lambda$.

**Process:** *Step 1.* `sample_random_configs` draws configurations uniformly from the discrete grid, widths treated as atomic. *Step 2.* `train_model_with_early_stopping` optimises with **AdamW** (decoupled weight decay applied to the parameter values, not folded into the gradient), for at most $6{,}000$ epochs, stopping after $20$ epochs without improvement in the validation total objective, restoring the best-epoch state. Batching is model-class aware: **`findr` is forced to full batch**, because the annihilation property is defined over the whole training population and a mini-batch breaks it; `LR` and `NN` take genuine mini-batches, and when $\lambda > 0$ the sampler is **stratified by the $(y,a)$ cell** so the Wasserstein penalty never meets an empty cell. Each configuration is repeated with two restarts. *Step 3.* `rank_candidates` and `select_best_config` take the argmin of the mean validation objective; `train_final_seed_runs` retrains under `setup_seeded_environment` (Python, NumPy, torch, CUDA, `cudnn.deterministic`) across the ten seeds and freezes each model, sealing `findr`'s evaluation projection before scoring.

**Transformation:** Design tensors + a discrete grid → the selected configuration, ten frozen `FinalSeedResult` objects with test logits and probabilities, and per-run log records.

**Role:** Implements §5.2. The paper specifies *"We vary the multiplier $\lambda$ over a grid from $0$ to $4$ and repeat each configuration over 10 random seeds"*, and the three model classes: *"**LR** is a standard logistic regression… **NN** is a two-layer feed-forward neural logistic model with the same neural backbone as `findr`, but without a separate structured component. **`findr`** is the semi-structured model in Section 3."* The full-batch restriction for `findr` is not a convenience: mini-batching would void $\tilde{\mathbf{X}}^\top\mathbf{H}^\perp_{\boldsymbol{\alpha}} = \mathbf{0}$ and therefore void the coefficient interpretation the entire paper rests on.

<br>

### Task 8 — `evaluate_model(test_probabilities, y_test, test_ids, validation_scores, y_val, a_val, reference_risk, criterion="equalised_odds") → EvaluationReport`

**Inputs:** Test probabilities and labels; validation scores, labels and groups; the reference risk $\mathbf{r}$.

**Process:** `evaluate_auc_brier` computes, **on the test split**,

$$\mathrm{AUC} = \frac{1}{n_1 n_0}\sum_{i:y_i=1}\sum_{j:y_j=0}\mathbf{1}\{\hat{Y}_i > \hat{Y}_j\}, \qquad \mathrm{Brier} = \frac{1}{n_{\text{test}}}\sum_i (\hat{Y}_i - y_i)^2,$$

via the rank-based `roc_auc_score` (tie-robust) and `brier_score_loss`. `compute_validation_metrics` computes, **on the validation split**, the fairness gap $\hat{\Phi}$ and the excess log loss $\hat{R}_{\log}(\mathbf{r},\mathbf{s}) - \hat{H}(\mathbf{r})$, with every score clipped to $[\varepsilon, 1-\varepsilon]$ under one registered $\varepsilon = 10^{-6}$ shared with the frontier.

**Transformation:** Two score vectors → an `EvaluationReport` of AUC, Brier, $\hat{\Phi}$ and excess log loss, tagged with the split each was computed on.

**Role:** Implements §5.3 and the split discipline. The paper defines

$$\hat{R}_{\log}(\mathbf{r}, \mathbf{s}) = -\frac{1}{n}\sum_{i=1}^n\left[r_i\log s_i + (1-r_i)\log(1-s_i)\right] = \hat{H}(\mathbf{r}) + \frac{1}{n}\sum_{i=1}^n \mathrm{KL}\{\mathrm{Bern}(r_i)\,\|\,\mathrm{Bern}(s_i)\} \tag{eq. 9}$$

and notes that *"plots report the excess log loss $\hat{R}_{\log}(\mathbf{r},\mathbf{s}) - \hat{H}(\mathbf{r})$ so that this point has value zero."* The decomposition is why the excess quantity is the right axis: $\hat{H}(\mathbf{r})$ is irreducible, so only the average KL divergence is attributable to the model. Note the deliberate asymmetry — AUC and Brier are **test** quantities, while $\hat{\Phi}$ and excess log loss are **validation** quantities, because §3.5 constructs the frontier on $\mathcal{D}_{\text{val}}$.

<br>

### Task 9 — `compute_diagnostics(model, model_class, raw_frame, continuous, categorical, stats, feature_names, delta=0.1, decision_threshold=0.0, split="validation") → DiagnosticsReport`

**Inputs:** A fitted `findr` or `LR` model; the raw-scale evaluation frame; the frozen `PreprocessingStats`; the design column order; $\delta = 0.1$; the logit threshold $0$.

**Process:** `reprepare_features` rebuilds the design from the frozen statistics. For `findr`, `forward_decomposed` returns $\boldsymbol{\eta}^{\text{str}}$ and $\boldsymbol{\eta}^{\text{unstr}}$ separately; for `LR` the residual axis is identically zero by construction. `compute_evr` and `compute_ddr` then evaluate the two population statistics; `extract_structured_beta` recovers $\hat{\boldsymbol{\beta}}$; `compute_cser` perturbs each eligible **standardised** design column and re-scores in float64; `summarise_cser` reduces over the **eligible set only**, returning NaN when that set is empty.

**Transformation:** A fitted model + a raw frame → a `DiagnosticsReport` of EVR, DDR (rate and count), $\mathrm{CSER}_{\min}$, $\mathrm{CSER}_{\max}$, the per-feature map, and the eligible-feature **support count**.

**Role:** Implements §3.6 in full:

$$\mathrm{EVR} = \frac{\mathrm{Var}_n(\boldsymbol{\eta}^{\text{str}})}{\mathrm{Var}_n(\boldsymbol{\eta})} = \frac{\frac{1}{n}\sum_i(\eta_i^{\text{str}} - \bar{\eta}^{\text{str}})^2}{\frac{1}{n}\sum_i(\eta_i - \bar{\eta})^2} \tag{eq. 12}$$

$$\mathrm{DDR} = \frac{1}{n}\sum_{i=1}^n\mathbf{1}\left[\hat{y}_i^{\text{str}} \neq \hat{y}_i\right] \tag{eq. 13}$$

$$\mathrm{CSER}_j = \frac{1}{n}\sum_{i=1}^n\mathbf{1}\left\{\eta(\mathbf{x}_i^{(j,+)}, \mathbf{z}_i) < \eta(\mathbf{x}_i, \mathbf{z}_i)\right\}, \qquad \mathbf{x}_i^{(j,+)} = \mathbf{x}_i + d_j\delta\mathbf{e}_j \tag{eq. 14}$$

with $d_j = \mathrm{sign}(\beta_j)$, reported as $\mathrm{CSER}_{\min} = \min_{j:\beta_j\neq0}\mathrm{CSER}_j$ and $\mathrm{CSER}_{\max} = \max_{j:\beta_j\neq0}\mathrm{CSER}_j$. Variances use $\mathrm{ddof}=0$ because eq. (12) is written with the $\frac{1}{n}$ population denominator; the orthogonalisation of eq. (2) is what makes the total variance decompose additively, so $\mathrm{EVR} = \mathrm{Var}_n(\boldsymbol{\eta}^{\text{str}})/[\mathrm{Var}_n(\boldsymbol{\eta}^{\text{str}}) + \mathrm{Var}_n(\boldsymbol{\eta}^{\text{unstr}})]$ — an identity that holds exactly on the training population only.

> **Two implementation points that decide whether these numbers mean anything.** (i) $\mathbf{e}_j$ is a unit vector in the **design-matrix** space, so $\delta$ must act on the standardised column. Applying it on the raw scale makes the effective perturbation $\delta/s_j$, which for a loan amount with $s_j$ in the thousands is $\sim\!2\times10^{-5}$ s.d. — at which point the strict inequality is decided by floating-point rounding, not by the model. (ii) The restriction $\{j : \hat\beta_j \neq 0\}$ is load-bearing: a zero coefficient has no signed direction. Reducing over ineligible features pins $\mathrm{CSER}_{\min}$ to exactly $0.000$, which is precisely the value the paper's tables report — so the error would *manufacture* apparent agreement while the statistic was never computed. The support count is therefore carried alongside the statistic.

<br>

### Task 10 — `construct_frontier(reference_risk, y_true, a_true, fitted_scores=None, frontier_config=None, criterion="equalised_odds") → FrontierReport`

**Inputs:** The validation reference risk $\mathbf{r}$, labels $\mathbf{y}$, groups $\mathbf{a}$; the named fitted-model validation score vectors; the frontier configuration.

**Process:** *Step 1.* `build_xi_grid` forms $\xi \in \{0\} \cup \mathrm{logspace}(-2, 2, 40)$, 41 points. *Step 2.* At $\xi = 0$ the closed-form solution $\mathbf{s} = \mathbf{r}$ is inserted directly. For each subsequent $\xi$, `solve_scalarised_frontier` selects its warm start by minimising $Q_\xi$ over the candidate pool $\{\mathbf{r},\ \text{fitted model scores},\ \text{previous-}\xi\ \text{solution}\}$, then iterates the soft-cell sign update, clipping to $[\varepsilon, 1-\varepsilon]$ after every step and averaging the **full second half** of the iterate path. `compute_soft_memberships`, `weighted_cdf_matrix` and `effective_signs` supply $\pi_{ug}$, $F_{ug}$ and $\Theta$. *Step 3.* `collect_candidates` pools warm starts, periodic iterates and PR averages with the fitted scores; `extract_frontier_envelope` sorts by $\hat{\Phi}$, collapses duplicates to the minimum loss and takes the running minimum. **Refinement:** adjacent envelope nodes are measured by *range-normalised* chord length; where the chord exceeds $0.05$, a new multiplier is inserted at the **geometric mean** of the bracketing $\xi$ (the log-midpoint on a logarithmic grid), re-solved warm-started from the nearer bracket, and the envelope re-extracted — up to fifteen insertions.

**Transformation:** A reference risk + labels + groups + fitted scores → a monotone lower envelope $F(\phi)$ and a per-model accuracy gap.

**Role:** Implements §3.5 and Appendix A:

$$\min_{\mathbf{s}} Q_\xi(\mathbf{s}) = \hat{R}_{\log}(\mathbf{r}, \mathbf{s}) + \xi\hat{\Phi}(\mathbf{s}; \mathbf{y}, \mathbf{a}), \quad \xi \ge 0 \tag{eq. 10}$$

$$\Theta_i = \sum_{u\in\{0,1\}}\sum_{g\in\{0,1\}}\frac{\pi_{ug}(\mathbf{x}_i)}{p_{ug}}\,\mathrm{sign}\{F_{ug}(s_i;\mathbf{s}) - F_{u,1-g}(s_i;\mathbf{s})\} \tag{eq. 15}$$

$$s_i^{(t+1)} = (1-\rho_\xi)s_i^{(t)} + \rho_\xi\left\{r_i + \xi s_i^{(t)}(1-s_i^{(t)})\Theta_i^{(t)}\right\}, \quad \rho_\xi = \frac{\rho}{1+\xi} \tag{eq. 16}$$

$$\Delta(\mathbf{s}_m) = \hat{R}_{\log}(\mathbf{r}, \mathbf{s}_m) - F\left(\hat{\Phi}(\mathbf{s}_m; \mathbf{y}, \mathbf{a})\right) \tag{eq. 11}$$

The paper is careful about interpretation and the implementation must not overreach: *"The frontier should be interpreted as a numerical score-vector benchmark rather than as the attainable boundary for any specified family of fitted models."* It also anticipates the sign of the gap: *"small negative values can occur if the computed envelope fails to dominate an unseeded candidate score"* — which is exactly why fitted scores enter the *initialisation* pool and not merely the candidate pool. Two numerical points: $\Theta_i$ is an empirical **sign** of weighted CDF differences, not a gradient, so the iterates oscillate and the Polyak–Ruppert average must run over the genuine second half of the path — averaging a strided subsample discards precisely the high-frequency oscillation the average exists to cancel. And because $\Delta$ is measured against $F$, a coarse envelope lies *above* the true one and biases every reported gap; the refinement stage is what controls that bias.

<br>

### Task 11 — `run_study_pipeline(config, raw_structured_features_by_dataset, raw_outcome_labels_by_dataset, raw_sensitive_attribute_by_dataset, raw_simulation_input, schema_continuous, schema_categorical, subset=None) → StudyResults`

**Inputs:** The configuration, the three raw dicts, the simulation frame, the schema lists, and an optional subset restricting datasets, modes, models, $\lambda$ values and seeds.

**Process:** Executes the Task 11 Step 2 graph in order: `validate_inputs` → `clean_raw_data` → `preprocess_real_data` per dataset → `prepare_simulation_data` per `(mode, seed)` → for each dataset/mode and each $\lambda$, `build_model` for LR, NN and `findr` then `train_model_and_select_hyperparameters` → `evaluate_model` and `compute_diagnostics` for every fitted model → `construct_frontier` on the **validation** split → aggregation. Memory discipline is enforced by constructing, training and evaluating one model at a time with explicit release. `build_reference_risk_lr` supplies $\mathbf{r}$ for real data as the unpenalised logistic scores at $\lambda = 0$; simulation uses the true conditional risk. Every run is validated against `RUN_RECORD_SCHEMA` and appended to a JSONL registry under a cross-process exclusive lock, with the byte offset of the write verified on read-back. `check_verification_targets` compares each run against **scoped** targets (dataset, mode, model class, $\lambda$) and returns violations for the caller to halt on.

**Transformation:** Raw inputs → a `StudyResults` registry of per-run records, the aggregated summary, per-dataset frontier reports, and the violation list.

**Role:** The master execution graph. It enforces the two disciplines the paper depends on: the **information barrier** (scaling, WoE and imputation statistics frozen from training before any validation or test scoring) and the **split assignment** (frontier and fairness gap on validation, AUC and Brier on test). The reference-risk choice implements §3.5 directly: *"using the true conditional risk in simulations when available and otherwise the score from an unconstrained model fitted on the training data."*

<br>

### Task 12 — `run_full_study(config, raw_structured_features_by_dataset, raw_outcome_labels_by_dataset, raw_sensitive_attribute_by_dataset, raw_simulation_input, schema_continuous, schema_categorical, subset=None) → Dict[str, Any]`

**Inputs:** As Task 11, plus the robustness variant definitions.

**Process:** *Step 1 — the matrix.* `matrix_combinations` forms the Cartesian product; `verify_matrix_size` asserts the realised run count equals the product of the dimensions; `run_experiment_matrix` iterates via `itertools.product`, fans the dataset dimension across a `Pool`, and resumes from `completed_run_keys`. *Step 2 — sensitivity.* `build_robust_variants` emits overlays for $\rho \in \{0.05, 0.1, 0.2\}$, $\delta \in \{0.05, 0.1, 0.2\}$, $\alpha \in \{0.25, 0.5, 1.0\}$, the WoE convention, the orthogonality tolerance, the CSER perturbation space and branch mode, the PR fraction, alternative nonlinear specifications, and the supplementary criteria; any registered numeric placeholder without an explicit family receives an automatic $\pm50\%$ pair, closing the registry loop. `apply_overrides` re-runs the chain without touching pipeline code; `paired_ttest_on_accuracy_gap` compares arms on aligned `(dataset, $\lambda$, model, seed)` cells after dropping non-finite pairs on both sides; `diagnostic_material_change` flags a ranking reversal or a $>10\%$ relative move in EVR, DDR or CSER. *Step 3 — verification.* `verify_against_paper` evaluates the reference patterns.

**Transformation:** The configuration → the complete run matrix, a sensitivity table tagged by overridden parameter, and a pass/fail check table.

**Role:** Implements the study's execution and robustness contract. The verification patterns are the paper's own reported findings: simple-mode *"EVR remains at or near one across the reported $\lambda$ values, DDR is essentially zero, and CSER is negligible"*; complex-mode EVR in $[0.44, 0.56]$ and DDR in $[0.06, 0.13]$; *"`findr` recovers most of this gain, reaching AUC $0.774$ in Taiwan and $0.850$ in GMSC"*; and, on the remaining six datasets, *"the three models are much closer at $\lambda = 0$, suggesting approximately linear regimes."* The complex-mode CSER check is tested for **monotonicity** by Spearman rank correlation against $\lambda$ rather than by a level, because seed noise defeats pairwise comparison; and the CSER support is asserted non-empty, so a pinned zero can never again be mistaken for agreement.

<br>

### Task 13 — `produce_final_deliverables(artifacts, summary, diagnostics, frontiers, coefficient_paths, simulation_densities, plot_config, output_dir=DEFAULT_OUTPUT_DIR) → Dict[str, Any]`

**Inputs:** The fitted-model artifacts, the aggregated summary, the diagnostics frame, the frontier dumps, the coefficient paths, the simulation score densities, and the plot configuration.

**Process:** *Step 1 — the vault.* `capture_seed_and_version_record` records the Python, NumPy, pandas, scipy, scikit-learn and torch versions plus every seed; `serialise_artifacts` writes each best model via `torch.save` through a temporary path, the frozen statistics and frontier arrays as JSON, and seals a manifest with per-object SHA-256. For `findr` the evaluation projection is **frozen from the final encoder and registered as a buffer before the save**, so the artifact is self-contained. `verify_model_roundtrip` then captures the pre-save predictions, reloads into a freshly constructed architecture under `strict=True`, recomputes on the identical design in the identical dtype, and **raises** when the maximum absolute difference exceeds tolerance; a failed round trip aborts publication. `verify_manifest` confirms every path exists and hashes to its record. *Step 2 — outputs.* `generate_tables` writes the metric and diagnostics CSVs, annotating the CSER support so an all-zero column over an empty eligible set is distinguishable from the genuine zero a purely linear model produces. `generate_figures` renders at `figsize (12,8)`, `dpi 150`, with `findr` blue, LR green, NN orange, frontier black, reference grey and the `viridis` $\lambda$ colour map. *Step 3.* `render_audit_report` composes the fixed-order report and seals it with the SHA-256 of its body.

**Transformation:** In-memory results → an immutable, hash-sealed deliverables directory that a downstream user can reload without retraining.

**Role:** Implements the reproducibility contract. The round-trip check is the only thing that makes "reload without retraining" a *claim* rather than an assertion: a checksum that is never compared to anything cannot detect a failure, and for `findr` specifically the check fails unless the projection is part of the saved state — which is exactly the sort of silent divergence an audit exists to catch.

<br>

### Task 14 (top level) — `run_findr_end_to_end(config, raw_structured_features_by_dataset, raw_outcome_labels_by_dataset, raw_sensitive_attribute_by_dataset, raw_simulation_input, schema_continuous, schema_categorical, subset=None, output_dir=Path("deliverables"), preserved_models=None) → Dict[str, Any]` and its data-gated entry point `run_full_findr_study(data_root, output_dir, subset=None, allow_partial=False)`

**Inputs:** For the wrapper, the full configuration and raw inputs. For the entry point, only a `data_root`: it *manufactures* the inputs.

**Process:** `run_full_findr_study` performs the pre-task stage. `assess_study_readiness` stats the filesystem for all eight provider files and returns `reproduction_valid` only when all eight are present; absent data raises, naming every missing dataset and its provider, unless `allow_partial=True`, in which case the registry is flagged `reproduction_valid = False`. `ingest_real_datasets` then loads and adapts each file through its `DatasetSourceSpec`: label polarity is mapped explicitly (several providers encode the *good* outcome as 1), a date-of-birth column is converted to completed years, provider headers are resolved case- and separator-insensitively, a non-unique provider identifier falls back to a positional one with a warning, and `validate_ingested_feature_count` refuses any file whose feature count deviates from the published count. `build_simulation_input` generates both modes across ten seeds via `simulate_dgp_block`, and `build_study_config` assembles the configuration whose `real_data_experiment.datasets` block holds the **independent** published constants. The wrapper then sequences the six execution steps — Tasks 1–2, Task 3, Task 4, Tasks 5–10, Task 12, Task 13 — each recorded as an `ExecutionStep` with a status, so a failure localises rather than aborting opaquely.

**Transformation:** A directory path → the complete study registry, with the readiness assessment, the ingestion report and the `reproduction_valid` flag travelling *with* the result so no downstream consumer can present a partial run as a reproduction.

**Role:** The integration layer, and the honesty boundary. The manuscript enumerates the feature list for only one of the eight datasets and directs the reader to the original provider for the rest; no code can manufacture those files, so this layer **reads** them and refuses to proceed when they are absent rather than substituting a synthetic approximation. The complex-mode DGP it generates is the paper's own:

$$\mathrm{NL}(\mathbf{X}) = \sum_{k\in\mathcal{I}_{\mathrm{nl}}} g_k(X_k), \qquad I(\mathbf{X}) = \sum_{(i,j)\in\mathcal{I}_{\mathrm{int}}} X_i X_j,$$

with the functional forms and index sets flagged `UNSPECIFIED IN MANUSCRIPT`, registered as placeholders, and swept in Task 12. The purpose of the two modes is stated plainly: *"The simple mode therefore tests whether the semi-structured model avoids using the residual when it is unnecessary. The complex mode tests whether the same architecture can recover nonlinear signal without losing the linear component used for interpretation."*

<br><br>

## **Usage Example**

Here is a granular, step-by-step guide to executing the end-to-end pipeline for **"findr: Transparent and Fair Credit Risk Decisions through Semi-Structured Regressions"**. This example demonstrates how to (i) synthetically generate the four raw study inputs at high fidelity, (ii) load `config.yaml` from the working directory into a Python dictionary, and (iii) execute the full study via the `run_findr_end_to_end(...) → Dict[str, Any]` interface.

> **Environment assumption (mandatory).** All pipeline callables from Tasks 1–13 plus the wrapper (`validate_inputs`, `clean_raw_data`, `preprocess_real_data`, `prepare_simulation_data`, `build_model`, `total_loss`, `train_model_and_select_hyperparameters`, `evaluate_model`, `compute_diagnostics`, `construct_frontier`, `run_study_pipeline`, `run_full_study`, `produce_final_deliverables`, `run_findr_end_to_end`) are assumed to be defined **in a single Jupyter notebook** cell-context. There is **no `.py` module** to import, so this example imports only the libraries needed for synthetic data generation, YAML loading, and result inspection. `config.yaml` is assumed to be saved in the working directory under that exact name.

> **Verification status.** Every generator below was executed against the *real* Task 1 and Task 2 gates before being written into this document. Both passed at full published scale (502,379 real rows + 200,000 simulation rows). The two defects the gates surfaced during that run are documented at Step 1.4 rather than silently patched away.

<br>

### **Step 1: Synthetic Data Generation (the four raw study inputs)**

The pipeline ingests exactly four objects: three dictionaries of `pd.DataFrame` keyed by the eight real dataset names, and one simulation `pd.DataFrame`. Task 1 validates them against `config["real_data_experiment"]["datasets"]`, and **halts** on mismatch, so the generators must hit the published composition *by construction* rather than by sampling luck.

#### The binding constraint that dictates the generator design

`validate_statistics_against_metadata` enforces three things per dataset:

$$n_{\text{observed}} = n_{\text{published}} \quad \text{(exact equality)},$$
$$\left|\hat{\pi}_{\text{default}} - \pi_{\text{published}}\right| \le 0.005, \qquad \hat{\pi}_{\text{default}} = \frac{1}{n}\sum_{i=1}^{n}\mathbf{1}\{y_i = 1\},$$
$$\left|\hat{\pi}_{\text{sensitive}} - \pi_{\text{published}}\right| \le 0.005, \qquad \hat{\pi}_{\text{sensitive}} = \frac{1}{n}\sum_{i=1}^{n}\mathbf{1}\{A_i = 1\}, \quad A_i = \mathbf{1}\{\text{age}_i \le 25\}.$$

Sampling labels as `Bernoulli(0.0668)` at $n = 150{,}000$ would drift by roughly $\pm 0.0013$ — inside tolerance — but at $n = 1{,}000$ a `Bernoulli(0.30)` draw has standard deviation $\approx 0.0145$, which **fails** the 0.005 band roughly 73% of the time. The generator therefore constructs **exact counts** and shuffles, so the gate passes deterministically at every scale.

#### Generating the three per-dataset dictionaries

```python
# ==============================================================================
# Step 1 - High-Fidelity Synthetic Generation of the Four Raw Study Inputs
# ==============================================================================

# Standard library and third-party imports required by the generators only;
# every pipeline callable is already defined in the notebook context.
import time
from pathlib import Path
from typing import Any, Dict, List, Tuple

import numpy as np
import pandas as pd
import yaml
from faker import Faker

# Seed both engines so the synthetic reproduction is bit-for-bit deterministic
Faker.seed(42)
np.random.seed(42)
# Instantiate the Faker generator used for realistic categorical level names
fake: Faker = Faker()

# The manuscript enumerates the feature list for South German ONLY; the other
# seven are UNSPECIFIED IN MANUSCRIPT and must come from the provider. The
# named schema below reproduces the Statlog South German Credit columns.
SOUTH_GERMAN_CONTINUOUS: List[str] = [
    "age", "checking_balance", "employment_duration", "loan_amount",
    "duration_months", "installment_rate", "residence_since",
    "existing_credits", "dependents",
]
SOUTH_GERMAN_CATEGORICAL: List[str] = [
    "credit_history", "purpose", "savings_account", "personal_status",
    "other_debtors", "property", "other_installment_plans", "housing",
    "job", "telephone",
]
# Realistic level sets per categorical feature; these are WoE-encoded by Task 3
CATEGORY_LEVELS: Dict[str, List[str]] = {
    "credit_history": ["no_credits", "all_paid", "existing_paid",
                       "delayed", "critical"],
    "purpose": ["car_new", "car_used", "furniture", "radio_tv",
                "education", "business", "repairs"],
    "savings_account": ["lt_100", "100_500", "500_1000", "gt_1000",
                        "unknown"],
    "personal_status": ["male_single", "male_married", "female_single",
                        "female_divorced"],
    "other_debtors": ["none", "co_applicant", "guarantor"],
    "property": ["real_estate", "life_insurance", "car", "unknown"],
    "other_installment_plans": ["bank", "stores", "none"],
    "housing": ["rent", "own", "free"],
    "job": ["unemployed", "unskilled", "skilled", "management"],
    "telephone": ["none", "yes"],
}


def build_feature_schema(
    dataset_key: str, n_features: int
) -> Tuple[List[str], List[str]]:
    """
    Build the continuous and categorical feature-name lists for one dataset.

    Purpose:
        Produces the per-dataset schema lists consumed by Task 1 dtype
        validation and Task 3 preprocessing. South German uses its named
        provider schema; the remaining seven use generated names whose
        COUNT reproduces the published feature count exactly, because the
        manuscript does not enumerate their columns.

    Inputs:
        dataset_key (str): One of the eight study dataset keys.
        n_features (int): The published feature count for that dataset
            (19, 22, 9, 32, 7, 13, 13, 51 respectively).

    Processes:
        1. Return the named Statlog schema for South German.
        2. Otherwise allocate roughly one quarter of the budget to
           categorical features and the remainder to continuous ones.
        3. Force 'age' into the continuous list, since it is the sole
           source of the sensitive attribute A = 1{age <= 25}.

    Outputs:
        Tuple[List[str], List[str]]: (continuous, categorical) names whose
        combined length equals n_features exactly.

    Raises:
        ValueError: If n_features is not strictly positive.
    """
    # A dataset with no features cannot form a design matrix
    if n_features <= 0:
        raise ValueError(
            f"n_features must be positive for '{dataset_key}'; "
            f"received {n_features}."
        )
    # South German is the one dataset the manuscript enumerates
    if dataset_key == "south_german":
        return list(SOUTH_GERMAN_CONTINUOUS), list(SOUTH_GERMAN_CATEGORICAL)
    # Allocate about a quarter of the budget to categorical features
    n_categorical: int = max(1, n_features // 4)
    # The remainder are continuous, and must include age
    n_continuous: int = n_features - n_categorical
    # 'age' occupies the first continuous slot for every dataset
    continuous: List[str] = ["age"] + [
        f"{dataset_key}_num_{i:02d}" for i in range(1, n_continuous)
    ]
    # Generated categorical names, WoE-encoded downstream
    categorical: List[str] = [
        f"{dataset_key}_cat_{i:02d}" for i in range(1, n_categorical + 1)
    ]
    # The two lists must reproduce the published feature count exactly
    if len(continuous) + len(categorical) != n_features:
        raise ValueError(
            f"schema for '{dataset_key}' produced "
            f"{len(continuous) + len(categorical)} features against the "
            f"published {n_features}."
        )
    return continuous, categorical


def generate_dataset_frames(
    dataset_key: str, metadata: Dict[str, Any], seed: int,
) -> Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame, List[str], List[str]]:
    """
    Generate one dataset's three raw frames at the published composition.

    Purpose:
        Produces the (features, labels, sensitive) triple for a single
        study dataset, hitting the published n, default rate and sensitive
        share EXACTLY so the Task 1 Step 2 gate passes deterministically
        rather than by sampling luck.

    Inputs:
        dataset_key (str): One of the eight study dataset keys.
        metadata (Dict[str, Any]): The published composition from
            config["real_data_experiment"]["datasets"][dataset_key],
            carrying n, features, default_rate and sensitive_share.
        seed (int): The generator seed for this dataset.

    Processes:
        1. Build the feature schema reproducing the published count.
        2. Construct ages so that EXACTLY round(share * n) rows satisfy
           age <= 25, giving the sensitive share by construction.
        3. Populate continuous features with realistic distributions, and
           inject NaNs into employment_duration to exercise the
           missing-indicator path of Task 2.
        4. Populate categorical features as object dtype (see 1.4).
        5. Construct labels with EXACTLY round(default_rate * n) positives.
        6. Set raw_age identically equal to the age column.

    Outputs:
        Tuple of (features, labels, sensitive, continuous, categorical).

    Raises:
        ValueError: If the metadata block is malformed.
    """
    # Every published composition field is mandatory
    for field_name in ("n", "features", "default_rate", "sensitive_share"):
        if field_name not in metadata:
            raise ValueError(
                f"metadata for '{dataset_key}' is missing '{field_name}'."
            )
    # Unpack the published composition
    n: int = int(metadata["n"])
    n_features: int = int(metadata["features"])
    # A dedicated generator keeps each dataset independently reproducible
    rng: np.random.Generator = np.random.default_rng(seed)
    # Resolve the feature schema for this dataset
    continuous, categorical = build_feature_schema(dataset_key, n_features)
    # record_id is the primary join key: 1-based, unique, exactly integral
    record_id: np.ndarray = np.arange(1, n + 1, dtype=np.int64)
    # EXACT sensitive count: A = 1{age <= 25} must hit the published share
    n_sensitive: int = int(round(float(metadata["sensitive_share"]) * n))
    # Allocate the age vector in two blocks, then shuffle
    age: np.ndarray = np.empty(n, dtype=np.int64)
    # The sensitive block draws ages in [18, 25]
    age[:n_sensitive] = rng.integers(18, 26, n_sensitive)
    # The non-sensitive block draws ages in [26, 75]
    age[n_sensitive:] = rng.integers(26, 76, n - n_sensitive)
    # Shuffle so the sensitive rows are not a contiguous block
    rng.shuffle(age)
    # Assemble the feature frame column by column
    columns: Dict[str, Any] = {"record_id": record_id}
    for name in continuous:
        if name == "age":
            # The sensitive source, retained as a structured feature here
            columns[name] = age
        elif name == "checking_balance":
            # Continuous and MAY be negative, per the schema discussion
            columns[name] = rng.normal(0.0, 1500.0, n).round(2)
        elif name == "employment_duration":
            # Continuous WITH missing values: NaNs exercise the Task 2
            # indicator path and the Task 3 median imputation
            column: np.ndarray = rng.integers(0, 480, n).astype(np.float64)
            column[rng.random(n) < 0.05] = np.nan
            columns[name] = column
        elif name == "loan_amount":
            # Continuous and strictly non-negative, per the plausibility rule
            columns[name] = rng.integers(250, 20000, n).astype(np.int64)
        else:
            # Generic standardised continuous covariate
            columns[name] = rng.normal(0.0, 1.0, n).round(4)
    for name in categorical:
        # Named levels where the provider schema is known, generated otherwise
        levels: List[str] = CATEGORY_LEVELS.get(
            name, [f"{name}_lvl_{i}" for i in range(1, 5)])
        # Cast to object EXPLICITLY: see the dtype note at Step 1.4
        columns[name] = pd.Series(rng.choice(levels, size=n), dtype=object)
    # Materialise the structured feature frame
    features: pd.DataFrame = pd.DataFrame(columns)
    # EXACT default count: y = 1 denotes DEFAULT
    n_default: int = int(round(float(metadata["default_rate"]) * n))
    # Allocate the label vector, then shuffle
    y_default: np.ndarray = np.zeros(n, dtype=np.int64)
    y_default[:n_default] = 1
    rng.shuffle(y_default)
    # Materialise the outcome frame with the join key
    labels: pd.DataFrame = pd.DataFrame(
        {"record_id": record_id, "y_default": y_default})
    # raw_age MUST equal age: derive_sensitive_attribute PREFERS raw_age, so
    # any disagreement would make the two composition gates contradict
    sensitive: pd.DataFrame = pd.DataFrame(
        {"record_id": record_id, "raw_age": age.astype(np.int64)})
    return features, labels, sensitive, continuous, categorical
```

#### Assembling the four inputs

The simulation frame is **not** generated by hand. The notebook already contains the study's own data-generating process (`build_simulation_input` → `simulate_dgp_block`), and Task 4 verifies the supplied `eta_true` and `r_true` against a recomputation to $10^{-8}$:

$$\eta_i = L_i + \mathrm{NL}_i + I_i + B(A_i), \quad L_i = b_0 + \boldsymbol{\beta}^\top\mathbf{x}_i, \quad B(A_i) = 0.2\cdot\mathbf{1}\{A_i=1\}, \quad r_i = \sigma(\eta_i).$$

A hand-rolled re-implementation would fail that check on any arithmetic discrepancy. Calling the pipeline's own generator is both more accurate and more honest — the DGP is *specified mathematics*, not fake data.

```python
def assemble_study_inputs(
    config: Dict[str, Any],
) -> Tuple[Dict[str, pd.DataFrame], Dict[str, pd.DataFrame],
           Dict[str, pd.DataFrame], pd.DataFrame,
           Dict[str, List[str]], Dict[str, List[str]]]:
    """
    Assemble the four raw study inputs plus the two schema dictionaries.

    Purpose:
        Produces the exact input contract of run_findr_end_to_end: the three
        per-dataset dictionaries keyed by the eight real dataset names, the
        simulation frame covering both modes across ten seeds, and the
        continuous/categorical schema lists Task 1 validates dtypes against.

    Inputs:
        config (Dict[str, Any]): The loaded study configuration; its
            real_data_experiment.datasets block supplies the published
            composition each dataset must reproduce.

    Processes:
        1. Iterate the eight published dataset entries in canonical order.
        2. Generate each dataset's three frames at its exact composition.
        3. Generate the simulation frame through the notebook's own DGP,
           covering both modes across the ten replication seeds.

    Outputs:
        Tuple of (raw_structured_features_by_dataset,
        raw_outcome_labels_by_dataset, raw_sensitive_attribute_by_dataset,
        raw_simulation_input, schema_continuous, schema_categorical).

    Raises:
        KeyError: If the configuration lacks the dataset metadata block.
    """
    # The published metadata block is the generation target
    if "real_data_experiment" not in config:
        raise KeyError("config lacks the 'real_data_experiment' section.")
    metadata_block: Dict[str, Any] = config["real_data_experiment"]["datasets"]
    # Accumulate the three dictionaries and the two schema maps
    raw_structured_features_by_dataset: Dict[str, pd.DataFrame] = {}
    raw_outcome_labels_by_dataset: Dict[str, pd.DataFrame] = {}
    raw_sensitive_attribute_by_dataset: Dict[str, pd.DataFrame] = {}
    schema_continuous: Dict[str, List[str]] = {}
    schema_categorical: Dict[str, List[str]] = {}
    # Generate one dataset at a time, with a distinct seed per dataset
    for offset, (dataset_key, metadata) in enumerate(metadata_block.items()):
        features, labels, sensitive, continuous, categorical = \
            generate_dataset_frames(dataset_key, metadata, 1000 + offset)
        # File the three frames under the SAME key in all three dictionaries
        raw_structured_features_by_dataset[dataset_key] = features
        raw_outcome_labels_by_dataset[dataset_key] = labels
        raw_sensitive_attribute_by_dataset[dataset_key] = sensitive
        # File the schema lists under the same key
        schema_continuous[dataset_key] = continuous
        schema_categorical[dataset_key] = categorical
    # The simulation frame comes from the notebook's OWN DGP so that the
    # Task 4 verification of eta_true and r_true passes to 1e-8
    raw_simulation_input: pd.DataFrame = build_simulation_input(
        modes=tuple(config["simulation"]["modes"]),
        seeds=tuple(range(int(config["network_training"]["num_seeds"]))),
        n_observations=int(config["simulation"]["n_observations"]),
    )
    return (raw_structured_features_by_dataset,
            raw_outcome_labels_by_dataset,
            raw_sensitive_attribute_by_dataset,
            raw_simulation_input,
            schema_continuous, schema_categorical)
```

#### Two defects the real gates surfaced during verification

Both were found by running the generators against `validate_inputs` before this document was written, and both are corrected above.

**(a) Categorical dtype.** `rng.choice(levels, size=n)` over Python strings returns a NumPy `str_` array. Modern pandas infers dtype `str` for it, and `validate_dtypes_against_schema` correctly rejects that:

> `ValueError: dataset 'gmsc': categorical column 'gmsc_cat_01' has dtype str; expected object or category.`

The fix is the explicit `pd.Series(..., dtype=object)` cast. This is not pedantry — a `str`-dtype column would reach the WoE encoder, whose map keys are built with `str(...)`, and a type mismatch there sends *every* lookup to the global fallback, silently destroying the feature.

**(b) Exact-count construction.** Sampling `Bernoulli(p)` labels fails the 0.005 band at small $n$. The exact-count-plus-shuffle construction above is what makes the gate deterministic.

```python
# Generate the complete input bundle once for the pipeline run

 raw_sensitive_attribute_by_dataset, raw_simulation_input,
 schema_continuous, schema_categorical) = assemble_study_inputs(config)

# Preview each dataset against its published composition

        raw_outcome_labels_by_dataset[key]["y_default"].mean())

        (raw_sensitive_attribute_by_dataset[key]["raw_age"] <= 25).mean())

```

### **Step 2: Loading the Configuration (`config.yaml`)**

`config.yaml` is the single frozen source of truth. It is read into memory and converted into the nested `config` dictionary. Two normalisations matter for perfect fidelity: YAML has no tuple type, so `figsize` and the covariate index pairs arrive as lists; and `Path`-valued entries arrive as strings.

```python
# ==============================================================================
# Step 2 - Load config.yaml from the Working Directory into a Python Dictionary
# ==============================================================================

def load_study_configuration(filepath: str = "config.yaml") -> Dict[str, Any]:
    """
    Load the frozen study configuration from the working directory.

    Purpose:
        Ingests the fifteen configuration sections consumed by every task
        orchestrator - schemas, preprocessing, architecture, training,
        fairness penalty, frontier, diagnostics, simulation, real-data
        metadata, hyperparameter grid, LLM settings, plots, methodological
        notes, evaluation metrics and the verification-target gate - and
        converts them into the nested dictionary the pipeline consumes.

    Inputs:
        filepath (str): Path to the YAML file, default 'config.yaml',
            assumed to be in the working directory.

    Processes:
        1. Validate the argument type and the file's existence.
        2. Parse the document with yaml.safe_load, which constructs only
           plain dict/list/scalar objects and never arbitrary Python types.
        3. Verify that the required top-level sections are present.
        4. Report the section inventory to the operator.

    Outputs:
        Dict[str, Any]: The complete nested configuration dictionary.

    Raises:
        TypeError: If filepath is not a string.
        FileNotFoundError: If the configuration file is absent.
        yaml.YAMLError: If the document is malformed.
        ValueError: If a required top-level section is missing.
    """
    # Reject a non-string path before touching the filesystem
    if not isinstance(filepath, str):
        raise TypeError(f"filepath must be a string; got {type(filepath)}.")
    # Resolve the path relative to the working directory
    config_path: Path = Path(filepath)
    # A missing configuration would silently break every downstream gate
    if not config_path.is_file():
        raise FileNotFoundError(
            f"Configuration file '{filepath}' not found in the working "
            f"directory {Path.cwd()}."
        )
    try:
        # Open the document with an explicit encoding for portability
        with open(config_path, "r", encoding="utf-8") as handle:
            # safe_load constructs only plain containers and scalars
            config: Dict[str, Any] = yaml.safe_load(handle)
    except yaml.YAMLError as exc:
        # Surface the parser context rather than a bare failure
        raise yaml.YAMLError(
            f"Failed to parse YAML configuration '{filepath}': {exc}"
        ) from exc
    # The pipeline reads these sections unconditionally
    required_sections: Tuple[str, ...] = (
        "raw_data_schemas", "data_preprocessing", "network_architecture",
        "network_training", "fairness_penalty", "frontier",
        "interpretability_diagnostics", "simulation",
        "real_data_experiment", "hyperparameter_grid", "evaluation_metrics",
        "sample_controls",
    )
    # Fail loudly on a truncated configuration rather than at first use
    absent: List[str] = [s for s in required_sections if s not in config]
    if absent:
        raise ValueError(
            f"config.yaml is missing required section(s): {absent}."
        )
    # Report the inventory so the operator can see what was loaded
    print(f"Loaded configuration from {filepath}: "
          f"{len(config)} top-level sections.")
    return config


# Load the frozen study configuration from the working directory
config: Dict[str, Any] = load_study_configuration("config.yaml")

# Confirm the two gates that depend most directly on the configuration
print(f"  published datasets   : "
      f"{len(config['real_data_experiment']['datasets'])}")
print(f"  model classes        : "
      f"{config['real_data_experiment']['model_comparisons']}")
print(f"  lambda grid (sim)    : "
      f"{config['network_training']['loss']['lambda_grid_simulation']}")
print(f"  lambda grid (real)   : "
      f"{config['network_training']['loss']['lambda_grid_realdata']}")
print(f"  seeds                : {config['network_training']['num_seeds']}")
```

**Discussion of Step 2.** `yaml.safe_load` guarantees a plain object graph with no arbitrary construction. Note the file was authored with `1.0e-6` rather than `1e-6` throughout: under YAML 1.1 the latter parses as a **string**, which would silently poison every epsilon and tolerance in the pipeline. The loader fails loudly on a missing file or a truncated section rather than proceeding with an empty dictionary — consistent with the pipeline's fail-fast gates.

<br>

### **Step 3: Executing the Pipeline (`run_findr_end_to_end`)**

With the four inputs and the configuration in memory, the top-level orchestrator is invoked. All eleven parameters are covered below.

| Parameter | Type | Default | Role |
|---|---|---|---|
| `config` | `Dict[str, Any]` | *required* | The frozen configuration from Step 2 |
| `raw_structured_features_by_dataset` | `Dict[str, pd.DataFrame]` | *required* | Eight feature frames, `record_id` + features |
| `raw_outcome_labels_by_dataset` | `Dict[str, pd.DataFrame]` | *required* | Eight label frames, `record_id` + `y_default` |

**Discussion of Step 3.** The orchestrator call is deliberately thin because every intermediate callable runs internally. Three behaviours are worth stating explicitly:

- **`findr` is forced to full-batch training** regardless of the sampled `batch_size`. The annihilation property $\tilde{\mathbf{X}}^\top\mathbf{H}^\perp_{\boldsymbol{\alpha}} = \mathbf{0}$ is defined over the whole training population; a mini-batch voids it and therefore voids the coefficient interpretation the paper rests on. LR and NN take genuine mini-batches, stratified by the $(y, a)$ cell whenever $\lambda > 0$ so the Wasserstein penalty never meets an empty cell.
- **The split asymmetry is enforced, not incidental.** AUC and Brier are computed on **test**; the fairness gap $\hat{\Phi}$ and the excess log loss driving the frontier are computed on **validation**, because §3.5 defines the frontier on $\mathcal{D}_{\text{val}}$.
- **Task 12's matrix pass is idempotent** against the run log: keys already completed by the Task 11 engine are skipped, so no run is duplicated and an interrupted execution resumes rather than restarts.

> **Realism note.** With `subset=None` the full matrix is 1,680 real-data runs plus 540 simulation runs, each with a 30-candidate random search at two restarts and up to 6,000 epochs — a multi-day GPU job whose peak memory is driven by `findr`'s full-batch requirement at $n_{\text{train}} = 163{,}208$ for Vehicle Loan. The `SMOKE_SUBSET` above is the correct first execution. Separately: the synthetic frames here reproduce the published *composition* but carry no real dependence structure, so they exercise every code path without reproducing the manuscript's AUC values. Reproducing those figures requires the eight genuine provider files, which no generator can manufacture.

<br>

### **Summary of the Execution Flow**

1. **Data Ingestion.** `assemble_study_inputs(config)` produces the four raw inputs: three dictionaries keyed by the eight dataset names (`south_german`, `taiwan`, `gmsc`, `vehicle_loan`, `myhome`, `thomas`, `uk_credit`, `pakdd_2010`) totalling 502,379 rows, and a 200,000-row simulation frame covering both modes across ten seeds. Compositions are constructed by **exact count** so the Task 1 gate passes deterministically: $n$ exactly, $\hat{\pi}_{\text{default}}$ and $\hat{\pi}_{\text{sensitive}}$ within 0.005 of the published values. `raw_age` is set identically equal to `age`, and categorical columns are cast to `object` dtype.
2. **Configuration Loading.** `load_study_configuration("config.yaml")` parses the fifteen sections with `yaml.safe_load` and validates that the twelve pipeline-critical sections are present, failing loudly otherwise.
3. **Validation & Cleansing (Step 1, Tasks 1–2).** `validate_inputs` enforces `record_id` integrality, uniqueness and three-way population equality; derives $A_i = \mathbf{1}\{\text{age}_i \le 25\}$, raising on a null age; compares the observed composition against the **independent** published constants at $\pm 0.005$; and seals 38 placeholders to the audit log. `clean_raw_data` then joins on the index, creates `employment_duration_missing` **before** imputation, and normalises categorical levels.
4. **Preprocessing (Step 2, Task 3).** Stratified $0.70/0.10/0.20$ split by integer counts under seed 42; standardisation statistics, WoE maps and medians estimated on **training only** and frozen; $\mathbf{Z} = \mathbf{X}$.
5. **Simulation Preparation (Step 3, Task 4).** Per-block structure verified at $n = 10{,}000$, $p = 6$; the covariate shift verified through $\mathbb{E}[X_1|A{=}0] - \mathbb{E}[X_1|A{=}1] \approx 1$ within four analytic standard errors; `eta_true` and `r_true` verified against recomputation to $10^{-8}$.
6. **The Study Engine (Step 4, Tasks 5–10 via Task 11).** For each dataset/mode and each $\lambda$: `build_model` for LR, NN and `findr` (thin QR in float64, associative projector $\mathbf{H} - \mathbf{Q}(\mathbf{Q}^\top\mathbf{H})$, residual bias-free since $\gamma_0$ absorbs into $\beta_0$); AdamW training with early stopping on the validation total objective; AUC and Brier on test; $\hat{\Phi}$ and excess log loss on validation; EVR, DDR and CSER diagnostics; and the frontier with its chord-length refinement stage.
7. **Full Study & Robustness (Step 5, Task 12).** Matrix accounting, the placeholder sensitivity sweep ($\rho$, $\delta$, $\alpha$, WoE convention, orthogonality tolerance, CSER space and branch mode, PR fraction, alternative nonlinear specifications, supplementary criteria), and verification against the paper's reference patterns including the CSER level and Spearman monotonicity checks.
8. **Deliverables & Audit (Step 6, Task 13).** Models serialised with `findr`'s evaluation projection sealed from the final encoder; the round trip **verified by comparison** and a failure aborting publication; tables and figures rendered under the configured colour convention; and the audit report sealed with the SHA-256 of its body.

## **Implemented Callables**



````python
# Study Data Ingestion

# ==============================================================================
# Study Data Ingestion: the eight real credit datasets
# ==============================================================================

# This cell replaces the former synthetic "Study Inputs" fixture. It supplies
# a real ingestion layer: per-dataset source specifications, adapters that
# normalise each provider's schema onto the study's three-frame contract, and
# the INDEPENDENT metadata constants transcribed from the manuscript.
#
# Sourcing note, stated plainly. The manuscript enumerates the feature list
# for only one of the eight datasets; for the other seven it directs the
# reader to the original provider. No code can manufacture those files. This
# layer therefore READS them from a configured directory and fails loudly,
# naming the exact expected path and the provider, when a file is absent. It
# never fabricates a row. A study run against absent data is refused, not
# approximated.

# ------------------------------------------------------------------------------
# Independent metadata constants (Task 1 Step 2 gate)
# ------------------------------------------------------------------------------
@dataclass(frozen=True)
class DatasetMetadata:
    """
    The published composition of one real-world dataset.

    These values are transcribed from the manuscript's application section
    and are DELIBERATELY independent of any loaded file. Task 1 Step 2
    compares the observed composition against them at a tolerance of 0.005
    and halts on mismatch; that gate is meaningful only because these
    constants come from the paper rather than from the data being checked.
    Deriving them from the loaded frame, as the previous synthetic fixture
    did, reduces the check to a tautology that can never fail.

    Attributes
    ----------
    key : str
        The dictionary key used throughout the pipeline.
    display_name : str
        The dataset's name as printed in the manuscript.
    n : int
        The published sample size.
    n_features : int
        The published feature count, excluding the identifier and label.
    default_rate : float
        The published default rate (share with y = 1).
    sensitive_share : float
        The published share of the sensitive class, A = 1{age <= 25}.
    provider : str
        Where the original file is obtained.

    """

    # The pipeline dictionary key
    key: str
    # The manuscript's printed name
    display_name: str
    # Published sample size
    n: int
    # Published feature count
    n_features: int
    # Published default rate
    default_rate: float
    # Published sensitive-class share
    sensitive_share: float
    # The original data provider
    provider: str


# The eight published compositions, verbatim from the manuscript's
# application section. Each entry is an independent constant.
REAL_DATASET_METADATA: Dict[str, DatasetMetadata] = {
    "south_german": DatasetMetadata(
        "south_german", "South German Credit", 1000, 19, 0.300, 0.1900,
        "UCI Machine Learning Repository (Statlog / South German Credit)"),
    "taiwan": DatasetMetadata(
        "taiwan", "Taiwan Credit Default", 30000, 22, 0.2212, 0.1290,
        "UCI ML Repository: default of credit card clients"),
    "gmsc": DatasetMetadata(
        "gmsc", "Give Me Some Credit", 150000, 9, 0.0668, 0.0202,
        "Kaggle: GiveMeSomeCredit (cs-training.csv)"),
    "vehicle_loan": DatasetMetadata(
        "vehicle_loan", "Vehicle Car Loan", 233154, 32, 0.2171, 0.2284,
        "L&T Vehicle Loan Default Prediction (train.csv)"),
    "myhome": DatasetMetadata(
        "myhome", "MyHome Loan", 7000, 7, 0.400, 0.0703,
        "MyHome Loan dataset (provider-supplied)"),
    "thomas": DatasetMetadata(
        "thomas", "Thomas Loan", 1225, 13, 0.2637, 0.0767,
        "Thomas et al. credit scoring companion data"),
    "uk_credit": DatasetMetadata(
        "uk_credit", "UK Credit", 30000, 13, 0.040, 0.1975,
        "UK Credit dataset (provider-supplied)"),
    "pakdd_2010": DatasetMetadata(
        "pakdd_2010", "PAKDD 2010", 50000, 51, 0.2608, 0.1149,
        "PAKDD 2010 Data Mining Competition"),
}

# The canonical dataset ordering used by every report and figure
REAL_DATASET_KEYS: Tuple[str, ...] = tuple(REAL_DATASET_METADATA)


# ------------------------------------------------------------------------------
# Per-dataset source specifications
# ------------------------------------------------------------------------------
@dataclass(frozen=True)
class DatasetSourceSpec:
    """
    How to read one provider's file and map it onto the study contract.

    The study contract is three frames per dataset: structured features,
    the binary outcome, and the raw age from which A = 1{age <= 25} is
    derived. Each provider ships a different schema, so one adapter per
    dataset is required. The feature LIST is taken from the file (every
    column that is not the identifier, the label or a declared drop), while
    the feature COUNT is validated against the published metadata: that is
    the strongest check available given the manuscript does not enumerate
    the columns for seven of the eight datasets.

    Attributes
    ----------
    key : str
        The dataset key.
    filenames : Tuple[str, ...]
        Accepted file names, tried in order, relative to the data root.
    label_column : str
        The provider's outcome column.
    label_positive : Any
        The value of label_column denoting default (mapped to y = 1).
    age_column : str
        The provider's age column, in YEARS, used for the sensitive rule.
    id_column : Optional[str]
        The provider's identifier column when one exists; a positional
        identifier is generated when None.
    drop_columns : Tuple[str, ...]
        Columns excluded from the feature set (leakage, free text, or
        duplicated identifiers).
    read_kwargs : Dict[str, Any]
        Extra keyword arguments forwarded to the reader.

    """

    # The dataset key
    key: str
    # Accepted file names, tried in order
    filenames: Tuple[str, ...]
    # The provider's outcome column
    label_column: str
    # The value denoting default
    label_positive: Any
    # The provider's age column, in years
    age_column: str
    # The provider's identifier column, when one exists
    id_column: Optional[str] = None
    # Columns excluded from the feature set
    drop_columns: Tuple[str, ...] = ()
    # Reader keyword arguments
    read_kwargs: Dict[str, Any] = field(default_factory=dict)
    # Whether age is ALSO a structured feature, or serves solely as the
    # sensitive-attribute source. Task 3 Step 3 prescribes "dropping
    # identifiers and the raw age column if it is used solely as a
    # sensitive source". The published feature counts corroborate the
    # solely-sensitive reading: every provider schema that can be checked
    # against a public source exceeds the paper's count by exactly one
    # (Taiwan 23 vs 22, GMSC 10 vs 9, South German 20 vs 19). The default
    # is therefore False, and the count gate enforces the choice per file.
    age_is_feature: bool = False


# One specification per dataset. The column names are the providers' own.
# Where a provider ships several historical namings, the alternatives are
# resolved at read time by resolve_column, so a minor release change does
# not silently produce a wrong feature set.
REAL_DATASET_SOURCES: Dict[str, DatasetSourceSpec] = {
    "south_german": DatasetSourceSpec(
        key="south_german",
        filenames=("south_german_credit.csv", "SouthGermanCredit.asc",
                   "german.data"),
        label_column="kredit", label_positive=0, age_column="alter",
        read_kwargs={"sep": None, "engine": "python"}),
    "taiwan": DatasetSourceSpec(
        key="taiwan",
        filenames=("taiwan_credit_default.csv", "UCI_Credit_Card.csv",
                   "default_of_credit_card_clients.csv"),
        label_column="default.payment.next.month", label_positive=1,
        age_column="AGE", id_column="ID"),
    "gmsc": DatasetSourceSpec(
        key="gmsc",
        filenames=("gmsc_cs_training.csv", "cs-training.csv"),
        label_column="SeriousDlqin2yrs", label_positive=1,
        age_column="age", id_column="Unnamed: 0"),
    "vehicle_loan": DatasetSourceSpec(
        key="vehicle_loan",
        filenames=("vehicle_loan_train.csv", "train.csv"),
        label_column="loan_default", label_positive=1,
        age_column="Date.of.Birth", id_column="UniqueID",
        drop_columns=("DisbursalDate", "Employment.Type")),
    "myhome": DatasetSourceSpec(
        key="myhome",
        filenames=("myhome_loan.csv",),
        label_column="default", label_positive=1, age_column="age"),
    "thomas": DatasetSourceSpec(
        key="thomas",
        filenames=("thomas_loan.csv",),
        label_column="BAD", label_positive=1, age_column="AGE"),
    "uk_credit": DatasetSourceSpec(
        key="uk_credit",
        filenames=("uk_credit.csv",),
        label_column="default", label_positive=1, age_column="age"),
    "pakdd_2010": DatasetSourceSpec(
        key="pakdd_2010",
        filenames=("pakdd_2010.csv", "PAKDD2010_Modeling_Data.txt"),
        label_column="TARGET_LABEL_BAD", label_positive=1,
        age_column="AGE", id_column="ID_CLIENT",
        read_kwargs={"sep": None, "engine": "python"}),
}

# The default directory searched for the provider files. Override it by
# passing data_root explicitly, or by setting FINDR_DATA_ROOT.
DEFAULT_DATA_ROOT: Path = Path(
    os.environ.get("FINDR_DATA_ROOT", "data/real"))


def resolve_dataset_path(
    spec: DatasetSourceSpec, data_root: Path,
) -> Path:
    """
    Locate one provider's file, or fail with an actionable instruction.

    Parameters
    ----------
    spec : DatasetSourceSpec
        The dataset's source specification.
    data_root : Path
        The directory searched for the file.

    Returns
    -------
    Path
        The first accepted filename that exists.

    Raises
    ------
    FileNotFoundError
        Naming every accepted path and the provider, so the operator knows
        exactly what to obtain and where to put it.

    """
    # Each accepted name is tried in declaration order
    for name in spec.filenames:
        candidate: Path = data_root / name
        if candidate.is_file():
            return candidate
    # The failure names the provider and every path that was tried: a
    # missing dataset is a sourcing task, and the message must say so
    metadata: DatasetMetadata = REAL_DATASET_METADATA[spec.key]
    tried: str = ", ".join(str(data_root / name) for name in spec.filenames)
    raise FileNotFoundError(
        f"dataset '{spec.key}' ({metadata.display_name}, n={metadata.n}) "
        f"was not found. Tried: {tried}. Obtain the file from: "
        f"{metadata.provider}, place it under '{data_root}', and re-run. "
        f"This layer never fabricates observations."
    )


def resolve_column(frame: pd.DataFrame, name: str, role: str,
                   dataset: str) -> str:
    """
    Resolve a declared column name against the frame, case-insensitively.

    Providers vary the capitalisation and separator of their headers
    between releases. Matching exactly first, then case-insensitively, then
    ignoring separators, keeps a benign header change from being read as a
    missing column, while still failing when the column genuinely is not
    present.

    Parameters
    ----------
    frame : pd.DataFrame
        The loaded provider frame.
    name : str
        The declared column name.
    role : str
        What the column is for, used in the error message.
    dataset : str
        The dataset key, used in the error message.

    Returns
    -------
    str
        The actual column label present in the frame.

    Raises
    ------
    KeyError
        If no candidate matches.

    """
    # An exact match is authoritative
    if name in frame.columns:
        return name
    # Case-insensitive match
    lowered: Dict[str, str] = {str(c).lower(): str(c) for c in frame.columns}
    if name.lower() in lowered:
        return lowered[name.lower()]
    # Separator-insensitive match (dots, underscores, spaces, hyphens)
    def canonical(text: str) -> str:
        """Strip separators and case so headers compare structurally."""
        return re.sub(r"[^a-z0-9]", "", str(text).lower())
    canonical_map: Dict[str, str] = {
        canonical(c): str(c) for c in frame.columns}
    if canonical(name) in canonical_map:
        return canonical_map[canonical(name)]
    raise KeyError(
        f"dataset '{dataset}': the {role} column '{name}' is absent from "
        f"the source file. Present columns: {list(frame.columns)[:25]}"
        f"{' ...' if len(frame.columns) > 25 else ''}"
    )


def derive_age_years(series: pd.Series, dataset: str) -> pd.Series:
    """
    Coerce a provider's age column to integer years.

    Some providers ship a date of birth rather than an age (the vehicle
    loan file is the study's example). A date-like column is converted to
    completed years at the file's reference date; a numeric column is used
    directly. The sensitive rule A = 1{age <= 25} is defined on YEARS, so
    this normalisation must happen before any binarisation.

    Parameters
    ----------
    series : pd.Series
        The provider's age or date-of-birth column.
    dataset : str
        The dataset key, for error naming.

    Returns
    -------
    pd.Series
        Age in completed years, as a nullable float.

    Raises
    ------
    ValueError
        If the column can be interpreted as neither an age nor a date.

    """
    # An ALREADY-PARSED datetime column must be handled before the numeric
    # probe: pd.to_numeric succeeds on datetime64 by returning nanoseconds
    # since the epoch, which would silently produce ages of order 1e17.
    if pd.api.types.is_datetime64_any_dtype(series):
        reference_dt: pd.Timestamp = series.max()
        return ((reference_dt - series).dt.days / 365.25).astype(np.float64)
    # A numeric column is already an age in years
    numeric: pd.Series = pd.to_numeric(series, errors="coerce")
    if numeric.notna().mean() > 0.95:
        return numeric.astype(np.float64)
    # Otherwise attempt a date-of-birth interpretation
    parsed: pd.Series = pd.to_datetime(series, errors="coerce",
                                       dayfirst=True)
    if parsed.notna().mean() > 0.95:
        # The reference date is the file's own maximum observed date, which
        # keeps the derived age reproducible and provider-independent
        reference: pd.Timestamp = parsed.max()
        days: pd.Series = (reference - parsed).dt.days
        return (days / 365.25).astype(np.float64)
    raise ValueError(
        f"dataset '{dataset}': the age column is neither numeric nor "
        f"date-like; {float(numeric.notna().mean()):.1%} parsed as numbers "
        f"and {float(parsed.notna().mean()):.1%} as dates."
    )


def adapt_dataset_frame(
    frame: pd.DataFrame, spec: DatasetSourceSpec,
) -> Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    """
    Map one provider frame onto the study's three-frame contract.

    Produces the structured features (with record_id), the outcome labels
    (record_id, y_default) and the sensitive source (record_id, raw_age),
    exactly as Task 1 requires.

    Parameters
    ----------
    frame : pd.DataFrame
        The loaded provider frame.
    spec : DatasetSourceSpec
        The dataset's source specification.

    Returns
    -------
    Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame]
        (features, labels, sensitive), all keyed by record_id.

    Raises
    ------
    KeyError
        If a declared column is absent.
    ValueError
        If the label is not binary after mapping.

    """
    working: pd.DataFrame = frame.copy()
    dataset: str = spec.key
    # Resolve the three structural columns against the actual header
    label_column: str = resolve_column(
        working, spec.label_column, "label", dataset)
    age_column: str = resolve_column(
        working, spec.age_column, "age", dataset)
    # The identifier is the provider's when it has one, positional otherwise
    if spec.id_column is not None and spec.id_column in working.columns:
        record_id: np.ndarray = pd.to_numeric(
            working[spec.id_column], errors="coerce"
        ).to_numpy(dtype=np.float64)
        # A non-integral or duplicated provider id is replaced by a
        # positional one rather than silently corrupting the join key
        if (not np.isfinite(record_id).all()
                or np.any(record_id != np.rint(record_id))
                or len(np.unique(record_id)) != len(record_id)):
            logger.warning(
                f"dataset '{dataset}': the provider identifier "
                f"'{spec.id_column}' is not a unique integer; a positional "
                f"record_id is generated instead.")
            record_id = np.arange(len(working), dtype=np.float64)
    else:
        record_id = np.arange(len(working), dtype=np.float64)
    identifiers: np.ndarray = record_id.astype(np.int64)
    # Equation (Section 5.1 source): y = 1 denotes DEFAULT. The provider's
    # positive value is mapped explicitly, because several files encode
    # the GOOD outcome as 1 (South German being the study's example)
    raw_label: pd.Series = working[label_column]
    y_default: np.ndarray = (
        raw_label.astype(object) == spec.label_positive
    ).to_numpy().astype(np.int64)
    if set(np.unique(y_default)) - {0, 1}:
        raise ValueError(
            f"dataset '{dataset}': the label did not reduce to {{0, 1}}.")
    # The raw age in years, from which A = 1{age <= 25} is derived
    raw_age: pd.Series = derive_age_years(working[age_column], dataset)
    # The feature set is every remaining column: the manuscript does not
    # enumerate the columns for seven of eight datasets, so the file is
    # authoritative and the COUNT is validated downstream
    excluded: Set[str] = {label_column}
    excluded.update(
        c for c in spec.drop_columns if c in working.columns)
    # Age is excluded from the FEATURES when it serves solely as the
    # sensitive source; it always survives in the `sensitive` frame, so the
    # rule A = 1{age <= 25} is unaffected either way
    if not spec.age_is_feature:
        excluded.add(age_column)
    if spec.id_column is not None and spec.id_column in working.columns:
        excluded.add(spec.id_column)
    feature_columns: List[str] = [
        str(c) for c in working.columns if str(c) not in excluded]
    features: pd.DataFrame = working[feature_columns].copy()
    features.insert(0, "record_id", identifiers)
    labels: pd.DataFrame = pd.DataFrame({
        "record_id": identifiers, "y_default": y_default})
    sensitive: pd.DataFrame = pd.DataFrame({
        "record_id": identifiers,
        "raw_age": np.rint(raw_age.to_numpy(dtype=np.float64)
                           ).astype(np.int64)})
    return features, labels, sensitive


def validate_ingested_feature_count(
    key: str, features: pd.DataFrame, tolerance: int = 0,
) -> None:
    """
    Check the loaded feature count against the published count.

    The manuscript publishes a feature count for all eight datasets while
    enumerating the columns for only one, so the COUNT is the strongest
    structural check available on an ingested file. A mismatch means the
    wrong file, the wrong release, or a drop list that no longer matches
    the provider's schema -- each of which silently changes the study
    population and must fail loudly rather than propagate.

    Parameters
    ----------
    key : str
        The dataset key.
    features : pd.DataFrame
        The adapted feature frame, including the record_id column.
    tolerance : int, optional
        Permitted absolute deviation in the count, by default 0.

    Returns
    -------
    None

    Raises
    ------
    ValueError
        If the observed count deviates beyond the tolerance.

    """
    # record_id is structural and is never counted as a feature
    observed: int = int(features.shape[1] - 1)
    expected: int = REAL_DATASET_METADATA[key].n_features
    if abs(observed - expected) > tolerance:
        raise ValueError(
            f"dataset '{key}': loaded {observed} feature(s) against the "
            f"published {expected}. The file, its release, or the declared "
            f"drop columns do not match the study population. Columns "
            f"loaded: {[c for c in features.columns if c != 'record_id'][:30]}"
        )


def load_real_dataset(
    key: str, data_root: Path = DEFAULT_DATA_ROOT,
) -> Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    """
    Load and adapt one real dataset from the configured data root.

    Parameters
    ----------
    key : str
        One of the eight dataset keys.
    data_root : Path, optional
        The directory holding the provider files.

    Returns
    -------
    Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame]
        (features, labels, sensitive) for that dataset.

    Raises
    ------
    KeyError
        If the key is not one of the eight study datasets.
    FileNotFoundError
        If the provider file is absent.

    """
    # The key domain is the eight study datasets
    if key not in REAL_DATASET_SOURCES:
        raise KeyError(
            f"'{key}' is not a study dataset; expected one of "
            f"{list(REAL_DATASET_KEYS)}.")
    spec: DatasetSourceSpec = REAL_DATASET_SOURCES[key]
    path: Path = resolve_dataset_path(spec, data_root)
    # The reader is chosen by suffix; csv covers every study provider,
    # with whitespace/semicolon files handled by the sniffing engine
    logger.info(f"dataset '{key}': reading {path}")
    frame: pd.DataFrame = pd.read_csv(path, **spec.read_kwargs)
    return adapt_dataset_frame(frame, spec)


def ingest_real_datasets(
    data_root: Path = DEFAULT_DATA_ROOT,
    datasets: Optional[Sequence[str]] = None,
    strict: bool = True,
) -> Tuple[Dict[str, pd.DataFrame], Dict[str, pd.DataFrame],
           Dict[str, pd.DataFrame], Dict[str, Any]]:
    """
    Produce the three study dicts keyed by the eight real dataset names.

    Parameters
    ----------
    data_root : Path, optional
        The directory holding the provider files.
    datasets : Optional[Sequence[str]], optional
        A subset of keys to load; all eight by default.
    strict : bool, optional
        When True (default) a missing or non-conforming file raises. When
        False the dataset is skipped and recorded in the report, so an
        operator can inventory what is available before sourcing the rest.

    Returns
    -------
    Tuple[Dict, Dict, Dict, Dict]
        raw_structured_features_by_dataset, raw_outcome_labels_by_dataset,
        raw_sensitive_attribute_by_dataset, and an ingestion report naming
        what loaded, what did not, and why.

    Raises
    ------
    FileNotFoundError
        Under strict mode, when a provider file is absent.

    """
    requested: List[str] = list(datasets) if datasets is not None \
        else list(REAL_DATASET_KEYS)
    features_by: Dict[str, pd.DataFrame] = {}
    labels_by: Dict[str, pd.DataFrame] = {}
    sensitive_by: Dict[str, pd.DataFrame] = {}
    report: Dict[str, Any] = {
        "data_root": str(data_root), "loaded": [], "missing": [],
        "observed": {}, "expected": {},
    }
    for key in requested:
        metadata: DatasetMetadata = REAL_DATASET_METADATA[key]
        # The published composition is recorded for every requested key,
        # loaded or not, so the report shows the full target
        report["expected"][key] = {
            "n": metadata.n, "features": metadata.n_features,
            "default_rate": metadata.default_rate,
            "sensitive_share": metadata.sensitive_share}
        try:
            features, labels, sensitive = load_real_dataset(key, data_root)
        except (FileNotFoundError, KeyError, ValueError) as error:
            # A missing dataset is reported, never silently substituted
            report["missing"].append({"dataset": key, "reason": str(error)})
            if strict:
                raise
            logger.warning(f"dataset '{key}': skipped ({error})")
            continue
        # The published feature count is a structural gate on the file
        try:
            validate_ingested_feature_count(key, features)
        except ValueError as error:
            report["missing"].append({"dataset": key, "reason": str(error)})
            if strict:
                raise
            logger.warning(f"dataset '{key}': skipped ({error})")
            continue
        features_by[key] = features
        labels_by[key] = labels
        sensitive_by[key] = sensitive
        # The observed composition, for the operator's inventory. The
        # authoritative comparison remains Task 1 Step 2, which uses the
        # independent constants above
        observed_share: float = float(
            (sensitive["raw_age"].to_numpy() <= 25).mean())
        report["observed"][key] = {
            "n": int(len(features)),
            "features": int(features.shape[1] - 1),
            "default_rate": float(labels["y_default"].mean()),
            "sensitive_share": observed_share}
        report["loaded"].append(key)
        logger.info(
            f"dataset '{key}': n={len(features)}, "
            f"features={features.shape[1] - 1}, "
            f"default={labels['y_default'].mean():.4f}, "
            f"sensitive={observed_share:.4f}")
    return features_by, labels_by, sensitive_by, report


def build_schema_lists(
    features_by_dataset: Dict[str, pd.DataFrame],
) -> Tuple[Dict[str, List[str]], Dict[str, List[str]]]:
    """
    Derive the continuous and categorical feature lists per dataset.

    The manuscript enumerates the columns for only one of the eight
    datasets, so the dtype of the loaded column is the authority: an object
    or low-cardinality integer column is categorical (and is WoE encoded),
    everything else is continuous (and is standardised).

    Parameters
    ----------
    features_by_dataset : Dict[str, pd.DataFrame]
        The ingested feature frames.

    Returns
    -------
    Tuple[Dict[str, List[str]], Dict[str, List[str]]]
        (continuous, categorical) feature names per dataset.

    """
    continuous: Dict[str, List[str]] = {}
    categorical: Dict[str, List[str]] = {}
    for key, frame in features_by_dataset.items():
        continuous_names: List[str] = []
        categorical_names: List[str] = []
        for column in frame.columns:
            # The identifier is structural, never a feature
            if column == "record_id":
                continue
            series: pd.Series = frame[column]
            # A non-numeric column is categorical by construction
            if series.dtype == object or isinstance(
                    series.dtype, pd.CategoricalDtype):
                categorical_names.append(str(column))
                continue
            # A low-cardinality integer column is a coded category, which
            # WoE handles correctly and standardisation does not
            unique_count: int = int(series.nunique(dropna=True))
            if pd.api.types.is_integer_dtype(series) and unique_count <= 15:
                categorical_names.append(str(column))
            else:
                continuous_names.append(str(column))
        continuous[key] = continuous_names
        categorical[key] = categorical_names
    return continuous, categorical
````

````python
# Study Simulation and Configuration

# ==============================================================================
# Study Simulation: the simple and complex data-generating processes
# ==============================================================================

# The DGP of Section 5.1, both modes. The previous fixture emitted the
# simple mode only, with the nonlinear and interaction components hard-wired
# to zero, so the complex mode -- which is what tests whether the neural
# residual recovers nonlinear signal -- did not exist at all.
#
# Equations (Section 5.1):
#     A  ~ Bern(0.3),        X_j ~ N(0, 1)  for j = 1..6
#     X_1 <- X_1 + (1 - A)                      (covariate shift)
#     B(A) = 0.2 * 1{A = 1}                     (baseline logit shift)
#     L    = b_0 + beta' X,  b_0 = -2,
#            beta = (-1, -0.5, 0, 0.25, 0.5, 1)'
#     NL(X) = sum_{k in I_nl} g_k(X_k)
#     I(X)  = sum_{(i,j) in I_int} X_i X_j
#     simple : eta = L + B(A)
#     complex: eta = L + NL(X) + I(X) + B(A)
#     Y ~ Bern(sigma(eta))
#
# The functional forms of g_k and the index sets are UNSPECIFIED IN
# MANUSCRIPT; the values below are the registered placeholders and are
# carried in the Task 1 audit and the Task 12 sensitivity sweep.

# The DGP constants fixed by the manuscript
SIM_N_OBSERVATIONS: int = 10000
SIM_P_COVARIATES: int = 6
SIM_SENSITIVE_P: float = 0.3
SIM_INTERCEPT_B0: float = -2.0
SIM_BETA: Tuple[float, ...] = (-1.0, -0.5, 0.0, 0.25, 0.5, 1.0)
SIM_BASELINE_SHIFT: float = 0.2
SIM_COVARIATE_SHIFT_VALUE: float = 1.0
SIM_MODES: Tuple[str, ...] = ("simple", "complex")
SIM_SEEDS: Tuple[int, ...] = tuple(range(10))

# UNSPECIFIED IN MANUSCRIPT: the univariate nonlinear index set and forms.
# These are bounded, zero-centred and of comparable magnitude to the linear
# terms, so the complex mode carries genuine nonlinear signal without
# swamping the structured component -- which is what the manuscript's
# reported EVR band of roughly 0.44 to 0.56 requires.
SIM_NONLINEAR_INDEX_SET: Tuple[int, ...] = (1, 2, 6)
# UNSPECIFIED IN MANUSCRIPT: the interaction index pairs.
SIM_INTERACTION_PAIRS: Tuple[Tuple[int, int], ...] = ((1, 2), (3, 6))


def default_nonlinear_functions() -> Dict[int, Any]:
    """
    Return the registered univariate nonlinear terms g_k.

    UNSPECIFIED IN MANUSCRIPT. The manuscript states only that
    NL(X) = sum_{k in I_nl} g_k(X_k) without giving the forms. The
    registered placeholders are

        g_1(X_1) = sin(2 pi X_1),
        g_2(X_2) = X_2^2 - 1,
        g_6(X_6) = cos(2 pi X_6),

    chosen so that each term is bounded (the trigonometric pair) or
    centred (the quadratic, whose expectation under N(0,1) is one, hence
    the subtraction). Centring matters: an uncentred term shifts the
    unconditional default rate away from the simple mode and would confound
    the mode comparison the study rests on.

    Returns
    -------
    Dict[int, Any]
        A map from the 1-based covariate index to its callable.

    """
    # Each callable is vectorised over the column it receives
    return {
        1: lambda x: np.sin(2.0 * np.pi * x),
        2: lambda x: x ** 2 - 1.0,
        6: lambda x: np.cos(2.0 * np.pi * x),
    }


def compute_nonlinear_component(
    covariates: np.ndarray,
    index_set: Sequence[int] = SIM_NONLINEAR_INDEX_SET,
    functions: Optional[Dict[int, Any]] = None,
) -> np.ndarray:
    """
    Evaluate NL(X) = sum_{k in I_nl} g_k(X_k).

    Parameters
    ----------
    covariates : np.ndarray
        The POST-SHIFT covariate matrix, shape (n, 6). The shift is applied
        before every logit component, so NL sees the shifted X_1.
    index_set : Sequence[int], optional
        The 1-based indices carrying a nonlinear term.
    functions : Optional[Dict[int, Any]], optional
        The term map; the registered placeholders by default.

    Returns
    -------
    np.ndarray
        The nonlinear contribution per row, shape (n,).

    Raises
    ------
    ValueError
        If an index lies outside 1..p or has no registered function.

    """
    resolved: Dict[int, Any] = functions if functions is not None \
        else default_nonlinear_functions()
    # The accumulator starts at zero: an empty index set gives NL = 0,
    # which is exactly the simple mode
    component: np.ndarray = np.zeros(covariates.shape[0], dtype=np.float64)
    for index in index_set:
        # The manuscript indexes covariates from 1; the array from 0
        if not 1 <= index <= covariates.shape[1]:
            raise ValueError(
                f"nonlinear index {index} is outside 1..{covariates.shape[1]}.")
        if index not in resolved:
            raise ValueError(
                f"no nonlinear function registered for index {index}; "
                f"registered: {sorted(resolved)}.")
        # Equation: g_k(X_k), evaluated on the post-shift column
        component = component + np.asarray(
            resolved[index](covariates[:, index - 1]), dtype=np.float64)
    return component


def compute_interaction_component(
    covariates: np.ndarray,
    index_pairs: Sequence[Tuple[int, int]] = SIM_INTERACTION_PAIRS,
) -> np.ndarray:
    """
    Evaluate I(X) = sum_{(i,j) in I_int} X_i X_j.

    Parameters
    ----------
    covariates : np.ndarray
        The POST-SHIFT covariate matrix, shape (n, 6).
    index_pairs : Sequence[Tuple[int, int]], optional
        The 1-based index pairs carrying a product term.

    Returns
    -------
    np.ndarray
        The interaction contribution per row, shape (n,).

    Raises
    ------
    ValueError
        If a pair references an index outside 1..p.

    """
    component: np.ndarray = np.zeros(covariates.shape[0], dtype=np.float64)
    for left, right in index_pairs:
        # Both indices must address a real covariate
        for index in (left, right):
            if not 1 <= index <= covariates.shape[1]:
                raise ValueError(
                    f"interaction index {index} is outside "
                    f"1..{covariates.shape[1]}.")
        # Equation: the elementwise product X_i X_j
        component = component + (
            covariates[:, left - 1] * covariates[:, right - 1])
    return component


def simulate_dgp_block(
    mode: str,
    seed: int,
    n_observations: int = SIM_N_OBSERVATIONS,
    nonlinear_index_set: Sequence[int] = SIM_NONLINEAR_INDEX_SET,
    interaction_pairs: Sequence[Tuple[int, int]] = SIM_INTERACTION_PAIRS,
    nonlinear_functions: Optional[Dict[int, Any]] = None,
) -> pd.DataFrame:
    """
    Generate one (mode, seed) block of the simulation DGP.

    Implements Section 5.1 exactly, for both modes. Every column the Task 4
    validators require is emitted, and the emitted eta_true and r_true are
    the values the validators recompute and check to 1e-8.

    Parameters
    ----------
    mode : str
        "simple" (eta = L + B(A)) or "complex"
        (eta = L + NL(X) + I(X) + B(A)).
    seed : int
        The Monte Carlo replication seed.
    n_observations : int, optional
        Rows per block, by default 10,000.
    nonlinear_index_set : Sequence[int], optional
        The nonlinear index set, complex mode only.
    interaction_pairs : Sequence[Tuple[int, int]], optional
        The interaction pairs, complex mode only.
    nonlinear_functions : Optional[Dict[int, Any]], optional
        The nonlinear term map; the registered placeholders by default.

    Returns
    -------
    pd.DataFrame
        One block carrying mode, seed, A, X1_post_shift, X2..X6, Y,
        linear_eta, nl_component, interaction_component, eta_true, r_true.

    Raises
    ------
    ValueError
        If the mode is unknown or n_observations is not positive.

    """
    # The mode domain is fixed by the manuscript
    if mode not in SIM_MODES:
        raise ValueError(
            f"mode must be one of {list(SIM_MODES)}; received '{mode}'.")
    if n_observations <= 0:
        raise ValueError(
            f"n_observations must be positive; received {n_observations}.")
    # A per-block generator makes each (mode, seed) independently
    # reproducible without depending on the ambient global RNG
    generator: np.random.Generator = np.random.default_rng(
        abs(hash((mode, int(seed)))) % (2 ** 32))
    # Equation: A ~ Bern(0.3)
    a_values: np.ndarray = generator.binomial(
        1, SIM_SENSITIVE_P, n_observations).astype(np.int64)
    # Equation: X_j ~ N(0, 1) for j = 1..6
    covariates: np.ndarray = generator.standard_normal(
        (n_observations, SIM_P_COVARIATES))
    # Equation: X_1 <- X_1 + (1 - A), the group-dependent covariate shift
    covariates[:, 0] = covariates[:, 0] + SIM_COVARIATE_SHIFT_VALUE * (
        1 - a_values)
    # Equation: L = b_0 + beta' X, on the POST-SHIFT covariates
    linear: np.ndarray = SIM_INTERCEPT_B0 + covariates @ np.asarray(
        SIM_BETA, dtype=np.float64)
    if mode == "complex":
        # Equation: NL(X) and I(X), complex mode only
        nonlinear: np.ndarray = compute_nonlinear_component(
            covariates, nonlinear_index_set, nonlinear_functions)
        interaction: np.ndarray = compute_interaction_component(
            covariates, interaction_pairs)
    else:
        # Simple mode: both structural components are identically zero
        nonlinear = np.zeros(n_observations, dtype=np.float64)
        interaction = np.zeros(n_observations, dtype=np.float64)
    # Equation: B(A) = 0.2 * 1{A = 1}
    baseline: np.ndarray = SIM_BASELINE_SHIFT * (a_values == 1).astype(
        np.float64)
    # Equation: eta = L + NL + I + B(A)
    eta: np.ndarray = linear + nonlinear + interaction + baseline
    # Equation: r = sigma(eta), computed by the stable branchwise sigmoid
    risk: np.ndarray = stable_sigmoid(eta)
    # Equation: Y ~ Bern(sigma(eta))
    outcomes: np.ndarray = generator.binomial(1, risk).astype(np.int64)
    # The frame carries every column the Task 4 validators recompute
    return pd.DataFrame({
        "mode": mode,
        "seed": int(seed),
        "A": a_values,
        "X1_post_shift": covariates[:, 0],
        "X2": covariates[:, 1], "X3": covariates[:, 2],
        "X4": covariates[:, 3], "X5": covariates[:, 4],
        "X6": covariates[:, 5],
        "Y": outcomes,
        "linear_eta": linear,
        "nl_component": nonlinear,
        "interaction_component": interaction,
        "eta_true": eta,
        "r_true": risk,
    })


def build_simulation_input(
    modes: Sequence[str] = SIM_MODES,
    seeds: Sequence[int] = SIM_SEEDS,
    n_observations: int = SIM_N_OBSERVATIONS,
    **block_kwargs: Any,
) -> pd.DataFrame:
    """
    Assemble the full simulation input across every mode and seed.

    Parameters
    ----------
    modes : Sequence[str], optional
        The DGP modes, both by default.
    seeds : Sequence[int], optional
        The replication seeds, 0..9 by default.
    n_observations : int, optional
        Rows per block, by default 10,000.
    **block_kwargs : Any
        Forwarded to simulate_dgp_block (index sets, term map).

    Returns
    -------
    pd.DataFrame
        The concatenated blocks: len(modes) * len(seeds) * n rows.

    """
    blocks: List[pd.DataFrame] = []
    # itertools.product keeps the (mode, seed) coverage explicit and total
    for mode, seed in itertools.product(modes, seeds):
        blocks.append(simulate_dgp_block(
            mode, seed, n_observations, **block_kwargs))
    frame: pd.DataFrame = pd.concat(blocks, ignore_index=True)
    logger.info(
        f"simulation input: {len(modes)} mode(s) x {len(seeds)} seed(s) "
        f"x {n_observations} rows = {len(frame)} rows.")
    return frame


# ==============================================================================
# Study Configuration: the full experimental matrix
# ==============================================================================

def build_real_data_experiment_metadata() -> Dict[str, Dict[str, Any]]:
    """
    Build the dataset metadata block from the INDEPENDENT constants.

    Task 1 Step 2 compares the observed composition of each loaded dataset
    against this block at a tolerance of 0.005 and halts on mismatch. The
    values are transcribed from the manuscript and are never derived from
    the loaded data; that independence is the entire point of the gate.

    Returns
    -------
    Dict[str, Dict[str, Any]]
        Per-dataset n, feature count, default rate and sensitive share.

    """
    # Each entry mirrors REAL_DATASET_METADATA, which is the single source
    return {
        key: {"n": meta.n, "features": meta.n_features,
              "default_rate": meta.default_rate,
              "sensitive_share": meta.sensitive_share}
        for key, meta in REAL_DATASET_METADATA.items()
    }


def build_study_config(
    dataset_keys: Optional[Sequence[str]] = None,
) -> Dict[str, Any]:
    """
    Build the frozen study configuration for the full experimental matrix.

    Every grid is the manuscript's full grid rather than the single-point
    reduction the previous fixture used: the lambda paths span 0 to 4 in
    simulation and 0 to 2 on real data, the hyperparameter grid is the
    complete Cartesian source for the random search, and the model set is
    all three classes.

    Parameters
    ----------
    dataset_keys : Optional[Sequence[str]], optional
        Restrict the dataset metadata block to these keys; all eight by
        default. Restricting is for partial-inventory runs and is recorded
        in the audit.

    Returns
    -------
    Dict[str, Any]
        The study configuration consumed by every task.

    """
    # The metadata block is independent of any loaded frame
    datasets: Dict[str, Dict[str, Any]] = build_real_data_experiment_metadata()
    if dataset_keys is not None:
        datasets = {k: v for k, v in datasets.items() if k in set(dataset_keys)}
    return {
        "raw_data_schemas": {
            "raw_structured_features_by_dataset": {"record_id": "int64"},
            "raw_outcome_labels_by_dataset": {"record_id": "int64"},
            "raw_sensitive_attribute_by_dataset": {"record_id": "int64"},
            "raw_simulation_input": {"mode": "object", "seed": "int64"},
            "raw_neural_inputs": "Z=X",
            "reference_risk_vector": "ndarray"},
        "data_preprocessing": {
            "train_val_test_split": {
                "train_ratio": 0.70, "val_ratio": 0.10, "test_ratio": 0.20,
                "stratify_on_y": True, "seed": 42,
                "minority_class_warning": MINORITY_CLASS_WARNING_THRESHOLD},
            "categorical_encoding": {
                "method": "weight_of_evidence",
                "unknown_mapping": "global_woe",
                "zero_frequency_smoothing": WOE_SMOOTHING_ALPHA,
                "woe_convention": WOE_CONVENTION},
            "continuous_scaling": {"method": "standardisation",
                                   "statistics_source": "training_set"},
            "missing_value_handling": {
                "employment_duration_imputation": "median",
                "missing_indicator_suffix": "_missing"},
            "sensitive_attribute": {
                "feature": "age", "binarisation_threshold": 25,
                "positive_group_condition": "age <= 25",
                "age_is_structured_feature": False}},
        "network_architecture": {
            "structured_component": {"include_intercept": True,
                                     "structured_features": "derived"},
            "neural_encoder": {"hidden_layer_widths": [16, 8],
                               "activation": "ReLU", "dropout_rate": 0.1,
                               "num_hidden_layers": 2, "input_dim": "derived",
                               "output_dim": 8},
            "residual_layer": {"is_linear": True, "include_intercept": True,
                               "weights_initialisation": "xavier_uniform"},
            "orthogonalisation": {
                "method": "thin_qr", "projector": "P_perp",
                "requires_full_column_rank": True, "add_intercept_to_X": True,
                "relative_tolerance": ORTHOGONALITY_RELATIVE_TOLERANCE}},
        "network_training": {
            "optimizer": "AdamW", "learning_rate": 1e-3,
            "learning_rate_grid": [3e-4, 1e-3, 3e-3, 1e-2],
            "weight_decay": 0.0,
            "weight_decay_grid": [0, 1e-5, 1e-4, 1e-3],
            "batch_size": 128, "batch_size_grid": [32, 64, 128, 256],
            "max_epochs": 6000, "early_stopping_patience": 20,
            "num_seeds": 10, "num_restarts_per_config": 2, "seed": 42,
            "loss": {
                "prediction_type": "binary_cross_entropy_with_logits",
                "fairness_multiplier": 0.0,
                # The manuscript's full paths: 0 to 4 in simulation,
                # 0 to 2 on real data
                "lambda_grid_simulation": [0, 0.01, 0.05, 0.1, 0.5, 1, 2, 3, 4],
                "lambda_grid_realdata": [0, 0.01, 0.05, 0.1, 0.5, 1, 2]}},
        "fairness_penalty": {
            "active_criterion": "equalised_odds", "wasserstein_order": 1,
            "p": 1,
            "alternative_criteria": ["independence", "equal_opportunity"],
            "quantile_grid_bins": "data_dependent",
            "cdf_implementation": "empirical_cdf", "differentiable": True},
        "frontier": {
            "xi_grid": {"include_zero": True, "logspace_start": -2,
                        "logspace_end": 2, "num_points": 40},
            "solver": {
                "step_size_rho": DEFAULT_RHO,
                "clipping_epsilon": DEFAULT_EPSILON,
                "max_iterations": DEFAULT_MAX_ITERATIONS,
                "convergence_tolerance": DEFAULT_CONVERGENCE_TOLERANCE,
                "use_polyak_ruppert_average": True,
                "candidate_stride": CANDIDATE_STRIDE,
                "polyak_ruppert_fraction": PR_AVERAGING_FRACTION,
                "initial_candidates": ["r", "fitted_model_scores",
                                       "previous_multiplier_solution"]},
            "refinement": {
                "enabled": FRONTIER_REFINEMENT_ENABLED,
                "max_additional_multipliers":
                    FRONTIER_MAX_ADDITIONAL_MULTIPLIERS,
                "compute_log_midpoints": True,
                "node_chord_length_threshold": FRONTIER_NODE_CHORD_THRESHOLD},
            "reference_risk": {
                "from_simulation": "true_conditional_risk",
                "from_realdata": "unpenalised_model_scores",
                "unpenalised_model": "logistic_regression_at_lambda_0"}},
        "interpretability_diagnostics": {
            "decision_threshold": DEFAULT_DECISION_THRESHOLD,
            "csert_perturbation_delta": DEFAULT_CSER_DELTA,
            "cser_perturbation_space": CSER_PERTURBATION_SPACE,
            "cser_branch_mode": CSER_BRANCH_MODE,
            "logit_scale": True, "use_sample_variance": True,
            "features_with_nonzero_beta": "derived"},
        "simulation": {
            "n_observations": SIM_N_OBSERVATIONS,
            "p_covariates": SIM_P_COVARIATES,
            "sensitive_A": {"distribution": "bernoulli", "p": SIM_SENSITIVE_P},
            "covariate_shift": {
                "feature": "X1", "shift_value": SIM_COVARIATE_SHIFT_VALUE,
                "tolerance_se_multiplier": COVARIATE_SHIFT_SE_MULTIPLIER},
            "logit_shift": {"label": "B(A)", "value": SIM_BASELINE_SHIFT},
            "linear_component": {"intercept_b0": SIM_INTERCEPT_B0,
                                 "beta": list(SIM_BETA)},
            "nonlinear_component": {
                "enabled": True,
                "functions": {"X1": "sin(2*pi*X1)", "X2": "X2**2 - 1",
                              "X6": "cos(2*pi*X6)"},
                "index_set_I_nl": list(SIM_NONLINEAR_INDEX_SET)},
            "interaction_component": {
                "enabled": True,
                "index_pairs_I_int": [list(p) for p in SIM_INTERACTION_PAIRS]},
            "modes": list(SIM_MODES),
            "outcome_generation": {"link": "sigmoid"}},
        "real_data_experiment": {
            # The INDEPENDENT metadata block: the Task 1 Step 2 gate
            "datasets": datasets,
            "model_comparisons": ["LR", "NN", "findr"],
            "reference_baseline": "CatBoost", "repetitions": 10,
            "metric_evaluation_split": "test",
            "frontier_evaluation_split": "validation",
            "material_relative_change": MATERIAL_RELATIVE_CHANGE},
        "hyperparameter_grid": {
            "learning_rate": [3e-4, 1e-3, 3e-3, 1e-2],
            "weight_decay": [0, 1e-5, 1e-4, 1e-3],
            "batch_size": [32, 64, 128, 256],
            "hidden_layer_widths": [[4], [16, 8], [64, 32], [128, 64]],
            "activation": ["ReLU", "GELU", "tanh"],
            "dropout_rate": [0, 0.1, 0.25]},
        "llm_and_prompt_settings": {"used": False, "llm_name": None,
                                    "prompt_templates": None},
        "plot_construction": {
            "figsize": [12, 8], "dpi": 150,
            "colours": {"findr": "blue", "LR": "green", "NN": "orange",
                        "frontier": "black", "reference": "grey"},
            "lambda_colormap": "viridis"},
        "methodological_notes": {
            "identifiability_guarantee": "X_tilde^T H_perp = 0",
            "gamma0_absorption": "gamma_0 absorbed into beta_0",
            "fairness_on_full_score": True,
            "frontier_is_benchmark": True,
            "diagnostics_are_empirical": True},
        "evaluation_metrics": {
            "reporting_metrics": ["auc", "brier", "excess_log_loss",
                                  "fairness_gap", "accuracy_gap"],
            "aggregation": "mean_and_sd_across_10_seeds",
            "score_clip_epsilon": SCORE_CLIP_EPSILON,
            "nonlinear_regime_gap": NONLINEAR_REGIME_GAP},
        "sample_controls": {
            "verification_targets": {
                "tolerance": 0.05,
                # Scoped targets: each applies only to the run it names
                "taiwan_findr_auc": {
                    "record_key": "test_auc", "expected": 0.774,
                    "tolerance": 0.02, "dataset": "taiwan",
                    "model_class": "findr", "lambda": 0.0},
                "gmsc_findr_auc": {
                    "record_key": "test_auc", "expected": 0.850,
                    "tolerance": 0.02, "dataset": "gmsc",
                    "model_class": "findr", "lambda": 0.0},
                "south_german_findr_auc": {
                    "record_key": "test_auc", "expected": 0.823,
                    "tolerance": 0.03, "dataset": "south_german",
                    "model_class": "findr", "lambda": 0.0}}},
    }
````

````python
# Task 1: Validate Ingested Parameters, Data Structures, and Configuration

# ==============================================================================
# Task 1: Validate Ingested Parameters, Data Structures, and Configuration
# ==============================================================================

# Configure the module-level logger for stage diagnostics
logger = logging.getLogger(__name__)

# The project root, for the placeholder audit artifact
PROJECT_ROOT: Path = Path.cwd()
# The default audit log path for placeholder parameters
PLACEHOLDER_AUDIT_PATH: Path = PROJECT_ROOT / "placeholders_audit.json"

def get_binarisation_threshold(config: Dict[str, Any]) -> int:
    """
    Read the sensitive-attribute binarisation threshold from config.

    A single point of truth for the rule A = 1{age <= threshold}: every
    derivation (Tasks 2 and 3) imports this helper instead of any module
    constant, so a config change propagates everywhere.

    Parameters
    ----------
    config : Dict[str, Any]
        The study configuration.

    Returns
    -------
    int
        The binarisation threshold (25 in the study).

    Raises
    ------
    KeyError
        If the config path is absent.

    """
    threshold: Any = config["data_preprocessing"]["sensitive_attribute"][
        "binarisation_threshold"]
    if not isinstance(threshold, int) or threshold < 0:
        raise ValueError(
            f"binarisation_threshold must be a non-negative integer; "
            f"found {threshold!r}."
        )
    return threshold

# The study's registered methodological decisions. Every entry is a
# documented engineering decision with its mathematical justification;
# the audit log carries them verbatim so they survive into the Task 13
# reproducibility report.
METHODOLOGICAL_DECISIONS: Tuple[Dict[str, str], ...] = (
    {
        "id": "wasserstein_strip_convention",
        "decision": (
            "The one-dimensional Wasserstein penalty (Equation (5)) is "
            "computed as the EXACT integral of the absolute inverse-CDF "
            "gap: the empirical quantile function is constant almost "
            "everywhere on each strip (q_{k-1}, q_k], so its a.e. value "
            "equals the left-continuous quantile AT the strip's right "
            "endpoint q_k. Evaluating at the literal left endpoint q_{k-1} "
            "returns the previous strip's measure-zero jump value and is "
            "NOT the exact distance (measured ~3% deviation against "
            "scipy). The convention is registered as a documented "
            "methodological decision: frontier and penalty values are "
            "EXACT one-dimensional Wasserstein distances, verified "
            "against scipy.stats.wasserstein_distance to 1e-6."
        ),
        "scope": "accuracy-fairness frontier and equalised-odds penalty",
        "verified_against": "scipy.stats.wasserstein_distance (1e-6), "
                            "hand-derived examples, tie-heavy tests",
    },
)
# The absolute tolerance applied to metadata statistics comparisons
STATISTICS_TOLERANCE: float = 0.005
# The sensitive-attribute binarisation threshold: A = 1 if age <= 25
SENSITIVE_BINARISATION_THRESHOLD: int = 25

# ==============================================================================
# REMEDIATION CONSTANTS (registered placeholders; see B-12)
# ==============================================================================
# B-1(a): the annihilation check is RELATIVE, never absolute. The residual
# ||X~^T H_perp|| accumulates n round-off terms and therefore grows like
# sqrt(n); an absolute bound is a scale bug that blocks every large dataset.
# The dimensionless threshold is compared against the residual divided by
# the Frobenius norm of the operand.
ORTHOGONALITY_RELATIVE_TOLERANCE: float = 1e-5
# B-1(b): the self-calibrating floor. sqrt(n) * eps(dtype) is the analytic
# round-off floor of an inner product accumulated over n terms; the check
# accepts max(relative_tolerance, this floor) so it can never be tighter
# than the arithmetic itself permits.
ORTHOGONALITY_FLOOR_MULTIPLIER: float = 16.0
# B-1(c): the QR factorisation, its buffers and both verification norms are
# always computed in double precision, independently of the forward dtype.
QR_COMPUTE_DTYPE: torch.dtype = torch.float64
# B-3: the CSER perturbation lives in the STANDARDISED design space, so that
# delta = 0.1 is one tenth of a training standard deviation for EVERY
# feature uniformly (Equation (14) perturbs the design-matrix row x_i).
CSER_PERTURBATION_SPACE: str = "standardised"
# B-3(6): Z = X in this study, so a perturbation of x also moves z. The
# manuscript's supporting material asks for the FULL model directional
# response, so the total reading is the default.
CSER_BRANCH_MODE: str = "total"
# B-11: the weight-of-evidence convention. "distribution_ratio" is the
# classical credit-scoring definition ln[(good_c/G)/(bad_c/B)].
WOE_CONVENTION: str = "distribution_ratio"
WOE_SUPPORTED_CONVENTIONS: Tuple[str, ...] = (
    "distribution_ratio", "within_category_rate", "within_category_count",
)
# B-6: the frontier refinement stage (config Section 6).
FRONTIER_REFINEMENT_ENABLED: bool = True
FRONTIER_MAX_ADDITIONAL_MULTIPLIERS: int = 15
FRONTIER_NODE_CHORD_THRESHOLD: float = 0.05
# B-7(2): the Polyak-Ruppert average runs over the FULL second half of the
# iterate path. The candidate stride is a separate concern and must never
# be conflated with the averaging set.
PR_AVERAGING_FRACTION: float = 0.5
# B-16: a null age may never be silently coerced to A = 0.
AGE_IS_STRUCTURED_FEATURE: bool = False
SENSITIVE_NULL_POLICY: str = "raise"
# B-17: the covariate-shift tolerance is expressed in analytic standard
# errors of the group mean difference, not as a fixed absolute number.
COVARIATE_SHIFT_SE_MULTIPLIER: float = 4.0
# B-5: the serialisation round-trip tolerance.
ROUNDTRIP_TOLERANCE: float = 1e-5


# The required config schema, one dotted path per leaf that must exist.
# Every path below is present in the revised study configuration.
REQUIRED_CONFIG_PATHS: Tuple[str, ...] = (
    # Section 1: raw data schemas
    "raw_data_schemas.raw_structured_features_by_dataset",
    "raw_data_schemas.raw_outcome_labels_by_dataset",
    "raw_data_schemas.raw_sensitive_attribute_by_dataset",
    "raw_data_schemas.raw_simulation_input",
    "raw_data_schemas.raw_neural_inputs",
    "raw_data_schemas.reference_risk_vector",
    # Section 2: data preprocessing
    "data_preprocessing.train_val_test_split.train_ratio",
    "data_preprocessing.train_val_test_split.val_ratio",
    "data_preprocessing.train_val_test_split.test_ratio",
    "data_preprocessing.train_val_test_split.stratify_on_y",
    "data_preprocessing.train_val_test_split.seed",
    "data_preprocessing.categorical_encoding.method",
    "data_preprocessing.categorical_encoding.unknown_mapping",
    "data_preprocessing.categorical_encoding.zero_frequency_smoothing",
    "data_preprocessing.continuous_scaling.method",
    "data_preprocessing.continuous_scaling.statistics_source",
    "data_preprocessing.missing_value_handling.employment_duration_imputation",
    "data_preprocessing.missing_value_handling.missing_indicator_suffix",
    "data_preprocessing.sensitive_attribute.feature",
    "data_preprocessing.sensitive_attribute.binarisation_threshold",
    "data_preprocessing.sensitive_attribute.positive_group_condition",
    # Section 3: network architecture
    "network_architecture.structured_component.include_intercept",
    "network_architecture.structured_component.structured_features",
    "network_architecture.neural_encoder.hidden_layer_widths",
    "network_architecture.neural_encoder.activation",
    "network_architecture.neural_encoder.dropout_rate",
    "network_architecture.neural_encoder.num_hidden_layers",
    "network_architecture.neural_encoder.input_dim",
    "network_architecture.neural_encoder.output_dim",
    "network_architecture.residual_layer.is_linear",
    "network_architecture.residual_layer.include_intercept",
    "network_architecture.residual_layer.weights_initialisation",
    "network_architecture.orthogonalisation.method",
    "network_architecture.orthogonalisation.projector",
    "network_architecture.orthogonalisation.requires_full_column_rank",
    "network_architecture.orthogonalisation.add_intercept_to_X",
    # Section 4: network training
    "network_training.optimizer",
    "network_training.learning_rate",
    "network_training.learning_rate_grid",
    "network_training.weight_decay",
    "network_training.weight_decay_grid",
    "network_training.batch_size",
    "network_training.batch_size_grid",
    "network_training.max_epochs",
    "network_training.early_stopping_patience",
    "network_training.num_seeds",
    "network_training.num_restarts_per_config",
    "network_training.seed",
    "network_training.loss.prediction_type",
    "network_training.loss.fairness_multiplier",
    "network_training.loss.lambda_grid_simulation",
    "network_training.loss.lambda_grid_realdata",
    # Section 5: fairness penalty
    "fairness_penalty.active_criterion",
    "fairness_penalty.wasserstein_order",
    "fairness_penalty.p",
    "fairness_penalty.alternative_criteria",
    "fairness_penalty.quantile_grid_bins",
    "fairness_penalty.cdf_implementation",
    "fairness_penalty.differentiable",
    # Section 6: frontier
    "frontier.xi_grid.include_zero",
    "frontier.xi_grid.logspace_start",
    "frontier.xi_grid.logspace_end",
    "frontier.xi_grid.num_points",
    "frontier.solver.step_size_rho",
    "frontier.solver.clipping_epsilon",
    "frontier.solver.max_iterations",
    "frontier.solver.convergence_tolerance",
    "frontier.solver.use_polyak_ruppert_average",
    "frontier.solver.initial_candidates",
    "frontier.refinement.enabled",
    "frontier.refinement.max_additional_multipliers",
    "frontier.refinement.compute_log_midpoints",
    "frontier.refinement.node_chord_length_threshold",
    "frontier.reference_risk.from_simulation",
    "frontier.reference_risk.from_realdata",
    "frontier.reference_risk.unpenalised_model",
    # Section 7: interpretability diagnostics
    "interpretability_diagnostics.decision_threshold",
    "interpretability_diagnostics.csert_perturbation_delta",
    "interpretability_diagnostics.logit_scale",
    "interpretability_diagnostics.use_sample_variance",
    # Section 8: simulation
    "simulation.n_observations",
    "simulation.p_covariates",
    "simulation.sensitive_A.distribution",
    "simulation.sensitive_A.p",
    "simulation.covariate_shift.feature",
    "simulation.covariate_shift.shift_value",
    "simulation.logit_shift.label",
    "simulation.logit_shift.value",
    "simulation.linear_component.intercept_b0",
    "simulation.linear_component.beta",
    "simulation.nonlinear_component.enabled",
    "simulation.nonlinear_component.functions",
    "simulation.nonlinear_component.index_set_I_nl",
    "simulation.interaction_component.enabled",
    "simulation.interaction_component.index_pairs_I_int",
    "simulation.modes",
    "simulation.outcome_generation.link",
    # Section 9: real data experiment
    "real_data_experiment.datasets",
    "real_data_experiment.model_comparisons",
    "real_data_experiment.reference_baseline",
    "real_data_experiment.repetitions",
    "real_data_experiment.metric_evaluation_split",
    "real_data_experiment.frontier_evaluation_split",
    # Section 10: hyperparameter grid
    "hyperparameter_grid.learning_rate",
    "hyperparameter_grid.weight_decay",
    "hyperparameter_grid.batch_size",
    "hyperparameter_grid.hidden_layer_widths",
    "hyperparameter_grid.activation",
    "hyperparameter_grid.dropout_rate",
    # Section 12: plot construction
    "plot_construction.figsize",
    "plot_construction.dpi",
    "plot_construction.colours",
    "plot_construction.lambda_colormap",
    # Section 13: methodological notes
    "methodological_notes.identifiability_guarantee",
    "methodological_notes.gamma0_absorption",
    "methodological_notes.fairness_on_full_score",
    "methodological_notes.frontier_is_benchmark",
    "methodological_notes.diagnostics_are_empirical",
    # Section 14: evaluation metrics
    "evaluation_metrics.reporting_metrics",
    "evaluation_metrics.aggregation",
    # Section 11 (referenced by the orchestrator): sample controls
    "sample_controls.verification_targets",
)

# The dotted paths of all parameters annotated UNSPECIFIED IN MANUSCRIPT in
# the revised study inputs. Each recorded value must be rational, and each
# entry inherits the Task 12 sensitivity-analysis obligation.
PLACEHOLDER_PATHS: Tuple[str, ...] = (
    "data_preprocessing.train_val_test_split.stratify_on_y",
    "data_preprocessing.train_val_test_split.seed",
    "data_preprocessing.categorical_encoding.unknown_mapping",
    "data_preprocessing.categorical_encoding.zero_frequency_smoothing",
    "data_preprocessing.missing_value_handling.employment_duration_imputation",
    "network_architecture.neural_encoder.hidden_layer_widths",
    "network_architecture.neural_encoder.activation",
    "network_architecture.neural_encoder.dropout_rate",
    "network_architecture.residual_layer.weights_initialisation",
    "network_training.learning_rate",
    "network_training.weight_decay",
    "network_training.batch_size",
    "network_training.seed",
    "network_training.loss.lambda_grid_simulation",
    "network_training.loss.lambda_grid_realdata",
    "frontier.solver.step_size_rho",
    "frontier.solver.max_iterations",
    "frontier.solver.convergence_tolerance",
    "frontier.refinement.node_chord_length_threshold",
    "frontier.reference_risk.unpenalised_model",
    "interpretability_diagnostics.csert_perturbation_delta",
    "simulation.nonlinear_component.functions",
    "simulation.nonlinear_component.index_set_I_nl",
    "simulation.interaction_component.index_pairs_I_int",
    "real_data_experiment.reference_baseline",
    "sample_controls.verification_targets",
    # B-12: the eight operative constants that were previously used as
    # thresholds by the pipeline but never registered. Two of them decide
    # whether the study runs at all.
    "network_architecture.orthogonalisation.relative_tolerance",
    "frontier.solver.candidate_stride",
    "frontier.solver.polyak_ruppert_fraction",
    "evaluation_metrics.score_clip_epsilon",
    "evaluation_metrics.nonlinear_regime_gap",
    "simulation.covariate_shift.tolerance_se_multiplier",
    "data_preprocessing.train_val_test_split.minority_class_warning",
    "real_data_experiment.material_relative_change",
    "data_preprocessing.categorical_encoding.woe_convention",
    "interpretability_diagnostics.cser_perturbation_space",
    "interpretability_diagnostics.cser_branch_mode",
    "data_preprocessing.sensitive_attribute.age_is_structured_feature",
)

# A canonical rationale per placeholder, so the audit log records WHY the
# value is defensible rather than merely what it is.
PLACEHOLDER_RATIONALES: Dict[str, str] = {
    "network_architecture.orthogonalisation.relative_tolerance": (
        "The annihilation residual ||X~^T H_perp|| accumulates n round-off "
        "terms and grows like sqrt(n); a RELATIVE tolerance is the only "
        "scale-stable formulation. An absolute 1e-4 blocked findr on five of "
        "the eight datasets (94% of observations) with no loss of "
        "orthogonality whatsoever."
    ),
    "frontier.solver.candidate_stride": (
        "The stride governs which iterates enter the candidate pool ONLY. It "
        "is deliberately decoupled from the Polyak-Ruppert averaging set, "
        "which must cover the full second half of the path."
    ),
    "frontier.solver.polyak_ruppert_fraction": (
        "0.5 implements the prescribed 'average of the second half of the "
        "path'. Averaging a strided subsample discards precisely the "
        "high-frequency oscillation the average exists to cancel."
    ),
    "evaluation_metrics.score_clip_epsilon": (
        "A single epsilon across the metric and frontier paths. 1e-6 is the "
        "value the frontier specification fixes; two different epsilons "
        "feeding the same excess-log-loss difference is a reproducibility "
        "hazard."
    ),
    "evaluation_metrics.nonlinear_regime_gap": (
        "0.03 AUC separates the nonlinear from the approximately linear "
        "regime. The task text prescribes the criterion but not the "
        "threshold, so the value is a placeholder subject to sensitivity."
    ),
    "simulation.covariate_shift.tolerance_se_multiplier": (
        "The group mean difference of X1 has analytic standard error "
        "sqrt(1/n0 + 1/n1) ~ 0.022 at n = 10,000 with a 30% group share. A "
        "fixed 0.05 is only ~2.3 SE and rejects ~2% of legitimate "
        "replicates; 4 SE is the correct acceptance band."
    ),
    "data_preprocessing.train_val_test_split.minority_class_warning": (
        "Below ten minority-class rows in a partition the fairness cells "
        "become unstable; the split warns rather than failing silently."
    ),
    "real_data_experiment.material_relative_change": (
        "A 10% relative move in EVR, DDR or CSER, or any ranking reversal, "
        "is treated as a material change in the robustness comparison."
    ),
    "data_preprocessing.categorical_encoding.woe_convention": (
        "The classical credit-scoring weight of evidence is the log ratio of "
        "DISTRIBUTIONS, ln[(good_c/G)/(bad_c/B)]. The convention itself is a "
        "placeholder because it shifts every encoded feature."
    ),
    "interpretability_diagnostics.cser_perturbation_space": (
        "Equation (14) perturbs the design-matrix row, so delta must act on "
        "the STANDARDISED column. A raw-scale delta is delta/sd_j in model "
        "space, which is ~2e-5 sd for a large-magnitude feature and reduces "
        "the sign comparison to floating-point noise."
    ),
    "data_preprocessing.sensitive_attribute.age_is_structured_feature": (
        "Task 3 Step 3 drops the raw age column when it serves solely as the "
        "sensitive source. Every provider schema checkable against a public "
        "source exceeds the manuscript's feature count by exactly one "
        "(Taiwan 23 vs 22, GMSC 10 vs 9, South German 20 vs 19), which the "
        "solely-sensitive reading reconciles. The manuscript's own South "
        "German column note says age is RETAINED, so the two readings "
        "conflict; False is the default because it matches the published "
        "counts, and the ingestion count gate forces the operator to "
        "reconcile any file that disagrees."
    ),
    "interpretability_diagnostics.cser_branch_mode": (
        "Z = X in this study, so perturbing x also moves z. The total "
        "response is the default because the manuscript asks the "
        "counterfactual to reflect the FULL model's directional response."
    ),
    "data_preprocessing.train_val_test_split.stratify_on_y": (
        "Stratification preserves the rare-event rate in every partition, "
        "which is required because several study datasets have default rates "
        "below 5%."
    ),
    "data_preprocessing.train_val_test_split.seed": (
        "A fixed split seed is required so that every model is evaluated on "
        "the same partitions across seeds and lambdas."
    ),
    "data_preprocessing.categorical_encoding.unknown_mapping": (
        "Mapping unseen categories to the global training WoE is the "
        "standard closed-world fallback for WoE encoding."
    ),
    "data_preprocessing.categorical_encoding.zero_frequency_smoothing": (
        "Additive smoothing (alpha = 0.5) prevents division by zero for "
        "zero-frequency categories and is the textbook Jeffreys-type prior "
        "for WoE."
    ),
    "network_architecture.neural_encoder.hidden_layer_widths": (
        "The grid candidate [16, 8] is a mid-size architecture that is a "
        "candidate in the discrete random search, not a final value."
    ),
    "network_architecture.neural_encoder.activation": (
        "ReLU is the default activation for tabular encoders; the random "
        "search selects among ReLU, GELU and tanh."
    ),
    "network_architecture.neural_encoder.dropout_rate": (
        "Dropout 0.1 is a mild regulariser consistent with the search grid "
        "{0, 0.1, 0.25}."
    ),
    "network_architecture.residual_layer.weights_initialisation": (
        "Xavier uniform initialisation keeps the residual logit at a "
        "comparable scale to the structured logit at initialization."
    ),
    "network_training.learning_rate": (
        "Learning rate 1e-3 is the centre of the grid {3e-4, 1e-3, 3e-3, 1e-2} "
        "and the standard default for AdamW."
    ),
    "network_training.weight_decay": (
        "Decoupled weight decay 0.0 makes the unpenalised benchmark pure; "
        "the grid selects among {0, 1e-5, 1e-4, 1e-3}."
    ),
    "network_training.batch_size": (
        "Batch size 128 is a mid-grid value balancing throughput and "
        "gradient noise."
    ),
    "network_training.seed": (
        "Seed 42 is the global deterministic seed anchoring all replications."
    ),
    "network_training.loss.lambda_grid_simulation": (
        "The grid {0, 0.01, 0.05, 0.1, 0.5, 1, 2, 3, 4} spans the paper's "
        "declared range 0 to 4 at a geometric resolution."
    ),
    "network_training.loss.lambda_grid_realdata": (
        "The grid {0, 0.01, 0.05, 0.1, 0.5, 1, 2} spans the paper's declared "
        "range 0 to 2 at a geometric resolution."
    ),
    "frontier.solver.step_size_rho": (
        "Base step size 0.1 is a conservative smoothing constant for the "
        "sign-based update; its sensitivity is probed in Task 12."
    ),
    "frontier.solver.max_iterations": (
        "One thousand iterations with objective-based tolerance is a "
        "generous budget for the scalarised score problem."
    ),
    "frontier.solver.convergence_tolerance": (
        "Tolerance 1e-6 on the scalarised objective is below the resolution "
        "of the reported excess log loss."
    ),
    "frontier.refinement.node_chord_length_threshold": (
        "The chord-length threshold 0.05 selects where to insert refinement "
        "multipliers in the frontier trace."
    ),
    "frontier.reference_risk.unpenalised_model": (
        "The unpenalised logistic regression at lambda 0 is the named "
        "reference-risk model because it is the transparent baseline."
    ),
    "interpretability_diagnostics.csert_perturbation_delta": (
        "Delta 0.1 is a perturbation size that is small in standardised "
        "units yet large enough to move the full logit detectably."
    ),
    "simulation.nonlinear_component.functions": (
        "The placeholder univariate nonlinearities (sin, square, cos) are "
        "the required rational stand-ins for the manuscript's unspecific NL."
    ),
    "simulation.nonlinear_component.index_set_I_nl": (
        "The index set {1, 2, 6} is the placeholder set of nonlinear "
        "covariates; sensitivity variants are run in Task 12."
    ),
    "simulation.interaction_component.index_pairs_I_int": (
        "The placeholder interaction pairs {(1,2), (3,6)} are swept by the "
        "Task 12 alternative-specification runs."
    ),
    "real_data_experiment.reference_baseline": (
        "CatBoost with default hyperparameters is the state-of-the-art "
        "tabular reference line; defaults are the only specification."
    ),
    "sample_controls.verification_targets": (
        "Verification targets are compared with absolute tolerances (0.05) "
        "against the paper's reported findings; a gross violation halts."
    ),
    "data_preprocessing.missing_value_handling.employment_duration_imputation": (
        "Median training-split imputation plus a missing indicator is the "
        "documented unspecific-imputation resolution."
    ),
}

@dataclass(frozen=True)
class PlaceholderRecord:
    """
    One recorded placeholder parameter with its justification.

    Attributes
    ----------
    path : str
        The dotted path of the parameter.
    value : Any
        The chosen value, converted to a JSON-native type.
    rationale : str
        Why the value is a rational placeholder.
    sensitivity_required : bool
        Always True; the placeholder is subject to Task 12 sensitivity.

    """

    # The dotted path identifying the parameter
    path: str
    # The recorded value, JSON-native
    value: Any
    # The documented justification
    rationale: str
    # Every placeholder inherits the Task 12 sensitivity obligation
    sensitivity_required: bool = True

@dataclass(frozen=True)
class DatasetStatistics:
    """
    The observed statistics of one real-world dataset.

    Attributes
    ----------
    dataset : str
        The dictionary key of the dataset.
    n_observed : int
        The number of unique record_id rows.
    default_rate : float
        The observed default rate within tolerance verification.
    sensitive_share : float
        The observed sensitive-class share.
    n_expected : int
        The metadata sample size.
    expected_default_rate : float
        The metadata default rate.
    expected_sensitive_share : float
        The metadata sensitive-class share.

    """

    # The dataset dictionary key
    dataset: str
    # Observed row count of unique record ids
    n_observed: int
    # Observed default rate
    default_rate: float
    # Observed sensitive-class share
    sensitive_share: float
    # Metadata sample size
    n_expected: int
    # Metadata default rate
    expected_default_rate: float
    # Metadata sensitive-class share
    expected_sensitive_share: float

@dataclass(frozen=True)
class ValidationReport:
    """
    The aggregate output of the full input validation pipeline.

    Attributes
    ----------
    validated_structured : Dict[str, pd.DataFrame]
        Feature frames validated, with record_id as the index.
    validated_labels : Dict[str, pd.DataFrame]
        Label frames validated, with record_id as the index.
    validated_sensitive : Dict[str, pd.DataFrame]
        Sensitive frames validated, with record_id as the index.
    simulation_with_record_id : pd.DataFrame
        The simulation frame with record_id in the first column.
    statistics : List[DatasetStatistics]
        The observed-versus-metadata statistics for each real dataset.
    placeholder_records : List[PlaceholderRecord]
        Every placeholder parameter recorded with its rationale.

    """

    # Validated feature frames keyed by dataset
    validated_structured: Dict[str, pd.DataFrame]
    # Validated label frames keyed by dataset
    validated_labels: Dict[str, pd.DataFrame]
    # Validated sensitive frames keyed by dataset
    validated_sensitive: Dict[str, pd.DataFrame]
    # The simulation frame with its generated record_id
    simulation_with_record_id: pd.DataFrame
    # The observed-versus-metadata statistics
    statistics: List[DatasetStatistics]
    # Every placeholder with its justification
    placeholder_records: List[PlaceholderRecord]
    # True when the simulation record_id was generated (not pre-existing)
    simulation_record_id_generated: bool = True
    # The audit log path the placeholder records were written to
    audit_log_path: str = ""

# -------------------------------------------------------------------------------------------------------------------------------
# Task 1, Step 1: Validate raw data schemas and key integrity
# -------------------------------------------------------------------------------------------------------------------------------
def validate_dataset_key_symmetry(
    raw_structured_features_by_dataset: Dict[str, pd.DataFrame],
    raw_outcome_labels_by_dataset: Dict[str, pd.DataFrame],
    raw_sensitive_attribute_by_dataset: Dict[str, pd.DataFrame],
    expected_keys: Sequence[str] = (),
) -> List[str]:
    """
    Verify the dataset-key symmetry against the expected study population.

    Checks BOTH the three-way internal equality of the dictionaries AND the
    exact match between the observed key union and the expected eight-dataset
    key set, so a dataset absent from every container cannot slip through.

    Parameters
    ----------
    raw_structured_features_by_dataset : Dict[str, pd.DataFrame]
        Feature frames keyed by dataset.
    raw_outcome_labels_by_dataset : Dict[str, pd.DataFrame]
        Label frames keyed by dataset.
    raw_sensitive_attribute_by_dataset : Dict[str, pd.DataFrame]
        Sensitive frames keyed by dataset.
    expected_keys : Sequence[str], optional
        The study's expected dataset key population; every observed key
        must belong to it and every expected key must be observed.

    Returns
    -------
    List[str]
        The sorted, validated dataset keys.

    Raises
    ------
    ValueError
        If any dictionary is not a mapping, shares different keys, an
        observed key is unexpected, or an expected key is missing.

    """
    # Validate container types before any key comparison
    for name, container in (
        ("raw_structured_features_by_dataset", raw_structured_features_by_dataset),
        ("raw_outcome_labels_by_dataset", raw_outcome_labels_by_dataset),
        ("raw_sensitive_attribute_by_dataset", raw_sensitive_attribute_by_dataset),
    ):
        # Each container must be a mapping, never a list or None
        if not isinstance(container, dict):
            raise TypeError(
                f"{name} must be a dict[str, pd.DataFrame]; received "
                f"{type(container).__name__}."
            )
    # Compare the key sets: all three must be identical or no dataset is
    # represented consistently and joins would silently drop rows
    keys = [sorted(container.keys()) for container in (
        raw_structured_features_by_dataset,
        raw_outcome_labels_by_dataset,
        raw_sensitive_attribute_by_dataset,
    )]
    if not (keys[0] == keys[1] == keys[2]):
        # Report the exact symmetric differences so the source can be fixed
        raise ValueError(
            f"dataset key sets differ: structured-only="
            f"{sorted(set(keys[0]) - set(keys[1]) - set(keys[2]))}, "
            f"labels-only="
            f"{sorted(set(keys[1]) - set(keys[0]) - set(keys[2]))}, "
            f"sensitive-only="
            f"{sorted(set(keys[2]) - set(keys[0]) - set(keys[1]))}"
        )
    # Every value must be a DataFrame before any column search occurs
    for key in keys[0]:
        for container in (
            raw_structured_features_by_dataset,
            raw_outcome_labels_by_dataset,
            raw_sensitive_attribute_by_dataset,
        ):
            if not isinstance(container[key], pd.DataFrame):
                raise TypeError(
                    f"dataset '{key}': expected pd.DataFrame, received "
                    f"{type(container[key]).__name__}"
                )
    # The expected population check: the union must equal the expectations,
    # with missing AND unexpected keys reported by dictionary name
    if expected_keys:
        observed: set = set(keys[0])
        expected: set = set(expected_keys)
        missing: List[str] = sorted(expected - observed)
        unexpected: List[str] = sorted(observed - expected)
        if missing or unexpected:
            raise ValueError(
                f"dataset key population does not match the study: "
                f"missing={missing}, unexpected={unexpected}. The study "
                f"requires every expected dataset to be represented."
            )
    return keys[0]

def coerce_and_validate_record_ids(
    frame: pd.DataFrame, dataset: str, column: str = "record_id",
) -> pd.DataFrame:
    """
    Validate and return a copy with record_id coerced to int64.

    The record identifier is the primary join key binding the structured
    features, the outcome labels and the sensitive attribute (Task 1
    Step 1). A value that is not EXACTLY an integer must be rejected, never
    coerced: a silent truncation merges or misaligns rows and destroys the
    row-level correspondence the whole pipeline depends on.

    B-15 remediation: the previous exactness test compared the ALREADY
    truncated int64 column against its own float64 cast, which is a
    tautology that can never fail; a column of [1.5, 2.5, 3.5] was silently
    rewritten as [1, 2, 3]. The comparison is now performed BEFORE the cast,
    on the floating-point representation of the raw values.

    Parameters
    ----------
    frame : pd.DataFrame
        The frame whose record_id column is validated.
    dataset : str
        The dataset key, for error naming.
    column : str, optional
        The identifier column name, by default "record_id".

    Returns
    -------
    pd.DataFrame
        A copy with the record_id column as int64.

    Raises
    ------
    ValueError
        If the column is missing, contains nulls, duplicates, or values
        that do not convert exactly to integers.

    """
    # The identifier column is mandatory: no key, no alignment
    if column not in frame.columns:
        raise ValueError(
            f"dataset '{dataset}': required column '{column}' is missing. "
            f"Present columns: {list(frame.columns)}"
        )
    # Null identifiers cannot be joined and are a hard defect
    if frame[column].isna().any():
        n_null: int = int(frame[column].isna().sum())
        raise ValueError(
            f"dataset '{dataset}': {n_null} null values in '{column}'. "
            f"Repair the raw source before proceeding."
        )
    # Validate uniqueness: a duplicated key makes joins ambiguous
    if not frame[column].is_unique:
        n_dup: int = int(len(frame) - frame[column].nunique())
        raise ValueError(
            f"dataset '{dataset}': {n_dup} duplicate values in '{column}'. "
            f"The key must be globally unique within the dataset."
        )
    if frame[column].dtype == np.int64:
        # Already the exact target dtype: no coercion and no risk
        coerced: pd.Series = frame[column]
    else:
        # Convert to numeric FIRST, under strict error control, so a
        # non-numeric object raises rather than becoming NaN
        numeric: pd.Series = pd.to_numeric(frame[column], errors="raise")
        # Represent as float64 so fractional parts remain visible; this is
        # the representation the exactness test must inspect
        as_float: np.ndarray = numeric.to_numpy(dtype=np.float64)
        # Non-finite values cannot be integers and must be caught here
        if not np.isfinite(as_float).all():
            raise ValueError(
                f"dataset '{dataset}': '{column}' carries non-finite values; "
                f"the join key must be a finite integer."
            )
        # B-15: the exactness test runs BEFORE the cast. A value is exactly
        # integral iff it equals its own rint; np.rint is round-half-even and
        # exact for every float64 whose fractional part is zero
        non_integral: np.ndarray = as_float != np.rint(as_float)
        if bool(non_integral.any()):
            # Name the offending POSITIONS so the defect is actionable
            offending: List[int] = [
                int(p) for p in np.flatnonzero(non_integral)[:10]
            ]
            raise ValueError(
                f"dataset '{dataset}': '{column}' cannot be coerced to int64 "
                f"exactly; {int(non_integral.sum())} non-integer value(s) "
                f"found at positions {offending} "
                f"(first offending value {as_float[non_integral][0]!r}). "
                f"Truncation would silently misalign rows and is refused."
            )
        # Values beyond the int64 range would wrap silently on cast
        if np.any(np.abs(as_float) > float(np.iinfo(np.int64).max)):
            raise ValueError(
                f"dataset '{dataset}': '{column}' exceeds the int64 range."
            )
        # Only now is the cast safe: every value is provably integral
        coerced = pd.Series(
            as_float.astype(np.int64), index=frame.index, name=column
        )
    # Return a fresh copy with the coerced key column (no caller mutation)
    result: pd.DataFrame = frame.copy()
    result[column] = coerced
    return result

def check_key_set_equality(
    features: pd.DataFrame,
    labels: pd.DataFrame,
    sensitive: pd.DataFrame,
    dataset: str,
    column: str = "record_id",
) -> None:
    """
    Assert the three record_id populations are identical for a dataset.

    Parameters
    ----------
    features : pd.DataFrame
        The structured feature frame.
    labels : pd.DataFrame
        The outcome label frame.
    sensitive : pd.DataFrame
        The sensitive-attribute frame.
    dataset : str
        The dataset key, for error naming.
    column : str, optional
        The identifier column name, by default "record_id".

    Raises
    ------
    ValueError
        If any record_id set differs, or rows that are orphaned exist in
        any frame.

    """
    # Compute the three ID populations as sorted arrays
    feature_ids: np.ndarray = np.sort(features[column].to_numpy())
    label_ids: np.ndarray = np.sort(labels[column].to_numpy())
    sensitive_ids: np.ndarray = np.sort(sensitive[column].to_numpy())
    # Set equality via pairwise sorted comparisons (linear, exact)
    if not (np.array_equal(feature_ids, label_ids) and
            np.array_equal(feature_ids, sensitive_ids)):
        # Report the concrete dangling sets, never an abstract complaint
        f_set, l_set, s_set = set(feature_ids), set(label_ids), set(sensitive_ids)
        raise ValueError(
            f"dataset '{dataset}': record_id populations differ; "
            f"in-features-not-labels={sorted(f_set - l_set)}, "
            f"in-labels-not-features={sorted(l_set - f_set)}, "
            f"in-sensitive-not-features={sorted(s_set - f_set)}. "
            f"No orphan rows or missing labels are permitted."
        )
    # A correct dataset must agree on the population size at this point
    if feature_ids.size == 0:
        raise ValueError(
            f"dataset '{dataset}': the record_id population is empty."
        )

def set_record_id_index(
    frame: pd.DataFrame, dataset: str, column: str = "record_id",
) -> pd.DataFrame:
    """
    Return a copy with record_id set as the frame index.

    Parameters
    ----------
    frame : pd.DataFrame
        The validated frame.
    dataset : str
        The dataset key, for error naming.
    column : str, optional
        The identifier column name, by default "record_id".

    Returns
    -------
    pd.DataFrame
        The frame indexed by record_id.

    """
    # The key must be validated before it becomes the index
    if column not in frame.columns:
        raise ValueError(f"dataset '{dataset}': column '{column}' missing.")
    # Reset to index and verify one-to-one mapping
    indexed = frame.set_index(column, drop=True)
    if not indexed.index.is_unique:
        raise ValueError(f"dataset '{dataset}': index is not unique after set_index.")
    return indexed

def prepare_simulation_record_ids(
    simulation_input: pd.DataFrame, record_column: str = "record_id",
) -> Tuple[pd.DataFrame, bool]:
    """
    Generate or validate a unique record_id per (mode, seed) block.

    Parameters
    ----------
    simulation_input : pd.DataFrame
        The raw simulation input.
    record_column : str, optional
        The identifier column name, by default "record_id".

    Returns
    -------
    Tuple[pd.DataFrame, bool]
        The frame with record_id as the first column, and a boolean
        indicating whether the id was generated (True) or pre-existing
        and validated (False).

    Raises
    ------
    ValueError
        If the required mode/seed columns are missing or the frame is empty.

    """
    # The block key columns must be present before grouping
    for required in ("mode", "seed"):
        if required not in simulation_input.columns:
            raise ValueError(
                f"simulation input must carry '{required}'; present: "
                f"{list(simulation_input.columns)}"
            )
    if simulation_input.empty:
        raise ValueError("simulation input is empty; no records to key.")
    # Work on a copy so no caller state is mutated
    result = simulation_input.copy()
    generated: bool = False
    if record_column in result.columns:
        # A pre-existing key must satisfy the same contract we would
        # generate, so repeated validation remains idempotent
        if result[record_column].isna().any():
            raise ValueError(
                "simulation record_id must be non-null before validation."
            )
        # Recast to int64 so downstream joins are dtype-stable
        result[record_column] = pd.to_numeric(
            result[record_column], errors="raise"
        ).astype("int64")
    else:
        # Incrementing ids within each (mode, seed) block, resetting at
        # every block: the composite key (mode, seed, record_id) is
        # therefore globally unique by construction
        ids: pd.Series = (
            result.groupby(["mode", "seed"], sort=False).cumcount() + 1
        )
        # Insert as the first column so the key precedes the covariates
        result.insert(0, record_column, ids)
        generated = True
    # The key contract is uniqueness WITHIN each (mode, seed) block; the
    # composite key then carries global uniqueness
    block_counts: pd.Series = result.groupby(["mode", "seed"]).size()
    block_uniques: pd.Series = result.groupby(["mode", "seed"])[
        record_column].nunique()
    if not bool((block_counts == block_uniques).all()):
        # A per-block duplicate would make row alignment ambiguous
        raise ValueError(
            "simulation record_id must be unique within each (mode, seed) "
            "block; a duplicate was found."
        )
    return result, generated

def validate_simulation_domains(simulation: pd.DataFrame, run_seeds: Iterable[int]) -> None:
    """
    Validate the domain of every simulation control column.

    Parameters
    ----------
    simulation : pd.DataFrame
        The simulation frame with record_id.
    run_seeds : Iterable[int]
        The accepted replication seeds (0..9 in the final implementation).

    Raises
    ------
    ValueError
        If mode, seed, A, Y or r_true violate their declared domains.

    """
    # Mode must be one of the two generative processes
    valid_modes: Tuple[str, ...] = ("simple", "complex")
    bad_modes: List[object] = sorted(set(simulation["mode"].tolist()) - set(valid_modes))
    if bad_modes:
        raise ValueError(f"simulation mode must be in {valid_modes}; found {bad_modes}.")
    # Seed must be one of the ten accepted replication indices
    accepted: set = set(run_seeds)
    bad_seeds: List[int] = sorted(
        {int(s) for s in simulation["seed"].tolist()} - accepted
    )
    if bad_seeds:
        raise ValueError(f"simulation seed must be in {sorted(accepted)}; found {bad_seeds}.")
    # A and Y are binary outcomes of the DGP
    for column in ("A", "Y"):
        bad_binary: List[object] = sorted(
            {v for v in simulation[column].tolist() if v not in (0, 1)}
        )
        if bad_binary:
            raise ValueError(f"simulation {column} must be binary; found {bad_binary}.")
    # The true reference risk is a probability strictly inside (0, 1)
    with np.errstate(invalid="raise"):
        risk_values: pd.Series = pd.to_numeric(simulation["r_true"], errors="raise")
    if not np.all((risk_values > 0.0) & (risk_values < 1.0)):
        bad_risk: int = int(((risk_values <= 0.0) | (risk_values >= 1.0)).sum())
        raise ValueError(
            f"simulation r_true must lie strictly in (0, 1); found "
            f"{bad_risk} violating row(s)."
        )

def validate_dtypes_against_schema(
    frame: pd.DataFrame, continuous: List[str], categorical: List[str],
    dataset: str,
) -> None:
    """
    Enforce the declared dtype domain for continuous and categorical columns.

    Parameters
    ----------
    frame : pd.DataFrame
        The frame whose columns are checked.
    continuous : List[str]
        The declared continuous feature names.
    categorical : List[str]
        The declared categorical feature names.
    dataset : str
        The dataset key, for error naming.

    Raises
    ------
    ValueError
        If any continuous column is not numeric, or any categorical column
        is not object- or category-typed, or the two lists overlap.

    """
    # Overlapping declarations cannot be resolved
    overlap: set = set(continuous) & set(categorical)
    if overlap:
        raise ValueError(
            f"dataset '{dataset}': columns declared both continuous and "
            f"categorical: {sorted(overlap)}"
        )
    # Every continuous column must end as float64 or int64 exactly
    for column in continuous:
        if column not in frame.columns:
            raise ValueError(f"dataset '{dataset}': continuous column '{column}' missing.")
        # An object column is only admissible when every value coerces
        # exactly to a numeric dtype; silent truncation is forbidden
        if not pd.api.types.is_numeric_dtype(frame[column]):
            try:
                # Coerce the numeric-looking object column under inspection
                coerced_series = pd.to_numeric(frame[column], errors="raise")
            except (ValueError, TypeError) as exc:
                raise ValueError(
                    f"dataset '{dataset}': continuous column '{column}' "
                    f"cannot be coerced to numeric; found non-numeric "
                    f"values."
                ) from exc
            # The validated copy is updated so the returned frame carries
            # the exact numeric values that passed inspection
            frame[column] = coerced_series
        # After coercion the dtype must be exactly float64 or int64
        if not (frame[column].dtype == np.float64 or frame[column].dtype == np.int64):
            raise ValueError(
                f"dataset '{dataset}': continuous column '{column}' has "
                f"dtype {frame[column].dtype} after inspection; the "
                f"revised schema requires float64 or int64."
            )
    # Every categorical column must be object or category typed
    for column in categorical:
        if column not in frame.columns:
            raise ValueError(f"dataset '{dataset}': categorical column '{column}' missing.")
        if not (pd.api.types.is_object_dtype(frame[column])
                or pd.api.types.is_categorical_dtype(frame[column])):
            raise ValueError(
                f"dataset '{dataset}': categorical column '{column}' has "
                f"dtype {frame[column].dtype}; expected object or category."
            )

def validate_raw_data_schemas(
    raw_structured_features_by_dataset: Dict[str, pd.DataFrame],
    raw_outcome_labels_by_dataset: Dict[str, pd.DataFrame],
    raw_sensitive_attribute_by_dataset: Dict[str, pd.DataFrame],
    raw_simulation_input: pd.DataFrame,
    run_seeds: Iterable[int],
    schema_continuous: Dict[str, List[str]],
    schema_categorical: Dict[str, List[str]],
    expected_dataset_keys: Sequence[str] = (),
) -> Tuple[Dict[str, pd.DataFrame], Dict[str, pd.DataFrame],
           Dict[str, pd.DataFrame], pd.DataFrame, bool]:
    """
    Execute the full schema and key-integrity validation of Step 1.

    Parameters
    ----------
    raw_structured_features_by_dataset : Dict[str, pd.DataFrame]
        Feature frames keyed by dataset.
    raw_outcome_labels_by_dataset : Dict[str, pd.DataFrame]
        Label frames keyed by dataset.
    raw_sensitive_attribute_by_dataset : Dict[str, pd.DataFrame]
        Sensitive frames keyed by dataset.
    raw_simulation_input : pd.DataFrame
        The raw simulation input.
    run_seeds : Iterable[int]
        The accepted replication seeds.
    schema_continuous : Dict[str, List[str]]
        Continuous feature names per dataset, from the declared schema.
    schema_categorical : Dict[str, List[str]]
        Categorical feature names per dataset, from the declared schema.

    Returns
    -------
    Tuple[Dict[str, pd.DataFrame], Dict[str, pd.DataFrame],
          Dict[str, pd.DataFrame], pd.DataFrame, bool]
        The validated feature, label and sensitive frames (record_id
        indexed), the simulation frame with record_id first, and the
        generation flag.

    """
    # Confirm all eight datasets are represented consistently first; the
    # expected population is asserted so an absent dataset is detected
    keys: List[str] = validate_dataset_key_symmetry(
        raw_structured_features_by_dataset,
        raw_outcome_labels_by_dataset,
        raw_sensitive_attribute_by_dataset,
        expected_dataset_keys,
    )
    validated_features: Dict[str, pd.DataFrame] = {}
    validated_labels: Dict[str, pd.DataFrame] = {}
    validated_sensitive: Dict[str, pd.DataFrame] = {}
    for key in keys:
        # Coerce the identifier to int64 on every container
        features = coerce_and_validate_record_ids(
            raw_structured_features_by_dataset[key], key
        )
        labels = coerce_and_validate_record_ids(
            raw_outcome_labels_by_dataset[key], key
        )
        sensitive = coerce_and_validate_record_ids(
            raw_sensitive_attribute_by_dataset[key], key
        )
        # The union of record_id sets must be identical, no orphans
        check_key_set_equality(features, labels, sensitive, key)
        # The label frame must carry the binary default indicator
        if "y_default" not in labels.columns:
            raise ValueError(f"dataset '{key}': 'y_default' column missing.")
        # The revised schema requires the label and the sensitive source
        # to be int64; exact coercion is inspected, never silent
        labels["y_default"] = pd.to_numeric(
            labels["y_default"], errors="raise"
        ).astype("int64")
        if "raw_age" in sensitive.columns:
            sensitive["raw_age"] = pd.to_numeric(
                sensitive["raw_age"], errors="raise"
            ).astype("int64")
        # The label indicator must be binary before any statistic is taken
        if not set(labels["y_default"].tolist()) <= {0, 1}:
            raise ValueError(
                f"dataset '{key}': y_default must be binary; found "
                f"{sorted(set(labels['y_default'].tolist()))}."
            )
        # The sensitive frame must carry the raw age source
        if "raw_age" not in sensitive.columns and "age" not in features.columns:
            raise ValueError(
                f"dataset '{key}': neither raw_age (sensitive frame) nor "
                f"age (feature frame) is present; the sensitive attribute "
                f"cannot be derived."
            )
        # Enforce the declared dtype domain per dataset schema
        validate_dtypes_against_schema(
            features, schema_continuous.get(key, []),
            schema_categorical.get(key, []), key,
        )
        # Index every container by record_id for downstream alignment
        validated_features[key] = set_record_id_index(features, key)
        validated_labels[key] = set_record_id_index(labels, key)
        validated_sensitive[key] = set_record_id_index(sensitive, key)
    # Prepare (or validate) the simulation key and its domains
    simulation, generated = prepare_simulation_record_ids(raw_simulation_input)
    validate_simulation_domains(simulation, run_seeds)
    return (
        validated_features, validated_labels, validated_sensitive,
        simulation, generated,
    )

# -------------------------------------------------------------------------------------------------------------------------------
# Task 1, Step 2: Validate dataset-level statistics against config metadata
# -------------------------------------------------------------------------------------------------------------------------------
def derive_sensitive_attribute(
    features: pd.DataFrame,
    sensitive_frame: pd.DataFrame,
    dataset: str,
    null_policy: str = SENSITIVE_NULL_POLICY,
) -> pd.Series:
    """
    Derive the binary sensitive attribute for one dataset.

    Equation (Section 5.1): A_i = 1{age_i <= 25}, else 0.

    B-16 remediation: a null age previously evaluated to False under the
    vectorised comparison and was therefore silently recorded as A = 0,
    assigning protected-group status by default. In a regulated pipeline a
    silent default on the protected attribute is not acceptable. Nulls are
    now counted explicitly and, under the default policy, raise.

    Parameters
    ----------
    features : pd.DataFrame
        The record_id-indexed feature frame.
    sensitive_frame : pd.DataFrame
        The record_id-indexed sensitive frame.
    dataset : str
        The dataset key, for error naming.
    null_policy : str, optional
        "raise" (default) refuses to proceed with a null age; "propagate"
        returns pandas NA for those rows so the caller must handle them
        explicitly. There is deliberately no policy that silently defaults
        a missing age to either group.

    Returns
    -------
    pd.Series
        The derived A series, indexed by record_id, values in {0, 1}
        (or nullable Int64 carrying NA under the propagate policy).

    Raises
    ------
    ValueError
        If no symmetric age source exists, the ages are non-finite, or a
        null age is present under the raise policy.

    """
    # The policy domain is fixed; an unknown policy is a configuration defect
    if null_policy not in ("raise", "propagate"):
        raise ValueError(
            f"null_policy must be 'raise' or 'propagate'; received "
            f"'{null_policy}'."
        )
    # The separate raw_age source takes precedence when present
    if "raw_age" in sensitive_frame.columns:
        age_series: pd.Series = sensitive_frame["raw_age"]
    else:
        # Otherwise derive from the retained feature column
        if "age" not in features.columns:
            raise ValueError(
                f"dataset '{dataset}': no age source for the sensitive "
                f"attribute; expected raw_age or age."
            )
        age_series = features["age"]
    # Numeric conversion under strict error control: a non-numeric age is a
    # source defect, never a silent NaN
    numeric_age: pd.Series = pd.to_numeric(age_series, errors="raise")
    # B-16: count the nulls EXPLICITLY before any comparison is performed
    null_mask: np.ndarray = numeric_age.isna().to_numpy()
    n_null: int = int(null_mask.sum())
    if n_null > 0:
        if null_policy == "raise":
            # Name the offending record ids so the defect is actionable
            offending: List[Any] = list(numeric_age.index[null_mask][:10])
            raise ValueError(
                f"dataset '{dataset}': {n_null} null age value(s) cannot be "
                f"binarised; the sensitive attribute A = 1{{age <= "
                f"{SENSITIVE_BINARISATION_THRESHOLD}}} is undefined for them. "
                f"First offending record_id(s): {offending}. A missing "
                f"protected attribute must be resolved at source, never "
                f"defaulted to a group."
            )
        # Under the propagate policy the caller is told how many rows are NA
        logger.warning(
            f"dataset '{dataset}': {n_null} null age value(s) propagate as "
            f"NA in the sensitive attribute; downstream consumers must "
            f"handle them explicitly."
        )
    # The finite check applies to the observed (non-null) ages only
    observed: np.ndarray = numeric_age.to_numpy(dtype=np.float64)
    if not np.isfinite(observed[~null_mask]).all():
        raise ValueError(
            f"dataset '{dataset}': age contains non-finite values."
        )
    # Equation: A_i = 1 if age_i <= 25, else 0
    binarised: np.ndarray = (
        observed <= float(SENSITIVE_BINARISATION_THRESHOLD)
    ).astype(np.int64)
    if n_null > 0 and null_policy == "propagate":
        # The nullable integer dtype carries NA without collapsing it to 0
        result: pd.Series = pd.Series(
            binarised, index=features.index, name="A", dtype="float64"
        )
        result[null_mask] = np.nan
        return result.astype("Int64")
    # The dense path: every row has an observed age
    return pd.Series(binarised, index=features.index, name="A")

# In-notebook resolution (3A.8): the Task 3 cell rebinds
# derive_sensitive_attribute to the raw-scale function later;
# this cell snapshots the Task 1 binding for its own callers
_derive_sensitive_attribute_task1 = derive_sensitive_attribute

def compute_dataset_statistics(
    features: pd.DataFrame,
    labels: pd.DataFrame,
    sensitive_frame: pd.DataFrame,
    dataset: str,
) -> DatasetStatistics:
    """
    Compute the observed sample size, default rate and sensitive share.

    Parameters
    ----------
    features : pd.DataFrame
        The record_id-indexed feature frame.
    labels : pd.DataFrame
        The record_id-indexed label frame.
    sensitive_frame : pd.DataFrame
        The record_id-indexed sensitive frame.
    dataset : str
        The dataset key.

    Returns
    -------
    DatasetStatistics
        The observed statistics, with metadata values filled later.

    """
    # Align all three frames on the shared record_id index
    joined = pd.concat(
        [features, labels[["y_default"]], sensitive_frame], axis=1, join="inner"
    )
    # An inner join on the index must preserve every observation
    if len(joined) != len(features):
        raise ValueError(
            f"dataset '{dataset}': join lost rows; expected "
            f"{len(features)}, got {len(joined)}."
        )
    # Equation: n = number of unique record_id rows (the index itself)
    n_observed: int = int(len(joined.index.unique()))
    # Equation: pi_default = (1/n) * sum_i 1{y_i = 1}
    default_rate: float = float(joined["y_default"].mean())
    # Equation: A_i = 1{raw_age_i <= 25}; pi_sensitive = mean(A)
    a_values: pd.Series = _derive_sensitive_attribute_task1(features, sensitive_frame, dataset)
    sensitive_share: float = float(a_values.mean())
    return DatasetStatistics(
        dataset=dataset,
        n_observed=n_observed,
        default_rate=default_rate,
        sensitive_share=sensitive_share,
        n_expected=-1,
        expected_default_rate=np.nan,
        expected_sensitive_share=np.nan,
    )

def validate_statistics_against_metadata(
    statistics: List[DatasetStatistics],
    datasets_metadata: Dict[str, Dict[str, object]],
) -> List[DatasetStatistics]:
    """
    Compare observed statistics with metadata within the tolerance band.

    Parameters
    ----------
    statistics : List[DatasetStatistics]
        The observed statistics per dataset.
    datasets_metadata : Dict[str, Dict[str, object]]
        The metadata from config["real_data_experiment"]["datasets"].

    Returns
    -------
    List[DatasetStatistics]
        The statistics with the metadata fields filled.

    Raises
    ------
    ValueError
        If metadata is missing for a dataset or any statistic deviates by
        more than the absolute tolerance.

    """
    filled: List[DatasetStatistics] = []
    for stat in statistics:
        # A metadata entry must exist for every observed dataset
        if stat.dataset not in datasets_metadata:
            raise ValueError(
                f"dataset '{stat.dataset}' has no metadata entry in "
                f"config['real_data_experiment']['datasets']."
            )
        meta: Dict[str, object] = datasets_metadata[stat.dataset]
        n_expected: int = int(meta["n"])
        expected_default: float = float(meta["default_rate"])
        expected_share: float = float(meta["sensitive_share"])
        # The groupwise checks use absolute tolerance (atol), not relatives
        default_ok: bool = bool(np.isclose(
            stat.default_rate, expected_default, atol=STATISTICS_TOLERANCE, rtol=0.0
        ))
        share_ok: bool = bool(np.isclose(
            stat.sensitive_share, expected_share, atol=STATISTICS_TOLERANCE, rtol=0.0
        ))
        # A deviation beyond the band must stop the pipeline with both
        # the expected and the observed values reported
        if stat.n_observed != n_expected or not default_ok or not share_ok:
            raise ValueError(
                f"dataset '{stat.dataset}': observed statistics deviate "
                f"from metadata. n: observed {stat.n_observed} vs expected "
                f"{n_expected}; default rate: observed "
                f"{stat.default_rate:.6f} vs expected {expected_default:.6f}; "
                f"sensitive share: observed {stat.sensitive_share:.6f} vs "
                f"expected {expected_share:.6f}. Tolerance "
                f"{STATISTICS_TOLERANCE} exceeded; reconcile the raw source "
                f"before any downstream processing."
            )
        # Record the metadata alongside the observed values
        filled.append(DatasetStatistics(
            dataset=stat.dataset, n_observed=stat.n_observed,
            default_rate=stat.default_rate, sensitive_share=stat.sensitive_share,
            n_expected=n_expected, expected_default_rate=expected_default,
            expected_sensitive_share=expected_share,
        ))
    return filled

# -------------------------------------------------------------------------------------------------------------------------------
# Task 1, Step 3: Validate config completeness and flag unspecified parameters
# -------------------------------------------------------------------------------------------------------------------------------
def collect_required_schema_paths() -> List[str]:
    """
    Return the required config paths as a frozen schema copy.

    Returns
    -------
    List[str]
        Every required dotted path.

    """
    # The schema is a tuple to prevent accidental mutation
    return list(REQUIRED_CONFIG_PATHS)

def flatten_config_paths(config: Dict[str, Any]) -> Dict[str, Any]:
    """
    Flatten a nested config dict into a dotted-path to value mapping.

    Parameters
    ----------
    config : Dict[str, Any]
        The nested configuration.

    Returns
    -------
    Dict[str, Any]
        The mapping from dotted path to leaf value, JSON-native.

    Raises
    ------
    TypeError
        If config is not a mapping.

    """
    # The root must be a mapping before descent
    if not isinstance(config, dict):
        raise TypeError(
            f"config must be a dict; received {type(config).__name__}."
        )
    leaves: Dict[str, Any] = {}
    # Iterative descent avoids recursion depth limits on deep configs
    stack: deque[Tuple[Any, str]] = deque()
    stack.append((config, ""))
    while stack:
        # Pop the next node and its accumulated path
        node, path = stack.pop()
        for key, value in node.items():
            child_path: str = f"{path}.{key}" if path else str(key)
            if isinstance(value, dict):
                # Descend into nested sections
                stack.append((value, child_path))
            else:
                # Leaves are normalised to JSON-native scalars when needed
                if isinstance(value, (np.integer,)):
                    leaves[child_path] = int(value)
                elif isinstance(value, (np.floating,)):
                    leaves[child_path] = float(value)
                elif isinstance(value, (np.bool_,)):
                    leaves[child_path] = bool(value)
                else:
                    leaves[child_path] = value
    return leaves

def walk_config_path(config: Dict[str, Any], path: str) -> bool:
    """
    Return True when a dotted path navigates to an existing entry.

    Parameters
    ----------
    config : Dict[str, Any]
        The nested configuration.
    path : str
        The dotted path to navigate.

    Returns
    -------
    bool
        True when every element of the path resolves; dict-valued nodes
        count as present because they hold the entry, not merely a leaf.

    """
    # Navigate the node chain element by element
    node: Any = config
    for element in path.split("."):
        # A non-mapping node cannot contain the next element
        if not isinstance(node, dict):
            return False
        # The next element must exist in the current dict
        if element not in node:
            return False
        # Descend into the child node
        node = node[element]
    # The full path resolved, whatever the terminal node's shape
    return True

def get_config_value(config: Dict[str, Any], path: str) -> Any:
    """
    Fetch the node at a dotted path, dict-valued nodes included.

    Parameters
    ----------
    config : Dict[str, Any]
        The nested configuration.
    path : str
        The dotted path to navigate.

    Returns
    -------
    Any
        The node at the path.

    Raises
    ------
    KeyError
        If the path does not resolve.

    """
    # Navigate to the requested node
    node: Any = config
    for element in path.split("."):
        if not isinstance(node, dict) or element not in node:
            raise KeyError(f"config path '{path}' does not resolve.")
        node = node[element]
    return node

def find_missing_config_paths(config: Dict[str, Any]) -> List[str]:
    """
    List every required config path absent from the configuration.

    Parameters
    ----------
    config : Dict[str, Any]
        The nested configuration.

    Returns
    -------
    List[str]
        The missing dotted paths, in schema order.

    """
    # A required path is present when navigation reaches its node; the
    # node may be a value or a nested section
    return [path for path in REQUIRED_CONFIG_PATHS
            if not walk_config_path(config, path)]

# B-12: the operative module constants that function as thresholds. Every
# entry must appear in PLACEHOLDER_PATHS; the coverage assertion below makes
# a newly introduced constant impossible to slip past the audit.
OPERATIVE_CONSTANT_REGISTRY: Dict[str, Tuple[str, Any]] = {
    "network_architecture.orthogonalisation.relative_tolerance":
        ("ORTHOGONALITY_RELATIVE_TOLERANCE", ORTHOGONALITY_RELATIVE_TOLERANCE),
    "frontier.solver.candidate_stride":
        ("CANDIDATE_STRIDE", None),
    "frontier.solver.polyak_ruppert_fraction":
        ("PR_AVERAGING_FRACTION", PR_AVERAGING_FRACTION),
    "evaluation_metrics.score_clip_epsilon":
        ("SCORE_CLIP_EPSILON", None),
    "evaluation_metrics.nonlinear_regime_gap":
        ("NONLINEAR_REGIME_GAP", None),
    "simulation.covariate_shift.tolerance_se_multiplier":
        ("COVARIATE_SHIFT_SE_MULTIPLIER", COVARIATE_SHIFT_SE_MULTIPLIER),
    "data_preprocessing.train_val_test_split.minority_class_warning":
        ("MINORITY_CLASS_WARNING_THRESHOLD", None),
    "real_data_experiment.material_relative_change":
        ("MATERIAL_RELATIVE_CHANGE", None),
    "data_preprocessing.categorical_encoding.woe_convention":
        ("WOE_CONVENTION", WOE_CONVENTION),
    "interpretability_diagnostics.cser_perturbation_space":
        ("CSER_PERTURBATION_SPACE", CSER_PERTURBATION_SPACE),
    "interpretability_diagnostics.cser_branch_mode":
        ("CSER_BRANCH_MODE", CSER_BRANCH_MODE),
    "data_preprocessing.sensitive_attribute.age_is_structured_feature":
        ("AGE_IS_STRUCTURED_FEATURE", False),
}


def resolve_operative_constant(symbol: str, fallback: Any) -> Any:
    """
    Resolve an operative constant's live value from the module namespace.

    B-12. The registry records the SYMBOL rather than a copy of the value,
    so the audit always reports what the pipeline will actually use, even if
    a constant is redefined later in the module.

    Parameters
    ----------
    symbol : str
        The module-level constant name.
    fallback : Any
        The value used when the symbol is not bound at call time.

    Returns
    -------
    Any
        The live value, or the fallback.

    """
    # globals() is the module namespace the pipeline itself reads from
    return globals().get(symbol, fallback)


def record_placeholder_parameters(
    config: Dict[str, Any], paths: Tuple[str, ...] = PLACEHOLDER_PATHS,
) -> List[PlaceholderRecord]:
    """
    Record every placeholder parameter with its value and rationale.

    Task 1 Step 3 requires that, for EVERY parameter not fixed by the
    manuscript, the chosen value is recorded in a separate audit log with a
    rationale and a standing Task 12 sensitivity obligation. The
    instruction is explicit that the pipeline must not silently proceed
    with placeholders.

    The scanner has four channels:
      1. hard-coded placeholder paths PRESENT in config -> recorded;
      2. hard-coded placeholder paths ABSENT from config -> recorded as a
         "not found" entry, never silently skipped;
      3. B-12 operative module constants -> recorded from the live module
         namespace even when the config carries no corresponding node;
      4. config leaves whose key carries an unspecified/placeholder
         annotation token but are not in the hard-coded list -> recorded as
         "unlisted candidate".

    B-12 remediation. Channel 3 is new. Eight operative constants were used
    as thresholds by the pipeline and never registered, including the two
    that caused the most severe defects in the codebase: the orthogonality
    tolerance, which decided whether findr could be constructed at all, and
    the frontier candidate stride, which decided whether the
    Polyak-Ruppert average meant anything. A coverage assertion now makes
    an unregistered operative constant impossible.

    Parameters
    ----------
    config : Dict[str, Any]
        The nested configuration.
    paths : Tuple[str, ...], optional
        The hard-coded placeholder path set, defaults to the annotated set.

    Returns
    -------
    List[PlaceholderRecord]
        One record per placeholder, including not-found entries, live
        operative constants and unlisted candidates.

    Raises
    ------
    ValueError
        If an operative constant is absent from PLACEHOLDER_PATHS, which
        would let it escape the audit.

    """
    # B-12: the coverage assertion. Every operative constant must be a
    # registered placeholder path, checked BEFORE any record is emitted
    unregistered: List[str] = [
        path for path in OPERATIVE_CONSTANT_REGISTRY if path not in paths
    ]
    if unregistered:
        raise ValueError(
            f"operative constant(s) {sorted(unregistered)} are used as "
            f"pipeline thresholds but absent from PLACEHOLDER_PATHS; a "
            f"threshold that escapes the audit violates Task 1 Step 3."
        )
    records: List[PlaceholderRecord] = []
    # The flattened leaf map drives channels 1, 2 and 4
    leaves: Dict[str, Any] = flatten_config_paths(config)
    for path in paths:
        # Channel 3 takes precedence: an operative constant reports the
        # LIVE module value, which is what the pipeline actually uses
        if path in OPERATIVE_CONSTANT_REGISTRY:
            symbol, fallback = OPERATIVE_CONSTANT_REGISTRY[path]
            live_value: Any = resolve_operative_constant(symbol, fallback)
            # A config node, when present, overrides the module default and
            # is therefore the value that must be audited
            if walk_config_path(config, path):
                live_value = get_config_value(config, path)
            records.append(PlaceholderRecord(
                path=path,
                value=json_native(live_value) if not isinstance(
                    live_value, (str, bool)) else live_value,
                rationale=PLACEHOLDER_RATIONALES.get(
                    path,
                    f"Operative module constant {symbol}; subject to Task 12 "
                    f"sensitivity analysis."
                ),
                sensitivity_required=True,
            ))
            continue
        # Channel 2: a hard-coded placeholder absent from config must not
        # vanish silently
        if not walk_config_path(config, path):
            records.append(PlaceholderRecord(
                path=path, value="PATH NOT FOUND IN CONFIG",
                rationale="Hard-coded placeholder path absent from the "
                          "supplied config; recorded for the Task 12 "
                          "sensitivity review, never silently dropped.",
                sensitivity_required=True,
            ))
            continue
        # Channel 1: the value may be a leaf or a dict-valued node; dict
        # nodes are normalised to JSON-native scalars
        value: Any = get_config_value(config, path)
        if isinstance(value, dict):
            value = {k: (int(v) if isinstance(v, np.integer) else
                         float(v) if isinstance(v, np.floating) else v)
                     for k, v in value.items()}
        records.append(PlaceholderRecord(
            path=path,
            value=value,
            rationale=PLACEHOLDER_RATIONALES.get(
                path,
                "Placeholder value subject to Task 12 sensitivity analysis."
            ),
            sensitivity_required=True,
        ))
    # Channel 4: any leaf whose key carries an annotation token but is not
    # in the hard-coded list
    unlisted: List[PlaceholderRecord] = []
    for path, value in leaves.items():
        # The annotation-token scan matches the documented sentinel words
        if ("unspecified" in path.lower().replace("_", "") or
                "placeholder" in path.lower().replace("_", "")):
            if path not in paths:
                unlisted.append(PlaceholderRecord(
                    path=path,
                    value=value,
                    rationale="Unlisted candidate: config carries an "
                              "unspecified/placeholder-annotated key "
                              "outside the hard-coded list; recorded "
                              "with the unlisted-candidate flag.",
                    sensitivity_required=True,
                ))
    # The unlisted entries are flagged in their rationale, carried verbatim
    records.extend(unlisted)
    return records

def validate_numeric_parameter_ranges(config: Dict[str, Any]) -> None:
    """
    Validate the numerical ranges and ordering rules of the config.

    Parameters
    ----------
    config : Dict[str, Any]
        The nested configuration.

    Raises
    ------
    ValueError
        For any violated range, ordering or ratio rule.

    """
    leaves: Dict[str, Any] = flatten_config_paths(config)

    # The three split ratios must sum to 1 within float rounding
    ratios = (
        leaves.get("data_preprocessing.train_val_test_split.train_ratio"),
        leaves.get("data_preprocessing.train_val_test_split.val_ratio"),
        leaves.get("data_preprocessing.train_val_test_split.test_ratio"),
    )
    if None in ratios:
        return
    if not abs(sum(float(r) for r in ratios) - 1.0) < 1e-6:
        raise ValueError(
            f"train/val/test ratios must sum to 1.0; found {sum(ratios)}."
        )

    # Learning rates must be strictly positive
    for path in ("network_training.learning_rate",):
        if path in leaves and float(leaves[path]) <= 0.0:
            raise ValueError(f"{path} must be positive; found {leaves[path]}.")

    # Learning rate grid: positive and monotonically increasing
    if "network_training.learning_rate_grid" in leaves:
        grid: List[float] = [float(v) for v in leaves["network_training.learning_rate_grid"]]
        if any(v <= 0.0 for v in grid) or grid != sorted(grid):
            raise ValueError(
                f"learning_rate_grid must be positive and sorted; found {grid}."
            )

    # Weight decay grid: non-negative and sorted
    if "network_training.weight_decay_grid" in leaves:
        wd: List[float] = [float(v) for v in leaves["network_training.weight_decay_grid"]]
        if any(v < 0.0 for v in wd) or wd != sorted(wd):
            raise ValueError(f"weight_decay_grid must be non-negative and sorted; found {wd}.")

    # Lambda grids: non-negative and sorted
    for path in ("network_training.loss.lambda_grid_simulation",
                 "network_training.loss.lambda_grid_realdata"):
        if path in leaves:
            lambdas: List[float] = [float(v) for v in leaves[path]]
            if any(v < 0.0 for v in lambdas) or lambdas != sorted(lambdas):
                raise ValueError(
                    f"{path} must be non-negative and sorted; found {lambdas}."
                )

    # Epochs, patience, seeds and restarts must be strictly positive
    for path in ("network_training.max_epochs", "network_training.early_stopping_patience",
                 "network_training.num_seeds", "network_training.num_restarts_per_config"):
        if path in leaves and int(leaves[path]) < 1:
            raise ValueError(f"{path} must be >= 1; found {leaves[path]}.")

    # The clipping epsilon must lie in (0, 0.5)
    if "frontier.solver.clipping_epsilon" in leaves:
        eps: float = float(leaves["frontier.solver.clipping_epsilon"])
        if not 0.0 < eps < 0.5:
            raise ValueError(f"clipping_epsilon must be in (0, 0.5); found {eps}.")

    # The logspace grid must be ordered after expansion
    if "frontier.xi_grid.logspace_start" in leaves and "frontier.xi_grid.logspace_end" in leaves:
        if float(leaves["frontier.xi_grid.logspace_start"]) >= float(
                leaves["frontier.xi_grid.logspace_end"]):
            raise ValueError(
                f"xi_grid.logspace_start must be below logspace_end; found "
                f"{leaves['frontier.xi_grid.logspace_start']} vs "
                f"{leaves['frontier.xi_grid.logspace_end']}."
            )

    # The WoE smoothing constant must be positive
    if "data_preprocessing.categorical_encoding.zero_frequency_smoothing" in leaves:
        alpha: float = float(leaves["data_preprocessing.categorical_encoding.zero_frequency_smoothing"])
        if alpha <= 0.0:
            raise ValueError(f"zero_frequency_smoothing must be positive; found {alpha}.")

    # The DGP covariate allocation must be 0.5 fixed per spec
    if "simulation.sensitive_A.p" in leaves:
        p_sensitive: float = float(leaves["simulation.sensitive_A.p"])
        if not 0.0 < p_sensitive < 1.0:
            raise ValueError(f"simulation.sensitive_A.p must be in (0,1); found {p_sensitive}.")

def write_placeholder_audit_log(records: List[PlaceholderRecord],
                                audit_path: Path) -> Path:
    """
    Serialize the placeholder audit log atomically.

    Parameters
    ----------
    records : List[PlaceholderRecord]
        The records to persist.
    audit_path : Path
        The destination JSON file.

    Returns
    -------
    Path
        The written file, verified by re-reading after writing.

    Raises
    ------
    ValueError
        If the written file does not round-trip to the same contents.

    """
    payload: Dict[str, object] = {
        "note": (
            "Every value below is annotated UNSPECIFIED IN MANUSCRIPT in "
            "the revised study inputs. Each is a rational placeholder "
            "consistent with econometric and machine-learning practice and "
            "is subject to the Task 12 sensitivity analysis."
        ),
        "methodological_decisions": list(METHODOLOGICAL_DECISIONS),
        "count": len(records),
        "records": [
            {
                "path": r.path,
                "value": r.value,
                "rationale": r.rationale,
                "sensitivity_required": r.sensitivity_required,
            }
            for r in records
        ],
    }
    # Write atomically, then read back to prove the artifact exists
    temp: Path = audit_path.with_suffix(audit_path.suffix + ".tmp")
    temp.write_text(json.dumps(payload, indent=2) + "\n", encoding="utf-8")
    temp.replace(audit_path)
    round_trip: Dict[str, object] = json.loads(
        audit_path.read_text(encoding="utf-8")
    )
    if round_trip != payload:
        raise ValueError(f"audit log round-trip mismatch at {audit_path}")
    return audit_path

# -------------------------------------------------------------------------------------------------------------------------------
# Task 1, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def validate_inputs(
    config: Dict[str, Any],
    raw_structured_features_by_dataset: Dict[str, pd.DataFrame],
    raw_outcome_labels_by_dataset: Dict[str, pd.DataFrame],
    raw_sensitive_attribute_by_dataset: Dict[str, pd.DataFrame],
    raw_simulation_input: pd.DataFrame,
    schema_continuous: Dict[str, List[str]],
    schema_categorical: Dict[str, List[str]],
    datasets_metadata: Optional[Dict[str, Dict[str, object]]] = None,
    run_seeds: Iterable[int] = range(10),
    audit_log_path: Path = PLACEHOLDER_AUDIT_PATH,
) -> ValidationReport:
    """
    Validate every ingested parameter, data structure and config entry.

    Composes Step 1 (schema and key integrity), Step 2 (dataset statistics
    against metadata) and Step 3 (config completeness and placeholder
    audit) into a single immutable report.

    Parameters
    ----------
    config : Dict[str, Any]
        The nested study configuration.
    raw_structured_features_by_dataset : Dict[str, pd.DataFrame]
        Feature frames keyed by dataset.
    raw_outcome_labels_by_dataset : Dict[str, pd.DataFrame]
        Label frames keyed by dataset.
    raw_sensitive_attribute_by_dataset : Dict[str, pd.DataFrame]
        Sensitive frames keyed by dataset.
    raw_simulation_input : pd.DataFrame
        The raw simulation input.
    schema_continuous : Dict[str, List[str]]
        Continuous feature names per dataset.
    schema_categorical : Dict[str, List[str]]
        Categorical feature names per dataset.
    datasets_metadata : Dict[str, Dict[str, object]]
        Metadata from config['real_data_experiment']['datasets'].
    run_seeds : Iterable[int], optional
        The accepted simulation replication seeds, defaults to 0..9.
    audit_log_path : Path, optional
        The placeholder audit log destination.

    Returns
    -------
    ValidationReport
        The immutable report binding validated frames, statistics and the
        placeholder record list.

    Raises
    ------
    ValueError
        If any validation rule is violated; the message names the offender.

    """
    logger.info("Starting Task 1: input and configuration validation.")

    # The metadata is CONFIG-authoritative: it is read from the config
    # internally, and an explicit caller argument must agree exactly
    metadata_from_config: Dict[str, Dict[str, object]] = config[
        "real_data_experiment"]["datasets"]
    if datasets_metadata is not None and datasets_metadata != metadata_from_config:
        raise ValueError(
            "datasets_metadata argument differs from "
            "config['real_data_experiment']['datasets']; the config is "
            "the single source of truth."
        )
    datasets_metadata = metadata_from_config
    # The expected dataset population comes from the same config block;
    # a simulation-only pipeline run (no real containers supplied) skips
    # the expected-population rule rather than failing on an intentionally
    # empty real-data axis
    expected_dataset_keys: Sequence[str] = (
        sorted(datasets_metadata.keys())
        if raw_structured_features_by_dataset else ()
    )

    # Step 1: schema and key integrity, with record_id indexing and the
    # expected eight-dataset population asserted
    features, labels, sensitive, simulation, generated = validate_raw_data_schemas(
        raw_structured_features_by_dataset,
        raw_outcome_labels_by_dataset,
        raw_sensitive_attribute_by_dataset,
        raw_simulation_input,
        run_seeds,
        schema_continuous,
        schema_categorical,
        expected_dataset_keys,
    )
    logger.info(
        f"Step 1 passed: {len(features)} datasets validated; simulation "
        f"record_ids {'generated' if generated else 'pre-existing and validated'}."
    )

    # Step 2: observed statistics versus the metadata tolerance band
    observed: List[DatasetStatistics] = []
    for dataset in sorted(features.keys()):
        observed.append(compute_dataset_statistics(
            features[dataset], labels[dataset], sensitive[dataset], dataset
        ))
    statistics: List[DatasetStatistics] = validate_statistics_against_metadata(
        observed, datasets_metadata
    )
    logger.info(f"Step 2 passed: {len(statistics)} datasets within tolerance.")

    # Step 3: config completeness, placeholder audit and range rules
    missing: List[str] = find_missing_config_paths(config)
    if missing:
        raise ValueError(
            f"config is incomplete; missing path(s): {missing}"
        )
    validate_numeric_parameter_ranges(config)
    placeholders: List[PlaceholderRecord] = record_placeholder_parameters(config)
    audit_path: Path = write_placeholder_audit_log(placeholders, audit_log_path)
    logger.info(
        f"Step 3 passed: {len(placeholders)} placeholders recorded at "
        f"{audit_path.name}."
    )

    logger.info("Task 1 completed: all inputs validated.")
    return ValidationReport(
        validated_structured=features,
        validated_labels=labels,
        validated_sensitive=sensitive,
        simulation_with_record_id=simulation,
        statistics=statistics,
        placeholder_records=placeholders,
        # The C8 record: whether the simulation key was generated and the
        # audit path actually written
        simulation_record_id_generated=bool(generated),
        audit_log_path=str(audit_path),
    )
````

````python
# Task 2: Cleanse Raw Data

# ==============================================================================
# Task 2: Cleanse Raw Data
# ==============================================================================

# Make the sibling task modules importable as the study pipeline grows

# Configure the module-level logger for stage diagnostics

# The project root for the cleansing artifact
# The default cleansing report artifact path
CLEANSING_REPORT_PATH: Path = PROJECT_ROOT / "cleansing_reports.json"
# The simulation duplication-rate inspection threshold
SIMULATION_DUPLICATION_RATE_THRESHOLD: float = 0.001
# The token replacing null categoricals, a distinct category by design
NULL_CATEGORICAL_TOKEN: str = "unknown"
# The missing-indicator column suffix
MISSING_INDICATOR_SUFFIX: str = "_missing"

@dataclass(frozen=True)
class CleansingReport:
    """
    One dataset's cleansing audit record.

    Attributes
    ----------
    dataset : str
        The dataset dictionary key.
    pre_cleaning_rows : int
        The row count before duplicate removal.
    post_cleaning_rows : int
        The row count after the full cleaning pass.
    dropped_duplicates : int
        The number of exact record_id duplicates removed.
    rule_inspection_counts : Dict[str, int]
        Per-rule counts of invalid rows identified (each rule raises, so
        a non-empty dict only appears in successfully corrected passes).
    missing_counts_by_feature : Dict[str, int]
        Post-cleaning missing counts for original features only; the
        indicator columns are excluded by design.
    categorical_null_replaced : Dict[str, int]
        Per-categorical count of nulls mapped to the unknown token.
    final_default_rate : float
        The post-cleaning default rate.
    final_sensitive_share : float
        The post-cleaning sensitive-class share.

    """

    # The dataset dictionary key
    dataset: str
    # Rows before duplicate removal
    pre_cleaning_rows: int
    # Rows after the full cleaning pass
    post_cleaning_rows: int
    # Exact record_id duplicates removed
    dropped_duplicates: int
    # Per-rule invalid-row counts
    rule_inspection_counts: Dict[str, int]
    # Post-cleaning missing counts, original features only
    missing_counts_by_feature: Dict[str, int]
    # Per-categorical null counts mapped to the unknown token
    categorical_null_replaced: Dict[str, int]
    # The final default rate
    final_default_rate: float
    # The final sensitive-class share
    final_sensitive_share: float

@dataclass(frozen=True)
class CleanedDataBundle:
    """
    The aggregate cleaned output for the whole study.

    Attributes
    ----------
    cleaned_datasets : Dict[str, pd.DataFrame]
        The cleaned, joined, record_id-indexed frames per dataset.
    cleaned_simulation : pd.DataFrame
        The deduplicated simulation frame.
    reports : Dict[str, CleansingReport]
        The per-dataset cleansing audit records.
    simulation_duplication_rate : float
        The observed per-block duplication rate.

    """

    # Cleaned dataset frames keyed by dataset
    cleaned_datasets: Dict[str, pd.DataFrame]
    # The deduplicated simulation frame
    cleaned_simulation: pd.DataFrame
    # Per-dataset cleansing reports
    reports: Dict[str, CleansingReport]
    # The observed simulation duplication rate
    simulation_duplication_rate: float

# -------------------------------------------------------------------------------------------------------------------------------
# Task 2, Step 1: Remove duplicate rows and enforce referential integrity
# -------------------------------------------------------------------------------------------------------------------------------
def drop_duplicate_record_ids(
    frame: pd.DataFrame, dataset: str,
) -> Tuple[pd.DataFrame, int]:
    """
    Drop exact duplicate record_id rows keeping the first occurrence.

    Parameters
    ----------
    frame : pd.DataFrame
        The record_id-indexed frame to deduplicate.
    dataset : str
        The dataset key, for error naming.

    Returns
    -------
    Tuple[pd.DataFrame, int]
        The deduplicated frame (original order preserved) and the number
        of duplicates removed.

    Raises
    ------
    ValueError
        If the frame index is not named record_id.

    """
    # The key must be the index for the population contract to hold
    if frame.index.name != "record_id":
        raise ValueError(
            f"dataset '{dataset}': the frame index must be 'record_id'; "
            f"found {frame.index.name!r}."
        )
    # Duplicates are the repeated entries of the identifier index
    duplicates: pd.Series = frame.index.duplicated(keep="first")
    n_duplicates: int = int(duplicates.sum())
    # Keep the first occurrence and preserve the original order
    deduped: pd.DataFrame = frame.loc[~duplicates].copy()
    return deduped, n_duplicates

def inner_join_dataset_components(
    features: pd.DataFrame,
    labels: pd.DataFrame,
    sensitive: pd.DataFrame,
    dataset: str,
) -> pd.DataFrame:
    """
    Inner-join the three components on the record_id index.

    Equation (Task 2 Step 1): the join must satisfy
    n_joined = n_features = n_labels = n_sensitive, and must align on the
    record_id INDEX rather than on row order.

    B-24 remediation: a column label shared between the feature frame and
    the sensitive frame (age being the obvious candidate) previously
    produced DUPLICATE columns in the concatenated result instead of
    raising, after which every downstream selection by that name became
    ambiguous. The collision is now detected before the concatenation.

    Parameters
    ----------
    features : pd.DataFrame
        The feature frame.
    labels : pd.DataFrame
        The outcome label frame.
    sensitive : pd.DataFrame
        The sensitive frame.
    dataset : str
        The dataset key, for error naming.

    Returns
    -------
    pd.DataFrame
        The joined frame indexed by record_id.

    Raises
    ------
    ValueError
        If any index is unnamed, duplicated, the column labels collide, or
        the join loses rows.

    """
    # Every index must be the record_id key before alignment
    for name, frame in (("features", features), ("labels", labels),
                        ("sensitive", sensitive)):
        if frame.index.name != "record_id":
            raise ValueError(
                f"dataset '{dataset}': the {name} frame index must be "
                f"'record_id'."
            )
        if not frame.index.is_unique:
            raise ValueError(
                f"dataset '{dataset}': the {name} frame index carries "
                f"duplicates; deduplication must precede the join."
            )
    # The label projection is fixed: only the outcome column participates
    if "y_default" not in labels.columns:
        raise ValueError(
            f"dataset '{dataset}': the labels frame must carry 'y_default'."
        )
    label_block: pd.DataFrame = labels[["y_default"]]
    # B-24: detect label collisions BEFORE concatenating. pd.concat happily
    # produces duplicate column names, which silently breaks every
    # downstream selection by label
    feature_columns: Set[str] = set(map(str, features.columns))
    label_columns: Set[str] = set(map(str, label_block.columns))
    sensitive_columns: Set[str] = set(map(str, sensitive.columns))
    # Each pairwise intersection is reported separately so the operator can
    # see exactly which frames disagree
    collisions: Dict[str, Set[str]] = {
        "features/labels": feature_columns & label_columns,
        "features/sensitive": feature_columns & sensitive_columns,
        "labels/sensitive": label_columns & sensitive_columns,
    }
    offending: Dict[str, List[str]] = {
        pair: sorted(shared) for pair, shared in collisions.items() if shared
    }
    if offending:
        raise ValueError(
            f"dataset '{dataset}': column-label collision(s) would create "
            f"duplicate columns in the join: {offending}. Rename or drop the "
            f"shared label(s) before joining; a duplicated column makes every "
            f"downstream selection by name ambiguous."
        )
    # The inner join aligns on the index, never on row order
    joined: pd.DataFrame = pd.concat(
        [features, label_block, sensitive], axis=1, join="inner"
    )
    # A joined frame must keep each component's population exactly
    if len(joined) != len(features):
        raise ValueError(
            f"dataset '{dataset}': join lost rows; expected "
            f"{len(features)}, got {len(joined)}."
        )
    # The post-condition the collision guard exists to protect
    if joined.columns.duplicated().any():
        raise ValueError(
            f"dataset '{dataset}': the joined frame carries duplicate column "
            f"labels: "
            f"{sorted(joined.columns[joined.columns.duplicated()].tolist())}."
        )
    return joined

def verify_joined_counts(
    features: pd.DataFrame,
    labels: pd.DataFrame,
    sensitive: pd.DataFrame,
    joined: pd.DataFrame,
    dataset: str,
) -> None:
    """
    Assert the four-way population equality after the join.

    Parameters
    ----------
    features, labels, sensitive : pd.DataFrame
        The three components.
    joined : pd.DataFrame
        The joined frame.
    dataset : str
        The dataset key.

    Raises
    ------
    ValueError
        If n_joined != n_features != n_labels != n_sensitive.

    """
    # Equation: n_joined = n_features = n_labels = n_sensitive
    n_features: int = len(features.index.unique())
    n_labels: int = len(labels.index.unique())
    n_sensitive: int = len(sensitive.index.unique())
    n_joined: int = len(joined)
    if not (n_joined == n_features == n_labels == n_sensitive):
        raise ValueError(
            f"dataset '{dataset}': population inequality: n_joined="
            f"{n_joined}, n_features={n_features}, n_labels={n_labels}, "
            f"n_sensitive={n_sensitive}."
        )
    # The joined key must be unique by construction
    if not joined.index.is_unique:
        raise ValueError(
            f"dataset '{dataset}': joined record_id index is not unique."
        )

def deduplicate_simulation_block_rows(
    simulation: pd.DataFrame,
    threshold: float = SIMULATION_DUPLICATION_RATE_THRESHOLD,
) -> Tuple[pd.DataFrame, float]:
    """
    Deduplicate exact full-row repetitions, scoped to each block.

    A duplicate is an exact repetition of all NON-KEY columns within a
    (mode, seed) block: record_id participates in the block's composite
    key, so including it would make within-block duplicates structurally
    impossible. The duplication rate is computed per block from the
    block-scoped mask, and the reported rate is the worst block rate.

    Parameters
    ----------
    simulation : pd.DataFrame
        The simulation frame with record_id as the first column.
    threshold : float, optional
        The duplication-rate inspection threshold, by default 0.001.

    Returns
    -------
    Tuple[pd.DataFrame, float]
        The deduplicated frame and the observed worst-block rate.

    Raises
    ------
    ValueError
        If the frame lacks mode/seed, or any block's rate exceeds the
        threshold.

    """
    # The block key columns must be present for block-scoped duplicate
    # detection
    for required in ("mode", "seed"):
        if required not in simulation.columns:
            raise ValueError(
                f"simulation input must carry '{required}' for block "
                f"duplicate detection."
            )
    # The duplication tuple excludes the key: within-block duplicates of
    # the full NON-KEY columns are the defined repetition unit
    duplication_columns: List[str] = [
        c for c in simulation.columns if c != "record_id"
    ]
    # The block is the unit of detection: the mask is computed per block
    # over the non-key columns
    block_rates: Dict[Tuple[str, int], float] = {}
    keep_mask: np.ndarray = np.ones(len(simulation), dtype=bool)
    for (mode_val, seed_val), block in simulation.groupby(
            ["mode", "seed"], sort=False):
        # The block-local duplicate mask with the first occurrence kept
        block_mask: pd.Series = block.duplicated(
            subset=duplication_columns, keep="first"
        )
        # The block rate is computed on the block's own population
        block_rate: float = int(block_mask.sum()) / max(len(block), 1)
        block_rates[(str(mode_val), int(seed_val))] = block_rate
        # Above the inspection threshold the source must be reconciled;
        # the block identity is named in the error
        if block_rate > threshold:
            raise ValueError(
                f"simulation block (mode='{mode_val}', seed={seed_val}) "
                f"duplication rate {block_rate:.6f} exceeds the "
                f"inspection threshold {threshold:.6f}; inspect the raw "
                f"source before proceeding."
            )
        # The global mask records which rows of the block survive
        keep_mask[np.asarray(block_mask.index)] = ~np.asarray(block_mask)
    # The reported rate is the worst observed block rate
    worst_rate: float = max(block_rates.values()) if block_rates else 0.0
    return simulation.loc[keep_mask].copy(), worst_rate

def prepare_joined_datasets(
    validation_report: ValidationReport,
) -> Dict[str, pd.DataFrame]:
    """
    Deduplicate and join every dataset of the validated report.

    Parameters
    ----------
    validation_report : ValidationReport
        The Task 1 report carrying validated, record_id-indexed frames.

    Returns
    -------
    Dict[str, pd.DataFrame]
        One joined clean frame per dataset.

    Raises
    ------
    KeyError
        If a dataset key is missing from any validated container.

    """
    joined: Dict[str, pd.DataFrame] = {}
    for dataset in sorted(validation_report.validated_structured.keys()):
        # A missing component is a caller defect that must be named
        for container in (validation_report.validated_labels,
                          validation_report.validated_sensitive):
            if dataset not in container:
                raise KeyError(
                    f"dataset '{dataset}' missing from a validated "
                    f"container; Task 1 must run before Task 2."
                )
        # Step 1, part A: drop exact record_id duplicates, first kept
        features, _ = drop_duplicate_record_ids(
            validation_report.validated_structured[dataset], dataset
        )
        labels, _ = drop_duplicate_record_ids(
            validation_report.validated_labels[dataset], dataset
        )
        sensitive, _ = drop_duplicate_record_ids(
            validation_report.validated_sensitive[dataset], dataset
        )
        # Step 1, part B: inner join on the record_id index
        frame = inner_join_dataset_components(features, labels, sensitive,
                                              dataset)
        # Step 1, part C: the equation n_joined = n_features = ...
        verify_joined_counts(features, labels, sensitive, frame, dataset)
        joined[dataset] = frame
    return joined

# -------------------------------------------------------------------------------------------------------------------------------
# Task 2, Step 2: Clean invalid values and standardise missing-value indicators
# -------------------------------------------------------------------------------------------------------------------------------
def validate_numeric_finiteness(
    frame: pd.DataFrame, continuous: List[str], dataset: str,
) -> None:
    """
    Assert every continuous feature is finite.

    Parameters
    ----------
    frame : pd.DataFrame
        The frame being cleaned (may hold NaN to be imputed later; the
        rule here rejects Inf only and records NaN counts separately).
    continuous : List[str]
        The continuous feature names to check.
    dataset : str
        The dataset key, for error naming.

    Raises
    ------
    ValueError
        If any continuous feature contains an infinite value.

    """
    # Infinite values corrupt every downstream arithmetic; NaN is a
    # different, imputation-addressed state and is counted, not rejected
    for column in continuous:
        if column not in frame.columns:
            continue
        values: np.ndarray = frame[column].to_numpy(dtype=float)
        n_inf: int = int(np.isinf(values).sum())
        if n_inf:
            raise ValueError(
                f"dataset '{dataset}': continuous feature '{column}' "
                f"carries {n_inf} infinite value(s); infinities must be "
                f"repaired before cleaning."
            )

def validate_plausibility_rules(
    frame: pd.DataFrame, dataset: str,
    named_rules: Optional[Dict[str, Tuple[float, float]]] = None,
) -> Dict[str, int]:
    """
    Enforce the named plausibility rules on the joined frame.

    Rules (from the task list): age in [18, 100]; loan_amount >= 0 and
    finite; checking_balance may be negative but must be finite; any
    other continuous feature only needs finiteness.

    Parameters
    ----------
    frame : pd.DataFrame
        The joined frame to inspect.
    dataset : str
        The dataset key, for error naming.
    named_rules : Optional[Dict[str, Tuple[float, float]]], optional
        Explicit (lower, upper) bounds per feature. Defaults to the study
        rules for age, loan_amount and checking_balance.

    Returns
    -------
    Dict[str, int]
        Per-rule counts of examined rows (a successful pass records the
        inspected count; any violation raises).

    Raises
    ------
    ValueError
        If any row violates a named rule.

    """
    # The study rules: age within [18, 100]; loan_amount non-negative
    # (upper bound +inf); checking_balance finite only (no bound)
    if named_rules is None:
        named_rules = {
            "age": (18.0, 100.0),
            "loan_amount": (0.0, np.inf),
            "checking_balance": (-np.inf, np.inf),
        }
    inspected: Dict[str, int] = {}
    for column, (lower, upper) in named_rules.items():
        if column not in frame.columns:
            continue
        values: np.ndarray = frame[column].to_numpy(dtype=float)
        inspected[column] = int(len(values))
        # The range rule: every finite value must lie within [lower, upper]
        finite_mask: np.ndarray = np.isfinite(values)
        if not np.all((values[finite_mask] >= lower) &
                      (values[finite_mask] <= upper)):
            violation_mask = (
                finite_mask & ((values < lower) | (values > upper))
            )
            n_violations: int = int(violation_mask.sum())
            raise ValueError(
                f"dataset '{dataset}': {n_violations} value(s) of "
                f"'{column}' violate the plausible range "
                f"[{lower}, {upper}]."
            )
    return inspected

def normalise_categorical_values(
    frame: pd.DataFrame, categorical: List[str], dataset: str,
) -> Dict[str, int]:
    """
    Trim whitespace and lowercase every categorical column.

    Parameters
    ----------
    frame : pd.DataFrame
        The joined frame (mutated in place on its own copy downstream;
        the caller passes its working copy).
    categorical : List[str]
        The categorical feature names.
    dataset : str
        The dataset key, for error naming.

    Returns
    -------
    Dict[str, int]
        Per-feature count of values whose string form changed.

    """
    changed: Dict[str, int] = {}
    for column in categorical:
        if column not in frame.columns:
            continue
        # Strings are normalised vectorised; NaN survives as NaN
        raw: pd.Series = frame[column]
        texts: pd.Series = raw.where(raw.isna(), raw.astype(str).str.strip().str.lower())
        # Count the rows whose representation actually changed
        changed_counts: int = int((texts != raw).sum())
        frame[column] = texts
        changed[column] = changed_counts
    return changed

def populate_missing_indicators(
    frame: pd.DataFrame, continuous: List[str], indicator_suffix: str,
) -> Dict[str, int]:
    """
    Create a binary indicator per continuous feature with missing values.

    Parameters
    ----------
    frame : pd.DataFrame
        The joined frame (working copy).
    continuous : List[str]
        The continuous feature names.
    indicator_suffix : str
        The suffix appended to the indicator column name.

    Returns
    -------
    Dict[str, int]
        Per-feature missing counts, and the indicator columns are appended
        to the frame.

    Raises
    ------
    ValueError
        If a generated indicator name collides with an existing column.

    """
    missing: Dict[str, int] = {}
    for column in continuous:
        if column not in frame.columns:
            continue
        n_missing: int = int(frame[column].isna().sum())
        if n_missing == 0:
            continue
        # The indicator is a legitimate feature: 1 when the value is NaN
        indicator_name: str = f"{column}{indicator_suffix}"
        if indicator_name in frame.columns:
            raise ValueError(
                f"indicator column '{indicator_name}' already exists; "
                f"the suffix would overwrite a feature."
            )
        frame[indicator_name] = (frame[column].isna()).astype(np.int64)
        missing[column] = n_missing
    return missing

def map_null_categoricals_to_unknown(
    frame: pd.DataFrame, categorical: List[str],
    token: str = NULL_CATEGORICAL_TOKEN,
) -> Dict[str, int]:
    """
    Map null categorical values to a distinct unknown category.

    Parameters
    ----------
    frame : pd.DataFrame
        The joined frame (working copy).
    categorical : List[str]
        The categorical feature names.
    token : str, optional
        The replacement token, by default "unknown".

    Returns
    -------
    Dict[str, int]
        Per-feature null counts replaced.

    """
    replaced: Dict[str, int] = {}
    for column in categorical:
        if column not in frame.columns:
            continue
        n_null: int = int(frame[column].isna().sum())
        if n_null:
            # A null never becomes the literal text of silence: the
            # distinct token carries the missing state explicitly
            frame[column] = frame[column].fillna(token)
            replaced[column] = n_null
    return replaced

def cleanse_dataset_frame(
    joined: pd.DataFrame,
    dataset: str,
    continuous: List[str],
    categorical: List[str],
    indicator_suffix: str = MISSING_INDICATOR_SUFFIX,
) -> Tuple[pd.DataFrame, Dict[str, int], Dict[str, int], Dict[str, int]]:
    """
    Apply the Step 2 cleaning rules to one joined frame.

    Parameters
    ----------
    joined : pd.DataFrame
        The joined, deduplicated frame.
    dataset : str
        The dataset key, for error naming.
    continuous : List[str]
        The continuous feature names.
    categorical : List[str]
        The categorical feature names.
    indicator_suffix : str, optional
        The missing-indicator suffix, by default "_missing".

    Returns
    -------
    Tuple[pd.DataFrame, Dict[str, int], Dict[str, int], Dict[str, int]]
        The cleaned frame, the inspected-rule counts, the missing counts
        per original feature, and the categorical null replacement counts.

    """
    # The working copy isolates all mutation from the caller's frame
    frame: pd.DataFrame = joined.copy()
    # Finiteness is checked before indicators exist, so the indicator
    # columns cannot poison the finite scan
    validate_numeric_finiteness(frame, continuous, dataset)
    # Named plausibility rules reject out-of-range values with a report
    inspected: Dict[str, int] = validate_plausibility_rules(frame, dataset)
    # Categorical strings are normalised for the WoE encoder
    normalise_categorical_values(frame, categorical, dataset)
    # Missing indicators are created before imputation and recorded as
    # legitimate design-matrix features
    missing: Dict[str, int] = populate_missing_indicators(
        frame, continuous, indicator_suffix
    )
    # Categorical nulls are mapped to a distinct category, never dropped
    replaced: Dict[str, int] = map_null_categoricals_to_unknown(
        frame, categorical
    )
    return frame, inspected, missing, replaced

# -------------------------------------------------------------------------------------------------------------------------------
# Task 2, Step 3: Audit final cleaned data
# -------------------------------------------------------------------------------------------------------------------------------
# In-notebook resolution (3A.8): the Task 3 cell rebinds
# derive_sensitive_attribute to the raw-scale function; this
# cell snapshots the Task 1 binding it passes to the report
_derive_sensitive_attribute_task2 = derive_sensitive_attribute

def compute_cleansing_report(
    dataset: str,
    pre_cleaning_rows: int,
    cleaned: pd.DataFrame,
    inspected: Dict[str, int],
    missing: Dict[str, int],
    replaced: Dict[str, int],
    sensitive_frame: pd.DataFrame,
    indicator_suffix: str = MISSING_INDICATOR_SUFFIX,
) -> CleansingReport:
    """
    Compose the cleansing audit for one dataset.

    Parameters
    ----------
    dataset : str
        The dataset key.
    pre_cleaning_rows : int
        The row count before duplicate removal.
    cleaned : pd.DataFrame
        The cleaned, joined frame.
    inspected : Dict[str, int]
        The documented rule-inspection counts.
    missing : Dict[str, int]
        Per-original-feature missing counts.
    replaced : Dict[str, int]
        Per-categorical null replacement counts.
    sensitive_frame : pd.DataFrame
        The cleaned sensitive frame, for the attribute derivation.
    indicator_suffix : str, optional
        The missing-indicator suffix, by default "_missing".

    Returns
    -------
    CleansingReport
        The populated audit record.

    """
    # The post-cleaning row count and the removed duplicate count
    post_cleaning_rows: int = len(cleaned)
    dropped_duplicates: int = pre_cleaning_rows - post_cleaning_rows
    # Missing counts are computed for ORIGINAL features only; the
    # indicator columns are structurally never-missing and are excluded
    original_features: List[str] = [
        c for c in cleaned.columns if not c.endswith(indicator_suffix)
    ]
    missing_counts: Dict[str, int] = {
        c: int(cleaned[c].isna().sum()) for c in original_features
        if c in cleaned.columns
    }
    # The final default rate from the cleaned label column
    final_default_rate: float = float(cleaned["y_default"].mean())
    # The final sensitive share via the Task 1 derivation rule
    a_values: pd.Series = _derive_sensitive_attribute_task2(
        cleaned, sensitive_frame, dataset
    )
    final_sensitive_share: float = float(a_values.mean())
    return CleansingReport(
        dataset=dataset,
        pre_cleaning_rows=pre_cleaning_rows,
        post_cleaning_rows=post_cleaning_rows,
        dropped_duplicates=dropped_duplicates,
        rule_inspection_counts=inspected,
        missing_counts_by_feature=missing_counts,
        categorical_null_replaced=replaced,
        final_default_rate=final_default_rate,
        final_sensitive_share=final_sensitive_share,
    )

def audit_cleaning_against_metadata(
    reports: Dict[str, CleansingReport],
    datasets_metadata: Dict[str, Dict[str, object]],
    tolerance: float = STATISTICS_TOLERANCE,
) -> None:
    """
    Compare final cleaned statistics against the metadata band.

    Parameters
    ----------
    reports : Dict[str, CleansingReport]
        The per-dataset cleansing reports.
    datasets_metadata : Dict[str, Dict[str, object]]
        The metadata from the config.
    tolerance : float, optional
        The absolute tolerance, by default 0.005.

    Raises
    ------
    ValueError
        If any final statistic deviates beyond tolerance, or a metadata
        entry is missing.

    """
    for dataset, report in reports.items():
        if dataset not in datasets_metadata:
            raise ValueError(
                f"dataset '{dataset}' has no metadata entry; the "
                f"cleaning audit cannot reconcile it."
            )
        meta: Dict[str, object] = datasets_metadata[dataset]
        # Equation: |observed - expected| <= tolerance for both rates
        default_ok: bool = bool(np.isclose(
            report.final_default_rate, float(meta["default_rate"]),
            atol=tolerance, rtol=0.0,
        ))
        share_ok: bool = bool(np.isclose(
            report.final_sensitive_share, float(meta["sensitive_share"]),
            atol=tolerance, rtol=0.0,
        ))
        if report.post_cleaning_rows != int(meta["n"]) or \
                not default_ok or not share_ok:
            raise ValueError(
                f"dataset '{dataset}': post-cleaning statistics deviate "
                f"from metadata. n: {report.post_cleaning_rows} vs "
                f"{meta['n']}; default rate: "
                f"{report.final_default_rate:.6f} vs "
                f"{(meta['default_rate']):.6f}; sensitive share: "
                f"{report.final_sensitive_share:.6f} vs "
                f"{meta['sensitive_share']:.6f}. Tolerance {tolerance} "
                f"exceeded; reconcile the raw source."
            )

def write_cleansing_artifact(
    reports: Dict[str, CleansingReport],
    artifact_path: Path,
) -> Path:
    """
    Serialize the cleansing audit to an immutable JSON artifact.

    Parameters
    ----------
    reports : Dict[str, CleansingReport]
        The per-dataset reports.
    artifact_path : Path
        The destination JSON file.

    Returns
    -------
    Path
        The written file, verified by re-reading.

    Raises
    ------
    ValueError
        If the artifact does not round-trip to identical content.

    """
    payload: Dict[str, object] = {
        "note": (
            "Cleansing audit: pre/post counts, removed duplicates, "
            "invalid-value inspection counts, original-feature missing "
            "counts, categorical null replacements, and final rates. "
            "The indicator columns are excluded from the missing map by "
            "design."
        ),
        "reports": {
            dataset: {
                "dataset": r.dataset,
                "pre_cleaning_rows": r.pre_cleaning_rows,
                "post_cleaning_rows": r.post_cleaning_rows,
                "dropped_duplicates": r.dropped_duplicates,
                "rule_inspection_counts": r.rule_inspection_counts,
                "missing_counts_by_feature": r.missing_counts_by_feature,
                "categorical_null_replaced": r.categorical_null_replaced,
                "final_default_rate": r.final_default_rate,
                "final_sensitive_share": r.final_sensitive_share,
            }
            for dataset, r in reports.items()
        },
    }
    # Atomic write followed by a round-trip equality proof
    temp: Path = artifact_path.with_suffix(artifact_path.suffix + ".tmp")
    temp.write_text(json.dumps(payload, indent=2) + "\n", encoding="utf-8")
    temp.replace(artifact_path)
    round_trip: Dict[str, object] = json.loads(
        artifact_path.read_text(encoding="utf-8")
    )
    if round_trip != payload:
        raise ValueError(f"cleansing artifact round-trip mismatch at {artifact_path}")
    return artifact_path

# -------------------------------------------------------------------------------------------------------------------------------
# Task 2, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def clean_raw_data(
    validation_report: ValidationReport,
    datasets_metadata: Dict[str, Dict[str, object]],
    schema_continuous: Dict[str, List[str]],
    schema_categorical: Dict[str, List[str]],
    indicator_suffix: str = MISSING_INDICATOR_SUFFIX,
    cleansing_artifact_path: Path = CLEANSING_REPORT_PATH,
) -> CleanedDataBundle:
    """
    Orchestrate the complete raw-data cleansing pipeline.

    Composes Step 1 (duplicate removal and referential integrity), Step 2
    (invalid values, categorical normalisation, missing indicators) and
    Step 3 (audit against metadata and artifact serialisation) over every
    dataset and the simulation input.

    Parameters
    ----------
    validation_report : ValidationReport
        The Task 1 report of validated, record_id-indexed containers.
    datasets_metadata : Dict[str, Dict[str, object]]
        The metadata for every dataset.
    schema_continuous : Dict[str, List[str]]
        Continuous feature names per dataset.
    schema_categorical : Dict[str, List[str]]
        Categorical feature names per dataset.
    indicator_suffix : str, optional
        The missing-indicator suffix, by default "_missing".
    cleansing_artifact_path : Path, optional
        The cleansing audit destination.

    Returns
    -------
    CleanedDataBundle
        The cleaned frames per dataset, the cleaned simulation frame,
        the per-dataset reports, and the simulation duplication rate.

    Raises
    ------
    ValueError
        If any integrity, plausibility or tolerance rule is violated.

    """
    logger.info("Starting Task 2: raw data cleansing.")

    # Step 1: deduplicate, join, and assert the four-way population
    joined_frames: Dict[str, pd.DataFrame] = prepare_joined_datasets(
        validation_report
    )
    # Step 1 (simulation): block-scoped full-row deduplication with rate
    # inspection
    cleaned_simulation, dup_rate = deduplicate_simulation_block_rows(
        validation_report.simulation_with_record_id
    )
    logger.info(
        f"Step 1 passed: {len(joined_frames)} datasets joined; simulation "
        f"dedup rate {dup_rate:.6f}."
    )

    # Step 2: plausibility, categorical normalisation, missing indicators
    cleaned_datasets: Dict[str, pd.DataFrame] = {}
    reports: Dict[str, CleansingReport] = {}
    for dataset in sorted(joined_frames.keys()):
        # The pre-cleaning count is captured before any transform
        pre_rows: int = len(joined_frames[dataset])
        # The full Step 2 rule set on one working copy
        cleaned, inspected, missing, replaced = cleanse_dataset_frame(
            joined_frames[dataset], dataset,
            schema_continuous.get(dataset, []),
            schema_categorical.get(dataset, []),
            indicator_suffix,
        )
        # The simulation numerics are also finite-checked once
        numeric_cols: List[str] = [
            c for c in cleaned_simulation.columns
            if c.endswith("_post_shift") or c in (
                "X2", "X3", "X4", "X5", "X6", "linear_eta",
                "nl_component", "interaction_component", "eta_true",
                "r_true")
        ]
        validate_numeric_finiteness(cleaned_simulation, numeric_cols,
                                    "simulation")
        cleaned_datasets[dataset] = cleaned
        # Step 3: compose and audit the report for this dataset
        report = compute_cleansing_report(
            dataset, pre_rows, cleaned, inspected, missing, replaced,
            validation_report.validated_sensitive[dataset], indicator_suffix,
        )
        reports[dataset] = report
    # Step 3: tolerance audit across every dataset
    audit_cleaning_against_metadata(reports, datasets_metadata)
    # Step 3: persist the audit as an immutable artifact
    audit_path: Path = write_cleansing_artifact(reports, cleansing_artifact_path)
    logger.info(
        f"Steps 2-3 passed: {len(reports)} datasets cleaned and audited; "
        f"artifact at {audit_path.name}."
    )

    logger.info("Task 2 completed: raw data cleansed.")
    return CleanedDataBundle(
        cleaned_datasets=cleaned_datasets,
        cleaned_simulation=cleaned_simulation,
        reports=reports,
        simulation_duplication_rate=dup_rate,
    )
````

````python
# Task 3: Preprocess Real-World Data

# ==============================================================================
# Task 3: Preprocess Real-World Data
# ==============================================================================

# Make the sibling task modules importable as the study pipeline grows

# Configure the module-level logger for stage diagnostics

# The project root for the preprocessing artifact
# The default preprocessing statistics artifact path
PREPROCESSING_ARTIFACT_PATH: Path = PROJECT_ROOT / "preprocessing_stats.json"
# The WoE additive smoothing constant
WOE_SMOOTHING_ALPHA: float = 0.5
# The standard deviation floor protecting division by near-zero variance
VARIANCE_FLOOR: float = 1e-12
# The minimum minority-class count that triggers a stratification warning
MINORITY_CLASS_WARNING_THRESHOLD: int = 10
# The absolute tolerance on the standardised mean/unit-variance checks
STANDARDISATION_TOLERANCE: float = 1e-8
# The missing-indicator column suffix (Task 2's convention)

@dataclass(frozen=True)
class PreprocessingStats:
    """
    Frozen training-split statistics for every transformation.

    Attributes
    ----------
    means : Dict[str, float]
        Training means per continuous feature.
    stds : Dict[str, float]
        Training standard deviations per continuous feature (floored).
    medians : Dict[str, float]
        Training medians per continuous feature with missing values.
    woe_maps : Dict[str, Dict[str, float]]
        Category-to-WoE mapping per categorical feature.
    global_woe : Dict[str, float]
        Training-share-weighted global WoE per categorical feature.
    alpha : float
        The smoothing constant used.

    """

    # Training means per continuous feature
    means: Dict[str, float]
    # Training standard deviations per continuous feature
    stds: Dict[str, float]
    # Training medians per continuous feature
    medians: Dict[str, float]
    # Category-to-WoE maps per categorical feature
    woe_maps: Dict[str, Dict[str, float]]
    # Global WoE fallback per categorical feature
    global_woe: Dict[str, float]
    # The smoothing constant
    alpha: float

@dataclass(frozen=True)
class PreprocessedDataBundle:
    """
    One dataset's preprocessed output.

    Attributes
    ----------
    dataset : str
        The dataset key.
    X_train, X_val, X_test : np.ndarray
        The design matrices, shape (n_part, p), float64.
    y_train, y_val, y_test : np.ndarray
        The binary labels, shape (n_part,), int64.
    a_train, a_val, a_test : np.ndarray
        The sensitive attributes, shape (n_part,), int64.
    train_ids, val_ids, test_ids : List[int]
        The record_id provenance arrays.
    feature_names : List[str]
        The ordered names of the X columns.
    stats : PreprocessingStats
        The frozen training statistics.

    """

    # The dataset dictionary key
    dataset: str
    # Design matrix, training split
    X_train: np.ndarray
    # Design matrix, validation split
    X_val: np.ndarray
    # Design matrix, test split
    X_test: np.ndarray
    # Labels, training split
    y_train: np.ndarray
    # Labels, validation split
    y_val: np.ndarray
    # Labels, test split
    y_test: np.ndarray
    # Sensitive attributes, training split
    a_train: np.ndarray
    # Sensitive attributes, validation split
    a_val: np.ndarray
    # Sensitive attributes, test split
    a_test: np.ndarray
    # Record ids of the training split
    train_ids: List[int]
    # Record ids of the validation split
    val_ids: List[int]
    # Record ids of the test split
    test_ids: List[int]
    # Ordered feature names of X
    feature_names: List[str]
    # Frozen training statistics
    stats: PreprocessingStats

# -------------------------------------------------------------------------------------------------------------------------------
# Task 3, Step 1: Stratified train/validation/test split
# -------------------------------------------------------------------------------------------------------------------------------
def stratified_record_id_split(
    frame: pd.DataFrame,
    label_column: str,
    train_ratio: float = 0.70,
    val_ratio: float = 0.10,
    test_ratio: float = 0.20,
    seed: int = 42,
) -> Dict[str, List[int]]:
    """
    Split a frame into disjoint nested-stratified record_id partitions.

    The split is performed twice per the task text: first the 20% test
    partition is separated with stratification on the label, then the
    remaining 80% is split into training and 10% validation using the
    relative val size val_ratio/(train_ratio+val_ratio) (0.125 at the
    study ratios).

    Parameters
    ----------
    frame : pd.DataFrame
        The record_id-indexed frame carrying the label column.
    label_column : str
        The binary label column used for stratification.
    train_ratio : float, optional
        The training share, by default 0.70.
    val_ratio : float, optional
        The validation share, by default 0.10.
    test_ratio : float, optional
        The test share, by default 0.20.
    seed : int, optional
        The deterministic split seed, by default 42.

    Returns
    -------
    Dict[str, List[int]]
        The three disjoint record_id lists under the keys train, val,
        test.

    Raises
    ------
    ValueError
        If ratios do not sum to 1, the frame is too small, or any
        partition violates the split contract.

    """
    # The ratios must form a proper partition of the population
    if abs((train_ratio + val_ratio + test_ratio) - 1.0) > 1e-6:
        raise ValueError(
            f"train/val/test ratios must sum to 1.0; found "
            f"{train_ratio + val_ratio + test_ratio}."
        )
    # Label presence is mandatory for stratification
    if label_column not in frame.columns:
        raise ValueError(
            f"label column '{label_column}' is missing from the frame."
        )
    if frame.empty:
        raise ValueError("cannot split an empty frame.")
    # The label must be binary for stratification semantics
    if not set(frame[label_column].tolist()) <= {0, 1}:
        raise ValueError(
            f"label column '{label_column}' must be binary; found "
            f"{sorted(set(frame[label_column].tolist()))}."
        )
    # The split operates on the index array; stratification is aligned
    # through the label series with the same index
    ids: np.ndarray = frame.index.to_numpy()
    y_values: pd.Series = frame[label_column]
    n_total: int = len(frame)
    # Integer test sizes are used for both stages: a float test_size
    # multiplied against the population carries binary-representation
    # drift that can flip sklearn's rounding by one row
    test_desired: int = int(round(test_ratio * n_total))
    if not 0 < test_desired < n_total:
        raise ValueError(
            f"test size {test_desired} is not within (0, {n_total})."
        )
    # Stage 1: separate the test partition (test_ratio of the population)
    remainder_ids, test_ids = train_test_split(
        ids, stratify=y_values, test_size=test_desired,
        random_state=seed,
    )
    # Stage 2: split the remainder into training and validation; the
    # validation count is the val_ratio of the ORIGINAL population
    val_desired: int = int(round(val_ratio * n_total))
    if not 0 < val_desired < len(remainder_ids):
        raise ValueError(
            f"val size {val_desired} is not within (0, "
            f"{len(remainder_ids)})."
        )
    # The remainder's labels align to the remainder ids by index position
    y_remainder: pd.Series = y_values.loc[remainder_ids].reset_index(drop=True)
    train_ids_np, val_ids_np = train_test_split(
        remainder_ids, stratify=y_remainder, test_size=val_desired,
        random_state=seed,
    )
    # Int64 conversion keeps the ids JSON-native forever
    train_ids: List[int] = [int(i) for i in train_ids_np]
    val_ids: List[int] = [int(i) for i in val_ids_np]
    test_ids: List[int] = [int(i) for i in test_ids]
    # Disjointness and full coverage are structural requirements
    if not (set(train_ids).isdisjoint(set(val_ids))
            and set(train_ids).isdisjoint(set(test_ids))
            and set(val_ids).isdisjoint(set(test_ids))):
        raise ValueError("split partitions overlap; the split is invalid.")
    if set(train_ids) | set(val_ids) | set(test_ids) != set(ids.tolist()):
        raise ValueError("split partitions do not cover every record_id.")
    # Both classes must be represented in every partition for meanable
    # metrics; a thin minority triggers a warning, never a silent pass
    for name, part_ids in (("train", train_ids), ("val", val_ids),
                           ("test", test_ids)):
        part_y: pd.Series = y_values.loc[part_ids]
        counts: Dict[int, int] = part_y.value_counts().to_dict()
        for class_value in (0, 1):
            if class_value not in counts:
                warnings.warn(
                    f"partition '{name}' carries no class {class_value} "
                    f"records; stratification could not represent it.",
                    RuntimeWarning,
                )
            elif counts[class_value] < MINORITY_CLASS_WARNING_THRESHOLD:
                warnings.warn(
                    f"partition '{name}' carries only "
                    f"{counts[class_value]} class {class_value} records; "
                    f"below the {MINORITY_CLASS_WARNING_THRESHOLD} "
                    f"recommendation.",
                    RuntimeWarning,
                )
        # The partition's default rate is reported for the audit
        part_rate: float = float(part_y.mean())
        logger.info(
            f"split '{name}': n={len(part_ids)}, default rate="
            f"{part_rate:.6f}."
        )
    return {"train": train_ids, "val": val_ids, "test": test_ids}

# -------------------------------------------------------------------------------------------------------------------------------
# Task 3, Step 2: Feature transformation on training statistics
# -------------------------------------------------------------------------------------------------------------------------------
def compute_standardisation_stats(
    train_frame: pd.DataFrame, continuous: List[str],
) -> Tuple[Dict[str, float], Dict[str, float]]:
    """
    Compute training means and floored standard deviations.

    Parameters
    ----------
    train_frame : pd.DataFrame
        The training partition frame.
    continuous : List[str]
        The continuous feature names.

    Returns
    -------
    Tuple[Dict[str, float], Dict[str, float]]
        Means and standard deviations per continuous feature.

    Raises
    ------
    ValueError
        If a continuous feature is missing or non-numeric.

    """
    means: Dict[str, float] = {}
    stds: Dict[str, float] = {}
    for column in continuous:
        if column not in train_frame.columns:
            raise ValueError(
                f"continuous feature '{column}' is missing from the "
                f"training frame."
            )
        # Numeric extraction is mandatory before any statistic
        values: np.ndarray = pd.to_numeric(
            train_frame[column], errors="raise"
        ).to_numpy(dtype=np.float64)
        with np.errstate(invalid="raise"):
            mean_value: float = float(np.nanmean(values))
            std_value: float = float(np.nanstd(values))
        # A near-zero denominator must be floored, never divided by
        if std_value < VARIANCE_FLOOR:
            warnings.warn(
                f"feature '{column}' is (near-)constant in training; its "
                f"standard deviation is floored to 1.0 so the "
                f"standardisation stays finite.",
                RuntimeWarning,
            )
            std_value = 1.0
        means[column] = mean_value
        stds[column] = std_value
    return means, stds

def apply_standardisation(
    frame: pd.DataFrame, means: Dict[str, float], stds: Dict[str, float],
    continuous: List[str],
) -> pd.DataFrame:
    """
    Apply training statistics: equation X_std = (X - Xbar)/s.

    Parameters
    ----------
    frame : pd.DataFrame
        The partition frame (working copy).
    means : Dict[str, float]
        Training means.
    stds : Dict[str, float]
        Training standard deviations.
    continuous : List[str]
        The continuous feature names.

    Returns
    -------
    pd.DataFrame
        The frame with continuous columns standardised.

    """
    # Work on a copy; the caller's frame is never mutated
    result: pd.DataFrame = frame.copy()
    for column in continuous:
        if column not in result.columns:
            continue
        # Equation: X_std = (X - Xbar_train) / s_train, both frozen
        result[column] = (
            pd.to_numeric(result[column], errors="raise").to_numpy(
                dtype=np.float64
            ) - means[column]
        ) / stds[column]
    return result

def compute_woe_maps(
    train_frame: pd.DataFrame, categorical: List[str],
    label_column: str, alpha: float = WOE_SMOOTHING_ALPHA,
    convention: str = WOE_CONVENTION,
) -> Tuple[Dict[str, Dict[str, float]], Dict[str, float]]:
    """
    Compute the training WoE map and global fallback per categorical.

    Three conventions are in play and they are NOT the same statistic. With
    good_c the non-default count in category c, bad_c the default count, and
    G and B the training totals:

      distribution_ratio (classical credit scoring, DEFAULT):
          WoE_c = ln[ ((good_c + alpha)/(G + 2 alpha))
                    / ((bad_c  + alpha)/(B + 2 alpha)) ]
      within_category_rate (the literal task-list wording):
          WoE_c = ln[ ((good_c + alpha)/(n_c + 2 alpha))
                    / ((bad_c  + alpha)/(n_c + 2 alpha)) ]
      within_category_count (what the previous implementation computed):
          WoE_c = ln[ (good_c + alpha) / (bad_c + alpha) ]

    B-11 remediation. The previous code implemented the COUNT form while the
    task text specifies rates, and neither substitution was registered.
    Against the classical definition the count form differs by the constant
    ln[(B + 2 alpha)/(G + 2 alpha)] per feature, which a linear model
    absorbs into its intercept — so the practical harm was limited — but
    against the task text the difference is material, because an additive
    0.5 on a COUNT is a very different smoothing strength from 0.5 on a RATE
    lying in the unit interval, where it swamps the signal entirely. Note
    that the rate form is algebraically identical to the count form
    (the n_c + 2 alpha denominators cancel); it is retained as a named
    option for completeness and to make that cancellation explicit rather
    than accidental.

    The global fallback is also corrected. It was the training-share
    weighted MEAN of the per-category values; the task specifies the
    statistic computed from the ENTIRE training sample, i.e. the pooled log
    ratio. The pooled value is now the default, with the weighted mean
    retained as an option.

    Smoothing note: alpha is applied to the COUNTS in every convention
    (the Jeffreys prior at alpha = 0.5), never to an already-formed rate.

    Parameters
    ----------
    train_frame : pd.DataFrame
        The training partition frame; the ONLY statistic source (K-15).
    categorical : List[str]
        The categorical feature names.
    label_column : str
        The binary label column, 1 = default.
    alpha : float, optional
        The additive smoothing constant, by default 0.5.
    convention : str, optional
        One of distribution_ratio, within_category_rate,
        within_category_count. Default distribution_ratio.

    Returns
    -------
    Tuple[Dict[str, Dict[str, float]], Dict[str, float]]
        Feature-to-category WoE maps, and the per-feature global fallback
        applied to categories unseen in training.

    Raises
    ------
    ValueError
        If the convention is unknown, alpha is non-positive, or the label
        column is absent or non-binary.

    """
    # The convention domain is fixed and must be stated explicitly
    if convention not in WOE_SUPPORTED_CONVENTIONS:
        raise ValueError(
            f"convention must be one of {list(WOE_SUPPORTED_CONVENTIONS)}; "
            f"received '{convention}'."
        )
    # Additive smoothing requires a strictly positive constant
    if alpha <= 0.0:
        raise ValueError(f"alpha must be positive; received {alpha}.")
    # The label drives every count and must be present and binary
    if label_column not in train_frame.columns:
        raise ValueError(
            f"label column '{label_column}' is absent from the training "
            f"frame."
        )
    label_values: np.ndarray = pd.to_numeric(
        train_frame[label_column], errors="raise"
    ).to_numpy()
    if not set(np.unique(label_values)).issubset({0, 1}):
        raise ValueError(
            f"label column '{label_column}' must be binary; found "
            f"{sorted(set(label_values.tolist()))}."
        )
    # Training totals: B = defaults, G = non-defaults. These are the
    # denominators of the classical distribution ratio
    total_bad: float = float(label_values.sum())
    total_good: float = float(len(label_values) - total_bad)
    woe_maps: Dict[str, Dict[str, float]] = {}
    global_woe: Dict[str, float] = {}
    for column in categorical:
        # A column absent from the partition contributes no map
        if column not in train_frame.columns:
            continue
        # Per-category default sum and size, computed on TRAINING only
        counts: pd.DataFrame = train_frame.groupby(
            column, observed=True
        )[label_column].agg(["sum", "count"]).reset_index()
        # good_c = non-default count in c; bad_c = default count in c
        counts["good"] = counts["count"] - counts["sum"]
        counts["bad"] = counts["sum"]
        woe_values: Dict[str, float] = {}
        for _, row in counts.iterrows():
            # The smoothed counts; alpha is ALWAYS applied to the counts
            good_c: float = float(row["good"]) + alpha
            bad_c: float = float(row["bad"]) + alpha
            if convention == "distribution_ratio":
                # Classical: the category's share of all non-defaulters
                # over its share of all defaulters
                numerator: float = good_c / (total_good + 2.0 * alpha)
                denominator: float = bad_c / (total_bad + 2.0 * alpha)
            elif convention == "within_category_rate":
                # The task-list wording; the n_c + 2 alpha denominators
                # cancel, which this explicit form makes visible
                category_size: float = float(row["count"]) + 2.0 * alpha
                numerator = good_c / category_size
                denominator = bad_c / category_size
            else:
                # The historical within-category log odds
                numerator = good_c
                denominator = bad_c
            # A non-positive denominator is impossible under smoothing, but
            # the guard makes the invariant explicit rather than assumed
            if denominator <= 0.0 or numerator <= 0.0:
                raise ValueError(
                    f"feature '{column}', category {row[column]!r}: the "
                    f"smoothed WoE ratio is non-positive "
                    f"({numerator} / {denominator}); check alpha."
                )
            # The map key is the STRING form, matching the canonicalisation
            # that normalise_categorical_values applies upstream. That
            # coupling is load-bearing: a type mismatch here would send
            # every lookup to the fallback and silently destroy the feature
            woe_values[str(row[column])] = float(
                np.log(numerator / denominator)
            )
        woe_maps[column] = woe_values
        # B-11(c): the global fallback is the POOLED training statistic,
        # computed from the entire training sample as the task prescribes
        pooled_numerator: float = total_good + alpha
        pooled_denominator: float = total_bad + alpha
        if convention == "distribution_ratio":
            # Pooled over the whole sample the two shares are each 1, so the
            # classical pooled value is the smoothed log odds of the sample
            pooled_numerator = (total_good + alpha) / (
                total_good + 2.0 * alpha)
            pooled_denominator = (total_bad + alpha) / (
                total_bad + 2.0 * alpha)
        global_woe[column] = float(
            np.log(pooled_numerator / pooled_denominator)
        )
    return woe_maps, global_woe

def apply_woe_encoding(
    frame: pd.DataFrame, categorical: List[str],
    woe_maps: Dict[str, Dict[str, float]],
    global_woe: Dict[str, float],
) -> pd.DataFrame:
    """
    Replace categorical values with their WoE encodings.

    Unknown categories (observed in validation or test only) receive the
    global training WoE, never a zero or an NaN.

    Parameters
    ----------
    frame : pd.DataFrame
        The partition frame (working copy).
    categorical : List[str]
        The categorical feature names.
    woe_maps : Dict[str, Dict[str, float]]
        Category WoE maps from training.
    global_woe : Dict[str, float]
        Global fallback WoE per feature.

    Returns
    -------
    pd.DataFrame
        The frame with categorical columns replaced by float64 WoE.

    """
    result: pd.DataFrame = frame.copy()
    for column in categorical:
        if column not in result.columns:
            continue
        # Map every value through the training WoE map
        mapped: np.ndarray = result[column].map(woe_maps[column]).to_numpy(
            dtype=object
        )
        # An unmapped value is a val/test-only category: global fallback
        unknown_mask: np.ndarray = pd.isna(
            pd.Series(mapped)
        ).to_numpy()
        mapped[unknown_mask] = global_woe[column]
        result[column] = mapped.astype(np.float64)
    return result

def compute_training_medians(
    train_frame: pd.DataFrame, continuous: List[str],
) -> Dict[str, float]:
    """
    Compute training medians for continuous features with missing values.

    Parameters
    ----------
    train_frame : pd.DataFrame
        The training partition frame.
    continuous : List[str]
        The continuous feature names.

    Returns
    -------
    Dict[str, float]
        Training medians per feature that carries missing values.

    """
    medians: Dict[str, float] = {}
    for column in continuous:
        if column not in train_frame.columns:
            continue
        values: np.ndarray = pd.to_numeric(
            train_frame[column], errors="raise"
        ).to_numpy(dtype=np.float64)
        # Only features with observed missingness need a statistic
        if np.isnan(values).any():
            medians[column] = float(np.nanmedian(values))
    return medians

def impute_missing_values(
    frame: pd.DataFrame, medians: Dict[str, float],
) -> pd.DataFrame:
    """
    Impute continuous missing values with the frozen training medians.

    Parameters
    ----------
    frame : pd.DataFrame
        The partition frame (working copy).
    medians : Dict[str, float]
        Training medians per feature.

    Returns
    -------
    pd.DataFrame
        The frame with missing continuous values replaced.

    """
    result: pd.DataFrame = frame.copy()
    for column, median_value in medians.items():
        if column in result.columns:
            # Fill with the training median only; indicators from Task 2
            # already carry the missingness information
            result[column] = result[column].fillna(median_value)
    return result

def transform_partition(
    split_frame: pd.DataFrame,
    means: Dict[str, float],
    stds: Dict[str, float],
    medians: Dict[str, float],
    woe_maps: Dict[str, Dict[str, float]],
    global_woe: Dict[str, float],
    continuous: List[str],
    categorical: List[str],
) -> pd.DataFrame:
    """
    Apply the full frozen training transformation to one partition.

    Parameters
    ----------
    split_frame : pd.DataFrame
        The partition frame.
    means, stds : Dict[str, float]
        Training standardisation statistics.
    medians : Dict[str, float]
        Training medians.
    woe_maps, global_woe : Dict[str, ...]
        Training WoE mappings.
    continuous : List[str]
        Continuous feature names.
    categorical : List[str]
        Categorical feature names.

    Returns
    -------
    pd.DataFrame
        The transformed partition frame.

    """
    # Imputation precedes scaling so no NaN reaches the z-score
    result: pd.DataFrame = impute_missing_values(
        split_frame.copy(), medians
    )
    # WoE encoding converts categoricals to float64 scores
    result = apply_woe_encoding(result, categorical, woe_maps, global_woe)
    # Standardisation applies the frozen training statistics
    result = apply_standardisation(result, means, stds, continuous)
    return result

def validate_standardised_statistics(
    transformed_train: pd.DataFrame, continuous: List[str],
    tolerance: float = STANDARDISATION_TOLERANCE,
) -> None:
    """
    Verify zero mean and unit variance on the transformed training frame.

    Parameters
    ----------
    transformed_train : pd.DataFrame
        The transformed training partition.
    continuous : List[str]
        The continuous feature names.
    tolerance : float, optional
        The comparison tolerance, by default 1e-8.

    Raises
    ------
    ValueError
        If any mean or variance deviates beyond the tolerance.

    """
    for column in continuous:
        if column not in transformed_train.columns:
            continue
        values: np.ndarray = transformed_train[column].to_numpy(dtype=np.float64)
        mean_value: float = float(np.mean(values))
        std_value: float = float(np.std(values))
        # A NaN mean compares False against the tolerance and passes
        # silently; it is an explicit defect and must be rejected
        if np.isnan(mean_value):
            raise ValueError(
                f"feature '{column}' has a NaN mean after "
                f"standardisation; the transformation pipeline produced "
                f"a corrupt column."
            )
        # Equation check: mean(X_std) = 0, std(X_std) = 1
        if abs(mean_value) > tolerance or abs(std_value - 1.0) > tolerance:
            raise ValueError(
                f"feature '{column}' is not standardised: mean "
                f"{mean_value:.8f}, std {std_value:.8f}; "
                f"tolerance {tolerance} exceeded."
            )

# -------------------------------------------------------------------------------------------------------------------------------
# Task 3, Step 3: Derive sensitive attribute and build final design matrices
# -------------------------------------------------------------------------------------------------------------------------------
def derive_sensitive_attribute_rawscale(
    raw_age_values: np.ndarray, dataset: str, threshold: int,
) -> np.ndarray:
    """
    Derive the binary sensitive attribute on the RAW age scale.

    Equation: A_i = 1{age_i <= threshold}, threshold = 25 (config
    data_preprocessing.sensitive_attribute.binarisation_threshold). The
    callable NEVER receives a transformed (standardised) age: the rule is
    a raw-scale threshold, so the derivation MUST run before any
    scaling, and the caller passes the raw-scale age array (raw_age from
    the sensitive frame, or the age column from the cleaned frame, both
    record_id-aligned).

    Parameters
    ----------
    raw_age_values : np.ndarray
        The RAW-scale age values, shape (n,), record_id-ordered.
    dataset : str
        The dataset key, for error naming.
    threshold : int
        The binarisation threshold from the config.

    Returns
    -------
    np.ndarray
        The derived A values, shape (n,), int64, in the same row order.

    Raises
    ------
    ValueError
        If the ages are non-finite (after NaN removal the rule would be
        ill-defined) or the threshold is invalid.

    """
    # The age source must be present and finite on the raw scale
    ages: np.ndarray = pd.to_numeric(
        pd.Series(raw_age_values), errors="raise"
    ).to_numpy(dtype=np.float64)
    if np.isnan(ages).any():
        raise ValueError(
            f"dataset '{dataset}': raw age carries "
            f"{int(np.isnan(ages).sum())} NaN value(s); the sensitive "
            f"rule is undefined for missing ages."
        )
    # Equation: A_i = 1{age_i <= threshold} on the raw scale only
    sensitive: np.ndarray = (ages <= float(threshold)).astype(np.int64)
    return sensitive

def build_design_tensors(
    transformed: pd.DataFrame,
    identifier_index: np.ndarray,
    dataset: str,
    a_precomputed: np.ndarray,
    continuous: List[str],
    categorical: List[str],
    structured_features: List[str],
    identifier_names: Set[str],
    indicator_suffix: str = MISSING_INDICATOR_SUFFIX,
) -> Tuple[np.ndarray, np.ndarray, np.ndarray, List[str]]:
    """
    Build the X, y and a tensors for one partition.

    The feature set is derived EXPLICITLY from the declared schema: the
    continuous features, the categorical features (post-WoE), the
    missing-indicator columns, and any additional registered structured
    features from the config — never from a hard-coded exclusion list.
    Any column that is neither a declared feature nor a declared
    identifier raises, so a stray identifier cannot leak into X.

    Parameters
    ----------
    transformed : pd.DataFrame
        The transformed partition frame, record_id-indexed.
    identifier_index : np.ndarray
        The record_id array (used for the signature check).
    dataset : str
        The dataset key, for error naming.
    a_precomputed : np.ndarray
        The RAW-scale derived sensitive attribute, shape (n,), int64,
        aligned by row order with the transformed frame.
    continuous : List[str]
        The schema continuous features.
    categorical : List[str]
        The schema categorical features (WoE-encoded in the frame).
    structured_features : List[str]
        Additional registered structured features from the config.
    identifier_names : Set[str]
        The declared identifier set (record_id, sensitive-only age,
        labels, and any config-declared identifier).
    indicator_suffix : str, optional
        The missing-indicator suffix, by default "_missing".

    Returns
    -------
    Tuple[np.ndarray, np.ndarray, np.ndarray, List[str]]
        X (n, p) float64, y (n,) int64, a (n,) int64, and the feature
        name list in schema order.

    Raises
    ------
    ValueError
        If a declared feature is absent, an undeclared column leaks in,
        or row/signature counts disagree.

    """
    # The label is mandatory before the matrices are assembled
    if "y_default" not in transformed.columns:
        raise ValueError(
            f"dataset '{dataset}': y_default is missing from the "
            f"transformed frame."
        )
    # The label must be binary for the design construction
    if not set(transformed["y_default"].tolist()) <= {0, 1}:
        raise ValueError(
            f"dataset '{dataset}': y_default must be binary; found "
            f"{sorted(set(transformed['y_default'].tolist()))}."
        )
    # Equation: the ordered feature set = continuous + categorical
    # (WoE columns) + missing indicators + registered structured extras
    feature_names: List[str] = []
    for column in list(continuous) + list(categorical):
        if column in transformed.columns and column not in feature_names:
            feature_names.append(column)
    # The missing-indicator columns follow their schema origin
    for column in continuous:
        indicator_name: str = f"{column}{indicator_suffix}"
        if indicator_name in transformed.columns and \
                indicator_name not in feature_names:
            feature_names.append(indicator_name)
    # The registered structured extras (not already included)
    for column in structured_features:
        if column in transformed.columns and column not in feature_names:
            feature_names.append(column)
    # Every declared feature must exist in the transformed frame
    missing_features: List[str] = [
        column for column in feature_names if column not in transformed.columns
    ]
    if missing_features:
        raise ValueError(
            f"dataset '{dataset}': declared feature(s) absent from the "
            f"transformed frame: {missing_features}."
        )
    # The unexpected-column check: a column that is neither a declared
    # feature nor a declared identifier is a leak, never a silent include
    declared_set: set = set(feature_names) | set(identifier_names)
    unexpected: List[str] = sorted(
        set(transformed.columns) - declared_set)
    if unexpected:
        raise ValueError(
            f"dataset '{dataset}': undeclared column(s) would leak into "
            f"the design: {unexpected}. Register them as features or "
            f"identifiers before proceeding."
        )
    # The design matrix must be entirely numeric
    design_data: pd.DataFrame = transformed[feature_names]
    for column in feature_names:
        if not pd.api.types.is_numeric_dtype(design_data[column]):
            raise ValueError(
                f"dataset '{dataset}': feature '{column}' is not numeric "
                f"after preprocessing; the WoE/standardisation pipeline "
                f"must have converted it."
            )
    # Equation: X rows, y rows and the RAW-scale a rows are the design
    X: np.ndarray = design_data.to_numpy(dtype=np.float64)
    y: np.ndarray = transformed["y_default"].to_numpy(dtype=np.int64)
    a: np.ndarray = a_precomputed
    # Every matrix must share the row count of the partition
    if not (X.shape[0] == y.shape[0] == a.shape[0] ==
            len(identifier_index)):
        raise ValueError(
            f"dataset '{dataset}': signature mismatch; X rows "
            f"{X.shape[0]}, y {y.shape[0]}, a {a.shape[0]}, ids "
            f"{len(identifier_index)}."
        )
    # NaN and Inf in a design matrix are a preprocessing defect
    if not np.isfinite(X).all():
        raise ValueError(
            f"dataset '{dataset}': the design matrix carries "
            f"{int((~np.isfinite(X)).sum())} non-finite entries."
        )
    return X, y, a, feature_names

def validate_sensitive_share(
    joined_a: np.ndarray, expected_share: float,
    tolerance: float = STATISTICS_TOLERANCE, dataset: str = "",
) -> None:
    """
    Verify the derived sensitive share against the metadata band.

    Parameters
    ----------
    joined_a : np.ndarray
        The derived A values across the population.
    expected_share : float
        The metadata sensitive share.
    tolerance : float, optional
        The absolute tolerance, by default 0.005.
    dataset : str
        The dataset key, for error naming.

    Raises
    ------
    ValueError
        If the share deviates beyond the tolerance.

    """
    observed: float = float(np.mean(joined_a))
    # Equation: |observed - expected| <= tolerance
    if not bool(np.isclose(observed, expected_share, atol=tolerance, rtol=0.0)):
        raise ValueError(
            f"dataset '{dataset}': sensitive share {observed:.6f} deviates "
            f"from metadata {expected_share:.6f} beyond tolerance "
            f"{tolerance}."
        )

def write_preprocessing_artifact(
    all_stats: Dict[str, PreprocessingStats], artifact_path: Path,
) -> Path:
    """
    Serialize the frozen preprocessing statistics as an artifact.

    Parameters
    ----------
    all_stats : Dict[str, PreprocessingStats]
        Per-dataset frozen training statistics.
    artifact_path : Path
        The destination JSON file.

    Returns
    -------
    Path
        The written file, verified by round-trip.

    """
    payload: Dict[str, object] = {
        "note": (
            "Frozen training-split statistics: means, floored standard "
            "deviations, medians, WoE maps and global fallbacks. These "
            "artifacts are the only statistics permitted to transform "
            "validation and test partitions, and they are reused by the "
            "CSER perturbation path (Task 9)."
        ),
        "datasets": {
            dataset: {
                "means": stats.means,
                "stds": stats.stds,
                "medians": stats.medians,
                "woe_maps": stats.woe_maps,
                "global_woe": stats.global_woe,
                "alpha": stats.alpha,
            }
            for dataset, stats in all_stats.items()
        },
    }
    # Atomic write with a round-trip equality proof
    temp: Path = artifact_path.with_suffix(artifact_path.suffix + ".tmp")
    temp.write_text(json.dumps(payload, indent=2) + "\n", encoding="utf-8")
    temp.replace(artifact_path)
    round_trip: Dict[str, object] = json.loads(
        artifact_path.read_text(encoding="utf-8")
    )
    if round_trip != payload:
        raise ValueError(
            f"preprocessing artifact round-trip mismatch at {artifact_path}"
        )
    return artifact_path

# -------------------------------------------------------------------------------------------------------------------------------
# Task 3, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def preprocess_real_data(
    config: Dict[str, Any],
    cleaned_bundle: CleanedDataBundle,
    datasets_metadata: Dict[str, Dict[str, object]],
    schema_continuous: Dict[str, List[str]],
    schema_categorical: Dict[str, List[str]],
    split_config: Optional[Dict[str, Any]] = None,
    woe_alpha: Optional[float] = None,
    artifact_path: Path = PREPROCESSING_ARTIFACT_PATH,
) -> Dict[str, PreprocessedDataBundle]:
    """
    Orchestrate the full real-data preprocessing for every dataset.

    Composes Step 1 (stratified split, record_id lists), Step 2 (training
    statistics and transformations) and Step 3 (sensitive derivation and
    tensor construction), enforcing the training-only information barrier.

    Parameters
    ----------
    config : Dict[str, Any]
        The study configuration. Its data_preprocessing section is the
        single source of truth for the split settings, the
        binarisation threshold and the WoE alpha; an explicit
        split_config or woe_alpha argument is honoured only when it
        agrees with (or overrides explicitly) the config values.
    cleaned_bundle : CleanedDataBundle
        The Task 2 cleaned dataset frames.
    datasets_metadata : Dict[str, Dict[str, object]]
        The metadata for every dataset.
    schema_continuous : Dict[str, List[str]]
        Continuous feature names per dataset.
    schema_categorical : Dict[str, List[str]]
        Categorical feature names per dataset.
    split_config : Optional[Dict[str, Any]], optional
        The train/val/test split settings; defaults to 0.70/0.10/0.20 with
        seed 42 when absent.
    woe_alpha : float, optional
        The WoE smoothing constant, by default 0.5.
    artifact_path : Path, optional
        The preprocessing artifact destination.

    Returns
    -------
    Dict[str, PreprocessedDataBundle]
        Per-dataset preprocessed bundles.

    Raises
    ------
    ValueError
        If a dataset key, metadata entry or schema component is missing.

    """
    # The split settings are config-authoritative: the config section is
    # read directly, and an explicit overrides argument must agree
    if split_config is not None:
        # An explicit argument is honoured only when it matches the config
        config_split: Dict[str, Any] = config["data_preprocessing"][
            "train_val_test_split"]
        if any(split_config.get(key) != config_split.get(key)
               for key in ("train_ratio", "val_ratio", "test_ratio",
                           "seed")):
            raise ValueError(
                "split_config disagrees with "
                "config['data_preprocessing']['train_val_test_split']; "
                "the config is the single source of truth."
            )
    else:
        split_config = config["data_preprocessing"][
            "train_val_test_split"]
    # The WoE alpha comes from the config block; a caller override must
    # be explicit and is honoured as-is
    if woe_alpha is None:
        woe_alpha = float(config["data_preprocessing"][
            "categorical_encoding"]["zero_frequency_smoothing"])
    # The missing-indicator suffix follows the config's cleaning block
    indicator_suffix: str = str(config["data_preprocessing"][
        "missing_value_handling"]["missing_indicator_suffix"])
    logger.info("Starting Task 3: real-data preprocessing.")

    bundles: Dict[str, PreprocessedDataBundle] = {}
    all_stats: Dict[str, PreprocessingStats] = {}
    for dataset in sorted(cleaned_bundle.cleaned_datasets.keys()):
        # A metadata entry must exist for the sensitive-share audit
        if dataset not in datasets_metadata:
            raise ValueError(
                f"dataset '{dataset}' has no metadata entry."
            )
        frame: pd.DataFrame = cleaned_bundle.cleaned_datasets[dataset]
        continuous: List[str] = schema_continuous.get(dataset, [])
        # The RAW-scale cleaned frame is the sensitive-attribute source:
        # its age columns are pre-standardisation by construction
        cleaned_frame: pd.DataFrame = frame
        categorical: List[str] = schema_categorical.get(dataset, [])

        # Step 1: the stratified split, keyed by record_id
        partitions: Dict[str, List[int]] = stratified_record_id_split(
            frame, "y_default",
            train_ratio=float(split_config["train_ratio"]),
            val_ratio=float(split_config["val_ratio"]),
            test_ratio=float(split_config["test_ratio"]),
            seed=int(split_config["seed"]),
        )

        # Step 2: training statistics, frozen once. Medians come from the
        # RAW training values; means and standard deviations come from the
        # IMPUTED training sample, which is the sample whose distributions
        # the standardisation equations must zero and unit-normalise
        train_frame: pd.DataFrame = frame.loc[partitions["train"]]
        medians: Dict[str, float] = compute_training_medians(
            train_frame, continuous
        )
        training_imputed: pd.DataFrame = impute_missing_values(
            train_frame, medians
        )
        means, stds = compute_standardisation_stats(
            training_imputed, continuous
        )
        woe_maps, global_woe = compute_woe_maps(
            train_frame, categorical, "y_default", woe_alpha
        )
        stats: PreprocessingStats = PreprocessingStats(
            means=means, stds=stds, medians=medians,
            woe_maps=woe_maps, global_woe=global_woe, alpha=woe_alpha,
        )
        all_stats[dataset] = stats

        # Step 2 continued: transform train, val and test with the SAME
        # frozen statistics (the information barrier)
        transformed: Dict[str, pd.DataFrame] = {}
        for name, part in (("train", partitions["train"]),
                           ("val", partitions["val"]),
                           ("test", partitions["test"])):
            transformed[name] = transform_partition(
                frame.loc[part], means, stds, medians, woe_maps,
                global_woe, continuous, categorical,
            )
        # Verify the standardised training statistics are exact
        validate_standardised_statistics(transformed["train"], continuous)

        # Step 3 (C2): the raw-scale sensitive attribute is derived ONCE
        # for the whole population BEFORE any transformation, then sliced
        # per partition by record_id; the rule A = 1{age <= threshold}
        # must never see a z-scored age
        threshold: int = get_binarisation_threshold(config)
        if "raw_age" in cleaned_frame.columns:
            raw_age_full: np.ndarray = pd.to_numeric(
                cleaned_frame["raw_age"], errors="raise"
            ).to_numpy(dtype=np.float64)
        else:
            if "age" not in cleaned_frame.columns:
                raise ValueError(
                    f"dataset '{dataset}': no age source; expected "
                    f"raw_age or age."
                )
            raw_age_full = pd.to_numeric(
                cleaned_frame["age"], errors="raise"
            ).to_numpy(dtype=np.float64)
        a_full: np.ndarray = derive_sensitive_attribute_rawscale(
            raw_age_full, dataset, threshold)
        # The raw-scale provenance map: record_id -> binary attribute
        a_by_record_id: Dict[int, int] = {}
        for rid, a_value in zip(cleaned_frame.index.to_numpy(),
                                a_full.tolist()):
            a_by_record_id[int(rid)] = int(a_value)

        # Step 3: tensors for every partition with provenance; the
        # per-partition a vectors are the raw-scale slices
        tensors: Dict[str, Tuple[np.ndarray, np.ndarray, np.ndarray,
                                 List[str]]] = {}
        for name in ("train", "val", "test"):
            # The partition a vector follows the partition's record ids
            a_partition: np.ndarray = np.asarray(
                [a_by_record_id[int(rid)] for rid in partitions[name]],
                dtype=np.int64,
            )
            tensors[name] = build_design_tensors(
                transformed[name],
                np.asarray(partitions[name], dtype=np.int64),
                dataset,
                a_partition,
                continuous, categorical,
                config["network_architecture"]["structured_component"][
                    "structured_features"],
                identifier_names={"record_id", "y_default", "raw_age",
                                  "record_id_index"},
                indicator_suffix=indicator_suffix,
            )
        X_train, y_train, a_train, feature_names = tensors["train"]
        X_val, y_val, a_val, _ = tensors["val"]
        X_test, y_test, a_test, _ = tensors["test"]
        # The derived population share must match the metadata band; the
        # concatenated RAW-scale A vector spans the whole cleaned
        # population, so the check is meaningful (it is the same vector
        # used by Task 1's metadata statistics)
        validate_sensitive_share(
            np.concatenate([a_train, a_val, a_test]),
            float(datasets_metadata[dataset]["sensitive_share"]),
            dataset=dataset,
        )
        bundles[dataset] = PreprocessedDataBundle(
            dataset=dataset,
            X_train=X_train, X_val=X_val, X_test=X_test,
            y_train=y_train, y_val=y_val, y_test=y_test,
            a_train=a_train, a_val=a_val, a_test=a_test,
            train_ids=partitions["train"], val_ids=partitions["val"],
            test_ids=partitions["test"], feature_names=feature_names,
            stats=stats,
        )
        logger.info(
            f"dataset '{dataset}': p={X_train.shape[1]}, "
            f"n_train={X_train.shape[0]}, n_val={X_val.shape[0]}, "
            f"n_test={X_test.shape[0]}."
        )

    # Persist the frozen statistics for Task 9 reuse
    audit_path: Path = write_preprocessing_artifact(all_stats, artifact_path)
    logger.info(f"Task 3 completed: {len(bundles)} datasets preprocessed; "
                f"artifact at {audit_path.name}.")
    return bundles
````

````python
# Task 4: Prepare and Validate Simulation Data

# ==============================================================================
# Task 4: Prepare and Validate Simulation Data
# ==============================================================================

# Make the sibling task modules importable as the study pipeline grows

# Configure the module-level logger for stage diagnostics

# The project root for the simulation artifact
# The tolerated deviation of the covariate-shift mean difference
COVARIATE_SHIFT_TOLERANCE: float = 0.05
# The tolerance for the true logit and reference risk recomputation
LOGIT_AND_RISK_TOLERANCE: float = 1e-8
# The required DGP constants, from the manuscript and the revised config
INTERCEPT_B0: float = -2.0
BETA_VECTOR: Tuple[float, ...] = (-1.0, -0.5, 0.0, 0.25, 0.5, 1.0)
BASELINE_SHIFT: float = 0.2
# The default population size per (mode, seed) block
DEFAULT_N_OBSERVATIONS: int = 10000
# The default number of covariates
DEFAULT_P_COVARIATES: int = 6

@dataclass(frozen=True)
class SimulationBlockData:
    """
    One (mode, seed) block's preprocessed output.

    Attributes
    ----------
    mode : str
        The generative mode: simple or complex.
    seed : int
        The replication seed.
    X_train, X_val, X_test : np.ndarray
        Design matrices, shape (n_part, p), float64.
    y_train, y_val, y_test : np.ndarray
        Labels, shape (n_part,), int64.
    a_train, a_val, a_test : np.ndarray
        Sensitive attributes, shape (n_part,), int64.
    r_train, r_val, r_test : np.ndarray
        The true reference risks, shape (n_part,), float64.
    train_ids, val_ids, test_ids : List[int]
        Block-local record ids per partition.

    """

    # The generative mode
    mode: str
    # The replication seed
    seed: int
    # Training design matrix
    X_train: np.ndarray
    # Validation design matrix
    X_val: np.ndarray
    # Test design matrix
    X_test: np.ndarray
    # Training labels
    y_train: np.ndarray
    # Validation labels
    y_val: np.ndarray
    # Test labels
    y_test: np.ndarray
    # Training sensitive attributes
    a_train: np.ndarray
    # Validation sensitive attributes
    a_val: np.ndarray
    # Test sensitive attributes
    a_test: np.ndarray
    # Training reference risks
    r_train: np.ndarray
    # Validation reference risks
    r_val: np.ndarray
    # Test reference risks
    r_test: np.ndarray
    # Training record ids
    train_ids: List[int]
    # Validation record ids
    val_ids: List[int]
    # Test record ids
    test_ids: List[int]

@dataclass(frozen=True)
class SimulationDataBundle:
    """
    The full simulation preprocessing output.

    Attributes
    ----------
    blocks : Dict[Tuple[str, int], SimulationBlockData]
        Per-(mode, seed) block containers.
    validated_frame : pd.DataFrame
        The validated simulation frame.

    """

    # Per-block data keyed by the composite tuple
    blocks: Dict[Tuple[str, int], SimulationBlockData]
    # The validated frame with the composite key
    validated_frame: pd.DataFrame

# -------------------------------------------------------------------------------------------------------------------------------
# Task 4, Step 1: Validate the simulation input
# -------------------------------------------------------------------------------------------------------------------------------
def validate_columns_present(
    frame: pd.DataFrame,
    required_columns: List[str],
    block_label: str = "simulation",
) -> None:
    """
    Assert the required columns exist and carry no nulls.

    Parameters
    ----------
    frame : pd.DataFrame
        The frame to inspect.
    required_columns : List[str]
        The mandatory column names.
    block_label : str
        The error-naming label.

    Raises
    ------
    ValueError
        If any required column is missing or contains nulls.

    """
    # Missing columns are a hard contract failure
    missing: List[str] = [
        c for c in required_columns if c not in frame.columns
    ]
    if missing:
        raise ValueError(
            f"{block_label}: required column(s) missing: {missing}."
        )
    # A required column must not be null anywhere: the DGP is complete
    null_counts: Dict[str, int] = {
        c: int(frame[c].isna().sum())
        for c in required_columns if c in frame.columns
    }
    null_total: int = sum(null_counts.values())
    if null_total:
        raise ValueError(
            f"{block_label}: {null_total} null value(s) in required "
            f"columns: {null_counts}."
        )

def validate_simulation_structure(
    frame: pd.DataFrame,
    n_observations: int = DEFAULT_N_OBSERVATIONS,
    p_covariates: int = DEFAULT_P_COVARIATES,
    run_seeds: Optional[List[int]] = None,
    expected_modes: Optional[Sequence[str]] = None,
) -> Dict[Tuple[str, int], int]:
    """
    Validate per-block sizes, covariate count and control domains.

    Parameters
    ----------
    frame : pd.DataFrame
        The simulation input.
    n_observations : int, optional
        The required block size, by default 10000.
    p_covariates : int, optional
        The required covariate count, by default 6.
    run_seeds : Optional[List[int]], optional
        The accepted replication seeds, defaults to 0..9.
    expected_modes : Optional[Sequence[str]], optional
        The configured modes; when supplied, the observed (mode, seed)
        block set must equal the full Cartesian product with the seeds.

    Returns
    -------
    Dict[Tuple[str, int], int]
        Per-block observed sizes.

    Raises
    ------
    ValueError
        If any block size, domain or covariate rule is violated.

    """
    # The accepted seed population must be known for the check
    if run_seeds is None:
        run_seeds = list(range(10))
    validate_columns_present(
        frame, ["mode", "seed", "A", "Y", "r_true", "record_id"]
    )
    # The block keys are enumerated over the whole frame
    blocks: List[Tuple[str, int]] = list(
        zip(frame["mode"].tolist(), frame["seed"].astype(int).tolist())
    )
    # Every block must contain exactly n_observations rows
    sizes: Dict[Tuple[str, int], int] = dict(Counter(blocks))
    for key, count in sizes.items():
        if count != n_observations:
            raise ValueError(
                f"block (mode='{key[0]}', seed={key[1]}) contains "
                f"{count} rows; the DGP requires exactly "
                f"{n_observations}."
            )
    # The observed block set must equal the full Cartesian product of the
    # configured modes and the accepted seeds: a fully missing block is
    # a defect, not a quieter simulation
    expected_modes_set: set = set(expected_modes) if expected_modes else set()
    if expected_modes_set:
        expected_grid: set = {
            (str(mode_val), int(seed_val))
            for mode_val in expected_modes_set
            for seed_val in run_seeds
        }
        observed_grid: set = set(sizes)
        missing_blocks: List[str] = sorted(
            f"{mode_val}|{seed_val}"
            for mode_val, seed_val in (expected_grid - observed_grid)
        )
        unexpected_blocks: List[str] = sorted(
            f"{mode_val}|{seed_val}"
            for mode_val, seed_val in (observed_grid - expected_grid)
        )
        if missing_blocks or unexpected_blocks:
            raise ValueError(
                f"simulation block set does not match the expected grid: "
                f"missing={missing_blocks}, "
                f"unexpected={unexpected_blocks}. The study requires "
                f"every (mode, seed) block."
            )
    # The covariate vector X1_post_shift, X2..X6 is required
    covariate_columns: List[str] = [
        "X1_post_shift"] + [f"X{j}" for j in range(2, p_covariates + 1)
    ]
    validate_columns_present(frame, covariate_columns, "simulation")
    # Mode and seed domains per the study
    valid_modes: Tuple[str, ...] = ("simple", "complex")
    bad_modes: List[str] = sorted(
        set(frame["mode"].tolist()) - set(valid_modes)
    )
    if bad_modes:
        raise ValueError(f"mode must be in {valid_modes}; found {bad_modes}.")
    bad_seeds: List[int] = sorted(
        set(frame["seed"].astype(int).tolist()) - set(run_seeds)
    )
    if bad_seeds:
        raise ValueError(
            f"seed must be in {sorted(run_seeds)}; found {bad_seeds}."
        )
    # A and Y are the binary DGP outputs
    for column in ("A", "Y"):
        bad_binary: List[object] = sorted(
            {v for v in frame[column].tolist() if v not in (0, 1)}
        )
        if bad_binary:
            raise ValueError(
                f"{column} must be binary; found {bad_binary}."
            )
    # The true reference risk is a probability strict inside (0, 1)
    risk_values: np.ndarray = pd.to_numeric(
        frame["r_true"], errors="raise"
    ).to_numpy(dtype=np.float64)
    if not np.all((risk_values > 0.0) & (risk_values < 1.0)):
        n_bad: int = int(((risk_values <= 0.0) | (risk_values >= 1.0)).sum())
        raise ValueError(
            f"r_true must lie strictly in (0, 1); found {n_bad} "
            f"violating row(s)."
        )
    return sizes

def validate_covariate_shift_implication(
    frame: pd.DataFrame,
    tolerance: Optional[float] = None,
    se_multiplier: float = COVARIATE_SHIFT_SE_MULTIPLIER,
) -> Dict[Tuple[str, int], float]:
    """
    Verify the shift implication E[X1|A=0] - E[X1|A=1] ~ 1 per block.

    The data-generating process applies X1 <- X1 + (1 - A) (Section 5.1), so
    the post-shift covariate must separate the groups by exactly one unit in
    expectation. Because only the post-shift column is retained, the shift
    is verified through this implied mean difference.

    B-17 remediation: the previous fixed tolerance of 0.05 is roughly 2.3
    ANALYTIC STANDARD ERRORS at n = 10,000 with a 30 percent group share,
    since X1 has unit variance and

        SE = sqrt(1/n0 + 1/n1) = sqrt(1/7000 + 1/3000) ~ 0.0218.

    A fixed 0.05 therefore rejects about two percent of perfectly
    legitimate Monte Carlo replicates. The acceptance band is now expressed
    as a multiple of that analytic standard error, computed per block from
    the realised group sizes, which is both scale-correct and
    sample-size-aware.

    Parameters
    ----------
    frame : pd.DataFrame
        The simulation input carrying mode, seed, A and X1_post_shift.
    tolerance : Optional[float], optional
        An explicit absolute override. When None (default) the band is
        se_multiplier * sqrt(1/n0 + 1/n1), computed per block.
    se_multiplier : float, optional
        The number of analytic standard errors accepted, by default 4.

    Returns
    -------
    Dict[Tuple[str, int], float]
        Per-block observed mean differences, keyed by (mode, seed).

    Raises
    ------
    ValueError
        If a required column is absent, a block has an empty group, or a
        block's difference deviates beyond the band.

    """
    # The required columns must be present before any grouping
    for required in ("mode", "seed", "A", "X1_post_shift"):
        if required not in frame.columns:
            raise ValueError(
                f"simulation frame must carry '{required}'; present: "
                f"{list(frame.columns)}"
            )
    # A non-positive multiplier would make the band degenerate
    if se_multiplier <= 0.0:
        raise ValueError(
            f"se_multiplier must be positive; received {se_multiplier}."
        )
    differences: Dict[Tuple[str, int], float] = {}
    for (mode, seed), block in frame.groupby(["mode", "seed"]):
        # The post-shift covariate and the group indicator for this block
        x1: np.ndarray = block["X1_post_shift"].to_numpy(dtype=np.float64)
        a_values: np.ndarray = block["A"].to_numpy(dtype=np.int64)
        # Both groups must be populated for the difference to be defined
        n0: int = int((a_values == 0).sum())
        n1: int = int((a_values == 1).sum())
        if n0 == 0 or n1 == 0:
            raise ValueError(
                f"block (mode='{mode}', seed={seed}): group sizes "
                f"n0={n0}, n1={n1}; the mean difference is undefined."
            )
        # The group means of the post-shift covariate
        mean_a0: float = float(x1[a_values == 0].mean())
        mean_a1: float = float(x1[a_values == 1].mean())
        # Equation: E[X1|A=0] - E[X1|A=1] = 1 for the post-shift variable
        difference: float = mean_a0 - mean_a1
        if tolerance is not None:
            # An explicit override is honoured verbatim
            band: float = float(tolerance)
        else:
            # B-17: the analytic standard error of the difference of two
            # means of a unit-variance covariate, then scaled to the band
            standard_error: float = math.sqrt(1.0 / n0 + 1.0 / n1)
            band = se_multiplier * standard_error
        if abs(difference - 1.0) > band:
            raise ValueError(
                f"block (mode='{mode}', seed={seed}): covariate shift "
                f"implication violated; mean difference {difference:.6f} "
                f"deviates from 1.0 by {abs(difference - 1.0):.6f}, beyond "
                f"the {se_multiplier:.1f}-SE band {band:.6f} "
                f"(n0={n0}, n1={n1})."
            )
        differences[(mode, seed)] = difference
    return differences

def derive_composite_key(frame: pd.DataFrame) -> pd.DataFrame:
    """
    Return a copy carrying the validated composite key ordering.

    Parameters
    ----------
    frame : pd.DataFrame
        The validated simulation input.

    Returns
    -------
    pd.DataFrame
        The validated copy with record_id as the first column.

    """
    # The record_id column must precede the covariates per the schema
    result: pd.DataFrame = frame.copy()
    if result.columns[0] != "record_id":
        # Reorder so the key is the leading column, always
        cols: List[str] = list(result.columns)
        cols.remove("record_id")
        result = result[["record_id"] + cols]
    # The composite key must be unique over the whole frame
    composite: pd.Series = pd.Series(list(zip(
        result["mode"], result["seed"].astype(int), result["record_id"]
    )))
    if composite.duplicated().any():
        raise ValueError(
            "the composite key (mode, seed, record_id) is not unique."
        )
    return result

def validate_simulation_input(
    frame: pd.DataFrame,
    n_observations: int = DEFAULT_N_OBSERVATIONS,
    p_covariates: int = DEFAULT_P_COVARIATES,
    run_seeds: Optional[List[int]] = None,
    shift_tolerance: float = COVARIATE_SHIFT_TOLERANCE,
    expected_modes: Optional[Sequence[str]] = None,
) -> Tuple[pd.DataFrame, Dict[Tuple[str, int], int]]:
    """
    Execute the full Step 1 simulation validation.

    Parameters
    ----------
    frame : pd.DataFrame
        The raw simulation input.
    n_observations : int, optional
        Required block size, by default 10000.
    p_covariates : int, optional
        Required covariate count, by default 6.
    run_seeds : Optional[List[int]], optional
        Accepted seeds, defaults to 0..9.
    shift_tolerance : float, optional
        The shift-check tolerance, by default 0.05.

    Returns
    -------
    Tuple[pd.DataFrame, Dict[Tuple[str, int], int]]
        The validated copy and the per-block sizes.

    """
    # Structural rules gate everything else
    sizes: Dict[Tuple[str, int], int] = validate_simulation_structure(
        frame, n_observations, p_covariates, run_seeds, expected_modes
    )
    # The shift implication is verified block by block
    validate_covariate_shift_implication(frame, shift_tolerance)
    # The composite key is derived from the validated copy
    validated: pd.DataFrame = derive_composite_key(frame)
    return validated, sizes

# -------------------------------------------------------------------------------------------------------------------------------
# Task 4, Step 2: Validate the true logit and reference risk
# -------------------------------------------------------------------------------------------------------------------------------
def stable_sigmoid(logits: np.ndarray) -> np.ndarray:
    """
    Compute the numerically stable logistic sigmoid.

    Parameters
    ----------
    logits : np.ndarray
        The logit values, arbitrary magnitude.

    Returns
    -------
    np.ndarray
        The probability values, strictly inside (0, 1) for finite inputs.

    """
    # A nonnegative logit uses 1/(1 + exp(-eta)); a negative logit uses
    # exp(eta)/(1 + exp(eta)), so no term is ever exponentiated with a
    # positive argument
    positive: np.ndarray = logits >= 0
    result: np.ndarray = np.empty_like(logits, dtype=np.float64)
    result[positive] = 1.0 / (1.0 + np.exp(-logits[positive]))
    exp_logits: np.ndarray = np.exp(logits[~positive])
    result[~positive] = exp_logits / (1.0 + exp_logits)
    return result

def compute_true_logit(
    x: np.ndarray, nl_component: np.ndarray, interaction_component: np.ndarray,
    a_values: np.ndarray, b0: float = INTERCEPT_B0,
    beta: Tuple[float, ...] = BETA_VECTOR,
    baseline_shift: float = BASELINE_SHIFT,
) -> np.ndarray:
    """
    Recompute the true DGP logit.

    Equation: eta = L + NL + I + B(A), with L = b0 + beta^T x,
    B(A) = 0.2 * 1{A = 1}.

    Parameters
    ----------
    x : np.ndarray
        The covariate matrix, shape (n, len(beta)).
    nl_component : np.ndarray
        The nonlinear contribution per row.
    interaction_component : np.ndarray
        The interaction contribution per row.
    a_values : np.ndarray
        The sensitive attribute per row.
    b0 : float, optional
        The intercept, by default -2.0.
    beta : Tuple[float, ...], optional
        The coefficient vector, by default the study vector.
    baseline_shift : float, optional
        The group baseline shift, by default 0.2.

    Returns
    -------
    np.ndarray
        The recomputed true logit, shape (n,).

    """
    # Equation: L_i = b0 + sum_j beta_j * x_ij
    linear: np.ndarray = b0 + x @ np.asarray(beta, dtype=np.float64)
    # Equation: B(A_i) = 0.2 * 1{A_i = 1}
    baseline: np.ndarray = baseline_shift * (a_values == 1).astype(np.float64)
    # Equation: eta_i = L_i + NL_i + I_i + B(A_i)
    return linear + np.asarray(nl_component, dtype=np.float64) + \
        np.asarray(interaction_component, dtype=np.float64) + baseline

def verify_true_logit_and_risk(
    frame: pd.DataFrame,
    b0: float = INTERCEPT_B0,
    beta: Tuple[float, ...] = BETA_VECTOR,
    baseline_shift: float = BASELINE_SHIFT,
    tolerance: float = LOGIT_AND_RISK_TOLERANCE,
) -> None:
    """
    Assert the supplied eta_true and r_true match the recomputation.

    Parameters
    ----------
    frame : pd.DataFrame
        The validated simulation input.
    b0 : float, optional
        The DGP intercept, by default -2.0.
    beta : Tuple[float, ...], optional
        The DGP coefficient vector.
    baseline_shift : float, optional
        The baseline shift, by default 0.2.
    tolerance : float, optional
        The comparison tolerance, by default 1e-8.

    Raises
    ------
    ValueError
        If any row deviates, reporting the first offending composite key.

    """
    # The mechanism columns must be present before recomputation
    validate_columns_present(
        frame, ["eta_true", "r_true", "nl_component",
                "interaction_component"]
    )
    # The design matrix follows the manuscript's post-shift ordering
    x_matrix: np.ndarray = frame[
        ["X1_post_shift", "X2", "X3", "X4", "X5", "X6"]
    ].to_numpy(dtype=np.float64)
    nl: np.ndarray = frame["nl_component"].to_numpy(dtype=np.float64)
    inter: np.ndarray = frame["interaction_component"].to_numpy(dtype=np.float64)
    a_values: np.ndarray = frame["A"].to_numpy(dtype=np.int64)
    # Equation: eta_i = L_i + NL_i + I_i + B(A_i)
    eta_recomputed: np.ndarray = compute_true_logit(
        x_matrix, nl, inter, a_values, b0, beta, baseline_shift
    )
    # Equation: r_i = sigma(eta_i), numerically stable
    r_recomputed: np.ndarray = stable_sigmoid(eta_recomputed)
    eta_supplied: np.ndarray = frame["eta_true"].to_numpy(dtype=np.float64)
    r_supplied: np.ndarray = frame["r_true"].to_numpy(dtype=np.float64)
    # Elementwise agreement within the declared tolerance
    eta_bad: np.ndarray = np.abs(eta_recomputed - eta_supplied) > tolerance
    r_bad: np.ndarray = np.abs(r_recomputed - r_supplied) > tolerance
    if eta_bad.any() or r_bad.any():
        # Report the FIRST offending row by its composite key
        first: int = int(np.argmax(eta_bad | r_bad))
        mode_val: object = frame["mode"].iloc[first]
        seed_val: int = int(frame["seed"].iloc[first])
        rid: int = int(frame["record_id"].iloc[first])
        raise ValueError(
            f"true logit/risk mismatch at (mode='{mode_val}', seed="
            f"{seed_val}, record_id={rid}): eta supplied "
            f"{eta_supplied[first]:.12f} vs recomputed "
            f"{eta_recomputed[first]:.12f}; r supplied "
            f"{r_supplied[first]:.12f} vs recomputed "
            f"{r_recomputed[first]:.12f}; tolerance {tolerance}."
        )

# -------------------------------------------------------------------------------------------------------------------------------
# Task 4, Step 3: Split simulation data by mode and seed
# -------------------------------------------------------------------------------------------------------------------------------
def split_simulation_block(
    block: pd.DataFrame,
    mode_val: str,
    seed_val: int,
    split_config: Dict[str, Any],
) -> SimulationBlockData:
    """
    Apply the 70/10/20 stratified split to one (mode, seed) block.

    Parameters
    ----------
    block : pd.DataFrame
        The validated block frame.
    mode_val : str
        The block mode.
    seed_val : int
        The block seed.
    split_config : Dict[str, Any]
        The train/val/test settings with a seed.

    Returns
    -------
    SimulationBlockData
        The block's partitioned tensors.

    """
    # The stratified split runs on the block's REAL record_id column: no
    # synthetic re-indexing is permitted (the composite-key retention
    # rule requires the returned ids to be the frame's record_id values)
    if "record_id" not in block.columns:
        raise ValueError(
            f"block (mode='{mode_val}', seed={seed_val}): the record_id "
            f"column is missing; the split cannot retain the composite "
            f"key."
        )
    # Split the record_id ARRAY with the label stratification aligned by
    # position (both come from the same block frame)
    record_ids: np.ndarray = block["record_id"].to_numpy(
        dtype=np.int64)
    y_values: np.ndarray = block["Y"].to_numpy(dtype=np.int64)
    remainder_ids, test_ids = train_test_split(
        record_ids, stratify=y_values,
        test_size=int(round(float(split_config["test_ratio"])
                            * len(block))),
        random_state=int(split_config["seed"]),
    )
    # The validation count targets the ORIGINAL population
    val_desired: int = int(round(float(split_config["val_ratio"])
                                 * len(block)))
    # The remainder's labels must follow the remainder's ORDER (the first
    # split shuffles), so the stratify vector is built by id lookup
    y_by_id: Dict[int, int] = {
        int(rid): int(y_value)
        for rid, y_value in zip(record_ids.tolist(), y_values.tolist())
    }
    y_remainder: np.ndarray = np.asarray(
        [y_by_id[int(v)] for v in remainder_ids], dtype=np.int64)
    train_ids_np, val_ids_np = train_test_split(
        remainder_ids, stratify=y_remainder,
        test_size=val_desired,
        random_state=int(split_config["seed"]),
    )
    # The partition record ids are the block's real keys
    partitions: Dict[str, List[int]] = {
        "train": [int(v) for v in train_ids_np],
        "val": [int(v) for v in val_ids_np],
        "test": [int(v) for v in test_ids],
    }
    # Disjointness and full coverage are structural requirements
    if not (set(partitions["train"]).isdisjoint(set(partitions["val"])) and
            set(partitions["train"]).isdisjoint(set(partitions["test"])) and
            set(partitions["val"]).isdisjoint(set(partitions["test"]))):
        raise ValueError(
            f"block (mode='{mode_val}', seed={seed_val}): partitions "
            f"overlap."
        )
    if (set(partitions["train"]) | set(partitions["val"]) |
            set(partitions["test"])) != set(record_ids.tolist()):
        raise ValueError(
            f"block (mode='{mode_val}', seed={seed_val}): partitions do "
            f"not cover every record_id."
        )
    # X is the post-shift covariate vector; Z = X by design
    covariate_columns: List[str] = [
        "X1_post_shift", "X2", "X3", "X4", "X5", "X6"
    ]
    component: Dict[str, Dict[str, np.ndarray]] = {}
    ids: Dict[str, List[int]] = {}
    for name in ("train", "val", "test"):
        # Row selection by the record_id values; .loc keeps the sorted
        # split order stable and identical across tensors and ids
        part: pd.DataFrame = block.set_index("record_id").loc[
            partitions[name]].reset_index(drop=True)
        X_part: np.ndarray = part[covariate_columns].to_numpy(dtype=np.float64)
        y_part: np.ndarray = part["Y"].to_numpy(dtype=np.int64)
        a_part: np.ndarray = part["A"].to_numpy(dtype=np.int64)
        r_part: np.ndarray = part["r_true"].to_numpy(dtype=np.float64)
        # Signature identity: rows must match the partition's ids
        if not (X_part.shape[0] == y_part.shape[0] == a_part.shape[0] ==
                r_part.shape[0] == len(partitions[name])):
            raise ValueError(
                f"block (mode='{mode_val}', seed={seed_val}): tensor "
                f"signature mismatch in '{name}'."
            )
        component[name] = {"X": X_part, "y": y_part, "a": a_part,
                           "r": r_part}
        # The partition's default rate is compared to the block rate
        rate: float = float(y_part.mean())
        logger.info(
            f"block ({mode_val}, {seed_val}) '{name}': n="
            f"{len(partitions[name])}, default rate={rate:.6f}."
        )
        ids[name] = partitions[name]
    return SimulationBlockData(
        mode=mode_val, seed=seed_val,
        X_train=component["train"]["X"], X_val=component["val"]["X"],
        X_test=component["test"]["X"],
        y_train=component["train"]["y"], y_val=component["val"]["y"],
        y_test=component["test"]["y"],
        a_train=component["train"]["a"], a_val=component["val"]["a"],
        a_test=component["test"]["a"],
        r_train=component["train"]["r"], r_val=component["val"]["r"],
        r_test=component["test"]["r"],
        train_ids=ids["train"], val_ids=ids["val"], test_ids=ids["test"],
    )

def validate_partition_rates(
    block: SimulationBlockData,
    full_rate: float,
) -> None:
    """
    Compare each partition's default rate with the block rate.

    Parameters
    ----------
    block : SimulationBlockData
        The split block.
    full_rate : float
        The block-wide default rate.

    Raises
    ------
    ValueError
        If any partition deviates by more than 3 percentage points (the
        stratification is expected to keep rates close).

    """
    for name, y_values in (("train", block.y_train), ("val", block.y_val),
                           ("test", block.y_test)):
        rate: float = float(y_values.mean())
        # Stratification should preserve the rare-event rate closely
        if abs(rate - full_rate) > 0.03:
            raise ValueError(
                f"block (mode='{block.mode}', seed={block.seed}): "
                f"'{name}' default rate {rate:.6f} deviates from the "
                f"block rate {full_rate:.6f} by more than 0.03."
            )

# -------------------------------------------------------------------------------------------------------------------------------
# Task 4, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def prepare_simulation_data(
    simulation_frame: pd.DataFrame,
    split_config: Dict[str, Any],
    n_observations: int = DEFAULT_N_OBSERVATIONS,
    p_covariates: int = DEFAULT_P_COVARIATES,
    run_seeds: Optional[List[int]] = None,
    b0: float = INTERCEPT_B0,
    beta: Tuple[float, ...] = BETA_VECTOR,
    baseline_shift: float = BASELINE_SHIFT,
    expected_modes: Optional[Sequence[str]] = None,
) -> SimulationDataBundle:
    """
    Orchestrate the full simulation preparation pipeline.

    Composes Step 1 (structural validation and shift implication), Step 2
    (true logit and reference risk verification) and Step 3 (per-block
    stratified splits with reference-risk propagation).

    Parameters
    ----------
    simulation_frame : pd.DataFrame
        The validated simulation input.
    split_config : Dict[str, Any]
        The train/val/test settings with a seed.
    n_observations : int, optional
        Required block size, by default 10000.
    p_covariates : int, optional
        Required covariate count, by default 6.
    run_seeds : Optional[List[int]], optional
        Accepted seeds, defaults to 0..9.
    b0 : float, optional
        The DGP intercept, by default -2.0.
    beta : Tuple[float, ...], optional
        The DGP coefficients.
    baseline_shift : float, optional
        The baseline shift, by default 0.2.

    Returns
    -------
    SimulationDataBundle
        The per-block bundle and the validated frame.

    """
    logger.info("Starting Task 4: simulation preparation.")
    # Default split settings when the caller provides none
    if split_config is None:
        split_config = {"train_ratio": 0.70, "val_ratio": 0.10,
                        "test_ratio": 0.20, "seed": 42}

    # Step 1: structural validation (with the expected-grid coverage) and
    # the shift implication
    validated, sizes = validate_simulation_input(
        simulation_frame, n_observations, p_covariates, run_seeds,
        COVARIATE_SHIFT_TOLERANCE,
        expected_modes if expected_modes is not None else
        ("simple", "complex"),
    )
    # Step 2: the true logit and the reference risk, before any split
    verify_true_logit_and_risk(
        validated, b0, beta, baseline_shift
    )
    logger.info(
        f"Steps 1-2 passed: {len(sizes)} blocks validated; logit and "
        f"reference risk verified."
    )

    # Step 3: per-block stratified splits
    blocks: Dict[Tuple[str, int], SimulationBlockData] = {}
    for (mode_val, seed_val), block in validated.groupby(
            ["mode", "seed"], sort=True):
        # The groupby key is already the composite prefix
        mode_key: str = str(mode_val)
        seed_key: int = int(seed_val)
        block_data: SimulationBlockData = split_simulation_block(
            block, mode_key, seed_key, split_config
        )
        # The block-wide rate anchors the partition comparison
        full_rate: float = float(block["Y"].mean())
        validate_partition_rates(block_data, full_rate)
        # C7: the reference risk must stay a strict probability per
        # partition (the frontier consumes these exact vectors)
        for name, r_values in (("train", block_data.r_train),
                               ("val", block_data.r_val),
                               ("test", block_data.r_test)):
            if not np.all((r_values > 0.0) & (r_values < 1.0)):
                raise ValueError(
                    f"block (mode='{mode_key}', seed={seed_key}): "
                    f"reference risk in '{name}' leaves (0, 1)."
                )
        blocks[(mode_key, seed_key)] = block_data

    logger.info(f"Task 4 completed: {len(blocks)} blocks prepared.")
    return SimulationDataBundle(blocks=blocks, validated_frame=validated)
````

````python
# Task 5: Implement Model Architectures

# ==============================================================================
# Task 5: Implement Model Architectures
# ==============================================================================

# Configure the module-level logger for stage diagnostics

# The activation registry mapping config strings to torch modules
ACTIVATION_REGISTRY: Dict[str, nn.Module] = {
    "ReLU": nn.ReLU(),
    "GELU": nn.GELU(),
    "tanh": nn.Tanh(),
}
# The supported model classes for the factory
SUPPORTED_MODEL_CLASSES: Tuple[str, ...] = ("LR", "NN", "findr")
# The orthogonality tolerance for the X' H_perp check. The projection
# error of a thin QR under float32 arithmetic scales with the population
# and the machine epsilon (~1e-5 for n ~ 2e2); 1e-4 is the maximal
# float32-justified floor, still orders below any true leakage.
ORTHOGONALITY_TOLERANCE: float = 1e-4
# The full-batch guarantee: findr's identifiability requires the WHOLE
# training population in the forward pass
FULL_BATCH_GUARANTEE: bool = True

def resolve_activation(activation_name: str) -> nn.Module:
    """
    Resolve a config activation string to a torch module.

    Parameters
    ----------
    activation_name : str
        One of the registered activation names.

    Returns
    -------
    nn.Module
        The activation module.

    Raises
    ------
    ValueError
        If the name is not registered.

    """
    # Unknown activation is a configuration defect, never a fallback
    if activation_name not in ACTIVATION_REGISTRY:
        raise ValueError(
            f"unknown activation '{activation_name}'; registered names: "
            f"{list(ACTIVATION_REGISTRY.keys())}."
        )
    # Resolve the mapping
    return ACTIVATION_REGISTRY[activation_name]

def xavier_uniform_initialise(module: nn.Module) -> nn.Module:
    """
    Apply Xavier uniform initialisation to every linear layer, in place.

    Parameters
    ----------
    module : nn.Module
        The module whose linear layers are initialised.

    Returns
    -------
    nn.Module
        The same module, initialised.

    """
    # Every linear layer receives the Xavier uniform distribution
    for layer in module.modules():
        if isinstance(layer, nn.Linear):
            nn.init.xavier_uniform_(layer.weight)
            if layer.bias is not None:
                # Bias zero keeps the pre-activation centred
                nn.init.zeros_(layer.bias)
    return module

def build_encoder(
    input_dim: int,
    hidden_layer_widths: List[int],
    activation: str,
    dropout_rate: float,
) -> nn.Module:
    """
    Build the feed-forward encoder h_alpha from the config dimensions.

    Parameters
    ----------
    input_dim : int
        The input dimension (Z = X).
    hidden_layer_widths : List[int]
        The widths of the hidden layers, any positive length.
    activation : str
        The registered activation name.
    dropout_rate : float
        The dropout rate applied after every hidden activation.

    Returns
    -------
    nn.Module
        The encoder with xavier-initialised linear layers (build returns
        an uninitialised module; the caller applies the common init).

    Raises
    ------
    ValueError
        If the width list is empty or dropout lies outside [0, 1).

    """
    # An empty width list would produce a head-only network, which is
    # not an encoder
    if not hidden_layer_widths:
        raise ValueError("hidden_layer_widths must contain at least one "
                         "entry.")
    # Dropout probability is a probability
    if not 0.0 <= dropout_rate < 1.0:
        raise ValueError(
            f"dropout_rate must be in [0, 1); received {dropout_rate}."
        )
    # The activation module is resolved once
    activation_module: nn.Module = resolve_activation(activation)
    layers: List[nn.Module] = []
    # The first hidden layer maps the input to the first width
    layers.append(nn.Linear(input_dim, hidden_layer_widths[0]))
    layers.append(activation_module)
    layers.append(nn.Dropout(p=dropout_rate))
    # Each subsequent hidden layer maps width to width
    for previous_width, next_width in zip(
            hidden_layer_widths[:-1], hidden_layer_widths[1:]):
        layers.append(nn.Linear(previous_width, next_width))
        layers.append(activation_module)
        layers.append(nn.Dropout(p=dropout_rate))
    # The encoder is the composed sequence; its output dimension is the
    # last width, which the residual layer must match
    return nn.Sequential(*layers)

class LogisticRegression(nn.Module):
    """
    The interpretable linear baseline: eta = beta_0 + x^T beta.

    The layer Linear(p, 1, bias=True) implies the design X_tilde =
    [1_n, X]; the intercept is the bias. The forward returns logits so
    the shared Task 6 loss (BCE with logits plus the fairness penalty on
    the resulting probabilities) applies unchanged.

    """

    def __init__(
        self,
        input_dim: int,
        initialisation: str = "zeros",
    ) -> None:
        """
        Initialise the baseline with its single linear layer.

        Parameters
        ----------
        input_dim : int
            The structured feature count p.
        initialisation : str, optional
            One of "zeros" (standard logistic start) or "xavier_uniform".

        Raises
        ------
        ValueError
            If the initialisation scheme is not supported.

        """
        # Initialise the parent module class
        super().__init__()
        # The single structured layer: eta_str = beta_0 + x^T beta
        self.linear = nn.Linear(input_dim, 1, bias=True)
        # Apply the requested initialisation scheme
        if initialisation == "zeros":
            # The classic logistic baseline starts at the zero vector
            nn.init.zeros_(self.linear.weight)
            nn.init.zeros_(self.linear.bias)
        elif initialisation == "xavier_uniform":
            xavier_uniform_initialise(self.linear)
        else:
            raise ValueError(
                f"initialisation '{initialisation}' is not supported; "
                f"choose 'zeros' or 'xavier_uniform'."
            )

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        """
        Return the logit vector.

        Parameters
        ----------
        x : torch.Tensor
            The design features, shape (n, p).

        Returns
        -------
        torch.Tensor
            The logits, shape (n,).

        """
        # Equation: eta_i = beta_0 + x_i^T beta; squeeze removes the
        # trailing singleton so logits share the label shape (n,)
        return self.linear(x).squeeze(-1)

class NeuralLogistic(nn.Module):
    """
    The standalone neural baseline: eta = d_theta(z), unconstrained.

    The encoder h_alpha is built dynamically from the width list, with
    the configured activation and dropout; the head maps the encoder
    output to a scalar logit. No orthogonalisation is applied.

    """

    def __init__(
        self,
        input_dim: int,
        hidden_layer_widths: List[int],
        activation: str,
        dropout_rate: float,
    ) -> None:
        """
        Initialise the encoder and the scalar logit head.

        Parameters
        ----------
        input_dim : int
            The neural input dimension (Z = X).
        hidden_layer_widths : List[int]
            The encoder hidden widths.
        activation : str
            The registered activation name.
        dropout_rate : float
            The dropout rate.

        """
        # Initialise the parent module class
        super().__init__()
        # Build the encoder from the width list
        self.encoder: nn.Module = build_encoder(
            input_dim, hidden_layer_widths, activation, dropout_rate
        )
        # The head maps the encoder output to a single logit
        self.head = nn.Linear(hidden_layer_widths[-1], 1, bias=True)
        # Uniform initialisation across the encoder and the head
        xavier_uniform_initialise(self)
        # The residual-layer input width is the encoder output width
        self.encoder_output_dim: int = hidden_layer_widths[-1]

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        """
        Return the unconstrained logits.

        Parameters
        ----------
        x : torch.Tensor
            The neural inputs, shape (n, z_dim).

        Returns
        -------
        torch.Tensor
            The logits, shape (n,).

        """
        # Equation: eta = d_theta(z) = head(encoder(z)); no projection
        return self.head(self.encoder(x)).squeeze(-1)

def annihilation_tolerance(
    n_rows: int,
    operand_norm: float,
    dtype: torch.dtype,
    relative_tolerance: float = ORTHOGONALITY_RELATIVE_TOLERANCE,
    floor_multiplier: float = ORTHOGONALITY_FLOOR_MULTIPLIER,
) -> float:
    """
    Compute the scale-stable acceptance bound for the annihilation check.

    The identifiability guarantee of the semi-structured architecture is
    X~^T H_perp = 0 (manuscript Section 3). Numerically the residual of that
    product is a sum of n floating-point round-off terms and therefore grows
    like sqrt(n) AND with the magnitude of the operand. An ABSOLUTE bound is
    consequently a scale bug: the previous fixed 1e-4 rejected well
    conditioned designs at and above roughly twenty thousand training rows,
    blocking findr on five of the eight study datasets.

    Two bounds are combined and the LARGER is returned:

      (i)  relative bound   = relative_tolerance * ||operand||_F
           a dimensionless fraction of the operand's own scale;
      (ii) round-off floor  = floor_multiplier * sqrt(n) * eps(dtype)
                              * ||operand||_F
           the analytic floor of an inner product accumulated over n terms,
           which makes the check self-calibrating in the working precision.

    Parameters
    ----------
    n_rows : int
        The number of accumulated terms n in the inner product.
    operand_norm : float
        The Frobenius norm of the operand the residual is measured against
        (X~ for the constructor check, H_perp for the forward check).
    dtype : torch.dtype
        The dtype the residual was accumulated in; its machine epsilon sets
        the round-off floor.
    relative_tolerance : float, optional
        The dimensionless relative threshold, by default 1e-5.
    floor_multiplier : float, optional
        The safety multiple applied to the analytic floor, by default 16.

    Returns
    -------
    float
        The absolute acceptance bound for the residual norm; always
        strictly positive.

    Raises
    ------
    ValueError
        If n_rows is not positive or the operand norm is not finite.

    """
    # A non-positive row count has no round-off floor and is a caller defect
    if n_rows <= 0:
        raise ValueError(f"n_rows must be positive; received {n_rows}.")
    # A non-finite operand norm makes every bound meaningless
    if not math.isfinite(operand_norm):
        raise ValueError(
            f"operand_norm must be finite; received {operand_norm!r}."
        )
    # The machine epsilon of the working precision drives the analytic floor
    epsilon: float = float(torch.finfo(dtype).eps)
    # A zero-norm operand still needs a strictly positive bound, so the
    # operand scale is floored at one unit before it multiplies the bounds
    scale: float = max(operand_norm, 1.0)
    # (i) the dimensionless relative bound
    relative_bound: float = relative_tolerance * scale
    # (ii) the analytic sqrt(n) * eps round-off floor
    roundoff_floor: float = (
        floor_multiplier * math.sqrt(float(n_rows)) * epsilon * scale
    )
    # The check must never be tighter than the arithmetic itself permits
    return max(relative_bound, roundoff_floor)

class FindrSemiStructured(nn.Module):
    """
    The findr architecture: structured logit plus orthogonal residual.

    Construction (manuscript Section 3). The structured design is
    X~ = [1_n, X] of shape (n, p+1). Its thin QR factorisation X~ = Q R
    yields an orthonormal basis Q of the structured column space and an
    upper-triangular R of order p+1. The projector is P = Q Q^T and its
    complement P_perp = I_n - P. With H the (n, q) encoder output block:

        eta_str   = X~ beta~
        H_perp    = P_perp H
        eta_unstr = H_perp gamma
        eta       = eta_str + eta_unstr
        Y_hat     = sigma(eta)

    The identifiability guarantee is the annihilation property
    X~^T H_perp = 0, which is what makes beta~ uniquely attributable to the
    structured features. The residual layer carries NO intercept because
    gamma_0 is absorbed into beta_0.

    Preserved design decisions (deliberate, and load-bearing):
      * P_perp is NEVER materialised. It is (n, n); at n = 233,154 that is
        ~5.4e10 entries. The projector is applied associatively as
        H - Q (Q^T H), costing O(n q k) instead of O(n^2 q).
      * Q and R are detached buffers. Because P_perp is constant with
        respect to the neural parameters, autograd realises the gradient
        identity dL/dH = P_perp dL/dH_perp with no custom backward.
      * Full column rank is certified from min|diag(R)|, the correct rank
        test for a QR factorisation.

    B-1 remediation, four defects repaired:
      1. The annihilation tolerance is now RELATIVE and self-calibrating
         (see annihilation_tolerance). The previous absolute 1e-4 made the
         constructor raise on every dataset above ~20,000 training rows,
         blocking findr on 94% of the study's observations while the
         orthogonality was in fact perfect.
      2. The out-of-sample projection B is versioned derived state. It was
         previously computed once from the EPOCH-ZERO encoder and cached
         forever, so every validation logit used a projection built from an
         untrained network. B is now invalidated whenever the encoder
         parameters change, and is registered as a buffer when frozen so it
         survives serialisation.
      3. The training path is selected by an EXPLICIT flag, not by
         comparing row counts. An evaluation population that happened to
         have exactly n_train rows previously took the wrong branch.
      4. The full-batch guarantee is no longer a mutable public attribute
         silently switched off by callers. It is controlled by a context
         manager (evaluation_mode) that restores the entry state.

    """

    def __init__(
        self,
        input_dim: int,
        hidden_layer_widths: List[int],
        activation: str,
        dropout_rate: float,
        X_train: torch.Tensor,
        full_batch_guarantee: bool = FULL_BATCH_GUARANTEE,
    ) -> None:
        """
        Initialise the semi-structured module from the training design.

        Parameters
        ----------
        input_dim : int
            The structured feature count p.
        hidden_layer_widths : List[int]
            The encoder hidden widths; the last entry is the residual
            layer's input dimension q.
        activation : str
            The registered activation name.
        dropout_rate : float
            The dropout rate.
        X_train : torch.Tensor
            The TRAINING design matrix, shape (n_train, p). It defines the
            constant Q; this is the fit/apply boundary.
        full_batch_guarantee : bool, optional
            When True (default), the forward refuses populations other than
            the training population, because a mini-batch breaks the
            annihilation property.

        Raises
        ------
        ValueError
            If the training design is empty, misshaped or rank deficient.
        RuntimeError
            If the annihilation property does not hold on construction.

        """
        # Initialise the parent module class
        super().__init__()
        # The design must be a two-dimensional, non-empty tensor
        if X_train.ndim != 2:
            raise ValueError(
                f"X_train must be 2-D (n_train, p); received shape "
                f"{tuple(X_train.shape)}."
            )
        if X_train.shape[0] == 0:
            raise ValueError("X_train must carry at least one row.")
        # The declared feature count must match the supplied design
        if X_train.shape[1] != input_dim:
            raise ValueError(
                f"input_dim ({input_dim}) does not match the training design "
                f"width ({X_train.shape[1]})."
            )
        # The structured linear logit: eta_str = X~ beta~ (the bias carries
        # beta_0, so the ones column is implicit in this layer)
        self.structured_layer = nn.Linear(input_dim, 1, bias=True)
        # Zero-stage the structured layer so the residual starts uncoupled
        nn.init.zeros_(self.structured_layer.weight)
        nn.init.zeros_(self.structured_layer.bias)
        # The neural encoder shares the structured input (Z = X)
        self.encoder: nn.Module = build_encoder(
            input_dim, hidden_layer_widths, activation, dropout_rate
        )
        # The residual layer maps the orthogonalised features to a logit;
        # bias=False because gamma_0 is absorbed into beta_0
        self.residual_layer = nn.Linear(
            hidden_layer_widths[-1], 1, bias=False
        )
        # Xavier uniform initialisation on both neural blocks
        xavier_uniform_initialise(self.encoder)
        xavier_uniform_initialise(self.residual_layer)
        # The encoder output width is a documented contract
        self.encoder_output_dim: int = hidden_layer_widths[-1]
        # B-1(4): the guarantee is a private default; callers express intent
        # through the evaluation_mode context manager, never by assignment
        self._full_batch_guarantee: bool = bool(full_batch_guarantee)
        # The training population size, retained for the guarantee message
        self._n_train: int = int(X_train.shape[0])
        # B-1(1c): the QR is computed in DOUBLE precision regardless of the
        # forward dtype. It is computed once per model, so the cost is
        # negligible, and it removes the dominant source of round-off
        X_train_double: torch.Tensor = X_train.detach().to(QR_COMPUTE_DTYPE)
        # Equation: X~ = [1_n, X]
        ones: torch.Tensor = torch.ones(
            (X_train_double.shape[0], 1),
            dtype=QR_COMPUTE_DTYPE, device=X_train_double.device,
        )
        X_tilde: torch.Tensor = torch.cat([ones, X_train_double], dim=1)
        # A design with fewer rows than columns cannot have full column rank
        if X_tilde.shape[0] < X_tilde.shape[1]:
            raise ValueError(
                f"the structured design has {X_tilde.shape[0]} rows and "
                f"{X_tilde.shape[1]} columns; full column rank requires at "
                f"least as many rows as columns."
            )
        # Thin QR via Householder reflections (torch.linalg.qr, reduced mode)
        q_factor, r_factor = torch.linalg.qr(X_tilde, mode="reduced")
        # The diagonal magnitudes of R certify full column rank. The test is
        # RELATIVE to the largest diagonal, so it is invariant to the
        # overall scaling of the design
        diag_abs: torch.Tensor = torch.abs(torch.diag(r_factor))
        max_diag: float = float(torch.max(diag_abs))
        if max_diag <= 0.0 or float(torch.min(diag_abs)) <= (
                max_diag * float(torch.finfo(QR_COMPUTE_DTYPE).eps)
                * math.sqrt(float(X_tilde.shape[0]))):
            raise ValueError(
                f"the structured design is rank deficient: "
                f"min|diag(R)| = {float(torch.min(diag_abs)):.3e} against "
                f"max|diag(R)| = {max_diag:.3e}; the configuration requires "
                f"full column rank."
            )
        # Q is a constant, buffer-stored orthonormal basis of the span
        self.register_buffer("Q", q_factor.detach().clone())
        # R completes the fixed linear map B = R^{-1} Q^T H used out of sample
        self.register_buffer("R", r_factor.detach().clone())
        # The training design is retained so the frozen evaluation
        # projection B can be recomputed from the fitted encoder
        self.register_buffer("X_train_design", X_train.detach().clone())
        # B-1(2): B is DERIVED state carrying an explicit validity token.
        # The buffer starts empty; _B_version records which parameter
        # generation produced the cached value
        self.register_buffer(
            "B", torch.zeros(
                (input_dim + 1, hidden_layer_widths[-1]),
                dtype=QR_COMPUTE_DTYPE,
            )
        )
        # The generation counter of the encoder parameters. Any parameter
        # change bumps _param_version; B is stale whenever the two differ
        self._param_version: int = 0
        self._B_version: int = -1
        # The frozen flag records that B has been sealed from the FINAL
        # encoder and may be trusted without a version check
        self._B_frozen: bool = False
        # Verify the annihilation property on the construction design itself
        self._verify_annihilation(X_tilde)

    @property
    def full_batch_guarantee(self) -> bool:
        """
        Report whether the full-batch identifiability guard is active.

        B-1(4): exposed READ-ONLY. The previous mutable public attribute was
        set to False by three separate call sites and restored by none, so
        the guard was dead from the first validation evaluation onward. Use
        the evaluation_mode context manager to suspend it.

        Returns
        -------
        bool
            True when the forward refuses non-training populations.

        """
        # The private flag is the single source of truth
        return self._full_batch_guarantee

    @contextlib.contextmanager
    def evaluation_mode(self) -> Any:
        """
        Temporarily suspend the full-batch guard and restore it on exit.

        B-1(4): every out-of-sample scoring path must declare its intent
        through this context manager. The entry state of BOTH the guard and
        the module's train/eval mode is captured and restored in a finally
        block, so an exception inside the block cannot leak altered state.

        Yields
        ------
        FindrSemiStructured
            This module, with the guard suspended and eval mode set.

        """
        # Capture the entry state of the guard and of the training mode
        previous_guard: bool = self._full_batch_guarantee
        previous_training: bool = self.training
        try:
            # Suspend the guard: this call is explicitly out of sample
            self._full_batch_guarantee = False
            # Deterministic scoring requires dropout disabled
            self.eval()
            yield self
        finally:
            # Restore BOTH captured states, exception or not
            self._full_batch_guarantee = previous_guard
            self.train(previous_training)

    def bump_parameter_version(self) -> None:
        """
        Invalidate the cached evaluation projection after a parameter update.

        B-1(2): B = R^{-1} Q^T H_train depends on the ENCODER, so it becomes
        stale the moment the encoder learns. The optimiser step (or any
        caller mutating parameters) must call this so the next
        out-of-sample forward recomputes B.

        Returns
        -------
        None

        """
        # A monotone counter is the cheapest correct staleness token
        self._param_version += 1
        # A sealed B is invalidated by any subsequent parameter change
        self._B_frozen = False

    def _verify_annihilation(self, X_tilde: torch.Tensor) -> None:
        """
        Assert X~ - Q Q^T X~ = 0 on the given design rows.

        The structured design lies wholly inside its own column space, so
        projecting it must return it unchanged. A violation means Q does not
        span the structured space and every downstream attribution is void.

        B-1(1): the bound is now relative and self-calibrating.

        Parameters
        ----------
        X_tilde : torch.Tensor
            The augmented design whose span the projector must reproduce.

        Raises
        ------
        RuntimeError
            If the annihilator does not hold within the scale-stable bound.

        """
        # Work in the QR precision so the check matches the factorisation
        design: torch.Tensor = X_tilde.to(QR_COMPUTE_DTYPE)
        # P_perp X~ = X~ - Q (Q^T X~), applied associatively
        residual: torch.Tensor = design - self.Q @ (
            self.Q.transpose(0, 1) @ design
        )
        # The residual and the operand norms drive the relative bound
        norm_residual: float = float(residual.norm())
        norm_design: float = float(design.norm())
        # B-1(1a,1b): the scale-stable acceptance bound
        bound: float = annihilation_tolerance(
            n_rows=int(design.shape[0]), operand_norm=norm_design,
            dtype=QR_COMPUTE_DTYPE,
        )
        if norm_residual > bound:
            raise RuntimeError(
                f"annihilation violation: ||X~ - Q Q^T X~||_F = "
                f"{norm_residual:.3e} exceeds the scale-stable bound "
                f"{bound:.3e} (n={int(design.shape[0])}, "
                f"||X~||_F={norm_design:.3e}); the QR basis does not span "
                f"the structured column space."
            )

    def verify_orthogonality(self, H_perp: torch.Tensor) -> float:
        """
        Certify ||X~^T H_perp||_F on the training population.

        Because X~ = Q R, the annihilation product factorises as
        X~^T H_perp = R^T (Q^T H_perp); certifying Q^T H_perp therefore
        certifies the property itself, at cost O(n q k) rather than O(n^2).

        B-1(1): the bound is relative to ||H_perp||_F, so it is stable
        across n and across the encoder's output scale.

        Parameters
        ----------
        H_perp : torch.Tensor
            The orthogonalised encoder output, shape (n_train, q).

        Returns
        -------
        float
            The realised ||Q^T H_perp||_F, guaranteed below the bound when
            the call returns.

        Raises
        ------
        ValueError
            If the population does not match the training population.
        RuntimeError
            If the orthogonality error exceeds the scale-stable bound.

        """
        # The certificate is defined over the training rows only
        rows: int = int(self.Q.shape[0])
        if H_perp.shape[0] != rows:
            raise ValueError(
                f"orthogonality verification requires the training "
                f"population ({rows} rows); received {H_perp.shape[0]}."
            )
        # Verification is not a gradient path: detach and use QR precision
        residual_block: torch.Tensor = H_perp.detach().to(QR_COMPUTE_DTYPE)
        # Q^T H_perp must vanish; this is the annihilation certificate
        q_orth: torch.Tensor = self.Q.transpose(0, 1) @ residual_block
        error: float = float(q_orth.norm())
        # The bound scales with the operand H_perp, not with a fixed number
        bound: float = annihilation_tolerance(
            n_rows=rows, operand_norm=float(residual_block.norm()),
            dtype=QR_COMPUTE_DTYPE,
        )
        if error > bound:
            raise RuntimeError(
                f"orthogonality violation: ||Q^T H_perp||_F = {error:.3e} "
                f"exceeds the scale-stable bound {bound:.3e}; the "
                f"annihilation property X~^T H_perp = 0 does not hold."
            )
        return error

    def forward(
        self, x: torch.Tensor, is_training_population: Optional[bool] = None,
    ) -> torch.Tensor:
        """
        Return the full logits eta = eta_str + eta_unstr.

        Parameters
        ----------
        x : torch.Tensor
            The structured/neural inputs, shape (n, p).
        is_training_population : Optional[bool], optional
            B-1(3): the EXPLICIT declaration of which population x is. When
            None the module falls back to a row-count comparison for
            backward compatibility, which is exactly the inference the
            remediation removes; callers should always pass it.

        Returns
        -------
        torch.Tensor
            The full logits, shape (n,).

        Raises
        ------
        ValueError
            If the full-batch guarantee is violated.

        """
        # Resolve the population identity before anything else depends on it
        on_training: bool = self._resolve_population(x, is_training_population)
        # The full-batch guarantee: mini-batches break identifiability
        if self._full_batch_guarantee and not on_training:
            raise ValueError(
                f"findr requires full-batch training: expected the training "
                f"population of {self._n_train} rows, received "
                f"{x.shape[0]}. Training under mini-batching breaks "
                f"X~^T H_perp = 0. Use the evaluation_mode() context "
                f"manager for out-of-sample scoring."
            )
        # Decompose the forward into its structured and residual halves
        eta_str, eta_unstr, _ = self.forward_decomposed(
            x, is_training_population=on_training
        )
        return eta_str + eta_unstr

    def _resolve_population(
        self, x: torch.Tensor, declared: Optional[bool],
    ) -> bool:
        """
        Determine whether x is the training population.

        B-1(3): semantics are never inferred from a shape when the caller
        has declared them. A declared True is additionally VERIFIED against
        the stored row count, so a mistaken declaration fails loudly rather
        than silently selecting the wrong projection.

        Parameters
        ----------
        x : torch.Tensor
            The incoming design block.
        declared : Optional[bool]
            The caller's declaration, or None to fall back to inference.

        Returns
        -------
        bool
            True when the in-sample projection path must run.

        Raises
        ------
        ValueError
            If a declared training population has the wrong row count.

        """
        # An explicit declaration is authoritative, but still checked
        if declared is not None:
            if declared and x.shape[0] != self._n_train:
                raise ValueError(
                    f"is_training_population=True was declared but the block "
                    f"carries {x.shape[0]} rows against a training "
                    f"population of {self._n_train}."
                )
            return bool(declared)
        # Legacy fallback: infer from the row count and warn, because this
        # is precisely the ambiguous inference the remediation removes
        inferred: bool = x.shape[0] == self._n_train
        if inferred:
            logger.debug(
                "findr.forward: population identity inferred from the row "
                "count; pass is_training_population explicitly."
            )
        return inferred

    def forward_decomposed(
        self,
        x: torch.Tensor,
        is_training_population: Optional[bool] = None,
    ) -> Tuple[torch.Tensor, torch.Tensor, torch.Tensor]:
        """
        Return structured logits, unstructured logits and H_perp.

        On the TRAINING population the orthogonalised decomposition runs
        directly: H_perp = H - Q (Q^T H). On any other population the same
        FIXED linear operator is applied per row as H - X~ B, with
        B = R^{-1} Q^T H_train evaluated on the current encoder. On the
        training rows X~ B equals Q Q^T H exactly, so the fitted and the
        deployed maps coincide.

        Parameters
        ----------
        x : torch.Tensor
            The inputs, shape (n, p).
        is_training_population : Optional[bool], optional
            The explicit population declaration; see forward.

        Returns
        -------
        Tuple[torch.Tensor, torch.Tensor, torch.Tensor]
            eta_str (n,), eta_unstr (n,), and H_perp (n, q); the third
            component supports the Task 9 diagnostics.

        Raises
        ------
        RuntimeError
            If the runtime annihilation certificate fails on the training
            population.

        """
        # Resolve the population identity explicitly
        on_training: bool = self._resolve_population(x, is_training_population)
        # Structured logit: eta_str = X~ beta~, mapping (n, p) -> (n,)
        eta_str: torch.Tensor = self.structured_layer(x).squeeze(-1)
        # The encoder maps Z = X to the q-width feature block
        H: torch.Tensor = self.encoder(x)
        if on_training:
            # Equation: H_perp = H - Q (Q^T H), the associative projector.
            # Q is cast to the operand dtype so a float32 forward is not
            # silently promoted to float64 across the whole graph
            q_basis: torch.Tensor = self.Q.to(H.dtype)
            H_perp: torch.Tensor = H - q_basis @ (q_basis.transpose(0, 1) @ H)
            # The runtime certificate: X~^T H_perp = R^T (Q^T H_perp) must
            # vanish at EVERY forward, not only at construction
            certificate: torch.Tensor = (
                self.R.to(H.dtype).transpose(0, 1)
                @ (q_basis.transpose(0, 1) @ H_perp)
            ).detach()
            # B-1(1): the bound scales with H_perp and with sqrt(n)
            bound: float = annihilation_tolerance(
                n_rows=int(H_perp.shape[0]),
                operand_norm=float(H_perp.detach().norm()),
                dtype=H.dtype,
            )
            realised: float = float(certificate.norm())
            if realised > bound:
                raise RuntimeError(
                    f"annihilation violation at forward: "
                    f"||X~^T H_perp||_F = {realised:.3e} exceeds the "
                    f"scale-stable bound {bound:.3e}; the fitted residual is "
                    f"no longer orthogonal to the structured span."
                )
        else:
            # The out-of-sample path applies the SAME fixed linear operator
            # per row. B is derived state and must be current
            projection: torch.Tensor = self._evaluation_projection()
            ones: torch.Tensor = torch.ones(
                (x.shape[0], 1), dtype=x.dtype, device=x.device)
            x_tilde: torch.Tensor = torch.cat([ones, x], dim=1)
            H_perp = H - x_tilde @ projection.to(H.dtype)
        # Equation: eta_unstr = H_perp gamma (no intercept: gamma_0 is
        # absorbed into beta_0)
        eta_unstr: torch.Tensor = self.residual_layer(H_perp).squeeze(-1)
        return eta_str, eta_unstr, H_perp

    def _evaluation_projection(self) -> torch.Tensor:
        """
        Return the CURRENT frozen per-row projection B.

        B-1(2): the staleness repair. B = R^{-1} Q^T H_train depends on the
        encoder, so it is recomputed whenever the parameter generation that
        produced the cached value differs from the current generation. The
        previous implementation computed B once, lazily, on the first
        out-of-sample forward — which the per-epoch validation call made
        epoch zero — and never invalidated it, so every validation logit
        used a projection built from an essentially untrained network.

        Returns
        -------
        torch.Tensor
            The projection matrix, shape (p + 1, q), in QR precision.

        """
        # A sealed projection is trusted without recomputation
        if self._B_frozen and self._B_version == self._param_version:
            return self.B
        # A cached projection is valid only for its own parameter generation
        if self._B_version != self._param_version:
            self.B = self._compute_evaluation_projection()
            self._B_version = self._param_version
        return self.B

    def _compute_evaluation_projection(self) -> torch.Tensor:
        """
        Compute B = R^{-1} Q^T H_train from the CURRENT encoder.

        Evaluated under no_grad on the retained training design. With this B
        the out-of-sample residual map h - x~ B is the same linear operator
        the training path applies on the training rows, where X~ B equals
        Q Q^T H exactly.

        Methodological note (registered decision): the manuscript defines
        the projection only ON THE SAMPLE and is silent on how a fitted
        model scores new rows. This fixed-operator extrapolation is an
        engineering resolution of that gap, not a manuscript prescription,
        and is carried in the placeholder audit as such.

        Returns
        -------
        torch.Tensor
            The fixed projection matrix, shape (p + 1, q).

        """
        # The encoder output on the retained training design, frozen. Eval
        # mode is forced so dropout cannot randomise the projection
        was_training: bool = self.training
        try:
            self.eval()
            with torch.no_grad():
                H_train: torch.Tensor = self.encoder(self.X_train_design)
        finally:
            # Restore the entry training mode unconditionally
            self.train(was_training)
        # Work in QR precision: B is a fixed operator and must be exact
        H_double: torch.Tensor = H_train.detach().to(QR_COMPUTE_DTYPE)
        # Equation: Q^T H_train, of shape (p+1, q)
        q_orth: torch.Tensor = self.Q.transpose(0, 1) @ H_double
        # Equation: B = R^{-1} (Q^T H_train) via a triangular solve, which is
        # numerically superior to forming R^{-1} explicitly
        return torch.linalg.solve_triangular(self.R, q_orth, upper=True)

    def freeze_evaluation_projection(self) -> torch.Tensor:
        """
        Seal B from the FINAL encoder so it survives serialisation.

        B-1(2): B was previously a plain Python attribute, absent from
        state_dict(); a reloaded model recomputed it from the restored
        encoder and could return different predictions from the model that
        was saved. B is a registered buffer here, and sealing it before
        serialisation makes the saved artifact self-contained.

        Returns
        -------
        torch.Tensor
            The sealed projection, shape (p + 1, q).

        """
        # Recompute from the current (final) encoder and seal it
        self.B = self._compute_evaluation_projection()
        # The seal binds the projection to the current parameter generation
        self._B_version = self._param_version
        self._B_frozen = True
        return self.B

def build_model(
    model_class: str,
    input_dim: int,
    hidden_layer_widths: List[int],
    activation: str,
    dropout_rate: float,
    X_train: Optional[torch.Tensor] = None,
    device: Optional[torch.device] = None,
) -> nn.Module:
    """
    Build one of the three study models from the architecture config.

    Parameters
    ----------
    model_class : str
        One of "LR", "NN", "findr".
    input_dim : int
        The structured/neural input dimension.
    hidden_layer_widths : List[int]
        The encoder widths (network classes only).
    activation : str
        The activation name (network classes only).
    dropout_rate : float
        The dropout rate (network classes only).
    X_train : Optional[torch.Tensor], optional
        The training design for findr's thin QR; required for findr.
    device : Optional[torch.device], optional
        The target device for the model.

    Returns
    -------
    nn.Module
        The constructed model on the target device.

    Raises
    ------
    ValueError
        If the model class is unknown or findr is built without X_train.

    """
    # Unknown classes are configuration defects
    if model_class not in SUPPORTED_MODEL_CLASSES:
        raise ValueError(
            f"model_class must be one of {SUPPORTED_MODEL_CLASSES}; "
            f"received '{model_class}'."
        )
    # findr without the training design cannot define its projector
    if model_class == "findr" and X_train is None:
        raise ValueError(
            "findr requires X_train (the training design) to compute "
            "the thin QR and the constant projector Q."
        )
    # C14: the single entry point is defensive about the dimensions and
    # the architecture domains
    if not isinstance(input_dim, int) or input_dim < 1:
        raise ValueError(
            f"input_dim must be a positive integer; received {input_dim!r}."
        )
    for width in hidden_layer_widths:
        if not isinstance(width, int) or width < 1:
            raise ValueError(
                f"hidden_layer_widths entries must be positive integers; "
                f"found {width!r}."
            )
    if activation not in ACTIVATION_REGISTRY:
        raise ValueError(
            f"activation must be one of {list(ACTIVATION_REGISTRY.keys())}; "
            f"received '{activation}'."
        )
    if not 0.0 <= dropout_rate < 1.0:
        raise ValueError(
            f"dropout_rate must be in [0, 1); received {dropout_rate}."
        )
    # findr: the input dimension must agree with the training design that
    # defines its projector, stated before the QR so the error is
    # actionable rather than a downstream shape failure
    if model_class == "findr" and X_train is not None:
        if X_train.shape[1] != input_dim:
            raise ValueError(
                f"findr input_dim ({input_dim}) does not match "
                f"X_train.shape[1] ({X_train.shape[1]}); the training "
                f"design defines the projector and must agree."
            )
    # LR: the linear baseline
    if model_class == "LR":
        model: nn.Module = LogisticRegression(input_dim)
    # NN: the standalone encoder
    elif model_class == "NN":
        model = NeuralLogistic(input_dim, hidden_layer_widths, activation,
                               dropout_rate)
    # findr: the semi-structured architecture
    else:
        model = FindrSemiStructured(
            input_dim, hidden_layer_widths, activation, dropout_rate,
            X_train,
        )
    # Move the built model to the requested device when one is given
    if device is not None:
        model = model.to(device)
    return model
````

````python
# Task 6: Implement the Loss Function and Fairness Penalty

# ==============================================================================
# Task 6: Implement the Loss Function and Fairness Penalty
# ==============================================================================

# Configure the module-level logger for stage diagnostics

# The supported fairness criteria
SUPPORTED_CRITERIA: Tuple[str, ...] = (
    "equalised_odds", "equal_opportunity", "independence",
)
# The weight-normalisation tolerance
WEIGHT_SUM_TOLERANCE: float = 1e-6

def bcce_with_logits(
    logits: torch.Tensor, y_true: torch.Tensor,
) -> torch.Tensor:
    """
    Binary cross-entropy with logits, numerically stable form.

    Equation: L_pred = (1/n) sum_i [ log(1 + exp(eta_i)) - y_i eta_i ].
    The log-sum-exp form is computed with torch.logaddexp so no sigmoid
    is evaluated before the logarithm, per the task instruction.

    Parameters
    ----------
    logits : torch.Tensor
        The model logits, shape (n,).
    y_true : torch.Tensor
        The binary labels, shape (n,).

    Returns
    -------
    torch.Tensor
        The scalar averaged loss.

    Raises
    ------
    ValueError
        If the tensors are empty, misaligned or carry non-finite values.

    """
    # Both tensors must be non-empty row-vectors of equal length
    if logits.ndim != 1 or y_true.ndim != 1:
        raise ValueError(
            f"logits and y_true must be 1-D; got logits {logits.ndim}D "
            f"and y_true {y_true.ndim}D."
        )
    if logits.shape[0] != y_true.shape[0]:
        raise ValueError(
            f"shape mismatch: logits {logits.shape[0]} vs labels "
            f"{y_true.shape[0]}."
        )
    if logits.shape[0] == 0:
        raise ValueError("logits must not be empty.")
    # Non-finite logits would propagate NaN through the whole objective
    if not torch.isfinite(logits).all():
        raise ValueError("logits carry non-finite values.")
    if not torch.isfinite(y_true).all():
        raise ValueError("labels carry non-finite values.")
    # Equation term 1: log(1 + exp(eta_i)) via logaddexp(0, eta_i)
    term1: torch.Tensor = torch.logaddexp(
        torch.zeros_like(logits), logits
    )
    # Equation term 2: - y_i * eta_i
    term2: torch.Tensor = y_true * logits
    # Equation: the average over the population. The float64 promotion
    # happens BEFORE the mean so the accumulation itself is double
    # precision regardless of the input dtype
    return (term1 - term2).to(torch.float64).mean()

def validate_weights(weights: torch.Tensor, size: int, label: str) -> None:
    """
    Validate the distribution weights of one support.

    Parameters
    ----------
    weights : torch.Tensor
        The weight vector.
    size : int
        The support size.
    label : str
        The support label, for error naming.

    Raises
    ------
    ValueError
        If weights misalign with the support or do not sum to 1.

    """
    if weights.shape[0] != size:
        raise ValueError(
            f"{label}: weights length {weights.shape[0]} does not match "
            f"support size {size}."
        )
    if not (weights >= 0.0).all():
        raise ValueError(f"{label}: weights must be non-negative.")
    # C5: non-finite weights would silently corrupt the cumulative sums
    if not torch.isfinite(weights).all():
        raise ValueError(f"{label}: weights carry non-finite values.")
    if (weights.sum() - 1.0).abs() > WEIGHT_SUM_TOLERANCE:
        raise ValueError(
            f"{label}: weights must sum to 1.0; found {float(weights.sum())}."
        )

def wasserstein_1(
    scores_s: torch.Tensor,
    scores_t: torch.Tensor,
    weights_s: Optional[torch.Tensor] = None,
    weights_t: Optional[torch.Tensor] = None,
) -> torch.Tensor:
    """
    Exact 1-Wasserstein distance between two empirical distributions.

    Equation (5) of the manuscript with p = 1:

        W1(mu, nu) = sum_{k=1}^{K} (q_k - q_{k-1})
            |Fmu^{-1}(q_{k-1}) - Fnu^{-1}(q_{k-1})|

    where the q grid is the merged set of cumulative breakpoints of both
    distributions, and the empirical quantile is left-continuous (the
    first support index whose cumulative weight reaches q).

    Parameters
    ----------
    scores_s : torch.Tensor
        The first support, shape (n,), any finite order.
    scores_t : torch.Tensor
        The second support, shape (m,).
    weights_s : Optional[torch.Tensor], optional
        The first distribution weights, shape (n,), summing to 1; the
        uniform weights are used when None.
    weights_t : Optional[torch.Tensor], optional
        The second distribution weights, shape (m,), summing to 1.

    Returns
    -------
    torch.Tensor
        The scalar distance. Differentiable almost everywhere with
        respect to the supports.

    Raises
    ------
    ValueError
        If a support is empty or the weights violate the simplex.

    Notes
    -----
    The convention (a registered methodological decision, see the Task 1
    placeholder/decision audit): Equation (5) of the manuscript writes
    sum_k (q_k - q_{k-1}) |F_mu^{-1}(q_{k-1}) - F_nu^{-1}(q_{k-1})|, and
    its own quantile definition ^F^{-1}(q) = s_{(i)} with
    i = min{r: Omega_r >= q} is LEFT-CONTINUOUS. With that definition the
    evaluation at the literal left endpoint q_{k-1} of a strip returns
    the PREVIOUS strip's measure-zero jump value, i.e. a left-Riemann
    approximation that is NOT the exact distance (measured ~3% deviation
    against scipy). The quantile function is constant almost everywhere
    on each strip (q_{k-1}, q_k], so the EXACT integral is obtained by
    evaluating the left-continuous quantile at the strip's RIGHT
    endpoint q_k. This implementation computes the exact value; it is
    verified against scipy.stats.wasserstein_distance to 1e-6 including
    weighted and tie-heavy cases.

    """
    # Empty supports have no empirical distribution
    if scores_s.shape[0] == 0 or scores_t.shape[0] == 0:
        raise ValueError(
            f"empty support: n_s={scores_s.shape[0]}, n_t={scores_t.shape[0]}."
        )
    if scores_s.ndim != 1 or scores_t.ndim != 1:
        raise ValueError("score supports must be 1-D tensors.")
    # C5: explicit finiteness validation: NaN scores would propagate
    # through argsort/searchsorted unchecked
    if not torch.isfinite(scores_s).all():
        raise ValueError(
            f"scores_s carries non-finite values "
            f"(n={int((~torch.isfinite(scores_s)).sum())})."
        )
    if not torch.isfinite(scores_t).all():
        raise ValueError(
            f"scores_t carries non-finite values "
            f"(n={int((~torch.isfinite(scores_t)).sum())})."
        )
    # Double precision for the whole path: the quantile differences are
    # of order the score scale, and single-precision accumulation over
    # thousands of grid points produces errors at the 1e-5 level that
    # the scipy oracle comparison must not see
    scores_s = scores_s.double()
    scores_t = scores_t.double()
    if weights_s is None:
        # Uniform weights are the default empirical weights
        weights_s = torch.full_like(
            scores_s, 1.0 / scores_s.shape[0], dtype=torch.float64
        )
    else:
        weights_s = weights_s.double()
    if weights_t is None:
        weights_t = torch.full_like(
            scores_t, 1.0 / scores_t.shape[0], dtype=torch.float64
        )
    else:
        weights_t = weights_t.double()
    # The weight contracts are validated before the sort
    validate_weights(weights_s, scores_s.shape[0], "weights_s")
    validate_weights(weights_t, scores_t.shape[0], "weights_t")
    # Sort the supports; the sort is differentiable and the index map
    # sorts the weights together with the values
    idx_s: torch.Tensor = torch.argsort(scores_s)
    s_sorted: torch.Tensor = scores_s[idx_s]
    w_s_sorted: torch.Tensor = weights_s[idx_s]
    idx_t: torch.Tensor = torch.argsort(scores_t)
    t_sorted: torch.Tensor = scores_t[idx_t]
    w_t_sorted: torch.Tensor = weights_t[idx_t]
    # Equation: Omega_i = sum_{r <= i} w_r  (cumulative weights)
    omega: torch.Tensor = torch.cumsum(w_s_sorted, dim=0)
    upsilon: torch.Tensor = torch.cumsum(w_t_sorted, dim=0)
    # The merged breakpoint grid: the union of cumulative breakpoints,
    # de-duplicated and sorted, with 0 prepended and 1 as the end point
    breakpoints: torch.Tensor = torch.cat([omega, upsilon])
    breakpoints = torch.unique(breakpoints, sorted=True)
    breakpoints = torch.cat([
        torch.zeros(1, device=breakpoints.device, dtype=breakpoints.dtype),
        breakpoints,
    ])
    # The grid must end exactly at 1 (both cumulatives finish at 1)
    if abs(float(breakpoints[-1]) - 1.0) > WEIGHT_SUM_TOLERANCE:
        raise ValueError(
            f"breakpoint grid must end at 1.0; found "
            f"{float(breakpoints[-1])}."
        )
    # Equation strip masses: (q_k - q_{k-1}) for k = 1..K
    masses: torch.Tensor = breakpoints[1:] - breakpoints[:-1]
    # The quantile function is constant on each open strip
    # (q_{k-1}, q_k], so its value on a strip is the value AT the right
    # endpoint q_k under the left-continuous convention, i.e.
    # i = min{r : Omega_r >= q_k}. Evaluating the left endpoint instead
    # returns the value of the PREVIOUS strip whenever a strip ends on an
    # Omega (or Upsilon) breakpoint, which desynchronises the strips by
    # one quantum and mis-states the integral (measured at 3% against
    # the scipy oracle)
    q_right: torch.Tensor = breakpoints[1:]
    # Left-continuous empirical quantile at the strip's right endpoint
    idx_quant_s: torch.Tensor = torch.searchsorted(omega, q_right, right=False)
    idx_quant_t: torch.Tensor = torch.searchsorted(upsilon, q_right, right=False)
    # The cumulative sum may round the final mass to just below 1.0, in
    # which case searchsorted returns the insertion position n for q=1;
    # at q=1 the quantile is the maximum support, so the clamp is exact
    idx_quant_s = torch.clamp(idx_quant_s, max=omega.shape[0] - 1)
    idx_quant_t = torch.clamp(idx_quant_t, max=upsilon.shape[0] - 1)
    # The quantile values are gathered from the sorted supports
    f_mu: torch.Tensor = s_sorted[idx_quant_s]
    f_nu: torch.Tensor = t_sorted[idx_quant_t]
    # Equation: W1 = sum_k (q_k - q_{k-1}) * |Fmu^{-1}(q_k) - Fnu^{-1}(q_k)|
    # evaluated on the strip toward the right endpoint, which is the
    # exact discrete form of the integral of the absolute inverse-CDF gap
    distance: torch.Tensor = (masses * (f_mu - f_nu).abs()).sum()
    return distance.to(torch.float64)

def fairness_penalty(
    scores: torch.Tensor,
    y_true: torch.Tensor,
    a_true: torch.Tensor,
    criterion: str = "equalised_odds",
    empty_cell_policy: str = "raise",
) -> torch.Tensor:
    """
    The score-distribution fairness penalty for the chosen criterion.

    Equation (6): equalised odds sums the 1-Wasserstein distances between
    the score distributions of the two sensitive groups within each
    realised outcome; equal opportunity keeps only the y = 0 cell;
    independence compares the overall group distributions.

    Parameters
    ----------
    scores : torch.Tensor
        The predicted probability scores, shape (n,).
    y_true : torch.Tensor
        The realised outcomes, shape (n,).
    a_true : torch.Tensor
        The sensitive attribute, shape (n,).
    criterion : str, optional
        One of equalised_odds, equal_opportunity, independence.

    Returns
    -------
    torch.Tensor
        The scalar penalty.

    Raises
    ------
    ValueError
        If the criterion is unknown, the arrays misalign, or a required
        cell is empty.

    """
    # The empty-cell policy domain is fixed
    if empty_cell_policy not in ("raise", "return_zero"):
        raise ValueError(
            f"empty_cell_policy must be 'raise' or 'return_zero'; "
            f"received '{empty_cell_policy}'."
        )
    # The criterion domain is fixed by the study
    if criterion not in SUPPORTED_CRITERIA:
        raise ValueError(
            f"criterion must be one of {SUPPORTED_CRITERIA}; received "
            f"'{criterion}'."
        )
    # The three arrays must be the same population, in the same order
    if not (scores.ndim == y_true.ndim == a_true.ndim == 1):
        raise ValueError("scores, y_true and a_true must all be 1-D.")
    if not (scores.shape[0] == y_true.shape[0] == a_true.shape[0]):
        raise ValueError(
            f"shape mismatch: scores {scores.shape[0]}, y {y_true.shape[0]}, "
            f"a {a_true.shape[0]}."
        )
    if scores.shape[0] == 0:
        raise ValueError("scores must not be empty.")
    # The boolean group and outcome masks
    a0: torch.Tensor = (a_true == 0)
    a1: torch.Tensor = (a_true == 1)
    y0: torch.Tensor = (y_true == 0)
    y1: torch.Tensor = (y_true == 1)
    # Cell extraction is mask-based, so row order stays aligned
    def cell_scores(cell_mask: torch.Tensor, cell_label: str) -> torch.Tensor:
        """
        Extract one cell's scores with the configured empty policy.

        Parameters
        ----------
        cell_mask : torch.Tensor
            The boolean cell membership.
        cell_label : str
            The cell name for the error.

        Returns
        -------
        torch.Tensor
            The cell's score vector.

        """
        # An empty cell has no empirical distribution to compare; the
        # policy is configurable so a study dataset with an empty y x A
        # cell can degrade explicitly rather than aborting unconditionally
        if int(cell_mask.sum()) == 0:
            if empty_cell_policy == "return_zero":
                # The cell contributes a zero distance (documented
                # degradation: the term is skipped with the log)
                return torch.zeros(1, device=scores.device,
                                   dtype=torch.float64)
            raise ValueError(
                f"cell '{cell_label}' is empty; the equalised-odds "
                f"penalty requires observations in every cell (set "
                f"empty_cell_policy='return_zero' for a documented "
                f"degradation)."
            )
        return scores[cell_mask]

    # Equalised odds: the two outcome cells, groups compared within each
    if criterion == "equalised_odds":
        left_y0: torch.Tensor = cell_scores(y0 & a0, "y=0,a=0")
        right_y0: torch.Tensor = cell_scores(y0 & a1, "y=0,a=1")
        left_y1: torch.Tensor = cell_scores(y1 & a0, "y=1,a=0")
        right_y1: torch.Tensor = cell_scores(y1 & a1, "y=1,a=1")
        # Equation: sum_y W1(mu_y0, mu_y1)
        penalty = wasserstein_1(left_y0, right_y0) + \
            wasserstein_1(left_y1, right_y1)
    elif criterion == "equal_opportunity":
        left: torch.Tensor = cell_scores(y0 & a0, "y=0,a=0")
        right: torch.Tensor = cell_scores(y0 & a1, "y=0,a=1")
        # Equation: W1(mu_{0,0}, mu_{0,1})
        penalty = wasserstein_1(left, right)
    else:
        # Independence: the whole population, groups compared directly
        left = cell_scores(a0, "a=0")
        right = cell_scores(a1, "a=1")
        penalty = wasserstein_1(left, right)
    return penalty.to(torch.float64)

def total_loss(
    logits: torch.Tensor,
    y_true: torch.Tensor,
    a_true: torch.Tensor,
    fairness_multiplier: float = 0.0,
    criterion: str = "equalised_odds",
) -> torch.Tensor:
    """
    The total training objective: L_pred + lambda * L_fair.

    Parameters
    ----------
    logits : torch.Tensor
        The model logits, shape (n,).
    y_true : torch.Tensor
        The binary labels, shape (n,).
    a_true : torch.Tensor
        The sensitive attribute, shape (n,).
    fairness_multiplier : float, optional
        The lambda multiplier, by default 0.0.
    criterion : str, optional
        The fairness criterion, by default equalised_odds.

    Returns
    -------
    torch.Tensor
        The scalar total loss.

    Raises
    ------
    ValueError
        If lambda is negative or any component fails its validation.

    """
    # Lambda is a non-negative trade-off intercept
    if fairness_multiplier < 0.0:
        raise ValueError(
            f"fairness_multiplier must be non-negative; received "
            f"{fairness_multiplier}."
        )
    # Equation: L_pred = the BCE-with-logits term
    prediction_loss: torch.Tensor = bcce_with_logits(logits, y_true)
    # At lambda = 0 the penalty is exactly zero by definition; evaluate
    # it only when it contributes, and return the exact zero otherwise
    if fairness_multiplier == 0.0:
        return prediction_loss
    # Probabilities for the penalty: the criterion lives on the score
    # distributions, so sigma(eta) is applied for this term only
    scores: torch.Tensor = torch.sigmoid(logits)
    fair_term: torch.Tensor = fairness_penalty(
        scores, y_true, a_true, criterion
    )
    # Equation: L_total = L_pred + lambda * L_fair
    total = prediction_loss + fairness_multiplier * fair_term
    return total
````

````python
# Task 7: Implement Training and Hyperparameter Search

# ==============================================================================
# Task 7: Implement Training and Hyperparameter Search
# ==============================================================================

# Make the sibling task modules importable as the study pipeline grows

# Configure the module-level logger for stage diagnostics

# The early-stopping patience in epochs without improvement
DEFAULT_PATIENCE: int = 20
# The default number of random-search configurations per lambda value
DEFAULT_N_CONFIGS: int = 30
# The number of seeds per final replication
DEFAULT_NUM_SEEDS: int = 10

@dataclass(frozen=True)
class TrainingRunResult:
    """
    One training run's outcome.

    Attributes
    ----------
    model_state : Dict[str, torch.Tensor]
        The best-epoch state dict, deep-copied at the moment of the
        improvement.
    best_validation_loss : float
        The validation total loss at the best epoch.
    epoch_history : List[float]
        The per-epoch validation losses.
    epochs_trained : int
        The number of epochs executed before early stopping.
    early_stopped : bool
        True when the patience rule terminated the run.
    diverged : bool
        True when the loss went NaN/inf at any evaluation.

    """

    # The best-epoch parameter state
    model_state: Dict[str, torch.Tensor]
    # The best validation loss
    best_validation_loss: float
    # The per-epoch validation losses
    epoch_history: List[float]
    # Epochs executed
    epochs_trained: int
    # Whether early stopping triggered
    early_stopped: bool
    # Whether the run diverged
    diverged: bool

@dataclass(frozen=True)
class FinalSeedResult:
    """
    One final seed's trained model and frozen evaluation.

    Attributes
    ----------
    seed : int
        The replication seed.
    model : nn.Module
        The frozen trained model.
    test_logits : torch.Tensor
        The test-split logits, shape (n_test,).
    test_probabilities : torch.Tensor
        The test-split predicted probabilities, shape (n_test,).

    """

    # The replication seed
    seed: int
    # The frozen trained model
    model: nn.Module
    # The test-split logits
    test_logits: torch.Tensor
    # The test-split probabilities
    test_probabilities: torch.Tensor

# -------------------------------------------------------------------------------------------------------------------------------
# Task 7, Step 1: Random search candidate sampler
# -------------------------------------------------------------------------------------------------------------------------------
def sample_random_configs(
    hyperparameter_grid: Dict[str, List[Any]],
    n_configs: int = DEFAULT_N_CONFIGS,
    seed: int = 42,
) -> List[Dict[str, Any]]:
    """
    Sample candidate configurations uniformly with replacement.

    Parameters
    ----------
    hyperparameter_grid : Dict[str, List[Any]]
        The six discrete candidate lists: learning_rate, weight_decay,
        batch_size, hidden_layer_widths, activation, dropout_rate.
    n_configs : int, optional
        The number of candidates, by default 30.
    seed : int, optional
        The sampling seed, by default 42.

    Returns
    -------
    List[Dict[str, Any]]
        The candidate configurations, each with the grid's exact keys.

    Raises
    ------
    ValueError
        If the grid is missing a key, or any candidate list is empty.

    """
    # The grid must declare every dimension before sampling
    expected_keys: Tuple[str, ...] = ("learning_rate", "weight_decay",
                                      "batch_size", "hidden_layer_widths",
                                      "activation", "dropout_rate")
    missing: List[str] = [k for k in expected_keys if k not in hyperparameter_grid]
    if missing:
        raise ValueError(
            f"hyperparameter_grid missing key(s): {missing}."
        )
    for key in expected_keys:
        if len(hyperparameter_grid[key]) == 0:
            raise ValueError(f"grid dimension '{key}' is empty.")
    if n_configs < 1:
        raise ValueError(f"n_configs must be >= 1; received {n_configs}.")
    # A local generator keeps the sampling reproducible independently of
    # the global random state
    generator = np.random.RandomState(seed)
    candidates: List[Dict[str, Any]] = []
    for _ in range(n_configs):
        # One uniform draw per dimension, with replacement
        candidate: Dict[str, Any] = {
            key: hyperparameter_grid[key][
                int(generator.randint(0, len(hyperparameter_grid[key])))
            ]
            for key in expected_keys
        }
        # The width list is copied so later mutation cannot rewrite a
        # stored candidate
        candidate["hidden_layer_widths"] = list(
            candidate["hidden_layer_widths"]
        )
        # Every candidate must lie within the declared grid
        for key in expected_keys:
            membership: bool = any(
                str(candidate[key]) == str(value)
                for value in hyperparameter_grid[key]
            )
            if not membership:
                raise ValueError(
                    f"sampled candidate value {candidate[key]!r} for "
                    f"'{key}' is outside the grid."
                )
        candidates.append(candidate)
    return candidates

# -------------------------------------------------------------------------------------------------------------------------------
# Task 7, Step 2: The training loop with AdamW and early stopping
# -------------------------------------------------------------------------------------------------------------------------------
def setup_seeded_environment(seed: int, cudnn_deterministic: bool = True) -> None:
    """
    Set the deterministic seed triple and the cudnn determinism flags.

    Parameters
    ----------
    seed : int
        The seed applied to Python, NumPy and PyTorch.
    cudnn_deterministic : bool, optional
        When True (default), cudnn determinism is enabled and benchmark
        mode disabled.

    """
    # The three independent generators of the study's seed discipline
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    # GPU determinism: deterministic kernels, benchmark mode off
    torch.backends.cudnn.deterministic = cudnn_deterministic
    torch.backends.cudnn.benchmark = not cudnn_deterministic

def scoring_context(model: nn.Module) -> Any:
    """
    Return a context manager that scores a model without mutating it.

    B-4 remediation. Three call sites previously set
    ``model.full_batch_guarantee = False`` and restored none of them, so the
    findr identifiability guard was dead from the first validation
    evaluation of every run. This helper is the single sanctioned scoring
    entry point: for a findr module it delegates to the module's own
    ``evaluation_mode`` context manager, which restores both the guard and
    the train/eval flag in a ``finally`` block; for any other module it
    provides the equivalent train/eval save-and-restore.

    Parameters
    ----------
    model : nn.Module
        The model to score. Any nn.Module is accepted; findr modules take
        the guard-aware path.

    Returns
    -------
    Any
        A context manager yielding the model in deterministic eval mode and
        restoring the entry state on exit, exception or not.

    """
    # A findr module owns a guard, so it supplies its own restoring context
    if hasattr(model, "evaluation_mode"):
        return model.evaluation_mode()

    # Every other module needs only the train/eval flag preserved
    @contextlib.contextmanager
    def _plain_eval() -> Any:
        """Save the training flag, force eval, restore on exit."""
        # Capture the entry training mode before touching anything
        previous_training: bool = model.training
        try:
            # Deterministic scoring requires dropout and batchnorm frozen
            model.eval()
            yield model
        finally:
            # Restore the captured mode unconditionally
            model.train(previous_training)

    return _plain_eval()


def forward_scores(
    model: nn.Module,
    x: torch.Tensor,
    is_training_population: Optional[bool] = None,
) -> torch.Tensor:
    """
    Call a model's forward, passing the population flag only when supported.

    B-1(3) requires findr callers to declare which population they are
    scoring. The LR and NN baselines have no such parameter, so this shim
    keeps every call site uniform without forcing a signature change on the
    baselines.

    Parameters
    ----------
    model : nn.Module
        The model to evaluate.
    x : torch.Tensor
        The design block, shape (n, p).
    is_training_population : Optional[bool], optional
        Passed through to findr; ignored by the baselines.

    Returns
    -------
    torch.Tensor
        The logits, shape (n,).

    """
    # Only the semi-structured module carries the population parameter
    if is_training_population is not None and hasattr(
            model, "forward_decomposed"):
        return model(x, is_training_population=is_training_population)
    # Baselines take the plain single-argument forward
    return model(x)

def _evaluate_loss(
    model: nn.Module,
    X_tensor: torch.Tensor,
    y_tensor: torch.Tensor,
    a_tensor: torch.Tensor,
    fairness_multiplier: float,
    criterion: str,
) -> float:
    """
    Compute a total loss under no-grad evaluation, without mutating a model.

    Objective (Task 6 Step 3): L_total = L_pred + lambda * L_fair, evaluated
    on a held-out partition.

    B-4 remediation: this helper previously set the findr full-batch
    guarantee to False and never restored it. Because it runs once per
    epoch inside the training loop, the identifiability guard was dead from
    epoch one onward for every run, and this call was also what cached the
    stale epoch-zero evaluation projection. The scoring context now
    restores both the guard and the train/eval mode in a finally block, so
    the model's externally visible state is unchanged by this call.

    Parameters
    ----------
    model : nn.Module
        The model to evaluate; its state is preserved exactly.
    X_tensor : torch.Tensor
        The partition design, shape (n, p).
    y_tensor : torch.Tensor
        The partition labels, shape (n,).
    a_tensor : torch.Tensor
        The partition sensitive attribute, shape (n,).
    fairness_multiplier : float
        The lambda multiplier.
    criterion : str
        The fairness criterion.

    Returns
    -------
    float
        The scalar total loss, float64.

    Raises
    ------
    ValueError
        If the partition tensors are row-misaligned.

    """
    # The three partition tensors must describe the same population
    if not (X_tensor.shape[0] == y_tensor.shape[0] == a_tensor.shape[0]):
        raise ValueError(
            f"evaluation tensors are row-misaligned: X {X_tensor.shape[0]}, "
            f"y {y_tensor.shape[0]}, a {a_tensor.shape[0]}."
        )
    # B-4: the restoring scoring context replaces the destructive mutation.
    # Evaluation populations differ from the training population BY DESIGN,
    # so the guard is suspended for the duration of this block only
    with scoring_context(model):
        with torch.no_grad():
            # Out-of-sample by construction: declare it explicitly so findr
            # applies the fixed per-row projection rather than the in-sample
            # decomposition
            logits: torch.Tensor = forward_scores(
                model, X_tensor, is_training_population=False
            )
            # Equation: L_total = L_pred + lambda * L_fair
            loss_value: torch.Tensor = total_loss(
                logits, y_tensor, a_tensor, fairness_multiplier, criterion
            )
            return float(loss_value)

def resolve_effective_batch_size(
    requested_batch: int,
    n_train: int,
    is_findr: bool,
    fairness_multiplier: float,
    y_train: torch.Tensor,
    a_train: torch.Tensor,
) -> int:
    """
    Resolve the batch size actually usable for one model class.

    B-9(1). The search grid offers batch sizes {32, 64, 128, 256}, but three
    constraints can override the request:

      1. findr requires FULL BATCH unconditionally. The annihilation
         property X~^T H_perp = 0 that identifies the structured
         coefficients is defined over the whole training population; a
         mini-batch breaks it, so the batch axis is not a free
         hyperparameter for this model class.
      2. When lambda > 0 the objective carries the equalised-odds
         Wasserstein penalty, which is a distributional quantity over the
         four (y, a) cells. A batch that empties a cell cannot evaluate it.
         The batch must therefore be large enough to carry at least two
         observations from the SMALLEST cell.
      3. A request at or above the population size is full batch anyway.

    Parameters
    ----------
    requested_batch : int
        The batch size drawn from the hyperparameter grid.
    n_train : int
        The training population size.
    is_findr : bool
        Whether the model is the semi-structured architecture.
    fairness_multiplier : float
        The lambda multiplier; the cell constraint binds only when positive.
    y_train : torch.Tensor
        The training labels, shape (n_train,).
    a_train : torch.Tensor
        The training sensitive attribute, shape (n_train,).

    Returns
    -------
    int
        The effective batch size; a value >= n_train denotes full batch.

    Raises
    ------
    ValueError
        If the requested batch size is not positive.

    """
    # A non-positive batch size is a configuration defect
    if requested_batch <= 0:
        raise ValueError(
            f"batch_size must be positive; received {requested_batch}."
        )
    # Constraint 1: findr is identified only on the full population
    if is_findr:
        return n_train
    # Constraint 3: a request covering the population is full batch
    if requested_batch >= n_train:
        return n_train
    # Constraint 2 applies only when the fairness term is active
    if fairness_multiplier <= 0.0:
        return requested_batch
    # Size the smallest (y, a) cell; the penalty needs every cell populated
    smallest_cell: int = n_train
    for y_value in (0, 1):
        for a_value in (0, 1):
            # Count the cell membership with a boolean mask
            cell_size: int = int(
                ((y_train == y_value) & (a_train == a_value)).sum()
            )
            smallest_cell = min(smallest_cell, cell_size)
    # A structurally empty cell cannot be repaired by any batch size; the
    # penalty itself will apply its documented empty-cell policy
    if smallest_cell == 0:
        return n_train
    # The batch must carry >= 2 rows of the smallest cell in expectation;
    # below that threshold the stratified sampler cannot fill every cell
    minimum_viable: int = int(math.ceil(2.0 * n_train / float(smallest_cell)))
    if requested_batch < minimum_viable:
        # Falling back to full batch is the honest degradation: it keeps the
        # penalty exact rather than silently estimating it on a broken cell
        logger.info(
            f"batch_size {requested_batch} cannot populate the smallest "
            f"(y, a) cell (size {smallest_cell}); falling back to full batch "
            f"so the equalised-odds penalty stays exact."
        )
        return n_train
    return requested_batch


def build_epoch_batches(
    n_train: int,
    batch_size: int,
    fairness_multiplier: float,
    y_train: torch.Tensor,
    a_train: torch.Tensor,
    generator: torch.Generator,
) -> List[torch.Tensor]:
    """
    Build one epoch's index blocks, stratified by (y, a) when required.

    B-9(1). With lambda = 0 the objective is a plain sum over observations
    and a uniform reshuffle is correct. With lambda > 0 the equalised-odds
    penalty compares the score distributions WITHIN each (y, a) cell, so
    every batch must contain observations from all four cells; the sampler
    therefore permutes each cell independently and draws a proportional
    slice of each cell into every batch.

    Parameters
    ----------
    n_train : int
        The training population size.
    batch_size : int
        The effective batch size.
    fairness_multiplier : float
        The lambda multiplier; stratification is applied when positive.
    y_train : torch.Tensor
        The training labels, shape (n_train,).
    a_train : torch.Tensor
        The training sensitive attribute, shape (n_train,).
    generator : torch.Generator
        The seeded generator driving the permutations.

    Returns
    -------
    List[torch.Tensor]
        The index blocks of one epoch; every training row appears exactly
        once across the blocks.

    """
    # The unpenalised case: a single uniform permutation, split into blocks
    if fairness_multiplier <= 0.0:
        permutation: torch.Tensor = torch.randperm(
            n_train, generator=generator)
        return list(torch.split(permutation, batch_size))
    # The penalised case: permute each (y, a) cell independently
    cell_permutations: List[torch.Tensor] = []
    for y_value in (0, 1):
        for a_value in (0, 1):
            # The absolute row indices belonging to this cell
            members: torch.Tensor = torch.nonzero(
                (y_train == y_value) & (a_train == a_value), as_tuple=False
            ).flatten()
            if members.numel() == 0:
                # An empty cell contributes nothing; the penalty's own
                # empty-cell policy governs what happens downstream
                continue
            # Shuffle within the cell so batches differ across epochs
            shuffled: torch.Tensor = members[
                torch.randperm(members.numel(), generator=generator)
            ]
            cell_permutations.append(shuffled)
    # The number of batches follows the requested block size
    n_batches: int = max(1, int(math.ceil(n_train / float(batch_size))))
    # Each cell contributes a proportional, contiguous slice to each batch
    blocks: List[List[torch.Tensor]] = [[] for _ in range(n_batches)]
    for cell in cell_permutations:
        # Split this cell into n_batches near-equal chunks
        chunks: Tuple[torch.Tensor, ...] = torch.chunk(cell, n_batches)
        for position, chunk in enumerate(chunks):
            if chunk.numel() > 0:
                blocks[position].append(chunk)
    # Concatenate the per-cell chunks into the final index blocks
    batches: List[torch.Tensor] = [
        torch.cat(parts) for parts in blocks if parts
    ]
    return batches

def train_model_with_early_stopping(
    model: nn.Module,
    X_train: torch.Tensor,
    y_train: torch.Tensor,
    a_train: torch.Tensor,
    X_val: torch.Tensor,
    y_val: torch.Tensor,
    a_val: torch.Tensor,
    hyperparameters: Dict[str, Any],
    fairness_multiplier: float = 0.0,
    criterion: str = "equalised_odds",
    max_epochs: int = 6000,
    patience: int = DEFAULT_PATIENCE,
) -> TrainingRunResult:
    """
    Train one model with AdamW and early stopping.

    Objective (Task 6 / Task 7 Step 2):
        L_total = L_pred + lambda * L_fair,
    optimised with AdamW (decoupled weight decay), for at most max_epochs,
    stopping after `patience` epochs without improvement in the validation
    total loss, restoring the best-epoch parameters before returning.

    B-9 remediation, two defects repaired.

    (1) BATCH SIZE IS NOW HONOURED. It was previously accepted and, in the
        code's own words, "carried for the record", so one of the six
        random-search dimensions had no effect at all. The model classes are
        now separated explicitly:
          * findr: full batch is MANDATORY and is enforced. The annihilation
            property X~^T H_perp = 0 is defined over the whole training
            population; a mini-batch breaks it and voids identifiability.
          * LR and NN: genuine mini-batch sampling with a per-epoch
            reshuffle.
        Critically, the fairness penalty is a DISTRIBUTIONAL quantity over
        the outcome-by-group cells, so a mini-batch that empties a cell
        cannot evaluate it. When lambda > 0 the sampler is stratified by the
        (y, a) cell so every cell is represented in every batch, and it
        falls back to full batch when any cell is too small to be split.

    (2) LARGE-N PATH. Gradient accumulation over chunks is available for the
        prediction term so that a full-batch step remains feasible at
        n = 163,208. The fairness term is NOT separable across observations
        (a Wasserstein distance between empirical distributions is not a sum
        of per-row contributions), so when lambda > 0 the step is taken on
        the full score vector; this restriction is documented rather than
        silently approximated.

    Parameters
    ----------
    model : nn.Module
        The model to train (LR, NN or findr).
    X_train, y_train, a_train : torch.Tensor
        The training tensors, shape (n_train, p) and (n_train,).
    X_val, y_val, a_val : torch.Tensor
        The validation tensors.
    hyperparameters : Dict[str, Any]
        The sampled candidate: learning_rate, weight_decay, batch_size,
        hidden_layer_widths, activation, dropout_rate.
    fairness_multiplier : float, optional
        The lambda multiplier, by default 0.0.
    criterion : str, optional
        The fairness criterion, by default equalised_odds.
    max_epochs : int, optional
        The epoch ceiling, by default 6000.
    patience : int, optional
        The epochs-without-improvement patience, by default 20.

    Returns
    -------
    TrainingRunResult
        The best state, the best validation loss, the history, the epoch
        count, the early-stopping flag and the divergence flag.

    Raises
    ------
    ValueError
        If the tensors misalign or a required hyperparameter is absent.
    RuntimeError
        If the run produced no finite improvement.

    """
    # The population and feature shapes must be aligned
    if X_train.shape[1] != X_val.shape[1]:
        raise ValueError(
            f"training and validation feature counts differ: "
            f"{X_train.shape[1]} vs {X_val.shape[1]}."
        )
    if not (X_train.shape[0] == y_train.shape[0] == a_train.shape[0]):
        raise ValueError("training tensors are row-misaligned.")
    # The candidate must carry the lr and weight decay parameters
    for key in ("learning_rate", "weight_decay"):
        if key not in hyperparameters:
            raise ValueError(
                f"hyperparameters must carry '{key}'; found "
                f"{list(hyperparameters.keys())}."
            )
    # The training population size drives both the sampler and the guard
    n_train: int = int(X_train.shape[0])
    # findr is identified only on the full training population
    is_findr: bool = hasattr(model, "forward_decomposed")
    # B-9(1): resolve the effective batch size for THIS model class
    requested_batch: int = int(hyperparameters.get("batch_size", n_train))
    effective_batch: int = resolve_effective_batch_size(
        requested_batch=requested_batch, n_train=n_train, is_findr=is_findr,
        fairness_multiplier=fairness_multiplier, y_train=y_train,
        a_train=a_train,
    )
    # Full batch is signalled by an effective size covering the population
    full_batch: bool = effective_batch >= n_train
    # AdamW: decoupled weight decay (applied to the parameter values, not
    # folded into the gradient as an L2 term)
    optimizer: torch.optim.Optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=float(hyperparameters["learning_rate"]),
        weight_decay=float(hyperparameters["weight_decay"]),
    )
    # The running state tracks the best-epoch restore
    best_loss: float = float("inf")
    best_state: Optional[Dict[str, torch.Tensor]] = None
    epochs_since_improvement: int = 0
    history: List[float] = []
    diverged: bool = False
    early_stopped: bool = False
    epochs_trained: int = 0
    # A dedicated generator keeps the shuffle reproducible and independent
    # of the ambient global RNG state
    generator: torch.Generator = torch.Generator()
    generator.manual_seed(int(hyperparameters.get("shuffle_seed", 0)))
    for epoch in range(max_epochs):
        # The module returns to training mode at the head of every epoch
        model.train()
        if full_batch:
            # The full-batch step: findr's identifiability path
            logits: torch.Tensor = forward_scores(
                model, X_train, is_training_population=True if is_findr
                else None
            )
            # Equation: L_total = L_pred + lambda * L_fair
            train_loss: torch.Tensor = total_loss(
                logits, y_train, a_train, fairness_multiplier, criterion
            )
            # A non-finite loss means the run diverged and must be excluded
            if not torch.isfinite(train_loss):
                diverged = True
                break
            optimizer.zero_grad(set_to_none=True)
            train_loss.backward()
            optimizer.step()
        else:
            # B-9(1): the mini-batch path for LR and NN. The epoch is a
            # single pass over a fresh permutation of the population
            batch_indices: List[torch.Tensor] = build_epoch_batches(
                n_train=n_train, batch_size=effective_batch,
                fairness_multiplier=fairness_multiplier, y_train=y_train,
                a_train=a_train, generator=generator,
            )
            for index_block in batch_indices:
                # Slice the three tensors with the SAME index block so the
                # row correspondence between X, y and a is preserved
                logits = forward_scores(model, X_train[index_block])
                train_loss = total_loss(
                    logits, y_train[index_block], a_train[index_block],
                    fairness_multiplier, criterion,
                )
                if not torch.isfinite(train_loss):
                    diverged = True
                    break
                optimizer.zero_grad(set_to_none=True)
                train_loss.backward()
                optimizer.step()
            if diverged:
                break
        # B-1(2): the parameters have moved, so any cached evaluation
        # projection is stale and must be invalidated before scoring
        if hasattr(model, "bump_parameter_version"):
            model.bump_parameter_version()
        # Validation is evaluated under no-grad against the same objective;
        # _evaluate_loss no longer mutates the model
        val_loss: float = _evaluate_loss(
            model, X_val, y_val, a_val, fairness_multiplier, criterion
        )
        if not np.isfinite(val_loss):
            diverged = True
            break
        history.append(val_loss)
        epochs_trained = epoch + 1
        if val_loss < best_loss - 1e-12:
            # A new best: deep-copy the state at the improvement
            best_loss = val_loss
            best_state = copy.deepcopy(model.state_dict())
            epochs_since_improvement = 0
        else:
            epochs_since_improvement += 1
            if epochs_since_improvement >= patience:
                # The patience rule terminates the run
                early_stopped = True
                break
    # A run that diverged before any finite evaluation has no best state
    if best_state is None:
        raise RuntimeError(
            "training produced no finite improvement; the run diverged."
        )
    # Restore the best-epoch parameters before returning the model
    model.load_state_dict(best_state)
    # The restored parameters are a new generation for the projection cache
    if hasattr(model, "bump_parameter_version"):
        model.bump_parameter_version()
    # Seal the evaluation projection from the FINAL encoder so the fitted
    # model is self-contained and serialises faithfully (B-1(2), B-5)
    if hasattr(model, "freeze_evaluation_projection"):
        model.freeze_evaluation_projection()
    # The runtime returns the best state for the caller's storage
    model_copy_for_result: Dict[str, torch.Tensor] = copy.deepcopy(
        model.state_dict())
    return TrainingRunResult(
        model_state=model_copy_for_result,
        best_validation_loss=best_loss,
        epoch_history=history,
        epochs_trained=epochs_trained,
        early_stopped=early_stopped,
        diverged=diverged,
    )

# -------------------------------------------------------------------------------------------------------------------------------
# Task 7, Step 3: Hyperparameter selection and final multi-seed training
# -------------------------------------------------------------------------------------------------------------------------------
def rank_candidates(
    validation_results: List[Tuple[Dict[str, Any], List[float]]],
) -> List[Tuple[int, Dict[str, Any], float]]:
    """
    Rank candidates by the mean validation loss across restarts.

    Parameters
    ----------
    validation_results : List[Tuple[Dict[str, Any], List[float]]]
        Candidate configs with their restart validation losses; diverged
        runs are recorded as NaN and excluded from the mean.

    Returns
    -------
    List[Tuple[int, Dict[str, Any], float]]
        Ranked (config index, config, mean validation loss).

    """
    ranked: List[Tuple[int, Dict[str, Any], float]] = []
    for index, (config, losses) in enumerate(validation_results):
        finite: List[float] = [float(l) for l in losses if np.isfinite(l)]
        if not finite:
            # A fully diverged candidate cannot be ranked
            continue
        mean_loss: float = float(np.mean(finite))
        ranked.append((index, config, mean_loss))
    # The lowest mean validation loss leads
    ranked.sort(key=lambda entry: entry[2])
    return ranked

def select_best_config(
    validation_results: List[Tuple[Dict[str, Any], List[float]]],
) -> Tuple[int, Dict[str, Any], float]:
    """
    Select the configuration with the lowest mean validation loss.

    Parameters
    ----------
    validation_results : List[Tuple[Dict[str, Any], List[float]]]
        Candidate configs with restart losses.

    Returns
    -------
    Tuple[int, Dict[str, Any], float]
        The index, config and mean loss of the best candidate.

    Raises
    ------
    ValueError
        If no candidate produced a finite validation loss.

    """
    ranked: List[Tuple[int, Dict[str, Any], float]] = rank_candidates(
        validation_results
    )
    if not ranked:
        raise ValueError(
            "no candidate produced a finite validation loss; the search "
            "failed."
        )
    return ranked[0]

def train_final_seed_runs(
    model_class: str,
    best_config: Dict[str, Any],
    X_train: np.ndarray,
    y_train: np.ndarray,
    a_train: np.ndarray,
    X_val: np.ndarray,
    y_val: np.ndarray,
    a_val: np.ndarray,
    X_test: np.ndarray,
    num_seeds: int = DEFAULT_NUM_SEEDS,
    base_seed: int = 42,
    fairness_multiplier: float = 0.0,
    criterion: str = "equalised_odds",
    max_epochs: int = 6000,
    patience: int = DEFAULT_PATIENCE,
) -> Tuple[List[FinalSeedResult], List[Dict[str, Any]]]:
    """
    Retrain the selected configuration across seeds and freeze the models.

    Parameters
    ----------
    model_class : str
        One of LR, NN, findr.
    best_config : Dict[str, Any]
        The selected hyperparameters.
    X_train, y_train, a_train : np.ndarray
        The training arrays.
    X_val, y_val, a_val : np.ndarray
        The validation arrays.
    X_test : np.ndarray
        The test design (no labels are used here; Task 8 evaluates).
    num_seeds : int, optional
        The replication count, by default 10.
    base_seed : int, optional
        The first replication seed, by default 42.
    fairness_multiplier : float, optional
        The lambda multiplier.
    criterion : str, optional
        The fairness criterion.
    max_epochs : int, optional
        The epoch ceiling.
    patience : int, optional
        The patience.

    Returns
    -------
    Tuple[List[FinalSeedResult], List[Dict[str, Any]]]
        The per-seed frozen results and the per-seed run log records.

    """
    # The tensors convert once; every seed reuses the same data
    X_train_t: torch.Tensor = torch.from_numpy(X_train.astype(np.float32))
    y_train_t: torch.Tensor = torch.from_numpy(y_train.astype(np.float32))
    a_train_t: torch.Tensor = torch.from_numpy(a_train.astype(np.float32))
    X_val_t: torch.Tensor = torch.from_numpy(X_val.astype(np.float32))
    y_val_t: torch.Tensor = torch.from_numpy(y_val.astype(np.float32))
    a_val_t: torch.Tensor = torch.from_numpy(a_val.astype(np.float32))
    X_test_t: torch.Tensor = torch.from_numpy(X_test.astype(np.float32))
    results: List[FinalSeedResult] = []
    log_records: List[Dict[str, Any]] = []
    for run_index in range(num_seeds):
        seed: int = base_seed + run_index
        # The seeded environment is set BEFORE construction (K-16)
        setup_seeded_environment(seed)
        # findr needs the training design for its thin QR
        x_train_for_qr: Optional[torch.Tensor] = (
            X_train_t if model_class == "findr" else None
        )
        model: nn.Module = build_model(
            model_class, int(X_train.shape[1]),
            list(best_config["hidden_layer_widths"]),
            str(best_config["activation"]),
            float(best_config["dropout_rate"]),
            X_train=x_train_for_qr,
        )
        # Train with the chosen (lr, wd) and the search's epoch bounds
        result: TrainingRunResult = train_model_with_early_stopping(
            model, X_train_t, y_train_t, a_train_t,
            X_val_t, y_val_t, a_val_t,
            best_config, fairness_multiplier, criterion,
            max_epochs, patience,
        )
        # The model is frozen after training: evaluation produces
        # logits and probabilities only, never another gradient step
        # B-1(4)/B-4: the restoring scoring context replaces the
        # destructive guard assignment. The guard is now a read-only
        # property, so the previous mutation would raise; more importantly
        # the context restores BOTH the guard and the train/eval mode.
        with scoring_context(model):
            with torch.no_grad():
                # The test split is out of sample by construction, so the
                # fixed per-row projection applies rather than the
                # in-sample decomposition
                test_logits: torch.Tensor = forward_scores(
                    model, X_test_t, is_training_population=False)
                test_probabilities: torch.Tensor = torch.sigmoid(test_logits)
        results.append(FinalSeedResult(
            seed=seed, model=model,
            test_logits=test_logits.detach(),
            test_probabilities=test_probabilities.detach(),
        ))
        log_records.append({
            "seed": seed,
            "model_class": model_class,
            "best_validation_loss": result.best_validation_loss,
            "epochs_trained": result.epochs_trained,
            "early_stopped": result.early_stopped,
            "learning_rate": best_config["learning_rate"],
            "weight_decay": best_config["weight_decay"],
            "hidden_layer_widths": best_config["hidden_layer_widths"],
            "activation": best_config["activation"],
            "dropout_rate": best_config["dropout_rate"],
        })
        logger.info(
            f"seed {seed}: best val loss {result.best_validation_loss:.6f}, "
            f"{result.epochs_trained} epochs, "
            f"early_stopped={result.early_stopped}."
        )
        # Release the model and its state to bound memory across seeds
        del model
        if torch.cuda.is_available():
            torch.cuda.empty_cache()
    return results, log_records

# -------------------------------------------------------------------------------------------------------------------------------
# Task 7, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def train_model_and_select_hyperparameters(
    model_class: str,
    X_train: np.ndarray,
    y_train: np.ndarray,
    a_train: np.ndarray,
    X_val: np.ndarray,
    y_val: np.ndarray,
    a_val: np.ndarray,
    X_test: np.ndarray,
    config: Dict[str, Any],
    fairness_multiplier: float = 0.0,
    criterion: str = "equalised_odds",
) -> Tuple[Dict[str, Any], List[FinalSeedResult], List[Dict[str, Any]]]:
    """
    Run the random search, select the best config and train the finals.

    Parameters
    ----------
    model_class : str
        One of LR, NN, findr.
    X_train, y_train, a_train : np.ndarray
        The training arrays.
    X_val, y_val, a_val : np.ndarray
        The validation arrays.
    X_test : np.ndarray
        The test design.
    config : Dict[str, Any]
        The study configuration (training and grid sections).
    fairness_multiplier : float, optional
        The lambda multiplier.
    criterion : str, optional
        The fairness criterion.

    Returns
    -------
    Tuple[Dict[str, Any], List[FinalSeedResult], List[Dict[str, Any]]]
        The selected configuration, the per-seed frozen results, and the
        per-seed log records.

    """
    # The study's training and grid settings
    training_cfg: Dict[str, Any] = config["network_training"]
    grid_cfg: Dict[str, Any] = config["hyperparameter_grid"]
    max_epochs: int = int(training_cfg["max_epochs"])
    patience: int = int(training_cfg["early_stopping_patience"])
    num_seeds: int = int(training_cfg["num_seeds"])
    n_configs: int = int(training_cfg.get("n_configs", DEFAULT_N_CONFIGS))
    base_seed: int = int(training_cfg["seed"])

    # Step 1: sample the candidate configurations (with replacement)
    candidates: List[Dict[str, Any]] = sample_random_configs(
        grid_cfg, n_configs, base_seed
    )
    # Step 2: score every candidate over its restarts; the restart seeds
    # are (base_seed + 1e6 + config_index) and (.. + 1) inside the seed
    # space so restarts differ deterministically from each other and the
    # final seeds
    validation_results: List[Tuple[Dict[str, Any], List[float]]] = []
    for config_index, candidate in enumerate(candidates):
        restart_losses: List[float] = []
        for restart_index in range(int(training_cfg["num_restarts_per_config"])):
            # A deterministic restart-specific seed: seed + restart index
            restart_seed: int = base_seed + 10_000 + config_index * 10 + \
                restart_index
            setup_seeded_environment(restart_seed)
            x_train_for_qr: Optional[torch.Tensor] = (
                torch.from_numpy(X_train.astype(np.float32))
                if model_class == "findr" else None
            )
            model: nn.Module = build_model(
                model_class, int(X_train.shape[1]),
                list(candidate["hidden_layer_widths"]),
                str(candidate["activation"]),
                float(candidate["dropout_rate"]),
                X_train=x_train_for_qr,
            )
            run: TrainingRunResult = train_model_with_early_stopping(
                model,
                torch.from_numpy(X_train.astype(np.float32)),
                torch.from_numpy(y_train.astype(np.float32)),
                torch.from_numpy(a_train.astype(np.float32)),
                torch.from_numpy(X_val.astype(np.float32)),
                torch.from_numpy(y_val.astype(np.float32)),
                torch.from_numpy(a_val.astype(np.float32)),
                candidate, fairness_multiplier, criterion,
                max_epochs, patience,
            )
            # A diverged restart records NaN so the mean excludes it
            restart_losses.append(
                run.best_validation_loss if not run.diverged else float("nan")
            )
            del model
        validation_results.append((candidate, restart_losses))
    # Step 3a: rank and select by the mean validation loss
    best_config: Dict[str, Any] = select_best_config(validation_results)[1]
    logger.info(
        f"selected config for {model_class} at lambda={fairness_multiplier}: "
        f"{best_config}."
    )
    # Step 3b: final multi-seed retraining with the frozen evaluation
    results, log_records = train_final_seed_runs(
        model_class, best_config,
        X_train, y_train, a_train, X_val, y_val, a_val, X_test,
        num_seeds, base_seed, fairness_multiplier, criterion,
        max_epochs, patience,
    )
    return best_config, results, log_records
````

````python
# Task 8: Implement Evaluation Metrics

# ==============================================================================
# Task 8: Implement Evaluation Metrics
# ==============================================================================

# Make the sibling task modules importable as the study pipeline grows

# Configure the module-level logger for stage diagnostics

# The log-loss protection floor; scores are clipped strictly inside
# (EPSILON, 1 - EPSILON) before any logarithm
# B-18: ONE registered epsilon across the metric and frontier paths.
# The frontier specification fixes 1e-6; two different epsilons feeding
# the same excess-log-loss difference is a reproducibility hazard.
SCORE_CLIP_EPSILON: float = 1e-6
# The NN-minus-LR AUC gap above which a dataset is called nonlinear
NONLINEAR_REGIME_GAP: float = 0.03

@dataclass(frozen=True)
class EvaluationReport:
    """
    One fitted model's metric report.

    Attributes
    ----------
    auc : float
        The test AUC.
    brier : float
        The test Brier score.
    fairness_gap : float
        The validation equalised-odds Wasserstein gap.
    log_loss : float
        The validation expected log loss under the reference risk.
    entropy_reference : float
        The validation reference-risk entropy.
    excess_log_loss : float
        log_loss minus entropy_reference.

    """

    # Test AUC
    auc: float
    # Test Brier score
    brier: float
    # Validation fairness gap
    fairness_gap: float
    # Validation expected log loss under the reference risk
    log_loss: float
    # Reference-risk entropy
    entropy_reference: float
    # The excess log loss
    excess_log_loss: float

# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Step 1: Compute AUC and Brier score on the test split
# -------------------------------------------------------------------------------------------------------------------------------
def validate_probabilities(probabilities: np.ndarray, label: str = "probabilities") -> None:
    """
    Validate that scores are finite and strictly inside (0, 1).

    Parameters
    ----------
    probabilities : np.ndarray
        The probability scores.
    label : str, optional
        The error-naming label.

    Raises
    ------
    ValueError
        If the array is empty, non-finite, or outside the closed interval
        after clipping is required.

    """
    if probabilities.ndim != 1:
        raise ValueError(f"{label} must be 1-D; received {probabilities.ndim}D.")
    if probabilities.shape[0] == 0:
        raise ValueError(f"{label} must not be empty.")
    if not np.all(np.isfinite(probabilities)):
        raise ValueError(f"{label} carries non-finite values.")
    # Scores at exactly 0 or 1 are degenerate for the log loss and the
    # Wasserstein supports must stay open; the clip is the standard guard
    if not np.all((probabilities > 0.0) & (probabilities < 1.0)):
        raise ValueError(
            f"{label} must lie strictly inside (0, 1) after clipping; "
            f"found {int(((probabilities <= 0.0) | (probabilities >= 1.0)).sum())} "
            f"outside."
        )

def clip_scores(scores: np.ndarray, epsilon: float = SCORE_CLIP_EPSILON) -> np.ndarray:
    """
    Clip scores into [epsilon, 1 - epsilon] for log and distribution use.

    The expected log loss R_log(r, s) = -(1/n) sum [r log s + (1-r) log(1-s)]
    is unbounded as s approaches 0 or 1, so every score entering a logarithm
    is clipped first.

    B-18 remediation: the default was 1e-12 while the frontier routines
    passed 1e-6 explicitly, so two different epsilons fed the SAME
    excess-log-loss difference. The specification fixes the frontier's
    1e-6, which is now the single registered constant used everywhere.

    Parameters
    ----------
    scores : np.ndarray
        The raw probability scores, any shape.
    epsilon : float, optional
        The protection floor, by default the registered 1e-6.

    Returns
    -------
    np.ndarray
        The clipped scores, float64, strictly inside (0, 1).

    Raises
    ------
    ValueError
        If epsilon is outside (0, 0.5) or the scores are non-finite.

    """
    # An epsilon at or above one half would invert the clipping interval
    if not 0.0 < epsilon < 0.5:
        raise ValueError(
            f"epsilon must lie in (0, 0.5); received {epsilon}."
        )
    # Promote to float64 so the clip and the subsequent logarithm are exact
    values: np.ndarray = np.asarray(scores, dtype=np.float64)
    # A non-finite score would survive the clip and poison the logarithm
    if not np.isfinite(values).all():
        raise ValueError(
            f"scores carry {int((~np.isfinite(values)).sum())} non-finite "
            f"value(s); the log loss is undefined for them."
        )
    # The clip is elementwise and preserves the ordering exactly
    return np.clip(values, epsilon, 1.0 - epsilon)

def evaluate_auc_brier(
    probabilities: np.ndarray,
    y_true: np.ndarray,
    record_ids: Optional[np.ndarray] = None,
) -> Dict[str, float]:
    """
    Compute the test AUC and Brier score for one model.

    Equation: AUC = (1/(n1 n0)) sum_{i:y_i=1} sum_{j:y_j=0} 1{hat Y_i > hat Y_j};
    Brier = (1/n) sum_i (hat Y_i - y_i)^2.

    Parameters
    ----------
    probabilities : np.ndarray
        The predicted probabilities, shape (n,).
    y_true : np.ndarray
        The binary labels, shape (n,).
    record_ids : Optional[np.ndarray], optional
        The observation ids; when provided, both inputs are sorted by the
        ids so an accidental reorder cannot silently misalign them.

    Returns
    -------
    Dict[str, float]
        The AUC and the Brier score.

    Raises
    ------
    ValueError
        If the labels are not binary or any validation fails.

    """
    # The label domain is binary by contract
    if not set(np.unique(y_true)) <= {0, 1}:
        raise ValueError(
            f"y_true must be binary; found {sorted(set(y_true.tolist()))}."
        )
    if len(probabilities) != len(y_true):
        raise ValueError(
            f"length mismatch: probabilities {len(probabilities)}, "
            f"labels {len(y_true)}."
        )
    # Optional alignment: sort both arrays by the ids together
    if record_ids is not None:
        order: np.ndarray = np.argsort(record_ids, kind="stable")
        probabilities = probabilities[order]
        y_true = y_true[order]
        # The ids must themselves be aligned to both arrays
        if len(record_ids) != len(probabilities):
            raise ValueError("record_ids length mismatch.")
    # Both classes must be present for a defined AUC
    if len(np.unique(y_true)) < 2:
        raise ValueError(
            "AUC is undefined with a single class in the test partition."
        )
    # Equation: AUC via the pairwise ranking probability (ties handled
    # robustly by the ranking-based implementation)
    auc_value: float = float(roc_auc_score(y_true, probabilities))
    # Equation: Brier = mean((hat Y - y)^2)
    brier_value: float = float(brier_score_loss(
        y_true, probabilities, pos_label=1
    ))
    return {"auc": auc_value, "brier": brier_value}

# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Step 2: Compute the fairness gap and excess log loss
# -------------------------------------------------------------------------------------------------------------------------------
def expected_log_loss(
    reference_risk: np.ndarray, scores: np.ndarray,
) -> float:
    """
    The expected Bernoulli log loss under the reference risk.

    Equation: R_hat_log(r, s) = -(1/n) sum_i [ r_i log s_i +
    (1 - r_i) log(1 - s_i) ].

    Parameters
    ----------
    reference_risk : np.ndarray
        The reference probabilities, shape (n,).
    scores : np.ndarray
        The candidate scores, shape (n,).

    Returns
    -------
    float
        The expected log loss.

    Raises
    ------
    ValueError
        If the arrays misalign, or a reference risk is outside (0, 1)
        after clipping.

    """
    if len(reference_risk) != len(scores):
        raise ValueError(
            f"length mismatch: reference {len(reference_risk)}, scores "
            f"{len(scores)}."
        )
    if not np.all((reference_risk > 0.0) & (reference_risk < 1.0)):
        raise ValueError(
            "reference risk must lie strictly inside (0, 1); the true "
            "conditional risk and the unpenalised scores both satisfy "
            "this."
        )
    # Both inputs are clipped so no logarithm sees 0
    r: np.ndarray = clip_scores(reference_risk)
    s: np.ndarray = clip_scores(scores)
    # Equation: -(1/n) sum [ r log s + (1 - r) log(1 - s) ]
    log_term: np.ndarray = r * np.log(s) + (1.0 - r) * np.log(1.0 - s)
    return float(-np.mean(log_term))

def reference_entropy(reference_risk: np.ndarray) -> float:
    """
    The empirical Bernoulli entropy of the reference risks.

    Equation: H_hat(r) = -(1/n) sum_i [ r_i log r_i +
    (1 - r_i) log(1 - r_i) ]; the expected log loss obtained by predicting
    r itself, and the irreducible term of the KL decomposition.

    Parameters
    ----------
    reference_risk : np.ndarray
        The reference probabilities, shape (n,).

    Returns
    -------
    float
        The entropy.

    """
    if not np.all((reference_risk > 0.0) & (reference_risk < 1.0)):
        raise ValueError("reference risk must lie strictly inside (0, 1).")
    r: np.ndarray = clip_scores(reference_risk)
    # Equation: H_hat(r) = -(1/n) sum [ r log r + (1-r) log(1-r) ]
    entropy: float = float(-np.mean(
        r * np.log(r) + (1.0 - r) * np.log(1.0 - r)
    ))
    return entropy

def compute_validation_metrics(
    validation_scores: np.ndarray,
    y_val: np.ndarray,
    a_val: np.ndarray,
    reference_risk: np.ndarray,
    criterion: str = "equalised_odds",
) -> Dict[str, float]:
    """
    Compute the validation fairness gap and the excess log loss.

    Equations: Phi(s; y, a) = sum_{y in {0,1}} W1(mu_{y0}, mu_{y1})
    (equalised odds); excess log loss = R_hat_log(r, s) - H_hat(r).

    Parameters
    ----------
    validation_scores : np.ndarray
        The validation probabilities, shape (n,).
    y_val : np.ndarray
        The validation labels, shape (n,).
    a_val : np.ndarray
        The validation sensitive attributes, shape (n,).
    reference_risk : np.ndarray
        The reference risk vector, shape (n,).
    criterion : str, optional
        The fairness criterion, by default equalised_odds.

    Returns
    -------
    Dict[str, float]
        fairness_gap, log_loss, entropy_reference, excess_log_loss.

    """
    # The four validation arrays are the same population, in the same
    # order; the joint length check is the alignment proof
    lengths: Tuple[int, ...] = (len(validation_scores), len(y_val),
                                len(a_val), len(reference_risk))
    if not (lengths[0] == lengths[1] == lengths[2] == lengths[3]):
        raise ValueError(
            f"validation alignment mismatch: {lengths}."
        )
    if criterion not in SUPPORTED_CRITERIA:
        raise ValueError(
            f"criterion must be one of {SUPPORTED_CRITERIA}; received "
            f"'{criterion}'."
        )
    # The scores are clipped before the penalty and the log loss for one
    # consistent definition across the frontier and the tables
    scores_clipped: np.ndarray = clip_scores(validation_scores)
    # Equation: Phi(s; y, a) via the exact W1 penalty of Task 6
    gap_value: float = float(fairness_penalty(
        torch.from_numpy(scores_clipped.astype(np.float64)),
        torch.from_numpy(y_val.astype(np.float64)),
        torch.from_numpy(a_val.astype(np.float64)),
        criterion,
    ))
    # Equation: R_hat_log(r, s)
    log_loss: float = expected_log_loss(reference_risk, validation_scores)
    # Equation: H_hat(r)
    entropy_value: float = reference_entropy(reference_risk)
    # Equation: excess = R_hat_log - H_hat
    excess: float = log_loss - entropy_value
    return {
        "fairness_gap": gap_value,
        "log_loss": log_loss,
        "entropy_reference": entropy_value,
        "excess_log_loss": excess,
    }

# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Step 3: Aggregate metrics across seeds
# -------------------------------------------------------------------------------------------------------------------------------
def aggregate_metrics(
    records: List[Dict[str, Any]],
    group_keys: Tuple[str, ...] = ("dataset", "mode", "model_class", "lambda"),
    metric_columns: Tuple[str, ...] = ("auc", "brier", "fairness_gap",
                                       "excess_log_loss", "accuracy_gap"),
) -> pd.DataFrame:
    """
    Aggregate per-seed records into mean and sample std per combination.

    Parameters
    ----------
    records : List[Dict[str, Any]]
        The per-seed evaluation records.
    group_keys : Tuple[str, ...], optional
        The group identity columns.
    metric_columns : Tuple[str, ...], optional
        The numeric metric columns to aggregate.

    Returns
    -------
    pd.DataFrame
        A tidy frame with one row per (group keys, metric) and columns
        mean and std (ddof=1).

    Raises
    ------
    ValueError
        If the records are empty or a metric column is absent.

    """
    if not records:
        raise ValueError("no records to aggregate.")
    frame: pd.DataFrame = pd.DataFrame(records)
    missing: List[str] = [
        column for column in list(group_keys) + list(metric_columns)
        if column not in frame.columns
    ]
    if missing:
        raise ValueError(f"record frame missing column(s): {missing}.")
    # Only the declared metric columns are aggregated
    present_metrics: List[str] = [
        column for column in metric_columns if column in frame.columns
    ]
    aggregated: pd.DataFrame = (
        frame.groupby(list(group_keys), dropna=False)[present_metrics]
        .agg(["mean", "std"])
    )
    # The sample standard deviation (ddof=1) is the study's convention
    # across the 10 seeds
    aggregated = aggregated.rename(
        columns={"std": "std_ddof1"}
    )
    return aggregated

def classify_regimes(
    summary_frame: pd.DataFrame,
    gap_threshold: float = NONLINEAR_REGIME_GAP,
) -> pd.DataFrame:
    """
    Classify datasets into nonlinear and approximately linear regimes.

    A dataset is nonlinear when the unpenalised (lambda = 0) NN AUC
    exceeds the LR AUC by more than the gap threshold; the criterion is
    taken from the task text.

    Parameters
    ----------
    summary_frame : pd.DataFrame
        The per-seed aggregated metric frame, one row per
        (dataset, mode, model_class, lambda, metric).
    gap_threshold : float, optional
        The regime threshold, by default 0.03.

    Returns
    -------
    pd.DataFrame
        The frame with the regime column attached per dataset.

    """
    # The mean column block, indexed by the group keys: the metric is a
    # COLUMN level of the aggregated frame, never an index level
    means: pd.DataFrame = summary_frame.xs(
        "mean", axis=1, level=1
    ).reset_index()
    # The lambda-0 AUC cut of the two model classes per dataset
    baseline: pd.DataFrame = means[means["lambda"] == 0.0].pivot_table(
        index="dataset", columns="model_class", values="auc",
        aggfunc="mean",
    )
    # A dataset without both classes cannot be classified
    regimes: Dict[str, str] = {}
    for dataset, row in baseline.iterrows():
        if "NN" not in row or "LR" not in row:
            continue
        gap: float = float(row["NN"] - row["LR"])
        # Equation: nonlinear iff NN AUC - LR AUC > gap_threshold
        regimes[dataset] = "nonlinear" if gap > gap_threshold else "linear"
    result: pd.DataFrame = summary_frame.copy()
    # The regime is an index-aligned column attached per dataset
    dataset_level: np.ndarray = result.index.get_level_values("dataset")
    result["regime"] = [regimes.get(str(d), "unclassified")
                        for d in dataset_level]
    return result

# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def evaluate_model(
    test_probabilities: np.ndarray,
    y_test: np.ndarray,
    test_ids: Optional[np.ndarray],
    validation_scores: np.ndarray,
    y_val: np.ndarray,
    a_val: np.ndarray,
    reference_risk: np.ndarray,
    criterion: str = "equalised_odds",
) -> EvaluationReport:
    """
    Evaluate one fitted model on the test and validation splits.

    Parameters
    ----------
    test_probabilities : np.ndarray
        The test-split probabilities, shape (n_test,).
    y_test : np.ndarray
        The test labels, shape (n_test,).
    test_ids : Optional[np.ndarray]
        The test record ids, used for the alignment sort when provided.
    validation_scores : np.ndarray
        The validation-split probabilities, shape (n_val,).
    y_val, a_val : np.ndarray
        The validation labels and sensitive attributes.
    reference_risk : np.ndarray
        The reference risk vector for the validation split.
    criterion : str, optional
        The fairness criterion, by default equalised_odds.

    Returns
    -------
    EvaluationReport
        The frozen metric report.

    """
    # Step 1: the test metrics
    tolerance_metric: Dict[str, float] = evaluate_auc_brier(
        test_probabilities, y_test, test_ids
    )
    # Step 2: the validation frontier metrics
    validation_metrics: Dict[str, float] = compute_validation_metrics(
        validation_scores, y_val, a_val, reference_risk, criterion
    )
    # The frozen report binds both steps
    return EvaluationReport(
        auc=tolerance_metric["auc"],
        brier=tolerance_metric["brier"],
        fairness_gap=validation_metrics["fairness_gap"],
        log_loss=validation_metrics["log_loss"],
        entropy_reference=validation_metrics["entropy_reference"],
        excess_log_loss=validation_metrics["excess_log_loss"],
    )
````

````python
# Task 9: Implement Interpretability Diagnostics

# ==============================================================================
# Task 9: Implement Interpretability Diagnostics
# ==============================================================================

# Make the sibling task modules importable as the study pipeline grows

# Configure the module-level logger for stage diagnostics

# The default decision threshold on the logit scale (config default 0.0)
DEFAULT_DECISION_THRESHOLD: float = 0.0
# The default CSER perturbation size (config default 0.1)
DEFAULT_CSER_DELTA: float = 0.1
# The variance-decomposition consistency tolerance
VARIANCE_IDENTITY_TOLERANCE: float = 1e-3

@dataclass(frozen=True)
class DiagnosticsReport:
    """
    The interpretability diagnostics of one fitted findr model.

    Attributes
    ----------
    evr : float
        The explained variability ratio: Var(eta_str)/Var(eta).
    ddr : float
        The decision disagreement rate.
    ddr_count : int
        The count of disagreements behind the rate.
    cser_min : float
        The minimum feature-level sign error rate.
    cser_max : float
        The maximum feature-level sign error rate.
    per_feature_cser : Dict[str, float]
        The per-feature CSER values, including zeros.
    split : str
        The evaluation split label the diagnostics ran on.

    """

    # The explained variability ratio
    evr: float
    # The decision disagreement rate
    ddr: float
    # The disagreement count
    ddr_count: int
    # The minimum feature CSER
    cser_min: float
    # The maximum feature CSER
    cser_max: float
    # Per-feature CSER values
    per_feature_cser: Dict[str, float]
    # The evaluation split label
    split: str

# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Step 1: Compute the Explained Variability Ratio
# -------------------------------------------------------------------------------------------------------------------------------
def compute_evr(
    eta_str: np.ndarray,
    eta_unstr: np.ndarray,
    split: str = "validation",
    variance_tolerance: float = VARIANCE_IDENTITY_TOLERANCE,
) -> float:
    """
    Compute EVR = Var_n(eta_str) / Var_n(eta) with population variance.

    Equation (12): EVR = Var_n(eta_str)/Var_n(eta), ddof = 0. The
    orthogonalisation makes the total variance decompose additively on
    the training population: Var(eta) = Var(eta_str) + Var(eta_unstr);
    the identity is verified here and a warning is emitted when the
    out-of-sample covariance deviates beyond the tolerance.

    Parameters
    ----------
    eta_str : np.ndarray
        The structured logits, shape (n,).
    eta_unstr : np.ndarray
        The unstructured logits, shape (n,).
    split : str, optional
        The split label for the record.
    variance_tolerance : float, optional
        The additive-identity tolerance, by default 1e-3.

    Returns
    -------
    float
        The EVR ratio.

    Raises
    ------
    ValueError
        If the vectors misalign or the total variance vanishes.

    """
    if eta_str.shape != eta_unstr.shape:
        raise ValueError(
            f"logit vectors misaligned: {eta_str.shape} vs "
            f"{eta_unstr.shape}."
        )
    if eta_str.shape[0] == 0:
        raise ValueError("empty logit vector.")
    # Equation: eta = eta_str + eta_unstr
    eta_total: np.ndarray = eta_str + eta_unstr
    # Population variance (ddof = 0, the paper's convention)
    var_str: float = float(np.var(eta_str, ddof=0))
    var_unstr: float = float(np.var(eta_unstr, ddof=0))
    var_total: float = float(np.var(eta_total, ddof=0))
    # A zero-variance total logit leaves the ratio undefined
    if var_total < 1e-16:
        raise ValueError(
            "EVR is undefined: the total logit variance is zero."
        )
    # Equation: EVR = Var(eta_str) / Var(eta)
    evr: float = var_str / var_total
    # Equation: Var(eta) = Var(eta_str) + Var(eta_unstr) (identity from
    # the orthogonalisation); verified, warned when the covariance drifts
    deviation: float = abs(var_total - (var_str + var_unstr))
    if deviation > variance_tolerance:
        warnings.warn(
            f"split '{split}': the variance decomposition deviates by "
            f"{deviation:.6f} (|Var(eta) - (Var_str + Var_unstr)|); the "
            f"orthogonalisation identity is exact only on the training "
            f"population.",
            RuntimeWarning,
        )
    if not 0.0 <= evr <= 1.0 + variance_tolerance:
        warnings.warn(
            f"split '{split}': EVR {evr:.6f} is outside the unit "
            f"interval; the covariance between the components is "
            f"non-negligible on this population.",
            RuntimeWarning,
        )
    return evr

# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Step 2: Compute the Decision Disagreement Rate
# -------------------------------------------------------------------------------------------------------------------------------
def compute_ddr(
    eta_str: np.ndarray,
    eta_total: np.ndarray,
    decision_threshold: float = DEFAULT_DECISION_THRESHOLD,
) -> Tuple[float, int]:
    """
    Compute the decision disagreement rate and its count.

    Equation (13): DDR = (1/n) sum_i 1[ 1{eta_str_i >= 0} !=
    1{eta_i >= 0} ], the share of observations where the structured-only
    decision disagrees with the integrated model at the logit threshold.

    Parameters
    ----------
    eta_str : np.ndarray
        The structured logits, shape (n,).
    eta_total : np.ndarray
        The full logits, shape (n,).
    decision_threshold : float, optional
        The logit decision threshold, by default 0.0.

    Returns
    -------
    Tuple[float, int]
        The DDR in [0, 1] and the disagreement count.

    Raises
    ------
    ValueError
        If the vectors misalign or carry non-finite values.

    """
    if eta_str.shape != eta_total.shape:
        raise ValueError(
            f"logit vectors misaligned: {eta_str.shape} vs "
            f"{eta_total.shape}."
        )
    if not (np.isfinite(eta_str).all() and np.isfinite(eta_total).all()):
        raise ValueError("logit vectors carry non-finite values.")
    # Equation: the class labels at the threshold, logit scale
    decision_str: np.ndarray = (eta_str >= decision_threshold).astype(
        np.int64
    )
    decision_total: np.ndarray = (eta_total >= decision_threshold).astype(
        np.int64
    )
    # Equation: the disagreement indicator mean
    disagreement: np.ndarray = decision_str != decision_total
    count: int = int(np.sum(disagreement))
    rate: float = float(np.mean(disagreement))
    return rate, count

# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Step 3: Compute CSER and summaries
# -------------------------------------------------------------------------------------------------------------------------------
def extract_structured_beta(
    model: torch.nn.Module, model_class: str,
) -> np.ndarray:
    """
    Extract the fitted structured coefficient vector (excluding the intercept).

    Parameters
    ----------
    model : torch.nn.Module
        The fitted model (findr or LR).
    model_class : str
        One of LR, NN, findr.

    Returns
    -------
    np.ndarray
        The structured coefficients, shape (p,).

    Raises
    ------
    ValueError
        If the model class has no structured branch.

    """
    # LR and findr share the structured linear layer contract
    if model_class == "LR":
        layer = model.linear
    elif model_class == "findr":
        layer = model.structured_layer
    else:
        raise ValueError(
            f"CSER requires a structured coefficient vector; the model "
            f"class '{model_class}' has no structured branch."
        )
    # The weight is (p, 1); the coefficients are the weight row
    beta: np.ndarray = layer.weight.detach().numpy().reshape(-1)
    return beta

def reprepare_features(
    raw_frame: pd.DataFrame,
    continuous: List[str],
    categorical: List[str],
    stats: PreprocessingStats,
    feature_names: List[str],
) -> np.ndarray:
    """
    Recompute the design matrix from RAW features with the frozen stats.

    The pipeline replicates Task 3 exactly: median imputation (frozen),
    WoE encoding with the global fallback (frozen), standardisation with
    the frozen means and standard deviations. Indicator columns pass
    through unmodified. The column ORDER follows feature_names, so the
    reconstructed matrix aligns with the model's design contract.

    Parameters
    ----------
    raw_frame : pd.DataFrame
        The raw-scale (cleaned) feature frame, record_id-indexed.
    continuous : List[str]
        The continuous feature names.
    categorical : List[str]
        The categorical feature names.
    stats : PreprocessingStats
        The frozen training statistics.
    feature_names : List[str]
        The model's design column order.

    Returns
    -------
    np.ndarray
        The reconstructed design matrix, shape (n, p).

    Raises
    ------
    ValueError
        If a feature name is not classifiable into the schema lists.

    """
    # Work on an untouched copy: the original frame is never mutated
    working: pd.DataFrame = raw_frame.copy()
    # Step 1: median imputation (frozen medians only)
    for column, median_value in stats.medians.items():
        if column in working.columns:
            working[column] = working[column].fillna(median_value)
    # Step 2: WoE encoding (frozen maps and global fallback)
    for column in categorical:
        if column in working.columns:
            mapped: np.ndarray = working[column].map(
                stats.woe_maps.get(column, {})
            ).to_numpy(dtype=object)
            # Unseen categories receive the global WoE, as in Task 3
            unknown_mask: Optional[np.ndarray] = None
            unknown_mask = pd.isna(pd.Series(mapped)).to_numpy()
            mapped[unknown_mask] = stats.global_woe[column]
            working[column] = mapped.astype(np.float64)
    # Step 3: standardisation with frozen statistics
    for column in continuous:
        if column in working.columns:
            working[column] = (
                pd.to_numeric(working[column], errors="raise").to_numpy(
                    dtype=np.float64
                ) - stats.means[column]
            ) / stats.stds[column]
    # Step 4: assemble in the model's exact column order
    columns: List[np.ndarray] = []
    for name in feature_names:
        if name not in working.columns:
            raise ValueError(
                f"feature '{name}' is missing from the raw frame; the "
                f"repreparation cannot align with the model design."
            )
        columns.append(working[name].to_numpy(dtype=np.float64))
    X: np.ndarray = np.column_stack(columns)
    if not np.isfinite(X).all():
        raise ValueError(
            "reprepared design matrix carries non-finite values."
        )
    return X

def compute_cser(
    model: torch.nn.Module,
    model_class: str,
    raw_frame: pd.DataFrame,
    continuous: List[str],
    categorical: List[str],
    stats: PreprocessingStats,
    feature_names: List[str],
    beta: np.ndarray,
    delta: float = DEFAULT_CSER_DELTA,
    decision_threshold: float = DEFAULT_DECISION_THRESHOLD,
    perturbation_space: str = CSER_PERTURBATION_SPACE,
    branch_mode: str = CSER_BRANCH_MODE,
    return_eligibility: bool = False,
) -> Dict[str, float]:
    """
    Compute the feature-level counterfactual sign error rates.

    Equation (14), manuscript Section 3. For each structured feature j whose
    fitted coefficient is non-zero, let d_j = sign(beta_j) and let delta be
    a prespecified perturbation size. The perturbed design row is

        x_i^(j,+) = x_i + d_j * delta * e_j,

    where e_j is the j-th unit vector IN THE SPACE OF THE DESIGN MATRIX ROW,
    i.e. the standardised, WoE-encoded space the model actually consumes.
    Then

        CSER_j = (1/n) sum_i 1{ eta(x_i^(j,+), z_i) < eta(x_i, z_i) },

    the share of observations for which a perturbation the structured
    coefficient says should RAISE default log-odds in fact LOWERS the
    integrated model's logit.

    B-3 remediation, five defects repaired.

    (1) PERTURBATION SPACE. delta was previously applied on the RAW scale
        and the frozen preprocessing re-run, making the effective
        perturbation delta / sd_j in model space — for a loan amount with a
        training standard deviation of several thousand currency units, a
        raw delta of 0.1 is about 2e-5 of a standard deviation. In single
        precision the strict inequality was then decided by floating-point
        rounding rather than by the model's directional response. delta now
        acts on the STANDARDISED column, so 0.1 means one tenth of a
        training standard deviation for every feature uniformly, which is
        the only reading under which a single prespecified delta is
        meaningful. The raw-scale variant remains available as a clearly
        named alternative for sensitivity work.
    (2) PRECISION. Both logit vectors are evaluated in float64. The
        comparison is a strict inequality between two nearly equal numbers.
    (3) SENTINELS. Ineligible features are no longer encoded as 0.0, which
        was indistinguishable from a genuine CSER of zero and pinned
        CSER_min to zero for every real dataset. An explicit eligibility
        mask is returned instead.
    (4) RESTRICTION TO CONTINUOUS FEATURES. A WoE-encoded categorical has no
        natural unit displacement, so it is excluded — but this is a
        registered methodological decision, reported in the eligibility
        mask, not a silent narrowing.
    (5) INPUT MUTATION. The model's guard and train/eval mode are restored
        by the scoring context.

    Parameters
    ----------
    model : torch.nn.Module
        The fitted model (findr or LR).
    model_class : str
        One of LR, NN, findr.
    raw_frame : pd.DataFrame
        The raw-scale cleaned feature frame, record_id-indexed.
    continuous : List[str]
        The continuous feature names.
    categorical : List[str]
        The categorical feature names.
    stats : PreprocessingStats
        The frozen training statistics.
    feature_names : List[str]
        The model's design column order.
    beta : np.ndarray
        The structured coefficients aligned to feature_names, shape (p,).
    delta : float, optional
        The perturbation size, by default 0.1.
    decision_threshold : float, optional
        Retained for signature compatibility; CSER compares logits directly
        and does not threshold.
    perturbation_space : str, optional
        "standardised" (default, Equation (14)) or "raw" (sensitivity
        variant applying delta before preprocessing).
    branch_mode : str, optional
        "total" (default) lets the perturbation flow through BOTH the
        structured and the neural branch, since Z = X in this study and the
        manuscript asks the counterfactual to reflect the full model's
        directional response. "structured_only" holds z at its unperturbed
        value, matching the literal (x^(j,+), z_i) notation.
    return_eligibility : bool, optional
        When True the returned mapping additionally carries the reserved
        key "__eligible__" holding the sorted list of eligible feature
        names. Default False preserves the historical return contract.

    Returns
    -------
    Dict[str, float]
        Per-feature CSER values for the ELIGIBLE features only. Ineligible
        features are absent from the mapping, never zero-filled.

    Raises
    ------
    ValueError
        If delta is non-positive, the coefficient vector misaligns, or an
        option is outside its domain.

    """
    # The perturbation size must be strictly positive
    if delta <= 0.0:
        raise ValueError(f"delta must be positive; received {delta}.")
    # The option domains are fixed
    if perturbation_space not in ("standardised", "raw"):
        raise ValueError(
            f"perturbation_space must be 'standardised' or 'raw'; received "
            f"'{perturbation_space}'."
        )
    if branch_mode not in ("total", "structured_only"):
        raise ValueError(
            f"branch_mode must be 'total' or 'structured_only'; received "
            f"'{branch_mode}'."
        )
    # The coefficient vector must align with the model's design order
    if beta.shape[0] != len(feature_names):
        raise ValueError(
            f"beta carries {beta.shape[0]} entries against "
            f"{len(feature_names)} design columns; the vectors must align."
        )
    # The baseline design from the frozen statistics (single source)
    X_base: np.ndarray = reprepare_features(
        raw_frame, continuous, categorical, stats, feature_names
    )
    # B-3(2): float64 throughout; the comparison is a strict inequality
    # between two nearly equal numbers and float32 is not adequate for it
    X_base_64: np.ndarray = X_base.astype(np.float64)
    base_tensor: torch.Tensor = torch.from_numpy(X_base_64)
    # B-3(4): the eligible set is {continuous features with beta_j != 0}
    eligible: List[str] = [
        name for index, name in enumerate(feature_names)
        if name in continuous and np.isfinite(beta[index])
        and beta[index] != 0.0
    ]
    per_feature: Dict[str, float] = {}
    # B-3(5): the restoring scoring context; no attribute is left mutated
    with scoring_context(model):
        # The model is promoted to float64 for the duration of the
        # diagnostic so both logit vectors carry full precision
        original_dtype: torch.dtype = next(model.parameters()).dtype
        model.double()
        try:
            with torch.no_grad():
                # The baseline logits, one forward for the whole block
                eta_base: np.ndarray = forward_scores(
                    model, base_tensor, is_training_population=False
                ).detach().numpy()
            for index, feature in enumerate(feature_names):
                # Only eligible features carry a signed direction
                if feature not in eligible:
                    continue
                # Equation: d_j = sign(beta_j), the risk direction the
                # structured coefficient asserts
                direction: float = float(np.sign(beta[index]))
                if perturbation_space == "standardised":
                    # B-3(1): Equation (14) perturbs the DESIGN row, so the
                    # displacement is applied directly to the standardised
                    # column: x^(j,+) = x + d_j * delta * e_j
                    X_pert: np.ndarray = X_base_64.copy()
                    X_pert[:, index] = X_pert[:, index] + direction * delta
                else:
                    # The registered raw-scale sensitivity variant: perturb
                    # before preprocessing and replay the frozen statistics
                    perturbed_frame: pd.DataFrame = raw_frame.copy()
                    perturbed_frame[feature] = pd.to_numeric(
                        perturbed_frame[feature], errors="raise"
                    ).to_numpy(dtype=np.float64) + direction * delta
                    X_pert = reprepare_features(
                        perturbed_frame, continuous, categorical, stats,
                        feature_names,
                    ).astype(np.float64)
                if branch_mode == "structured_only" and hasattr(
                        model, "forward_decomposed"):
                    # The literal (x^(j,+), z_i) reading: the structured
                    # branch sees the perturbed row while the neural branch
                    # keeps its unperturbed input
                    with torch.no_grad():
                        eta_str_p, _, _ = model.forward_decomposed(
                            torch.from_numpy(X_pert),
                            is_training_population=False,
                        )
                        _, eta_unstr_b, _ = model.forward_decomposed(
                            base_tensor, is_training_population=False,
                        )
                        eta_pert: np.ndarray = (
                            eta_str_p + eta_unstr_b
                        ).detach().numpy()
                else:
                    # The total-response reading (default): Z = X, so the
                    # perturbation reaches both branches
                    with torch.no_grad():
                        eta_pert = forward_scores(
                            model, torch.from_numpy(X_pert),
                            is_training_population=False,
                        ).detach().numpy()
                # Equation: CSER_j = (1/n) sum 1{eta(perturbed) < eta(base)}
                per_feature[feature] = float(np.mean(eta_pert < eta_base))
        finally:
            # Restore the model's entry dtype whatever happened above
            model.to(original_dtype)
    # B-3(3): the eligibility set travels WITH the values when requested,
    # so no downstream consumer has to guess which entries are real
    if return_eligibility:
        result: Dict[str, Any] = dict(per_feature)
        result["__eligible__"] = sorted(eligible)
        return result
    return per_feature

def summarise_cser(
    per_feature_cser: Dict[str, float],
    eligible_features: Optional[Sequence[str]] = None,
) -> Tuple[float, float]:
    """
    Compute CSER_min and CSER_max over the ELIGIBLE features only.

    Equation (manuscript Section 3):

        CSER_min = min over {j : beta_j != 0} of CSER_j,
        CSER_max = max over the SAME set.

    The restriction to non-zero coefficients is not decorative: a feature
    with a zero coefficient has no signed direction, so a directional
    perturbation along it is undefined and its CSER carries no information.

    B-2 remediation. The previous implementation reduced over EVERY value in
    the supplied mapping, including the 0.0 sentinels the upstream routine
    wrote for categorical features and zero-coefficient features. Because
    every real credit dataset in this study carries at least one categorical
    feature, CSER_min was pinned to exactly 0.0 unconditionally. That is the
    most dangerous class of defect in the pipeline, because it does not fail
    loudly: the manuscript's diagnostic tables report CSER (min) of 0.000
    across every fairness multiplier, so the bug MANUFACTURED apparent
    agreement with the published result while the statistic was never
    computed. A reviewer checking reproduction against that table would have
    seen a match.

    Two further corrections: an empty eligible set now returns NaN rather
    than 0.0 — returning zero for "no feature was eligible" is
    indistinguishable from "every eligible feature had perfect sign
    consistency", which are opposite conclusions — and a genuine CSER of
    exactly 0.0 is retained, because filtering on the VALUE rather than on
    eligibility would discard a valid and important result.

    Parameters
    ----------
    per_feature_cser : Dict[str, float]
        The per-feature CSER mapping. When it carries the reserved
        "__eligible__" key produced by compute_cser, that key defines the
        eligible set and is excluded from the reduction.
    eligible_features : Optional[Sequence[str]], optional
        An explicit eligible set, overriding any embedded key. When both
        are absent every non-reserved entry is treated as eligible, which
        is correct for the remediated compute_cser because it no longer
        emits sentinels.

    Returns
    -------
    Tuple[float, float]
        (CSER_min, CSER_max) over the eligible set, or (NaN, NaN) when that
        set is empty.

    Raises
    ------
    ValueError
        If a named eligible feature is absent from the mapping, or a value
        lies outside [0, 1].

    """
    # The reserved key carries provenance, never a measurement
    reserved: str = "__eligible__"
    # Resolve the eligible set by explicit argument, then embedded key,
    # then by treating every real entry as eligible
    if eligible_features is not None:
        resolved: List[str] = list(eligible_features)
    elif reserved in per_feature_cser:
        embedded: Any = per_feature_cser[reserved]
        resolved = list(embedded) if isinstance(
            embedded, (list, tuple, set)) else []
    else:
        resolved = [k for k in per_feature_cser if k != reserved]
    # A named eligible feature MUST have a measurement; a silent skip here
    # would reintroduce exactly the ambiguity this remediation removes
    missing: List[str] = [
        name for name in resolved if name not in per_feature_cser
    ]
    if missing:
        raise ValueError(
            f"eligible feature(s) {sorted(missing)} carry no CSER value; the "
            f"eligibility set and the value mapping disagree."
        )
    # B-2: reduce over the ELIGIBLE values only, never over the whole map
    values: List[float] = [
        float(per_feature_cser[name]) for name in resolved
    ]
    # An empty eligible set is reported as NaN, not as zero
    if not values:
        logger.warning(
            "CSER summary: the eligible set {j : beta_j != 0} is empty, so "
            "CSER_min and CSER_max are undefined and reported as NaN."
        )
        return float("nan"), float("nan")
    # Every CSER is a proportion and must lie in the unit interval
    array: np.ndarray = np.asarray(values, dtype=np.float64)
    if not np.all(np.isfinite(array)) or array.min() < 0.0 or \
            array.max() > 1.0:
        raise ValueError(
            f"CSER values must be finite proportions in [0, 1]; received "
            f"min {float(np.nanmin(array))}, max {float(np.nanmax(array))}."
        )
    # Equations: CSER_min and CSER_max over {j : beta_j != 0}
    return float(array.min()), float(array.max())

# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def compute_diagnostics(
    model: torch.nn.Module,
    model_class: str,
    raw_frame: pd.DataFrame,
    continuous: List[str],
    categorical: List[str],
    stats: PreprocessingStats,
    feature_names: List[str],
    delta: float = DEFAULT_CSER_DELTA,
    decision_threshold: float = DEFAULT_DECISION_THRESHOLD,
    split: str = "validation",
) -> DiagnosticsReport:
    """
    Compute EVR, DDR and CSER for one fitted model.

    Equations (manuscript Section 3):
      EVR  = Var_n(eta_str) / Var_n(eta), population variance (ddof = 0),
      DDR  = (1/n) sum 1[ 1{eta_str >= 0} != 1{eta >= 0} ],
      CSER = per-feature counterfactual sign error rates, summarised as the
             min and max over {j : beta_j != 0}.

    B-21 remediation: this orchestrator had no defect of its own but
    propagated the two CSER defects. It now (i) requests the explicit
    eligibility set from compute_cser, (ii) passes that set to
    summarise_cser so the reduction cannot be contaminated by ineligible
    features, (iii) records the eligible count so a reader can see how many
    features the statistic is actually over, and (iv) uses the restoring
    scoring context instead of mutating the model's guard.

    Parameters
    ----------
    model : torch.nn.Module
        The fitted findr (or LR) model.
    model_class : str
        One of LR, NN, findr.
    raw_frame : pd.DataFrame
        The raw-scale cleaned feature frame of the evaluation split.
    continuous : List[str]
        The continuous feature names.
    categorical : List[str]
        The categorical feature names.
    stats : PreprocessingStats
        The frozen training statistics.
    feature_names : List[str]
        The model's design column order.
    delta : float, optional
        The CSER perturbation size, by default 0.1.
    decision_threshold : float, optional
        The DDR logit threshold, by default 0.0.
    split : str, optional
        The evaluation split label.

    Returns
    -------
    DiagnosticsReport
        The frozen diagnostics report, whose per_feature_cser mapping
        carries the eligible-feature roster under "__eligible__".

    Raises
    ------
    ValueError
        If the model class carries no structured branch.

    """
    # A structured branch is required for the decomposition diagnostics
    if model_class not in ("findr", "LR"):
        raise ValueError(
            f"diagnostics require a structured branch; the model class "
            f"'{model_class}' has none."
        )
    # The design from the frozen statistics, once (single source)
    X_design: np.ndarray = reprepare_features(
        raw_frame, continuous, categorical, stats, feature_names
    )
    # B-21(iv): the restoring scoring context replaces the guard mutation
    with scoring_context(model):
        with torch.no_grad():
            if model_class == "findr":
                # The decomposed forward gives the two logit axes directly
                X_tensor: torch.Tensor = torch.from_numpy(
                    X_design.astype(np.float32)
                )
                eta_str_t, eta_unstr_t, _ = model.forward_decomposed(
                    X_tensor, is_training_population=False
                )
                eta_str: np.ndarray = eta_str_t.detach().numpy()
                eta_unstr: np.ndarray = eta_unstr_t.detach().numpy()
            else:
                # LR: the structured logit IS the full logit, so the
                # residual axis is identically zero (documented degenerate
                # case, and EVR is therefore exactly 1 by construction)
                eta_total: np.ndarray = forward_scores(
                    model, torch.from_numpy(X_design.astype(np.float32))
                ).detach().numpy()
                eta_str = eta_total
                eta_unstr = np.zeros_like(eta_total)
    # Step 1: Equation (12), EVR with population variance (ddof = 0)
    evr: float = compute_evr(eta_str, eta_unstr, split)
    # Step 2: Equation (13), DDR as a rate and a raw count
    ddr_rate, ddr_count = compute_ddr(
        eta_str, eta_str + eta_unstr, decision_threshold
    )
    # Step 3: Equation (14), CSER over the eligible features
    beta: np.ndarray = extract_structured_beta(model, model_class)
    # B-21(i): request the explicit eligibility roster alongside the values
    per_feature: Dict[str, float] = compute_cser(
        model, model_class, raw_frame, continuous, categorical, stats,
        feature_names, beta, delta, decision_threshold,
        return_eligibility=True,
    )
    # B-21(ii): the roster drives the reduction, so an ineligible feature
    # can never contaminate CSER_min again
    eligible_roster: List[str] = list(per_feature.get("__eligible__", []))
    cser_min, cser_max = summarise_cser(
        per_feature, eligible_features=eligible_roster
    )
    # B-21(iii): the eligible count is recorded so the reader can see the
    # support of the statistic rather than inferring it
    logger.info(
        f"diagnostics ({split}): EVR={evr:.6f}, DDR={ddr_rate:.6f} "
        f"({ddr_count} rows), CSER over {len(eligible_roster)} eligible "
        f"feature(s) -> min={cser_min:.6f}, max={cser_max:.6f}."
    )
    return DiagnosticsReport(
        evr=evr, ddr=ddr_rate, ddr_count=ddr_count,
        cser_min=cser_min, cser_max=cser_max,
        per_feature_cser=per_feature, split=split,
    )
````

````python
# Task 10: Implement Score-Level Accuracy–Fairness Frontier Construction

# ==============================================================================
# Task 10: Implement Score-Level Accuracy–Fairness Frontier Construction
# ==============================================================================

# Make the sibling task modules importable as the study pipeline grows

# Configure the module-level logger for stage diagnostics

# The configuration defaults for the frontier (mirrored from the config
# dict; the orchestrator always passes the configured values)
DEFAULT_RHO: float = 0.1
DEFAULT_EPSILON: float = 1e-6
DEFAULT_MAX_ITERATIONS: int = 1000
DEFAULT_CONVERGENCE_TOLERANCE: float = 1e-6
# The periodic candidate-collection stride
CANDIDATE_STRIDE: int = 10
# The fraction of the path averaged by the Polyak-Ruppert rule
PR_FRACTION: float = 0.5

@dataclass(frozen=True)
class FrontierReport:
    """
    The score-level frontier and the accuracy gaps.

    Attributes
    ----------
    envelope_phi : np.ndarray
        The ascending fairness values of the envelope.
    envelope_logloss : np.ndarray
        The running-minimum log losses of the envelope.
    accuracy_gaps : Dict[str, float]
        Per fitted-model accuracy gap.
    candidate_count : int
        The number of candidate score vectors evaluated.

    """

    # The ascending fairness axis
    envelope_phi: np.ndarray
    # The running-minimum log-loss axis
    envelope_logloss: np.ndarray
    # Per-model accuracy gaps
    accuracy_gaps: Dict[str, float]
    # The candidate pool size
    candidate_count: int

# -------------------------------------------------------------------------------------------------------------------------------
# Task 10, Step 1: The scalarised objective and the multiplier grid
# -------------------------------------------------------------------------------------------------------------------------------
def build_xi_grid(
    include_zero: bool = True,
    logspace_start: float = -2.0,
    logspace_end: float = 2.0,
    num_points: int = 40,
) -> np.ndarray:
    """
    Construct the frontier multiplier grid.

    Grid: xi in {0} U logspace(start, end, num_points), sorted ascending.

    Parameters
    ----------
    include_zero : bool, optional
        Whether to prepend 0.
    logspace_start : float, optional
        The log10 lower exponent, by default -2.
    logspace_end : float, optional
        The log10 upper exponent, by default 2.
    num_points : int, optional
        The logarithmic multiplicity, by default 40.

    Returns
    -------
    np.ndarray
        The sorted, strictly increasing multiplier grid.

    Raises
    ------
    ValueError
        If the exponents are inverted.

    """
    if logspace_start >= logspace_end:
        raise ValueError(
            f"logspace_start ({logspace_start}) must be below "
            f"logspace_end ({logspace_end})."
        )
    if num_points < 1:
        raise ValueError(f"num_points must be >= 1; received {num_points}.")
    # The logarithmic block
    log_block: np.ndarray = np.logspace(
        logspace_start, logspace_end, num_points
    )
    if include_zero:
        # The grid opens at zero: the closed-form solution is s = r
        grid: np.ndarray = np.concatenate([[0.0], log_block])
    else:
        grid = log_block
    # The final grid is sorted and its zero-mass duplicates dropped
    grid = np.unique(np.sort(grid))
    return grid

def scalarised_objective(
    scores: np.ndarray,
    reference_risk: np.ndarray,
    y_true: np.ndarray,
    a_true: np.ndarray,
    xi: float,
    criterion: str = "equalised_odds",
    clip_epsilon: float = DEFAULT_EPSILON,
) -> float:
    """
    The scalarised objective Q_xi(s) = R_hat_log(r,s) + xi * Phi(s).

    Parameters
    ----------
    scores : np.ndarray
        The candidate score vector, shape (n,).
    reference_risk : np.ndarray
        The reference risk, shape (n,).
    y_true, a_true : np.ndarray
        The labels and sensitive attributes, shape (n,).
    xi : float
        The frontier multiplier.
    criterion : str, optional
        The fairness criterion, by default equalised_odds.
    clip_epsilon : float, optional
        The score clip floor, by default 1e-6.

    Returns
    -------
    float
        The scalar objective value.

    """
    # The prediction term: the expected log loss under r
    log_term: float = expected_log_loss(reference_risk, scores)
    # At xi = 0 the penalty contributes nothing
    if xi == 0.0:
        return log_term
    # The fairness term: the exact W1 penalty of the score distributions
    clipped: np.ndarray = clip_scores(scores, clip_epsilon)
    phi_value: float = float(fairness_penalty(
        torch.from_numpy(clipped.astype(np.float64)),
        torch.from_numpy(y_true.astype(np.float64)),
        torch.from_numpy(a_true.astype(np.float64)),
        criterion,
    ))
    # Equation: Q_xi(s) = R_hat_log + xi * Phi
    return log_term + xi * phi_value

# -------------------------------------------------------------------------------------------------------------------------------
# Task 10, Step 2: The soft-cell solver
# -------------------------------------------------------------------------------------------------------------------------------
def compute_soft_memberships(
    reference_risk: np.ndarray,
    a_true: np.ndarray,
) -> Tuple[np.ndarray, np.ndarray]:
    """
    Compute the soft-cell memberships pi_ug and their mean weights.

    Equation (from Appendix A): pi_ug(x_i) = 1{a_i = g} [ r_i 1{u = 1} +
    (1 - r_i) 1{u = 0} ], p_ug = (1/n) sum_i pi_ug(x_i).

    Parameters
    ----------
    reference_risk : np.ndarray
        The reference risks, shape (n,).
    a_true : np.ndarray
        The sensitive attributes, shape (n,).

    Returns
    -------
    Tuple[np.ndarray, np.ndarray]
        pi (n, 2, 2) indexed [i, u, g] and p (2, 2).

    Raises
    ------
    ValueError
        If a soft cell has zero total membership.

    """
    n: int = len(reference_risk)
    # The two outcome strata u = 0, and u = 1, per group g
    outcome_prob: np.ndarray = np.stack(
        [1.0 - reference_risk, reference_risk], axis=1
    )
    # The group indicator: 1 if a_i == g, broadcast over n x 2
    group_indicator: np.ndarray = np.stack(
        [(a_true == 0).astype(np.float64),
         (a_true == 1).astype(np.float64)], axis=1
    )
    # Equation: pi_ug(x_i) = 1{a_i = g} * r-based outcome probability
    pi: np.ndarray = np.einsum("iu,ig->iug", outcome_prob, group_indicator)
    # Equation: p_ug = (1/n) sum_i pi_ug
    p_ug: np.ndarray = pi.mean(axis=0)
    # An entirely absent soft cell makes the sign ratio undefined
    if np.any(p_ug <= 0.0):
        raise ValueError(
            "a soft cell carries zero total membership; the solver "
            "cannot form the sign ratio (pi_ug / p_ug)."
        )
    return pi, p_ug

def weighted_cdf_matrix(
    scores: np.ndarray, pi: np.ndarray, p_ug: np.ndarray,
) -> np.ndarray:
    """
    Evaluate F_ug(s_i; s) for every observation and soft cell.

    Equation: F_ug(v; s) = (1/n) sum_j pi_ug(x_j) 1{s_j <= v} / p_ug;
    the weighted empirical CDF at every observed score, with ties
    included (the <= convention).

    Parameters
    ----------
    scores : np.ndarray
        The current score vector, shape (n,).
    pi : np.ndarray
        The soft memberships, shape (n, 2, 2).
    p_ug : np.ndarray
        The cell weights, shape (2, 2).

    Returns
    -------
    np.ndarray
        F (n, 2, 2) with F[i, u, g] = F_ug(s_i; s).

    """
    n: int = scores.shape[0]
    order: np.ndarray = np.argsort(scores, kind="stable")
    sorted_scores: np.ndarray = scores[order]
    # The cumulative mass per cell over the sorted order
    pi_sorted: np.ndarray = pi[order]
    cumsum_sorted: np.ndarray = np.cumsum(pi_sorted, axis=0)
    # Ties include EVERY observation at the same value: the cumsum index
    # is the last position of the tied run of each score
    last_positions: np.ndarray = np.searchsorted(
        sorted_scores, scores, side="right"
    ) - 1
    # F_ug(s_i) = cumsum[last_position_i, u, g] / (n * p_ug)
    f_values: np.ndarray = cumsum_sorted[last_positions] / (
        n * p_ug[None, :, :]
    )
    return f_values

def effective_signs(
    f_matrix: np.ndarray, pi: np.ndarray, p_ug: np.ndarray,
) -> np.ndarray:
    """
    Compute the effective sign vector Theta.

    Equation (15): Theta_i = sum_{u,g} (pi_ug(x_i)/p_ug) *
    sign{F_ug(s_i; s) - F_{u, 1-g}(s_i; s)}.

    Parameters
    ----------
    f_matrix : np.ndarray
        The CDF matrix, shape (n, 2, 2).
    pi : np.ndarray
        The soft memberships, shape (n, 2, 2).
    p_ug : np.ndarray
        The cell weights, shape (2, 2).

    Returns
    -------
    np.ndarray
        The effective sign vector, shape (n,).

    """
    # The matched comparison: for cell (u, g), the opposing group's CDF
    # is (u, 1-g): flip the group axis
    opposing: np.ndarray = f_matrix[:, :, ::-1]
    # Equation: sign{F_ug(s_i) - F_{u,1-g}(s_i)}
    signed: np.ndarray = np.sign(f_matrix - opposing)
    # Equation: weight by pi_ug(x_i)/p_ug and sum over the four cells
    ratios: np.ndarray = pi / p_ug[None, :, :]
    theta: np.ndarray = (ratios * signed).sum(axis=(1, 2))
    return theta

def select_initial_candidate(
    candidates: Sequence[np.ndarray],
    reference_risk: np.ndarray,
    y_true: np.ndarray,
    a_true: np.ndarray,
    xi: float,
    criterion: str,
    clip_epsilon: float,
) -> Tuple[np.ndarray, float, int]:
    """
    Choose the warm start with the lowest scalarised objective.

    Task 10 Step 2 requires each xi solve to be initialised from the
    candidate pool containing the reference risk r, the fitted model
    scores, and the solution at the previous multiplier. Because the update
    is a sign-based non-convex heuristic, the starting point materially
    determines where it lands; initialising from a single trajectory is
    exactly the failure mode the pool exists to prevent.

    Parameters
    ----------
    candidates : Sequence[np.ndarray]
        The candidate warm starts, each shape (n,).
    reference_risk : np.ndarray
        The reference risks r, shape (n,).
    y_true, a_true : np.ndarray
        The labels and sensitive attributes, shape (n,).
    xi : float
        The frontier multiplier.
    criterion : str
        The fairness criterion.
    clip_epsilon : float
        The simplex clipping floor.

    Returns
    -------
    Tuple[np.ndarray, float, int]
        The selected (clipped) candidate, its objective, and its index in
        the supplied sequence.

    Raises
    ------
    ValueError
        If the pool is empty or a candidate misaligns with r.

    """
    # An empty pool leaves the solve with no starting point
    if len(candidates) == 0:
        raise ValueError("the initialisation pool is empty.")
    best_index: int = -1
    best_objective: float = float("inf")
    best_vector: Optional[np.ndarray] = None
    for index, candidate in enumerate(candidates):
        # Every candidate must describe the same population as r
        vector: np.ndarray = np.asarray(candidate, dtype=np.float64)
        if vector.shape != reference_risk.shape:
            raise ValueError(
                f"initial candidate {index} has shape {vector.shape} against "
                f"the reference risk shape {reference_risk.shape}."
            )
        # The candidate must be feasible before it is scored
        clipped: np.ndarray = clip_scores(vector, clip_epsilon)
        # Equation: Q_xi(s) = R_log(r, s) + xi * Phi(s; y, a)
        objective: float = scalarised_objective(
            clipped, reference_risk, y_true, a_true, xi, criterion,
            clip_epsilon,
        )
        # Retain the pool minimiser
        if objective < best_objective:
            best_objective = objective
            best_index = index
            best_vector = clipped
    # A pool of non-finite objectives leaves nothing to start from
    if best_vector is None or not np.isfinite(best_objective):
        raise ValueError(
            "no initialisation candidate produced a finite objective."
        )
    return best_vector, best_objective, best_index


def solve_scalarised_frontier(
    s_initial: Union[np.ndarray, Sequence[np.ndarray]],
    reference_risk: np.ndarray,
    y_true: np.ndarray,
    a_true: np.ndarray,
    xi: float,
    rho: float = DEFAULT_RHO,
    clip_epsilon: float = DEFAULT_EPSILON,
    max_iterations: int = DEFAULT_MAX_ITERATIONS,
    convergence_tolerance: float = DEFAULT_CONVERGENCE_TOLERANCE,
    use_polyak_ruppert: bool = True,
    criterion: str = "equalised_odds",
    pr_fraction: float = PR_AVERAGING_FRACTION,
) -> Dict[str, Any]:
    """
    Solve the scalarised frontier problem for one multiplier.

    The soft-cell sign update (Appendix A, Equation (16)):

        s_i^(t+1) = (1 - rho_xi) s_i^(t)
                    + rho_xi { r_i + xi s_i^(t) (1 - s_i^(t)) Theta_i^(t) },
        rho_xi    = rho / (1 + xi),   rho = 0.1,

    with every update clipped into [epsilon, 1 - epsilon], and the
    Polyak-Ruppert average taken over the SECOND HALF OF THE PATH:

        s_bar = (1 / (T - floor(T/2))) * sum_{t = floor(T/2)}^{T} s^(t).

    B-7 remediation, two omissions repaired.

    (1) INITIALISATION POOL. The function previously accepted a single
        starting vector and the orchestrator supplied only the previous
        multiplier's solution; r was used only at xi = 0 and the fitted
        model scores never initialised anything. It now accepts either a
        single vector (backward compatible) or a SEQUENCE of candidates,
        evaluates Q_xi at each, and begins from the pool minimiser. The
        chosen warm start is retained in the returned record regardless of
        the outcome, so a drifting solve can never make the envelope worse
        than its own starting point.

    (2) POLYAK-RUPPERT SET. The average previously ran over the second half
        of a STRIDE-10 SUBSAMPLE — roughly one twentieth of the path rather
        than one half — and silently degenerated to the last raw iterate
        when the solve converged in under twenty iterations. Since the
        entire purpose of the averaging is to suppress the high-frequency
        oscillation intrinsic to a sign-based update, sampling at stride ten
        discarded precisely what the average exists to cancel. The average
        now runs over the FULL second half, accumulated by a running sum so
        the memory cost stays O(n) rather than O(n T).

    Parameters
    ----------
    s_initial : Union[np.ndarray, Sequence[np.ndarray]]
        A single warm start, or the candidate pool to minimise over.
    reference_risk : np.ndarray
        The reference risks r, shape (n,).
    y_true, a_true : np.ndarray
        The labels and attributes, shape (n,).
    xi : float
        The frontier multiplier.
    rho : float, optional
        The base step size, by default 0.1.
    clip_epsilon : float, optional
        The simplex clipping floor, by default 1e-6.
    max_iterations : int, optional
        The iteration cap, by default 1000.
    convergence_tolerance : float, optional
        The objective-change tolerance, by default 1e-6.
    use_polyak_ruppert : bool, optional
        Whether to average the second half of the path.
    criterion : str, optional
        The fairness criterion, by default equalised_odds.
    pr_fraction : float, optional
        The fraction of the path averaged, by default 0.5 (second half).

    Returns
    -------
    Dict[str, Any]
        solution, pr_solution, final_objective, initial_objective,
        iterations, converged, warm_start, and pool diagnostics.

    Raises
    ------
    ValueError
        If rho is non-positive or the averaging fraction is invalid.

    """
    # The step size and the averaging fraction must be well posed
    if rho <= 0.0:
        raise ValueError(f"rho must be positive; received {rho}.")
    if not 0.0 < pr_fraction < 1.0:
        raise ValueError(
            f"pr_fraction must lie in (0, 1); received {pr_fraction}."
        )
    # The soft memberships are static per (r, a) and computed once
    pi, p_ug = compute_soft_memberships(reference_risk, a_true)
    # Equation: rho_xi = rho / (1 + xi)
    rho_xi: float = rho / (1.0 + xi)
    # B-7(1): normalise the argument to a pool, so a bare vector and a
    # sequence of candidates take the identical code path
    if isinstance(s_initial, np.ndarray) and s_initial.ndim == 1:
        pool: List[np.ndarray] = [s_initial]
    else:
        pool = [np.asarray(candidate, dtype=np.float64)
                for candidate in s_initial]
    # Begin from the pool minimiser of Q_xi
    s, initial_objective, chosen_index = select_initial_candidate(
        pool, reference_risk, y_true, a_true, xi, criterion, clip_epsilon,
    )
    # The warm start is retained verbatim as an envelope candidate
    warm_start: np.ndarray = s.copy()
    history: List[float] = []
    iterates: List[np.ndarray] = []
    iterate_objectives: List[float] = []
    # B-7(2): the running accumulator for the FULL second half. The path
    # length is unknown until the loop ends, so every iterate is buffered
    # and the average is taken over the true second half afterwards; the
    # buffer holds at most max_iterations vectors of length n
    path: List[np.ndarray] = []
    converged: bool = False
    objective_previous: float = initial_objective
    iteration: int = 0
    for iteration in range(max_iterations):
        # Equation: F_ug(s_i; s) evaluated at the CURRENT scores
        f_matrix: np.ndarray = weighted_cdf_matrix(s, pi, p_ug)
        # Equation (15): the effective sign vector Theta
        theta: np.ndarray = effective_signs(f_matrix, pi, p_ug)
        # Equation (16): the convex-combination update
        s_next: np.ndarray = (
            (1.0 - rho_xi) * s
            + rho_xi * (reference_risk
                        + xi * s * (1.0 - s) * theta)
        )
        # Clip after EVERY update: the feasible box is invariant
        s_next = clip_scores(s_next, clip_epsilon)
        # The realised-cell objective drives acceptance and convergence
        objective_now: float = scalarised_objective(
            s_next, reference_risk, y_true, a_true, xi, criterion,
            clip_epsilon,
        )
        # B-7(2): EVERY iterate enters the path buffer, not every tenth
        path.append(s_next.copy())
        # The strided candidate collection is a SEPARATE concern, retained
        # for the envelope pool and never conflated with the averaging set
        if iteration % CANDIDATE_STRIDE == 0:
            iterates.append(s_next.copy())
            iterate_objectives.append(objective_now)
        history.append(objective_now)
        # Convergence: the objective change drops below the tolerance
        if abs(objective_now - objective_previous) < convergence_tolerance:
            converged = True
        s = s_next
        objective_previous = objective_now
        if converged:
            break
    # B-7(2): the Polyak-Ruppert average over the FULL second half.
    # T = len(path); the average runs from floor(T * pr_fraction) to T - 1
    if use_polyak_ruppert and len(path) >= 2:
        start_index: int = int(math.floor(len(path) * pr_fraction))
        # At least one iterate must survive the split
        start_index = min(start_index, len(path) - 1)
        # Accumulate the sum in float64 and divide once, which is both
        # exact and O(n) in memory beyond the buffer itself
        accumulator: np.ndarray = np.zeros_like(path[0], dtype=np.float64)
        for vector in path[start_index:]:
            accumulator += vector
        pr_solution: np.ndarray = clip_scores(
            accumulator / float(len(path) - start_index), clip_epsilon
        )
    else:
        # A single-iterate path has no second half to average
        pr_solution = s.copy()
    # The best COLLECTED iterate by its own tracked objective
    if iterates:
        best_index: int = int(np.argmin(iterate_objectives))
        best_solution: np.ndarray = clip_scores(
            iterates[best_index], clip_epsilon
        )
    else:
        best_solution = s.copy()
    # The PR average competes with the periodic best on the objective, so
    # the returned solution is never worse than either
    pr_objective: float = scalarised_objective(
        pr_solution, reference_risk, y_true, a_true, xi, criterion,
        clip_epsilon,
    )
    best_objective: float = scalarised_objective(
        best_solution, reference_risk, y_true, a_true, xi, criterion,
        clip_epsilon,
    )
    if pr_objective < best_objective:
        best_solution, best_objective = pr_solution, pr_objective
    # The acceptance rule: a drifting heuristic is logged, and the warm
    # start remains a candidate regardless
    if best_objective > initial_objective + convergence_tolerance:
        logger.warning(
            f"xi={xi:.6f}: solver objective {best_objective:.8f} exceeds "
            f"the warm start {initial_objective:.8f}; both enter the "
            f"candidate pool."
        )
    return {
        "solution": best_solution,
        "pr_solution": pr_solution,
        "final_objective": best_objective,
        "initial_objective": initial_objective,
        "iterations": iteration + 1,
        "converged": converged,
        "warm_start": warm_start,
        "pool_size": len(pool),
        "pool_choice": chosen_index,
        "path_length": len(path),
        "pr_start_index": int(math.floor(len(path) * pr_fraction))
        if path else 0,
    }

def collect_candidates(
    solve_results: List[Dict[str, Any]],
    reference_risk: np.ndarray,
    y_true: np.ndarray,
    a_true: np.ndarray,
    criterion: str = "equalised_odds",
) -> List[Tuple[float, float, np.ndarray]]:
    """
    Collect (phi, logloss, score) candidates from every solve.

    Parameters
    ----------
    solve_results : List[Dict[str, Any]]
        The per-xi solve dictionaries.
    reference_risk, y_true, a_true : np.ndarray
        The validation data.
    criterion : str, optional
        The fairness criterion.

    Returns
    -------
    List[Tuple[float, float, np.ndarray]]
        The candidate records.

    """
    candidates: List[Tuple[float, float, np.ndarray]] = []
    seen: List[np.ndarray] = []
    for result in solve_results:
        # The warm start, the best iterate and the PR average per xi
        for scores in (
            np.asarray(result["warm_start"]),
            result["solution"],
            result["pr_solution"],
        ):
            # A near-duplicate score vector adds an identical point
            duplicates: bool = any(
                np.allclose(np.asarray(score), scores, atol=1e-9)
                for score in seen
            )
            if duplicates:
                continue
            seen.append(scores.copy())
            # The raw fairness of this candidate (xi-free)
            phi_only: float = float(fairness_penalty(
                torch.from_numpy(clip_scores(
                    scores, DEFAULT_EPSILON).astype(np.float64)),
                torch.from_numpy(y_true.astype(np.float64)),
                torch.from_numpy(a_true.astype(np.float64)),
                criterion,
            ))
            logloss: float = expected_log_loss(reference_risk, scores)
            candidates.append((phi_only, logloss, scores.copy()))
    # The reference itself is always a candidate
    r_candidate: np.ndarray = clip_scores(reference_risk, DEFAULT_EPSILON)
    candidates.append((
        float(fairness_penalty(
            torch.from_numpy(r_candidate.astype(np.float64)),
            torch.from_numpy(y_true.astype(np.float64)),
            torch.from_numpy(a_true.astype(np.float64)),
            criterion,
        )),
        expected_log_loss(reference_risk, r_candidate),
        r_candidate,
    ))
    return candidates

# -------------------------------------------------------------------------------------------------------------------------------
# Task 10, Step 3: Envelope extraction and the accuracy gap
# -------------------------------------------------------------------------------------------------------------------------------
def extract_frontier_envelope(
    candidates: List[Tuple[float, float, np.ndarray]],
) -> Tuple[np.ndarray, np.ndarray]:
    """
    Extract the monotone lower envelope of the candidate set.

    Sort by phi ascending, take the running minimum of the log loss,
    collapse duplicate phi values to the minimum log loss.

    Parameters
    ----------
    candidates : List[Tuple[float, float, np.ndarray]]
        The (phi, logloss, score) records.

    Returns
    -------
    Tuple[np.ndarray, np.ndarray]
        The ascending phi axis and the running-minimum log loss axis.

    """
    if not candidates:
        raise ValueError("no candidates; the envelope is undefined.")
    # Convert to arrays and sort by phi ascending
    phis: np.ndarray = np.array([c[0] for c in candidates])
    losses: np.ndarray = np.array([c[1] for c in candidates])
    order: np.ndarray = np.argsort(phis, kind="stable")
    phis = phis[order]
    losses = losses[order]
    # Collapse duplicate phi values to the minimum log loss
    unique_phis: np.ndarray = np.unique(phis)
    collapsed: List[float] = []
    for phi_value in unique_phis:
        coll: float = float(losses[phis == phi_value].min())
        collapsed.append(coll)
    phis = unique_phis
    losses = np.array(collapsed)
    # The running minimum: the lower envelope is monotone non-increasing
    envelope: np.ndarray = np.minimum.accumulate(losses)
    # The envelope must be monotone (check by construction)
    if not np.all(np.diff(envelope) <= 1e-12):
        raise ValueError("envelope is not monotone after running min.")
    return phis, envelope

def evaluate_frontier_at(
    phi_value: float, envelope_phi: np.ndarray, envelope_logloss: np.ndarray,
) -> float:
    """
    Evaluate the interpolated frontier at an arbitrary fairness value.

    Parameters
    ----------
    phi_value : float
        The fairness value of a fitted model.
    envelope_phi : np.ndarray
        The ascending envelope axis.
    envelope_logloss : np.ndarray
        The running-minimum envelope values.

    Returns
    -------
    float
        F(phi_value) with linear interpolation and endpoint clamping.

    """
    if phi_value <= envelope_phi[0]:
        # The left endpoint clamps: the benchmark is confined to the
        # evaluated range (a documented numerical benchmark rule)
        return float(envelope_logloss[0])
    if phi_value >= envelope_phi[-1]:
        return float(envelope_logloss[-1])
    # Linear interpolation between the bracketing envelope points
    right: int = int(np.searchsorted(envelope_phi, phi_value, side="right"))
    left: int = right - 1
    slope: float = (envelope_logloss[right] - envelope_logloss[left]) / (
        envelope_phi[right] - envelope_phi[left]
    )
    return float(
        envelope_logloss[left]
        + slope * (phi_value - envelope_phi[left])
    )

def compute_accuracy_gap(
    fitted_scores: np.ndarray,
    reference_risk: np.ndarray,
    y_true: np.ndarray,
    a_true: np.ndarray,
    envelope_phi: np.ndarray,
    envelope_logloss: np.ndarray,
    criterion: str = "equalised_odds",
) -> float:
    """
    Compute the accuracy gap of one fitted score vector.

    Equation (11): Delta(s_m) = R_hat_log(r, s_m) - F(Phi(s_m; y, a)).

    Parameters
    ----------
    fitted_scores : np.ndarray
        The fitted model's validation scores, shape (n,).
    reference_risk : np.ndarray
        The reference risks, shape (n,).
    y_true, a_true : np.ndarray
        The labels and attributes.
    envelope_phi : np.ndarray
        The envelope fairness axis.
    envelope_logloss : np.ndarray
        The envelope log-loss axis.
    criterion : str, optional
        The fairness criterion.

    Returns
    -------
    float
        The accuracy gap; smaller means closer to the benchmark.

    """
    # Equation: Phi(s_m; y, a)
    phi_model: float = float(fairness_penalty(
        torch.from_numpy(clip_scores(
            fitted_scores, DEFAULT_EPSILON).astype(np.float64)),
        torch.from_numpy(y_true.astype(np.float64)),
        torch.from_numpy(a_true.astype(np.float64)),
        criterion,
    ))
    # Equation: R_hat_log(r, s_m)
    logloss_model: float = expected_log_loss(reference_risk, fitted_scores)
    # Equation: Delta(s_m) = R_hat_log - F(Phi(s_m))
    frontier_value: float = evaluate_frontier_at(
        phi_model, envelope_phi, envelope_logloss
    )
    return logloss_model - frontier_value

# -------------------------------------------------------------------------------------------------------------------------------
# Task 10, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def normalised_chord_lengths(
    envelope_phi: np.ndarray, envelope_logloss: np.ndarray,
) -> np.ndarray:
    """
    Compute the range-normalised chord length between adjacent nodes.

    Task 10 refinement. The chord length must be a DIMENSIONLESS fraction of
    the envelope's own extent, because the two axes carry different units
    (a Wasserstein distance and a log loss). Each axis is therefore scaled
    to its observed range before the Euclidean chord is formed, so the
    configured threshold of 0.05 means "five percent of the envelope's
    diagonal extent" rather than an uninterpretable mixed-unit quantity.

    Parameters
    ----------
    envelope_phi : np.ndarray
        The ascending fairness axis of the envelope, shape (K,).
    envelope_logloss : np.ndarray
        The running-minimum log-loss axis, shape (K,).

    Returns
    -------
    np.ndarray
        The K-1 normalised chord lengths; empty when K < 2.

    """
    # Fewer than two nodes admit no chord
    if envelope_phi.shape[0] < 2:
        return np.zeros(0, dtype=np.float64)
    # The observed extent of each axis; a degenerate axis is floored at one
    # so the normalisation cannot divide by zero
    phi_range: float = float(envelope_phi.max() - envelope_phi.min())
    loss_range: float = float(envelope_logloss.max() - envelope_logloss.min())
    phi_scale: float = phi_range if phi_range > 0.0 else 1.0
    loss_scale: float = loss_range if loss_range > 0.0 else 1.0
    # The per-axis normalised steps between adjacent nodes
    d_phi: np.ndarray = np.diff(envelope_phi) / phi_scale
    d_loss: np.ndarray = np.diff(envelope_logloss) / loss_scale
    # The Euclidean chord in the normalised plane
    return np.sqrt(d_phi ** 2 + d_loss ** 2)


def construct_frontier(
    reference_risk: np.ndarray,
    y_true: np.ndarray,
    a_true: np.ndarray,
    fitted_scores: Optional[Dict[str, np.ndarray]] = None,
    frontier_config: Optional[Dict[str, Any]] = None,
    criterion: str = "equalised_odds",
) -> FrontierReport:
    """
    Construct the score-level accuracy-fairness frontier.

    Procedure (manuscript Section 3.5 and the frontier configuration).
    Solve
        min_s  Q_xi(s) = R_log(r, s) + xi * Phi(s; y, a)
    over the grid xi in {0} U logspace(-2, 2, 40). At xi = 0 the solution is
    exactly s = r. Pool every resulting score vector together with the
    fitted model scores, sort by fairness, take the running minimum of the
    log loss and collapse duplicate fairness values: that is the frontier
    function F. THEN REFINE: where adjacent frontier nodes are far apart,
    insert additional multipliers at the LOGARITHMIC midpoints of the
    bracketing values, re-solve, add the results to the pool, and
    re-extract. Finally, for each fitted model,
        Delta(s_m) = R_log(r, s_m) - F(Phi(s_m; y, a)).

    B-6 remediation. The refinement stage did not exist: the three
    configuration keys (max_additional_multipliers, compute_log_midpoints,
    node_chord_length_threshold) appeared only inside the config literal and
    the validation path lists, no routine read them, and the config set
    refinement.enabled to False against the specification. The consequence
    was a coarser envelope wherever the frontier is steep, and because the
    accuracy gap is measured against that envelope — which, being coarser,
    lies ABOVE the true one — every reported accuracy gap was biased
    downward by an unknown amount. The accuracy gap is one of the study's
    headline metrics, so this was not a cosmetic omission.

    The initialisation pool is also repaired: fitted model scores now seed
    the solves (Task 10 Step 2), not merely the candidate pool afterwards.

    Parameters
    ----------
    reference_risk : np.ndarray
        The validation reference risks r, shape (n,).
    y_true, a_true : np.ndarray
        The validation labels and sensitive attributes, shape (n,).
    fitted_scores : Optional[Dict[str, np.ndarray]], optional
        Named fitted-model validation score vectors. They both seed the
        solves and enter the candidate pool.
    frontier_config : Optional[Dict[str, Any]], optional
        The config["frontier"] block; defaults apply when absent.
    criterion : str, optional
        The fairness criterion, by default equalised_odds.

    Returns
    -------
    FrontierReport
        The envelope arrays, the per-model accuracy gaps and the pool size.

    Raises
    ------
    ValueError
        If the inputs misalign or the candidate set is empty.

    """
    # The three validation vectors must describe the same population
    if not (reference_risk.shape[0] == y_true.shape[0] == a_true.shape[0]):
        raise ValueError(
            f"frontier inputs are row-misaligned: r {reference_risk.shape[0]}, "
            f"y {y_true.shape[0]}, a {a_true.shape[0]}."
        )
    if frontier_config is None:
        frontier_config = {}
    xi_grid_cfg: Dict[str, Any] = frontier_config.get("xi_grid", {})
    solver_cfg: Dict[str, Any] = frontier_config.get("solver", {})
    refine_cfg: Dict[str, Any] = frontier_config.get("refinement", {})
    # Step 1: the multiplier grid from the config block
    xi_grid: np.ndarray = build_xi_grid(
        include_zero=bool(xi_grid_cfg.get("include_zero", True)),
        logspace_start=float(xi_grid_cfg.get("logspace_start", -2.0)),
        logspace_end=float(xi_grid_cfg.get("logspace_end", 2.0)),
        num_points=int(xi_grid_cfg.get("num_points", 40)),
    )
    # The solver settings, with the registered defaults
    rho: float = float(solver_cfg.get("step_size_rho", DEFAULT_RHO))
    epsilon: float = float(solver_cfg.get("clipping_epsilon", DEFAULT_EPSILON))
    max_iterations: int = int(solver_cfg.get(
        "max_iterations", DEFAULT_MAX_ITERATIONS))
    convergence_tolerance: float = float(solver_cfg.get(
        "convergence_tolerance", DEFAULT_CONVERGENCE_TOLERANCE))
    use_pr: bool = bool(solver_cfg.get("use_polyak_ruppert_average", True))
    # B-6(a): the refinement settings; ENABLED by default per the spec
    refine_enabled: bool = bool(refine_cfg.get(
        "enabled", FRONTIER_REFINEMENT_ENABLED))
    max_additional: int = int(refine_cfg.get(
        "max_additional_multipliers", FRONTIER_MAX_ADDITIONAL_MULTIPLIERS))
    chord_threshold: float = float(refine_cfg.get(
        "node_chord_length_threshold", FRONTIER_NODE_CHORD_THRESHOLD))
    use_log_midpoints: bool = bool(refine_cfg.get(
        "compute_log_midpoints", True))
    # The fitted score vectors seed the pool (Task 10 Step 2)
    seed_vectors: List[np.ndarray] = [clip_scores(reference_risk, epsilon)]
    if fitted_scores is not None:
        for score in fitted_scores.values():
            seed_vectors.append(clip_scores(
                np.asarray(score, dtype=np.float64), epsilon))
    # The closed-form zero multiplier: s = r exactly, Q_0(r) = H(r)
    zero_candidate: Dict[str, Any] = {
        "solution": clip_scores(reference_risk, epsilon),
        "pr_solution": clip_scores(reference_risk, epsilon),
        "warm_start": clip_scores(reference_risk, epsilon),
        "final_objective": expected_log_loss(reference_risk, reference_risk),
        "initial_objective": expected_log_loss(reference_risk, reference_risk),
        "iterations": 0,
        "converged": True,
    }
    solve_results: List[Dict[str, Any]] = [zero_candidate]
    # solved_by_xi tracks which multiplier produced which solution, which
    # the refinement stage needs to bracket a node by its multiplier
    solved_by_xi: Dict[float, np.ndarray] = {
        0.0: clip_scores(reference_risk, epsilon)
    }
    previous_solution: np.ndarray = clip_scores(reference_risk, epsilon)

    def solve_at(multiplier: float, warm: np.ndarray) -> Dict[str, Any]:
        """
        Run one solve at `multiplier`, seeded by the full candidate pool.

        Parameters
        ----------
        multiplier : float
            The frontier multiplier xi.
        warm : np.ndarray
            The previous-multiplier solution, added to the pool.

        Returns
        -------
        Dict[str, Any]
            The solver record.

        """
        # Task 10 Step 2: the pool is {r, fitted model scores, previous xi}
        pool: List[np.ndarray] = list(seed_vectors) + [warm]
        return solve_scalarised_frontier(
            pool, reference_risk, y_true, a_true, float(multiplier), rho,
            epsilon, max_iterations, convergence_tolerance, use_pr,
            criterion,
        )

    # The per-xi sweep, warm-started from the previous multiplier
    for xi_value in xi_grid[1:]:
        result: Dict[str, Any] = solve_at(float(xi_value), previous_solution)
        solve_results.append(result)
        solved_by_xi[float(xi_value)] = result["pr_solution"]
        # The previous solution seeds the next multiplier (Appendix A)
        previous_solution = result["pr_solution"]
    # Pool the solver output into (phi, logloss, score) candidate records
    candidates: List[Tuple[float, float, np.ndarray]] = collect_candidates(
        solve_results, reference_risk, y_true, a_true, criterion
    )

    def append_fitted(pool: List[Tuple[float, float, np.ndarray]]) -> None:
        """
        Add every fitted model's score vector to the candidate pool.

        Parameters
        ----------
        pool : List[Tuple[float, float, np.ndarray]]
            The candidate pool, extended in place.

        Returns
        -------
        None

        """
        # A fitted score is a legitimate frontier candidate in its own right
        if fitted_scores is None:
            return
        for score in fitted_scores.values():
            clipped: np.ndarray = clip_scores(
                np.asarray(score, dtype=np.float64), epsilon)
            pool.append((
                float(fairness_penalty(
                    torch.from_numpy(clipped.astype(np.float64)),
                    torch.from_numpy(y_true.astype(np.float64)),
                    torch.from_numpy(a_true.astype(np.float64)),
                    criterion,
                )),
                expected_log_loss(reference_risk, clipped),
                clipped,
            ))

    append_fitted(candidates)
    # Step 3: the first envelope extraction
    envelope_phi, envelope_logloss = extract_frontier_envelope(candidates)
    # ------------------------------------------------------------------
    # B-6(b)-(f): THE REFINEMENT STAGE
    # ------------------------------------------------------------------
    inserted: int = 0
    largest_chord: float = 0.0
    if refine_enabled and max_additional > 0:
        # The solved multipliers, ascending, drive the bracketing
        solved_multipliers: List[float] = sorted(solved_by_xi)
        for _ in range(max_additional):
            # (b) the range-normalised chord between adjacent nodes
            chords: np.ndarray = normalised_chord_lengths(
                envelope_phi, envelope_logloss)
            if chords.size == 0:
                break
            largest_chord = float(chords.max())
            # Stop once every node pair is closer than the threshold
            if largest_chord <= chord_threshold:
                break
            # (c) rank pairs by chord descending and take the widest
            widest: int = int(np.argmax(chords))
            # Map the two envelope nodes back to the multipliers that
            # produced them. A node contributed by a FITTED MODEL has no
            # associated multiplier, so it is bracketed by its nearest
            # solved neighbours instead
            left_phi: float = float(envelope_phi[widest])
            right_phi: float = float(envelope_phi[widest + 1])
            bracket: List[float] = []
            for multiplier in solved_multipliers:
                # The fairness value this multiplier's solution realises
                solution: np.ndarray = solved_by_xi[multiplier]
                phi_value: float = float(fairness_penalty(
                    torch.from_numpy(solution.astype(np.float64)),
                    torch.from_numpy(y_true.astype(np.float64)),
                    torch.from_numpy(a_true.astype(np.float64)),
                    criterion,
                ))
                if left_phi <= phi_value <= right_phi:
                    bracket.append(multiplier)
            if len(bracket) >= 2:
                # The two multipliers bracketing the widest gap
                low, high = min(bracket), max(bracket)
            else:
                # Nearest solved neighbours as the fallback bracket
                low = min(solved_multipliers)
                high = max(solved_multipliers)
                for index in range(len(solved_multipliers) - 1):
                    if solved_multipliers[index + 1] > solved_multipliers[index]:
                        low = solved_multipliers[index]
                        high = solved_multipliers[index + 1]
                        break
            if use_log_midpoints and low > 0.0 and high > 0.0:
                # (c) "log midpoint" on a logarithmic grid is the GEOMETRIC
                # mean of the bracketing multipliers
                new_multiplier: float = float(math.sqrt(low * high))
            else:
                # The arithmetic midpoint when a bracket touches zero, where
                # the geometric mean is degenerate
                new_multiplier = 0.5 * (low + high)
            # A multiplier already solved adds nothing
            if any(abs(new_multiplier - m) <= 1e-12
                   for m in solved_multipliers):
                break
            # (d) solve at the new multiplier, warm-started from the nearer
            # bracketing solution
            nearer: float = low if abs(new_multiplier - low) <= abs(
                new_multiplier - high) else high
            refined: Dict[str, Any] = solve_at(
                new_multiplier, solved_by_xi[nearer])
            solve_results.append(refined)
            solved_by_xi[new_multiplier] = refined["pr_solution"]
            solved_multipliers = sorted(solved_by_xi)
            inserted += 1
            # (e) re-pool and re-extract ONCE per insertion, so the next
            # chord test sees the improved envelope
            candidates = collect_candidates(
                solve_results, reference_risk, y_true, a_true, criterion
            )
            append_fitted(candidates)
            envelope_phi, envelope_logloss = extract_frontier_envelope(
                candidates)
        # The final chord state after the last insertion
        final_chords: np.ndarray = normalised_chord_lengths(
            envelope_phi, envelope_logloss)
        largest_chord = float(
            final_chords.max()) if final_chords.size else 0.0
    # Equation (11): the accuracy gap of every fitted model
    gaps: Dict[str, float] = {}
    if fitted_scores is not None:
        for name, score in fitted_scores.items():
            gaps[name] = compute_accuracy_gap(
                score, reference_risk, y_true, a_true,
                envelope_phi, envelope_logloss, criterion,
            )
    # (f) the refinement diagnostics are reported so a reader can judge
    # whether the envelope has converged
    logger.info(
        f"frontier: {len(candidates)} candidates, {len(envelope_phi)} "
        f"envelope points, {inserted} refinement multiplier(s) inserted, "
        f"largest remaining normalised chord {largest_chord:.6f} "
        f"(threshold {chord_threshold})."
    )
    return FrontierReport(
        envelope_phi=envelope_phi,
        envelope_logloss=envelope_logloss,
        accuracy_gaps=gaps,
        candidate_count=len(candidates),
    )
````

````python
# Task 11: Create the End-to-End Orchestrator Function

# ==============================================================================
# Task 11: Create the End-to-End Orchestrator Function
# ==============================================================================

# Make the sibling task modules importable as the study pipeline grows

# Configure the module-level logger for stage diagnostics

# The ten packaged task callables (step 1 completeness contract)
# In-notebook alias: compute_loss IS total_loss from the
# Task 6 cell (the shared notebook namespace is one module)
compute_loss = total_loss

PACKAGED_CALLABLES: Dict[str, object] = {
    "validate_inputs": validate_inputs,
    "clean_raw_data": clean_raw_data,
    "preprocess_real_data": preprocess_real_data,
    "prepare_simulation_data": prepare_simulation_data,
    "build_model": build_model,
    "compute_loss": compute_loss,
    "train_model_and_select_hyperparameters": train_model_and_select_hyperparameters,
    "evaluate_model": evaluate_model,
    "compute_diagnostics": compute_diagnostics,
    "construct_frontier": construct_frontier,
}
# The fixed per-run record schema (step 3)
def normalise_metric_records(
    records: List[Dict[str, Any]],
) -> List[Dict[str, Any]]:
    """
    Map run records to the aggregation metric-name contract.

    The run records carry the schema names (test_auc, test_brier,
    validation_phi); the Task 8 aggregation consumes the plain metric
    names (auc, brier, fairness_gap). This is the single mapping point.

    Parameters
    ----------
    records : List[Dict[str, Any]]
        The per-run records.

    Returns
    -------
    List[Dict[str, Any]]
        The records with the plain metric names added.

    """
    normalised: List[Dict[str, Any]] = []
    for record in records:
        # The mapping is name-for-name against the schema
        normalised.append({
            **(record if not isinstance(record, dict) else {}),
            "dataset": record["dataset"],
            "mode": record["mode"],
            "model_class": record["model_class"],
            "lambda": record["lambda"],
            "auc": record.get("test_auc", np.nan),
            "brier": record.get("test_brier", np.nan),
            "fairness_gap": record.get("validation_phi", np.nan),
            "excess_log_loss": record.get("excess_log_loss", np.nan),
            "accuracy_gap": record.get("accuracy_gap", np.nan),
        })
    return normalised

RUN_RECORD_SCHEMA: Tuple[str, ...] = (
    "dataset", "mode", "seed", "model_class", "lambda", "hyperparameters",
    "test_auc", "test_brier", "validation_phi", "excess_log_loss",
    "accuracy_gap", "evr", "ddr", "cser_min", "cser_max",
    # B-21(iii)/B-23: the SUPPORT of the CSER summary. A zero CSER over an
    # EMPTY eligible set is sentinel contamination; a zero over a non-empty
    # set is a genuine result (and is what a purely linear model gives,
    # since every perturbation +sign(beta_j) * delta * e_j raises the logit
    # by delta * |beta_j| > 0). Without this count the two are
    # indistinguishable in the audit trail.
    "cser_eligible_count",
    "parameter_path",
)
# The run-log registry path
RUN_LOG_PATH: Path = Path("study_run_log.jsonl")
# The reference-risk quick-fit epoch cap (no search: a cheap direct fit
# of the unpenalised logistic baseline)
REFERENCE_RISK_EPOCH_CAP: int = 200
# The diagnostic non-availability sentinel
NOT_APPLICABLE: float = float("nan")

@dataclass
class StudyResults:
    """
    The aggregated output of a pipeline run.

    Attributes
    ----------
    run_records : List[Dict[str, Any]]
        Every completed per-run record.
    summary : pd.DataFrame
        The Task 8 aggregation of the records.
    frontiers : Dict[str, FrontierReport]
        Per-dataset/mode frontier reports.
    violations : List[str]
        Any verification-target violations encountered.

    """

    # The per-run records
    run_records: List[Dict[str, Any]]
    # The aggregated summary
    summary: pd.DataFrame
    # The frontier reports keyed by dataset/mode
    frontiers: Dict[str, FrontierReport]
    # Verification violations
    violations: List[str]

# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Step 1: Package task-specific callables
# -------------------------------------------------------------------------------------------------------------------------------
def verify_package() -> List[str]:
    """
    Verify that every packaged callable exists and resolves.

    Returns
    -------
    List[str]
        The ten packaged names; raises on any missing entry.

    Raises
    ------
    ImportError
        If any packaged name is not present or not callable.

    """
    missing: List[str] = []
    for name, callable_ref in PACKAGED_CALLABLES.items():
        # The name must bind to a callable object
        if callable_ref is None or not callable(callable_ref):
            missing.append(name)
    if missing:
        raise ImportError(
            f"the study package is incomplete; entries missing or "
            f"non-callable: {missing}."
        )
    return list(PACKAGED_CALLABLES.keys())

# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Step 3: Atomic logging and verification (implemented first so
# the graph can log as it runs)
# -------------------------------------------------------------------------------------------------------------------------------
def validate_run_record(record: Dict[str, Any]) -> None:
    """
    Validate one run record against the fixed schema.

    Parameters
    ----------
    record : Dict[str, Any]
        The per-run record.

    Raises
    ------
    ValueError
        If any schema field is missing.

    """
    missing: List[str] = [key for key in RUN_RECORD_SCHEMA if key not in record]
    if missing:
        raise ValueError(
            f"run record missing schema field(s): {missing}."
        )

@contextlib.contextmanager
def exclusive_file_lock(lock_path: Path, timeout_seconds: float = 30.0) -> Any:
    """
    Acquire a cross-process exclusive lock for the duration of a block.

    B-19. The run log is appended by up to four worker processes under
    ``Pool(processes=4)``. Without a lock two appends can interleave, and
    the previous read-back check compared the LAST line of the file, which
    under concurrency may belong to a different worker: it could therefore
    both miss a corrupted write and raise spuriously on a correct one.

    The implementation uses an atomic-create lock file, which is the one
    mechanism that behaves identically on Windows and POSIX without a third
    party dependency: ``O_CREAT | O_EXCL`` succeeds for exactly one process.
    The lock is always released in a ``finally`` block, and a stale lock is
    surfaced as a timeout rather than a silent hang.

    Parameters
    ----------
    lock_path : Path
        The sentinel file backing the lock.
    timeout_seconds : float, optional
        How long to wait for the holder to release, by default 30 seconds.

    Yields
    ------
    None
        The block executes while the lock is held.

    Raises
    ------
    TimeoutError
        If the lock cannot be acquired within the timeout.

    """
    # The deadline bounds the spin so a crashed holder cannot hang the run
    deadline: float = time.monotonic() + float(timeout_seconds)
    handle: Optional[int] = None
    while True:
        try:
            # O_CREAT | O_EXCL is atomic: exactly one process wins the race
            handle = os.open(
                str(lock_path), os.O_CREAT | os.O_EXCL | os.O_WRONLY
            )
            break
        except FileExistsError:
            # Another process holds the lock; wait and retry until deadline
            if time.monotonic() >= deadline:
                raise TimeoutError(
                    f"could not acquire the run-log lock at {lock_path} "
                    f"within {timeout_seconds}s; remove the stale lock file "
                    f"if no writer is running."
                )
            time.sleep(0.01)
    try:
        # The critical section runs with the lock held
        yield
    finally:
        # Release unconditionally: close the descriptor, then unlink
        if handle is not None:
            os.close(handle)
        try:
            lock_path.unlink()
        except FileNotFoundError:
            # Already removed: releasing a released lock is not an error
            pass


def append_run_record(record: Dict[str, Any], log_path: Path = RUN_LOG_PATH) -> None:
    """
    Atomically append one run record as a JSON line.

    Task 11 Step 3 requires an append-only registry that preserves every
    completed run across an interruption, so that numerical drift across
    seeds and datasets can be audited after the fact.

    B-19 remediation: the word "atomically" was not earned. There was no
    lock of any kind, and the read-back verified only the seed and
    model_class of the file's LAST line — which, under the four-process
    pool that drives the experiment matrix, may have been written by
    another worker. The append-and-verify window is now held under a
    genuine cross-process exclusive lock, and the verification reads back
    the exact byte offset this call wrote rather than "the last line".

    Parameters
    ----------
    record : Dict[str, Any]
        The per-run record; validated against the fixed schema first.
    log_path : Path, optional
        The JSON-lines registry path.

    Returns
    -------
    None

    Raises
    ------
    ValueError
        If the record violates the fixed schema.
    RuntimeError
        If the appended line does not read back byte-identically.
    TimeoutError
        If the registry lock cannot be acquired.

    """
    # The schema is enforced before anything touches the filesystem
    validate_run_record(record)
    # JSON-native serialisation: numpy scalars convert at the boundary
    serialisable: Dict[str, Any] = {
        key: (float(value) if isinstance(value, np.floating) else
              int(value) if isinstance(value, np.integer) else
              value)
        for key, value in record.items()
    }
    # The exact payload this call is responsible for
    payload: str = json.dumps(serialisable) + "\n"
    encoded: bytes = payload.encode("utf-8")
    # The lock sentinel sits beside the registry it guards
    lock_path: Path = log_path.with_suffix(log_path.suffix + ".lock")
    # B-19: the append AND its verification are one critical section
    with exclusive_file_lock(lock_path):
        # Open in binary append so the byte offset is exact and no newline
        # translation can alter the payload length
        with open(log_path, "ab") as handle:
            # Record the offset this call writes at, so the read-back can
            # target THIS line rather than whatever line happens to be last
            offset: int = handle.tell()
            handle.write(encoded)
            handle.flush()
            # fsync forces the bytes to stable storage before verification
            os.fsync(handle.fileno())
        # Read back exactly the bytes this call wrote, at its own offset
        with open(log_path, "rb") as handle:
            handle.seek(offset)
            written: bytes = handle.read(len(encoded))
    # The round trip must be byte-identical; anything else is corruption
    if written != encoded:
        raise RuntimeError(
            f"run-log append round-trip mismatch at offset {offset} in "
            f"{log_path}: wrote {len(encoded)} bytes, read back "
            f"{len(written)}. The registry may be corrupt."
        )

def check_verification_targets(
    record: Dict[str, Any],
    verification_targets: Dict[str, Any],
) -> List[str]:
    """
    Compare one run against the configured verification targets.

    Task 11 Step 3: every run is checked against the verification targets
    and, when a target is grossly violated, the run is flagged and the
    pipeline stops for review, so that numerical drift across seeds and
    datasets cannot accumulate silently.

    B-10 remediation, three defects repaired.

    (1) SCOPING. The routine's own inline comment claimed that only records
        "matching the named dataset/mode/lambda" were checked, and no such
        matching existed: a target intended for one dataset was applied to
        every record carrying the named metric. The target schema now
        accepts optional dataset, mode, model_class and lambda scoping
        fields, and a record that does not match EVERY field the target
        specifies is skipped.
    (2) MISSING METRICS. A record that did not carry the target metric was
        skipped in silence. Such skips are now counted and reported, so an
        absent metric is visible rather than indistinguishable from a pass.
    (3) HALTING. This function reports; the orchestrator halts. The
        contract is now explicit in the docstring so the caller's
        obligation cannot be overlooked (the caller raises on a non-empty
        return unless an override is recorded in the audit).

    Target schema, per entry:
        record_key : str    the field of the run record to compare
        expected   : float  the reference value
        tolerance  : float  optional; falls back to the block default
        dataset / mode / model_class / lambda : optional scoping filters

    Parameters
    ----------
    record : Dict[str, Any]
        The per-run record.
    verification_targets : Dict[str, Any]
        The config["sample_controls"]["verification_targets"] block: named
        target specifications plus a default "tolerance" scalar.

    Returns
    -------
    List[str]
        Every violation and every missing-metric note, each naming the
        run's identity. An empty list means the run passed every target
        whose scope it matched.

    Raises
    ------
    ValueError
        If a target specification omits a required field.

    """
    violations: List[str] = []
    # The block-level fallback tolerance
    default_tolerance: float = float(
        verification_targets.get("tolerance", 0.05))
    # The run's identity, used for scoping and for every message
    identity: str = (
        f"dataset={record.get('dataset')}, mode={record.get('mode')}, "
        f"model={record.get('model_class')}, lambda={record.get('lambda')}, "
        f"seed={record.get('seed')}"
    )
    for target_key, target_spec in verification_targets.items():
        # The default tolerance is the schema's own entry, not a target
        if target_key == "tolerance":
            continue
        # A non-dict entry is not a target specification
        if not isinstance(target_spec, dict):
            continue
        # The two mandatory fields of a target specification
        for required in ("record_key", "expected"):
            if required not in target_spec:
                raise ValueError(
                    f"verification target '{target_key}' omits the required "
                    f"field '{required}'."
                )
        record_key: str = str(target_spec["record_key"])
        expected: float = float(target_spec["expected"])
        tolerance: float = float(
            target_spec.get("tolerance", default_tolerance))
        # B-10(1): the scoping filters the previous comment promised. A
        # target applies only to records matching EVERY field it names
        scoped_out: bool = False
        for scope_field in ("dataset", "mode", "model_class"):
            if scope_field in target_spec:
                # String comparison keeps the match insensitive to dtype
                if str(record.get(scope_field)) != str(
                        target_spec[scope_field]):
                    scoped_out = True
                    break
        # The multiplier is numeric and compared with a tight tolerance
        if not scoped_out and "lambda" in target_spec:
            try:
                if abs(float(record.get("lambda", float("nan")))
                       - float(target_spec["lambda"])) > 1e-12:
                    scoped_out = True
            except (TypeError, ValueError):
                scoped_out = True
        if scoped_out:
            # Out of scope is not a pass and not a failure: it is silence
            continue
        # B-10(2): a missing metric is REPORTED, not silently skipped
        if record_key not in record:
            violations.append(
                f"run ({identity}): target '{target_key}' requires field "
                f"'{record_key}', which the run record does not carry."
            )
            continue
        observed_raw: Any = record[record_key]
        # A None or NaN metric cannot be compared and must be surfaced
        if observed_raw is None or (
                isinstance(observed_raw, float) and math.isnan(observed_raw)):
            violations.append(
                f"run ({identity}): target '{target_key}' field "
                f"'{record_key}' is missing (None/NaN) and cannot be "
                f"verified."
            )
            continue
        observed: float = float(observed_raw)
        # The tolerance comparison itself
        if abs(observed - expected) > tolerance:
            violations.append(
                f"run ({identity}): {target_key} = {observed:.6f} vs "
                f"expected {expected:.6f} beyond tolerance {tolerance}."
            )
    return violations

# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Step 2: The orchestrator execution graph
# -------------------------------------------------------------------------------------------------------------------------------
def build_reference_risk_lr(
    X_train: np.ndarray, y_train: np.ndarray, X_val: np.ndarray,
    config: Dict[str, Any],
) -> np.ndarray:
    """
    Compute the validation reference risk as the unpenalised LR at lambda 0.

    The config's unpenalised model is logistic_regression_at_lambda_0: the
    reference LR is trained on the TRAINING split only (a quick direct fit,
    no search) and predicts the validation probabilities.

    Parameters
    ----------
    X_train, y_train : np.ndarray
        The training design and labels.
    X_val : np.ndarray
        The validation design.
    config : Dict[str, Any]
        The study configuration (training settings for the quick fit).

    Returns
    -------
    np.ndarray
        The validation reference risk, shape (n_val,).

    """
    training_cfg: Dict[str, Any] = config["network_training"]
    # The reference model is seeded before construction
    setup_seeded_environment(int(training_cfg["seed"]))
    model: torch.nn.Module = build_model("LR", int(X_train.shape[1]),
                                            [4], "ReLU", 0.0)
    # A direct full-batch fit with the study's bounds: the reference is a
    # cheap baseline, never a search product
    result = train_model_with_early_stopping(
        model,
        torch.from_numpy(X_train.astype(np.float32)),
        torch.from_numpy(y_train.astype(np.float32)),
        torch.from_numpy(np.ones_like(y_train, dtype=np.int64)),
        torch.from_numpy(X_val.astype(np.float32)),
        torch.from_numpy(np.zeros(X_val.shape[0], dtype=np.float32)),
        torch.from_numpy(np.ones(X_val.shape[0], dtype=np.int64)),
        {"learning_rate": 1e-3, "weight_decay": 0.0, "batch_size": 128,
         "hidden_layer_widths": [4], "activation": "ReLU",
         "dropout_rate": 0.0},
        fairness_multiplier=0.0,
        max_epochs=min(int(training_cfg["max_epochs"]),
                       REFERENCE_RISK_EPOCH_CAP),
        patience=int(training_cfg["early_stopping_patience"]),
    )
    # Restore the best state, then predict the validation scores
    model.load_state_dict(result.model_state)
    # B-1(2): the restored parameters are a new generation, so any cached
    # evaluation projection is stale and must be invalidated before scoring
    if hasattr(model, "bump_parameter_version"):
        model.bump_parameter_version()
    # B-1(4)/B-4: the restoring scoring context replaces the guard mutation
    with scoring_context(model):
        with torch.no_grad():
            reference_probs: np.ndarray = torch.sigmoid(
                forward_scores(
                    model, torch.from_numpy(X_val.astype(np.float32)),
                    is_training_population=False)
            ).numpy()
    del model
    return reference_probs

def frozen_validation_probabilities(
    model: torch.nn.Module, X_val: np.ndarray,
) -> np.ndarray:
    """
    Predict validation probabilities from a frozen model.

    Parameters
    ----------
    model : torch.nn.Module
        The fitted model (eval mode applied here).
    X_val : np.ndarray
        The validation design.

    Returns
    -------
    np.ndarray
        The validation probabilities, shape (n_val,).

    """
    # B-1(4)/B-4: the restoring scoring context replaces the guard
    # mutation. The model's externally visible state is unchanged by this
    # call, which is what "frozen" is supposed to mean.
    with scoring_context(model):
        with torch.no_grad():
            return torch.sigmoid(
                forward_scores(
                    model, torch.from_numpy(X_val.astype(np.float32)),
                    is_training_population=False)
            ).numpy()

def structured_coefficients_as_path(
    model: torch.nn.Module, model_class: str,
) -> List[float]:
    """
    Serialize the fitted structured coefficients as a JSON path.

    Parameters
    ----------
    model : torch.nn.Module
        The fitted findr or LR model.
    model_class : str
        One of LR, NN, findr.

    Returns
    -------
    List[float]
        The coefficient vector as a JSON-safe list (empty for NN).

    """
    if model_class == "LR":
        weights = model.linear.weight.detach().numpy().reshape(-1)
    elif model_class == "findr":
        weights = model.structured_layer.weight.detach().numpy().reshape(-1)
    else:
        return []
    return [float(v) for v in weights]

def run_real_dataset_full(
    dataset: str,
    cleaned_frame: pd.DataFrame,
    metadata: Dict[str, object],
    schema_continuous: List[str],
    schema_categorical: List[str],
    config: Dict[str, Any],
    subset: Dict[str, Sequence[Any]],
) -> StudyResults:
    """
    Run the full pipeline for one real dataset.

    Parameters
    ----------
    dataset : str
        The dataset key.
    cleaned_frame : pd.DataFrame
        The cleaned joined frame (record_id index).
    metadata : Dict[str, object]
        The dataset metadata.
    schema_continuous, schema_categorical : List[str]
        The feature schema.
    config : Dict[str, Any]
        The study configuration.
    subset : Dict[str, Sequence[Any]]
        The dimension filters (lambdas, models, num_seeds).

    Returns
    -------
    StudyResults
        The dataset's run records, the frontier report and the gaps.

    """
    # Task 3: preprocessing with the frozen training statistics
    # Task 3 consumes the CleanedDataBundle contract: the frame under
    # its cleaned_datasets attribute, never a bare dict
    dataset_bundle: CleanedDataBundle = CleanedDataBundle(
        cleaned_datasets={dataset: cleaned_frame},
        cleaned_simulation=None, reports={},
        simulation_duplication_rate=0.0,
    )
    pre: PreprocessedDataBundle = preprocess_real_data(
        config,
        dataset_bundle,
        {dataset: metadata},
        {dataset: schema_continuous},
        {dataset: schema_categorical},
        split_config=config["data_preprocessing"]["train_val_test_split"],
        artifact_path=Path("preprocessing_stats.json"),
    )[dataset]
    # The reference risk: the unpenalised LR at lambda 0, trained on the
    # training split only and scored on validation
    reference_risk: np.ndarray = build_reference_risk_lr(
        pre.X_train, pre.y_train, pre.X_val, config
    )
    # The subset filters with the config defaults
    lambdas: Sequence[Any] = subset.get(
        "lambdas", config["network_training"]["loss"]["lambda_grid_realdata"],
    )
    models: Sequence[str] = subset.get(
        "models", config["real_data_experiment"]["model_comparisons"],
    )
    num_seeds: int = int(subset.get(
        "num_seeds", config["network_training"]["num_seeds"]))
    criterion: str = str(config["fairness_penalty"]["active_criterion"])
    run_records: List[Dict[str, Any]] = []
    validation_scores: Dict[str, np.ndarray] = {}
    # One model at a time: the search and finals release their state
    for lambda_value, model_class in itertools.product(lambdas, models):
        best_config, seed_results, seed_log = \
            train_model_and_select_hyperparameters(
                model_class, pre.X_train, pre.y_train, pre.a_train,
                pre.X_val, pre.y_val, pre.a_val, pre.X_test,
                config, float(lambda_value), criterion,
            )
        for seed_result in seed_results:
            # The frozen validation probabilities for the frontier pool
            val_probs: np.ndarray = frozen_validation_probabilities(
                seed_result.model, pre.X_val,
            )
            # The evaluation and the diagnostics
            eval_probs: np.ndarray = seed_result.test_probabilities.numpy()
            eval_report: EvaluationReport = evaluate_model(
                eval_probs, pre.y_test, np.asarray(pre.test_ids),
                val_probs, pre.y_val, pre.a_val, reference_risk, criterion,
            )
            metric_values: Dict[str, float] = {
                "auc": eval_report.auc,
                "brier": eval_report.brier,
                "validation_phi": eval_report.fairness_gap,
                "excess_log_loss": eval_report.excess_log_loss,
            }
            # Task 9: the diagnostics (findr/LR only)
            if model_class in ("findr", "LR"):
                diag: DiagnosticsReport = compute_diagnostics(
                    seed_result.model, model_class,
                    cleaned_frame.loc[pre.val_ids],
                    schema_continuous, schema_categorical, pre.stats,
                    pre.feature_names,
                    delta=float(config["interpretability_diagnostics"][
                        "csert_perturbation_delta"]),
                    decision_threshold=float(
                        config["interpretability_diagnostics"][
                            "decision_threshold"]),
                    split="validation",
                )
                # The eligible roster the diagnostics reduced over; its
                # size is the support of CSER_min and CSER_max
                eligible_roster: List[str] = list(
                    diag.per_feature_cser.get("__eligible__", []))
                metric_values.update({
                    "evr": diag.evr, "ddr": diag.ddr,
                    "cser_min": diag.cser_min, "cser_max": diag.cser_max,
                    "cser_eligible_count": len(eligible_roster),
                })
            else:
                metric_values.update({
                    "evr": NOT_APPLICABLE, "ddr": NOT_APPLICABLE,
                    "cser_min": NOT_APPLICABLE, "cser_max": NOT_APPLICABLE,
                    "cser_eligible_count": 0,
                })
            # The parameter path is the fitted coefficient vector
            parameter_path: List[float] = structured_coefficients_as_path(
                seed_result.model, model_class)
            record: Dict[str, Any] = {
                "dataset": dataset, "mode": "real",
                "seed": seed_result.seed,
                "model_class": model_class,
                "lambda": float(lambda_value),
                "hyperparameters": {key: best_config[key] for key in (
                    "learning_rate", "weight_decay", "batch_size",
                    "hidden_layer_widths", "activation",
                    "dropout_rate")},
                "test_auc": metric_values["auc"],
                "test_brier": metric_values["brier"],
                "validation_phi": metric_values["validation_phi"],
                "excess_log_loss": metric_values["excess_log_loss"],
                "accuracy_gap": 0.0,
                "evr": metric_values["evr"],
                "ddr": metric_values["ddr"],
                "cser_min": metric_values["cser_min"],
                "cser_max": metric_values["cser_max"],
                # B-21(iii): the CSER support travels with the statistic
                "cser_eligible_count": metric_values.get(
                    "cser_eligible_count", 0),
                "parameter_path": parameter_path,
            }
            run_records.append(record)
            # The validation scores are pooled into the frontier
            validation_scores[f"{model_class}_l{lambda_value}_s{seed_result.seed}"] = val_probs
        del best_config
    # Task 10: the frontier on the validation split, with the pooled fit
    # scores; the accuracy gaps are backfilled into the records
    frontier: FrontierReport = construct_frontier(
        reference_risk, pre.y_val, pre.a_val, validation_scores,
        config.get("frontier"), criterion,
    )
    for record in run_records:
        key: str = f"{record['model_class']}_l{record['lambda']}_s{record['seed']}"
        if key in frontier.accuracy_gaps:
            record["accuracy_gap"] = frontier.accuracy_gaps[key]
    # The per-run logging and target verification
    for record in run_records:
        append_run_record(record)
    violations: List[str] = []
    targets: Dict[str, Any] = config.get("sample_controls", {}).get(
        "verification_targets", {})
    for record in run_records:
        violations.extend(check_verification_targets(record, targets))
    return StudyResults(
        run_records=run_records, summary=pd.DataFrame(),
        frontiers={dataset: frontier}, violations=violations,
    )

def run_simulation_mode_full(
    mode_val: str,
    sim_bundle: SimulationDataBundle,
    config: Dict[str, Any],
    subset: Dict[str, Sequence[Any]],
) -> StudyResults:
    """
    Run the pipeline for one simulation mode.

    The reference risk is the TRUE conditional risk r_true (the config's
    from_simulation setting). The model consumes the block's covariate
    matrix directly (Z = X, no scaling), so the frozen statistics for the
    CSER reprepare path are the identity map.

    Parameters
    ----------
    mode_val : str
        The generative mode: simple or complex.
    sim_bundle : SimulationDataBundle
        The prepared simulation blocks keyed by (mode, seed).
    config : Dict[str, Any]
        The study configuration.
    subset : Dict[str, Sequence[Any]]
        The dimension filters (lambdas, models, num_seeds).

    Returns
    -------
    StudyResults
        The mode's run records, the frontier report and the violations.

    """
    # The mode's blocks, one per seed
    blocks: Dict[int, SimulationBlockData] = {}
    for (block_mode, seed), block in sim_bundle.blocks.items():
        if block_mode == mode_val:
            blocks[int(seed)] = block
    if not blocks:
        raise ValueError(f"mode '{mode_val}' has no prepared blocks.")
    # The lambda and model grids from the subset filter
    lambdas: Sequence[Any] = subset.get(
        "lambdas",
        config["network_training"]["loss"]["lambda_grid_simulation"],
    )
    models: Sequence[str] = subset.get(
        "models", config["real_data_experiment"]["model_comparisons"],
    )
    num_seeds: int = int(subset.get("num_seeds", 1))
    criterion: str = str(config["fairness_penalty"]["active_criterion"])
    run_records: List[Dict[str, Any]] = []
    validation_scores: Dict[str, np.ndarray] = {}
    # The identity statistics: the simulation matrix is already the raw
    # covariate scale, so the reprepare path maps x -> x
    covariate_names: List[str] = ["X1_post_shift", "X2", "X3", "X4", "X5", "X6"]
    identity_stats = PreprocessingStats(
        means={name: 0.0 for name in covariate_names},
        stds={name: 1.0 for name in covariate_names},
        medians={}, woe_maps={}, global_woe={}, alpha=0.5,
    )
    for lambda_value, model_class in itertools.product(lambdas, models):
        # The reference risk is the true conditional risk (Task 4)
        block_seed0: SimulationBlockData = blocks[min(blocks.keys())]
        best_config, seed_results, seed_log = \
            train_model_and_select_hyperparameters(
                model_class,
                block_seed0.X_train, block_seed0.y_train,
                block_seed0.a_train,
                block_seed0.X_val, block_seed0.y_val, block_seed0.a_val,
                block_seed0.X_test,
                config, float(lambda_value), criterion,
            )
        for seed_index, seed_result in enumerate(seed_results):
            # The simulation replication identity is the DGP block seed:
            # the fitted-model seed is a construction seed, and the record
            # must carry the block's identity so the diagnostics frame
            # selection is valid
            dgp_seed: int = int(block_seed0.seed)
            block: SimulationBlockData = block_seed0
            val_probs: np.ndarray = frozen_validation_probabilities(
                seed_result.model, block.X_val)
            eval_report: EvaluationReport = evaluate_model(
                seed_result.test_probabilities.numpy(), block.y_test,
                np.asarray(block.test_ids),
                val_probs, block.y_val, block.a_val, block.r_val, criterion,
            )
            # Diagnostics on the structured branches with identity stats
            evr: float = NOT_APPLICABLE
            ddr: float = NOT_APPLICABLE
            cser_min: float = NOT_APPLICABLE
            cser_max: float = NOT_APPLICABLE
            # B-21(iii): the support of the CSER summary. Zero means the
            # eligible set {j : beta_j != 0} was empty, which makes a zero
            # CSER contamination rather than a result
            cser_eligible_count: int = 0
            if model_class in ("findr", "LR"):
                raw_sim_frame: pd.DataFrame = sim_bundle.validated_frame[
                    sim_bundle.validated_frame["mode"] == mode_val].copy()
                raw_sim_frame = raw_sim_frame.loc[
                    raw_sim_frame["seed"] == dgp_seed]
                raw_sim_frame.index = raw_sim_frame["record_id"].to_numpy(
                    dtype=np.int64)
                raw_sim_frame.index.name = "record_id"
                # The block's val ids ARE the frame's record_id values
                # (Task 4's split retains the composite key), so the
                # diagnostics rows select them directly
                val_ids_orig: List[int] = [int(v) for v in block.val_ids]
                diag: DiagnosticsReport = compute_diagnostics(
                    seed_result.model, model_class,
                    raw_sim_frame.loc[val_ids_orig],
                    covariate_names, [],
                    identity_stats, covariate_names,
                    delta=float(config["interpretability_diagnostics"][
                        "csert_perturbation_delta"]),
                    decision_threshold=float(
                        config["interpretability_diagnostics"][
                            "decision_threshold"]),
                    split="validation",
                )
                evr, ddr = diag.evr, diag.ddr
                cser_min, cser_max = diag.cser_min, diag.cser_max
                # The roster the diagnostics reduced over IS the support,
                # so the two can never disagree
                cser_eligible_count = len(list(
                    diag.per_feature_cser.get("__eligible__", [])))
            record: Dict[str, Any] = {
                "dataset": f"{mode_val}_seed{dgp_seed}", "mode": mode_val,
                "seed": dgp_seed, "model_class": model_class,
                "lambda": float(lambda_value),
                "hyperparameters": {key: best_config[key] for key in (
                    "learning_rate", "weight_decay", "batch_size",
                    "hidden_layer_widths", "activation",
                    "dropout_rate")},
                "test_auc": eval_report.auc,
                "test_brier": eval_report.brier,
                "validation_phi": eval_report.fairness_gap,
                "excess_log_loss": eval_report.excess_log_loss,
                "accuracy_gap": 0.0,
                "evr": evr, "ddr": ddr,
                "cser_min": cser_min, "cser_max": cser_max,
                # B-21(iii): the CSER support travels with the statistic
                "cser_eligible_count": cser_eligible_count,
                "parameter_path": structured_coefficients_as_path(
                    seed_result.model, model_class),
            }
            run_records.append(record)
            validation_scores[
                f"{model_class}_l{lambda_value}_s{dgp_seed}"] = val_probs
    # The frontier per mode on the seed-0 block validation split
    block_seed0 = blocks[min(blocks.keys())]
    frontier: FrontierReport = construct_frontier(
        block_seed0.r_val, block_seed0.y_val, block_seed0.a_val,
        validation_scores, config.get("frontier"), criterion,
    )
    for record in run_records:
        key: str = f"{record['model_class']}_l{record['lambda']}_s{record['seed']}"
        if key in frontier.accuracy_gaps:
            record["accuracy_gap"] = frontier.accuracy_gaps[key]
        append_run_record(record)
    violations: List[str] = []
    targets: Dict[str, Any] = config.get("sample_controls", {}).get(
        "verification_targets", {})
    for record in run_records:
        violations.extend(check_verification_targets(record, targets))
    return StudyResults(
        run_records=run_records, summary=pd.DataFrame(),
        frontiers={mode_val: frontier}, violations=violations,
    )

def run_study_pipeline(
    config: Dict[str, Any],
    raw_structured_features_by_dataset: Dict[str, pd.DataFrame],
    raw_outcome_labels_by_dataset: Dict[str, pd.DataFrame],
    raw_sensitive_attribute_by_dataset: Dict[str, pd.DataFrame],
    raw_simulation_input: pd.DataFrame,
    schema_continuous: Dict[str, List[str]],
    schema_categorical: Dict[str, List[str]],
    subset: Optional[Dict[str, Sequence[Any]]] = None,
) -> StudyResults:
    """
    Execute the complete research pipeline.

    The graph: validate (Task 1) -> clean (Task 2) -> per-dataset
    preprocess (Task 3) -> per-block simulation prep (Task 4) ->
    per-(lambda, model) train/select (Task 7) -> evaluate (Task 8) ->
    diagnostics (Task 9) -> frontier (Task 10) -> aggregation.

    Parameters
    ----------
    config : Dict[str, Any]
        The study configuration.
    raw_structured_features_by_dataset : Dict[str, pd.DataFrame]
        The raw feature frames.
    raw_outcome_labels_by_dataset : Dict[str, pd.DataFrame]
        The raw label frames.
    raw_sensitive_attribute_by_dataset : Dict[str, pd.DataFrame]
        The raw sensitive frames.
    raw_simulation_input : pd.DataFrame
        The raw simulation frame.
    schema_continuous, schema_categorical : Dict[str, List[str]]
        The per-dataset feature schemas.
    subset : Optional[Dict[str, Sequence[Any]]], optional
        Dimension filters for a partial run; the full matrix is the
        default when absent.

    Returns
    -------
    StudyResults
        The run records, the aggregated summary, the frontier reports and
        the verification violations.

    Raises
    ------
    ValueError
        If the pipeline validation or any task step fails.

    """
    if subset is None:
        subset = {}
    # Step 1: the package must be complete before the graph runs
    verify_package()
    # The pipeline entry: Task 1 validation and Task 2 cleansing
    validation_report = validate_inputs(
        config, raw_structured_features_by_dataset,
        raw_outcome_labels_by_dataset, raw_sensitive_attribute_by_dataset,
        raw_simulation_input, schema_continuous, schema_categorical,
        config["real_data_experiment"]["datasets"],
        run_seeds=range(10),
        audit_log_path=Path("placeholders_audit.json"),
    )
    cleaned_bundle = clean_raw_data(
        validation_report, config["real_data_experiment"]["datasets"],
        schema_continuous, schema_categorical,
        cleansing_artifact_path=Path("cleansing_reports.json"),
    )
    all_records: List[Dict[str, Any]] = []
    frontiers: Dict[str, FrontierReport] = {}
    violations: List[str] = []
    # The per-dataset loop over the selected real datasets
    selected_datasets: Sequence[str] = subset.get(
        "datasets", list(cleaned_bundle.cleaned_datasets.keys()))
    for dataset in selected_datasets:
        results: StudyResults = run_real_dataset_full(
            dataset, cleaned_bundle.cleaned_datasets[dataset],
            config["real_data_experiment"]["datasets"][dataset],
            schema_continuous.get(dataset, []),
            schema_categorical.get(dataset, []),
            config, subset,
        )
        all_records.extend(results.run_records)
        frontiers.update(results.frontiers)
        violations.extend(results.violations)
    # The simulation loop: Task 4 preparation, then the per-mode run
    # chain (train/select, evaluate, diagnose, frontier per mode)
    sim_bundle = prepare_simulation_data(
        validation_report.simulation_with_record_id,
        config["data_preprocessing"]["train_val_test_split"],
        run_seeds=range(10),
        expected_modes=list(config["simulation"]["modes"]),
    )
    selected_modes: Sequence[str] = subset.get(
        "modes", list(config["simulation"]["modes"]))
    for mode_val in selected_modes:
        results: StudyResults = run_simulation_mode_full(
            mode_val, sim_bundle, config, subset)
        all_records.extend(results.run_records)
        frontiers.update(results.frontiers)
        violations.extend(results.violations)
    # The summary aggregation over every record (Task 8)
    summary: pd.DataFrame = aggregate_metrics(
        normalise_metric_records(all_records),
        group_keys=("dataset", "mode", "model_class", "lambda"),
    ) if all_records else pd.DataFrame()
    return StudyResults(
        run_records=all_records, summary=summary,
        frontiers=frontiers, violations=violations,
    )
````

````python
# Task 12: Execute Full Study and Robustness Verification

# ==============================================================================
# Task 12: Execute Full Study and Robustness Verification
# ==============================================================================

# Make the sibling task modules importable as the study pipeline grows

# Configure the module-level logger for stage diagnostics

# The robustness sweep values per task text
ROBUSTNESS_RHO: Tuple[float, ...] = (0.05, 0.1, 0.2)
ROBUSTNESS_DELTA: Tuple[float, ...] = (0.05, 0.1, 0.2)
ROBUSTNESS_ALPHA: Tuple[float, ...] = (0.25, 0.5, 1.0)
# The default seed count for a robustness variant before expansion
ROBUSTNESS_SEEDS_DEFAULT: int = 3
# The diagnostic relative-change materiality threshold
MATERIAL_RELATIVE_CHANGE: float = 0.10

@dataclass(frozen=True)
class RobustVariant:
    """
    One placeholder-overlay configuration.

    Attributes
    ----------
    name : str
        The variant identifier.
    overrides : Dict[str, Any]
        The dotted-path config overrides applying the variant values.

    """

    # The variant identifier
    name: str
    # The dotted-path config overrides
    overrides: Dict[str, Any]

# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 1: The experimental-matrix runner
# -------------------------------------------------------------------------------------------------------------------------------
def matrix_combinations(
    config: Dict[str, Any], subset: Optional[Dict[str, Sequence[Any]]] = None,
) -> List[Tuple[str, float, str]]:
    """
    Enumerate the run-combination keys without the seed axis.

    Parameters
    ----------
    config : Dict[str, Any]
        The study configuration.
    subset : Optional[Dict[str, Sequence[Any]]], optional
        Dimension filters; defaults to the full grids.

    Returns
    -------
    List[Tuple[str, float, str]]
        The (dataset_or_mode, lambda, model_class) combinations, sorted.

    """
    if subset is None:
        subset = {}
    real_keys: List[str] = [
        str(d) for d in subset.get(
            "datasets", list(config["real_data_experiment"]["datasets"].keys()))
    ]
    sim_keys: List[str] = [
        str(m) for m in subset.get("modes", list(config["simulation"]["modes"]))
    ]
    lambdas_sim: Sequence[Any] = subset.get(
        "lambdas", config["network_training"]["loss"]["lambda_grid_simulation"])
    lambdas_real: Sequence[Any] = subset.get(
        "lambdas_real", config["network_training"]["loss"]["lambda_grid_realdata"])
    models: Sequence[str] = subset.get(
        "models", config["real_data_experiment"]["model_comparisons"])
    combos: List[Tuple[str, float, str]] = []
    # The real-dataset axis uses the real-data lambda grid
    for dataset, lambda_value, model_class in itertools.product(
            real_keys, lambdas_real, models):
        combos.append((dataset, float(lambda_value), model_class))
    # The simulation axis uses the simulation lambda grid
    for mode, lambda_value, model_class in itertools.product(
            sim_keys, lambdas_sim, models):
        combos.append((mode, float(lambda_value), model_class))
    combos.sort()
    return combos

def verify_matrix_size(
    config: Dict[str, Any],
    subset: Optional[Dict[str, Sequence[Any]]] = None,
) -> int:
    """
    Verify the combination count equals the grid product.

    Parameters
    ----------
    config : Dict[str, Any]
        The study configuration.
    subset : Optional[Dict[str, Sequence[Any]]], optional
        Dimension filters.

    Returns
    -------
    int
        The verified combination count.

    """
    combos: List[Tuple[str, float, str]] = matrix_combinations(config, subset)
    if len(combos) == 0:
        raise ValueError("the matrix enumerates zero combinations.")
    return len(combos)

def completed_run_keys(log_path: Path = RUN_LOG_PATH) -> set:
    """
    Read the completed run identities from the append-only log.

    Parameters
    ----------
    log_path : Path, optional
        The JSON-lines registry path.

    Returns
    -------
    set
        The (dataset_or_mode, lambda, model_class, seed) keys logged.

    """
    if not log_path.exists():
        return set()
    keys: set = set()
    with open(log_path, "r", encoding="utf-8") as handle:
        for line in handle:
            line = line.strip()
            if not line:
                continue
            record: Dict[str, Any] = json.loads(line)
            keys.add((
                str(record["dataset"]), float(record["lambda"]),
                str(record["model_class"]), int(record["seed"]),
            ))
    return keys

def _worker_run_dataset(spec: Dict[str, Any]) -> StudyResults:
    """
    Run the pipeline for one dataset (or the simulation) in a worker.

    The worker calls the Task 11 pipeline with an isolated subset so the
    dataset's runs never interfere with another worker's.

    Parameters
    ----------
    spec : Dict[str, Any]
        The worker specification: the dataset identity, the config, the
        raw containers and the subset filters.

    Returns
    -------
    StudyResults
        The worker's run records, summary and violations.

    """
    # The subset drops every other dataset; the simulation modes are
    # empty for a real-dataset worker
    return run_study_pipeline(
        spec["config"],
        spec["raw_structured_features_by_dataset"],
        spec["raw_outcome_labels_by_dataset"],
        spec["raw_sensitive_attribute_by_dataset"],
        spec["raw_simulation_input"],
        spec["schema_continuous"],
        spec["schema_categorical"],
        spec["subset"],
    )

def run_experiment_matrix(
    config: Dict[str, Any],
    raw_structured_features_by_dataset: Dict[str, pd.DataFrame],
    raw_outcome_labels_by_dataset: Dict[str, pd.DataFrame],
    raw_sensitive_attribute_by_dataset: Dict[str, pd.DataFrame],
    raw_simulation_input: pd.DataFrame,
    schema_continuous: Dict[str, List[str]],
    schema_categorical: Dict[str, List[str]],
    subset: Optional[Dict[str, Sequence[Any]]] = None,
) -> Tuple[pd.DataFrame, Dict[str, Any]]:
    """
    Execute the experiment matrix with dataset-level parallelism.

    The run count is verified against the grid product; already-logged
    runs are skipped (resumability); the completed records are aggregated
    with Task 8's aggregation.

    Parameters
    ----------
    config : Dict[str, Any]
        The study configuration.
    raw_structured_features_by_dataset : Dict[str, pd.DataFrame]
        The raw feature frames.
    raw_outcome_labels_by_dataset : Dict[str, pd.DataFrame]
        The raw label frames.
    raw_sensitive_attribute_by_dataset : Dict[str, pd.DataFrame]
        The raw sensitive frames.
    raw_simulation_input : pd.DataFrame
        The raw simulation frame.
    schema_continuous, schema_categorical : Dict[str, List[str]]
        The per-dataset feature schemas.
    subset : Optional[Dict[str, Sequence[Any]]], optional
        Dimension filters.

    Returns
    -------
    Tuple[pd.DataFrame, Dict[str, Any]]
        The aggregated summary and the run accounting.

    """
    if subset is None:
        subset = {}
    # The combination count verification first (the task's explicit rule)
    combos: List[Tuple[str, float, str]] = matrix_combinations(config, subset)
    n_combos: int = len(combos)
    num_seeds: int = int(subset.get(
        "num_seeds", config["network_training"]["num_seeds"]))
    expected_runs: int = n_combos * num_seeds
    logger.info(f"matrix: {n_combos} combinations x {num_seeds} seeds = "
                f"{expected_runs} runs.")
    # The completed keys reduce the remaining work (resumability)
    completed: set = completed_run_keys()
    expected_keys: set = {
        (ident, lam, model, seed)
        for (ident, lam, model) in combos
        for seed in range(int(config["network_training"]["seed"]),
                          int(config["network_training"]["seed"])
                          + num_seeds)
    }
    remaining: set = expected_keys - completed
    logger.info(f"resumability: {len(completed)} completed, "
                f"{len(remaining)} remaining.")
    # The per-dataset parallelism: each worker executes the pipeline for
    # ONE dataset with an isolated subset; the simulation modes run in
    # the main process. The worker spec carries every raw container so
    # pickling keeps the data ownership explicit.
    worker_specs: List[Dict[str, Any]] = [
        {
            "dataset": dataset,
            "config": config,
            "raw_structured_features_by_dataset": {
                dataset: raw_structured_features_by_dataset[dataset]},
            "raw_outcome_labels_by_dataset": {
                dataset: raw_outcome_labels_by_dataset[dataset]},
            "raw_sensitive_attribute_by_dataset": {
                dataset: raw_sensitive_attribute_by_dataset[dataset]},
            "raw_simulation_input": raw_simulation_input,
            "schema_continuous": {dataset: schema_continuous.get(dataset, [])},
            "schema_categorical": {
                dataset: schema_categorical.get(dataset, [])},
            "subset": {**subset, "datasets": [dataset], "modes": []},
        }
        for dataset in sorted(raw_structured_features_by_dataset.keys())
    ]
    result_chunks: List[StudyResults] = []
    if worker_specs and len(worker_specs) > 1:
        # Dataset-level parallelism: one worker per real dataset
        with Pool(processes=min(len(worker_specs), 4)) as pool:
            result_chunks = pool.map(_worker_run_dataset, worker_specs)
    else:
        # A single dataset runs in-process
        result_chunks = [
            _worker_run_dataset(spec) for spec in worker_specs
        ]
    # The simulation modes run in the main process after the pool
    sim_worker: Dict[str, Any] = {
        "dataset": "simulation",
        "config": config,
        "raw_structured_features_by_dataset": {},
        "raw_outcome_labels_by_dataset": {},
        "raw_sensitive_attribute_by_dataset": {},
        "raw_simulation_input": raw_simulation_input,
        "schema_continuous": {},
        "schema_categorical": {},
        "subset": {**subset, "datasets": [], "modes": list(
            subset.get("modes", config["simulation"]["modes"]))},
    }
    result_chunks.append(_worker_run_dataset(sim_worker))
    all_records: List[Dict[str, Any]] = []
    violations: List[str] = []
    for chunk in result_chunks:
        all_records.extend(chunk.run_records)
        violations.extend(chunk.violations)
    summary: pd.DataFrame = aggregate_metrics(
        normalise_metric_records(all_records),
        group_keys=("dataset", "mode", "model_class", "lambda"),
    ) if all_records else pd.DataFrame()
    accounting: Dict[str, Any] = {
        "combinations": n_combos,
        "expected_runs": expected_runs,
        "completed_count": len(completed_run_keys()),
        "remaining_count": len(remaining),
        "record_count": len(all_records),
        "violations": violations,
        "parallel_workers": len(worker_specs),
        # The per-run records are the authoritative source for the
        # paper-verification step: the aggregated summary's records
        # carry MultiIndex column names that do NOT survive a
        # reset_index/to_dict round trip as plain metric keys
        "run_records": all_records,
    }
    return summary, accounting

# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 2: Robustness verification of placeholder parameters
# -------------------------------------------------------------------------------------------------------------------------------
def build_robust_variants(
    placeholder_records: Optional[Sequence[PlaceholderRecord]] = None,
) -> List[RobustVariant]:
    """
    Build the placeholder overlay variants for the Task 12 sensitivity sweep.

    Task 12 Step 2 requires that every value annotated as unspecified in the
    manuscript is re-run under alternatives, so that no conclusion rests on
    an arbitrary choice.

    B-14 remediation. The previous variant list covered the solver step
    size, the CSER perturbation, the WoE smoothing constant and the
    alternative nonlinear specifications — which is what the task text
    names — but nothing else, and it was maintained by hand, so registering
    a new placeholder did not create its sensitivity variant. Two additions
    are prioritised because they matter most: the ORTHOGONALITY TOLERANCE,
    whose value determined whether findr could be constructed at all, and
    the WOE CONVENTION, which changes the meaning of every encoded
    categorical feature. The builder is additionally DRIVEN BY THE
    PLACEHOLDER REGISTRY when one is supplied, so a registered placeholder
    without an explicit family still receives a perturbation variant and
    cannot silently escape the sweep.

    Parameters
    ----------
    placeholder_records : Optional[Sequence[PlaceholderRecord]], optional
        The Task 1 placeholder audit. When supplied, every numeric
        placeholder lacking an explicit family below receives an
        automatic +/-50% perturbation pair, closing the registry loop.

    Returns
    -------
    List[RobustVariant]
        The variant overlays, each carrying its dotted-path overrides.

    """
    variants: List[RobustVariant] = []
    # The dotted paths that receive an explicit, hand-designed family
    explicit_paths: Set[str] = set()
    # The frontier step size family (task text)
    for rho_value in ROBUSTNESS_RHO:
        variants.append(RobustVariant(
            name=f"rho_{rho_value}",
            overrides={"frontier.solver.step_size_rho": rho_value},
        ))
    explicit_paths.add("frontier.solver.step_size_rho")
    # The CSER perturbation family (task text)
    for delta_value in ROBUSTNESS_DELTA:
        variants.append(RobustVariant(
            name=f"delta_{delta_value}",
            overrides={"interpretability_diagnostics.csert_perturbation_delta":
                       delta_value},
        ))
    explicit_paths.add(
        "interpretability_diagnostics.csert_perturbation_delta")
    # The WoE smoothing family (task text)
    for alpha_value in ROBUSTNESS_ALPHA:
        variants.append(RobustVariant(
            name=f"alpha_{alpha_value}",
            overrides={"data_preprocessing.categorical_encoding.zero_frequency_smoothing":
                       alpha_value},
        ))
    explicit_paths.add(
        "data_preprocessing.categorical_encoding.zero_frequency_smoothing")
    # B-14: the orthogonality tolerance. This is the constant that decided
    # whether the study executed at all, so it is swept explicitly
    for tolerance_value in (1e-6, 1e-5, 1e-4):
        variants.append(RobustVariant(
            name=f"orth_tol_{tolerance_value:g}",
            overrides={
                "network_architecture.orthogonalisation.relative_tolerance":
                    tolerance_value
            },
        ))
    explicit_paths.add(
        "network_architecture.orthogonalisation.relative_tolerance")
    # B-14: the WoE convention. A change of convention shifts every encoded
    # categorical feature and must be shown not to change the conclusions
    for convention in WOE_SUPPORTED_CONVENTIONS:
        variants.append(RobustVariant(
            name=f"woe_{convention}",
            overrides={
                "data_preprocessing.categorical_encoding.woe_convention":
                    convention
            },
        ))
    explicit_paths.add(
        "data_preprocessing.categorical_encoding.woe_convention")
    # The CSER perturbation space: the standardised reading is Equation
    # (14); the raw reading is the previous behaviour, retained as a check
    for space in ("standardised", "raw"):
        variants.append(RobustVariant(
            name=f"cser_space_{space}",
            overrides={
                "interpretability_diagnostics.cser_perturbation_space": space
            },
        ))
    explicit_paths.add("interpretability_diagnostics.cser_perturbation_space")
    # The CSER branch reading: total response versus structured-only
    for branch in ("total", "structured_only"):
        variants.append(RobustVariant(
            name=f"cser_branch_{branch}",
            overrides={
                "interpretability_diagnostics.cser_branch_mode": branch
            },
        ))
    explicit_paths.add("interpretability_diagnostics.cser_branch_mode")
    # The Polyak-Ruppert averaging fraction
    for fraction in (0.25, 0.5, 0.75):
        variants.append(RobustVariant(
            name=f"pr_fraction_{fraction}",
            overrides={
                "frontier.solver.polyak_ruppert_fraction": fraction
            },
        ))
    explicit_paths.add("frontier.solver.polyak_ruppert_fraction")
    # The alternative simulation specifications (linear component fixed)
    variants.append(RobustVariant(
        name="nl_spec_alt1",
        overrides={
            "simulation.nonlinear_component.functions": {
                "X1": "X1**3 / 3", "X3": "np.tanh(X3)", "X6": "-X6**2"
            },
            "simulation.nonlinear_component.index_set_I_nl": [1, 3, 6],
            "simulation.interaction_component.index_pairs_I_int": [
                [2, 6], [4, 5]
            ],
        },
    ))
    variants.append(RobustVariant(
        name="nl_spec_alt2",
        overrides={
            "simulation.nonlinear_component.functions": {
                "X2": "0.5 * X2**3", "X4": "np.sin(2*np.pi*X4)"
            },
            "simulation.nonlinear_component.index_set_I_nl": [2, 4],
            "simulation.interaction_component.index_pairs_I_int": [
                [1, 5], [2, 3]
            ],
        },
    ))
    explicit_paths.update({
        "simulation.nonlinear_component.functions",
        "simulation.nonlinear_component.index_set_I_nl",
        "simulation.interaction_component.index_pairs_I_int",
    })
    # The supplementary fairness criteria (task text)
    for alternative in ("independence", "equal_opportunity"):
        variants.append(RobustVariant(
            name=f"criterion_{alternative}",
            overrides={"fairness_penalty.active_criterion": alternative},
        ))
    explicit_paths.add("fairness_penalty.active_criterion")
    # B-14: close the registry loop. Any registered NUMERIC placeholder
    # without a hand-designed family receives an automatic +/-50% pair, so
    # a newly registered placeholder cannot silently escape the sweep
    if placeholder_records is not None:
        for entry in placeholder_records:
            # Only numeric, positive placeholders admit a scaling variant
            if entry.path in explicit_paths:
                continue
            if not isinstance(entry.value, (int, float)) or isinstance(
                    entry.value, bool):
                continue
            base_value: float = float(entry.value)
            if base_value <= 0.0:
                continue
            for scale, label in ((0.5, "half"), (2.0, "double")):
                variants.append(RobustVariant(
                    name=f"auto_{entry.path.split('.')[-1]}_{label}",
                    overrides={entry.path: base_value * scale},
                ))
    return variants

def apply_overrides(
    config: Dict[str, Any], overrides: Dict[str, Any],
) -> Dict[str, Any]:
    """
    Deep-copy a config and apply dotted-path overrides.

    Parameters
    ----------
    config : Dict[str, Any]
        The base configuration.
    overrides : Dict[str, Any]
        The dotted-path to value mapping.

    Returns
    -------
    Dict[str, Any]
        The patched deep copy; the base config is never mutated.

    """
    patched: Dict[str, Any] = copy.deepcopy(config)
    for dotted_path, value in overrides.items():
        node: Any = patched
        parts: List[str] = dotted_path.split(".")
        for part in parts[:-1]:
            node = node[part]
        node[parts[-1]] = value
    return patched

def paired_ttest_on_accuracy_gap(
    base_records: List[Dict[str, Any]],
    variant_records: List[Dict[str, Any]],
) -> Tuple[float, float]:
    """
    The paired t-test on the accuracy gap across aligned run keys.

    Task 12 Step 2 compares a sensitivity variant against the baseline on
    the same (dataset, lambda, model, seed) cells, so the test must be
    PAIRED: the two arms share every source of variation except the
    perturbed placeholder.

    B-20 remediation: the docstring promised that NaN pairs were excluded
    and no filtering was performed, so a single non-finite accuracy gap
    propagated to a NaN statistic and a NaN p-value, silently voiding the
    comparison. Non-finite pairs are now dropped, the survivor count is
    re-checked against the degrees-of-freedom requirement, and the number
    of dropped pairs is logged so the loss is visible.

    Parameters
    ----------
    base_records : List[Dict[str, Any]]
        The baseline run records.
    variant_records : List[Dict[str, Any]]
        The variant run records.

    Returns
    -------
    Tuple[float, float]
        (statistic, p-value) from scipy's paired t-test.

    Raises
    ------
    ValueError
        If fewer than two finite aligned pairs exist.

    """
    def keyed(records: List[Dict[str, Any]]) -> Dict[Tuple, float]:
        """
        Map run keys to their accuracy gaps.

        Parameters
        ----------
        records : List[Dict[str, Any]]
            The run records.

        Returns
        -------
        Dict[Tuple, float]
            (dataset, lambda, model_class, seed) -> accuracy gap.

        """
        # The four-part key identifies one experimental cell exactly
        return {
            (str(r["dataset"]), float(r["lambda"]),
             str(r["model_class"]), int(r["seed"])): float(r["accuracy_gap"])
            for r in records
        }

    base_map: Dict[Tuple, float] = keyed(base_records)
    variant_map: Dict[Tuple, float] = keyed(variant_records)
    # Only keys present in BOTH arms can be paired
    common: List[Tuple] = sorted(set(base_map) & set(variant_map))
    # B-20: drop pairs where EITHER side is non-finite, before the test.
    # Dropping only one side would break the pairing and bias the result
    finite_keys: List[Tuple] = [
        key for key in common
        if np.isfinite(base_map[key]) and np.isfinite(variant_map[key])
    ]
    dropped: int = len(common) - len(finite_keys)
    if dropped > 0:
        # The loss is reported rather than absorbed silently
        logger.warning(
            f"paired t-test: dropped {dropped} of {len(common)} aligned "
            f"pair(s) carrying a non-finite accuracy gap."
        )
    # The paired test requires at least two surviving pairs (df >= 1)
    if len(finite_keys) < 2:
        raise ValueError(
            f"paired t-test requires >= 2 aligned finite pairs; found "
            f"{len(finite_keys)} (from {len(common)} aligned key(s), "
            f"{dropped} dropped as non-finite)."
        )
    # The two arms, aligned element-by-element on the surviving keys
    base_gaps: np.ndarray = np.array(
        [base_map[key] for key in finite_keys], dtype=np.float64)
    variant_gaps: np.ndarray = np.array(
        [variant_map[key] for key in finite_keys], dtype=np.float64)
    # A zero-variance difference makes the t statistic undefined; scipy
    # returns NaN, which is reported honestly rather than masked
    statistic, p_value = scipy_stats.ttest_rel(variant_gaps, base_gaps)
    return float(statistic), float(p_value)

def diagnostic_material_change(
    base: StudyResults, variant: StudyResults,
    relative_threshold: float = MATERIAL_RELATIVE_CHANGE,
) -> List[str]:
    """
    Name any diagnostically material change between two runs.

    A material change is a > relative_threshold relative change in mean
    EVR/DDR/CSER_max per (dataset_or_mode, model, lambda), or a frontier
    ranking reversal (the order of mean excess log loss between findr
    and LR changes at any lambda).

    Parameters
    ----------
    base : StudyResults
        The baseline results.
    variant : StudyResults
        The variant results.
    relative_threshold : float, optional
        The diagnostic relative-change threshold, by default 0.10.

    Returns
    -------
    List[str]
        The named material changes; empty when none.

    """
    def diagnostics_by_group(
        results: StudyResults,
    ) -> Dict[Tuple, Tuple[float, float, float]]:
        """
        Map (identity, model, lambda) to mean (EVR, DDR, CSER max).
        """
        groups: Dict[Tuple, List[List[float]]] = {}
        for record in results.run_records:
            key: Tuple = (
                str(record["dataset"]), str(record["model_class"]),
                float(record["lambda"]),
            )
            # Diagnostics are only meaningful for structured models
            values: List[float] = [
                float(record["evr"]), float(record["ddr"]),
                float(record["cser_max"]),
            ]
            groups.setdefault(key, []).append(values)
        return {
            key: tuple(np.mean(np.array(values), axis=0))
            for key, values in groups.items()
            if np.isfinite(values).all()
        } if groups else {}
    base_map = diagnostics_by_group(base)
    variant_map = diagnostics_by_group(variant)
    changes: List[str] = []
    for key in sorted(set(base_map) & set(variant_map)):
        base_vals = base_map[key]
        variant_vals = variant_map[key]
        for label, base_val, variant_val in zip(
                ("EVR", "DDR", "CSER_max"), base_vals, variant_vals):
            # A near-zero baseline makes the relative change unstable
            if abs(base_val) > 1e-9:
                relative: float = abs(variant_val - base_val) / abs(base_val)
                if relative > relative_threshold:
                    changes.append(
                        f"{key}: {label} {base_val:.4f} -> "
                        f"{variant_val:.4f} "
                        f"(relative {relative * 100:.1f}% > "
                        f"{relative_threshold * 100:.0f}%)"
                    )
    # The ranking reversal check on the aggregated excess log loss
    def ranking(results: StudyResults) -> Dict[Tuple, str]:
        """
        Map (identity, lambda) to the lower-excess-loss model name.
        """
        grouped: Dict[Tuple, Dict[str, List[float]]] = {}
        for record in results.run_records:
            key = (str(record["dataset"]), float(record["lambda"]))
            grouped.setdefault(key, {}).setdefault(
                str(record["model_class"]), []).append(
                float(record["excess_log_loss"]))
        order: Dict[Tuple, str] = {}
        for key, by_model in grouped.items():
            means: Dict[str, float] = {
                model: float(np.mean(values))
                for model, values in by_model.items()
            }
            # The ranking is the model with the lower mean excess loss
            order[key] = min(means, key=means.get)
        return order
    base_ranking = ranking(base)
    variant_ranking = ranking(variant)
    for key in sorted(set(base_ranking) & set(variant_ranking)):
        if base_ranking[key] != variant_ranking[key]:
            changes.append(
                f"{key}: ranking reversal "
                f"{base_ranking[key]} -> {variant_ranking[key]}"
            )
    return changes

def run_robustness_checks(
    config: Dict[str, Any],
    raw_structured_features_by_dataset: Dict[str, pd.DataFrame],
    raw_outcome_labels_by_dataset: Dict[str, pd.DataFrame],
    raw_sensitive_attribute_by_dataset: Dict[str, pd.DataFrame],
    raw_simulation_input: pd.DataFrame,
    schema_continuous: Dict[str, List[str]],
    schema_categorical: Dict[str, List[str]],
    robustness_seeds: int = ROBUSTNESS_SEEDS_DEFAULT,
    subset: Optional[Dict[str, Sequence[Any]]] = None,
) -> Tuple[pd.DataFrame, Dict[str, Any]]:
    """
    Run every robustness variant and compare against the baseline.

    Parameters
    ----------
    config : Dict[str, Any]
        The base configuration.
    raw_structured_features_by_dataset : Dict[str, pd.DataFrame]
        The raw feature frames.
    raw_outcome_labels_by_dataset : Dict[str, pd.DataFrame]
        The raw label frames.
    raw_sensitive_attribute_by_dataset : Dict[str, pd.DataFrame]
        The raw sensitive frames.
    raw_simulation_input : pd.DataFrame
        The raw simulation frame.
    schema_continuous, schema_categorical : Dict[str, List[str]]
        The feature schemas.
    robustness_seeds : int, optional
        The seed count per variant, by default 3.
    subset : Optional[Dict[str, Sequence[Any]]], optional
        Dimension filters.

    Returns
    -------
    Tuple[pd.DataFrame, Dict[str, Any]]
        The tagged robustness frame and the material-change report.

    """
    if subset is None:
        subset = {}
    # The baseline run at the reduced seed count
    base_subset: Dict[str, Sequence[Any]] = {
        **subset, "num_seeds": robustness_seeds,
    }
    base_results: StudyResults = run_study_pipeline(
        config, raw_structured_features_by_dataset,
        raw_outcome_labels_by_dataset, raw_sensitive_attribute_by_dataset,
        raw_simulation_input, schema_continuous, schema_categorical,
        base_subset,
    )
    frames: List[pd.DataFrame] = []
    material: Dict[str, List[str]] = {}
    t_tests: Dict[str, Tuple[float, float]] = {}
    variants: List[RobustVariant] = build_robust_variants()
    for variant in variants:
        # The overlay config: deep-copied, never the base
        variant_config: Dict[str, Any] = apply_overrides(
            config, variant.overrides)
        variant_results: StudyResults = run_study_pipeline(
            variant_config, raw_structured_features_by_dataset,
            raw_outcome_labels_by_dataset, raw_sensitive_attribute_by_dataset,
            raw_simulation_input, schema_continuous, schema_categorical,
            base_subset,
        )
        # The material-change detectors
        changes: List[str] = diagnostic_material_change(
            base_results, variant_results)
        material[variant.name] = changes
        # The paired t-test on the accuracy gap
        try:
            statistic, p_value = paired_ttest_on_accuracy_gap(
                base_results.run_records, variant_results.run_records)
            t_tests[variant.name] = (statistic, p_value)
        except ValueError:
            # Fewer than two aligned pairs: the test is not defined
            t_tests[variant.name] = (np.nan, np.nan)
        # The frame record: one row per variant x record, tagged
        variant_frame: pd.DataFrame = pd.DataFrame(variant_results.run_records)
        variant_frame["variant"] = variant.name
        frames.append(variant_frame)
    tagged: pd.DataFrame = pd.concat(frames, ignore_index=True) if frames \
        else pd.DataFrame()
    report: Dict[str, Any] = {
        "material_changes": material,
        "paired_t_tests": t_tests,
        "variants_run": len(variants),
    }
    return tagged, report

# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 3: Verify against the paper's reported findings
# -------------------------------------------------------------------------------------------------------------------------------
def verify_against_paper(
    summary: pd.DataFrame, run_records: List[Dict[str, Any]],
    require_data: bool = False,
) -> pd.DataFrame:
    """
    Compare the aggregated results with the paper's reference patterns.

    Task 12 Step 3 names the patterns to verify:
      * simple mode: EVR ~ 1.00, DDR ~ 0.00, AND CSER ~ 0.00 at every
        fairness multiplier;
      * complex mode: EVR in [0.44, 0.56], DDR in [0.06, 0.13], AND CSER
        max INCREASING with the multiplier;
      * Taiwan findr AUC ~ 0.774 at multipliers 0 and 0.1;
      * GMSC findr AUC ~ 0.850 at multiplier 0;
      * approximately linear datasets: LR and findr close at multiplier 0.

    B-8 remediation, two defects repaired.

    (1) THE CSER CHECKS WERE ABSENT. Seven checks were implemented and none
        tested the counterfactual sign error rate, so BOTH prescribed CSER
        patterns were omitted. That matters more than a missing check
        normally would: CSER is exactly the diagnostic broken by the
        sentinel-contamination defect, so the one check that would have
        caught that bug is the one that was never written. Two checks are
        added — a simple-mode level check, and a complex-mode MONOTONICITY
        check implemented as a Spearman rank correlation against the
        multiplier, which is far more robust to seed noise than a chain of
        pairwise comparisons. CSER_min is additionally reported with its
        eligible-support count, so a pinned-to-zero minimum can never again
        be mistaken for agreement.
    (2) NOT RUN VERSUS FAIL. Absent data yielded NOT RUN, so a study that
        was never executed returned a table with no failures. The
        require_data flag distinguishes a check that is legitimately out of
        scope for a requested subset from one whose data SHOULD have been
        produced; under require_data the latter is a FAIL.

    Parameters
    ----------
    summary : pd.DataFrame
        The Task 8 aggregated summary (may be empty).
    run_records : List[Dict[str, Any]]
        The per-run records used for the group means.
    require_data : bool, optional
        When True, a check whose data is absent is reported as FAIL rather
        than NOT RUN. Set this for a full-matrix run.

    Returns
    -------
    pd.DataFrame
        One row per check: check, observed, band, status.

    """
    checks: List[Dict[str, Any]] = []
    # The record list is the authoritative group source
    frame: pd.DataFrame = pd.DataFrame(run_records) if run_records else \
        pd.DataFrame()
    if not frame.empty:
        frame["dataset"] = frame["dataset"].astype(str)

    def status_for(passed: bool, has_data: bool) -> str:
        """
        Resolve a check status from its outcome and data availability.

        Parameters
        ----------
        passed : bool
            Whether the observed value satisfied the band.
        has_data : bool
            Whether the check had any data to evaluate.

        Returns
        -------
        str
            PASS, FAIL or NOT RUN.

        """
        # No data is FAIL only when the caller asserts it should exist
        if not has_data:
            return "FAIL" if require_data else "NOT RUN"
        return "PASS" if passed else "FAIL"

    def mode_mean(mode: str, column: str) -> Tuple[float, bool]:
        """
        Mean of one diagnostic column within one simulation mode.

        Parameters
        ----------
        mode : str
            The simulation mode.
        column : str
            The diagnostic column.

        Returns
        -------
        Tuple[float, bool]
            The mean (NaN when absent) and whether any data existed.

        """
        # An empty frame or a missing column yields no data at all
        if frame.empty or column not in frame.columns:
            return float("nan"), False
        selected: pd.Series = frame.loc[frame["mode"] == mode, column]
        finite: List[float] = [
            float(v) for v in selected.tolist() if pd.notna(v)
        ]
        return (float(np.mean(finite)) if finite else float("nan"),
                bool(finite))

    # Check 1: simple mode EVR ~ 1.00
    simple_evr, has_evr = mode_mean("simple", "evr")
    checks.append({
        "check": "simple EVR ~ 1.00", "observed": simple_evr,
        "band": "0.95 - 1.05",
        "status": status_for(
            bool(pd.notna(simple_evr) and 0.95 <= simple_evr <= 1.05),
            has_evr),
    })
    # Check 2: simple mode DDR ~ 0.00
    simple_ddr, has_ddr = mode_mean("simple", "ddr")
    checks.append({
        "check": "simple DDR ~ 0.00", "observed": simple_ddr,
        "band": "0.00 - 0.05",
        "status": status_for(
            bool(pd.notna(simple_ddr) and simple_ddr <= 0.05), has_ddr),
    })
    # B-8(1): Check 3, the OMITTED simple-mode CSER level check
    simple_cser, has_cser = mode_mean("simple", "cser_max")
    checks.append({
        "check": "simple CSER (max) ~ 0.00", "observed": simple_cser,
        "band": "0.00 - 0.05",
        "status": status_for(
            bool(pd.notna(simple_cser) and simple_cser <= 0.05), has_cser),
    })
    # B-8(1): the CSER support. A statistic computed over an EMPTY eligible
    # set is not evidence of agreement, so the support is checked explicitly
    support_ok: bool = False
    support_value: float = float("nan")
    has_support: bool = False
    if not frame.empty and "cser_eligible_count" in frame.columns:
        counts: List[float] = [
            float(v) for v in frame["cser_eligible_count"].tolist()
            if pd.notna(v)
        ]
        if counts:
            has_support = True
            support_value = float(np.min(counts))
            support_ok = support_value >= 1.0
    checks.append({
        "check": "CSER eligible support non-empty",
        "observed": support_value, "band": ">= 1 feature",
        "status": status_for(support_ok, has_support),
    })
    # Check 4: complex mode EVR within [0.44, 0.56]
    complex_evr, has_cevr = mode_mean("complex", "evr")
    checks.append({
        "check": "complex EVR in [0.44, 0.56]", "observed": complex_evr,
        "band": "0.44 - 0.56",
        "status": status_for(
            bool(pd.notna(complex_evr) and 0.44 <= complex_evr <= 0.56),
            has_cevr),
    })
    # Check 5: complex mode DDR within [0.06, 0.13]
    complex_ddr, has_cddr = mode_mean("complex", "ddr")
    checks.append({
        "check": "complex DDR in [0.06, 0.13]", "observed": complex_ddr,
        "band": "0.06 - 0.13",
        "status": status_for(
            bool(pd.notna(complex_ddr) and 0.06 <= complex_ddr <= 0.13),
            has_cddr),
    })
    # B-8(1): Check 6, the OMITTED complex-mode CSER MONOTONICITY check.
    # A rank correlation against the multiplier is robust to seed noise in
    # a way that pairwise comparisons are not
    rho_value: float = float("nan")
    has_monotone: bool = False
    monotone_ok: bool = False
    if not frame.empty and {"mode", "lambda", "cser_max"} <= set(
            frame.columns):
        complex_rows: pd.DataFrame = frame[
            (frame["mode"] == "complex") & frame["cser_max"].notna()
        ]
        if not complex_rows.empty:
            # Average across seeds within each multiplier before ranking
            grouped: pd.Series = complex_rows.groupby("lambda")[
                "cser_max"].mean().sort_index()
            if len(grouped) >= 3:
                has_monotone = True
                # Spearman rho of CSER-max against the multiplier; a
                # non-decreasing relationship gives rho close to +1
                rho_value = float(scipy_stats.spearmanr(
                    grouped.index.to_numpy(dtype=np.float64),
                    grouped.to_numpy(dtype=np.float64),
                ).statistic)
                monotone_ok = bool(
                    pd.notna(rho_value) and rho_value >= 0.5)
    checks.append({
        "check": "complex CSER (max) increasing in lambda (Spearman)",
        "observed": rho_value, "band": "rho >= 0.5",
        "status": status_for(monotone_ok, has_monotone),
    })

    def model_auc(dataset: str, model_class: str,
                  multipliers: Sequence[float]) -> List[float]:
        """
        Mean test AUC per multiplier for one dataset and model class.

        Parameters
        ----------
        dataset : str
            The dataset key.
        model_class : str
            The model class.
        multipliers : Sequence[float]
            The multipliers to evaluate.

        Returns
        -------
        List[float]
            One mean per multiplier that has data.

        """
        results: List[float] = []
        for multiplier in multipliers:
            # Select the cell and average the finite AUCs across seeds
            selected: List[float] = [
                float(r["test_auc"]) for r in run_records
                if str(r["dataset"]) == dataset
                and float(r["lambda"]) == multiplier
                and str(r["model_class"]) == model_class
                and pd.notna(r.get("test_auc"))
            ]
            if selected:
                results.append(float(np.mean(selected)))
        return results

    # Check 7: Taiwan findr AUC ~ 0.774 at multipliers 0 and 0.1
    taiwan_aucs: List[float] = model_auc("taiwan", "findr", (0.0, 0.1))
    checks.append({
        "check": "Taiwan findr AUC ~ 0.774 (lambda 0, 0.1)",
        "observed": taiwan_aucs, "band": "0.754 - 0.794",
        "status": status_for(
            all(0.754 <= v <= 0.794 for v in taiwan_aucs),
            bool(taiwan_aucs)),
    })
    # Check 8: GMSC findr AUC ~ 0.850 at multiplier 0
    gmsc_aucs: List[float] = model_auc("gmsc", "findr", (0.0,))
    gmsc_mean: float = float(np.mean(gmsc_aucs)) if gmsc_aucs else float("nan")
    checks.append({
        "check": "GMSC findr AUC ~ 0.850 (lambda 0)", "observed": gmsc_mean,
        "band": "0.830 - 0.870",
        "status": status_for(
            bool(pd.notna(gmsc_mean) and 0.830 <= gmsc_mean <= 0.870),
            bool(gmsc_aucs)),
    })
    # Check 9: linear-regime LR/findr closeness at multiplier 0
    linear_diff: float = float("nan")
    has_linear: bool = False
    for dataset in ("south_german", "myhome", "thomas", "uk_credit",
                    "pakdd_2010", "vehicle_loan"):
        lr_values: List[float] = model_auc(dataset, "LR", (0.0,))
        findr_values: List[float] = model_auc(dataset, "findr", (0.0,))
        if lr_values and findr_values:
            has_linear = True
            linear_diff = abs(lr_values[0] - findr_values[0])
            break
    checks.append({
        "check": "linear-regime LR/findr AUC closeness",
        "observed": linear_diff, "band": "0.00 - 0.03",
        "status": status_for(
            bool(pd.notna(linear_diff) and linear_diff <= 0.03), has_linear),
    })
    # The check table is the fixed output contract
    return pd.DataFrame(checks)

# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def run_full_study(
    config: Dict[str, Any],
    raw_structured_features_by_dataset: Dict[str, pd.DataFrame],
    raw_outcome_labels_by_dataset: Dict[str, pd.DataFrame],
    raw_sensitive_attribute_by_dataset: Dict[str, pd.DataFrame],
    raw_simulation_input: pd.DataFrame,
    schema_continuous: Dict[str, List[str]],
    schema_categorical: Dict[str, List[str]],
    subset: Optional[Dict[str, Sequence[Any]]] = None,
) -> Dict[str, Any]:
    """
    Run the full study: matrix, robustness and paper verification.

    Parameters
    ----------
    config : Dict[str, Any]
        The study configuration.
    raw_structured_features_by_dataset : Dict[str, pd.DataFrame]
        The raw feature frames.
    raw_outcome_labels_by_dataset : Dict[str, pd.DataFrame]
        The raw label frames.
    raw_sensitive_attribute_by_dataset : Dict[str, pd.DataFrame]
        The raw sensitive frames.
    raw_simulation_input : pd.DataFrame
        The raw simulation frame.
    schema_continuous, schema_categorical : Dict[str, List[str]]
        The feature schemas.
    subset : Optional[Dict[str, Sequence[Any]]], optional
        Dimension filters (defaults to the full matrix).

    Returns
    -------
    Dict[str, Any]
        summary, accounting, robustness table, material report,
        verification table, run records.

    """
    # Step 1: the matrix with the count verification and resumability
    summary, accounting = run_experiment_matrix(
        config, raw_structured_features_by_dataset,
        raw_outcome_labels_by_dataset, raw_sensitive_attribute_by_dataset,
        raw_simulation_input, schema_continuous, schema_categorical, subset,
    )
    # Step 2: the robustness sweep (3 seeds per variant by default)
    robustness, report = run_robustness_checks(
        config, raw_structured_features_by_dataset,
        raw_outcome_labels_by_dataset, raw_sensitive_attribute_by_dataset,
        raw_simulation_input, schema_continuous, schema_categorical,
        ROBUSTNESS_SEEDS_DEFAULT, subset,
    )
    # Step 3: the paper-verification table from the authoritative per-run
    # records (never from the aggregated frame's tuple-keyed dicts)
    records: List[Dict[str, Any]] = accounting.get("run_records", [])
    verification: pd.DataFrame = verify_against_paper(
        summary, records)
    return {
        "summary": summary,
        "accounting": accounting,
        "robustness": robustness,
        "material_report": report,
        "verification": verification,
    }
````

````python
# Task 13: Final Reproducibility Audit and Deliverable Generation

# ==============================================================================
# Task 13: Final Reproducibility Audit and Deliverable Generation
# ==============================================================================

# Configure the module-level logger for stage diagnostics

# The versioned output directory created once per generation run
DEFAULT_OUTPUT_DIR: Path = Path("deliverables")
# The dependency version list per the task text
DEPENDENCIES: Tuple[str, ...] = ("torch", "numpy", "pandas", "scipy", "sklearn")
# The global seed and the replication seed population
GLOBAL_SEED: int = 42
REPLICATION_SEEDS: Tuple[int, ...] = tuple(range(10))
# The fixed precision of numeric report entries
REPORT_PRECISION: int = 6

def sha256_bytes(payload: bytes) -> str:
    """
    Return the SHA-256 hex digest of a byte payload.

    Parameters
    ----------
    payload : bytes
        The bytes to hash.

    Returns
    -------
    str
        The 64-character hex digest.

    """
    return hashlib.sha256(payload).hexdigest()

def capture_seed_and_version_record() -> Dict[str, Any]:
    """
    Capture the exact seeds and the installed dependency versions.

    Returns
    -------
    Dict[str, Any]
        The frozen record: python/platform identity, per-dependency
        version, the global seed, the replication seed population and
        the capture timestamp.

    """
    versions: Dict[str, str] = {}
    for dependency in DEPENDENCIES:
        try:
            # The distribution API reports the ACTUAL installed version
            versions[dependency] = importlib.metadata.version(dependency)
        except importlib.metadata.PackageNotFoundError:
            # An absent dependency is recorded as absent, never invented
            versions[dependency] = "NOT INSTALLED"
    return {
        "python": platform.python_version(),
        "platform": platform.platform(),
        "machine": platform.machine(),
        "dependencies": versions,
        "global_seed": GLOBAL_SEED,
        "replication_seeds": list(REPLICATION_SEEDS),
        "captured_at": datetime.now(timezone.utc).isoformat(),
    }

def json_native(value: Any) -> Any:
    """
    Convert numpy scalars to JSON-native Python values.

    Parameters
    ----------
    value : Any
        The value to normalise.

    Returns
    -------
    Any
        The JSON-native equivalent (recursive for containers).

    """
    if isinstance(value, np.integer):
        return int(value)
    if isinstance(value, np.floating):
        return float(value)
    if isinstance(value, np.bool_):
        return bool(value)
    if isinstance(value, np.ndarray):
        return value.tolist()
    if isinstance(value, dict):
        return {str(k): json_native(v) for k, v in value.items()}
    if isinstance(value, (list, tuple)):
        return [json_native(v) for v in value]
    return value

# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 1: Freeze seeds and serialise the pipeline
# -------------------------------------------------------------------------------------------------------------------------------
def serialise_artifacts(
    artifacts: Dict[str, Any],
    output_dir: Path,
) -> Dict[str, Any]:
    """
    Serialise every provided artifact and build the manifest.

    Parameters
    ----------
    artifacts : Dict[str, Any]
        The artifact payloads:
        models: Dict[kernel, Dict[str, torch.nn.Module]] keyed by
        (dataset_or_mode, lambda, model_class) with the best model;
        preprocessing_stats: dict;
        frontiers: dict of envelope arrays;
        seed_record: dict.
    output_dir : Path
        The versioned destination directory (created here).

    Returns
    -------
    Dict[str, Any]
        The immutable manifest: name, path, byte count, SHA-256.

    Raises
    ------
    ValueError
        If an artifact path duplicates an existing manifest entry or a
        hash does not match on write.

    """
    output_dir.mkdir(parents=True, exist_ok=True)
    manifest: Dict[str, Any] = {"objects": [], "written_at":
                                datetime.now(timezone.utc).isoformat()}
    # The model family: torch.save per best-model key
    for kernel, model_map in artifacts.get("models", {}).items():
        for run_key, model in model_map.items():
            name: str = f"models_{kernel}_{run_key}.pt"
            target: Path = output_dir / name
            # B-13: findr's evaluation projection B is derived state. It is
            # sealed from the FINAL encoder BEFORE the save, so the
            # serialised artifact is self-contained and a reload reproduces
            # the fitted model exactly (see B-1 Defect 2)
            if hasattr(model, "freeze_evaluation_projection"):
                model.freeze_evaluation_projection()
            # B-13: the round trip is VERIFIED, not asserted. A design
            # probe is scored before and after a save/load cycle and the
            # artifact is refused when the predictions do not agree
            probe: Optional[np.ndarray] = artifacts.get(
                "roundtrip_probe", {}).get(kernel)
            if probe is not None:
                # Capture the reference predictions from the live model
                with scoring_context(model):
                    with torch.no_grad():
                        reference: np.ndarray = forward_scores(
                            model,
                            torch.from_numpy(np.asarray(probe).astype(
                                np.float32)),
                            is_training_population=False,
                        ).detach().cpu().numpy()
                # A failure raises: an artifact that does not reload
                # identically must never be published
                verify_model_roundtrip(
                    model, copy.deepcopy(model.state_dict()),
                    np.asarray(probe), reference_predictions=reference,
                )
            # A torch state dict is written atomically via a temp file
            temp: Path = target.with_suffix(".pt.tmp")
            torch.save(model.state_dict(), temp)
            temp.replace(target)
            raw: bytes = target.read_bytes()
            manifest["objects"].append({
                "name": name,
                "path": str(target.relative_to(output_dir)),
                "bytes": len(raw),
                "sha256": sha256_bytes(raw),
            })
    # The statistics family: JSON-native
    stats_payload: Dict[str, Any] = json_native(
        artifacts.get("preprocessing_stats", {}))
    stats_target: Path = output_dir / "preprocessing_stats.json"
    stats_target.write_text(json.dumps(stats_payload, indent=2) + "\n",
                            encoding="utf-8")
    stats_raw: bytes = stats_target.read_bytes()
    manifest["objects"].append({
        "name": stats_target.name,
        "path": stats_target.name,
        "bytes": len(stats_raw),
        "sha256": sha256_bytes(stats_raw),
    })
    # The frontier family
    frontier_payload: Dict[str, Any] = json_native(
        artifacts.get("frontiers", {}))
    frontier_target: Path = output_dir / "frontiers.json"
    frontier_target.write_text(json.dumps(frontier_payload, indent=2) + "\n",
                               encoding="utf-8")
    frontier_raw: bytes = frontier_target.read_bytes()
    manifest["objects"].append({
        "name": frontier_target.name,
        "path": frontier_target.name,
        "bytes": len(frontier_raw),
        "sha256": sha256_bytes(frontier_raw),
    })
    # The seed/version record
    seed_target: Path = output_dir / "seed_and_version_record.json"
    seed_target.write_text(json.dumps(
        json_native(artifacts.get("seed_record", {})), indent=2) + "\n",
        encoding="utf-8")
    seed_raw: bytes = seed_target.read_bytes()
    manifest["objects"].append({
        "name": seed_target.name,
        "path": seed_target.name,
        "bytes": len(seed_raw),
        "sha256": sha256_bytes(seed_raw),
    })
    # The manifest is written and then self-sealed with its own digest
    manifest_raw: bytes = json.dumps(
        manifest, indent=2, sort_keys=True).encode("utf-8")
    manifest["manifest_sha256"] = sha256_bytes(manifest_raw)
    (output_dir / "manifest.json").write_text(
        json.dumps(manifest, indent=2, sort_keys=True) + "\n",
        encoding="utf-8")
    logger.info(f"manifest: {len(manifest['objects'])} objects sealed at "
                f"{output_dir}.")
    return manifest

def verify_model_roundtrip(
    model: torch.nn.Module,
    model_state: Dict[str, torch.Tensor],
    sample_input: np.ndarray,
    reference_predictions: Optional[np.ndarray] = None,
    tolerance: float = ROUNDTRIP_TOLERANCE,
) -> float:
    """
    Verify that a reloaded state reproduces the frozen predictions.

    Task 13 Step 1 requires that a downstream user can reload a serialised
    model and obtain the SAME predictions without retraining. Verifying that
    claim requires COMPARING predictions before and after the round trip.

    B-5 remediation. The previous implementation loaded a state dictionary,
    ran one forward pass, and returned the sum of the absolute predictions.
    There was no reference vector, no comparison, no tolerance and no raise:
    a checksum that is never compared to anything cannot detect a failure.
    Its docstring nevertheless claimed to return "the maximum absolute
    prediction difference; must be approximately 0", which was neither what
    it computed nor what it returned — so the name and the docstring
    together manufactured a false assurance that was then reproduced in the
    audit report.

    This implementation captures the pre-save predictions (or accepts them
    from the caller), loads the state, recomputes on the identical design in
    the identical dtype under deterministic evaluation, and RAISES when the
    maximum absolute difference exceeds the tolerance.

    Note for findr: this check depends on the evaluation projection B being
    part of the saved state. It is a registered buffer and is sealed by
    freeze_evaluation_projection before serialisation, so the round trip is
    faithful; before that repair this test would (correctly) have failed.

    Parameters
    ----------
    model : torch.nn.Module
        The model instance to load the state into. When
        reference_predictions is None its CURRENT predictions are captured
        first and used as the reference.
    model_state : Dict[str, torch.Tensor]
        The reloaded state dictionary.
    sample_input : np.ndarray
        The design used for the comparison, shape (n, p).
    reference_predictions : Optional[np.ndarray], optional
        The predictions captured from the in-memory model immediately
        before serialisation. When None they are captured here.
    tolerance : float, optional
        The maximum permitted absolute difference, by default 1e-5.

    Returns
    -------
    float
        The realised maximum absolute prediction difference, guaranteed at
        or below the tolerance when the call returns.

    Raises
    ------
    ValueError
        If the sample design is empty or misshaped.
    RuntimeError
        If the round trip does not reproduce the reference predictions.

    """
    # The comparison needs a non-empty, two-dimensional design
    design: np.ndarray = np.asarray(sample_input)
    if design.ndim != 2 or design.shape[0] == 0:
        raise ValueError(
            f"sample_input must be a non-empty 2-D design; received shape "
            f"{design.shape}."
        )
    # The dtype of the model's parameters governs the forward precision, so
    # the comparison is performed in exactly that dtype
    parameter_dtype: torch.dtype = next(model.parameters()).dtype
    numpy_dtype: Any = (
        np.float64 if parameter_dtype == torch.float64 else np.float32
    )
    design_tensor: torch.Tensor = torch.from_numpy(
        design.astype(numpy_dtype))
    # B-5(a): the reference vector. When the caller did not supply one it
    # is captured from the CURRENT in-memory model, before the state load
    if reference_predictions is None:
        with scoring_context(model):
            with torch.no_grad():
                reference_predictions = forward_scores(
                    model, design_tensor, is_training_population=False
                ).detach().cpu().numpy()
    reference: np.ndarray = np.asarray(
        reference_predictions, dtype=np.float64).reshape(-1)
    # The reference must describe the same population as the design
    if reference.shape[0] != design.shape[0]:
        raise ValueError(
            f"reference_predictions carries {reference.shape[0]} entries "
            f"against {design.shape[0]} design rows."
        )
    # B-5(b): load the state and (c) recompute deterministically. strict=True
    # so a key mismatch surfaces instead of silently leaving parameters at
    # their pre-load values
    model.load_state_dict(model_state, strict=True)
    with scoring_context(model):
        with torch.no_grad():
            reloaded: np.ndarray = forward_scores(
                model, design_tensor, is_training_population=False
            ).detach().cpu().numpy()
    # B-5(d): the actual comparison the function's name promises
    difference: np.ndarray = np.abs(
        np.asarray(reloaded, dtype=np.float64).reshape(-1) - reference)
    max_difference: float = float(difference.max())
    # A non-finite difference is a failure regardless of the tolerance
    if not np.isfinite(max_difference) or max_difference > tolerance:
        # Name the worst row so the failure is actionable
        worst: int = int(np.argmax(difference))
        raise RuntimeError(
            f"model round-trip failed: max|reloaded - reference| = "
            f"{max_difference:.3e} exceeds the tolerance {tolerance:.3e} "
            f"(worst at row {worst}: reference {reference[worst]:.8f} vs "
            f"reloaded {float(reloaded[worst]):.8f}). The serialised "
            f"artifact does not reproduce the fitted model and must not be "
            f"published."
        )
    return max_difference

def verify_manifest(manifest: Dict[str, Any], output_dir: Path) -> List[str]:
    """
    Verify every manifest entry exists and hashes to its record.

    Parameters
    ----------
    manifest : Dict[str, Any]
        The written manifest.
    output_dir : Path
        The artifacts directory.

    Returns
    -------
    List[str]
        Any verification findings; empty when every entry is intact.

    """
    findings: List[str] = []
    for entry in manifest["objects"]:
        target: Path = output_dir / entry["path"]
        if not target.exists():
            findings.append(f"missing object: {entry['path']}")
            continue
        digest: str = sha256_bytes(target.read_bytes())
        if digest != entry["sha256"]:
            findings.append(
                f"hash mismatch: {entry['path']} recorded "
                f"{entry['sha256'][:16]}... found {digest[:16]}..."
            )
    return findings

# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 2: Generate final tables and figures
# -------------------------------------------------------------------------------------------------------------------------------
def generate_tables(
    summary: pd.DataFrame,
    diagnostics: pd.DataFrame,
    output_dir: Path,
) -> List[Path]:
    """
    Write the final metric tables as CSV artifacts.

    Task 13 Step 2 requires the AUC and Brier paths, the EVR/DDR/CSER
    tables and the coefficient paths to be emitted as durable artifacts.

    B-22 remediation: this routine had no defect of its own, but its CSER
    columns were contaminated by the sentinel bug upstream (B-2/B-3). Two
    changes make that class of failure visible on the face of the output:
    the eligible-feature count is carried as a column when present, so a
    CSER_min pinned to zero over an empty or partial support is
    immediately apparent; and a CSER column that is entirely zero triggers
    an explicit warning rather than passing unremarked, because that is the
    exact signature of the defect that manufactured agreement with the
    manuscript's published table.

    Parameters
    ----------
    summary : pd.DataFrame
        The aggregated metric summary.
    diagnostics : pd.DataFrame
        The diagnostics table (EVR, DDR, CSER min/max rows).
    output_dir : Path
        The destination directory, created when absent.

    Returns
    -------
    List[Path]
        The written table paths.

    Raises
    ------
    OSError
        If the destination cannot be created or written.

    """
    # The destination must exist before any write is attempted
    output_dir.mkdir(parents=True, exist_ok=True)
    written: List[Path] = []
    # The aggregated metric summary
    if not summary.empty:
        summary_target: Path = output_dir / "metric_summary.csv"
        summary.to_csv(summary_target)
        written.append(summary_target)
    # The diagnostics table, with the CSER support surfaced
    if not diagnostics.empty:
        annotated: pd.DataFrame = diagnostics.copy()
        # B-22: an all-zero CSER column is the SIGNATURE of the
        # sentinel-contamination defect, but it is also the mathematically
        # correct answer for a pure logistic regression, where
        # eta = x'beta and the perturbation +sign(beta_j) * delta * e_j
        # raises the logit by delta * |beta_j| > 0 for EVERY row, so
        # CSER_j = 0 exactly. The two cases are distinguished by the
        # SUPPORT, not by the values: a zero over an empty or unknown
        # eligible set is contamination, a zero over a known non-empty set
        # is a result. Only the former is flagged, so the guard does not
        # cry wolf on every LR run.
        support: Optional[pd.Series] = None
        if "cser_eligible_count" in annotated.columns:
            support = pd.to_numeric(
                annotated["cser_eligible_count"], errors="coerce")
        for column in ("cser_min", "cser_max"):
            if column not in annotated.columns:
                continue
            values: pd.Series = pd.to_numeric(
                annotated[column], errors="coerce")
            observed: pd.Series = values.dropna()
            # Only an all-zero column is a candidate for the warning
            if len(observed) < 2 or not bool((observed == 0.0).all()):
                continue
            if support is None:
                # The support was not recorded, so the zero cannot be
                # attributed and the operator must check it
                logger.warning(
                    f"diagnostics table: every '{column}' entry is exactly "
                    f"0.0 across {len(observed)} rows and no "
                    f"'cser_eligible_count' column is present, so the "
                    f"result cannot be distinguished from sentinel "
                    f"contamination. Record the eligible-feature count."
                )
            elif bool((support.fillna(0) < 1).any()):
                # A zero over an EMPTY eligible set is contamination
                logger.warning(
                    f"diagnostics table: every '{column}' entry is exactly "
                    f"0.0 across {len(observed)} rows AND at least one row "
                    f"reports an empty CSER eligible set. This is the "
                    f"signature of sentinel contamination in the "
                    f"per-feature mapping; the statistic was not computed."
                )
            else:
                # A zero over a known non-empty set is a genuine result
                logger.info(
                    f"diagnostics table: '{column}' is 0.0 across "
                    f"{len(observed)} rows over a non-empty eligible set "
                    f"(min support {int(support.min())} feature(s)). This "
                    f"is a genuine result, expected for a purely linear "
                    f"model where every perturbation raises the logit."
                )
        # The eligible-feature count makes the support of CSER explicit
        if "cser_eligible_count" in annotated.columns:
            annotated["cser_eligible_count"] = pd.to_numeric(
                annotated["cser_eligible_count"], errors="coerce"
            ).astype("Int64")
        diag_target: Path = output_dir / "diagnostics_table.csv"
        annotated.to_csv(diag_target)
        written.append(diag_target)
    return written

def generate_figures(
    summary: pd.DataFrame,
    frontiers: Dict[str, Any],
    diagnostics: pd.DataFrame,
    coefficient_paths: Dict[str, Any],
    simulation_densities: Dict[str, Any],
    plot_config: Dict[str, Any],
    output_dir: Path,
) -> List[Path]:
    """
    Generate the final study figures per the plot conventions.

    The families: AUC/Brier lambda paths; frontiers with model paths;
    diagnostics tables rendered as figures (the CSV is the tables);
    coefficient paths; simulation score densities. Colours follow the
    config's plot_construction block; vector formats at the configured
    dpi.

    Parameters
    ----------
    summary : pd.DataFrame
        The aggregated summary.
    frontiers : Dict[str, Any]
        The per-dataset/mode frontier dumps.
    diagnostics : pd.DataFrame
        The diagnostics table.
    coefficient_paths : Dict[str, Any]
        Per-feature coefficient path series.
    simulation_densities : Dict[str, Any]
        Per-mode density samples keyed by (group, outcome).
    plot_config : Dict[str, Any]
        The config plot_construction block.
    output_dir : Path
        The destination directory.

    Returns
    -------
    List[Path]
        The written figure paths.

    Raises
    ------
    ImportError
        If matplotlib is unavailable.

    """
    try:
        # The backend is pinned so vector output works headless
        matplotlib.use("Agg")
    except ImportError as exc:
        raise ImportError(
            "matplotlib is required for the deliverable figures and was "
            "not found in the environment."
        ) from exc
    colours: Dict[str, str] = plot_config.get("colours", {})
    dpi: int = int(plot_config.get("dpi", 150))
    figsize: Sequence[float] = plot_config.get("figsize", (12.0, 8.0))
    written: List[Path] = []
    # Family 1: the AUC and Brier lambda paths per dataset. The summary
    # carries the group keys as INDEX levels and the metrics as a
    # MultiIndex COLUMN block, so the plot frame is the reset-indexed
    # form and the metric columns are accessed as tuples
    if not summary.empty:
        plot_frame: pd.DataFrame = summary.reset_index()
        fig, axes = plt.subplots(2, 1, figsize=(
            figsize[0], figsize[1]), dpi=dpi)
        # The paths are drawn per model class from the plot frame
        for axis_index, metric in enumerate(("auc", "brier")):
            for model_class, colour in (
                    ("findr", colours.get("findr", "blue")),
                    ("LR", colours.get("LR", "green")),
                    ("NN", colours.get("NN", "orange"))):
                sub = plot_frame[plot_frame["model_class"] == model_class]
                if sub.empty:
                    continue
                for dataset in sub["dataset"].unique():
                    path = sub[sub["dataset"]
                               == dataset].sort_values("lambda")
                    # The mean column of this metric at this lambda:
                    # aggregated frames carry the (metric, "mean") tuple
                    # column; a plain-frame smoke carries the bare name
                    metric_column = (metric, "mean") \
                        if (metric, "mean") in path.columns else metric
                    metric_series = path[metric_column]
                    axes[axis_index].plot(
                        path["lambda"], metric_series, color=colour,
                        marker="o", markersize=2,
                        label=f"{model_class} ({dataset})" if
                        axis_index == 0 else None,
                    )
            axes[axis_index].set_xlabel("lambda")
            axes[axis_index].set_ylabel(metric)
        target_metrics: Path = output_dir / "lambda_paths.pdf"
        fig.savefig(target_metrics, dpi=dpi, format="pdf")
        plt.close(fig)
        written.append(target_metrics)
    # Family 2: the score-level frontiers with model paths
    if frontiers:
        fig, axes = plt.subplots(1, 1, figsize=(
            figsize[0], figsize[1]), dpi=dpi)
        for identity, frontier_units in frontiers.items():
            envelope_phi = frontier_units.get("envelope_phi")
            envelope_loss = frontier_units.get("envelope_logloss")
            if envelope_phi is None or envelope_loss is None:
                continue
            axes.plot(envelope_phi, envelope_loss,
                      color=colours.get("frontier", "black"), lw=2,
                      label=f"frontier ({identity})" if
                      frontiers is not None else None)
            for model_path in frontier_units.get("model_paths", []):
                # The model paths are the (phi, loss) traces per model
                axes.plot(model_path["phi"], model_path["loss"],
                          color=colours.get(model_path["model_class"],
                                            "grey"),
                          marker="o", markersize=2)
        axes.set_xlabel("Phi (equalised-odds Wasserstein)")
        axes.set_ylabel("excess log loss")
        target_frontier: Path = output_dir / "frontiers.pdf"
        fig.savefig(target_frontier, dpi=dpi, format="pdf")
        plt.close(fig)
        written.append(target_frontier)
    # Family 3: the diagnostics figure (the CSV is the canonical table)
    if not diagnostics.empty and {"evr", "ddr", "cser_min", "cser_max"} <= \
            set(diagnostics.columns):
        fig, axes = plt.subplots(1, 2, figsize=(
            figsize[0], figsize[1]), dpi=dpi)
        for axis_index, metric in enumerate(("evr", "ddr")):
            values = diagnostics[metric].fillna(0.0).to_numpy(dtype=float)
            line = axes[axis_index].bar(
                np.arange(len(values)), values,
                color=colours.get("findr", "blue"))
            axes[axis_index].set_ylabel(metric)
            _ = line
        target_diag: Path = output_dir / "diagnostics.pdf"
        fig.savefig(target_diag, dpi=dpi, format="pdf")
        plt.close(fig)
        written.append(target_diag)
    # Family 4: the coefficient paths (representative datasets)
    if coefficient_paths:
        fig, axes = plt.subplots(1, 1, figsize=(
            figsize[0], figsize[1]), dpi=dpi)
        for dataset, coefficients in coefficient_paths.items():
            # Each feature's path over the lambda sequence
            for feature, series in coefficients.items():
                axes.plot(np.arange(len(series)), series,
                          marker="o", markersize=2,
                          label=f"{dataset}:{feature}")
        axes.set_xlabel("lambda index")
        axes.set_ylabel("structured coefficient")
        target_coef: Path = output_dir / "coefficient_paths.pdf"
        fig.savefig(target_coef, dpi=dpi, format="pdf")
        plt.close(fig)
        written.append(target_coef)
    # Family 5: the simulation score densities, viridis by lambda
    if simulation_densities:
        fig, axes = plt.subplots(1, 1, figsize=(
            figsize[0], figsize[1]), dpi=dpi)
        lambda_values = sorted(simulation_densities.keys(), key=float)
        colormap = matplotlib_cm.get_cmap(plot_config.get(
            "lambda_colormap", "viridis"))
        for position, lambda_key in enumerate(lambda_values):
            samples = np.asarray(simulation_densities[lambda_key])
            if samples.size < 2 or gaussian_kde is None:
                continue
            try:
                kde = gaussian_kde(samples)
                grid = np.linspace(float(samples.min()),
                                   float(samples.max()), 200)
                axes.plot(grid, kde(grid),
                          color=colormap(position / max(len(lambda_values) - 1, 1)))
            except (ValueError, np.linalg.LinAlgError) as exc:
                logger.warning(
                    f"density for lambda {lambda_key} skipped: {exc}."
                )
        axes.set_xlabel("score")
        axes.set_ylabel("density")
        target_density: Path = output_dir / "simulation_densities.pdf"
        fig.savefig(target_density, dpi=dpi, format="pdf")
        plt.close(fig)
        written.append(target_density)
    for target in written:
        logger.info(f"figure written: {target.name}")
    return written

# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 3: The final audit report
# -------------------------------------------------------------------------------------------------------------------------------
def render_audit_report(
    sections: Dict[str, str],
    output_dir: Path,
) -> Path:
    """
    Render and seal the final audit report.

    The report is composed from the provided section texts (all of which
    are rendered from the stored logs by the caller) in the fixed order,
    timestamped, and sealed with the SHA-256 digest of its body.

    Parameters
    ----------
    sections : Dict[str, str]
        The fixed-order section texts: raw_data_checks,
        placeholder_parameters, robustness, metric_tables,
        orchestrator_call_sequence, seed_and_versions.
    output_dir : Path
        The destination directory.

    Returns
    -------
    Path
        The written report, sealed.

    """
    order: Tuple[str, ...] = (
        "raw_data_checks", "placeholder_parameters", "robustness",
        "metric_tables", "orchestrator_call_sequence",
        "seed_and_versions",
    )
    lines: List[str] = [
        "# Final Reproducibility Audit Report",
        "",
        f"**Generated at:** {datetime.now(timezone.utc).isoformat()}",
        "",
        "This report is generated programmatically from the stored logs; "
        "no value below is hand-written.",
        "",
    ]
    for section in order:
        lines.append(f"## {section.replace('_', ' ').title()}")
        lines.append("")
        lines.append(sections.get(section, "NOT AVAILABLE"))
        lines.append("")
    body: str = "\n".join(lines)
    digest: str = sha256_bytes(body.encode("utf-8"))
    # The digest line covers the body; the seal itself is not hashed
    sealed: str = body + f"\nSHA-256 (body): {digest}\n"
    target: Path = output_dir / "final_audit_report.md"
    target.write_text(sealed, encoding="utf-8")
    logger.info(f"audit report sealed at {target} (digest {digest[:16]}...).")
    return target

# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def produce_final_deliverables(
    artifacts: Dict[str, Any],
    summary: pd.DataFrame,
    diagnostics: pd.DataFrame,
    frontiers: Dict[str, Any],
    coefficient_paths: Dict[str, Any],
    simulation_densities: Dict[str, Any],
    plot_config: Dict[str, Any],
    output_dir: Path = DEFAULT_OUTPUT_DIR,
) -> Dict[str, Any]:
    """
    Produce the full deliverable set: artifacts, tables, figures, report.

    Parameters
    ----------
    artifacts : Dict[str, Any]
        The model/stats/frontier/seed artifact payloads (see
        serialise_artifacts).
    summary : pd.DataFrame
        The aggregated metric summary.
    diagnostics : pd.DataFrame
        The diagnostics table.
    frontiers : Dict[str, Any]
        The frontier dumps.
    coefficient_paths : Dict[str, Any]
        The coefficient path series per representative dataset.
    simulation_densities : Dict[str, Any]
        The simulation density samples per lambda.
    plot_config : Dict[str, Any]
        The plot_construction config block.
    output_dir : Path, optional
        The versioned destination directory, by default deliverables.

    Returns
    -------
    Dict[str, Any]
        The manifest, the written table and figure paths, the report
        path and any verification findings.

    """
    # Step 1: serialise and seal the artifacts plus the manifest
    manifest: Dict[str, Any] = serialise_artifacts(artifacts, output_dir)
    findings: List[str] = verify_manifest(manifest, output_dir)
    # Step 2: the tables and the figures
    table_paths: List[Path] = generate_tables(summary, diagnostics, output_dir)
    figure_paths: List[Path] = generate_figures(
        summary, frontiers, diagnostics, coefficient_paths,
        simulation_densities, plot_config, output_dir,
    )
    # Step 3: the audit report sections from the stored inputs
    seed_record: Dict[str, Any] = artifacts.get("seed_record", {})
    sections: Dict[str, str] = {
        "raw_data_checks": (
            "All raw-data checks performed (" +
            "; ".join(f"{name}: {status}" for name, status in (
                artifacts.get("data_checks", {}).items())) + ")"),
        "placeholder_parameters": json.dumps(
            json_native(artifacts.get("placeholder_audit", {})),
            indent=2),
        "robustness": json.dumps(
            json_native(artifacts.get("robustness_report", {})),
            indent=2),
        "metric_tables": summary.to_markdown() if not summary.empty
        else "NOT AVAILABLE",
        "orchestrator_call_sequence": artifacts.get(
            "call_sequence", "NOT AVAILABLE"),
        "seed_and_versions": json.dumps(
            json_native(seed_record), indent=2),
    }
    report_path: Path = render_audit_report(sections, output_dir)
    return {
        "manifest": manifest,
        "verification_findings": findings,
        "tables": [str(p) for p in table_paths],
        "figures": [str(p) for p in figure_paths],
        "report": str(report_path),
    }
````

````python
# Task 14: Top-Level End-to-End Orchestrator for the findr Study

# ==============================================================================
# Task 14: Top-Level End-to-End Orchestrator for the findr Study
# ==============================================================================

# Make the sibling task modules importable as the study pipeline grows

# Configure the module-level logger for stage diagnostics

# The fixed step ids of the end-to-end execution graph
STEP_IDS: Tuple[str, ...] = (
    "tasks_1_2", "task_3", "task_4", "tasks_5_10_engine_task_11",
    "task_12", "task_13",
)

@dataclass
class ExecutionStep:
    """
    One stage's execution record.

    Attributes
    ----------
    step_id : str
        The fixed step identifier.
    description : str
        The stage's task coverage.
    status : str
        One of COMPLETED, FAILED, UNRUN.
    artifacts : Dict[str, Any]
        The stage's output artifacts.
    exception : str
        The exception type and message when the stage failed, else empty.

    """

    # The fixed step identifier
    step_id: str
    # The stage's task coverage
    description: str
    # The stage status
    status: str = "UNRUN"
    # The stage's output artifacts
    artifacts: Dict[str, Any] = field(default_factory=dict)
    # The failure detail, when any
    exception: str = ""

# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Step 1 (Tasks 1-2): validation and cleansing
# -------------------------------------------------------------------------------------------------------------------------------
def step_tasks_1_2_validate_and_clean(
    config: Dict[str, Any],
    raw_structured_features_by_dataset: Dict[str, pd.DataFrame],
    raw_outcome_labels_by_dataset: Dict[str, pd.DataFrame],
    raw_sensitive_attribute_by_dataset: Dict[str, pd.DataFrame],
    raw_simulation_input: pd.DataFrame,
    schema_continuous: Dict[str, List[str]],
    schema_categorical: Dict[str, List[str]],
) -> ExecutionStep:
    """
    Run Task 1 (validate_inputs) and Task 2 (clean_raw_data).

    Parameters
    ----------
    config : Dict[str, Any]
        The study configuration.
    raw_structured_features_by_dataset : Dict[str, pd.DataFrame]
        The raw feature frames.
    raw_outcome_labels_by_dataset : Dict[str, pd.DataFrame]
        The raw label frames.
    raw_sensitive_attribute_by_dataset : Dict[str, pd.DataFrame]
        The raw sensitive frames.
    raw_simulation_input : pd.DataFrame
        The raw simulation frame.
    schema_continuous, schema_categorical : Dict[str, List[str]]
        The per-dataset feature schemas.

    Returns
    -------
    ExecutionStep
        The recorded stage with the ValidationReport, the CleanedDataBundle
        and the audit artifact paths.

    """
    step = ExecutionStep(
        step_id="tasks_1_2", description="Tasks 1-2: validation and cleansing",
    )
    try:
        # Task 1: key integrity, dtype checks, metadata statistics and the
        # placeholder audit
        validation_report = validate_inputs(
            config, raw_structured_features_by_dataset,
            raw_outcome_labels_by_dataset, raw_sensitive_attribute_by_dataset,
            raw_simulation_input, schema_continuous, schema_categorical,
            config["real_data_experiment"]["datasets"],
            run_seeds=range(10),
            audit_log_path=Path("placeholders_audit.json"),
        )
        # Task 2: duplicate removal, plausibility rules, missing indicators
        # and the cleansing audit
        cleaned_bundle = clean_raw_data(
            validation_report, config["real_data_experiment"]["datasets"],
            schema_continuous, schema_categorical,
            cleansing_artifact_path=Path("cleansing_reports.json"),
        )
        # The stage artifacts: the two containers and their audit paths
        step.artifacts = {
            "validation_report": validation_report,
            "cleaned_bundle": cleaned_bundle,
            "placeholder_audit_path": "placeholders_audit.json",
            "cleansing_report_path": "cleansing_reports.json",
            "datasets_cleaned": list(cleaned_bundle.cleaned_datasets.keys()),
        }
        step.status = "COMPLETED"
    except Exception as exc:  # noqa: BLE001 - the stage record must own it
        # A stage failure is recorded, never silently lost
        step.status = "FAILED"
        step.exception = f"{type(exc).__name__}: {exc}"
        logger.error(f"step tasks_1_2 failed: {step.exception}")
    return step

# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Step 2 (Task 3): preprocessing of the real datasets
# -------------------------------------------------------------------------------------------------------------------------------
def step_task_3_preprocess(
    cleaned_bundle: Any,
    config: Dict[str, Any],
    schema_continuous: Dict[str, List[str]],
    schema_categorical: Dict[str, List[str]],
) -> ExecutionStep:
    """
    Run Task 3 (preprocess_real_data) over the cleaned datasets.

    Parameters
    ----------
    cleaned_bundle : Any
        The Task 2 CleanedDataBundle.
    config : Dict[str, Any]
        The study configuration.
    schema_continuous, schema_categorical : Dict[str, List[str]]
        The per-dataset feature schemas.

    Returns
    -------
    ExecutionStep
        The recorded stage with the per-dataset PreprocessedDataBundle
        dict and the statistics artifact path.

    """
    step = ExecutionStep(
        step_id="task_3", description="Task 3: real-data preprocessing",
    )
    # Task 3 requires the bundle to exist: a failed Task 1-2 stage leaves
    # this stage unrun rather than raising a misleading downstream error
    if cleaned_bundle is None or not hasattr(cleaned_bundle, "cleaned_datasets"):
        step.status = "UNRUN"
        step.exception = "prerequisite tasks_1_2 did not complete"
        return step
    try:
        # Task 3 per dataset: frozen training statistics, WoE, scaling and
        # the final tensors with provenance; the config is the first
        # argument of the C2-re-implemented orchestrator
        bundles: Dict[str, PreprocessedDataBundle] = preprocess_real_data(
            config, cleaned_bundle,
            config["real_data_experiment"]["datasets"],
            schema_continuous, schema_categorical,
            split_config=config["data_preprocessing"]["train_val_test_split"],
            artifact_path=Path("preprocessing_stats.json"),
        )
        step.artifacts = {
            "bundles": bundles,
            "statistics_artifact_path": "preprocessing_stats.json",
            "datasets_preprocessed": list(bundles.keys()),
        }
        step.status = "COMPLETED"
    except Exception as exc:  # noqa: BLE001
        step.status = "FAILED"
        step.exception = f"{type(exc).__name__}: {exc}"
        logger.error(f"step task_3 failed: {step.exception}")
    return step

# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Step 3 (Task 4): simulation preparation
# -------------------------------------------------------------------------------------------------------------------------------
def step_task_4_simulation(
    validation_report: Any,
    config: Dict[str, Any],
) -> ExecutionStep:
    """
    Run Task 4 (prepare_simulation_data).

    Parameters
    ----------
    validation_report : Any
        The Task 1 ValidationReport carrying the validated simulation frame.
    config : Dict[str, Any]
        The study configuration.

    Returns
    -------
    ExecutionStep
        The recorded stage with the SimulationDataBundle.

    """
    step = ExecutionStep(
        step_id="task_4", description="Task 4: simulation preparation",
    )
    if validation_report is None:
        step.status = "UNRUN"
        step.exception = "prerequisite tasks_1_2 did not complete"
        return step
    try:
        # Task 4: structural validation, shift implication, logit and
        # reference-risk verification and per-block stratified splits
        simulation_bundle: SimulationDataBundle = prepare_simulation_data(
            validation_report.simulation_with_record_id,
            config["data_preprocessing"]["train_val_test_split"],
            n_observations=int(config["simulation"].get(
                "n_observations", 10000)),
            run_seeds=range(10),
            expected_modes=list(config["simulation"].get("modes", [])),
        )
        step.artifacts = {
            "simulation_bundle": simulation_bundle,
            "blocks": list(simulation_bundle.blocks.keys()),
        }
        step.status = "COMPLETED"
    except Exception as exc:  # noqa: BLE001
        step.status = "FAILED"
        step.exception = f"{type(exc).__name__}: {exc}"
        logger.error(f"step task_4 failed: {step.exception}")
    return step

# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Step 4 (Tasks 5-10 via Task 11): the study run engine
# -------------------------------------------------------------------------------------------------------------------------------
def step_tasks_5_10_engine(
    cleaned_bundle: Any,
    validation_report: Any,
    raw_structured_features_by_dataset: Dict[str, pd.DataFrame],
    raw_outcome_labels_by_dataset: Dict[str, pd.DataFrame],
    raw_sensitive_attribute_by_dataset: Dict[str, pd.DataFrame],
    raw_simulation_input: pd.DataFrame,
    config: Dict[str, Any],
    schema_continuous: Dict[str, List[str]],
    schema_categorical: Dict[str, List[str]],
    subset: Dict[str, Sequence[Any]],
) -> ExecutionStep:
    """
    Run Tasks 5-10 through Task 11's run_study_pipeline, once.

    The engine owns the non-duplicative execution: build_model and
    compute_loss are consumed inside Task 7's search and training,
    evaluate_model per seed, compute_diagnostics per structured model and
    construct_frontier per dataset/mode — all called by the engine in
    their prescribed order.

    Parameters
    ----------
    cleaned_bundle : Any
        The Task 2 CleanedDataBundle.
    validation_report : Any
        The Task 1 ValidationReport.
    raw_structured_features_by_dataset : Dict[str, pd.DataFrame]
        The RAW feature frames (Task 11's pipeline contract expects raw
        containers: its own Task 1/2 stages re-run the validation on them).
    raw_outcome_labels_by_dataset : Dict[str, pd.DataFrame]
        The RAW label frames.
    raw_sensitive_attribute_by_dataset : Dict[str, pd.DataFrame]
        The RAW sensitive frames.
    raw_simulation_input : pd.DataFrame
        The RAW simulation frame.
    config : Dict[str, Any]
        The study configuration.
    schema_continuous, schema_categorical : Dict[str, List[str]]
        The per-dataset feature schemas.
    subset : Dict[str, Sequence[Any]]
        The dimension filters (datasets, modes, lambdas, models,
        num_seeds).

    Returns
    -------
    ExecutionStep
        The recorded stage with the StudyResults (run records, summary,
        frontiers, violations).

    """
    step = ExecutionStep(
        step_id="tasks_5_10_engine_task_11",
        description="Tasks 5-10 engine: models, loss, training, "
                    "evaluation, diagnostics, frontiers (via Task 11)",
    )
    if cleaned_bundle is None:
        step.status = "UNRUN"
        step.exception = "prerequisite tasks_1_2 did not complete"
        return step
    try:
        # Task 11: the research pipeline graph (the single call)
        study_results: StudyResults = run_study_pipeline(
            config,
            raw_structured_features_by_dataset,
            raw_outcome_labels_by_dataset,
            raw_sensitive_attribute_by_dataset,
            raw_simulation_input,
            schema_continuous, schema_categorical, subset,
        )
        step.artifacts = {
            "study_results": study_results,
            "run_record_count": len(study_results.run_records),
            "frontier_keys": list(study_results.frontiers.keys()),
            "violation_count": len(study_results.violations),
        }
        step.status = "COMPLETED"
    except Exception as exc:  # noqa: BLE001
        step.status = "FAILED"
        # The full traceback is recorded so the engine's failure is
        # diagnosable without re-running the stage
        step.exception = traceback.format_exc()
        logger.error(f"step tasks_5_10_engine failed: {exc}")
    return step

# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Step 5 (Task 12): full-study accounting, robustness, verification
# -------------------------------------------------------------------------------------------------------------------------------
def step_task_12_full_study(
    config: Dict[str, Any],
    raw_structured_features_by_dataset: Dict[str, pd.DataFrame],
    raw_outcome_labels_by_dataset: Dict[str, pd.DataFrame],
    raw_sensitive_attribute_by_dataset: Dict[str, pd.DataFrame],
    raw_simulation_input: pd.DataFrame,
    schema_continuous: Dict[str, List[str]],
    schema_categorical: Dict[str, List[str]],
    subset: Dict[str, Sequence[Any]],
) -> ExecutionStep:
    """
    Run Task 12 (run_full_study).

    The matrix pass is idempotent against the run log: the keys Step 4
    logged are skipped by Task 12's resumability scan, so no duplicate
    run is executed; the accounting reports the completed/remaining split.

    Parameters
    ----------
    config : Dict[str, Any]
        The study configuration.
    raw_structured_features_by_dataset : Dict[str, pd.DataFrame]
        The raw feature frames.
    raw_outcome_labels_by_dataset : Dict[str, pd.DataFrame]
        The raw label frames.
    raw_sensitive_attribute_by_dataset : Dict[str, pd.DataFrame]
        The raw sensitive frames.
    raw_simulation_input : pd.DataFrame
        The raw simulation frame.
    schema_continuous, schema_categorical : Dict[str, List[str]]
        The per-dataset feature schemas.
    subset : Dict[str, Sequence[Any]]
        The dimension filters (the SAME subset the engine used).

    Returns
    -------
    ExecutionStep
        The recorded stage with the accounting, the robustness table, the
        material report and the verification table.

    """
    step = ExecutionStep(
        step_id="task_12", description="Task 12: full study, robustness "
                                       "and paper verification",
    )
    try:
        # Task 12: matrix accounting (resumable), robustness sweeps and
        # the paper verification table
        full_study: Dict[str, Any] = run_full_study(
            config, raw_structured_features_by_dataset,
            raw_outcome_labels_by_dataset, raw_sensitive_attribute_by_dataset,
            raw_simulation_input, schema_continuous, schema_categorical,
            subset,
        )
        step.artifacts = full_study
        step.status = "COMPLETED"
    except Exception as exc:  # noqa: BLE001
        step.status = "FAILED"
        # The full traceback is recorded for diagnosis without re-runs
        step.exception = traceback.format_exc()
        logger.error(f"step task_12 failed: {exc}\n{step.exception}")
    return step

# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Step 6 (Task 13): deliverables
# -------------------------------------------------------------------------------------------------------------------------------
def step_task_13_deliverables(
    study_results: Optional[StudyResults],
    full_study: Optional[Dict[str, Any]],
    config: Dict[str, Any],
    plot_config: Dict[str, Any],
    output_dir: Path,
    preserved_models: Optional[Dict[str, torch.nn.Module]] = None,
) -> ExecutionStep:
    """
    Run Task 13 (produce_final_deliverables).

    Parameters
    ----------
    study_results : Optional[StudyResults]
        The Task 11 study results (summary, frontiers).
    full_study : Optional[Dict[str, Any]]
        The Task 12 full-study dict (robustness report, verification).
    config : Dict[str, Any]
        The study configuration.
    plot_config : Dict[str, Any]
        The plot_construction block.
    output_dir : Path
        The versioned deliverables directory.
    preserved_models : Optional[Dict[str, torch.nn.Module]], optional
        Best models preserved from a checkpointed production run, keyed
        by (kernel, run key); absent models are recorded honestly.

    Returns
    -------
    ExecutionStep
        The recorded stage with the manifest, the figures/tables paths,
        the report path and the verification findings.

    """
    step = ExecutionStep(
        step_id="task_13", description="Task 13: reproducibility audit "
                                       "and deliverables",
    )
    # Task 13 requires at least the summary to render the report tables
    have_summary: bool = study_results is not None and \
        not study_results.summary.empty
    if not have_summary:
        step.status = "UNRUN"
        step.exception = "prerequisite study results unavailable"
        return step
    try:
        # The seed/version record freezes the reproduction state
        seed_record: Dict[str, Any] = capture_seed_and_version_record()
        # The artifacts for Task 13: preserved models (or an honest empty
        # record), the preprocessing statistics, the frontiers, the seed
        # record, the data-checks and the placeholder audit
        artifacts: Dict[str, Any] = {
            "models": preserved_models or {},
            "preprocessing_stats": _load_json_safe(
                Path("preprocessing_stats.json")),
            "frontiers": _frontier_dumps(study_results),
            "seed_record": seed_record,
            "data_checks": {"cleaning_audit":
                            _load_json_safe(Path("cleansing_reports.json"))},
            "placeholder_audit": _load_json_safe(
                Path("placeholders_audit.json")),
            "robustness_report": full_study.get("material_report", {}) if
            full_study else {},
            "call_sequence": ("validate_inputs -> clean_raw_data -> "
                              "preprocess_real_data -> "
                              "prepare_simulation_data -> "
                              "run_study_pipeline (Tasks 5-10) -> "
                              "run_full_study (Task 12) -> "
                              "produce_final_deliverables (Task 13)"),
        }
        # Diagnostics and coefficient paths from the study records
        diagnostics: pd.DataFrame = _diagnostics_frame(study_results)
        coefficient_paths: Dict[str, Any] = _coefficient_paths(study_results)
        density_samples: Dict[str, Any] = _density_samples(study_results)
        # Task 13: the deliverable generation
        deliverables: Dict[str, Any] = produce_final_deliverables(
            artifacts, study_results.summary, diagnostics,
            artifacts["frontiers"], coefficient_paths, density_samples,
            plot_config, output_dir,
        )
        step.artifacts = deliverables
        step.status = "COMPLETED"
    except Exception as exc:  # noqa: BLE001
        step.status = "FAILED"
        step.exception = f"{type(exc).__name__}: {exc}"
        logger.error(f"step task_13 failed: {step.exception}")
    return step

# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, helper callables (JSON loading, derivations)
# -------------------------------------------------------------------------------------------------------------------------------
def _load_json_safe(path: Path) -> Dict[str, Any]:
    """
    Load a JSON artifact, tolerant of its absence.

    Parameters
    ----------
    path : Path
        The artifact path.

    Returns
    -------
    Dict[str, Any]
        The parsed content, or a NOT AVAILABLE marker when absent.

    """
    # An absent audit file is recorded, never fabricated
    if not path.exists():
        return {"status": "NOT AVAILABLE", "path": str(path)}
    return json.loads(path.read_text(encoding="utf-8"))

def _frontier_dumps(study_results: StudyResults) -> Dict[str, Any]:
    """
    Render the frontier reports as JSON-native dumps.

    Parameters
    ----------
    study_results : StudyResults
        The study results carrying the per-identity FrontierReports.

    Returns
    -------
    Dict[str, Any]
        The envelope arrays keyed by identity.

    """
    dumps: Dict[str, Any] = {}
    for identity, report in study_results.frontiers.items():
        # Report objects without envelope data are skipped honestly
        if report is None:
            continue
        dumps[str(identity)] = {
            "envelope_phi": report.envelope_phi.tolist(),
            "envelope_logloss": report.envelope_logloss.tolist(),
            "accuracy_gaps": report.accuracy_gaps
            if isinstance(report.accuracy_gaps, dict) else {},
            "candidate_count": int(report.candidate_count),
        }
    return dumps

def _diagnostics_frame(study_results: StudyResults) -> pd.DataFrame:
    """
    Derive the diagnostics table from the run records.

    B-23 remediation: no defect of its own, but the CSER columns it carries
    were contaminated upstream. The frame now also carries the
    CSER eligible-feature count when the run record supplies it, so the
    support of CSER_min and CSER_max is visible in every downstream table
    rather than having to be inferred.

    Parameters
    ----------
    study_results : StudyResults
        The study results carrying the per-run records.

    Returns
    -------
    pd.DataFrame
        One row per structured-model record with the EVR, DDR and CSER
        fields plus the CSER support count.

    """
    rows: List[Dict[str, Any]] = []
    for record in study_results.run_records:
        # Only records carrying a structured decomposition have diagnostics
        if not pd.notna(record.get("evr")):
            continue
        # The eligible-feature roster is carried when the record has it,
        # so the CSER support travels with the statistic
        eligible_count: Any = record.get("cser_eligible_count")
        if eligible_count is None:
            roster: Any = record.get("cser_eligible")
            eligible_count = len(roster) if isinstance(
                roster, (list, tuple, set)) else pd.NA
        rows.append({
            "dataset": record["dataset"], "mode": record["mode"],
            "model_class": record["model_class"], "lambda": record["lambda"],
            "evr": record["evr"], "ddr": record["ddr"],
            "cser_min": record["cser_min"], "cser_max": record["cser_max"],
            "cser_eligible_count": eligible_count,
        })
    return pd.DataFrame(rows)

def _coefficient_paths(study_results: StudyResults) -> Dict[str, Any]:
    """
    Derive the structured coefficient paths per dataset.

    Parameters
    ----------
    study_results : StudyResults
        The study results.

    Returns
    -------
    Dict[str, Any]
        Per-dataset feature-to-path mappings (the record's
        parameter_path lists).

    """
    paths: Dict[str, Any] = {}
    for record in study_results.run_records:
        dataset: str = str(record["dataset"])
        parameter_path = record.get("parameter_path")
        if not isinstance(parameter_path, list) or not parameter_path:
            continue
        paths.setdefault(dataset, {})[f"feature_{record.get('lambda')}"] = [
            float(v) for v in parameter_path
        ]
    return paths

def _density_samples(study_results: StudyResults) -> Dict[str, Any]:
    """
    Derive the simulation score-density samples per lambda.

    Parameters
    ----------
    study_results : StudyResults
        The study results (simulation records and frontiers).

    Returns
    -------
    Dict[str, Any]
        Per-lambda density samples where available (empty when the study
        results carry no simulation predictions in the record schema).

    """
    # The run records do not persist the score vectors by design (Task 7
    # releases them); the density step consumes the persisted arrays the
    # caller supplies through the frontier dumps when present
    return {}

# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
def run_findr_end_to_end(
    config: Dict[str, Any],
    raw_structured_features_by_dataset: Dict[str, pd.DataFrame],
    raw_outcome_labels_by_dataset: Dict[str, pd.DataFrame],
    raw_sensitive_attribute_by_dataset: Dict[str, pd.DataFrame],
    raw_simulation_input: pd.DataFrame,
    schema_continuous: Dict[str, List[str]],
    schema_categorical: Dict[str, List[str]],
    subset: Optional[Dict[str, Sequence[Any]]] = None,
    output_dir: Path = Path("deliverables"),
    preserved_models: Optional[Dict[str, torch.nn.Module]] = None,
) -> Dict[str, Any]:
    """
    Execute the complete findr study end to end.

    The graph invokes each task-scoped orchestrator exactly once:
    Tasks 1-2 (validation and cleansing), Task 3 (preprocessing),
    Task 4 (simulation), Tasks 5-10 through Task 11's engine (models,
    loss, training, evaluation, diagnostics, frontiers), Task 12 (matrix
    accounting, robustness, paper verification) and Task 13 (deliverables
    with the reproducibility audit). The returned dictionary contains
    the artifacts of every executed step.

    Parameters
    ----------
    config : Dict[str, Any]
        The study configuration.
    raw_structured_features_by_dataset : Dict[str, pd.DataFrame]
        The raw feature frames.
    raw_outcome_labels_by_dataset : Dict[str, pd.DataFrame]
        The raw label frames.
    raw_sensitive_attribute_by_dataset : Dict[str, pd.DataFrame]
        The raw sensitive frames.
    raw_simulation_input : pd.DataFrame
        The raw simulation frame.
    schema_continuous, schema_categorical : Dict[str, List[str]]
        The per-dataset feature schemas.
    subset : Optional[Dict[str, Sequence[Any]]], optional
        Dimension filters; the full matrix is the default.
    output_dir : Path, optional
        The deliverables directory, by default deliverables.
    preserved_models : Optional[Dict[str, torch.nn.Module]], optional
        Best models preserved from a checkpointed production run.

    Returns
    -------
    Dict[str, Any]
        The step registry: one ExecutionStep per stage, plus the registry
        block with the subset, the output directory and the run identity.

    """
    if subset is None:
        subset = {}
    logger.info("Starting the findr end-to-end orchestrator.")

    # Step 1 (Tasks 1-2): validation and cleansing
    step_1: ExecutionStep = step_tasks_1_2_validate_and_clean(
        config, raw_structured_features_by_dataset,
        raw_outcome_labels_by_dataset, raw_sensitive_attribute_by_dataset,
        raw_simulation_input, schema_continuous, schema_categorical,
    )
    logger.info(f"tasks_1_2: {step_1.status}")

    # The Task 1 report and the Task 2 bundle feed the next stages
    validation_report = step_1.artifacts.get("validation_report") if \
        step_1.status == "COMPLETED" else None
    cleaned_bundle = step_1.artifacts.get("cleaned_bundle") if \
        step_1.status == "COMPLETED" else None

    # Step 2 (Task 3): preprocessing of the real datasets
    step_2: ExecutionStep = step_task_3_preprocess(
        cleaned_bundle, config, schema_continuous, schema_categorical)
    logger.info(f"task_3: {step_2.status}")

    # Step 3 (Task 4): simulation preparation
    step_3: ExecutionStep = step_task_4_simulation(
        validation_report, config)
    logger.info(f"task_4: {step_3.status}")

    # Step 4 (Tasks 5-10 via Task 11): the study run engine
    step_4: ExecutionStep = step_tasks_5_10_engine(
        cleaned_bundle, validation_report,
        raw_structured_features_by_dataset, raw_outcome_labels_by_dataset,
        raw_sensitive_attribute_by_dataset, raw_simulation_input,
        config, schema_continuous, schema_categorical, subset,
    )
    logger.info(f"tasks_5_10_engine: {step_4.status}")

    # Step 5 (Task 12): full-study accounting, robustness, verification.
    # The matrix pass is idempotent against the run log: Step 4's keys
    # are skipped, so no run is duplicated.
    step_5: ExecutionStep = step_task_12_full_study(
        config, raw_structured_features_by_dataset,
        raw_outcome_labels_by_dataset, raw_sensitive_attribute_by_dataset,
        raw_simulation_input, schema_continuous, schema_categorical, subset,
    )
    logger.info(f"task_12: {step_5.status}")

    # Step 6 (Task 13): deliverables and the reproducibility audit
    study_results: Optional[StudyResults] = step_4.artifacts.get(
        "study_results") if step_4.status == "COMPLETED" else None
    full_study: Optional[Dict[str, Any]] = step_5.artifacts if \
        step_5.status == "COMPLETED" else None
    step_6: ExecutionStep = step_task_13_deliverables(
        study_results, full_study, config,
        config["plot_construction"], output_dir, preserved_models,
    )
    logger.info(f"task_13: {step_6.status}")

    # The registry: every step plus the run identity
    registry: Dict[str, Any] = {
        step.step_id: step for step in (step_1, step_2, step_3, step_4,
                                        step_5, step_6)
    }
    registry["registry"] = {
        "subset": subset,
        "output_dir": str(output_dir),
        "steps": list(STEP_IDS),
        "completed_steps": [step_id for step_id in STEP_IDS
                            if registry[step_id].status == "COMPLETED"],
        "failed_steps": [step_id for step_id in STEP_IDS
                         if registry[step_id].status == "FAILED"],
    }
    logger.info(
        f"findr pipeline finished: "
        f"{len(registry['registry']['completed_steps'])} completed, "
        f"{len(registry['registry']['failed_steps'])} failed."
    )
    return registry
````

# Study Run Configuration

````python
# Study Run Configuration

# ==============================================================================
# Study Run Configuration: the full experimental matrix, data-gated
# ==============================================================================

# The former "Study Execution" cell ran the pipeline unconditionally against
# a synthetic fixture with one dataset, one model class, one lambda and one
# seed, and printed a completion banner indistinguishable from a real run.
# This cell replaces it with a matrix declaration plus a GATE: the study runs
# only against the eight real datasets, and any partial inventory is reported
# as such rather than presented as a reproduction.

# The full matrix the manuscript specifies (Section 5): three model classes,
# ten seeds, the complete lambda paths, both simulation modes.
FULL_STUDY_SUBSET: Dict[str, Any] = {
    "datasets": list(REAL_DATASET_KEYS),
    "modes": list(SIM_MODES),
    "models": ["LR", "NN", "findr"],
    "lambdas": [0, 0.01, 0.05, 0.1, 0.5, 1, 2, 3, 4],
    "lambdas_real": [0, 0.01, 0.05, 0.1, 0.5, 1, 2],
    "num_seeds": 10,
}


def assess_study_readiness(
    data_root: Path = DEFAULT_DATA_ROOT,
) -> Dict[str, Any]:
    """
    Report whether the eight real datasets are present, without loading them.

    A reproduction claim is only meaningful when the study population is the
    study population. This inventory is deliberately cheap -- it stats the
    filesystem rather than parsing -- so an operator can see what remains to
    be sourced before committing to a full run.

    Parameters
    ----------
    data_root : Path, optional
        The directory searched for the provider files.

    Returns
    -------
    Dict[str, Any]
        available, missing, the searched root, and a reproduction_valid flag
        that is True only when all eight datasets are present.

    """
    available: List[str] = []
    missing: List[Dict[str, str]] = []
    for key in REAL_DATASET_KEYS:
        spec: DatasetSourceSpec = REAL_DATASET_SOURCES[key]
        # A dataset is available when any accepted filename resolves
        if any((data_root / name).is_file() for name in spec.filenames):
            available.append(key)
        else:
            metadata: DatasetMetadata = REAL_DATASET_METADATA[key]
            missing.append({
                "dataset": key, "display_name": metadata.display_name,
                "expected_any_of": ", ".join(
                    str(data_root / n) for n in spec.filenames),
                "provider": metadata.provider})
    return {
        "data_root": str(data_root),
        "available": available,
        "missing": missing,
        # The reproduction gate: all eight, or the run is not the study
        "reproduction_valid": len(available) == len(REAL_DATASET_KEYS),
    }


def run_full_findr_study(
    data_root: Path = DEFAULT_DATA_ROOT,
    output_dir: Path = Path("deliverables_full"),
    subset: Optional[Dict[str, Any]] = None,
    allow_partial: bool = False,
) -> Dict[str, Any]:
    """
    Ingest the real datasets, build the simulation, and run the full study.

    Execution graph: ingest -> simulate (both modes, 10 seeds) -> configure
    -> run_findr_end_to_end over Tasks 1 to 13.

    Parameters
    ----------
    data_root : Path, optional
        The directory holding the provider files.
    output_dir : Path, optional
        The deliverables destination.
    subset : Optional[Dict[str, Any]], optional
        The experimental matrix; FULL_STUDY_SUBSET by default.
    allow_partial : bool, optional
        When False (default) a missing dataset aborts. When True the study
        runs on the available subset and the returned registry is flagged
        reproduction_valid = False, so no downstream report can present a
        partial run as a reproduction of the manuscript.

    Returns
    -------
    Dict[str, Any]
        The pipeline registry, augmented with the readiness assessment and
        the reproduction_valid flag.

    Raises
    ------
    RuntimeError
        When datasets are missing and allow_partial is False.

    """
    # The inventory is taken before any expensive work begins
    readiness: Dict[str, Any] = assess_study_readiness(data_root)
    if not readiness["reproduction_valid"]:
        # The message names every missing dataset and its provider, because
        # closing this gap is a sourcing task, not a coding one
        lines: List[str] = [
            f"  - {m['dataset']} ({m['display_name']}): expected any of "
            f"{m['expected_any_of']}; obtain from {m['provider']}"
            for m in readiness["missing"]]
        detail: str = "\n".join(lines)
        if not allow_partial:
            raise RuntimeError(
                f"the study cannot run: {len(readiness['missing'])} of "
                f"{len(REAL_DATASET_KEYS)} real datasets are absent from "
                f"'{data_root}'.\n{detail}\n"
                f"Pass allow_partial=True to run on the available subset; "
                f"the result will be flagged as NOT a reproduction."
            )
        logger.warning(
            f"running on a PARTIAL inventory: "
            f"{len(readiness['available'])} of {len(REAL_DATASET_KEYS)} "
            f"datasets. This is not a reproduction of the manuscript.\n"
            f"{detail}")
    # Ingest whatever is available; strict when a full run was demanded
    features_by, labels_by, sensitive_by, ingestion_report = \
        ingest_real_datasets(data_root, readiness["available"],
                             strict=not allow_partial)
    # The simulation is generated in full: both modes, ten seeds each
    active_subset: Dict[str, Any] = dict(
        subset if subset is not None else FULL_STUDY_SUBSET)
    simulation_frame: pd.DataFrame = build_simulation_input(
        modes=active_subset.get("modes", SIM_MODES),
        seeds=tuple(range(int(active_subset.get("num_seeds", 10)))),
    )
    # The configuration's metadata block covers the LOADED datasets, so the
    # Task 1 Step 2 gate compares like with like while remaining independent
    config: Dict[str, Any] = build_study_config(readiness["available"])
    # The schema lists follow from the loaded dtypes
    continuous, categorical = build_schema_lists(features_by)
    # The subset's dataset list is narrowed to what actually loaded
    active_subset["datasets"] = list(features_by)
    # Tasks 1 to 13, end to end
    registry: Dict[str, Any] = run_findr_end_to_end(
        config, features_by, labels_by, sensitive_by, simulation_frame,
        continuous, categorical, active_subset, output_dir=output_dir,
    )
    # The provenance travels with the result: a consumer of this registry
    # can always tell whether it constitutes a reproduction
    registry["readiness"] = readiness
    registry["ingestion_report"] = ingestion_report
    registry["reproduction_valid"] = bool(readiness["reproduction_valid"])
    return registry


# ------------------------------------------------------------------------------
# Readiness report (no execution): running the study is an explicit call
# ------------------------------------------------------------------------------
# Importing this module must never launch a multi-hour job, so the cell
# reports the inventory and stops. Start the study with:
#     registry = run_full_findr_study(Path("data/real"))
STUDY_READINESS: Dict[str, Any] = assess_study_readiness()
print(
    f"findr study readiness | data root: {STUDY_READINESS['data_root']} | "
    f"{len(STUDY_READINESS['available'])} of {len(REAL_DATASET_KEYS)} "
    f"datasets present | reproduction_valid="
    f"{STUDY_READINESS['reproduction_valid']}"
)
if STUDY_READINESS["missing"]:
    print("Missing datasets (sourcing required; no substitute is generated):")
    for entry in STUDY_READINESS["missing"]:
        print(f"  - {entry['dataset']:<14} {entry['display_name']:<24} "
              f"from {entry['provider']}")
    print("Place the files under the data root, then call "
          "run_full_findr_study().")
else:
    print("All eight datasets present. Start the study with: "
          "run_full_findr_study()")
````
